# Теоретические сведения

## Часть 1. Анализ дискретных систем во временной и z-областях

### 1. Импульсная характеристика и передаточная функция
* **Импульсная характеристика (ИХ) $h[n]$** — реакция линейной дискретной стационарной (ЛДС) системы на единичный импульс $u_0[n]$ при нулевых начальных условиях.
* **Передаточная функция $H(z)$** — $\mathcal{Z}$-преобразование импульсной характеристики:
  $$H(z) = \sum_{n=0}^{\infty} h[n] z^{-n} = \frac{\sum_{i=0}^{N} b_i z^{-i}}{1 + \sum_{k=1}^{M} a_k z^{-k}}$$
* **Длина ИХ:** для рекурсивного звена, где значение ипульсной характеристики не равны нулю


### 2. Дискретная свёртка и реакция на прямоугольный импульс
* **Формула свёртки:**
  $$y_1[n] = x[n] * h[n] = \sum_{m=0}^{N_2-1} x[m] h[n - m]$$
* **Длина воздействия:** $N_2$ отсчётов (на интервале $n \in [0, N_2-1]$).
* **Длина свёртки:**
  * Аналитически и по полному графику длина переходного процесса свёртки строго равна:
    $$L_y = N_1 + N_2 - 1$$
  * **Причина ограничения до длины воздействия $N_2$:** в прикладных задачах и лабораторном анализе реакцию часто рассматривают только на интервале действия входного сигнала, отсекая затухающий переходный «хвост» свободных колебаний после окончания импульса.


### 3. Разностное уравнение
* **Разностное уравнение звена 2-го порядка:**
  $$y[n] = b_0 x[n] + b_1 x[n-1] + b_2 x[n-2] - a_1 y[n-1] - a_2 y[n-2]$$
* **Длины сигналов:** длина воздействия — $N_2$; при расчёте через разностное уравнение реакция вычисляется на том же интервале длины $N_2$.


### 4. Нули, полюсы и факторизация передаточной функции
* **Определения:**
  * **Нули ($z_{0i}$):** корни числителя $B(z) = 0$, при которых $H(z) = 0$.
  * **Полюсы ($p_k$):** корни знаменателя $A(z) = 0$, при которых $H(z) \to \infty$.
* **Формы представления:**
  * *Алгебраическая:* $z = \text{Re}(z) + j\,\text{Im}(z)$.
  * *Показательная:* $z = r e^{j\theta}$, где $r = \sqrt{\text{Re}^2 + \text{Im}^2}$ — радиус, $\theta = \operatorname{arctg2}(\text{Im}, \text{Re})$ — аргумент (выражается в долях $\pi$: $\theta = c \cdot \pi$).
* **Факторизованная форма ПФ:**
  $$H(z) = b_0 \frac{(1 - z_{01} z^{-1})(1 - z_{02} z^{-1})}{(1 - p_1 z^{-1})(1 - p_2 z^{-1})}$$

### 5. Карта нулей и полюсов и устойчивость
* **Критерий устойчивости:** причинная линейная дискретная система устойчива тогда и только тогда, когда все её полюсы лежат **строго внутри единичной окружности** на $z$-плоскости:
  $$|p_k| < 1 \quad \forall k$$
  Положение нулей на устойчивость не влияет.
* **Верификация:** координаты нулей и полюсов на комплексной плоскости должны строго совпадать с корнями полиномов числителя и знаменателя.


### 6. Частотные характеристики в шкале нормированных частот $\hat{\omega}$
* **АЧХ и ФЧХ:** получаются подстановкой $z = e^{j\hat{\omega}}$:
  $$H(e^{j\hat{\omega}}) = |H(e^{j\hat{\omega}})| e^{j \arg H(e^{j\hat{\omega}})}$$
* **Границы основной полосы:** интервал однозначности составляет $\hat{\omega} \in [0, \pi]$ рад/отсчёт (для полного периода — $[0, 2\pi]$).
* **Связь карты с АЧХ:**
  * Близость полюса к единичной окружности вызывает резонансный подъём АЧХ на частоте $\hat{\omega} \approx \arg(p)$.
  * Близость нуля формирует провал (подавление) АЧХ на частоте $\hat{\omega} \approx \arg(z_0)$.
* **Фазовый скачок на $\pi$:** происходит в точках, где АЧХ обращается в точный ноль (нуль лежит строго на единичной окружности, $|z_0| = 1$), так как комплексный коэффициент передачи меняет знак.
* **Характер фильтрации:** в зависимости от расположения полюсов и нулей фильтр подавляет низкочастотные (ФВЧ) либо высокочастотные (ФНЧ) гармоники входного импульса.


### 7. Частотные характеристики в шкале физических частот $f$ (Гц)
* **Связь частот:**
  $$\hat{\omega} = 2\pi \frac{f}{f_д} \iff f = \frac{\hat{\omega}}{2\pi} f_д$$
* **Границы основной полосы:** диапазон Найквиста от $0$ до $f_д / 2$ Гц.


## Часть 2. Дискретное преобразование Фурье (ДПФ) и спектральный анализ

### 1. Связь шкал времени и частоты
* **Время:** непрерывное время $t$ связано с номером отсчёта $n$ через период дискретизации $T = 1/f_д$:
  $$t = n T = \frac{n}{f_д}$$
* **Частота:** физическая частота $f$ (Гц) связана с нормированной частотой $\hat{f}$ (в долях от $f_д$) соотношением:
  $$\hat{f} = \frac{k}{N} \implies f = \hat{f} \cdot f_д = \frac{k}{N} f_д$$
  где $\Delta f = f_д / N$ — шаг дискретизации по частоте (разрешение ДПФ).


### 2. Прямое и обратное ДПФ
* **Прямое ДПФ:**
  $$X[k] = \sum_{n=0}^{N-1} x[n] e^{-j \frac{2\pi}{N} k n}, \quad k = 0, \dots, N-1$$
* **Обратное ДПФ (ОДПФ):**
  $$x[n] = \frac{1}{N} \sum_{k=0}^{N-1} X[k] e^{j \frac{2\pi}{N} k n}, \quad n = 0, \dots, N-1$$
* Тождественность синтеза гарантирует совпадение $x[n]$ и $\text{ОДПФ}(\text{ДПФ}(x))$ с точностью до машинного эпсилон ($\sim 10^{-15}$).


### 3. Свойства амплитудного и фазового спектров
* **Симметрия (для вещественных сигналов):**
  * Амплитудный спектр чётно-симметричен: $|X[k]| = |X[N - k]|$.
  * Фазовый спектр нечётно-симметричен: $\arg X[k] = -\arg X[N - k]$.
* **Растекание спектра:**
  * Отсутствует, если в окно $N$ укладывается целое число периодов каждой гармоники ($f_{\text{гарм}} = k \cdot \Delta f$). Спектр состоит из строго изолированных линий.
  * При некратности частот шагу ДПФ энергия гармоники растекается по соседним отсчётам.


### 4. Связь модуля ДПФ и амплитудного спектра периодического сигнала
* ДПФ конечной последовательности длины $N$ соответствует дискретному ряду Фурье (ДРФ) её периодического повторения:
$$\frac{X[k]}{N} = \frac{1}{N} X(e^{j\hat{\omega}})\Big\vert{}_{\hat{\omega}=\frac{2\pi}{N}k} \implies \frac{\vert{}X[k]\vert{}}{N} = \frac{1}{N}\left\vert{}X(e^{j\hat{\omega}})\right\vert{}_{\hat{\omega}=\frac{2\pi}{N}k}$$
* Форма и пропорции спектров совпадают, различие заключается в нормировке:
  * Модуль ДПФ $|X[k]|$ суммирует значения по окну, его пики пропорциональны $N$ ($|X| = \frac{A N}{2}$).
  * Спектр периодического сигнала  нормирован на $N$ и равен физическим полуамплитудам гармоник ($\frac{A}{2}$).


### 5. Восстановление непрерывного сигнала по отсчётам ДПФ
* **Центрирование спектра $X_{\text{a}}(k)$:** перенос отсчётов $k \in [N/2, N-1]$ в область отрицательных индексов $[-N/2, -1]$ переводит спектр в симметричный двухсторонний диапазон частот $[-f_д/2, f_д/2)$.
* $$X_{\text{a}}(k) = \begin{cases} X[k], & 0 \leqslant k \leqslant \frac{N}{2} - 1 \\  X[N + k], & -\frac{N}{2} \leqslant k \leqslant -1 \end{cases}$$
* **Интерполяционная формула:**
  $$x(t) = \frac{1}{N} \sum_{k=-N/2}^{N/2-1} X_{\text{a}}(k) \exp\left(j \frac{2\pi}{N T} k t\right)$$
* **Теорема Котельникова–Шеннона:** восстановление выполняется точно, если максимальная частота сигнала не превышает частоту Найквиста ($f_{\text{макс}} < f_д / 2$). В этом случае тригонометрический базис ДПФ формирует идеальную интерполяцию без наложения спектров.


### 6. Равенство Парсеваля
* **Математическая запись:**
  $$\sum_{n=0}^{N-1} |x[n]|^2 = \frac{1}{N} \sum_{k=0}^{N-1} |X[k]|^2$$

```python
# реализация разностного уравнения

def difference_equation(x, b, a):
  """вычисление реакции y(n) по разностному уравнению:
  """
  x = np.asarray(x, dtype=float)
  b = np.asarray(b, dtype=float)
  a = np.asarray(a, dtype=float)

  N = len(b)
  M = len(a)
  length = len(x)

  y = np.zeros(length)

  for n in range(length):
   
    sum_b = sum(b[i] * x[n - i] for i in range(N) if n - i >= 0)

    
    sum_a = sum(a[k] * y[n - k] for k in range(1, M) if n - k >= 0)

    y[n] = sum_b - sum_a

  return y
```

```python
u0 = np.zeros(N1)
u0[0] = 1

# ипульсная характеристика
h = difference_equation(u0, b, a)
```

```python
def g(n):
  """
  g(n) = (0.83)^n * (cos(2.2*n) - 0.73*sin(2.2*n)) * u_1(n)
  """
  n = np.asarray(n, dtype=float)
  res = np.zeros_like(n)

  # Множитель u_1(n): функция единичного скачка (равна 1 при n >= 0 и 0 при n < 0)
  mask = n >= 0
  n_pos = n[mask]

  res[mask] = (0.83**n_pos) * (
      np.cos(2.2 * n_pos) - 0.73 * np.sin(2.2 * n_pos)
  )
  return res


def h_analytic(n):
  """Итоговое аналитическое решение h(n):
  h(n) = 0.66*g(n) - 0.74*g(n-1) + 0.92*g(n-2)
  """
  n = np.asarray(n)
  return 0.66 * g(n) - 0.74 * g(n - 1) + 0.92 * g(n - 2)
```

### Сверточная функция 


```python
def discrete_convolution(x, h, mode='full'):
  """Вычисление дискретной свёртки двух одномерных сигналов.
     'full' — полная свертка (длина Lx + Lh - 1),
     'same' — свертка, усеченная до длины входного сигнала Lx
  """
  x = np.asarray(x, dtype=float)
  h = np.asarray(h, dtype=float)

  Lx = len(x)
  Lh = len(h)
  Ly = Lx + Lh - 1

  y = np.zeros(Ly)

  for n in range(Ly):
    k_min = max(0, n - Lh + 1)
    k_max = min(n, Lx - 1)

    acc = 0.0
    for k in range(k_min, k_max + 1):
      acc += x[k] * h[n - k]
    y[n] = acc

  if mode == 'same':
    return y[:Lx]

  return y
```

```python
# Входное воздействие 
n = np.arange(N2)
x = np.zeros(N2)
x[0:19] = 1.0

n_axis = np.arange(N1)

# Вычисление импульсной характеристики y1(n)
h = h_analytic(n_axis)

# Вычисление свертки y1(n)
y1 = discrete_convolution(x, h, 'same')
```

$n$ - это порядковый номер такта (отсчёта): $0, 1, 2, \dots, 63$.   $nT$ - реальное время в секундах.   Связь прямая: время получается простым умножением номера отсчёта на шаг дискретизации: $t_n = n \cdot T = \frac{n}{f_д}$.  

b. дискретная частота ($f$) и нормированная ($\hat{f}$)

Нормированная частота $\hat{f} = \frac{k}{N}$ показывает долю от частоты дискретизации ($0 \le \hat{f} < 1$).   дискретная частота $f$ измеряется в герцах (Гц).Связь: чтобы перевести долю в герцы, её умножают на частоту дискретизации: $f = \hat{f} \cdot f_д$.

c. Вид амплитудного и фазового спектров.


Амплитудный спектр: Линейчатый без растекания (частоты строго кратны шагу ДПФ) и чётно-симметричный относительно частоты ($4000$ Гц / $\hat{f} = 0.5$). Содержит два основных пика на $1000$ Гц ($\approx 34.6$) и $1500$ Гц ($\approx 69.1$, вдвое выше из-за $A_2 = 2A_1$), а также их зеркальные копии равной высоты на $6500$ Гц и $7000$ Гц.<br>Фазовый спектр: Строго нулевой во всех точках, кроме частот гармоник, и обладает нечётной симметрией относительно центра. На частотах $1000$ Гц и $1500$ Гц отсчёты равны начальным фазам гармоник ($+\pi/4$ и $+\pi/8$), а на зеркальных частотах $7000$ Гц и $6500$ Гц принимают противоположные по знаку значения ($-\pi/4$ и $-\pi/8$).

Вывести графики в шкале дискретных нормированных частот модуля ДПФ конечной последовательности и амплитудного спектра периодической последовательности. Пояснить связь модуля ДПФ конечной последовательности с амплитудным спектром периодической последовательности.

```python
# Вычисление ДПФ конечной последовательности длины N
X = np.fft.fft(x)
mag_X = np.abs(X)  # Модуль ДПФ конечной последовательности

# Амплитудный спектр периодической последовательности (коэффициенты ряда Фурье ДРФ)
# (1 / N) * X[k]
amp_periodic = mag_X / N

# Шкала дискретных нормированных частот (в долях от f_d)
k = np.arange(N)
f_hat = k / N

# Построение графиков
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(9, 7))

# Модуль ДПФ конечной последовательности
markerline1, stemlines1, baseline1 = ax1.stem(f_hat, mag_X, basefmt="black")
plt.setp(stemlines1, color="royalblue", linewidth=1.4)
plt.setp(markerline1, color="royalblue", markersize=5)
ax1.set_title(r"Модуль ДПФ конечной последовательности $|X[k]|$")
ax1.set_xlabel(r"Дискретная нормированная частота $\hat{f} = k/N$")
ax1.set_ylabel(r"$|X[k]|$")
ax1.set_xlim(-0.02, 1.02)
ax1.grid(True, linestyle="--", alpha=0.5)

# Амплитудный спектр периодической последовательности
markerline2, stemlines2, baseline2 = ax2.stem(
    f_hat, amp_periodic, basefmt="black"
)
plt.setp(stemlines2, color="crimson", linewidth=1.4)
plt.setp(markerline2, color="crimson", markersize=5)
ax2.set_title(
    r"Амплитудный спектр периодической последовательности $"
    r" \frac{1}{N}|X[k]|$"
)
ax2.set_xlabel(r"Дискретная нормированная частота $\hat{f} = k/N$")
ax2.set_ylabel(r"$|C_k|$")
ax2.set_xlim(-0.02, 1.02)
ax2.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()

# Контрольный вывод амплитуд
print(
    f"Пики модуля ДПФ: |X[8]| = {mag_X[8]:.2f}, |X[12]| = {mag_X[12]:.2f}"
)  # A*N/2
print(
    f"Пики спектра периодического сигнала: |C[8]| = {amp_periodic[8]:.3f},"
    f" |C[12]| = {amp_periodic[12]:.3f}"
)  # A/2
```

Связь между модулем ДПФ конечной последовательности $\vert{}X[k]\vert{}$ и амплитудным спектром периодического сигнала выражается формулой:$$\frac{\vert{}X[k]\vert{}}{N} = \frac{1}{N}\left\vert{}X(e^{j\hat{\omega}})\right\vert{}_{\hat{\omega}=\frac{2\pi}{N}k}$$Форма и частотные положения линий обоих спектров полностью идентичны, так как периодический сигнал получен повторением исходного блока длины $N$. Различие заключается исключительно в вертикальном масштабе: модуль ДПФ $\vert{}X[k]\vert{}$ накапливает сумму отсчётов по окну и растёт пропорционально длине выборки ($\vert{}X\vert{} = \frac{AN}{2} \approx 34.6$ и $69.1$), а спектр периодического сигнала $ \frac{\vert{}X[k]\vert{}}{N}$ усреднён по периоду и отражает истинные физические полуамплитуды гармоник ($ \frac{A}{2} \approx 0.54$ и $1.08$).

```python
# Исходные параметры и ДПФ
X = np.fft.fft(x)

# Центрирование спектра согласно формуле:
k_arr = np.arange(-N // 2, N // 2)
X_a = np.fft.fftshift(X)

# Временная сетка для аналогового сигнала 
# Шаг Delta_t = 0.25 * T на интервале [0, (N - 1) * T]
delta_t = 0.25 * T
t = np.arange(0, (N - 1) * T + delta_t / 2, delta_t)

# Восстановление x(t) по формуле 
# Реализация через матричное умножение:
exp_matrix = np.exp(1j * (2 * np.pi / (N * T)) * np.outer(t, k_arr))
x_rec = np.real((1.0 / N) * np.dot(exp_matrix, X_a))

# Исходный аналоговый сигнал x(t) = x(nT)|_{nT=t}
x_orig = A1 * np.cos(2 * np.pi * f1 * t + np.pi / 4) + A2 * np.cos(
    2 * np.pi * f2 * t + np.pi / 8
)

# Амплитудный спектр непрерывного периодического сигнала|X_a(k)| / N
amp_spectrum_a = np.abs(X_a) / N
f_analog = k_arr / (N * T)  # частоты от -f_d/2 до f_d/2 - delta_f

# Построение графиков
fig, axes = plt.subplots(3, 1, figsize=(10, 11))

# Последовательность x[n] и модуль её ДПФ |X[k]|
axes[0].stem(n, x, linefmt="royalblue", markerfmt="bo", basefmt="black")
axes[0].set_title(r"а) Исходная последовательность $x[n]$ на периоде $N$")
axes[0].set_xlabel(r"Дискретное нормированное время $n$ (отсчёты)")
axes[0].set_ylabel(r"$x[n]$")
axes[0].set_xlim(-1, N)
axes[0].grid(True, linestyle="--", alpha=0.5)

# (b, c) Сравнение исходного x(t) и восстановленного сигнала
t_ms = t * 1e3
axes[1].plot(
    t_ms,
    x_orig,
    color="silver",
    linewidth=4,
    label=r"в) Исходный аналоговый $x(t)$",
)
axes[1].plot(
    t_ms,
    x_rec,
    color="crimson",
    linestyle="--",
    linewidth=1.8,
    label=r"б) Восстановленный сигнал $x(t)$",
)
axes[1].stem(
    n * T * 1e3,
    x,
    linefmt="black",
    markerfmt="k.",
    basefmt="none",
    label=r"Отсчёты $x(nT)$",
)
axes[1].set_title(
    r"б, в) Восстановленный $x(t)$ и исходный аналоговый сигнал"
    rf" ($\Delta t = 0.25T$)"
)
axes[1].set_xlabel(r"Время $t$, мс")
axes[1].set_ylabel(r"$x(t)$")
axes[1].set_xlim(t_ms[0], t_ms[-1])
axes[1].legend(loc="upper right")
axes[1].grid(True, linestyle="--", alpha=0.5)

# Спектры: модуль ДПФ и амплитудный спектр аналогового сигнала
markerline, stemlines, baseline = axes[2].stem(
    f_analog, amp_spectrum_a, basefmt="black"
)
plt.setp(stemlines, color="forestgreen", linewidth=1.5)
plt.setp(markerline, color="forestgreen", markersize=5)
axes[2].set_title(
    r"б) Амплитудный спектр восстановленного сигнала $|C_k| ="
    r" \frac{1}{N}|X_a(k)|$"
)
axes[2].set_xlabel(r"Частота $f$, Гц")
axes[2].set_ylabel(r"$|C_k|$")
axes[2].grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()

# Контрольная проверка ошибки восстановления
err = np.max(np.abs(x_orig - x_rec))
print(f"Максимальная погрешность восстановления: {err:.2e}")
```

```python
import numpy as np

# Расчёт энергии во временной области
E_time = np.sum(np.abs(x) ** 2)

# Расчёт энергии в частотной области через отсчёты ДПФ
E_freq = (1.0 / N) * np.sum(np.abs(X) ** 2)

# 3. Вывод численных результатов проверки
print(f"Энергия во временной области (E_time):  {E_time:.8f}")
print(f"Энергия в частотной области  (E_freq):  {E_freq:.8f}")
print(f"Абсолютная разность:                   {abs(E_time - E_freq):.2e}")
```

Равенство Парсеваля выражает закон сохранения энергии: полная энергия сигнала во времени строго равна суммарной энергии его спектральных составляющих (с нормирующим множителем $1/N$).

ДПФ не рассеивает и не добавляет энергию, поэтому мощность и энергию сигнала можно одинаково точно рассчитывать как по отсчётам во времени, так и напрямую по спектру.
