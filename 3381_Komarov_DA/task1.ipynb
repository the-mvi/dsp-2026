## Основные теоретические положения

### Время и частота

Момент отсчёта равен $t=nT$, где $T=1/f_d$. Номер $n$ безразмерный. Нормированные частоты связаны с частотой в герцах так
$$\hat f=\frac{f}{f_d},\qquad \hat\omega=2\pi\hat f.$$
Основная полоса $-\pi\le\hat\omega<\pi$ соответствует $-f_d/2\le f<f_d/2$. Частота Найквиста равна $f_d/2$.

### Импульсная характеристика и свёртка

Единичный импульс $\delta[n]$ равен единице при $n=0$ и нулю при остальных $n$. Реакция на него при нулевых начальных условиях называется импульсной характеристикой $h[n]$. В работе вход и выход при $n<0$ равны нулю.

Для линейной системы с неизменными во времени параметрами
$$y[n]=\sum_{m=0}^{\infty}h[m]x[n-m].$$
Свёртка конечных массивов длины $N_h$ и $N_x$ содержит $N_h+N_x-1$ отсчётов. Это длина массива результата. У БИХ-системы сам отклик продолжается бесконечно.

### Разностное уравнение и передаточная функция

Звено второго порядка описывается уравнением
$$y[n]=b_0x[n]+b_1x[n-1]+b_2x[n-2]-a_1y[n-1]-a_2y[n-2].$$
Выход зависит от своих прошлых значений, поэтому система рекурсивная. При нулевых начальных условиях
$$H(z)=\frac{Y(z)}{X(z)}=\mathcal Z\{h[n]\}=\frac{b_0+b_1z^{-1}+b_2z^{-2}}{1+a_1z^{-1}+a_2z^{-2}}.$$
После сокращения общих множителей нули находятся из числителя, а полюса из знаменателя
$$b_0z^2+b_1z+b_2=0,\qquad z^2+a_1z+a_2=0.$$
Две формы записи корня связаны формулами
$$z=\xi+j\eta=re^{j\varphi},\qquad r=\sqrt{\xi^2+\eta^2},\qquad \varphi=\operatorname{atan2}(\eta,\xi).$$
Модуль задаёт расстояние до начала координат, аргумент задаёт угол.

### Аналитический отклик и устойчивость

Для сопряжённых полюсов $re^{\pm j\varphi}$ знаменатель равен $1-2r\cos\varphi\,z^{-1}+r^2z^{-2}$. Поэтому $r=\sqrt{a_2}$ и $\varphi=\arccos(-a_1/(2r))$.

По таблице $Z$-преобразований изображению $G(z)=1/(1+a_1z^{-1}+a_2z^{-2})$ соответствует последовательность
$$g[n]=\begin{cases}r^n\dfrac{\sin((n+1)\varphi)}{\sin\varphi},&n\ge0,\\0,&n<0.\end{cases}$$
Числитель задаёт сумму её сдвигов
$$h[n]=b_0g[n]+b_1g[n-1]+b_2g[n-2].$$
Отрицательные индексы сдвинутых последовательностей дают нулевые значения.

Устойчивая система сохраняет ограниченность выхода при ограниченном входе. Для причинной системы условие имеет вид
$$\sum_{n=0}^{\infty}|h[n]|<\infty.$$
Для рациональной передаточной функции это равносильно расположению всех полюсов внутри единичного круга. Множитель $r^n$ при $r<1$ задаёт затухание отклика.

### АЧХ и ФЧХ

Частотная характеристика получается подстановкой $z=e^{j\hat\omega}$. АЧХ равна $|H(e^{j\hat\omega})|$ и задаёт изменение амплитуды гармоники в установившемся режиме. ФЧХ равна $\arg H(e^{j\hat\omega})$ и задаёт фазовый сдвиг.

Полюс рядом с единичной окружностью обычно даёт подъём АЧХ, а нуль даёт спад в области своего аргумента. Нуль на самой окружности даёт точное обнуление АЧХ. Фаза там не определена. Скачок на $2\pi$ также может появиться из-за границы интервала главного аргумента.

### ДПФ и нормировка спектра

Прямое и обратное преобразования для $N$ отсчётов
$$X[k]=\sum_{n=0}^{N-1}x[n]e^{-j2\pi kn/N},\qquad x[n]=\frac1N\sum_{k=0}^{N-1}X[k]e^{j2\pi kn/N}.$$
Шаг частотной сетки равен $\Delta f=f_d/N$. До частоты Найквиста $f_k=k\Delta f$. Индексы $k>N/2$ соответствуют частотам $(k-N)\Delta f$.

Косинусоида $A\cos(2\pi k_0n/N+\varphi_0)$ при целом $0<k_0<N/2$ даёт две линии
$$X[k_0]=\frac{NA}{2}e^{j\varphi_0},\qquad X[N-k_0]=\frac{NA}{2}e^{-j\varphi_0}.$$
У вещественного сигнала $X[(N-k)\bmod N]=\overline{X[k]}$. Модули сопряжённых линий равны, а фазы противоположны по знаку.

ДПФ конечной записи даёт отсчёты её преобразования Фурье на частотной сетке. При периодическом повторении записи коэффициенты ряда равны $C_k=X[k]/N$. Двусторонние амплитуды равны $|X[k]|/N$. Амплитуда косинусоиды в одностороннем спектре равна $2|X[k]|/N$. Постоянную составляющую и линию Найквиста не удваивают.

### Восстановление и энергия

Для чётного $N$ отрицательным индексам сопоставляются отсчёты $X_a[k]=X[N+k]$. При $0\le k<N/2$ берутся $X_a[k]=X[k]$. Непрерывный сигнал восстанавливается по сумме
$$z(t)=\frac1N\sum_{k=-N/2}^{N/2-1}X_a[k]\exp\left(j\frac{2\pi k}{NT}t\right).$$
При $t=nT$ она совпадает с ОДПФ. Для гармоник из задания совпадение сохраняется и между отсчётами. Их частоты ниже Найквиста и попадают на сетку ДПФ.

Равенство Парсеваля связывает энергию записи с её спектром
$$E=\sum_{n=0}^{N-1}|x[n]|^2=\frac1N\sum_{k=0}^{N-1}|X[k]|^2.$$
Оба способа расчёта должны давать одно значение.

#### Импульсная характеристика


```python
def system_response(signal, numerator, denominator):
    # Нулевые значения перед началом входа и выхода
    x = np.pad(signal, (2, 0))
    y = np.zeros(len(x))
    for k in range(2, len(x)):
        y[k] = (numerator[0] * x[k] + numerator[1] * x[k-1] + numerator[2] * x[k-2]
                - denominator[1] * y[k-1] - denominator[2] * y[k-2]) / denominator[0]
    return y[2:]

poles = np.roots(a)
radius = np.abs(poles[0])
pole_angle = np.abs(np.angle(poles[0]))

extended_impulse = np.zeros(tail_window)
extended_impulse[0] = 1.0
extended_h = system_response(extended_impulse, b, a)
h = extended_h[:N1]
sample_numbers = np.arange(N1)
# Запаздывающие слагаемые начинаются с первого и второго отсчётов.
g = radius ** sample_numbers * np.sin((sample_numbers + 1) * pole_angle) / np.sin(pole_angle)
analytic_h = b0 * g
analytic_h[1:] += b1 * g[:-1]
analytic_h[2:] += b2 * g[:-2]

show_table(['Величина', 'Значение'], [
    ['Модуль полюса', f'{radius:.9f}'],
    ['Аргумент полюса / π', f'{pole_angle / np.pi:.9f}'],
    ['Расхождение с аналитической формулой', f'{np.max(np.abs(h - analytic_h)):.3e}'],
])
```

```python
figure, axes = plt.subplots(2, 1, figsize=(10, 5.8))
axes[0].stem(sample_numbers, h)
axes[0].set(title='Вычисленный фрагмент характеристики', xlabel='n', ylabel='h[n]')
axes[1].plot(np.arange(tail_window), extended_h)
axes[1].axvline(N1, linestyle=':', color='gray', label='Конец основного фрагмента')
axes[1].set(title='Продолжение отклика', xlabel='n', ylabel='h[n]')
axes[1].legend()
finish_figure(figure, axes)
```

На верхнем графике видны затухающие колебания. Продолжение до 83 отсчётов показывает оставшийся хвост. Его амплитуда уменьшается из-за множителя $r^n$ при $r<1$.

Численный и аналитический расчёты совпадают. Изображению $G(z)=1/(1+a_1z^{-1}+a_2z^{-2})$ соответствует последовательность
$$g[n]=\begin{cases}r^n\dfrac{\sin((n+1)\varphi)}{\sin\varphi},&n\ge0,\\0,&n<0.\end{cases}$$
По сдвигам этой последовательности получаем
$$h[n]=0{,}68g[n]+0{,}776832g[n-1]+1{,}088g[n-2],$$
$$r=\sqrt{0{,}694},\qquad \varphi=\arccos\frac{1{,}0028}{2\sqrt{0{,}694}}.$$
Второе слагаемое начинается при $n=1$, третье начинается при $n=2$. Все значения до начала последовательности равны нулю.

Основной массив содержит 29 отсчётов. Сам отклик бесконечен, поскольку полюса не сокращаются с нулями. Дополнительное окно тоже ограничивает только наблюдение.

#### 2. Воздействие и реакции (свёртка и рекурсия)


```python
pulse_width = int(np.floor(N2 / 2 + 0.5))
input_signal = np.zeros(N2)
input_signal[:pulse_width] = 1.0
convolution = np.convolve(input_signal, h)

# Выход с продолжением входа нулями
check_output = np.convolve(input_signal, extended_h)[:N2]
long_output = system_response(np.pad(input_signal, (0, tail_window - N2)), b, a)
recursive_output = long_output[:N2]
method_error = np.abs(convolution[:N2] - recursive_output)

show_table(['Величина', 'Значение'], [
    ['Единичных отсчётов входа', pulse_width],
    ['Длина массива свёртки', convolution.size],
    ['Длина ненулевой части свёртки', pulse_width + N1 - 1],
    ['Характеристика для дополнительной проверки', tail_window],
    ['Максимальная ошибка с короткой h[n]', f'{method_error.max():.8f}'],
    ['Ошибка с длинной h[n] на первых 39 отсчётах', f'{np.max(np.abs(check_output - recursive_output)):.3e}'],
])
```

```python
figure, axes = plt.subplots(4, 1, figsize=(10, 10))
axes[0].stem(np.arange(N2), input_signal)
axes[0].set(title='Прямоугольное воздействие', ylabel='x[n]')
axes[1].plot(np.arange(tail_window), long_output, label='Рекурсия с продолжением')
axes[1].plot(np.arange(convolution.size), convolution, '--', label='Полная свёртка с короткой h[n]')
axes[1].set(title='Полная свёртка и затухание выхода', ylabel='y[n]')
axes[1].legend()
axes[2].plot(np.arange(N2), convolution[:N2], 'o', markersize=3, label='Свёртка')
axes[2].plot(np.arange(N2), recursive_output, '--', label='Разностное уравнение')
axes[2].axvline(N1, color='gray', linestyle=':')
axes[2].set(title='Сравнение на длине входной записи', ylabel='y[n]')
axes[2].legend()
axes[3].stem(np.arange(N2), method_error)
axes[3].set(title='Расхождение двух способов', ylabel='Абсолютная ошибка')
for axis in axes:
    axis.set_xlabel('n')
finish_figure(figure, axes)
```

Вход содержит 20 единиц и 19 нулей. Число единиц получено округлением $39/2=19{,}5$ до 20. Полная свёртка массивов имеет длину $39+29-1=67$. Ненулевая часть занимает 48 отсчётов, последние 19 значений равны нулю из-за нулевого конца входа.

Рекурсия на основном интервале даёт 39 значений. На графике продолжения выход рассчитан до $n=82$. Он остаётся ненулевым после выключения входа из-за обратной связи. Его колебания затухают, но конечный момент завершения не возникает.

Разностное уравнение имеет вид

При включении и выключении импульса видны переходные колебания. Они связаны с теми же полюсами, что и колебания импульсной характеристики.

#### 4. АЧХ и ФЧХ


```python
normalized_angles = np.linspace(-np.pi, np.pi, 2005)
delay = np.exp(-1j * normalized_angles)
transfer = (b0 + b1 * delay + b2 * delay ** 2) / (a0 + a1 * delay + a2 * delay ** 2)
magnitude = np.abs(transfer)
phase_response = np.where(magnitude > 1e-10, np.angle(transfer), np.nan)

peak_index = len(magnitude) // 2 + magnitude[len(magnitude) // 2:].argmax()
show_table(['Показатель', 'Значение'], [
    ['АЧХ на нулевой частоте', f'{magnitude[len(magnitude) // 2]:.6f}'],
    ['АЧХ на частоте Найквиста', f'{magnitude[-1]:.6f}'],
    ['Минимум АЧХ на сетке', f'{magnitude.min():.6f}'],
    ['Максимум АЧХ на сетке', f'{magnitude[peak_index]:.6f}'],
    ['Частота максимума в положительной полуполосе, Гц', f'{normalized_angles[peak_index] * fd / (2 * np.pi):.3f}'],
])
```

```python
frequency_axes = {
    r'$\hat\omega/\pi$': normalized_angles / np.pi,
    'f, Гц': normalized_angles * fd / (2 * np.pi),
}
figure, axes = plt.subplots(2, 2, figsize=(12, 6.5))
for column, (label, axis_frequencies) in enumerate(frequency_axes.items()):
    axes[0, column].plot(axis_frequencies, magnitude)
    axes[1, column].plot(axis_frequencies, phase_response)
    axes[0, column].set(title='Амплитудная характеристика', xlabel=label, ylabel='|H|')
    axes[1, column].set(title='Фазовая характеристика', xlabel=label, ylabel='Фаза, рад')
finish_figure(figure, axes)
```

На АЧХ виден подъём рядом с аргументами полюсов $\pm0{,}29442\pi$. В этой области знаменатель частотной характеристики уменьшается. Точное положение максимума зависит также от числителя.

В области аргументов нулей усиление спадает. При этом точных нулей АЧХ нет, поскольку нули передаточной функции лежат вне единичной окружности. Поэтому скачков фазы на $\pi$ из-за обнуления амплитуды здесь нет.

На ФЧХ возможен переход между значениями около $\pi$ и $-\pi$. Такой скачок на $2\pi$ связан с границей главного аргумента. Он не означает обнуление усиления.

Обе частотные шкалы показывают одинаковую форму кривых. Между ними действует связь
$$f=\frac{\hat\omega f_d}{2\pi},\qquad \hat f=\frac{f}{f_d}.$$
Основная полоса $-\pi\le\hat\omega<\pi$ при $f_d=9000$ Гц соответствует $-4500\le f<4500$ Гц. На нормированной оси используется $\hat\omega/\pi$. Значения на добавленной правой границе совпадают с левой границей из-за периодичности.

Высокие частоты проходят слабее низких. Это видно по спаду АЧХ к частоте Найквиста и по значениям усиления в таблице.

### Часть 2. ДПФ
#### Параметры


```python
def harmonic_signal(time):
    return (A1 * np.cos(2 * np.pi * f1 * time + initial_phases[0])
            + A2 * np.cos(2 * np.pi * f2 * time + initial_phases[1]))

show_table(['Параметр записи', 'Значение'], [
    ['Число отсчётов', N], ['Частота дискретизации, Гц', fd2],
    ['Период дискретизации, с', f'{T:.6f}'],
])
show_table(['Гармоника', 'Амплитуда', 'Частота, Гц', 'Начальная фаза / π'], [
    [1, A1, f1, '0.250'], [2, A2, f2, '0.125']
])
```

#### 5. Период, ДПФ и ОДПФ


```python
sample_index = np.arange(N)
sample_times = sample_index * T
samples = harmonic_signal(sample_times)
spectrum = np.fft.fft(samples)
restored_samples = np.fft.ifft(spectrum)
```

```python
figure, axes = plt.subplots(2, 1, figsize=(10, 5.8))
axes[0].stem(sample_index, samples, label='Исходные отсчёты')
axes[0].plot(sample_index, restored_samples.real, '--', label='После ОДПФ')
axes[0].set(title='Исходные и восстановленные отсчёты', xlabel='n', ylabel='x[n]')
axes[0].legend()
axes[1].stem(sample_times, samples)
axes[1].set(title='Та же запись в секундах', xlabel='t, с', ylabel='x[n]')
finish_figure(figure, axes)
```

Кривые до и после ОДПФ совпадают. Прямое и обратное преобразования сохраняют исходные отсчёты с вычислительной погрешностью.

В записи видны четыре одинаковых участка. Наименьший период суммы равен 16 отсчётам, а длина записи равна 64. Периоды отдельных гармоник равны 8 и 16 отсчётам.

Временные шкалы связаны равенством $t=nT$. Поэтому графики в отсчётах и в секундах показывают одни и те же значения.

#### 6. Амплитудный и фазовый спектры


```python
centered_spectrum = np.fft.fftshift(spectrum)
spectral_indices = np.arange(-N // 2, N // 2)
normalized_frequencies = spectral_indices / N
line_magnitudes = np.abs(centered_spectrum)
line_weights = line_magnitudes / N
significant = line_weights > 1e-10
line_phases = np.angle(centered_spectrum[significant])

active_bins = np.flatnonzero(np.abs(spectrum) > 1e-10)
signed_indices = np.where(active_bins > N // 2, active_bins - N, active_bins)
show_table(['Индекс ДПФ', 'Частота, Гц', 'Модуль ДПФ', 'Вес линии', 'Фаза / π'], [
    [int(index), f'{signed * fd2 / N:.2f}', f'{abs(spectrum[index]):.2f}',
     f'{abs(spectrum[index]) / N:.3f}', f'{np.angle(spectrum[index]) / np.pi:.3f}']
    for index, signed in zip(active_bins, signed_indices)
])
```

```python
figure, axes = plt.subplots(2, 2, figsize=(12, 6.5))
for column, (label, positions) in enumerate([
    ('f/fд', normalized_frequencies), ('f, Гц', normalized_frequencies * fd2),
]):
    axes[0, column].stem(positions, line_weights)
    axes[1, column].stem(positions[significant], line_phases)
    axes[0, column].set(title='Амплитудный спектр', xlabel=label, ylabel='Вес линии')
    axes[1, column].set(title='Фазы значимых линий', xlabel=label, ylabel='Фаза, рад')
finish_figure(figure, axes)
```

```python
figure, axes = plt.subplots(2, 1, figsize=(10, 5), sharex=True)
for axis, values, caption in zip(axes, [line_magnitudes, line_weights], ['|X[k]|', '|X[k]| / N']):
    axis.stem(normalized_frequencies, values)
    axis.set_ylabel(caption)
axes[0].set_title('Модуль ДПФ и амплитуды периодического ряда')
axes[-1].set_xlabel('f/fд')
finish_figure(figure, axes)
```

В спектре видны четыре значимые линии. Каждая косинусоида даёт пару на противоположных частотах. Модули внутри пары равны, а фазы имеют разные знаки. Это свойство вещественного сигнала.

Остальные линии близки к нулю. Частоты гармоник точно попадают в узлы ДПФ с индексами 8 и 12, поэтому спектральной утечки нет. Фазы численно нулевых линий не показаны, поскольку они не имеют определённого значения.

Два набора частотных осей связаны формулами
$$f_k=\frac{kf_d}{N},\qquad \hat f_k=\frac{k}{N},\qquad \hat\omega_k=\frac{2\pi k}{N}.$$
После центрирования отрицательные частоты расположены слева. В исходном порядке индексы выше 32 соответствуют частотам $(k-N)f_d/N$. Шаг сетки равен $156{,}25$ Гц.

На сравнительных графиках линии стоят в тех же точках, но отличаются по высоте в 64 раза. Для конечной записи ДПФ даёт отсчёты преобразования Фурье на сетке. Для периодического повторения коэффициенты ряда равны $C_k=X[k]/N$. Поэтому двусторонние амплитуды получаются делением модуля ДПФ на длину записи.

В одностороннем спектре положительные линии нужно удвоить. Получаются амплитуды исходных косинусоид $1{,}09$ и $2{,}18$. Частоты, модули и фазы четырёх линий приведены в таблице.

#### 7. Восстановление аналогового сигнала


```python
def restore_continuous(time, coefficients, indices, sampling_period):
    # Сумма комплексных гармоник с весами из ДПФ
    result = np.zeros(len(time), dtype=complex)
    period = len(coefficients) * sampling_period
    for k in range(len(coefficients)):
        result += coefficients[k] * np.exp(2j * np.pi * indices[k] * time / period)
    return result / len(coefficients)

dense_time = np.arange(4 * (N - 1) + 1) * (T / 4)
reconstructed = restore_continuous(dense_time, centered_spectrum, spectral_indices, T)
assert np.max(np.abs(reconstructed.imag)) < 1e-10
original_curve = harmonic_signal(dense_time)
sample_error = np.abs(reconstructed[::4].real - samples)
```

```python
figure, axes = plt.subplots(2, 3, figsize=(15, 7))
axes[0, 0].stem(sample_index, samples)
axes[0, 0].set(title='Исходная запись', xlabel='n', ylabel='x[n]')
axes[0, 1].stem(normalized_frequencies, line_magnitudes)
axes[0, 1].set(title='Модуль ДПФ', xlabel='f/fд', ylabel='|X[k]|')
axes[0, 2].plot(dense_time, original_curve)
axes[0, 2].set(title='Исходный непрерывный сигнал', xlabel='t, с', ylabel='x(t)')
axes[1, 0].plot(dense_time, reconstructed.real, label='Восстановление')
axes[1, 0].plot(dense_time, original_curve, '--', label='Исходный сигнал')
axes[1, 0].scatter(sample_times, samples, s=9)
axes[1, 0].set(title='Сравнение кривых', xlabel='t, с', ylabel='x(t), z(t)')
axes[1, 0].legend()
axes[1, 1].stem(normalized_frequencies * fd2, line_weights)
axes[1, 1].set(title='Спектр восстановленного сигнала', xlabel='f, Гц', ylabel='Вес линии')
axes[1, 2].stem(sample_index, sample_error)
axes[1, 2].set(title='Ошибка в моменты отсчётов', xlabel='n', ylabel='Абсолютная ошибка')
finish_figure(figure, axes)
```

Исходная и восстановленная кривые совпадают. Разность в моменты отсчётов остаётся на уровне вычислительной погрешности. Восстановление выполнено суммой
$$z(t)=\frac1{64}\sum_{k=-32}^{31}X_a[k]\exp\left(j\frac{2\pi k}{64T}t\right).$$
Для отрицательных индексов берётся $X_a[k]=X[N+k]$. Временная сетка охватывает интервал от 0 до $63T=0{,}0063$ с с шагом $T/4=0{,}000025$ с.

В спектре восстановленного сигнала остаются те же четыре частоты. Веса линий равны $|X[k]|/64$, а каждая сопряжённая пара задаёт одну косинусоиду с полной амплитудой.

Совпадение сохраняется и между отсчётами. Обе частоты ниже Найквиста 5000 Гц и точно попадают на сетку ДПФ, поэтому восстановление сохраняет их амплитуды и начальные фазы.

#### 8. Равенство Парсеваля


```python
energy_time = np.sum(samples ** 2)
energy_spectrum = np.sum(np.abs(spectrum) ** 2) / N
energy_amplitudes = N * (A1 ** 2 + A2 ** 2) / 2

show_table(['Способ расчёта', 'Энергия'], [
    ['По временным отсчётам', f'{energy_time:.6f}'],
    ['По ДПФ с нормировкой 1/N', f'{energy_spectrum:.6f}'],
    ['Через амплитуды гармоник', f'{energy_amplitudes:.6f}'],
])
show_table(['Показатель', 'Значение'], [
    ['Расхождение двух расчётов энергии', f'{abs(energy_time - energy_spectrum):.3e}'],
    ['Ошибка ОДПФ', f'{np.max(np.abs(restored_samples - samples)):.3e}'],
    ['Ошибка восстановления в отсчётах', f'{sample_error.max():.3e}'],
    ['Ошибка на всей сетке восстановления', f'{np.max(np.abs(reconstructed.real - original_curve)):.3e}'],
])
```

Три значения энергии в таблице совпадают. Это подтверждает равенство Парсеваля и нормировку спектральной энергии на $N$.

Проверка по амплитудам даёт
$$E=\frac{N}{2}(A_1^2+A_2^2)=\frac{64}{2}(1{,}09^2+2{,}18^2)=190{,}096.$$
Полученное значение относится к записи из четырёх наименьших периодов. Средняя мощность равна $E/N=2{,}97025$.
