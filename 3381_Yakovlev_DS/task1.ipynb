### Линейные дискретные системы

#### Линейная дискретная система (ЛДС) – это система, обладающая свойствами аддитивности и однородности. Так как система дискретна, рассматриваются дискретные сигналы.

####Система обработки сигналов (система)


– это любой процесс, дающий выходной сигнал (реакцию) в ответ на входной сигнал (воздействие).

####Сигнал


 – это физический процесс, параметры которого изменяются в соответствии с передаваемой информацией. Они бывают аналоговыми (непрерывными по времени и состоянию), дискретными (дискретными по времени и непрерывными по состоянию) и цифровыми (дискретными по времени и квантованными по состоянию).

####Свойства систем:


- Однородность – изменения в амплитуде входного сигнала приводят к соответствующим изменениям в амплитуде выходного сигнала.
- Аддитивность – если в системе входной сигнал $x_1(nT)$ даёт на выходе $y_1(nT)$, а сигнал $x_2(nT)$ даёт на выходе $y_2(nT)$, то входной сигнал $x_1(nT) + x_2(nT)$ даёт на выходе $y_1(nT)+ y_2(nT)$.
- Стационарность (инвариантность) – сдвиг входного сигнала не даёт ничего больше, чем сдвиг выходного сигнала ($x(nT) \to y(nT) \Rightarrow x((n+s)T) \to y((n+s)T)$).

### Типовые дискретные сигналы

u_0(n) = \left\{

#### Дискретный гармонический сигнал

$x(nT) = x(n) = Acos(2 \pi fnT) = Acos(\omega nT)$

$T$ - период дискретизации (обратен частоте дискретизации - частоте взятих отсчётов непрерывного по времени сигнала при его дискретизации,)

$A$ - амплитуда

$j$ - мнимая единица

### Импульсная характеристика

Импульсная характеристика (ИХ) - это реакция на цифровой единичный импульс ($ u_0(nT) $) при нулевых начальных условиях.

$ y(n) = x(n) * h(n) = \sum_{m=0}^{\infty} h(n-m)x(m) = \sum_{m=0}^{\infty} h(m)x(n-m) $

##### Переходная характеристика

Переходная характеристика (ПХ) - это реакция на цифровой единичный скачок ($ u_1(nT) $) при нулевых начальных условиях, $ g(nT) $.

$ g(t) = \int_{0}^{t} h(\tau ) \ d\tau  \Rightarrow  g(nT) = \sum_{m=0}^{n} h(mT) \$

### Передаточная функция

Передаточная функция (ПФ) - это $\mathcal{Z}$-изображение импульсной характеристики ($h(nT)$).

$ H(z) = \mathcal{Z} \{ h(n) \} = \sum_{n=0}^{\infty} h(n) z^{-n} $

$ y(n) = x(n) * h(n) = \sum_{m=0}^{\infty} h(n-m)x(m) $

$ \mathcal{Z} \{ y(n) \} = \mathcal{Z} \{ x(n) * h(n) \} = X(z) \cdot H(z) = Y(z)$

Т. е. ПФ - это отношение $\mathcal{Z}$-изображения реакции ($Y(z)$) к $\mathcal{Z}$-изображению воздействия ($X(z)$) при ННУ (нулевых начальных условиях):

$ H(z) = \frac{Y(z)}{X(z)} $

#### $\mathcal{Z}$-изображение

$\mathcal{Z}$-изображение - это функция, к которой было применено прямое $\mathcal{Z}$-преобразование, это ряд $ X(z) = \mathcal{Z} \{ x(nT) \} = \sum_{n=0}^{\infty} x(nT) z^{-n} $, т. е. $X(z)$ - это $\mathcal{Z}$-изображение функции $x(nT)$.

#### Обратное $\mathcal{Z}$-преобразование

$ \mathcal{Z}^{-1} \{ X(z) \} = \frac{1}{2 \pi j} \oint X(z) z^{n-1} \ dz$

Обратное $\mathcal{Z}$-преобразование вычисляется с помощью:

##### Таблиц соответствий

##### Теоремы Коши о вычетах

$ x(n) = \sum_{k=1}^{K} {Res}_{\alpha_k} ( X(z) z^{n-1} ) $

$ \alpha_k $  - простой корень (полюс)

$ {Res}_{\alpha_k} ( X(z) z^{n-1} ) =  lim_{z \to \alpha_k} (z - \alpha_k) X(z) z^{n-1} $

Если $ \alpha_k $ кратный ($ l _k $ - степень кратности):

$ {Res}_{\alpha_k} ( X(z) z^{n-1} ) = \frac{1}{(l_k -1)!} lim_{z \to \alpha_k} \frac{d^{l_k - 1}}{dz^{l_k - 1}} ( (z - \alpha_k)^{l_k} X(z) z^{n-1} ) $

##### Разложения на простейшие дроби

Если $X(z)$ - дробно-рациональная функция, то её можно представить как $ X(z) = \sum_{k=1}^{M-1} \frac{A_k}{1-\alpha_k z^{-1}} $

$ x(n) = \mathcal{Z}^{-1} \{ \sum_{k=1}^{M-1} \frac{A_k}{1-\alpha_k z^{-1}} \} = \sum_{k=1}^{M-1} A_k \alpha_k^n$

#### Особые точки ПФ

##### Нули

Нули - это значения $ z = z_0 $, при которых ПФ равна нулю.

##### Полюса

Полюса - это значения $z = z_*$, при которых знаменатель ПФ равен нулю

#### Разностное уравнение

$ y(n) = \sum_{i=0}^{N-1} b_i x(n-i) - \sum_{k=1}^{M-1} a_k y(n-k) $

$ H(z) = \frac{\sum_{i=0}^{N-1} b_i z^{-i}}{1 + \sum_{k=1}^{M-1} a_k z^{-k}} $

Числитель ПФ связан с отсчётами воздействия ($ b_i x(n-i) \Leftrightarrow b_i z^{-i}$)

Знаменатель ПФ связан с отсчётами реакции ($ a_k x(n-k) \Leftrightarrow -a_k z^{-k}$)

### Преобразование Фурье

Прямое:

$ X(j\omega) = \int_{-\infty}^{\infty} x(t) e^{-j\omega t} \ dt $

Обратное:

$ x(t) = \frac{1}{2\pi} \int_{-\infty}^{\infty} X(j\omega) e^{j\omega t} \ dt $

#### Спектры

##### Комплексный спектр

$ X(e^{j\omega T}) =  \left| X(e^{j\omega T}) \right|  \cdot exp(j \cdot arg \{ X(e^{j\omega T}) \}) $



##### Амплитудный спектр

$ \left| X(e^{j\omega T}) \right|  = \sqrt{Re^2 + Im^2}$

##### Фазовый спектр

$ arg \{ X(e^{j\omega T}) \}) = arctg(\frac{Im}{Re})$

##### Равенство Парсеваля

Энергия в частотной области = энергия во временной области

$ \sum_{m=0}^{\infty} \left| x(nT) \right|^2 = \frac{T}{2\pi} \int_{\frac{-\pi}{T}}^{\frac{\pi}{T}} \left| X(e^{j \omega T}) \right| ^ 2 \ d\omega $

#### Дискретное преобразование Фурье

##### Для периодических последовательностей

Прямое:

$ X_p(k) = \sum_{n=0}^{N-1} x_p(n) e^{-j\frac{2\pi}{N} nT}$

Обратное:

$ x_p(n) = \frac{1}{N} \sum_{k=0}^{N-1} X_p(k) e^{j\frac{2\pi}{N} nT}$

$k,n = 0, 1, ..., N-1$

##### Для конечных последовательностей

Прямое:

$ X(k) = \sum_{n=0}^{N-1} x(n) W_{N}^{nk}$

Обратное:

$ x(n) = \frac{1}{N} \sum_{k=0}^{N-1} X(k) W_{N}^{-nk}$

$W_{N}^{nk} = exp(-j\frac{2\pi}{N}nk)$

$k,n = 0, 1, ..., N-1$

##### Частотные характеристики

$ H(e^{j\hat\omega}) = \left| H(e^{j\hat\omega}) \right| \cdot exp(j \cdot arg\{ H(e^{j\hat\omega}) \} ) $

Амплитудно-частотная характеристика - это $ A(\hat\omega) = \left| H(e^{j\hat\omega}) \right|$

Фазочастотная характеристика - это $ \phi(\hat\omega) = arg\{ H(e^{j\hat\omega}) \} $

### Часть 1

#### 1. Импульсная характеристика

Импульсная характеристика ($h(n)$) — это реакция системы на цифровой единичный импульс при нулевых начальных условиях.

$$h(n) = \mathcal{Z}^{-1}\{H(z)\}$$

Передаточная функция рекурсивного звена 2-го порядка:

$$H(z) = \frac{b_0 + b_1 z^{-1} + b_2 z^{-2}}{1 + a_1 z^{-1} + a_2 z^{-2}} = \frac{0.92 + 1.24752\, z^{-1} + 0.92\, z^{-2}}{1 - 1.3028\, z^{-1} + 0.766\, z^{-2}}$$

При $x(n) = u_0(n)$ рекуррентная формула для ИХ принимает вид:

$$
h(n) =
\begin{cases}
b_0, & n = 0,\\
b_n - a_1 h(n-1) - a_2 h(n-2), & n \geq 1,
\end{cases}
$$

где $b_n = b_i$ при $n = i \in \{0,1,2\}$ и $b_n = 0$ при $n > 2$. По этой формуле ИХ рассчитывается рекурсивно на $N_1 = 21$ отсчётов.

**Аналитическая формула** для звена 2-го порядка с комплексно-сопряжёнными полюсами $p_{1,2} = r e^{\pm j\varphi}$:

$$
h(n) =
\begin{cases}
b_0, & n = 0,\\
2|A_1|\, r^{\,n-1} \cos\!\big((n-1)\varphi + \arg A_1\big), & n \geq 1,
\end{cases}
$$

где $A_1 = \dfrac{c_1 p_1 + c_2}{p_1 - p_2}$, $c_1 = b_1 - a_1 b_0$, $c_2 = b_2 - a_2 b_0$, а полюса находятся из $z^2 + a_1 z + a_2 = 0$.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

# Параметры варианта
N_var = 21
N1 = N_var % 10 + 20                                   # 21

b0 = 0.5 + 0.02 * N_var                                # 0.92
b1 = b0 * (-1)**(N_var + 1) * (0.9822 + 0.0178*N_var)  # 1.24752
b2 = b0 * (0.8 + 0.2 * (N_var % 5))                    # 0.92
a0 = 1
a1 = (-1)**N_var * (0.7778 + 0.025*N_var)              # -1.3028
a2 = 0.64 + 0.006 * N_var                              # 0.766

print(f"N1 = {N1}")
print(f"b = [{b0}, {b1}, {b2}]")
print(f"a = [{a0}, {a1}, {a2}]")

# ИХ через рекуррентную формулу (lfilter)
impulse = np.zeros(N1); impulse[0] = 1.0
h = signal.lfilter([b0, b1, b2], [a0, a1, a2], impulse)

print("\nИХ h(n), рассчитанная рекурсивно:")
for k, v in enumerate(h):
    print(f"  h({k:2d}) = {v:+.6f}")

# Аналитическая формула через полюса
p1 = (-a1 + np.sqrt(a1**2 - 4*a2 + 0j)) / 2
p2 = (-a1 - np.sqrt(a1**2 - 4*a2 + 0j)) / 2
r   = np.abs(p1)
phi = np.angle(p1)

c1 = b1 - a1*b0
c2 = b2 - a2*b0
A1 = (c1*p1 + c2) / (p1 - p2)

h_an = np.zeros(N1)
h_an[0] = b0
for k in range(1, N1):
    h_an[k] = 2*np.abs(A1) * r**(k-1) * np.cos((k-1)*phi + np.angle(A1))

print(f"\nПолюса:  p1 = {p1:.6f},  p2 = {p2:.6f}")
print(f"r = |p1| = {r:.6f},  phi = arg(p1) = {phi:.6f} рад")
print(f"A1 = {A1:.6f}")

print("\nСравнение рекуррентной и аналитической формул:")
print(f"{'n':>3} | {'h(n) числ.':>13} | {'h(n) анал.':>13} | {'|Δ|':>10}")
for k in range(N1):
    print(f"{k:>3} | {h[k]:>13.6f} | {h_an[k]:>13.6f} | {abs(h[k]-h_an[k]):>10.2e}")

# График
n = np.arange(N1)
plt.figure(figsize=(11, 4))
plt.stem(n, h, basefmt="k-", linefmt="b-", markerfmt="bo")
plt.xlabel("n"); plt.ylabel("h(n)")
plt.title(f"Импульсная характеристика (N\u2081 = {N1})")
plt.grid(True, alpha=0.4)
plt.tight_layout()
plt.show()
```

**Анализ результатов.**

Полюса передаточной функции оказались комплексно-сопряжёнными:

$$p_{1,2} = 0.6514 \pm j\,0.5845, \qquad r = |p_1| = 0.8752, \qquad \varphi = \arg p_1 = 0.7313\ \text{рад} \approx 41.9^\circ.$$

Так как $r = 0.8752 < 1$, оба полюса лежат внутри единичной окружности. Это означает, что система устойчива, а её импульсная характеристика с ростом $n$ экспоненциально затухает с огибающей $\sim r^{\,n}$. Наличие ненулевого аргумента $\varphi \neq 0$ приводит к **колебательному** характеру ИХ: период колебаний в отсчётах

$$T_h = \frac{2\pi}{\varphi} = \frac{2\pi}{0.7313} \approx 8.59\ \text{отсчётов},$$

что хорошо видно на графике — локальные максимумы располагаются около $n \approx 2;\ 11;\ 20$ (расстояние $\sim 9$ отсчётов).

С учётом $|A_1| = 1.9721$ и $\arg A_1 = -0.9016$ рад аналитическая формула принимает вид:

$$
h(n) =
\begin{cases}
0.92, & n = 0,\\[4pt]
3.9442 \cdot 0.8752^{\,n-1} \cos\!\big(0.7313\,(n-1) - 0.9016\big), & n \geq 1.
\end{cases}
$$

Начальные отсчёты ИХ: $h(0) = 0.92$, $h(1) \approx 2.446$, $h(2) \approx 3.402$ — амплитуда нарастает за счёт суперпозиции косинуса и экспоненты огибающей, а затем плавно убывает. К $n = 21$ значение падает до $h(20) \approx 0.285$.

Сравнение рекуррентного и аналитического расчётов показывает, что расхождение не превышает $|\Delta| \sim 10^{-16}$, то есть находится на уровне **машинной погрешности** для чисел двойной точности. Это подтверждает корректность обеих формул и справедливость аналитического представления ИХ через полюса.

**Чему в действительности равна длина ИХ.** Рекурсивное звено имеет импульсную характеристику **бесконечной длины**. Величина $N_1 = 21$ — это лишь заданное число отсчётов, отображаемых на графике, а не «физическая» длина ИХ. Поскольку $r < 1$, ИХ затухает, но **не обращается в точный ноль** ни при каком конечном $n$; практически её можно считать пренебрежимо малой, когда огибающая $r^{\,n}$ опускается ниже уровня вычислительной точности.

#### 2. Реакция через свёртку

Реакция системы на произвольное воздействие вычисляется по формуле свёртки:

$$y(n) = x(n) * h(n) = \sum_{m=0}^{\infty} h(m)\,x(n-m)$$

В качестве воздействия $x(n)$ задан дискретный прямоугольный импульс длины $N_2 = 31$:

$$
x(n) =
\begin{cases}
1, & 0 \leqslant n < int(N_2/2),\\
0, & int(N_2/2) \leqslant n \leqslant N_2-1.
\end{cases}
$$

Для вычисления реакции $y_1(n)$ используется свёртка воздействия $x(n)$ с импульсной характеристикой $h(n)$, полученной в предыдущем пункте. Поскольку $h(n)$ рассчитана на $N_1 = 21$ отсчётов, результат свёртки будет содержать $N_1 + N_2 - 1 = 51$ отсчёт.

```python
import numpy as np
import matplotlib.pyplot as plt

# Параметры
N2 = 31
N_pulse = int(N2 / 2)

# Воздействие
x = np.zeros(N2)
x[:N_pulse] = 1.0

# Реакция через свёртку (h берется из п. 1)
y1 = np.convolve(x, h)

# Графики
plt.figure(figsize=(11, 6))

plt.subplot(2, 1, 1)
plt.stem(range(N2), x, basefmt="k-")
plt.xlabel("n"); plt.ylabel("x(n)")
plt.title(f"Воздействие — прямоугольный импульс (N\u2082 = {N2})")
plt.grid(alpha=0.4)

plt.subplot(2, 1, 2)
plt.stem(range(len(y1)), y1, basefmt="k-")
plt.xlabel("n"); plt.ylabel("y\u2081(n)")
plt.title("Реакция, вычисленная через свёртку")
plt.grid(alpha=0.4)

plt.tight_layout()
plt.show()

print(f"Длина воздействия x: {len(x)}")
print(f"Длина импульсной характеристики h: {len(h)}")
print(f"Длина свёртки (аналитически): {len(x) + len(h) - 1}")
print(f"Длина свёртки (по графику): {len(y1)}")
```

**a) Чему равна длина импульса $x$.** Общая длина последовательности $x(n)$ составляет $N_2 = 31$ отсчёт. Однако ненулевая часть (собственно прямоугольный импульс) занимает только первые $int(N_2/2) = 15$ отсчётов (от $n = 0$ до $n = 14$). На интервале от $n = 15$ до $n = 30$ значения сигнала равны нулю. Это хорошо видно на верхнем графике.

**b) Чему равна длина свёртки аналитически и по графику.** Аналитически длина свёртки двух последовательностей конечной длины $N_2$ и $N_1$ равна $N_2 + N_1 - 1$. Поскольку импульсная характеристика $h(n)$ имеет длину $N_1 = 21$, длина свёртки составляет $31 + 21 - 1 = 51$ отсчёт. По графику это подтверждается: массив `y1` содержит 51 элемент (от $n = 0$ до $n = 50$), но после $n \approx 35$ значения становятся пренебрежимо малыми.

**c) Почему её ограничивают до длины воздействия.** Теоретически, поскольку импульсная характеристика $h(n)$ бесконечна, свёртка также имеет бесконечную длину. Однако после окончания воздействия (при $n \geqslant 15$) система переходит в режим свободных затухающих колебаний. Амплитуда этих колебаний быстро уменьшается, и к $n \approx 35$ реакция становится практически нулевой. При анализе систем основной интерес представляет интервал, на котором воздействие активно, поэтому для наглядного представления реакции на графике её часто ограничивают длиной воздействия $N_2 = 31$.

#### 3. Реакция через разностное уравнение

Разностное уравнение связывает текущее значение реакции с текущим и прошлыми значениями воздействия, а также с прошлыми значениями реакции:

$$y(n) = \sum_{i=0}^{N-1} b_i x(n-i) - \sum_{k=1}^{M-1} a_k y(n-k)$$

Для заданного рекурсивного звена 2-го порядка ($a_0 = 1$):

$$y(n) = b_0 x(n) + b_1 x(n-1) + b_2 x(n-2) - a_1 y(n-1) - a_2 y(n-2)$$

При нулевых начальных условиях ($y(n) = 0$ при $n < 0$) реакция вычисляется рекурсивно. Поскольку воздействие ($x(n)$) имеет длину 31, а реакция ($y_1(n)$) содержит 51 отсчёт, дополним воздействие нулями до длины 51 для корректного сравнения обеих реакций.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

# Параметры
N_y = len(y1)

# Воздействие, дополненное нулями до длины реакции
x_padded = np.zeros(N_y)
x_padded[:len(x)] = x

# Реакция по разностному уравнению
y2 = signal.lfilter([b0, b1, b2], [1, a1, a2], x_padded)

# Графики
n_y = np.arange(N_y)

plt.figure(figsize=(11, 6))

plt.subplot(2, 1, 1)
plt.stem(n_y, y2, basefmt="k-")
plt.xlabel("n"); plt.ylabel("y\u2082(n)")
plt.title("Реакция, вычисленная по разностному уравнению")
plt.grid(alpha=0.4)

plt.subplot(2, 1, 2)
plt.stem(n_y, y1, basefmt="k-", linefmt='C0-', markerfmt='C0o', label='y₁ (свёртка)')
plt.stem(n_y, y2, basefmt="k-", linefmt='C1--', markerfmt='C1x', label='y₂ (разностное уравнение)')
plt.xlabel("n"); plt.ylabel("y(n)")
plt.title("Сравнение реакций y\u2081(n) и y\u2082(n)")
plt.legend()
plt.grid(alpha=0.4)

plt.tight_layout()
plt.show()

print(f"Максимальное расхождение между y1 и y2: {np.max(np.abs(y1 - y2)):.2e}")
```

**Разностное уравнение** для заданных коэффициентов принимает вид:

$$y(n) = 0.92 x(n) + 1.24752 x(n-1) + 0.92 x(n-2) + 1.3028 y(n-1) - 0.766 y(n-2)$$

**Сравнение реакций.** На начальном участке (при $n \leqslant 20$) графики реакций $y_1(n)$ и $y_2(n)$ полностью совпадают. Однако при $n > 20$ наблюдается расхождение, максимальное значение которого составляет $0.42$.

Это объясняется тем, что при вычислении свёртки в п. 2 импульсная характеристика $h(n)$ была ограничена длиной $N_1 = 21$. В разностном уравнении используется полная (бесконечная) импульсная характеристика, поэтому $y_2(n)$ точнее описывает реакцию системы на больших временах. Расхождение равно вкладу отброшенных отсчётов $h(n)$ при $n \geqslant 21$.

**Чему равны длины воздействия и реакции.** Длина воздействия $x(n)$ составляет $N_2 = 31$ отсчёт. Длина реакции $y_2(n)$ теоретически бесконечна, так как система является рекурсивной. На практике её вычисляют на конечном интервале, где она затухает до пренебрежимо малых значений. Для наглядного сравнения с $y_1(n)$ реакция $y_2(n)$ была рассчитана на 51 отсчёте.

#### 4. Нули и полюсы передаточной функции

Нули передаточной функции находятся из условия равенства числителя нулю:
$$b_0 z^2 + b_1 z + b_2 = 0$$
Полюсы определяются из условия равенства знаменателя нулю:
$$z^2 + a_1 z + a_2 = 0$$

Найденные корни можно представить в алгебраической форме ($z = \alpha \pm j\beta$) и в показательной форме ($z = r e^{\pm j\varphi}$), где $r = |z|$ — модуль, а $\varphi = \arg z$ — аргумент. Значение аргумента выражается относительно $\pi$.

Передаточная функция может быть представлена в виде произведения простейших множителей через её нули и полюсы:
$$H(z) = \frac{b_0 (z - z_1)(z - z_2)}{(z - p_1)(z - p_2)} = b_0 \frac{(1 - z_1 z^{-1})(1 - z_2 z^{-1})}{(1 - p_1 z^{-1})(1 - p_2 z^{-1})}$$

```python
import numpy as np

# Параметры (используются определенные ранее коэффициенты b0, b1, b2, a1, a2)

# Расчет нулей
zeros = np.roots([b0, b1, b2])
z1, z2 = zeros[0], zeros[1]

# Расчет полюсов
poles = np.roots([1, a1, a2])
p1, p2 = poles[0], poles[1]

# Вывод нулей
print("Нули (z):")
for i, z in enumerate([z1, z2]):
    r = np.abs(z)
    phi = np.angle(z)
    print(f"  z_{i+1} = {z.real:+.4f} {z.imag:+.4f}j")
    print(f"      = {r:.4f} * exp({phi:+.4f}j)")
    print(f"      arg = {phi:+.4f} рад = {phi/np.pi:+.4f}*pi")

# Вывод полюсов
print("\nПолюсы (p):")
for i, p in enumerate([p1, p2]):
    r = np.abs(p)
    phi = np.angle(p)
    print(f"  p_{i+1} = {p.real:+.4f} {p.imag:+.4f}j")
    print(f"      = {r:.4f} * exp({phi:+.4f}j)")
    print(f"      arg = {phi:+.4f} рад = {phi/np.pi:+.4f}*pi")
```

**Нули** передаточной функции находятся из условия равенства числителя нулю. В результате решения получены комплексно-сопряжённые корни в алгебраической форме:
$$z_{1,2} = -0.6780 \pm j0.7351$$

В показательной форме они имеют единичный модуль ($r_z = 1.0000$) и аргумент $\varphi_z = 2.3158$ рад. Относительно $\pi$ значение аргумента составляет $\pm 0.7372\pi$:
$$z_{1,2} = 1.0000 \cdot e^{\pm j 2.3158}$$

**Полюсы** определяются из условия равенства знаменателя нулю. Они также являются комплексно-сопряжёнными и в алгебраической форме имеют вид:
$$p_{1,2} = 0.6514 \pm j0.5845$$

В показательной форме их модуль $r_p = 0.8752$, а аргумент $\varphi_p = 0.7313$ рад, что составляет $\pm 0.2328\pi$:
$$p_{1,2} = 0.8752 \cdot e^{\pm j 0.7313}$$

**Связь между формами** заключается в геометрической интерпретации: алгебраическая форма показывает координаты точки на комплексной плоскости (действительную и мнимую части), а показательная форма отражает её расстояние от начала координат (модуль) и угол поворота (аргумент).

С учётом найденных нулей и полюсов передаточная функция может быть записана в виде произведения простейших множителей:
$$H(z) = 0.92 \frac{(1 - e^{j 2.3158} z^{-1})(1 - e^{-j 2.3158} z^{-1})}{(1 - 0.8752 e^{j 0.7313} z^{-1})(1 - 0.8752 e^{-j 0.7313} z^{-1})}$$

#### 5. Карта нулей и полюсов

Карта нулей и полюсов — это графическое представление расположения нулей и полюсов передаточной функции на комплексной плоскости. Нули обозначаются кружками, а полюсы — крестиками. Для наглядности на график наносится единичная окружность ($|z| = 1$).

Для оценки устойчивости рекурсивного звена необходимо проанализировать положение полюсов:
*   если все полюсы лежат **внутри** единичной окружности ($|p_i| < 1$), система устойчива;
*   если хотя бы один полюс лежит **на** единичной окружности, система находится на границе устойчивости;
*   если хотя бы один полюс лежит **вне** единичной окружности ($|p_i| > 1$), система неустойчива.

```python
import numpy as np
import matplotlib.pyplot as plt

# Параметры
# zeros и poles уже рассчитаны в п. 4

# Построение единичной окружности
theta = np.linspace(0, 2 * np.pi, 400)
x_circle = np.cos(theta)
y_circle = np.sin(theta)

plt.figure(figsize=(6, 6))

# Оси координат и единичная окружность
plt.plot(x_circle, y_circle, 'k--', label='Единичная окружность')
plt.axhline(0, color='gray', linewidth=0.8)
plt.axvline(0, color='gray', linewidth=0.8)

# Нули и полюсы
plt.plot(zeros.real, zeros.imag, 'bo', markersize=8, label='Нули (z)')
plt.plot(poles.real, poles.imag, 'rx', markersize=8, label='Полюсы (p)')

plt.xlabel('Re(z)')
plt.ylabel('Im(z)')
plt.title('Карта нулей и полюсов')
plt.grid(True, alpha=0.4)
plt.legend()
plt.axis('equal')
plt.tight_layout()
plt.show()
```

**Является ли рекурсивное звено устойчивым.** Да, звено устойчиво. На карте нулей и полюсов видно, что оба полюса $p_{1,2} = 0.6514 \pm j0.5845$ лежат строго внутри единичной окружности. Их модули равны $r_p = 0.8752 < 1$. Поскольку все полюсы находятся внутри круга радиуса 1, система является устойчивой.

**Совпадают ли значения с вычисленными в п. 4.** Да, совпадают. На графике чётко видны две пары комплексно-сопряжённых точек:
*   Нули (синие кружки) расположены на единичной окружности в левой полуплоскости: $z_{1,2} = -0.6780 \pm j0.7351$. Это подтверждает, что их модуль равен 1.
*   Полюсы (красные крестики) расположены внутри единичной окружности в правой полуплоскости: $p_{1,2} = 0.6514 \pm j0.5845$.

Координаты этих точек на комплексной плоскости полностью соответствуют значениям, полученным аналитически в предыдущем пункте.

#### 6. АЧХ и ФЧХ в нормированных частотах

Нормированная частота определяется как $\hat{\omega} = \omega T = 2\pi f / f_{\text{д}}$. Основная полоса частот для дискретных систем лежит в пределах от $-\pi$ до $\pi$.

Комплексная частотная характеристика вычисляется путем подстановки $z = e^{j\hat{\omega}}$ в передаточную функцию:
$$H(e^{j\hat{\omega}}) = \frac{b_0 + b_1 e^{-j\hat{\omega}} + b_2 e^{-j2\hat{\omega}}}{1 + a_1 e^{-j\hat{\omega}} + a_2 e^{-j2\hat{\omega}}}$$

Амплитудно-частотная характеристика (АЧХ) — это модуль комплексной частотной характеристики:
$$A(\hat{\omega}) = \left| H(e^{j\hat{\omega}}) \right|$$

Фазочастотная характеристика (ФЧХ) — это аргумент комплексной частотной характеристики:
$$\phi(\hat{\omega}) = \arg\left\{ H(e^{j\hat{\omega}}) \right\}$$

```python
import numpy as np
import matplotlib.pyplot as plt

# Параметры (b0, b1, b2, a1, a2 предполагаются заданными ранее)

# Диапазон основной полосы частот [-pi, pi]
w_narrow = np.linspace(-np.pi, np.pi, 1024)
z_narrow = np.exp(1j * w_narrow)
H_narrow = (b0 + b1 * z_narrow**-1 + b2 * z_narrow**-2) / (1 + a1 * z_narrow**-1 + a2 * z_narrow**-2)

A_narrow = np.abs(H_narrow)
phi_narrow = np.unwrap(np.angle(H_narrow))

# Расширенный диапазон частот [-3pi, 3pi]
w_wide = np.linspace(-3*np.pi, 3*np.pi, 2048)
z_wide = np.exp(1j * w_wide)
H_wide = (b0 + b1 * z_wide**-1 + b2 * z_wide**-2) / (1 + a1 * z_wide**-1 + a2 * z_wide**-2)

A_wide = np.abs(H_wide)
phi_wide = np.unwrap(np.angle(H_wide))

# Графики
plt.figure(figsize=(12, 8))

# 1. АЧХ (основная полоса)
plt.subplot(2, 2, 1)
plt.plot(w_narrow, A_narrow, color='C0')
plt.title('АЧХ (основная полоса)')
plt.xlabel(r'$\hat{\omega}$, рад')
plt.ylabel(r'$|H(e^{j\hat{\omega}})|$')
plt.grid(alpha=0.4)
plt.xlim([-np.pi, np.pi])

# 2. ФЧХ (основная полоса)
plt.subplot(2, 2, 2)
plt.plot(w_narrow, phi_narrow, color='C1')
plt.title('ФЧХ (основная полоса)')
plt.xlabel(r'$\hat{\omega}$, рад')
plt.ylabel(r'$\arg H(e^{j\hat{\omega}})$, рад')
plt.grid(alpha=0.4)
plt.xlim([-np.pi, np.pi])

# 3. АЧХ (расширенная полоса)
plt.subplot(2, 2, 3)
plt.plot(w_wide, A_wide, color='C0')
plt.title(r'АЧХ (расширенная полоса $[-3\pi, 3\pi]$)')
plt.xlabel(r'$\hat{\omega}$, рад')
plt.ylabel(r'$|H(e^{j\hat{\omega}})|$')
plt.grid(alpha=0.4)
plt.xlim([-3*np.pi, 3*np.pi])
plt.xticks([-3*np.pi, -2*np.pi, -np.pi, 0, np.pi, 2*np.pi, 3*np.pi],
           [r'$-3\pi$', r'$-2\pi$', r'$-\pi$', '0', r'$\pi$', r'$2\pi$', r'$3\pi$'])

# 4. ФЧХ (расширенная полоса)
plt.subplot(2, 2, 4)
plt.plot(w_wide, phi_wide, color='C1')
plt.title(r'ФЧХ (расширенная полоса $[-3\pi, 3\pi]$)')
plt.xlabel(r'$\hat{\omega}$, рад')
plt.ylabel(r'$\arg H(e^{j\hat{\omega}})$, рад')
plt.grid(alpha=0.4)
plt.xlim([-3*np.pi, 3*np.pi])
plt.xticks([-3*np.pi, -2*np.pi, -np.pi, 0, np.pi, 2*np.pi, 3*np.pi],
           [r'$-3\pi$', r'$-2\pi$', r'$-\pi$', '0', r'$\pi$', r'$2\pi$', r'$3\pi$'])

plt.tight_layout()
plt.show()
```

```python
# Ключевые значения АЧХ в характерных точках

def H_at(w_hat):
    z = np.exp(1j * w_hat)
    return (b0 + b1 * z**-1 + b2 * z**-2) / (1 + a1 * z**-1 + a2 * z**-2)

# Аргументы полюсов и нулей (координаты уже получены в п. 4)
w_p = np.angle(poles[0])
w_z = np.angle(zeros[0])

print("Характерные частоты:")
print(f"  arg(p1) = {w_p:.4f} рад = {w_p/np.pi:+.4f}·π")
print(f"  arg(z1) = {w_z:.4f} рад = {w_z/np.pi:+.4f}·π")

print("\nЗначения АЧХ в характерных точках:")
print(f"  |H(1)|  = |H(e^(j·0))|          = {np.abs(H_at(0.0)):.4f}")
print(f"  |H(-1)| = |H(e^(j·π))|          = {np.abs(H_at(np.pi)):.4f}")
print(f"  |H(e^(j·arg(p1)))|              = {np.abs(H_at(w_p)):.4f}  (на резонансе)")
print(f"  |H(e^(j·arg(z1)))|              = {np.abs(H_at(w_z)):.2e}  (в нуле)")

# Точное положение максимума АЧХ в основной полосе
idx_max = np.argmax(A_narrow)
print(f"\nМаксимум АЧХ в основной полосе:")
print(f"  ω̂_max = {w_narrow[idx_max]:.4f} рад")
print(f"  A_max = {A_narrow[idx_max]:.4f}")

# Точное положение минимума (провала) в основной полосе
idx_min = np.argmin(A_narrow)
print(f"\nМинимум АЧХ в основной полосе:")
print(f"  ω̂_min = {w_narrow[idx_min]:.4f} рад")
print(f"  A_min = {A_narrow[idx_min]:.2e}")
```

**a) Границы основной полосы частот.** Основная полоса частот дискретной системы лежит в пределах от $-\pi$ до $\pi$ рад (что соответствует интервалу $[-f_\text{д}/2;\ f_\text{д}/2]$ в герцах). За её пределами частотная характеристика **периодически повторяется с периодом $2\pi$** — это наглядно подтверждают графики в расширенной полосе $[-3\pi, 3\pi]$.

**b) Соответствие между картой нулей и полюсов и видом АЧХ.**
* **Полюсы** $p_{1,2} = 0.6514 \pm j0.5845$ имеют модуль $r_p = 0.8752$ и аргумент $\varphi_p \approx \pm 0.7313$ рад. Близость полюсов к единичной окружности даёт **резонансные пики** АЧХ. На обоих графиках (и в основной, и в расширенной полосе) видны максимумы АЧХ со значением $\approx 16.8528$ в точках $\hat{\omega} \approx \pm 0.7155 + 2\pi k$.

$$f = \frac{\hat{\omega}}{2\pi} f_{\text{д}}, \qquad \hat{\omega} = \frac{2\pi f}{f_{\text{д}}}$$

где $f_{\text{д}} = 21000$ Гц — частота дискретизации.

Основная полоса частот дискретной системы в герцах лежит в пределах от $0$ до $f_{\text{д}}/2$ (или симметрично от $-f_{\text{д}}/2$ до $f_{\text{д}}/2$). За её пределами частотная характеристика периодически повторяется с периодом $f_{\text{д}}$.

Комплексная частотная характеристика вычисляется подстановкой $z = e^{j 2\pi f / f_{\text{д}}}$ в передаточную функцию:

$$H(e^{j 2\pi f / f_{\text{д}}}) = \frac{b_0 + b_1 e^{-j 2\pi f / f_{\text{д}}} + b_2 e^{-j 4\pi f / f_{\text{д}}}}{1 + a_1 e^{-j 2\pi f / f_{\text{д}}} + a_2 e^{-j 4\pi f / f_{\text{д}}}}$$

АЧХ и ФЧХ в абсолютных частотах:

$$A(f) = \left| H(e^{j 2\pi f / f_{\text{д}}}) \right|, \qquad \phi(f) = \arg\left\{ H(e^{j 2\pi f / f_{\text{д}}}) \right\}$$

#### 1. График последовательности x(n)

Последовательность задана на интервале $n=0,\ldots,N-1$:

$$
x(nT)=A_1\cos\left(2\pi f_1 nT+\frac{\pi}{4}\right)+A_2\cos\left(2\pi f_2 nT+\frac{\pi}{8}\right).
$$

Используются шкалы дискретного времени $n$ и дискретного нормированного времени $\hat{t}=n/N$, при котором один период $N$ отображается на интервал $[0;1)$.

```python
import numpy as np
import matplotlib.pyplot as plt

# Параметры
N = 64
f_d = 4000.0
T = 1.0 / f_d
A1 = 1.21
A2 = 2.42
f1 = 500.0
f2 = 750.0

# Расчёт
n = np.arange(N)
x = (A1 * np.cos(2 * np.pi * f1 * n * T + np.pi / 4)
     + A2 * np.cos(2 * np.pi * f2 * n * T + np.pi / 8))
t_norm = n / N

print(f"N = {N}, f_d = {f_d:.0f} Гц, T = {T:.6f} с")
print(f"x_min = {x.min():.4f}, x_max = {x.max():.4f}")
print(f"Повторяемость с периодом 16: {np.allclose(x[:16], x[16:32])}")

# Графики
plt.figure(figsize=(11, 6))

plt.subplot(2, 1, 1)
plt.stem(n, x, basefmt="k-")
plt.xlabel("n")
plt.ylabel("x(n)")
plt.title("Последовательность x(n) в шкале дискретного времени")
plt.grid(alpha=0.4)

plt.subplot(2, 1, 2)
plt.stem(t_norm, x, basefmt="k-")
plt.xlabel(r"$\hat{t}=n/N$")
plt.ylabel("x(n)")
plt.title("Последовательность x(n) в шкале дискретного нормированного времени")
plt.grid(alpha=0.4)

plt.tight_layout()
plt.show()
```

На интервале $N=64$ последовательность содержит четыре повторения фактического периода $16$ отсчётов. Это подтверждается проверкой `np.allclose(x[:16], x[16:32])`, которая даёт `True`, и видно по графикам: форма сигнала повторяется через каждые $16$ отсчётов.

- $x_{\max} \approx 3.0914$, максимум достигается при $n=0,16,32,48$;
- $x_{\min} \approx -3.0914$, минимум достигается при $n=2,18,34,50$.

В шкале дискретного нормированного времени $\hat{t}=n/N$ период $16$ отсчётов соответствует интервалу $16/64=0.25$. Поэтому на $\hat{t}\in[0;1)$ укладывается четыре одинаковых участка сигнала.

Сигнал является суммой двух гармоник с частотами $f_1=500$ Гц и $f_2=750$ Гц. При $f_\text{д}=4000$ Гц нормированные частоты равны $f_1/f_\text{д}=1/8$ и $f_2/f_\text{д}=3/16$, поэтому суммарная последовательность периодична с периодом $16$ отсчётов, а на заданном интервале $N=64$ показаны четыре её периода.

#### 2. ОДПФ от ДПФ последовательности

Прямое ДПФ последовательности $x(n)$ вычисляется по формуле:

$$
X(k) = \sum_{n=0}^{N-1} x(n) e^{-j\frac{2\pi}{N}nk}, \qquad k=0,1,\ldots,N-1.
$$

Обратное ДПФ (ОДПФ) восстанавливает исходную последовательность:

$$
x(n) = \frac{1}{N} \sum_{k=0}^{N-1} X(k) e^{j\frac{2\pi}{N}nk}.
$$

Восстановленная последовательность выводится в шкале дискретного нормированного времени $\hat{t}=n/N$.

```python
import numpy as np
import matplotlib.pyplot as plt

# Расчёт
X = np.fft.fft(x)
x_ifft = np.fft.ifft(X).real

# Оценка точности восстановления
max_err = np.max(np.abs(x - x_ifft))

# Графики
plt.figure(figsize=(11, 4))
plt.stem(t_norm, x_ifft, basefmt="k-", linefmt="C1-", markerfmt="C1o")
plt.xlabel(r"$\hat{t}=n/N$")
plt.ylabel("x(n)")
plt.title("Последовательность, восстановленная через ОДПФ")
plt.grid(alpha=0.4)
plt.tight_layout()
plt.show()

print(f"Максимальное отклонение восстановленной последовательности от исходной: {max_err:.2e}")
```

Максимальное отклонение восстановленной последовательности от исходной по результатам расчёта составило 1.33e-15, что соответствует уровню машинной погрешности чисел двойной точности. График восстановленной последовательности полностью совпадает с исходным сигналом из п. 1.

Это подтверждает, что пара преобразований ДПФ и ОДПФ образует взаимно обратное преобразование: ОДПФ по отсчётам ДПФ $X(k)$ точно восстанавливает исходную конечную последовательность $x(n)$ без потери информации.

#### 3. Амплитудный и фазовый спектры

ДПФ последовательности позволяет представить её в частотной области:

$$
X(k) = \sum_{n=0}^{N-1} x(n) e^{-j\frac{2\pi}{N}nk}, \qquad k=0,1,\ldots,N-1.
$$

**Амплитудный спектр** — это модуль ДПФ:

$$
|X(k)| = \sqrt{\operatorname{Re}^2 X(k) + \operatorname{Im}^2 X(k)}.
$$

**Фазовый спектр** — это аргумент ДПФ:

$$
\varphi(k) = \arg X(k) = \operatorname{arctg}\frac{\operatorname{Im} X(k)}{\operatorname{Re} X(k)}.
$$

Для наглядности спектры выводятся симметрично относительно нуля с помощью `np.fft.fftshift`. При этом используются две шкалы частот: дискретная нормированная частота $\hat{f} = k/N$ и абсолютная частота $f = k f_\text{д}/N$ (Гц).

```python
import numpy as np
import matplotlib.pyplot as plt

# Расчёт
X = np.fft.fft(x)
X_shifted = np.fft.fftshift(X)

# Шкалы частот
k = np.arange(N)
k_shifted = k - N // 2
f_norm = k_shifted / N
f_abs = f_norm * f_d

A_spec = np.abs(X_shifted)
phi_spec = np.angle(X_shifted)

# Графики
plt.figure(figsize=(12, 8))

# Амплитудный спектр (нормированная частота)
plt.subplot(2, 2, 1)
plt.stem(f_norm, A_spec, basefmt="k-")
plt.xlabel(r"$\hat{f}=k/N$")
plt.ylabel(r"$|X(k)|$")
plt.title("Амплитудный спектр (нормированная частота)")
plt.grid(alpha=0.4)

# Фазовый спектр (нормированная частота)
plt.subplot(2, 2, 2)
plt.stem(f_norm, phi_spec, basefmt="k-")
plt.xlabel(r"$\hat{f}=k/N$")
plt.ylabel(r"$\arg X(k)$, рад")
plt.title("Фазовый спектр (нормированная частота)")
plt.grid(alpha=0.4)

# Амплитудный спектр (абсолютная частота)
plt.subplot(2, 2, 3)
plt.stem(f_abs, A_spec, basefmt="k-")
plt.xlabel("f, Гц")
plt.ylabel(r"$|X(k)|$")
plt.title("Амплитудный спектр (абсолютная частота)")
plt.grid(alpha=0.4)

# Фазовый спектр (абсолютная частота)
plt.subplot(2, 2, 4)
plt.stem(f_abs, phi_spec, basefmt="k-")
plt.xlabel("f, Гц")
plt.ylabel(r"$\arg X(k)$, рад")
plt.title("Фазовый спектр (абсолютная частота)")
plt.grid(alpha=0.4)

plt.tight_layout()
plt.show()

# Вывод пиковых значений
non_zero = np.where(A_spec > 1e-10)[0]
print("Отличные от нуля компоненты спектра:")
for idx in non_zero:
    print(f"  f = {f_abs[idx]:+.0f} Гц (k={k_shifted[idx]:+d}): |X| = {A_spec[idx]:.2f}, phi = {phi_spec[idx]:+.4f} рад")
```

**a) Связь дискретного нормированного времени с дискретным временем.**
Дискретное время $n$ измеряется в отсчётах. Дискретное нормированное время $\hat{t}$ показывает, какую долю от длины периода $N$ составляет текущий отсчёт:
$$
\hat{t} = \frac{n}{N} = \frac{nT}{NT}.
$$
В пределах одного периода $n$ меняется от $0$ до $N-1$, а $\hat{t}$ — от $0$ до $(N-1)/N$. На графиках спектров аналогичным образом используются дискретная нормированная частота $\hat{f}=k/N$ и абсолютная частота $f$.

**b) Связь частоты $f$ (Гц) с дискретной нормированной частотой.**
Шкала абсолютных частот получается умножением нормированной частоты на частоту дискретизации:
$$
f = \hat{f} f_\text{д} = \frac{k}{N} f_\text{д}.
$$
Шаг по частоте между соседними отсчётами спектра равен $\Delta f = f_\text{д}/N = 4000/64 = 62.5$ Гц.

**c) Вид амплитудного и фазового спектров.**
Поскольку сигнал $x(n)$ является вещественным, его спектр симметричен: амплитудный спектр чётный ($|X(-k)| = |X(k)|$), а фазовый — нечётный ($\arg X(-k) = -\arg X(k)$).

#### 4. ДПФ конечной последовательности

Для конечной последовательности $x(n)$ длины $N$ прямое ДПФ определяется как:

$$
X(k) = \sum_{n=0}^{N-1} x(n) e^{-j\frac{2\pi}{N}nk}, \qquad k=0,1,\ldots,N-1.
$$

Если рассматривать $x(n)$ как один период бесконечной периодической последовательности $x_p(n)$, то коэффициенты её дискретного ряда Фурье (ДРФ) вычисляются по той же формуле. Таким образом, ДПФ конечной последовательности длины $N$ совпадает с одним периодом ДРФ её периодического продолжения.

Амплитудный спектр периодической последовательности определяется модулями коэффициентов ДРФ, нормированными на $N$:

$$
A_p(k) = \frac{|X(k)|}{N}.
$$

Для наглядного сравнения строятся графики модуля ДПФ $|X(k)|$ и амплитудного спектра периодической последовательности $|X(k)|/N$ в шкале дискретных нормированных частот $\hat{f} = k/N$.

```python
import numpy as np
import matplotlib.pyplot as plt

# Расчёт
X = np.fft.fft(x)
X_shifted = np.fft.fftshift(X)
A_finite = np.abs(X_shifted)      # Модуль ДПФ конечной последовательности
A_periodic = A_finite / N         # Амплитудный спектр периодической последовательности

k = np.arange(N)
k_shifted = k - N // 2
f_norm = k_shifted / N

# Графики
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.stem(f_norm, A_finite, basefmt="k-")
plt.title("Модуль ДПФ конечной последовательности")
plt.xlabel(r"$\hat{f} = k/N$")
plt.ylabel(r"$|X(k)|$")
plt.grid(alpha=0.4)

plt.subplot(1, 2, 2)
plt.stem(f_norm, A_periodic, basefmt="k-", linefmt="C1-", markerfmt="C1o")
plt.title("Амплитудный спектр периодической последовательности")
plt.xlabel(r"$\hat{f} = k/N$")
plt.ylabel(r"$|X(k)| / N$")
plt.grid(alpha=0.4)

plt.tight_layout()
plt.show()

# Сравнение пиковых значений
non_zero = np.where(A_finite > 1e-10)[0]
print("Сравнение амплитуд на ненулевых частотах:")
for idx in non_zero:
    print(f"  f = {f_norm[idx]:+.3f} (k={k_shifted[idx]:+d}): |X(k)| = {A_finite[idx]:.2f}, |X(k)|/N = {A_periodic[idx]:.4f}")
```

Графики демонстрируют, что форма модуля ДПФ конечной последовательности и амплитудного спектра периодической последовательности полностью идентична. Отличие состоит лишь в масштабном коэффициенте $N$.

Связь между ними заключается в следующем: ДПФ конечной последовательности длины $N$ представляет собой один период дискретного ряда Фурье её периодического продолжения. Модуль ДПФ $|X(k)|$ равен амплитуде соответствующей гармоники ДРФ, умноженной на $N$. Поэтому для получения физических амплитуд гармоник периодической последовательности модуль ДПФ нормируют на $N$: $A_p(k) = |X(k)|/N$.

Согласно выводу программы, на частотах $f = \pm 500$ Гц (нормированная частота $\hat{f} = \pm 0.125$) модуль ДПФ равен $38.72$, а амплитуда периодической последовательности составляет $0.6050$. На частотах $f = \pm 750$ Гц ($\hat{f} = \pm 0.1875$) модуль ДПФ равен $77.44$, а амплитуда периодической последовательности — $1.2100$.

Полученные амплитуды $0.605$ и $1.21$ в точности равны половинам заданных амплитуд $A_1/2 = 1.21/2 = 0.605$ и $A_2/2 = 2.42/2 = 1.21$, что соответствует представлению вещественного косинуса в виде суммы двух комплексных экспонент. Таким образом, ДПФ конечной последовательности позволяет точно определить спектральный состав исходного периодического сигнала.

$$

$$
X_a(k) = \begin{cases}
X(N+k), & -N/2 \le k \le -1, \\
X(k), & 0 \le k \le N/2 - 1.
\end{cases}

#### 7. Равенство Парсеваля

Равенство Парсеваля связывает энергию последовательности во временной области с энергией её ДПФ в частотной области:

$$
\sum_{n=0}^{N-1} |x(n)|^2 = \frac{1}{N} \sum_{k=0}^{N-1} |X(k)|^2.
$$

Левая часть представляет собой суммарную энергию отсчётов последовательности $x(n)$, правая — суммарную энергию спектральных составляющих, нормированную на число отсчётов $N$. Равенство утверждает, что ДПФ не изменяет энергию сигнала: она лишь перераспределяется между временным и частотным представлениями. Множитель $1/N$ возникает из-за того, что ДПФ каждой гармоники даёт амплитуду, в $N$ раз превышающую её физическое значение.

Проверка выполняется для исходной последовательности $x(n)$ длины $N=64$. Энергия во временной области и энергия в частотной области вычисляются по формулам:

$$
E_t = \sum_{n=0}^{N-1} x^2(n), \qquad E_f = \frac{1}{N}\sum_{k=0}^{N-1} |X(k)|^2.
$$

```python
import numpy as np

# Расчёт
X = np.fft.fft(x)

E_time = np.sum(np.abs(x) ** 2)          # Энергия во временной области
E_freq = np.sum(np.abs(X) ** 2) / N      # Энергия в частотной области

abs_err = abs(E_time - E_freq)
rel_err = abs_err / E_time

print(f"Энергия во временной области:   E_t = {E_time:.10f}")
print(f"Энергия в частотной области:    E_f = {E_freq:.10f}")
print(f"Абсолютная разность:            |E_t - E_f| = {abs_err:.2e}")
print(f"Относительная погрешность:      {rel_err:.2e}")

# Проверка равенства Парсеваля через составляющие спектра
non_zero = np.where(np.abs(X) > 1e-10)[0]
E_from_peaks = np.sum(np.abs(X[non_zero]) ** 2) / N
print(f"\nЭнергия, посчитанная только по ненулевым спектральным составляющим: {E_from_peaks:.10f}")
```

В результате расчёта получены значения энергий:

- энергия во временной области: $E_t = 234.2560000000$;
- энергия в частотной области: $E_f = 234.2560000000$;
- абсолютная разность: $|E_t - E_f| \approx 2.84 \cdot 10^{-14}$;
- относительная погрешность: $\approx 1.21 \cdot 10^{-16}$.

Разность находится на уровне машинной погрешности чисел двойной точности, что подтверждает выполнение равенства Парсеваля:

$$
\sum_{n=0}^{N-1} |x(n)|^2 = \frac{1}{N} \sum_{k=0}^{N-1} |X(k)|^2.
$$

**Смысл равенства Парсеваля.** Оно утверждает сохранение энергии сигнала при переходе от временного представления к частотному. Суммарная энергия отсчётов последовательности равна суммарной энергии её спектральных составляющих (с учётом нормировки на $N$). Это означает, что ДПФ не теряет и не добавляет энергию — оно лишь раскладывает сигнал по базису комплексных экспонент, сохраняя полную энергию.

Проверка также показывает, что в данном сигнале практически вся энергия сосредоточена в четырёх ненулевых спектральных составляющих: на частотах $\pm 500$ Гц и $\pm 750$ Гц. Энергия, посчитанная только по этим компонентам, оказалась равной $234.2560000000$ — это в точности совпадает с полной энергией сигнала. Остальные $60$ отсчётов ДПФ равны нулю с точностью до машинной погрешности, поэтому не дают вклада в энергию. Это согласуется со спектральным анализом п. 3: последовательность $x(n)$ является суммой двух гармоник, поэтому её спектр содержит ровно четыре ненулевых отсчёта.

Физически равенство Парсеваля утверждает сохранение энергии за период при переходе к частотному представлению; средняя мощность при этом равна E/N и также может вычисляться в обеих областях.

Пара преобразований ДПФ–ОДПФ образует взаимно обратное преобразование: восстановленная последовательность совпадает с исходной с точностью $\approx 1.33\cdot10^{-15}$, что соответствует машинной погрешности чисел двойной точности.

Амплитудный спектр содержит ровно четыре ненулевые компоненты на частотах $\pm 500$ Гц ($|X| = 38.72$) и $\pm 750$ Гц ($|X| = 77.44$), а фазовый спектр нечётен и принимает значения $\pm\pi/4$ и $\pm\pi/8$ соответственно. Эти значения точно соответствуют теоретическим: $A_1 N/2 = 38.72$, $A_2 N/2 = 77.44$. Поскольку частоты сигнала попадают ровно в отсчёты ДПФ ($k = \pm 8$ и $k = \pm 12$), растекание спектра отсутствует.

Установлена связь модуля ДПФ конечной последовательности с амплитудным спектром периодической последовательности: $|X(k)|/N$ даёт амплитуды гармоник, равные $A_1/2 = 0.605$ и $A_2/2 = 1.21$. Это соответствует представлению косинуса суммой двух комплексных экспонент.

Периодический аналоговый сигнал восстановлен по отсчётам ДПФ с шагом $\Delta t = 0.25T$; максимальное отклонение от исходного сигнала составило $\approx 4.01\cdot10^{-14}$. Совпадение объясняется отсутствием наложения спектров ($f_\text{д} = 4000$ Гц $> 2f_2 = 1500$ Гц), целым числом периодов на интервале $N$ и использованием полного набора отсчётов $X_a(k)$.

Проверено равенство Парсеваля: энергия во временной и частотной областях совпала ($E_t = E_f = 234.256$) с абсолютной разностью $\approx 2.84\cdot10^{-14}$ и относительной погрешностью $\approx 1.21\cdot10^{-16}$. Вся энергия сигнала сосредоточена в четырёх ненулевых спектральных составляющих, остальные $60$ отсчётов ДПФ равны нулю с точностью до машинной погрешности. Равенство Парсеваля подтверждает сохранение энергии сигнала при переходе между временным и частотным представлениями и позволяет вычислять энергию (и среднюю мощность) последовательности в любой из областей.

Все полученные результаты согласуются с теоретическими положениями цифровой обработки сигналов.
