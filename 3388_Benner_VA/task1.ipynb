## Основные теоретические положения

### Дискретное время и типовые сигналы

Частота дискретизации fд определяет период T = 1/fд. Дискретное время t = nT; нормированное дискретное время — индекс n. Цифровой единичный импульс обозначается u₀(n):

$$
u_0(n)=\begin{cases}1,&n=0,\\0,&n\ne0.\end{cases}
$$

### Импульсная характеристика и свёртка

Импульсная характеристика h(n) — реакция системы на единичный импульс при нулевых начальных условиях. Входные и выходные отсчёты до начала воздействия принимаются равными нулю. Реакция определяется свёрткой:

$$
y(n)=\sum_{m=0}^{\infty}h(m)x(n-m).
$$

Для конечных массивов длиной Nₓ и Nₕ длина полной свёртки составляет Nₓ + Nₕ − 1.

### Разностное уравнение и передаточная функция

Общее разностное уравнение имеет вид:

$$
y(n)=\sum_{i=0}^{L-1}b_i x(n-i)-\sum_{k=1}^{M-1}a_k y(n-k).
$$

Здесь L и M — количества коэффициентов числителя и знаменателя; коэффициент a₀ равен 1. Обозначение L использовано вместо N, чтобы отличать число коэффициентов от длины ДПФ во второй части работы.

Z-преобразование последовательности:

$$
X(z)=\sum_{n=0}^{\infty}x(n)z^{-n}.
$$

При нулевых начальных условиях передаточная функция:

$$
H(z)=\frac{Y(z)}{X(z)}=\mathcal{Z}\{h(n)\}
=\frac{\displaystyle\sum_{i=0}^{L-1}b_i z^{-i}}
{1+\displaystyle\sum_{k=1}^{M-1}a_k z^{-k}}.
$$

В работе L = M = 3, поэтому получается звено второго порядка. Его формулы с коэффициентами варианта приведены в выполнении работы.

Нерекурсивная система не содержит слагаемых с предыдущими значениями выхода и имеет конечную импульсную характеристику. Рекурсивная система содержит такие слагаемые; её импульсная характеристика может быть бесконечной.

### Обратное Z-преобразование звена второго порядка

Для звена второго порядка:

$$
G(z)=\frac{1}{1-2r\cos\varphi\,z^{-1}+r^2z^{-2}},
$$

$$
g(n)=r^n\frac{\sin((n+1)\varphi)}{\sin\varphi},\quad n\geq0.
$$

При отрицательных индексах g(n) = 0. Для знаменателя 1 + a₁z⁻¹ + a₂z⁻²:

$$
r=\sqrt{a_2},\qquad \cos\varphi=-\frac{a_1}{2r}.
$$

Умножение G(z) на числитель и применение свойства задержки дают:

$$
h(n)=b_0g(n)+b_1g(n-1)+b_2g(n-2).
$$

Эта формула применяется в данной работе, где корни знаменателя комплексно-сопряжённые и sin φ ≠ 0.

### Нули, полюса и устойчивость

Нули — значения z, при которых передаточная функция равна нулю; полюса соответствуют обращению знаменателя в нуль. Алгебраическая и показательная формы комплексного числа связаны соотношениями:

$$
z=\xi+j\eta=re^{j\varphi},\qquad r=\sqrt{\xi^2+\eta^2},
$$

$$
\xi=r\cos\varphi,\qquad\eta=r\sin\varphi,\qquad\varphi=\arg z.
$$

Устойчивость рассматриваемого звена проверяется расположением всех полюсов строго внутри единичной окружности:

$$
|p_k|<1.
$$

### Частотные характеристики и нормирование частот

23:

$$
\widehat{\omega}=\omega T=\frac{2\pi f}{f_{\mathrm{d}}},\qquad
\widehat f=\frac{f}{f_{\mathrm{d}}}.
$$

Основная полоса для положительных частот: 0 ≤ f ≤ fд/2; 0 ≤ f̂ ≤ 0.5; 0 ≤ ω̂ ≤ π. При изображении обеих половин спектра используется интервал −π ≤ ω̂ < π. На графике с осью ω̂/π границы равны −1 и 1.

Комплексная частотная характеристика получается подстановкой z = exp(jω̂):

$$
H(e^{j\widehat{\omega}})=\sum_{n=0}^{\infty}h(n)e^{-j\widehat{\omega}n}.
$$

$$
A(\widehat{\omega})=|H(e^{j\widehat{\omega}})|,\qquad
\psi(\widehat{\omega})=\arg H(e^{j\widehat{\omega}}).
$$

### ДПФ, ОДПФ и спектры


$$
X(k)=\sum_{n=0}^{N-1}x(n)e^{-j2\pi kn/N},\quad k=0,\ldots,N-1,
$$

$$
x(n)=\frac{1}{N}\sum_{k=0}^{N-1}X(k)e^{j2\pi kn/N}.
$$

Коэффициенты ряда Фурье периодической последовательности равны C(k) = X(k)/N. Двусторонний амплитудный спектр периодической последовательности — |C(k)|, фазовый — arg C(k). Модуль ДПФ конечного фрагмента равен |X(k)|. Поэтому модуль ДПФ в N раз больше амплитуд линий периодического спектра. Спектр действительного сигнала обладает симметрией: модули на противоположных частотах равны, а фазы противоположны.

### Равенство Парсеваля


$$
\sum_{n=0}^{N-1}|x(n)|^2=\frac1N\sum_{k=0}^{N-1}|X(k)|^2.
$$

Смысл равенства — совпадение суммы квадратов отсчётов, вычисленной по временному и спектральному представлениям.

### 1.1. Импульсная характеристика

```python
delta = np.zeros(N1); delta[0] = 1
h = signal.lfilter(b, a, delta)
poles = np.roots(a); zeros = np.roots(b)
r = np.sqrt(a[2])
theta = np.arccos(-a[1]/(2*r))
nh = np.arange(N1)
def g(indices):
    # Аналитическое обратное Z-преобразование звена второго порядка.
    result = np.zeros(len(indices))
    valid = indices >= 0
    q = indices[valid]
    result[valid] = r**q*np.sin((q+1)*theta)/np.sin(theta)
    return result
h_formula = sum(b[i]*g(nh-i) for i in range(3))
assert np.allclose(h, h_formula, atol=1e-13)
print(f'r={r:.10f}; φ={theta:.10f}; φ/π={theta/np.pi:.10f}')
print('Первые отсчёты h:', h[:5])
print('Ошибка аналитической формулы:', np.max(abs(h-h_formula)))
fig, ax = plt.subplots(); stem(ax, nh, h, 'Импульсная характеристика: первые 22 отсчёта', 'n', 'h(n)'); plt.show()
```

При нулевых начальных условиях h(n) = 0 для n < 0. Подстановка единичного импульса u₀(n) в разностное уравнение даёт:

$$
\begin{aligned}
h(n)={}&b_0u_0(n)+b_1u_0(n-1)+b_2u_0(n-2)\\
&-a_1h(n-1)-a_2h(n-2).
\end{aligned}
$$

Для варианта 2:

$$
\begin{aligned}
h(n)={}&0.54u_0(n)-0.549612u_0(n-1)+0.648u_0(n-2)\\
&-0.8278h(n-1)-0.652h(n-2).
\end{aligned}
$$

По формуле обратного Z-преобразования и свойству задержки:

$$
h(n)=0.54g(n)-0.549612g(n-1)+0.648g(n-2),
$$

$$
g(n)=\begin{cases}
r^n\dfrac{\sin((n+1)\varphi)}{\sin\varphi},&n\geq0,\\
0,&n<0,
\end{cases}
$$

$$
r=\sqrt{0.652}=0.8074651695,\qquad
\varphi=\arccos\left(-\frac{0.8278}{2\sqrt{0.652}}\right)=2.1089968922.
$$

В частности, h(0) = 0.54, h(1) = −0.996624, h(2) = 1.1209253472. Аналитическая формула проверена по всем 22 отсчётам. Действительная длина импульсной характеристики бесконечна; N₁ = 22 задаёт длину вычисляемого фрагмента. Для данного звена полюсы не сокращаются нулями, а последовательность после этого фрагмента продолжает затухать.

### 1.2. Реакция по свёртке

```python
nx = np.arange(N2)
M = int(np.floor(N2/2+0.5))  # Округление до ближайшего целого; здесь 16.
x1 = (nx < M).astype(float)
y_full = np.convolve(x1, h, mode='full')
y1 = y_full[:N2]
print(f'Длина массива x={len(x1)}, ненулевая часть={M}; длина полной свёртки={len(y_full)}')
fig, ax = plt.subplots(2, 1, figsize=(11, 7))
stem(ax[0], nx, x1, 'Воздействие', 'n', 'x[n]')
stem(ax[1], np.arange(len(y_full)), y_full, 'Полная свёртка с усечённой характеристикой', 'n', 'y₁[n]')
ax[1].axvline(N2-0.5, color='r', linestyle='--', label='Граница окна 32 отсчёта'); ax[1].legend()
plt.tight_layout(); plt.show()
```

$$
x[n]=\begin{cases}1,&0\leq n<16,\\0,&16\leq n\leq31.\end{cases}
$$

Длительность ненулевого прямоугольного импульса составляет 16 отсчётов, или 8 мс; длина массива воздействия — 32 отсчёта. Линейная свёртка массивов имеет длину 32+22-1=53. Поскольку последние 16 элементов воздействия нулевые, последние 16 элементов этой свёртки также нулевые; потенциально ненулевая часть содержит 37 отсчётов.

Ограничение реакции первыми 32 отсчётами позволяет сравнить выходы в одном временном окне с воздействием. Оно не означает физического прекращения реакции: истинный рекурсивный выход имеет бесконечный затухающий хвост. Дополнительно усечение h до 22 отсчётов вносит приближение.

### 1.3. Реакция по разностному уравнению и сравнение

```python
def recurrence(x, b, a):
    y = np.zeros(len(x))
    for n in range(len(x)):
        y[n] = sum(b[i]*x[n-i] for i in range(len(b)) if n >= i)
        y[n] -= sum(a[i]*y[n-i] for i in range(1, len(a)) if n >= i)
        y[n] /= a[0]
    return y
y2 = recurrence(x1, b, a)
h32 = signal.lfilter(b, a, np.r_[1., np.zeros(N2-1)])
y_exact_conv = np.convolve(x1, h32)[:N2]
assert np.allclose(y2, signal.lfilter(b, a, x1), atol=1e-13)
assert np.allclose(y2, y_exact_conv, atol=1e-13)
print('Ошибка свёртки с h длины 22:', np.max(np.abs(y1-y2)))
print('Ошибка свёртки с h длины 32:', np.max(np.abs(y_exact_conv-y2)))
fig, ax = plt.subplots(3, 1, figsize=(11, 9))
stem(ax[0], nx, x1, 'Вход', 'n', 'x[n]')
ax[1].plot(nx, y1, 'o-', label='Свёртка, h длины 22')
ax[1].plot(nx, y2, '.--', label='Разностное уравнение'); ax[1].legend(); ax[1].set(xlabel='n', ylabel='y[n]')
stem(ax[2], nx, y1-y2, 'Разность выходов', 'n', 'y₁−y₂')
plt.tight_layout(); plt.show()
```

Для варианта 2:

$$
y[n]=0.54x[n]-0.549612x[n-1]+0.648x[n-2]-0.8278y[n-1]-0.652y[n-2],
$$

где все входные и выходные отсчёты с отрицательными индексами нулевые. Длины входного и выходного массивов при рекурсивном расчёте равны 32. Физическая реакция после конца окна продолжается.

Свёртка и разностное уравнение описывают одну и ту же систему. При характеристике длины 22 первые 22 выходных отсчёта совпадают, затем возникает ошибка усечения. Увеличение длины h до 32 обеспечивает совпадение во всём окне с точностью вычислений. Небольшое расхождение первоначальных графиков связано именно с усечением h.

### 1.6–1.7. АЧХ и ФЧХ

```python
w = np.linspace(0, np.pi, 4097)
_, H = signal.freqz(b, a, worN=w)
phase = np.angle(H)
imax = np.argmax(np.abs(H))
print(f'Максимум АЧХ: {abs(H[imax]):.8f}, ω/π={w[imax]/np.pi:.8f}, f={w[imax]*fs1/(2*np.pi):.5f} Гц')
print('АЧХ при 0 и π:', np.abs(H[[0, -1]]))
fig, ax = plt.subplots(2, 2, figsize=(12, 7))
for col, freq, xlabel in [(0, w/np.pi, 'ω̂/π'), (1, w*fs1/(2*np.pi), 'f, Гц')]:
    ax[0,col].plot(freq, abs(H)); ax[0,col].set(xlabel=xlabel, ylabel='|H|', title='АЧХ')
    ax[1,col].plot(freq, phase, label='Фаза')
    ax[1,col].set(xlabel=xlabel, ylabel='Фаза, рад', title='ФЧХ'); ax[1,col].legend()
plt.tight_layout(); plt.show()
```

Положительная основная полоса: 0 ≤ ω̂ ≤ π, то есть 0 ≤ f ≤ 1000 Гц. На графике показана ось ω̂/π от 0 до 1. Связь шкал: f = ω̂fд/(2π) = 1000·(ω̂/π) Гц. Для двустороннего изображения соответствующие интервалы равны −π ≤ ω̂ < π и −1000 ≤ f < 1000 Гц.

Подстановка z = exp(jω̂) в передаточную функцию позволяет сопоставить карту корней и АЧХ: вблизи направления полюсов знаменатель уменьшается, поэтому передача возрастает; вблизи направления нулей числитель уменьшается. Полюсы имеют аргументы около ±0.6713π, нули — около ±0.3462π. Нули не лежат на единичной окружности, поэтому АЧХ в этих точках не обращается в нуль.

Максимум вычисленной АЧХ составляет около 4.0034 при частоте 688.72 Гц. При нулевой частоте АЧХ равна 0.2574, при 1000 Гц — 2.1082. Следовательно, низкие частоты воздействия подавляются сильнее высоких. На положительной полуполосе скачка ФЧХ на π не обнаружено. ФЧХ определяется аргументом комплексной частотной характеристики.

На ФЧХ около f = 444.82 Гц главное значение аргумента переходит от −π к +π. Видимый скачок составляет приблизительно 2π; он соответствует АЧХ около 0.3098. Это переход через границу представления аргумента, а не обращение АЧХ в нуль.


### 2.1. Последовательность и временные шкалы

```python
n = np.arange(N)
tn = n*T
x = A1*np.cos(2*np.pi*f1*tn+np.pi/4)+A2*np.cos(2*np.pi*f2*tn+np.pi/8)
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
stem(ax[0], tn*1000, x, 'Дискретное время t=nT', 't, мс', 'x[n]')
stem(ax[1], n, x, 'Дискретное нормированное время', 'n=t/T', 'x[n]')
plt.tight_layout(); plt.show()
print(f'Период NT={N*T:.10f} с; последний отсчёт {(N-1)*T:.10f} с; Δf={fs2/N:.5f} Гц')
```

$$
x[n]=1.02\cos(\pi n/4+\pi/4)+2.04\cos(3\pi n/8+\pi/8).
$$

Нормированное дискретное время — целый индекс n=t/T; физическое время tₙ=nT. Частоты точно попадают в бины k₁=Nf₁/fₛ=8 и k₂=12. N = 64 является периодом последовательности, но её минимальный период равен 16 отсчётам: периоды отдельных гармоник равны 8 и 16 отсчётам.

### 2.2. ДПФ и ОДПФ

```python
X = np.fft.fft(x)
x_back = np.fft.ifft(X)
err_dft = np.max(np.abs(x_back-x))
print('Максимальная ошибка ОДПФ:', err_dft)
assert np.allclose(x_back, x, atol=1e-12)
fig, ax = plt.subplots(); ax.plot(n, x, 'o', label='Исходная'); ax.plot(n, x_back.real, '.--', label='ОДПФ от ДПФ')
ax.set(xlabel='n=t/T', ylabel='x[n]', title='Восстановление дискретной последовательности'); ax.legend(); plt.show()
```

ОДПФ восстанавливает исходные отсчёты с погрешностью округления. Мнимая часть результата численно близка к нулю, поскольку исходный сигнал действительный.

### 2.3. Амплитудный и фазовый спектры

```python
ks = np.arange(-N//2, N//2)
Xs = np.fft.fftshift(X)
nu = ks/N
ck = Xs/N
mask = abs(ck) > 1e-12
phases = np.where(mask, np.angle(ck), np.nan)
fig, ax = plt.subplots(2, 2, figsize=(12, 7))
for col, freq, xlabel in [(0, nu, 'f̂=k/N'), (1, nu*fs2, 'f, Гц')]:
    stem(ax[0,col], freq, abs(ck), 'Двусторонний амплитудный спектр', xlabel, '|X[k]|/N')
    stem(ax[1,col], freq[mask], phases[mask], 'Фазовый спектр ненулевых линий', xlabel, 'Фаза, рад')
plt.tight_layout(); plt.show()
for k, c in zip(ks[mask], ck[mask]):
    print(f'k={k:+3d}; ν={k/N:+.5f}; f={k*fs2/N:+.2f} Гц; |X|={abs(c)*N:.8f}; |c|={abs(c):.8f}; arg/π={np.angle(c)/np.pi:+.8f}')
```

Связь частот: f=f̂ fₛ=kfₛ/N; шаг ДПФ равен 93.75 Гц. В центрированном представлении отрицательные частоты явно показаны; в исходном массиве им соответствуют индексы 52 и 56.

| k | f̂ | f, Гц | Модуль ДПФ | Двусторонняя амплитуда | Фаза |
|---:|---:|---:|---:|---:|---:|
| −12 | −0.1875 | −1125 | 65.28 | 1.02 | −π/8 |
| −8 | −0.125 | −750 | 32.64 | 0.51 | −π/4 |
| 8 | 0.125 | 750 | 32.64 | 0.51 | π/4 |
| 12 | 0.1875 | 1125 | 65.28 | 1.02 | π/8 |

У действительного сигнала амплитуды симметричны, фазы ненулевых линий противоположны. Односторонние амплитуды гармоник равны 1.02 и 2.04. Остальные линии теоретически нулевые, поэтому их фазы не интерпретируются. Частоты гармоник соответствуют целым номерам отсчётов ДПФ.

### 2.4. Конечная и периодическая последовательности

```python
Xfinite = np.fft.fft(x)  # ДПФ конечного массива из тех же 64 отсчётов.
fig, ax = plt.subplots(2, 1, figsize=(11, 7))
stem(ax[0], nu, abs(np.fft.fftshift(Xfinite)), 'Модуль ДПФ конечной последовательности', 'f̂=k/N', '|X(k)|')
stem(ax[1], nu, abs(ck), 'Амплитудный спектр периодической последовательности', 'f̂=k/N', '|C(k)|')
plt.tight_layout(); plt.show()
assert np.allclose(abs(np.fft.fftshift(Xfinite))/N, abs(ck), atol=1e-12)
print('Максимальная ошибка связи |X|/N=|C|:', np.max(abs(abs(np.fft.fftshift(Xfinite))/N-abs(ck))))
```

ДПФ конечного массива из N отсчётов совпадает с ДПФ одного периода периодической последовательности с теми же отсчётами. Коэффициенты ряда Фурье равны C(k) = X(k)/N. Поэтому модуль ДПФ конечной последовательности в N = 64 раза больше двусторонней амплитуды периодического спектра. Это подтверждается графиками и численной проверкой.

### 2.5–2.6. Восстановление аналогового сигнала и сравнение спектров

Используется требуемая формула

$$
x_r(t)=\frac{1}{N}\sum_{k=-N/2}^{N/2-1}X_a[k]\,e^{j2\pi kt/(NT)}.
$$

$$
X_a[k]=\begin{cases}
X[N+k], & -N/2\leq k\leq -1, \\
X[k], & 0\leq k\leq N/2-1.
\end{cases}
$$

Переупорядочивание реализуется `fftshift`. Интервал восстановления: 0≤ t≤(N-1)T, шаг Δ t=T/4. Исходный аналоговый сигнал определяется той же суммой двух косинусоид с непрерывным временем.

```python
t = np.arange(4*(N-1)+1)*(T/4)
xr_complex = np.exp(2j*np.pi*np.outer(t/(N*T), ks)) @ Xs/N
xr = xr_complex.real
xa = A1*np.cos(2*np.pi*f1*t+np.pi/4)+A2*np.cos(2*np.pi*f2*t+np.pi/8)
err_analog = np.max(abs(xr_complex-xa))
print('Число точек:', len(t), '; шаг, с:', T/4)
print('Максимальная ошибка аналогового восстановления:', err_analog)
print('Максимальная мнимая часть:', np.max(abs(xr_complex.imag)))
assert np.allclose(xr_complex, xa, atol=1e-11)
assert np.allclose(xr[::4], x, atol=1e-11)
fig, ax = plt.subplots(3, 2, figsize=(13, 11))
stem(ax[0,0], tn*1000, x, 'Период дискретной последовательности', 't, мс', 'x[n]')
stem(ax[0,1], nu*fs2, abs(Xs), 'Модуль ДПФ', 'f, Гц', '|X[k]|')
ax[1,0].plot(t*1000, xr); ax[1,0].set(title='Восстановленный аналоговый сигнал', xlabel='t, мс', ylabel='xᵣ(t)')
stem(ax[1,1], nu[mask]*fs2, abs(ck[mask]), 'Амплитуды линий восстановленного сигнала', 'f, Гц', '|cₖ|')
ax[2,0].plot(t*1000, xa, label='Исходный'); ax[2,0].plot(t*1000, xr, '--', label='Восстановленный'); ax[2,0].scatter(tn*1000,x,s=10,label='Отсчёты'); ax[2,0].legend(); ax[2,0].set(xlabel='t, мс',ylabel='x(t)',title='Сравнение с исходным аналоговым сигналом')
ax[2,1].plot(t*1000, xr-xa); ax[2,1].set(xlabel='t, мс',ylabel='xᵣ−x',title='Ошибка восстановления')
plt.tight_layout(); plt.show()
```

Коэффициенты восстановленного аналогового ряда равны Xₐ(k)/N: это следует из формулы задания и согласуется с нормировкой ряда Фурье. Двусторонние амплитуды равны 0.51 на частотах ±750 Гц и 1.02 на частотах ±1125 Гц. Модуль ДПФ в 64 раза больше этих амплитуд. На графике амплитудного спектра показаны величины соответствующих коэффициентов ряда.

Обе гармоники имеют частоты ниже 3000 Гц, то есть ниже половины частоты дискретизации. Выполняется условие восстановления по теореме Котельникова. В рассматриваемом случае частоты также соответствуют целым номерам гармоник периода NT: k₁ = 8, k₂ = 12. Восстановленный по формуле задания сигнал совпадает с исходной суммой косинусоид с точностью вычислений; в моменты nT совпадают и дискретные отсчёты.

### 2.7. Равенство Парсеваля

```python
E_time = np.sum(abs(x)**2)
E_freq = np.sum(abs(X)**2)/N
P_theory = (A1**2+A2**2)/2
print(f'Энергия периода во времени: {E_time:.12f}')
print(f'Энергия периода по спектру: {E_freq:.12f}')
print(f'Абсолютная невязка: {abs(E_time-E_freq):.3e}')
print(f'Средняя мощность: {E_time/N:.12f}; теоретически: {P_theory:.12f}')
assert np.isclose(E_time, E_freq, rtol=1e-13)
assert np.isclose(E_time/N, P_theory, rtol=1e-13)
```

Энергия одного периода равна 166.464, средняя мощность — 2.601. Равенство Парсеваля подтверждает согласованность временного и спектрального представлений с принятой нормировкой ДПФ.
