## **Теоретическая справка**

### 1. Линейные дискретные системы:

**Линейная дискретная система (ЛДС)** преобразует входную последовательность $x(n)$ (воздействие) в выходную $y(n)$ (реакцию).

**Импульсная характеристика (ИХ)** $h(n)$ - это реакция ЛДС на цифровой единичный импульс при нулевых начальных условиях:
$$u_0(n) = \begin{cases} 1, & n = 0 \\ 0, & n \neq 0 \end{cases}$$

**Нулевые начальные условия:** $x(n-i) = 0$ при $n-i < 0$, $y(n-k) = 0$ при $n-k < 0$.

### 2. Формула свертки:
Реакция ЛДС при нулевых начальных условиях определяется дискретной сверткой:
$$y(n) = \sum_{m=0}^{\infty} h(m)\, x(n-m) = \sum_{m=0}^{\infty} x(m)\, h(n-m).$$
Если $h(n)$ имеет длину $N_h$, а $x(n)$ - длину $N_x$, то длина свертки $N_y = N_h + N_x - 1$.

Системы с бесконечной ИХ называют БИХ-системами.

### 3. Разностное уравнение:
$$y(n) = \sum_{i=0}^{M} b_i\, x(n-i) - \sum_{k=1}^{L} a_k\, y(n-k).$$

Система **рекурсивная**, если хотя бы один коэффициент $a_k \ne 0$. **Звено 2-го порядка** - ЛДС, у которой запаздывания воздействия и реакции не превышают двух отсчетов:
$$y(n) = b_0 x(n) + b_1 x(n-1) + b_2 x(n-2) - a_1 y(n-1) - a_2 y(n-2).$$

### 4. Передаточная функция, нули и полюсы:
**Передаточная функция** - Z-преобразование ИХ, или отношение Z-изображений реакции и воздействия при нулевых начальных условиях:
$$H(z) = \sum_{n=0}^{\infty} h(n) z^{-n} = \frac{Y(z)}{X(z)} = \frac{b_0 + b_1 z^{-1} + b_2 z^{-2}}{1 + a_1 z^{-1} + a_2 z^{-2}}.$$
- Нули $z_0$ - значения $z$, при которых $H(z) = 0$.
- Полюсы $z_*$ - значения $z$, при которых знаменатель равен нулю.

Алгебраическая форма $z = \xi + j\eta$, показательная $z = re^{j\varphi}$, где $r = \sqrt{\xi^2+\eta^2}$, $\varphi = \text{atan2}(\eta, \xi)$.

**ИХ звена 2-го порядка.** Полюсы звена $r_* e^{\pm j\varphi_*}$, где $a_1 = -2r_*\cos\varphi_*$, $a_2 = r_*^2$. ИХ:
$$h(n) = b_0\, g(n) + b_1\, g(n-1) + b_2\, g(n-2), \qquad g(n) = r_*^{\,n}\frac{\sin((n+1)\varphi_*)}{\sin\varphi_*}, \quad g(n) = 0 \text{ при } n < 0.$$

**Устойчивость.** Система устойчива, если при ограниченном воздействии реакция ограничена. Необходимо и достаточно $\sum_{n=0}^{\infty}|h(n)| < \infty$, что выполняется, если все полюсы лежат внутри единичного круга $|z_*| < 1$.

### 5. Частотные характеристики:
**Частотная характеристика** - значение ПФ на единичной окружности:
$$H(e^{j\hat\omega}) = H(z)\big|_{z = e^{j\hat\omega}}.$$
- $|H(e^{j\hat\omega})|$ - амплитудно-частотная характеристика (АЧХ);
- $\arg H(e^{j\hat\omega})$ - фазочастотная характеристика (ФЧХ).

Нормированная частота: $\hat\omega = \omega T = 2\pi f / f_д$. Частота Найквиста $f_д/2$.

### 6. Дискретное время и дискретная частота:
- **Дискретное время:** $nT$, где $T = 1/f_д$ - период дискретизации.
- **Дискретное нормированное время:** $n = nT/T$, номер отсчета.
- **Дискретная нормированная частота:** $k$, номер отсчета ДПФ.
- **Абсолютная дискретная частота:** $f_k = k \dfrac{f_д}{N} = \dfrac{k}{NT}$.

Шаг по частоте (разрешение ДПФ) равен $\Delta f = f_д/N$.

### 7. Дискретное преобразование Фурье:
Прямое ДПФ:
$$X(k) = \sum_{n=0}^{N-1} x(n)\, e^{-j\frac{2\pi}{N} kn}, \quad k = 0, 1, \dots, N-1.$$

Обратное ДПФ (ОДПФ):
$$x(n) = \frac{1}{N} \sum_{k=0}^{N-1} X(k)\, e^{j\frac{2\pi}{N} kn}.$$

**Амплитудный спектр** - $|X(k)|$, **фазовый спектр** - $\arg X(k)$.

Для вещественной $x(n)$: $X(N-k) = X^*(k)$. Отсчеты $k > N/2$ соответствуют отрицательным частотам $f_k - f_д$.

**Дискретная гармоника** $A\cos\left(\frac{2\pi}{N} k_0 n + \varphi\right)$, $0 < k_0 < N/2$, имеет два ненулевых отсчета ДПФ:
$$X(k_0) = \frac{AN}{2} e^{j\varphi}, \qquad X(N-k_0) = \frac{AN}{2} e^{-j\varphi}.$$

### 8. Теорема Котельникова:
Сигнал с ограниченным спектром, не содержащий частот выше $f_{max}$, восстанавливается по своим отсчетам без искажений, если частота дискретизации удовлетворяет условию
$$f_д > 2 f_{max}, \quad \text{т. е.} \quad f_{max} < f_д/2.$$

### 9. Равенство Парсеваля:
Для ДПФ выполняется равенство Парсеваля:
$$\sum_{n=0}^{N-1} |x(n)|^2 = \frac{1}{N} \sum_{k=0}^{N-1} |X(k)|^2.$$

```python
import numpy as np
import scipy.signal as sig
import matplotlib.pyplot as plt

# оформление графиков 
plt.rcParams.update({
    "figure.figsize": (10, 3.6),
    "axes.grid": True,
    "grid.alpha": 0.3,
    "axes.spines.top": False,
    "axes.spines.right": False,
})
C1, C2 = "#2a6fdb", "#e07b39"

N_var = 10

# коэффициенты числителя (b) и знаменателя (a) передаточной функции
b0 = 0.5 + 0.02 * N_var
b1 = b0 * (-1) ** (N_var + 1) * (0.9822 + 0.0178 * N_var)
b2 = b0 * (0.8 + 0.2 * (N_var % 5))
a0 = 1
a1 = (-1) ** N_var * (0.7778 + 0.025 * N_var)
a2 = 0.64 + 0.006 * N_var

b = np.array([b0, b1, b2])
a = np.array([a0, a1, a2])

N1 = N_var % 10 + 20 # длина импульсной характеристики
N2 = N_var % 10 + 30 # длина воздействия
fs = 1000 * N_var # частота дискретизации f_д

print(f"b=[{b0:.5f}, {b1:.5f}, {b2:.5f}]")
print(f"a=[{a0}, {a1:.5f}, {a2:.5f}]")
print(f"N1={N1}, N2={N2}, f_д={fs} Гц")
```

Запишем **аналитическую формулу ИХ рекурсивного звена 2-го порядка с учетом нулевых начальных условий**. Поясним, чему в действительности равна **длина ИХ**.

ИХ - реакция на единичный импульс $u_0(n)$, вычисляем функцией `lfilter(b, a, u0)` для $N_1 = 20$ отсчетов.


```python
n1 = np.arange(N1) # номера отсчетов 0, 1,..., N1-1
u0 = np.zeros(N1); u0[0] = 1 # цифровой единичный импульс
h = sig.lfilter(b, a, u0) # реакция на единичный импульс = ИХ

print("Отсчеты h(n):", np.round(h, 4))

plt.stem(n1, h, linefmt=C1, markerfmt="o", basefmt="k-")
plt.title("Импульсная характеристика h(n), N1 = 20")
plt.xlabel("n"); plt.ylabel("h(n)")
plt.show()
```

На графике показаны первые $N_1 = 20$ отсчетов импульсной характеристики $h(n)$.

$h(0) = 0.7$, $h(1) = -1.5316$, $h(2) = 1.6442$ формируются коэффициентами числителя $b_0, b_1, b_2$ и обратной связью. Начиная с $n = 3$ входной импульс уже закончился, и отсчеты порождаются только рекурсивной частью $-a_1 h(n-1) - a_2 h(n-2)$.

Аналитическая формула задает $h(n)$ явно, в отличие от рекурсивного разностного уравнения. Для звена 2-го порядка с учетом нулевых начальных условий (см. теорию, раздел 4):
$$h(n) = b_0\, g(n) + b_1\, g(n-1) + b_2\, g(n-2), \qquad g(n) = r_*^{\,n}\frac{\sin((n+1)\varphi_*)}{\sin\varphi_*}.$$
Модуль и аргумент полюса находим из $a_2 = r_*^2$, $a_1 = -2r_*\cos\varphi_*$.

```python
r_p = np.sqrt(a2) # модуль полюса
phi_p = np.arccos(-a1 / (2 * r_p)) # аргумент полюса
print(f"r*={r_p:.4f}, phi*={phi_p:.4f} рад={phi_p/np.pi:.4f}π")

def g(m):
    # r^m * sin((m+1)phi) / sin(phi), равна нулю при m < 0
    return np.where(m >= 0, r_p**m * np.sin((m + 1) * phi_p) / np.sin(phi_p), 0.0)

h_an = b0 * g(n1) + b1 * g(n1 - 1) + b2 * g(n1 - 2)
print("Максимальное расхождение формулы и численной ИХ:", np.max(np.abs(h - h_an)))
```

Знакопеременный характер ИХ (плюс, минус, плюс и т.д.) - следствие того, что аргумент полюса

#### 2. Вычислим **реакцию $y_1(n)$ по формуле свертки**. В качестве воздействия $x(n)$ длины $N_2$ выберем **дискретный прямоугольный импульс**. Выведем **графики воздействия и реакции**.

**Воздействие:**
$$x(n) = \begin{cases} 1, & 0 \le n < int(N_2/2) \\ 0, & int(N_2/2) \le n \le N_2 - 1 \end{cases}$$

При $N_2 = 30$: $int(30/2) = 15$. Значит, массив воздействия имеет длину 30, из них первые 15 отсчетов - единицы, остальные 15 - нули.

Функция $int(t)$ по заданию - округление до ближайшего целого. Встроенные `round` и `np.round` в Python округляют половинки к четному (например, `round(16.5) = 16`), поэтому используем округление `floor(t + 0.5)`.

Свертку двух конечных массивов вычисляет функция `np.convolve(h, x)`. Используем ИХ длины $N_1 = 20$ из п. 1.

```python
def int_round(t):
    return int(np.floor(t + 0.5))

n2 = np.arange(N2) # номера отсчетов воздействия 0..N2-1
L_imp = int_round(N2 / 2) # число единиц в импульсе
x = np.where(n2 < L_imp, 1.0, 0.0) # прямоугольный импульс

y1_full = np.convolve(h, x) # полная свертка
n_full = np.arange(len(y1_full))
last_nz = np.nonzero(np.abs(y1_full) > 1e-12)[0][-1] # последний ненулевой отсчет

print(f"Длина массива воздействия: N2={len(x)}")
print(f"Длина полной свертки: {len(y1_full)} (N1+N2-1={N1 + N2 - 1})")
print(f"Последний ненулевой отсчет: n={last_nz}, ненулевых отсчетов {last_nz + 1} (N1+{L_imp}-1={N1 + L_imp - 1})")

y1 = y1_full[:N2] # ограничиваем длиной воздействия

fig, ax = plt.subplots(3, 1, figsize=(10, 9))
ax[0].stem(n2, x, linefmt=C1, basefmt="k-")
ax[0].set_title("Воздействие x(n): прямоугольный импульс")
ax[1].stem(n_full, y1_full, linefmt=C2, basefmt="k-")
ax[1].axvline(last_nz, ls="--", c="gray")
ax[1].set_title(f"Полная свертка, длина {len(y1_full)}; пунктир - последний ненулевой отсчет")
ax[2].stem(n2, y1, linefmt=C2, basefmt="k-")
ax[2].set_title(f"Реакция y1(n), ограниченная длиной воздействия N2 = {N2}")
for a_ in ax:
    a_.set_xlabel("n")
plt.tight_layout(); plt.show()
```

Реакция не повторяет форму импульса: наибольшие выбросы возникают на **фронтах** ($n = 0 \dots 2$ и $n = 15 \dots 17$), на участке постоянного воздействия реакция колеблется вокруг $\approx 0.16 = H(1)$, а после окончания импульса затухающе колеблется, как ИХ. Звено подавляет постоянную (низкочастотную) составляющую и выделяет высокочастотные составляющие на фронтах, что согласуется с п. 1.

#### 3. Вычислим **реакцию $y_2(n)$ по разностному уравнению**. Воздействие $x(n)$ - прямоугольный импульс из п. 2. Сравним **графики реакций $y_2(n)$ и $y_1(n)$**. $$y(n) = 0.7\,x(n) - 0.81214\,x(n-1) + 0.56\,x(n-2) - 1.0278\,y(n-1) - 0.7\,y(n-2).$$

Вычисляем функцией `lfilter(b, a, x)` и для проверки вручную циклом.

```python
def diff_eq(b, a, x):
    # прямая реализация разностного уравнения 2-го порядка
    y = np.zeros(len(x))
    for n in range(len(x)):
        xn1 = x[n-1] if n >= 1 else 0.0 # нулевые начальные условия
        xn2 = x[n-2] if n >= 2 else 0.0
        yn1 = y[n-1] if n >= 1 else 0.0
        yn2 = y[n-2] if n >= 2 else 0.0
        y[n] = b[0]*x[n] + b[1]*xn1 + b[2]*xn2 - a[1]*yn1 - a[2]*yn2
    return y

y2 = sig.lfilter(b, a, x) # реакция по разностному уравнению
print("Цикл и lfilter совпадают:", np.allclose(y2, diff_eq(b, a, x)))
print(f"Длина воздействия: {len(x)}, длина реакции: {len(y2)}")

d = np.abs(y1 - y2) # расхождение свертки и разностного уравнения
first = np.nonzero(d > 1e-12)[0][0]
print(f"y1 и y2 совпадают при n=0...{first - 1}")
print(f"Максимальное расхождение при n >= {first}: {d.max():.4f}")

fig, ax = plt.subplots(3, 1, figsize=(10, 9))
ax[0].stem(n2, x, linefmt=C1, basefmt="k-")
ax[0].set_title("Воздействие x(n)")
ax[1].stem(n2, y2, linefmt=C2, basefmt="k-")
ax[1].set_title("Реакция y2(n) по разностному уравнению")
ax[2].plot(n2, y1, "o-", c=C1, ms=4, label="y1 - свертка с усеченной ИХ")
ax[2].plot(n2, y2, "s--", c=C2, ms=4, label="y2 - разностное уравнение")
ax[2].axvline(N1 - 0.5, ls=":", c="gray")
ax[2].text(N1, 0.9, " n = N1 = 20", color="gray")
ax[2].set_title("Сравнение y1(n) и y2(n)")
ax[2].legend(loc="lower right")
for a_ in ax:
    a_.set_xlabel("n")
plt.tight_layout(); plt.show()
```

Форма $y_2(n)$ такая же, как у $y_1(n)$ из п. 2.

При $n = 0 \dots 19$ реакции совпадают точно: для свертки нужны $h(0) \dots h(n)$, и при $n < N_1$ они все есть.При $n \ge 20$ появляется расхождение (до $\approx 0.055$), т. к. свертке нужны отброшенные отсчеты $h(20), h(21), \dots$ Разностное уравнение от усечения ИХ не зависит, поэтому $y_2(n)$ - точная реакция, а $y_1(n)$ - ее приближение.

Длина воздействия и длина реакции равны $N_2 = 30$: `lfilter` дает один выходной отсчет на каждый входной. Истинная реакция рекурсивного звена бесконечна (после импульса звено продолжает колебаться), 30 - это интервал вычисления.


#### 4. Вычислим **нули и полюсы передаточной функции** в алгебраической и показательной формах, поясним связь между ними. Выразим **значение аргумента полюса и нуля относительно $\pi$**. Представим **передаточную функцию в виде произведения простейших множителей с нулями и полюсами** в показательной форме.

Домножим числитель и знаменатель $H(z)$ на $z^2$:
$$H(z) = \frac{0.7 z^2 - 0.81214 z + 0.56}{z^2 + 1.0278 z + 0.7}.$$
Нули - корни числителя, полюсы - корни знаменателя. Находим их функцией `np.roots`.

```python
zeros = np.roots(b) # корни числителя - нули
poles = np.roots(a) # корни знаменателя - полюсы

def show(name, zs):
    for z in sorted(zs, key=lambda c: -c.imag):
        print(f"{name}: {z.real:+.4f} {z.imag:+.4f}j={abs(z):.4f}*exp(j*{np.angle(z)/np.pi:+.4f}π)")

print("Нули:"); show("z0", zeros)
print("Полюсы:"); show("z*", poles)

# проверка соотношений из теории
print(f"\n|z*|^2={abs(poles[0])**2:.4f}, a2={a2:.4f}")
print(f"|z0|^2={abs(zeros[0])**2:.4f}, b2/b0={b2/b0:.4f}")

# проверка произведения множителей в произвольной точке z
z_t = 0.3 + 1.1j
H_orig = np.polyval(b, z_t) / np.polyval(a, z_t)
H_prod = b0 * np.prod(1 - zeros / z_t) / np.prod(1 - poles / z_t)
print(f"\nH(z) исходная: {H_orig:.6f}, произведением: {H_prod:.6f}")
```

**Связь алгебраической и показательной форм:**

$z = \text{Re}\,z + j\,\text{Im}\,z = r e^{j\varphi}$, где $r = |z|$, $\varphi = \arg z$ (формула Эйлера).

Нули и полюсы - комплексно-сопряженные пары, т. к. коэффициенты $H(z)$ вещественные.

**Аргументы относительно $\pi$:** нули $\pm 0.2754\pi$, полюсы $\pm 0.7105\pi$.

**Произведение простейших множителей:**
$$H(z) = 0.7\,\frac{\left(1 - 0.8944\,e^{j0.2754\pi} z^{-1}\right)\left(1 - 0.8944\,e^{-j0.2754\pi} z^{-1}\right)}{\left(1 - 0.8367\,e^{j0.7105\pi} z^{-1}\right)\left(1 - 0.8367\,e^{-j0.7105\pi} z^{-1}\right)}.$$

#### 4. Вычислим **ДПФ конечной последовательности** $x(n)$ длины $N$. Выведем графики в шкале дискретных нормированных частот модуля ДПФ конечной последовательности и амплитудного спектра периодической последовательности. Поясним связь модуля ДПФ конечной последовательности с амплитудным спектром периодической последовательности.

Конечная последовательность - один период $x(n)$, $n = 0 \dots N-1$, вне этого интервала нули. Ее спектральная плотность $X(e^{j\hat\omega})$ - непрерывная функция частоты, а ДПФ - ее $N$ отсчетов на частотах $\hat\omega_k = 2\pi k/N$.

```python
x_fin = x2.copy() # конечная последовательность длины N
X_fin = np.fft.fft(x_fin) # ДПФ конечной последовательности

# спектральная плотность: ДПФ с дополнением нулями до 4096 точек
M = 4096
S = np.abs(np.fft.fft(x_fin, M))
k_dense = np.arange(M) * N / M # та же шкала k, но с мелким шагом

print("Модуль ДПФ конечной = амплитудный спектр периодической:", np.allclose(np.abs(X_fin), amp))

fig, ax = plt.subplots(2, 1, figsize=(10, 7), sharex=True)
ax[0].stem(k, amp, linefmt=C1, basefmt="k-", label="амплитудный спектр периодической")
ax[0].plot(k, np.abs(X_fin), "x", c=C2, ms=8, mew=2, label="модуль ДПФ конечной")
ax[0].set_title("Модуль ДПФ конечной и амплитудный спектр периодической последовательности")
ax[0].legend()
ax[1].plot(k_dense, S, c="gray", lw=1, label="спектральная плотность |X(e^jω)|")
ax[1].plot(k, np.abs(X_fin), "o", c=C2, ms=5, label="ДПФ - отсчеты плотности")
ax[1].set_title("Спектральная плотность конечной последовательности и ее отсчеты")
ax[1].set_xlabel("k"); ax[1].legend()
plt.tight_layout(); plt.show()
```

Модуль ДПФ конечной последовательности **совпадает** с амплитудным спектром периодической (проверка дала `True`, крестики ложатся на линии). Формулы ДПФ для конечной последовательности и для одного периода периодической одинаковы, а периодическая последовательность - это периодическое продолжение конечной.

У периодической последовательности спектр дискретный: существуют только частоты $k\,\Delta f$. У конечной спектр непрерывный (серая кривая), с главными лепестками на частотах гармоник и боковыми между ними. ДПФ берет отсчеты этой кривой в точках $k = 0 \dots N-1$. Т. к. $f_1$ и $f_2$ кратны $\Delta f$, отсчеты попадают точно на вершины главных лепестков и в нули боковых, поэтому ДПФ выглядит как 4 чистые линии.

#### 5. Восстановим **периодический аналоговый сигнал $x(t)$ по отсчетам ДПФ** $X(k)$.

$$x(t) = \frac{1}{N}\sum_{k=-N/2}^{N/2-1} X_а(k)\, e^{j\frac{2\pi}{NT}kt}, \qquad X_а(k) = \begin{cases} X(N+k), & -N/2 \le k \le -1 \\ X(k), & 0 \le k \le N/2-1 \end{cases}$$

Время $t \in [0;\,(N-1)T]$ с шагом $\Delta t = 0.25T$, т. е. между соседними отсчетами $x(n)$ вычисляем еще 3 промежуточных значения.

```python
ka = np.arange(-N//2, N//2) # k от -N/2 до N/2-1
Xa = X2[ka % N] # X(N+k) при k<0, X(k) при k>=0

t = np.arange(0, (N-1)*T + T/8, 0.25*T) # t от 0 до (N-1)T с шагом 0.25T
x_t = (np.exp(1j * 2*np.pi * np.outer(t, ka) / (N*T)) @ Xa).real / N # восстановленный

x_orig = A1*np.cos(2*np.pi*f1*t + np.pi/4) + A2*np.cos(2*np.pi*f2*t + np.pi/8) # исходный

fa = ka * df # частоты от -f_д/2 до f_д/2 - Δf
amp_a = np.abs(Xa) / N # амплитудный спектр аналогового сигнала

print(f"Число точек t: {len(t)}, шаг {0.25*T*1e3} мс")
print("Макс. расхождение восстановленного и исходного:", np.max(np.abs(x_t - x_orig)))

# a. периодическая последовательность и модуль ДПФ
fig, ax = plt.subplots(2, 1, figsize=(10, 7))
ax[0].stem(n, x2, linefmt=C1, basefmt="k-")
ax[0].set_title("Периодическая последовательность x(n)"); ax[0].set_xlabel("n")
ax[1].stem(k, amp, linefmt=C1, basefmt="k-")
ax[1].set_title("Модуль ДПФ |X(k)|"); ax[1].set_xlabel("k")
plt.tight_layout(); plt.show()

# b. восстановленный сигнал и его амплитудный спектр
fig, ax = plt.subplots(2, 1, figsize=(10, 7))
ax[0].plot(t*1e3, x_t, c=C2, lw=1.5, label="восстановленный x(t)")
ax[0].plot(nT*1e3, x2, "o", c=C1, ms=4, label="отсчеты x(nT)")
ax[0].set_title("Восстановленный аналоговый сигнал x(t)"); ax[0].set_xlabel("t, мс"); ax[0].legend(loc="lower right")
ax[1].stem(fa, amp_a, linefmt=C2, basefmt="k-")
ax[1].set_title("Амплитудный спектр восстановленного сигнала |Xa(k)|/N"); ax[1].set_xlabel("f, Гц")
plt.tight_layout(); plt.show()

# c. исходный аналоговый сигнал
plt.plot(t*1e3, x_orig, c=C1, lw=1.5)
plt.title("Исходный аналоговый сигнал x(t) = x(nT) при nT = t")
plt.xlabel("t, мс")
plt.show()
```

**a.** Последовательность $x(n)$ и модуль ее ДПФ - те же, что в п. 1 и п. 3.

**b.** Восстановленный сигнал - непрерывная кривая, проходящая через все отсчеты $x(nT)$. Его амплитудный спектр построен в диапазоне $[-f_д/2;\, f_д/2)$ и содержит 4 линии: $\pm 250$ Гц высотой $A_1/2 = 0.55$ и $\pm 375$ Гц высотой $A_2/2 = 1.1$. Каждая гармоника $A\cos(\cdot)$ раскладывается на две экспоненты с амплитудой $A/2$ на частотах $\pm f$.

**c.** Исходный аналоговый сигнал получен подстановкой $nT = t$ в формулу $x(nT)$.

#### 6. Проанализируем **результаты п. 5.**

```python
fig, ax = plt.subplots(2, 1, figsize=(10, 7))
ax[0].plot(t*1e3, x_orig, c=C1, lw=3, alpha=0.5, label="исходный")
ax[0].plot(t*1e3, x_t, "--", c=C2, lw=1.5, label="восстановленный")
ax[0].set_title("Сравнение исходного и восстановленного сигналов")
ax[0].set_xlabel("t, мс"); ax[0].legend(loc="lower right")
ax[1].plot(t*1e3, x_t - x_orig, c="gray")
ax[1].set_title("Разность восстановленного и исходного сигналов")
ax[1].set_xlabel("t, мс")
plt.tight_layout(); plt.show()
```

**a. Связь модуля ДПФ со спектром аналогового сигнала.** Амплитудный спектр аналогового сигнала равен модулю ДПФ, деленному на $N$, в основной полосе $[-f_д/2;\, f_д/2)$:
$$|X_а(k)|/N, \quad f = k\,\Delta f.$$
Отсчеты ДПФ $k = 0 \dots N/2-1$ соответствуют положительным частотам, $k = N/2 \dots N-1$ - отрицательным частотам $(k-N)\Delta f$. Например, линия $|X(8)| = 35.2$ дает $35.2/64 = 0.55 = A_1/2$ на частоте 250 Гц, а $|X(56)|$ - ту же амплитуду на частоте $-250$ Гц.

**b. Сравнение сигналов.** Восстановленный и исходный сигналы визуально совпадают, разность порядка $10^{-14}$ (погрешность вычислений). Восстановление точное, т. к. выполнено условие теоремы Котельникова: $f_{max} = f_2 = 375$ Гц $< f_д/2 = 1000$ Гц, и частоты гармоник кратны $\Delta f$, поэтому спектр полностью описывается отсчетами ДПФ.

#### 7. Проверим **равенство Парсеваля**. Поясним его смысл.

$$\sum_{n=0}^{N-1} |x(n)|^2 = \frac{1}{N}\sum_{k=0}^{N-1} |X(k)|^2.$$

```python
E_time = np.sum(np.abs(x2)**2) # энергия во временной области
E_freq = np.sum(np.abs(X2)**2) / N # энергия в частотной области
E_theor = N * (A1**2 / 2 + A2**2 / 2) # теоретическое значение

print(f"Во временной области: {E_time:.4f}")
print(f"В частотной области:  {E_freq:.4f}")
print(f"Теоретически N(A1²/2 + A2²/2): {E_theor:.4f}")
print("Равенство выполняется:", np.isclose(E_time, E_freq))
```

Обе части равны $193.6$, равенство Парсеваля выполняется.

Энергия последовательности на периоде одинакова, считать ли ее во временной области (по отсчетам $x(n)$) или в частотной (по отсчетам $X(k)$). ДПФ не теряет и не добавляет энергию, множитель $1/N$ связан только с нормировкой ДПФ.

Для нашего сигнала энергия распределена по 4 спектральным линиям: $\frac{1}{N}(2 \cdot 35.2^2 + 2 \cdot 70.4^2) = 193.6$. Это совпадает с теоретическим значением $N(A_1^2/2 + A_2^2/2)$: энергия суммы гармоник равна сумме их энергий. Вторая гармоника вдвое больше по амплитуде, поэтому несет в 4 раза больше энергии.
