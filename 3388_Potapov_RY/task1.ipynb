## Основные теоретические положения

**Линейная дискретная система (ЛДС)** второго порядка с нулевыми начальными условиями описывается разностным уравнением

$$
y[n]=b_0x[n]+b_1x[n-1]+b_2x[n-2]-a_1y[n-1]-a_2y[n-2],
$$

где $x[n]$ — входная последовательность, $y[n]$ — реакция системы. Её **передаточная функция** при $a_0=1$ имеет вид

$$
H(z)=\frac{Y(z)}{X(z)}=
\frac{b_0+b_1z^{-1}+b_2z^{-2}}{1+a_1z^{-1}+a_2z^{-2}}.
$$

**Импульсная характеристика** $h[n]$ — реакция ЛДС на единичный дискретный импульс $\delta[n]$ при нулевых начальных условиях. Для звена с двумя различными комплексно-сопряжёнными полюсами $p_{1,2}=r e^{\pm j\theta}$ обратное $Z$-преобразование позволяет записать

$$
H(z)=K+\frac{R_1}{1-p_1z^{-1}}+\frac{R_2}{1-p_2z^{-1}},
\qquad
h[n]=K\delta[n]+2|R_1|r^n\cos(n\theta+\arg R_1)u[n],
$$

где $u[n]$ — единичная ступенчатая последовательность, $R_2=\overline{R_1}$. В данном случае импульсная характеристика имеет **бесконечную длительность** (БИХ — *бесконечная импульсная характеристика*); при построении графиков используется конечное число отсчётов.

**Линейная свёртка** определяет реакцию ЛДС на произвольное воздействие:

$$
y[n]=(x*h)[n]=\sum_{m=-\infty}^{+\infty}x[m]h[n-m].
$$

При свёртке *двух конечных массивов* длиной $N_x$ и $N_h$ полный результат имеет длину $N_x+N_h-1$. Если импульсная характеристика бесконечна, то реакция на конечное воздействие также может продолжаться неограниченно долго. Поэтому свёртка с усечённой импульсной характеристикой является приближением.

**Нули** $z_k$ передаточной функции — корни её числителя, **полюсы** $p_k$ — корни знаменателя. Для рассматриваемой причинной дискретной системы устойчивость обеспечивается расположением всех полюсов передаточной функции внутри единичной окружности: $|p_k|<1$. Алгебраическая и показательная формы комплексного числа связаны равенством

$$
z=x+jy=r e^{j\varphi},\qquad
r=\sqrt{x^2+y^2},\qquad \varphi=\operatorname{atan2}(y,x).
$$

**Частотная характеристика** получается подстановкой $z=e^{j\hat\omega}$:

$$
H(e^{j\hat\omega})=
\frac{b_0+b_1e^{-j\hat\omega}+b_2e^{-2j\hat\omega}}
{1+a_1e^{-j\hat\omega}+a_2e^{-2j\hat\omega}},
\quad
A(\hat\omega)=|H(e^{j\hat\omega})|,
\quad
\Phi(\hat\omega)=\arg H(e^{j\hat\omega}).
$$

Здесь $A$ — **АЧХ**, $\Phi$ — **ФЧХ**. В силу периодичности с периодом $2\pi$ и сопряжённой симметрии для вещественных коэффициентов достаточно положительной половины основной полосы $\hat\omega\in[0,\pi]$; полная основная полоса может быть задана как $[-\pi,\pi]$. Переход к абсолютной частоте:

$$
\hat\omega=\frac{2\pi f}{f_{\mathrm d}},\qquad
f=\frac{\hat\omega f_{\mathrm d}}{2\pi}.
$$

**Дискретное преобразование Фурье (ДПФ)** конечной последовательности из $N$ отсчётов и **обратное ДПФ (ОДПФ)** задаются формулами

$$
X[k]=\sum_{n=0}^{N-1}x[n]e^{-j2\pi kn/N},\qquad
x[n]=\frac{1}{N}\sum_{k=0}^{N-1}X[k]e^{j2\pi kn/N}.
$$

Дискретное нормированное время на периоде $NT$ обозначается $n/N=t_n/(NT)$, где $t_n=nT$. Дискретная нормированная частота $k/N$ связана с абсолютной частотой $f_k=kf_{\mathrm d}/N$; индексы $k>N/2$ описывают отрицательные частоты $f_k=(k-N)f_{\mathrm d}/N$.

Для периодического продолжения последовательности коэффициенты **дискретного ряда Фурье** равны $C[k]=X[k]/N$. Его двухсторонний амплитудный спектр представляет собой $|C[k]|=|X[k]|/N$, а для ненулевых гармоник с сопряжёнными частотами односторонние физические амплитуды равны $2|X[k]|/N$ (при $0<k<N/2$). Фазовый спектр описывается $\arg X[k]$; фазы отсчётов с нулевой амплитудой не определены.

**Восстановление аналогового периодического сигнала** по отсчётам ДПФ для чётного $N$ выполняется по формуле

$$
x_{\mathrm{восст}}(t)=\frac{1}{N}\sum_{k=-N/2}^{N/2-1}X_a[k]\exp\!\left(j\frac{2\pi}{NT}kt\right),\quad
X_a[k]=\begin{cases}
X[N+k],&-N/2\le k\le-1,\\
X[k],&0\le k\le N/2-1.
\end{cases}
$$

**Равенство Парсеваля** для ДПФ в принятой нормировке устанавливает равенство энергии за $N$ отсчётов во временной и частотной областях:

$$
\sum_{n=0}^{N-1}|x[n]|^2=\frac{1}{N}\sum_{k=0}^{N-1}|X[k]|^2.
$$

### Часть 1. Исходные данные

Параметры вычисляются по таблице вариантов практической работы №1.

| Параметр | Формула | Значение |
|:--|:--|--:|
| $b_0$ | $0.5+0.02N_{\mathrm{вар}}$ | $0.74$ |
| $b_1$ | $b_0(-1)^{N_{\mathrm{вар}}+1}(0.9822+0.0178N_{\mathrm{вар}})$ | $-0.884892$ |

| $f_{\mathrm d}$ | $1000N_{\mathrm{вар}}$ | $12000$ Гц |

Передаточная функция исследуемого фильтра:

$$
\boxed{H(z)=\frac{0.74-0.884892z^{-1}+0.888z^{-2}}
{1+1.0778z^{-1}+0.712z^{-2}}}.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

# Расчёт исходных параметров по формулам из таблицы вариантов.
N_var = 12
b0 = 0.5 + 0.02 * N_var
b1 = b0 * (-1) ** (N_var + 1) * (0.9822 + 0.0178 * N_var)
b2 = b0 * (0.8 + 0.2 * (N_var % 5))
a0 = 1.0
a1 = (-1) ** N_var * (0.7778 + 0.025 * N_var)
a2 = 0.64 + 0.006 * N_var
N1 = N_var % 10 + 20
N2 = N_var % 10 + 30
f_d = 1000 * N_var

b = np.array([b0, b1, b2], dtype=float)
a = np.array([a0, a1, a2], dtype=float)

print(f'Вариант: {N_var}')
print('b =', b, '; a =', a)
print(f'N1 = {N1}, N2 = {N2}, f_d = {f_d} Гц')
assert np.allclose(b, [0.74, -0.884892, 0.888])
assert np.allclose(a, [1.0, 1.0778, 0.712])
```

### 1. Импульсная характеристика

Требуется вычислить первые $N_1=22$ отсчёта $h[n]$ и построить их график. Входной сигнал $x[n]=\delta[n]$ имеет вид $\delta[0]=1$ и $\delta[n]=0$ при $n\neq0$; начальные условия $h[-1]=h[-2]=0$.

Из разностного уравнения при воздействии $\delta[n]$:

$$
h[n]=0.74\delta[n]-0.884892\delta[n-1]+0.888\delta[n-2]
-1.0778h[n-1]-0.712h[n-2].
$$

Для записи **аналитической формулы** используем разложение $H(z)$ на простейшие дроби в форме $Z^{-1}$:

$$
H(z)=K+\frac{R_1}{1-p_1z^{-1}}+\frac{R_2}{1-p_2z^{-1}}.
$$

После вычисления вычетов и полюсов получаем (коэффициенты округлены до шести знаков):

$$
\boxed{h[n]=1.247191\,\delta[n]
+3.054566\cdot(0.843801)^n
\cos(2.263549n+1.737613)\,u[n]}
$$

Данная формула справедлива для всех $n\geqslant0$; ниже проверяется её совпадение с результатом расчёта разностного уравнения.

```python
# Реакция на единичный импульс при нулевых начальных условиях.
n_h = np.arange(N1)
delta = np.zeros(N1)
delta[0] = 1.0
h_N1 = signal.lfilter(b, a, delta)

# Разложение H(z) на простейшие дроби по z^(-1).
residues, poles_res, direct = signal.residuez(b, a)
pos = np.flatnonzero(poles_res.imag > 0)[0]
R1, p1 = residues[pos], poles_res[pos]
r_h, theta_h = np.abs(p1), np.angle(p1)
A_h, psi_h = 2 * np.abs(R1), np.angle(R1)
K_h = float(np.real(direct[0]))

h_analytical = (K_h * (n_h == 0)
                + A_h * r_h**n_h * np.cos(n_h * theta_h + psi_h))
max_error_h = np.max(np.abs(h_N1 - h_analytical))
print(f'K = {K_h:.9f}; A = {A_h:.9f}; r = {r_h:.9f}')
print(f'Угол полюса = {theta_h:.9f} рад; фаза вычета = {psi_h:.9f} рад')
print(f'Максимальное расхождение двух формул: {max_error_h:.18f}')
assert max_error_h < 1e-11

fig, ax = plt.subplots(figsize=(10, 4))
ax.stem(n_h, h_N1, basefmt='k-')
ax.set(title='Импульсная характеристика (первые 22 отсчёта)',
       xlabel='n, номер отсчёта', ylabel='h[n]', xlim=(-0.5, N1 - 0.5))
ax.set_xticks(np.arange(0, N1, 2))
ax.grid(True, alpha=0.35)
plt.tight_layout()
plt.show()
```

**Интерпретация.** На графике приведены $N_1=22$ отсчёта, но действительная длина импульсной характеристики **бесконечна**, поскольку рекурсивная составляющая с ненулевыми полюсами не обрывается после 22-го отсчёта. Модуль полюсов равен $r\approx0.843801<1$, поэтому колебания затухают, а не возрастают. Максимальное расхождение аналитической формулы и программного расчёта составляет примерно $3.719\cdot10^{-15}$ (в программе оно выводится десятичной дробью), что соответствует погрешности вычислений с плавающей точкой.

### 2. Реакция $y_1[n]$ по формуле свёртки

Воздействие длины $N_2=32$ задаётся прямоугольным импульсом:

$$
x[n]=\begin{cases}
1, & 0\le n<16,\\
0, & 16\le n\le31.
\end{cases}
$$

В условии задания граница ненулевой части задана выражением $\operatorname{int}(N_2/2)$. При $N_2=32$ она равна $16$. Длина **всей последовательности** $x[n]$ составляет $32$ отсчёта, длительность **ненулевой части** — $16$ отсчётов.

Реакция вычисляется по формуле

$$
y_1[n]=\sum_{m=0}^{n}x[m]h[n-m],\qquad n\geqslant0,
$$

где $x[m]=0$ за пределами $0\leqslant m<32$.

Для **полной свёртки двух заданных конечных массивов** $x[n]$ длины $N_2=32$ и $h[n]$ длины $N_1=22$ получается

$$
N_{\mathrm{св}}=N_1+N_2-1=22+32-1=\boxed{53}.
$$

Однако для точного вычисления первых $N_2=32$ отсчётов выходного сигнала одной импульсной характеристики длины $22$ недостаточно: значение $y_1[31]$ включает $h[31]x[0]$. Поэтому **дополнительно** рассчитываются первые $32$ отсчёта ИХ, после чего свёртка ограничивается интервалом $n=0,\ldots,31$. Это позволяет корректно сравнить $y_1[n]$ с разностным уравнением из пункта 3.

```python
# Воздействие длины N2: 16 единиц, затем 16 нулей.
n_x = np.arange(N2)
nonzero_length = N2 // 2  # N2 = 32 — чётное число
x = np.zeros(N2)
x[:nonzero_length] = 1.0

# Полная свёртка двух конечных массивов из пп. 1 и 2 (длина 53).
y1_full_53 = np.convolve(x, h_N1, mode='full')

# Для правильных первых 32 отсчётов требуется h[0], ..., h[31].
delta_N2 = np.zeros(N2)
delta_N2[0] = 1.0
h_N2 = signal.lfilter(b, a, delta_N2)
y1 = np.convolve(x, h_N2, mode='full')[:N2]

print(f'Длина x: {len(x)}; ненулевых отсчётов x: {nonzero_length}')
print(f'Длина полной свёртки x * h_N1: {len(y1_full_53)} (расчёт: {N2} + {N1} - 1 = {N1+N2-1})')
print(f'Длина вычисляемой реакции y1: {len(y1)}')
print(f'Влияние усечения h до N1 при n < N2: '
      f'{np.max(np.abs(y1_full_53[:N2] - y1)):.6f}')
assert len(y1_full_53) == 53
assert len(y1) == N2

fig, axes = plt.subplots(3, 1, figsize=(10, 10))
axes[0].stem(n_x, x, basefmt='k-')
axes[0].set(title='Входное воздействие', ylabel='x[n]', xlim=(-0.5, 32))
axes[1].stem(np.arange(len(y1_full_53)), y1_full_53, basefmt='k-')
axes[1].set(title='Полная свёртка x * h_N1 (53 отсчёта)', ylabel='Значение свёртки',
            xlim=(-0.5, 52.5))
axes[2].stem(n_x, y1, basefmt='k-')
axes[2].set(title='Реакция y1[n] по свёртке (первые 32 точных отсчёта)',
            xlabel='n, номер отсчёта', ylabel='y1[n]', xlim=(-0.5, 31.5))
for ax in axes:
    ax.grid(True, alpha=0.35)
plt.tight_layout()
plt.show()
```

**Интерпретация.** На втором графике показаны $53$ отсчёта полной свёртки **двух конечных массивов**, что совпадает с аналитическим расчётом $22+32-1=53$. На третьем графике показаны $32$ отсчёта реакции с продлённой до $32$ отсчётов ИХ.

Ограничение реакции длиной воздействия $N_2=32$ принято для сопоставления двух способов расчёта на одинаковом временном интервале. **Оно не означает, что реакция становится равной нулю после $n=31$**: у рассматриваемого фильтра затухающие колебания продолжаются и после окончания прямоугольного импульса. Конечная свёртка длины $53$ также не описывает бесконечный хвост истинной реакции.

### 3. Реакция $y_2[n]$ по разностному уравнению

Разностное уравнение с найденными коэффициентами имеет вид

$$
\boxed{y_2[n]=0.74x[n]-0.884892x[n-1]+0.888x[n-2]
-1.0778y_2[n-1]-0.712y_2[n-2]}.
$$

Используется то же воздействие $x[n]$, что в пункте 2, и **нулевые начальные условия**: $x[n]=0$, $y_2[n]=0$ при $n<0$. Реакция рассчитывается последовательно для $n=0,\dots,31$.

```python
# Прямое вычисление реакции по разностному уравнению, без lfilter.
y2 = np.zeros(N2)
for n in range(N2):
    xn_1 = x[n - 1] if n >= 1 else 0.0
    xn_2 = x[n - 2] if n >= 2 else 0.0
    yn_1 = y2[n - 1] if n >= 1 else 0.0
    yn_2 = y2[n - 2] if n >= 2 else 0.0
    y2[n] = (b0 * x[n] + b1 * xn_1 + b2 * xn_2
             - a1 * yn_1 - a2 * yn_2)

difference = np.max(np.abs(y1 - y2))
print(f'Длина воздействия: {len(x)}; длина рассчитанной реакции: {len(y2)}')
print(f'Максимальное отличие свёртки от разностного уравнения: {difference:.18f}')
assert difference < 1e-11

fig, axes = plt.subplots(2, 1, figsize=(10, 7), sharex=True)
axes[0].stem(n_x, x, basefmt='k-')
axes[0].set(ylabel='x[n]', title='Входное воздействие')
axes[1].plot(n_x, y1, 'o-', markersize=4, label='y1[n] — свёртка')
axes[1].plot(n_x, y2, 'x--', markersize=5, label='y2[n] — разностное уравнение')
axes[1].set(xlabel='n, номер отсчёта', ylabel='Реакция',
            title='Сравнение двух способов вычисления реакции')
axes[1].legend()
for ax in axes:
    ax.grid(True, alpha=0.35)
plt.tight_layout()
plt.show()
```

**Интерпретация.** Последовательности $y_1[n]$ и $y_2[n]$ на интервале $0\leqslant n<32$ совпадают с точностью вычислений. Это подтверждает эквивалентность свёртки с полной для данного интервала импульсной характеристикой и решения разностного уравнения. Длина входного воздействия — $32$ отсчёта; длина **вычисленной** реакции — также $32$ отсчёта, а ненулевой выходной сигнал может продолжаться после окончания воздействия.

### 4. Нули и полюсы передаточной функции

Для вычисления нулей и полюсов числитель и знаменатель передаточной функции умножаются на $z^2$:

$$
H(z)=\frac{0.74z^2-0.884892z+0.888}{z^2+1.0778z+0.712}.
$$

**Нули** — корни числителя, **полюсы** — корни знаменателя. Получаются две пары комплексно-сопряжённых корней (значения округлены до шести знаков).

**Нули — алгебраическая и показательная формы:**

$$
\begin{aligned}
z_1&=0.597900+j0.917886=1.095445\,e^{\,j0.316224\pi},\\
z_2&=0.597900-j0.917886=1.095445\,e^{-j0.316224\pi}.
\end{aligned}
$$

**Полюсы — алгебраическая и показательная формы:**

$$
\begin{aligned}
p_1&=-0.538900+j0.649297=0.843801\,e^{\,j0.720510\pi},\\
p_2&=-0.538900-j0.649297=0.843801\,e^{-j0.720510\pi}.
\end{aligned}
$$

Здесь $x+jy=|z|e^{j\arg z}$ по формуле Эйлера, аргументы выражены в долях $\pi$.

**Представление передаточной функции произведением простейших множителей:**

$$
\boxed{
H(z)=0.74\,
\frac{
(1-1.095445e^{\,j0.316224\pi}z^{-1})
(1-1.095445e^{-j0.316224\pi}z^{-1})
}{
(1-0.843801e^{\,j0.720510\pi}z^{-1})
(1-0.843801e^{-j0.720510\pi}z^{-1})
}}
$$

Коэффициенты в последней формуле округлены; численно в коде используются неокруглённые значения корней.

```python
zeros, poles, gain = signal.tf2zpk(b, a)
# Упорядочиваем: сначала корень с положительной мнимой частью.
zeros = zeros[np.argsort(-zeros.imag)]
poles = poles[np.argsort(-poles.imag)]

for label, roots in [('Нули', zeros), ('Полюсы', poles)]:
    print(label)
    for i, root in enumerate(roots, 1):
        print(f'  {i}: {root.real:+.9f}{root.imag:+.9f}j; '
              f'|z| = {np.abs(root):.9f}; '
              f'arg = {np.angle(root):+.9f} рад = '
              f'{np.angle(root) / np.pi:+.9f}π')
print(f'Коэффициент перед факторизованной функцией: {gain:.6f}')

# Проверка восстановления коэффициентов по вычисленным корням.
num_reconstructed = gain * np.poly(zeros)
den_reconstructed = np.poly(poles)
print('Числитель из множителей:', np.real_if_close(num_reconstructed))
print('Знаменатель из множителей:', np.real_if_close(den_reconstructed))
assert np.allclose(num_reconstructed, b)
assert np.allclose(den_reconstructed, a)
```

Координаты нулей и полюсов на карте совпадают с корнями из пункта 4.

### 6. АЧХ и ФЧХ в шкале нормированных частот

Частотная характеристика исследуется в шкале $\hat\omega\in[0,\pi]$ (положительная половина основной полосы частот) по формуле

$$
H(e^{j\hat\omega})=
\frac{0.74-0.884892e^{-j\hat\omega}+0.888e^{-2j\hat\omega}}
{1+1.0778e^{-j\hat\omega}+0.712e^{-2j\hat\omega}}.
$$

**АЧХ** равна $A(\hat\omega)=|H(e^{j\hat\omega})|$, **ФЧХ** — $\Phi(\hat\omega)=\arg H(e^{j\hat\omega})$. Для отображения непрерывного изменения фазы используется развёртка аргумента (`np.unwrap`), устраняющая искусственные скачки $2\pi$ при переходе через границу главного значения аргумента.

```python
# Равномерная сетка положительной половины основной полосы 0 ... pi.
omega = np.linspace(0, np.pi, 8193)
_, H_w = signal.freqz(b, a, worN=omega)
amplitude = np.abs(H_w)
phase = np.unwrap(np.angle(H_w))

imin = np.argmin(amplitude)
imax = np.argmax(amplitude)
print(f'Границы отображаемой полосы: 0 .. π рад/отсчёт')
print(f'Минимум АЧХ: {amplitude[imin]:.6f}, ω/π = {omega[imin]/np.pi:.6f}')
print(f'Максимум АЧХ: {amplitude[imax]:.6f}, ω/π = {omega[imax]/np.pi:.6f}')
print(f'A(0) = {amplitude[0]:.6f}; A(π) = {amplitude[-1]:.6f}')

fig, axes = plt.subplots(2, 1, figsize=(10, 8), sharex=True)
axes[0].plot(omega / np.pi, amplitude, lw=1.7, label='АЧХ')
axes[0].scatter([omega[imin]/np.pi, omega[imax]/np.pi],
                [amplitude[imin], amplitude[imax]],
                color=['tab:blue', 'tab:red'], zorder=3,
                label='Минимум и максимум')
axes[0].axvline(np.angle(zeros[0]) / np.pi, color='tab:blue', ls=':', alpha=0.6,
                label='Аргумент нуля')
axes[0].axvline(np.angle(poles[0]) / np.pi, color='tab:red', ls=':', alpha=0.6,
                label='Аргумент полюса')
axes[0].set(ylabel=r'$A(\hat\omega)$', title='Амплитудно-частотная характеристика')
axes[0].legend()
axes[1].plot(omega / np.pi, phase, lw=1.7)
axes[1].set(xlabel=r'$\hat\omega/\pi$', ylabel=r'$\Phi(\hat\omega)$, рад',
            title='Фазочастотная характеристика (развёрнутая фаза)', xlim=(0, 1))
axes[1].set_yticks([0, -np.pi, -2*np.pi], ['0', '−π', '−2π'])
for ax in axes:
    ax.grid(True, alpha=0.35)
plt.tight_layout()
plt.show()
```

**Интерпретация.** Полная основная полоса нормированных частот может быть задана интервалом $[-\pi,\pi]$; для вещественных коэффициентов достаточно графиков на $[0,\pi]$.

- Минимум АЧХ составляет примерно **$0.06095$** при $\hat\omega\approx0.3135\pi$. Он расположен вблизи аргумента нулей $0.316224\pi$: нули вызывают провал АЧХ.
- Максимум АЧХ составляет примерно **$8.81948$** при $\hat\omega\approx0.7304\pi$. Он расположен вблизи аргумента полюсов $0.720510\pi$: полюсы обусловливают усиление в этом диапазоне. Положение экстремумов и аргументов нулей/полюсов не обязано совпадать точно.
- Скачок ФЧХ на $\pi$, связанный с нулём АЧХ, **отсутствует**: на рассматриваемой полосе $A(\hat\omega)>0$, поскольку нули передаточной функции не лежат на единичной окружности. Если строить главные значения аргумента без развёртки, возможен скачок на $2\pi$ из-за выбора ветви функции $\arg$; он не указывает на нулевую АЧХ.
- $A(0)\approx0.2664$, тогда как $A(\pi)\approx3.9623$. Следовательно, **низкочастотные составляющие** воздействия преимущественно подавляются относительно высокочастотных. Система обладает выраженным усилением в верхней части рассматриваемой полосы.

### 7. АЧХ и ФЧХ в шкале частот $f$ (Гц)

Частота дискретизации равна $f_{\mathrm d}=12000$ Гц. Связь с нормированной частотой:

$$
\hat\omega=\frac{2\pi f}{12000},\qquad
f=\frac{12000\hat\omega}{2\pi}.
$$

Поэтому $\hat\omega\in[0,\pi]$ соответствует $f\in[0,6000]$ Гц. Частота Найквиста $f_{\mathrm d}/2=6000$ Гц. Графики ниже содержат те же значения АЧХ и ФЧХ, что в пункте 6, но с другой шкалой горизонтальной оси.

```python
freq_hz = omega * f_d / (2 * np.pi)

fig, axes = plt.subplots(2, 1, figsize=(10, 8), sharex=True)
axes[0].plot(freq_hz, amplitude, lw=1.7)
axes[0].scatter([freq_hz[imin], freq_hz[imax]],
                [amplitude[imin], amplitude[imax]],
                color=['tab:blue', 'tab:red'], zorder=3)
axes[0].set(ylabel='A(f)', title='Амплитудно-частотная характеристика в герцах')
axes[1].plot(freq_hz, phase, lw=1.7)
axes[1].set(xlabel='f, Гц', ylabel='Φ(f), рад',
            title='Фазочастотная характеристика в герцах', xlim=(0, f_d/2))
axes[1].set_yticks([0, -np.pi, -2*np.pi], ['0', '−π', '−2π'])
for ax in axes:
    ax.grid(True, alpha=0.35)
plt.tight_layout()
plt.show()

print(f'Основная отображаемая полоса: 0 .. {f_d / 2:.0f} Гц')
print(f'Минимум АЧХ: f ≈ {freq_hz[imin]:.2f} Гц; A ≈ {amplitude[imin]:.6f}')
print(f'Максимум АЧХ: f ≈ {freq_hz[imax]:.2f} Гц; A ≈ {amplitude[imax]:.6f}')
print(f'π/2 рад/отсчёт соответствует {f_d/4:.0f} Гц')
```

**Интерпретация.** Положительная половина основной полосы составляет $0\leqslant f\leqslant6000$ Гц; полную основную полосу можно представить как $[-6000,6000]$ Гц. Нормированным частотам $0$, $\pi/2$ и $\pi$ соответствуют $0$, $3000$ и $6000$ Гц. Минимум АЧХ наблюдается приблизительно на **$1881$ Гц**, максимум — приблизительно на **$4382$ Гц**. Графики имеют ту же форму, что и в пункте 6, потому что значения АЧХ и ФЧХ не пересчитываются: меняется только горизонтальная ось по соотношению $f=\hat\omega f_{\mathrm d}/(2\pi)$. Например, $\hat\omega=\pi/2$ соответствует $3000$ Гц.

### Часть 2. Исходные данные

Исследуется периодическая последовательность $x[n]$ длины $N=64$, состоящая из двух гармоник. Параметры вычисляются по формулам таблицы вариантов.

| Параметр | Формула | Значение |
|:--|:--|--:|
| $N$ | $64$ | $64$ |
| $f_{\mathrm d2}$ | $2000(N_{\mathrm{вар}}\bmod5+1)$ | $6000$ Гц |
| $T$ | $1/f_{\mathrm d2}$ | $1/6000\approx0.000166667$ с |
| $A_1$ | $1+0.01N_{\mathrm{вар}}$ | $1.12$ |
| $A_2$ | $2A_1$ | $2.24$ |
| $f_1$ | $f_{\mathrm d2}/8$ | $750$ Гц |
| $f_2$ | $1.5f_1$ | $1125$ Гц |

Исходный дискретный сигнал:

$$
\boxed{x[n]=1.12\cos\!\left(\frac{\pi n}{4}+\frac{\pi}{4}\right)+2.24\cos\!\left(\frac{3\pi n}{8}+\frac{\pi}{8}\right)},\qquad n=0,\ldots,63.
$$

Частотное разрешение ДПФ: $\Delta f=f_{\mathrm d2}/N=93.75$ Гц. Период сигнала $NT=64/6000$ с.

```python
# Параметры второй части (с отдельными именами, чтобы не менять расчёты части 1).
N = 64
f_d2 = 2000 * (N_var % 5 + 1)
T = 1 / f_d2
A1 = 1 + 0.01 * N_var
A2 = 2 * A1
f1 = f_d2 / 8
f2 = 1.5 * f1

delta_f = f_d2 / N
period = N * T
k1 = round(f1 / delta_f)
k2 = round(f2 / delta_f)

print(f'N = {N}; fд = {f_d2} Гц; T = {T:.9f} с')
print(f'A1 = {A1:.2f}; A2 = {A2:.2f}')
print(f'f1 = {f1:.1f} Гц; f2 = {f2:.1f} Гц; Δf = {delta_f:.2f} Гц')
print(f'Номера спектральных отсчётов: k1 = {k1}; k2 = {k2}')
assert (N, f_d2, k1, k2) == (64, 6000, 8, 12)
assert np.isclose(A1, 1.12) and np.isclose(A2, 2.24)
```

### 1. Графики периодической последовательности в двух шкалах времени

Согласно заданию, сигнал задаётся выражением

$$
x(nT)=A_1\cos\left(2\pi f_1nT+\frac{\pi}{4}\right)
+A_2\cos\left(2\pi f_2nT+\frac{\pi}{8}\right).
$$

Построены графики на одном периоде $n=0,\ldots,N-1$ в шкале **дискретного времени $n$** и **дискретного нормированного времени $n/N$**. Соответствующие моменты физического времени равны $t_n=nT$.

```python
# Отсчёты двухгармонического сигнала на одном периоде.
n2 = np.arange(N)
t_discrete = n2 * T
t_norm = n2 / N
x2 = (A1 * np.cos(2 * np.pi * f1 * t_discrete + np.pi / 4)
      + A2 * np.cos(2 * np.pi * f2 * t_discrete + np.pi / 8))

fig, axes = plt.subplots(2, 1, figsize=(11, 8))
axes[0].stem(n2, x2, basefmt='k-')
axes[0].set(title='Периодическая последовательность в шкале дискретного времени',
            xlabel='n, номер отсчёта', ylabel='x[n]', xlim=(-0.5, N-0.5))
axes[1].stem(t_norm, x2, basefmt='k-')
axes[1].set(title='Периодическая последовательность в шкале нормированного времени',
            xlabel='n/N', ylabel='x[n]', xlim=(-0.01, 1))
for ax in axes:
    ax.grid(True, alpha=0.35)
plt.tight_layout()
plt.show()

print(f'Длина последовательности: {len(x2)} отсчёта')
print(f'Количество периодов гармоник на длине N: {f1*period:.0f} и {f2*period:.0f}')
```

**Интерпретация.** На интервале из $64$ отсчётов укладываются ровно $8$ периодов первой гармоники и $12$ периодов второй. Поэтому полученная последовательность периодична с периодом $N=64$ (хотя её наименьший период составляет $16$ отсчётов). Оба графика отображают одни и те же значения, изменяется только масштаб оси времени: $n/N=t_n/(NT)$.

### 2. Обратное ДПФ от ДПФ последовательности

Прямое и обратное дискретные преобразования Фурье:

$$
X[k]=\sum_{n=0}^{N-1}x[n]e^{-j2\pi kn/N},\qquad
\widetilde{x}[n]=\frac1N\sum_{k=0}^{N-1}X[k]e^{j2\pi kn/N}.
$$

Вычисляется $\widetilde{x}[n]=\mathrm{ОДПФ}(\mathrm{ДПФ}(x[n]))$ и выводится её график по оси $n/N$. Совпадение с $x[n]$ проверяется численно.

```python
# ДПФ и ОДПФ средствами NumPy.
X2 = np.fft.fft(x2)
x2_inverse_complex = np.fft.ifft(X2)
x2_inverse = x2_inverse_complex.real
error_ifft = np.max(np.abs(x2 - x2_inverse))

fig, ax = plt.subplots(figsize=(11, 4.5))
ax.stem(t_norm, x2_inverse, basefmt='k-', label='ОДПФ(ДПФ(x[n]))')
ax.plot(t_norm, x2, 'r.', markersize=4, label='Исходная последовательность')
ax.set(title='Восстановленная последовательность после ОДПФ',
       xlabel='n/N', ylabel='x[n]', xlim=(-0.01, 1))
ax.legend()
ax.grid(True, alpha=0.35)
plt.tight_layout()
plt.show()

print(f'Максимальное отклонение от исходной последовательности: {error_ifft:.18f}')
print(f'Максимальная мнимая часть результата ОДПФ: '
      f'{np.max(np.abs(x2_inverse_complex.imag)):.18f}')
assert np.allclose(x2, x2_inverse, rtol=1e-12, atol=1e-12)
```

**Интерпретация.** ОДПФ восстанавливает исходную дискретную последовательность: расхождение находится на уровне погрешности вычислений с плавающей точкой. Мнимая часть результата, возникающая численно, пренебрежимо мала, поскольку исходный сигнал вещественный.

### 3. Амплитудный и фазовый спектры в нормированных и абсолютных частотах

Для периодической последовательности амплитудный спектр удобно задавать через коэффициенты дискретного ряда Фурье:

$$
C[k]=\frac{X[k]}{N},\qquad A[k]=|C[k]|=\frac{|X[k]|}{N},\qquad
\varphi[k]=\arg X[k].
$$

Нормированная частота $\nu_k=k/N$, абсолютная $f_k=kf_{\mathrm d2}/N$. В стандартном порядке ДПФ $k=0,\ldots,63$ отсчёты $k=52$ и $56$ отвечают отрицательным частотам $-1125$ и $-750$ Гц (то есть их представлению с переносом на $f_{\mathrm d2}-|f|$).

Фаза строится **только для спектрально значимых отсчётов**: при нулевой амплитуде фаза не имеет физического смысла, а вычисляемый `np.angle` определяется шумом округления.

```python
k_bins = np.arange(N)
nu_bins = k_bins / N
f_bins = k_bins * delta_f
spec_ampl = np.abs(X2) / N
spec_phase = np.angle(X2)

# Для нулевых (численно) спектральных отсчётов фаза не определена.
significant = np.abs(X2) > 1e-10 * np.max(np.abs(X2))
print('Спектрально значимые отсчёты (k, k/N, частота с учётом знака, амплитуда, фаза):')
for kk in np.flatnonzero(significant):
    signed_f = (kk if kk < N/2 else kk-N) * delta_f
    print(f'k={kk:2d}; k/N={kk/N:.5f}; f={signed_f:+.1f} Гц; '
          f'|X|/N={spec_ampl[kk]:.6f}; фаза={spec_phase[kk]/np.pi:+.6f}π')

fig, axes = plt.subplots(2, 2, figsize=(14, 9))
axes[0, 0].stem(nu_bins, spec_ampl, basefmt='k-')
axes[0, 0].set(title='Амплитудный спектр', xlabel='k/N', ylabel='|X[k]|/N', xlim=(-0.01, 1))
axes[0, 1].stem(f_bins, spec_ampl, basefmt='k-')
axes[0, 1].set(title='Амплитудный спектр', xlabel='f, Гц', ylabel='|X[k]|/N', xlim=(-50, f_d2))
axes[1, 0].stem(nu_bins[significant], spec_phase[significant], basefmt='k-')
axes[1, 0].set(title='Фазовый спектр (значимые линии)', xlabel='k/N',
               ylabel='φ[k], рад', xlim=(-0.01, 1))
axes[1, 1].stem(f_bins[significant], spec_phase[significant], basefmt='k-')
axes[1, 1].set(title='Фазовый спектр (значимые линии)', xlabel='f, Гц',
               ylabel='φ[k], рад', xlim=(-50, f_d2))
for ax in axes.flat:
    ax.grid(True, alpha=0.35)
for ax in axes[1]:
    ax.set_yticks([-np.pi/4, -np.pi/8, 0, np.pi/8, np.pi/4],
                  ['−π/4', '−π/8', '0', 'π/8', 'π/4'])
plt.tight_layout()
plt.show()

assert np.array_equal(np.flatnonzero(significant), [k1, k2, N-k2, N-k1])
assert np.allclose(spec_ampl[[k1, k2]], [A1/2, A2/2], atol=1e-12)
```

**Интерпретация.** По отношению $t_n=nT$, $n/N=t_n/(NT)$ определяется нормированная временная шкала; частоты связаны равенством $f_k=(k/N)f_{\mathrm d2}$ с шагом $\Delta f=93.75$ Гц.

Спектр содержит четыре ненулевые линии в двустороннем представлении: при $k=8,56$ (частоты $\pm750$ Гц) их амплитуда равна $0.56$, а при $k=12,52$ (частоты $\pm1125$ Гц) — $1.12$. Вторая гармоника имеет вдвое большую амплитуду. На положительных частотах фазы составляют $\pi/4$ и $\pi/8$, на отрицательных — соответственно $-\pi/4$ и $-\pi/8$ из-за комплексного сопряжения. В одностороннем представлении амплитуды исходных косинусоид равны $1.12$ и $2.24$.

Поскольку частоты обеих гармоник точно попадают в узлы сетки ДПФ ($k=8$ и $12$), спектрального растекания в данном случае не возникает.

### 4. ДПФ конечной последовательности и спектр периодической последовательности

Для конечной последовательности $x[n]$ длиной $N$ рассчитывается

$$
X_{\mathrm{кон}}[k]=\sum_{n=0}^{N-1}x[n]e^{-j2\pi kn/N}.
$$

Если те же $N$ отсчётов рассматриваются как один период бесконечной периодической последовательности, её комплексные коэффициенты ряда Фурье равны

$$
C[k]=\frac{1}{N}\sum_{n=0}^{N-1}x[n]e^{-j2\pi kn/N},\qquad
\boxed{|X_{\mathrm{кон}}[k]|=N|C[k]|}.
$$

Сравниваются **модуль ДПФ конечной последовательности** $|X_{\mathrm{кон}}[k]|$ и **двусторонний амплитудный спектр периодической последовательности** $|C[k]|$ в шкале $k/N$.

```python
# Вычисление ДПФ конечного массива и независимое вычисление
# коэффициентов дискретного ряда Фурье периодического продолжения.
X_finite = np.fft.fft(x2)
C_periodic = (np.sum(
    x2[:, np.newaxis] * np.exp(-2j * np.pi * np.outer(n2, k_bins) / N),
    axis=0) / N)
amp_finite = np.abs(X_finite)
amp_periodic = np.abs(C_periodic)
comparison_error = np.max(np.abs(X_finite / N - C_periodic))

fig, axes = plt.subplots(2, 1, figsize=(11, 8), sharex=True)
axes[0].stem(nu_bins, amp_finite, basefmt='k-')
axes[0].set(title='Модуль ДПФ конечной последовательности длиной N',
            ylabel='|Xкон[k]|')
axes[1].stem(nu_bins, amp_periodic, basefmt='k-')
axes[1].set(title='Амплитудный спектр периодической последовательности',
            xlabel='k/N', ylabel='|C[k]|', xlim=(-0.01, 1))
for ax in axes:
    ax.grid(True, alpha=0.35)
plt.tight_layout()
plt.show()

for kk in (k1, k2):
    print(f'k = {kk}: |Xкон| = {amp_finite[kk]:.6f}; '
          f'|C| = {amp_periodic[kk]:.6f}; отношение = '
          f'{amp_finite[kk]/amp_periodic[kk]:.6f}')
print(f'Максимальное расхождение Xкон/N и C: {comparison_error:.18f}')
assert np.allclose(X_finite/N, C_periodic, atol=1e-12)
```

**Интерпретация.** Положения всех спектральных линий совпадают. При $k=8$ получены $|X_{\mathrm{кон}}|=35.84$ и $|C|=0.56$; при $k=12$ — $71.68$ и $1.12$. В обоих случаях отношение равно $N=64$. Таким образом, амплитудный спектр периодического продолжения получается из ДПФ конечного массива делением на длину периода.

При необходимости получить амплитуды двух исходных косинусоид, следует объединить вклад каждой пары сопряжённых линий: $2|C[8]|=1.12$ и $2|C[12]|=2.24$. ДПФ конечной последовательности также можно рассматривать как отсчёты её непрерывного по частоте преобразования Фурье в $N$ равноотстоящих точках.

### 5. Восстановление периодического аналогового сигнала по ДПФ

Аналоговый сигнал восстанавливается **по заданной в методических указаниях формуле**

$$
x_{\mathrm{восст}}(t)=\frac1N\sum_{k=-N/2}^{N/2-1}X_a[k]\exp\!\left(j\frac{2\pi}{NT}kt\right),
$$

где

$$
X_a[k]=\begin{cases}
X[N+k],&-N/2\le k\le-1,\\
X[k],&0\le k\le N/2-1.
\end{cases}
$$

Время изменяется на интервале $t\in[0,(N-1)T]$ с шагом $\Delta t=0.25T$, то есть формируется $4(N-1)+1=253$ точки. Для контроля рассчитывается исходный аналоговый сигнал

$$
x_{\mathrm{исх}}(t)=1.12\cos(2\pi\cdot750t+\pi/4)+2.24\cos(2\pi\cdot1125t+\pi/8).
$$

Амплитудный спектр восстановленного аналогового сигнала изображается как **двусторонний линейчатый спектр** $|X_a[k]|/N$ на частотах $f_k=k/(NT)$.

```python
# X_a[k] по кусочной формуле задания; fftshift даёт тот же порядок.
k_signed = np.arange(-N // 2, N // 2)
X_a = np.concatenate((X2[N//2:], X2[:N//2]))
assert np.allclose(X_a, np.fft.fftshift(X2))

# Дискретизация непрерывной временной оси с шагом ровно T/4:
# t=0, T/4, ..., (N-1)T — всего 253 точки.
t_analog = np.arange(4*(N-1)+1) * (T/4)

# Прямое применение суммы из методических указаний.
exp_matrix = np.exp(1j * 2*np.pi / (N*T) * np.outer(t_analog, k_signed))
x_reconstructed_complex = (exp_matrix @ X_a) / N
x_reconstructed = np.real(x_reconstructed_complex)

# Исходная аналитическая сумма двух косинусоид на той же сетке времени.
x_analog_original = (A1 * np.cos(2*np.pi*f1*t_analog + np.pi/4)
                     + A2 * np.cos(2*np.pi*f2*t_analog + np.pi/8))

# Спектр ряда Фурье восстановленного аналогового сигнала.
f_signed = k_signed / (N*T)
analog_spectrum = np.abs(X_a) / N

fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))
axes[0].stem(t_discrete*1000, x2, basefmt='k-')
axes[0].set(title='а) Периодическая последовательность x[n]',
            xlabel='nT, мс', ylabel='x[n]')
axes[1].stem(k_bins, np.abs(X2), basefmt='k-')
axes[1].set(title='б) Модуль ДПФ периодической последовательности',
            xlabel='k, номер спектрального отсчёта', ylabel='|X[k]|')
for ax in axes:
    ax.grid(True, alpha=0.35)
plt.tight_layout()
plt.show()

fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))
axes[0].plot(t_analog*1000, x_reconstructed, lw=1.8)
axes[0].set(title='в) Восстановленный аналоговый сигнал',
            xlabel='t, мс', ylabel='xвосст(t)')
axes[1].stem(f_signed, analog_spectrum, basefmt='k-')
axes[1].set(title='г) Амплитудный спектр аналогового сигнала',
            xlabel='f, Гц', ylabel='|Xа[k]|/N', xlim=(-f_d2/2, f_d2/2))
for ax in axes:
    ax.grid(True, alpha=0.35)
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(11, 4.5))
ax.plot(t_analog*1000, x_analog_original, lw=1.8)
ax.set(title='д) Исходный аналоговый сигнал (аналитическая формула)',
       xlabel='t, мс', ylabel='xисх(t)')
ax.grid(True, alpha=0.35)
plt.tight_layout()
plt.show()

error_analog = np.max(np.abs(x_analog_original - x_reconstructed))
print(f'Число временных точек: {len(t_analog)}')
print(f'Шаг времени Δt = {T/4:.9f} с')
print(f'Интервал: [0, {t_analog[-1]:.9f}] с; (N−1)T = {(N-1)*T:.9f} с')
print(f'Максимальная ошибка восстановления: {error_analog:.18f}')
print(f'Максимальная мнимая часть реконструкции: '
      f'{np.max(np.abs(x_reconstructed_complex.imag)):.18f}')
assert len(t_analog) == 253
assert np.isclose(t_analog[-1], (N-1)*T)
assert np.allclose(x_reconstructed, x_analog_original, atol=1e-11)
```

**Интерпретация.** Восстановленная по ДПФ функция совпадает с исходной суммой двух косинусоид на выбранном интервале с точностью численных вычислений. Аналоговый амплитудный спектр содержит четыре линии при $f=\pm750$ Гц и $f=\pm1125$ Гц с двухсторонними амплитудами $0.56$ и $1.12$. Физические амплитуды гармоник исходного сигнала равны $1.12$ и $2.24$.

### 6. Связь спектров и сравнение восстановленного и исходного сигналов

При частоте дискретизации $f_{\mathrm d2}=6000$ Гц частота Найквиста составляет $3000$ Гц. Максимальная частота исходного сигнала — $f_2=1125$ Гц, поэтому условие $f_{\mathrm d2}>2f_{\max}$ выполняется: $6000>2250$.

**Связь спектров.** Модуль ДПФ $|X[k]|$ связан с комплексными амплитудами аналоговых гармоник соотношением $|X[k]|/N=|C[k]|$. Каждая вещественная гармоника образует пару симметричных линий при положительной и отрицательной частотах. В данном случае компоненты попадают точно на сетку ДПФ и не накладываются друг на друга при дискретизации.

**Сравнение сигналов.** Для наглядной проверки исходная и восстановленная функции совмещаются на одном графике, дополнительно отмечаются исходные дискретные отсчёты.

```python
fig, ax = plt.subplots(figsize=(11, 5))
ax.plot(t_analog*1000, x_analog_original, lw=2.0,
        label='Исходный аналоговый сигнал')
ax.plot(t_analog*1000, x_reconstructed, '--', lw=1.6,
        label='Восстановленный по ДПФ сигнал')
ax.plot(t_discrete*1000, x2, 'o', markersize=3.5,
        label='Исходные дискретные отсчёты')
ax.set(title='Визуальное сравнение исходного и восстановленного сигналов',
       xlabel='t, мс', ylabel='x(t)')
ax.legend()
ax.grid(True, alpha=0.35)
plt.tight_layout()
plt.show()

sample_error = np.max(np.abs(x_reconstructed[::4] - x2))
print(f'Частота Найквиста: {f_d2/2:.0f} Гц; максимальная частота сигнала: {f2:.0f} Гц')
print(f'Максимальная ошибка по всей временной сетке: {error_analog:.18f}')
print(f'Максимальная ошибка на исходных отсчётах: {sample_error:.18f}')
assert np.allclose(x_reconstructed[::4], x2, atol=1e-11)
```

**Интерпретация.** Графики исходного и восстановленного сигналов визуально накладываются друг на друга. Ошибка порядка машинной точности подтверждает корректность восстановления для данного примера. Это объясняется тем, что исходный сигнал состоит только из двух частот ниже частоты Найквиста, причём их спектральные линии полностью представлены коэффициентами ДПФ. Равенство исходных дискретных отсчётов восстановленному сигналу отдельно проверено в точках $t=nT$.

### 7. Проверка равенства Парсеваля

Для ДПФ, определённого без множителя $1/N$ в прямом преобразовании, равенство Парсеваля имеет вид

$$
\boxed{\sum_{n=0}^{N-1}|x[n]|^2=\frac1N\sum_{k=0}^{N-1}|X[k]|^2}.
$$

Так как на одном периоде присутствуют две ортогональные гармоники с целочисленными номерами частотных отсчётов $k_1=8$ и $k_2=12$, энергия вычисляется также аналитически:

$$
E=\frac{N}{2}(A_1^2+A_2^2)
=\frac{64}{2}(1.12^2+2.24^2)=\boxed{200.704}.
$$

```python
# Сравнение энергий за один период последовательности.
energy_time = np.sum(np.abs(x2)**2)
energy_freq = np.sum(np.abs(X2)**2) / N
energy_analytic = N * (A1**2 + A2**2) / 2
mean_power = energy_time / N

print(f'Энергия во временной области: {energy_time:.12f}')
print(f'Энергия в частотной области:  {energy_freq:.12f}')
print(f'Аналитическое значение:       {energy_analytic:.12f}')
print(f'Абсолютное расхождение:       {abs(energy_time-energy_freq):.18f}')
print(f'Средняя мощность за период:   {mean_power:.6f}')

assert np.isclose(energy_time, energy_freq, atol=1e-11)
assert np.isclose(energy_time, energy_analytic, atol=1e-11)
```

**Интерпретация.** Энергия последовательности за один период равна $200.704$ как при суммировании квадратов её отсчётов, так и по спектральным коэффициентам с нормировкой $1/N$. Равенство Парсеваля означает, что переход от временного представления к ДПФ не изменяет энергии при правильном учёте нормировки. Поскольку сигнал периодический и рассматривается один период, средняя мощность равна $E/N=3.136$, тогда как полная энергия бесконечно продолженного ненулевого периодического сигнала бесконечна.
