## Основные теоретические положения

### Дискретные сигналы, шкалы времени и частоты

Дискретный сигнал $x(nT)$ - последовательность отсчетов аналогового сигнала $x(t)$, взятых с периодом дискретизации $T$; частота дискретизации $f_d = 1/T$.

- **Дискретное время** - моменты $nT$ (с); **дискретное нормированное время** - номер отсчета $n = nT/T$ (безразмерный). Сокращенная запись: $x(n) \equiv x(nT)$.
- Частоты: абсолютная $f$ (Гц), круговая $\omega = 2\pi f$ (рад/с), нормированная $\hat{f} = f/f_d$, нормированная круговая $\hat{\omega} = \omega T = 2\pi f/f_d$ (рад).
- Спектр дискретного сигнала периодичен с периодом $f_d$ (по $\hat{\omega}$ - с периодом $2\pi$). Для вещественных сигналов достаточно **основной полосы** $f \in [0;\, f_d/2]$, т. е. $\hat{\omega} \in [0;\, \pi]$.
- **Теорема Котельникова:** аналоговый сигнал со спектром, ограниченным частотой $f_B$, однозначно восстанавливается по отсчетам, если $f_d \ge 2 f_B$.

Типовые последовательности: цифровой единичный импульс $u_0(n) = 1$ при $n = 0$ и $0$ при $n \ne 0$; цифровой единичный скачок $u_1(n) = 1$ при $n \ge 0$ и $0$ при $n < 0$.

### Линейные дискретные системы (ЛДС)

ЛДС подчиняется принципу суперпозиции (реакция на взвешенную сумму воздействий равна взвешенной сумме реакций); стационарная ЛДС, кроме того, инвариантна к сдвигу. Все характеристики ниже определяются при **нулевых начальных условиях**: $x(n-i) = 0$, $y(n-k) = 0$ при отрицательных аргументах.

**Импульсная характеристика (ИХ)** $h(n)$ - реакция ЛДС на $u_0(n)$.

**Формула свертки** выражает реакцию через ИХ и воздействие:
$$ y(n) = \sum_{m=0}^{n} h(m)\, x(n-m) = \sum_{m=0}^{n} x(m)\, h(n-m). $$
Линейная свертка последовательностей длиной $N_x$ и $N_h$ имеет длину $N_x + N_h - 1$.

**Разностное уравнение (РУ)** - рекуррентное соотношение между отсчетами воздействия и реакции:
$$ y(n) = \sum_{i=0}^{N-1} b_i\, x(n-i) - \sum_{k=1}^{M-1} a_k\, y(n-k), \qquad a_0 = 1. $$
Если все $a_k = 0$, ЛДС нерекурсивная, ее ИХ конечна (КИХ); если хотя бы один $a_k \ne 0$ - рекурсивная, ИХ бесконечна (БИХ).

### Передаточная функция, нули и полюсы

$Z$-преобразование последовательности: $X(z) = \sum_{n=0}^{\infty} x(n) z^{-n}$. Основные свойства: линейность, задержка $x(n-m) \leftrightarrow z^{-m} X(z)$, свертка $x_1 * x_2 \leftrightarrow X_1(z) X_2(z)$; соответствия $u_0(n) \leftrightarrow 1$, $\alpha^n u_1(n) \leftrightarrow 1/(1 - \alpha z^{-1})$.

**Передаточная функция (ПФ)** - отношение $Z$-изображений реакции и воздействия, она же $Z$-изображение ИХ: $H(z) = Y(z)/X(z) = \sum_{n} h(n) z^{-n}$. Для рекурсивного звена 2-го порядка
$$ H(z) = \frac{b_0 + b_1 z^{-1} + b_2 z^{-2}}{1 + a_1 z^{-1} + a_2 z^{-2}} = \frac{b_0 z^2 + b_1 z + b_2}{z^2 + a_1 z + a_2}. $$

**Нули** $z_{\circ i}$ - корни числителя, **полюсы** $z_{* k}$ - корни знаменателя. При вещественных коэффициентах комплексные нули и полюсы образуют сопряженные пары $r e^{\pm j\varphi}$. Комплексное число в алгебраической форме $z = \xi + j\eta$ связано с показательной формой $z = r e^{j\varphi}$ соотношениями $r = \sqrt{\xi^2 + \eta^2}$, $\varphi = \arg z$, $\xi = r\cos\varphi$, $\eta = r\sin\varphi$. Для пары полюсов $a_1 = -2 r_* \cos\varphi_*$, $a_2 = r_*^2$; для пары нулей $b_1/b_0 = -2 r_\circ \cos\varphi_\circ$, $b_2/b_0 = r_\circ^2$.

ПФ в виде произведения простейших множителей:
$$ H(z) = b_0\, \frac{(1 - z_{\circ 1} z^{-1})(1 - z_{\circ 2} z^{-1})}{(1 - z_{*1} z^{-1})(1 - z_{*2} z^{-1})}. $$

ПФ в виде суммы простых дробей (степени числителя и знаменателя по $z^{-1}$ равны, поэтому выделяется целая часть $C_0$):
$$ H(z) = C_0 + \sum_{k} \frac{C_k}{1 - z_{*k} z^{-1}} \quad\Rightarrow\quad h(n) = C_0\, u_0(n) + \sum_{k} C_k\, z_{*k}^{\,n}\, u_1(n). $$

**Устойчивость.** ЛДС устойчива, если любое ограниченное воздействие вызывает ограниченную реакцию. Критерий во временной области: $\sum_{n=0}^{\infty} |h(n)| < \infty$; в $z$-области: все полюсы ПФ лежат внутри единичного круга $|z_{*k}| < 1$. Положение нулей на устойчивость не влияет. **Карта нулей и полюсов** - их изображение на $z$-плоскости (нули - «o», полюсы - «x») вместе с единичной окружностью.

### Частотные характеристики

Частотная характеристика - ПФ на единичной окружности:
$$ H(e^{j\hat{\omega}}) = H(z)\big|_{z = e^{j\hat{\omega}}} = \sum_{n=0}^{\infty} h(n) e^{-j\hat{\omega} n} = A(\hat{\omega})\, e^{j\varphi(\hat{\omega})}, $$
где $A(\hat{\omega}) = |H(e^{j\hat{\omega}})|$ - **АЧХ** (амплитудно-частотная характеристика), $\varphi(\hat{\omega}) = \arg H(e^{j\hat{\omega}})$ - **ФЧХ** (фазочастотная характеристика). Обе периодичны с периодом $2\pi$; при вещественной ИХ АЧХ - четная, ФЧХ - нечетная функция. Переход к абсолютной частоте: $f = \hat{\omega} f_d / (2\pi)$.

Геометрический смысл: $A(\hat{\omega}) = |b_0| \dfrac{\prod_i |e^{j\hat{\omega}} - z_{\circ i}|}{\prod_k |e^{j\hat{\omega}} - z_{*k}|}$ - отношение произведений расстояний от точки $e^{j\hat{\omega}}$ единичной окружности до нулей и до полюсов. Поэтому АЧХ имеет максимум вблизи аргумента полюса и минимум вблизи аргумента нуля; если нуль лежит на окружности, АЧХ обращается в ноль, а ФЧХ в этой точке скачком меняется на $\pi$. Скачки ФЧХ на $2\pi$ - следствие представления фазы главным значением на интервале $(-\pi;\, \pi]$.

### Дискретное преобразование Фурье

ДПФ и обратное ДПФ (ОДПФ) $N$-точечной последовательности:
$$ X(k) = \sum_{n=0}^{N-1} x(n)\, W_N^{nk}, \qquad x(n) = \frac{1}{N}\sum_{k=0}^{N-1} X(k)\, W_N^{-nk}, \qquad W_N = e^{-j 2\pi / N}. $$
Отсчету $X(k)$ соответствуют дискретная нормированная частота $\hat{\omega}_k = 2\pi k/N$ и абсолютная частота $f_k = k f_d / N = k\,\Delta f$, где $\Delta f = f_d/N = 1/(NT)$ - шаг по частоте. **Амплитудный спектр** - $|X(k)|$, **фазовый спектр** - $\arg X(k)$.

- **Периодическая последовательность** $x(n) = x(n + N)$ имеет дискретный (линейчатый) спектр, полностью описываемый ДПФ одного периода.
- **Последовательность конечной длины** $N$ имеет непрерывный спектр $X(e^{j\hat{\omega}}) = \sum_{n=0}^{N-1} x(n) e^{-j\hat{\omega} n}$; ее ДПФ - отсчеты этого спектра с шагом $2\pi/N$: $X(k) = X(e^{j\hat{\omega}})\big|_{\hat{\omega} = 2\pi k/N}$.
- Свойства ДПФ: периодичность $X(k) = X(k + N)$, линейность, для вещественной $x(n)$ - симметрия $|X(N-k)| = |X(k)|$, $\arg X(N-k) = -\arg X(k)$.
- Гармоника $A\cos(2\pi k_0 n/N + \theta)$ с целым $k_0$ дает два ненулевых отсчета: $X(k_0) = \frac{NA}{2} e^{j\theta}$ и $X(N - k_0) = \frac{NA}{2} e^{-j\theta}$.

### Восстановление аналогового сигнала по ДПФ

Периодический аналоговый сигнал с периодом $NT$, спектр которого лежит в основной полосе, представляется конечным рядом Фурье, коэффициенты которого равны $X_a(k)/N$:
$$ x(t) = \frac{1}{N}\sum_{k=-N/2}^{N/2-1} X_a(k) \exp\Big(j \frac{2\pi}{NT} k t\Big), \qquad X_a(k) = \begin{cases} X(N + k), & -N/2 \le k \le -1, \\ X(k), & 0 \le k \le N/2 - 1. \end{cases} $$
Переход от $X(k)$ к $X_a(k)$ использует периодичность ДПФ и переносит отсчеты верхней половины периода на отрицательные частоты.

### Равенство Парсеваля

$$ \sum_{n=0}^{N-1} |x(n)|^2 = \frac{1}{N}\sum_{k=0}^{N-1} |X(k)|^2 . $$
Энергия последовательности на периоде, вычисленная во временной области, равна энергии, вычисленной по отсчетам ДПФ.

#### 1. Импульсная характеристика

ИХ вычисляется как реакция звена на цифровой единичный импульс $u_0(n)$ длины $N_1$ по разностному уравнению (функция `lds_response`, она же используется в п. 3).

```python
def lds_response(b, a, x):
    """Реакция ЛДС по разностному уравнению при нулевых начальных условиях:
    y(n) = sum_i b_i x(n-i) - sum_{k>=1} a_k y(n-k), a_0 = 1."""
    y = np.zeros(len(x))
    for n in range(len(x)):
        s = 0.0
        for i in range(len(b)):
            if n - i >= 0:
                s += b[i] * x[n - i]
        for k in range(1, len(a)):
            if n - k >= 0:
                s -= a[k] * y[n - k]
        y[n] = s / a[0]
    return y


n1 = np.arange(N1)
u0 = np.where(n1 == 0, 1.0, 0.0)      # цифровой единичный импульс
h = lds_response(b, a, u0)            # ИХ - реакция на u0(n)

fig, ax = plt.subplots()
draw_stem(ax, n1, h)
ax.set(title=f'Импульсная характеристика h(n), N1 = {N1}', xlabel='n', ylabel='h(n)')
plt.show()
print('h(n) =', h)
print(f'max|h(n)| = {np.abs(h).max():.4f} при n = {np.abs(h).argmax()}')
```

**Аналитическая формула ИХ.** Дискриминант знаменателя $z^2 + a_1 z + a_2$ отрицателен ($a_1^2 - 4a_2 = 0.9561 - 2.752 < 0$), поэтому полюсы комплексно-сопряженные: $z_{*1,2} = r_* e^{\pm j\varphi_*}$, $r_* = \sqrt{a_2}$, $\varphi_* = \arccos\big(-a_1/(2 r_*)\big)$.

Степени числителя и знаменателя ПФ по $z^{-1}$ одинаковы, поэтому при разложении на простые дроби выделяется целая часть $C_0 = b_2/a_2$ (значение $H(z)$ при $z^{-1} \to \infty$):
$$ H(z) = C_0 + \frac{C_1}{1 - z_{*1} z^{-1}} + \frac{C_1^{*}}{1 - z_{*1}^{*} z^{-1}}, \qquad C_1 = \frac{b_0 + b_1 z_{*1}^{-1} + b_2 z_{*1}^{-2}}{1 - z_{*1}^{*}/z_{*1}}. $$
Коэффициент $C_1$ получен умножением $H(z)$ на $(1 - z_{*1} z^{-1})$ и подстановкой $z = z_{*1}$; второй коэффициент сопряжен с первым, так как коэффициенты ПФ вещественны. По соответствиям $1 \leftrightarrow u_0(n)$ и $\alpha^n u_1(n) \leftrightarrow 1/(1 - \alpha z^{-1})$ при нулевых начальных условиях:
$$ h(n) = C_0\, u_0(n) + \big(C_1 z_{*1}^{\,n} + C_1^{*} (z_{*1}^{*})^{n}\big) u_1(n) = C_0\, u_0(n) + 2|C_1|\, r_*^{\,n} \cos(\varphi_* n + \psi)\, u_1(n), \qquad \psi = \arg C_1 . $$

```python
# Полюсы и коэффициенты разложения на простые дроби
p1 = np.roots(a)
p1 = p1[np.argmax(p1.imag)]                    # полюс с Im > 0
r_p, phi_p = abs(p1), np.angle(p1)
C0 = b2 / a2
C1 = (b0 + b1 / p1 + b2 / p1**2) / (1 - np.conj(p1) / p1)
psi = np.angle(C1)
print(f'r* = {r_p:.4f};  phi* = {phi_p:.4f} рад = {phi_p/np.pi:.4f}*pi')
print(f'C0 = {C0:.4f};  C1 = {C1:.4f};  2|C1| = {2*abs(C1):.4f};  psi = arg C1 = {psi/np.pi:.4f}*pi')


def h_formula(n):
    """ИХ по аналитической формуле (n >= 0)."""
    n = np.asarray(n, dtype=float)
    return C0 * (n == 0) + 2 * abs(C1) * r_p**n * np.cos(phi_p * n + psi)


print(f'max|h_РУ(n) - h_формула(n)| = {np.max(np.abs(h - h_formula(n1))):.1e}')
print(f'h(0) по формуле = C0 + 2|C1|cos(psi) = {C0 + 2*abs(C1)*np.cos(psi):.4f} = b0')

# Затухание: огибающая 2|C1| r*^n и ИХ на длинном интервале
env = 2 * abs(C1) * r_p**n1
h_long = lds_response(b, a, np.r_[1.0, np.zeros(999)])
last_1pct = np.max(np.nonzero(np.abs(h_long) >= 0.01 * np.abs(h_long).max())[0])
print(f'огибающая при n = N1-1 = {N1-1}: {env[-1]:.4f} ({100*env[-1]/np.abs(h).max():.1f}% от max|h|)')
print(f'|h(n)| < 1% от max|h| при всех n > {last_1pct};  sum|h(n)| (1000 отсчетов) = {np.abs(h_long).sum():.4f}')
```

Значения ИХ по разностному уравнению и по аналитической формуле совпадают с точностью до погрешности вычислений. Для варианта 8 формула в числовом виде:
$$ h(n) = 1.3430\, u_0(n) + 2.6584 \cdot 0.8295^{\,n} \cos(0.7006\pi n + 0.5827\pi)\, u_1(n). $$
Проверка первого отсчета: $h(0) = C_0 + 2|C_1|\cos\psi = 1.3430 - 0.6830 = 0.66 = b_0$; следующие отсчеты по РУ: $h(1) = b_1 - a_1 h(0) = -1.3876$, $h(2) = b_2 - a_1 h(1) - a_2 h(0) = 1.8267$.

**Длина ИХ.** Звено рекурсивное ($a_1, a_2 \ne 0$), поэтому ИХ в действительности имеет **бесконечную длину**: каждый следующий отсчет вычисляется через предыдущие и в ноль не обращается. Длина $N_1 = 28$ - только выбранный интервал расчета. ИХ - затухающее колебание с огибающей $2.6584 \cdot 0.8295^{\,n}$ и нормированной частотой $\varphi_* = 0.7006\pi$, т. е. с периодом около $2/0.7006 \approx 2.9$ отсчета: знак ИХ меняется почти на каждом отсчете (полюсы лежат в левой полуплоскости, ближе к $z = -1$, чем к $z = 1$). К концу интервала огибающая уменьшается до $\approx 0.017$ (около 1% от $\max|h| = 1.827$), поэтому практически ИХ затухает за 25-30 отсчетов. Малое значение $h(27) \approx 10^{-5}$ - не конец ИХ, а случайное попадание отсчета вблизи нуля колебания.

#### 2. Реакция по формуле свертки

Длина импульса: $int(N_2/2) = int(38/2) = 19$, т. е. $x(n) = 1$ при $0 \le n \le 18$ и $x(n) = 0$ при $19 \le n \le 37$.

```python
n2 = np.arange(N2)
L = round_half_up(N2 / 2)                  # int(N2/2)
x = np.where(n2 < L, 1.0, 0.0)             # дискретный прямоугольный импульс


def convolve_direct(x, h):
    """Линейная свертка y(n) = sum_m h(m) x(n-m); длина len(x) + len(h) - 1."""
    y = np.zeros(len(x) + len(h) - 1)
    for n in range(len(y)):
        for m in range(len(h)):
            if 0 <= n - m < len(x):
                y[n] += h[m] * x[n - m]
    return y


y_conv = convolve_direct(x, h)             # полная свертка
y1 = y_conv[:N2]                           # реакция, ограниченная длиной воздействия
nz = np.nonzero(np.abs(y_conv) > 1e-12)[0]
print(f'длина последовательности x: {len(x)};  длина импульса (единичных отсчетов): {int(x.sum())}')
print(f'длина полной свертки: {len(y_conv)} = N1 + N2 - 1 = {N1 + N2 - 1}')
print(f'ненулевые отсчеты свертки: n = {nz[0]}..{nz[-1]} ({nz[-1] - nz[0] + 1} = {int(x.sum())} + {N1} - 1)')
print('совпадение с numpy.convolve:', np.allclose(y_conv, np.convolve(x, h)))

fig, axs = plt.subplots(1, 3, figsize=(15, 3.6))
draw_stem(axs[0], n2, x)
axs[0].set(title=f'Воздействие x(n), N2 = {N2}', xlabel='n', ylabel='x(n)')
draw_stem(axs[1], np.arange(len(y_conv)), y_conv, color='C2')
axs[1].axvline(N2 - 0.5, color='r', ls='--', lw=1, label='граница n = N2')
axs[1].set(title=f'Полная свертка, длина {len(y_conv)}', xlabel='n')
axs[1].legend(fontsize=8)
draw_stem(axs[2], n2, y1, color='C1')
axs[2].set(title=f'Реакция y1(n), ограниченная до N2 = {N2}', xlabel='n', ylabel='y1(n)')
plt.tight_layout()
plt.show()
```

**Пояснения.**
- **Длина импульса** $x$ - $int(N_2/2) = 19$ единичных отсчетов ($n = 0 \dots 18$); длина всей последовательности воздействия - $N_2 = 38$ отсчетов (последние 19 - нули).
- **Длина свертки аналитически** - $N_1 + N_2 - 1 = 28 + 38 - 1 = 65$ отсчетов ($n = 0 \dots 64$). **По графику** ненулевые отсчеты занимают $n = 0 \dots 45$, т. е. $46 = 19 + 28 - 1$: нулевая часть воздействия ничего не добавляет, и фактическая длина определяется длиной самого импульса и длиной ИХ.
- **Ограничение длиной воздействия.** Реакцию рассматривают на том же интервале $n = 0 \dots N_2 - 1$, что и воздействие: так ее можно сравнить отсчет в отсчет с реакцией по РУ (п. 3). Кроме того, свертка вычислена с ИХ, усеченной до $N_1$ отсчетов, поэтому чем дальше от начала, тем сильнее она отличается от истинной реакции бесконечной ИХ: после $n = 45$ свертка дает нули, хотя реальная реакция рекурсивного звена продолжает затухать.

Разностное уравнение рекурсивного звена 2-го порядка в общем виде и с коэффициентами варианта 8:
$$ y(n) = b_0 x(n) + b_1 x(n-1) + b_2 x(n-2) - a_1 y(n-1) - a_2 y(n-2), $$
$$ y(n) = 0.66\, x(n) - 0.742236\, x(n-1) + 0.924\, x(n-2) - 0.9778\, y(n-1) - 0.688\, y(n-2). $$

```python
y2 = lds_response(b, a, x)

fig, axs = plt.subplots(1, 3, figsize=(15, 3.6))
draw_stem(axs[0], n2, x)
axs[0].set(title='Воздействие x(n)', xlabel='n', ylabel='x(n)')
draw_stem(axs[1], n2, y2, color='C3')
axs[1].set(title='Реакция y2(n) по РУ', xlabel='n', ylabel='y2(n)')
axs[2].plot(n2, y1, 'o-', ms=5, lw=1, label='y1(n) - свертка')
axs[2].plot(n2, y2, 'x--', ms=7, lw=1, label='y2(n) - РУ')
axs[2].axhline(b.sum() / a.sum(), color='gray', ls=':', lw=1, label='H(1)')
axs[2].set(title='Сравнение y1(n) и y2(n)', xlabel='n')
axs[2].legend(fontsize=8)
plt.tight_layout()
plt.show()

d = np.abs(y1 - y2)
print(f'длина воздействия: {len(x)};  длина реакции y2: {len(y2)}')
print(f'max|y1 - y2| при n < N1: {d[:N1].max():.1e};  при n >= N1: {d[N1:].max():.4f} (n = {N1 + d[N1:].argmax()})')
print(f'H(1) = sum(b)/sum(a) = {b.sum()/a.sum():.4f};  y2(14..18) = {y2[14:19]}')
print(f'y2(0..2) = {y2[:3]};  y2(19..21) = {y2[19:22]}')
```

**Пояснения.**

- **Сравнение $y_1(n)$ и $y_2(n)$.** Графики совпадают. При $n < N_1 = 28$ отсчеты равны с точностью $\sim 10^{-16}$; при $n \ge 28$ появляется расхождение до $0.0114$: свертка использует ИХ, усеченную до 28 отсчетов, а РУ неявно учитывает всю бесконечную ИХ. Это еще одно подтверждение бесконечной длины ИХ.
- **Форма реакции.** На интервале импульса реакция колеблется около значения $H(1) = \sum b_i / \sum a_k = 0.3158$ - коэффициента передачи на нулевой частоте, т. е. постоянная составляющая импульса ослаблена примерно в 3 раза. На фронте ($y_2(1) = -0.73$, $y_2(2) = 1.10$) и на срезе импульса ($y_2(20) = 1.05$, $y_2(21) = -0.81$) возникают знакопеременные колебания с размахом больше единицы: звено выделяет резкие перепады воздействия, т. е. его высокочастотные составляющие (подробнее - в п. 6).

#### 4. Нули и полюсы передаточной функции

Нули - корни $b_0 z^2 + b_1 z + b_2 = 0$, полюсы - корни $z^2 + a_1 z + a_2 = 0$.

```python
zeros = np.roots(b)
poles = np.roots(a)
zeros = zeros[np.argsort(-zeros.imag)]     # сначала корень с Im > 0
poles = poles[np.argsort(-poles.imag)]

for name, roots in (('нуль ', zeros), ('полюс', poles)):
    for i, c in enumerate(roots, 1):
        print(f'{name} {i}: {c.real:+.4f} {c.imag:+.4f}j  =  {abs(c):.4f} exp(j {np.angle(c):+.4f})'
              f'  =  {abs(c):.4f} exp(j {np.angle(c)/np.pi:+.4f} pi)   ({np.degrees(np.angle(c)):+.1f} град)')

# Связь модулей и аргументов с коэффициентами ПФ
r_z, phi_z = abs(zeros[0]), np.angle(zeros[0])
print(f'\nнули:   r^2 = {r_z**2:.4f} (b2/b0 = {b2/b0:.4f});  -2 r cos(phi) = {-2*r_z*np.cos(phi_z):.4f} (b1/b0 = {b1/b0:.4f})')
print(f'полюсы: r^2 = {r_p**2:.4f} (a2 = {a2:.4f});  -2 r cos(phi) = {-2*r_p*np.cos(phi_p):.4f} (a1 = {a1:.4f})')

# Проверка разложения на простейшие множители в произвольной точке z
z_test = 0.3 - 0.8j
H_coef = np.polyval(b, z_test) / np.polyval(a, z_test)
H_fact = b0 * np.prod(1 - zeros / z_test) / np.prod(1 - poles / z_test)
print(f'\nH({z_test}) по коэффициентам: {H_coef:.6f};  по произведению множителей: {H_fact:.6f}')
```

**Связь алгебраической и показательной форм.** Алгебраическая форма $z = \xi + j\eta$ задает точку на $z$-плоскости декартовыми координатами, показательная $z = r e^{j\varphi}$ - полярными: модулем $r = \sqrt{\xi^2 + \eta^2}$ (расстояние от начала координат) и аргументом $\varphi$ (угол к положительной вещественной полуоси); $\xi = r\cos\varphi$, $\eta = r\sin\varphi$. Коэффициенты ПФ вещественны, поэтому нули и полюсы образуют комплексно-сопряженные пары: модули одинаковы, аргументы противоположны по знаку. Модули и аргументы однозначно выражаются через коэффициенты: $r_\circ^2 = b_2/b_0$, $-2 r_\circ \cos\varphi_\circ = b_1/b_0$, $r_*^2 = a_2$, $-2 r_* \cos\varphi_* = a_1$ - проверка в выводе выше.

| | Алгебраическая форма | Показательная форма |
|:--|:--|:--|
| Нули $z_{\circ 1,2}$ | $0.5623 \pm 1.0411j$ | $1.1832\, e^{\pm j\, 0.3424\pi}$ ($\pm 61.6°$) |
| Полюсы $z_{* 1,2}$ | $-0.4889 \pm 0.6701j$ | $0.8295\, e^{\pm j\, 0.7006\pi}$ ($\pm 126.1°$) |

**ПФ в виде произведения простейших множителей:**
$$ H(z) = b_0 \frac{(1 - z_{\circ 1} z^{-1})(1 - z_{\circ 2} z^{-1})}{(1 - z_{*1} z^{-1})(1 - z_{*2} z^{-1})} = 0.66\, \frac{\big(1 - 1.1832\, e^{j0.3424\pi} z^{-1}\big)\big(1 - 1.1832\, e^{-j0.3424\pi} z^{-1}\big)}{\big(1 - 0.8295\, e^{j0.7006\pi} z^{-1}\big)\big(1 - 0.8295\, e^{-j0.7006\pi} z^{-1}\big)}. $$
Значения ПФ, вычисленные по коэффициентам и по этому произведению, совпадают.

#### 6. АЧХ и ФЧХ в шкале нормированных частот

Частотная характеристика вычисляется подстановкой $z = e^{j\hat{\omega}}$ в ПФ:
$$ H(e^{j\hat{\omega}}) = \frac{b_0 + b_1 e^{-j\hat{\omega}} + b_2 e^{-j2\hat{\omega}}}{1 + a_1 e^{-j\hat{\omega}} + a_2 e^{-j2\hat{\omega}}}. $$

```python
def freq_response(b, a, w):
    """H(e^{jw}) = sum_i b_i e^{-jwi} / sum_k a_k e^{-jwk}."""
    num = sum(bi * np.exp(-1j * w * i) for i, bi in enumerate(b))
    den = sum(ak * np.exp(-1j * w * k) for k, ak in enumerate(a))
    return num / den


w = np.linspace(0, np.pi, 4001)            # основная полоса [0; pi]
Hw = freq_response(b, a, w)
A_w = np.abs(Hw)                           # АЧХ
phi_w = np.angle(Hw)                       # ФЧХ (главное значение)
i_max, i_min = A_w.argmax(), A_w.argmin()
jumps = np.nonzero(np.abs(np.diff(phi_w)) > np.pi)[0]
i_one = np.nonzero(np.diff(np.sign(A_w - 1)))[0]

fig, axs = plt.subplots(1, 2, figsize=(15, 4))
axs[0].plot(w / np.pi, A_w, label=r'$A(\hat{\omega})$')
axs[0].plot(w[i_max] / np.pi, A_w[i_max], 'r^', label='максимум')
axs[0].plot(w[i_min] / np.pi, A_w[i_min], 'gv', label='минимум')
axs[0].axvline(phi_p / np.pi, color='C3', ls='--', lw=1, label='аргумент полюса')
axs[0].axvline(phi_z / np.pi, color='C0', ls='--', lw=1, label='аргумент нуля')
axs[0].axhline(1, color='gray', ls=':', lw=1)
axs[0].set(title=r'АЧХ $A(\hat{\omega})$', xlabel=r'$\hat{\omega}/\pi$', ylabel=r'$A(\hat{\omega})$')
axs[0].legend(fontsize=8)
axs[1].plot(w / np.pi, phi_w / np.pi, label='главное значение')
axs[1].plot(w / np.pi, np.unwrap(phi_w) / np.pi, '--', label=r'без скачков на $2\pi$')
axs[1].set(title=r'ФЧХ $\varphi(\hat{\omega})$', xlabel=r'$\hat{\omega}/\pi$', ylabel=r'$\varphi/\pi$')
axs[1].legend(fontsize=8)
plt.tight_layout()
plt.show()

print(f'A(0) = {A_w[0]:.4f};  A(pi) = {A_w[-1]:.4f}')
print(f'максимум A = {A_w[i_max]:.4f} при w = {w[i_max]/np.pi:.4f}*pi  (аргумент полюса {phi_p/np.pi:.4f}*pi)')
print(f'минимум  A = {A_w[i_min]:.4f} при w = {w[i_min]/np.pi:.4f}*pi  (аргумент нуля {phi_z/np.pi:.4f}*pi)')
print(f'A(w) = 1 при w = {w[i_one[0]]/np.pi:.4f}*pi; A < 1 левее, A > 1 правее')
for k in jumps:
    print(f'скачок ФЧХ при w = {w[k]/np.pi:.4f}*pi: {(phi_w[k+1]-phi_w[k])/np.pi:+.3f}*pi, A = {A_w[k]:.4f}')
```

**Пояснения.**
- **Границы основной полосы:** $\hat{\omega} \in [0;\, \pi]$. За ее пределами характеристики повторяются с периодом $2\pi$, на $[-\pi;\, 0]$ АЧХ симметрична, а ФЧХ антисимметрична относительно нуля.
- **Соответствие карте нулей и полюсов.** АЧХ пропорциональна отношению расстояний от точки $e^{j\hat{\omega}}$ до нулей и до полюсов. Полюс под углом $0.7006\pi$ достаточно близок к окружности ($r_* = 0.8295$), поэтому около этой частоты АЧХ имеет **резонансный максимум** $A = 6.830$ при $\hat{\omega} = 0.7133\pi$. Нуль под углом $0.3424\pi$ дает **минимум** $A = 0.127$ при $\hat{\omega} = 0.3325\pi$; так как нуль не лежит на окружности ($r_\circ = 1.1832$), АЧХ в ноль не обращается. Экстремумы немного смещены относительно аргументов полюса и нуля, так как на АЧХ влияют и остальные корни: сопряженные полюс и нуль и вторая пара.
- **Скачок ФЧХ на $\pi$** происходит на частоте, где АЧХ равна нулю (нуль ПФ лежит на единичной окружности): при переходе через ноль множитель $e^{j\hat{\omega}} - z_\circ$ меняет знак. В варианте 8 нули не лежат на окружности, поэтому **скачков на $\pi$ нет**. Скачок при $\hat{\omega} = 0.5088\pi$ равен $2\pi$ (при $A = 0.844 \ne 0$): это не свойство звена, а результат приведения фазы к главному значению $(-\pi;\, \pi]$; пунктирная кривая показывает ту же ФЧХ без этого скачка.
- **Подавленные составляющие.** $A(0) = 0.316$, $A(\pi) = 3.275$; $A < 1$ при $\hat{\omega} < 0.5245\pi$ и $A > 1$ выше. Следовательно, звено преимущественно **подавляет низкие частоты** (сильнее всего - в окрестности $0.33\pi$) и усиливает высокие. В реакции на прямоугольный импульс (пп. 2-3) это проявилось так: постоянная составляющая ослаблена до уровня $0.316$, а резкие фронт и срез импульса, содержащие высокочастотные составляющие, вызвали колебания с частотой около $0.7\pi$ (знак меняется почти на каждом отсчете).

#### 7. АЧХ и ФЧХ в шкале частот f (Гц)

Переход к абсолютной частоте: $f = \dfrac{\hat{\omega}}{2\pi} f_d$, где $f_d = 8000$ Гц.

```python
f_hz = w / (2 * np.pi) * fs1               # Гц

fig, axs = plt.subplots(1, 2, figsize=(15, 4))
axs[0].plot(f_hz, A_w)
axs[0].plot(f_hz[i_max], A_w[i_max], 'r^')
axs[0].plot(f_hz[i_min], A_w[i_min], 'gv')
axs[0].set(title=f'АЧХ A(f), f_d = {fs1} Гц', xlabel='f, Гц', ylabel='A(f)', xlim=(0, fs1 / 2))
axs[1].plot(f_hz, phi_w / np.pi)
axs[1].set(title=r'ФЧХ $\varphi(f)$', xlabel='f, Гц', ylabel=r'$\varphi/\pi$', xlim=(0, fs1 / 2))
plt.tight_layout()
plt.show()

print(f'основная полоса: [0; f_d/2] = [0; {fs1/2:.0f}] Гц')
for name, i in (('максимум АЧХ', i_max), ('минимум АЧХ', i_min), ('A = 1', i_one[0]), ('скачок 2pi', jumps[0])):
    print(f'{name:13s}: w = {w[i]/np.pi:.4f}*pi  ->  f = {f_hz[i]:.0f} Гц')
```

**Пояснения.**
- **Границы основной полосы** в шкале $f$: $[0;\, f_d/2] = [0;\, 4000]$ Гц; $f_d/2$ - частота Найквиста.
- **Соответствие частот** $\hat{\omega} = 2\pi f / f_d$, т. е. $f = \hat{\omega} f_d / (2\pi)$: $\hat{\omega} = 0 \leftrightarrow 0$ Гц, $\hat{\omega} = \pi \leftrightarrow f_d/2 = 4000$ Гц, $\hat{\omega} = 2\pi \leftrightarrow f_d = 8000$ Гц. Графики в шкале $f$ отличаются от графиков п. 6 только масштабом оси абсцисс ($0.1\pi$ соответствует 400 Гц): максимум $0.7133\pi \leftrightarrow 2853$ Гц, минимум $0.3325\pi \leftrightarrow 1330$ Гц, граница $A = 1$: $0.5245\pi \leftrightarrow 2098$ Гц. При той же ПФ, но другой $f_d$ форма характеристик сохранится, а частоты в герцах изменятся пропорционально $f_d$.

### Часть 2. Дискретное преобразование Фурье

```python
# Параметры варианта №8, часть 2
N = 64
fs2 = 2000 * (N_var % 5 + 1)        # частота дискретизации, Гц
T = 1 / fs2                         # период дискретизации, с
A1 = 1 + 0.01 * N_var
A2 = 2 * A1
f1 = fs2 / 8
f2 = 1.5 * f1
df = fs2 / N                        # шаг по частоте ДПФ
print(f'N = {N}, f_d = {fs2} Гц, T = {T*1e6:.0f} мкс, A1 = {A1:g}, A2 = {A2:g}, f1 = {f1:g} Гц, f2 = {f2:g} Гц')
print(f'шаг по частоте df = f_d/N = {df:g} Гц;  номера гармоник k1 = f1/df = {f1/df:g}, k2 = f2/df = {f2/df:g}')
```

#### 1. Последовательность на периоде N

```python
def x_analog(t):
    """Исходный аналоговый сигнал x(t)."""
    return A1 * np.cos(2 * np.pi * f1 * t + np.pi / 4) + A2 * np.cos(2 * np.pi * f2 * t + np.pi / 8)


n = np.arange(N)
xn = x_analog(n * T)                        # x(nT)

fig, axs = plt.subplots(1, 2, figsize=(15, 3.6))
draw_stem(axs[0], n * T * 1e3, xn)
axs[0].set(title='x(nT) в шкале дискретного времени', xlabel='nT, мс', ylabel='x(nT)')
draw_stem(axs[1], n, xn, color='C1')
axs[1].set(title='x(n) в шкале дискретного нормированного времени', xlabel='n', ylabel='x(n)')
plt.tight_layout()
plt.show()
print(f'интервал времени: [0; (N-1)T] = [0; {(N-1)*T*1e3:.3f}] мс;  min x = {xn.min():.4f}, max x = {xn.max():.4f}')
print('период последовательности 16 отсчетов: x(n) == x(n+16):', np.allclose(xn[:48], xn[16:]))
```

Графики в двух шкалах одинаковы по форме и различаются только подписями оси абсцисс: на левом - моменты времени $nT$ с шагом $T = 125$ мкс (интервал $[0;\, 7.875]$ мс), на правом - номера отсчетов $n = 0 \dots 63$. Нормированные частоты гармоник $f_1/f_d = 1/8$ и $f_2/f_d = 3/16$, поэтому первая гармоника повторяется через 8 отсчетов, вторая - через 16 (3 своих периода), а их сумма - через 16 отсчетов. На интервале $N = 64$ укладываются 4 периода последовательности: ровно 8 периодов первой гармоники и 12 периодов второй.

#### 2. ОДПФ от ДПФ

ДПФ и ОДПФ вычисляются по определению через матрицу $W_N^{nk}$.

```python
k = np.arange(N)
W = np.exp(-2j * np.pi * np.outer(k, n) / N)     # матрица ДПФ: W[k, n] = W_N^{nk}
X = W @ xn                                        # ДПФ
x_idft = (W.conj() @ X) / N                       # ОДПФ
print('ДПФ по определению совпадает с numpy.fft.fft:', np.allclose(X, np.fft.fft(xn)))
print(f'max|x(n) - ОДПФ{{ДПФ{{x(n)}}}}| = {np.max(np.abs(xn - x_idft)):.1e};  max|Im| = {np.max(np.abs(x_idft.imag)):.1e}')

fig, ax = plt.subplots()
draw_stem(ax, n, x_idft.real, color='C2', label='ОДПФ{ДПФ{x(n)}}')
ax.plot(n, xn, 'k.', ms=5, label='исходная x(n)')
ax.set(title='Последовательность, полученная ОДПФ от ДПФ', xlabel='n', ylabel='x(n)')
ax.legend(fontsize=8, loc='upper right')
plt.show()
```

ОДПФ от ДПФ возвращает исходную последовательность: расхождение $\sim 10^{-14}$ и мнимая часть того же порядка - погрешность вычислений с плавающей точкой. ДПФ и ОДПФ образуют пару взаимно однозначных преобразований: по $N$ отсчетам ДПФ последовательность восстанавливается полностью.

#### 3. Амплитудный и фазовый спектры

```python
A_k = np.abs(X)                                   # амплитудный спектр
tol = 1e-9 * A_k.max()
phi_k = np.where(A_k > tol, np.angle(X), 0.0)     # фаза нулевых отсчетов не определена -> 0
w_k = 2 * np.pi * k / N                           # дискретные нормированные частоты
f_k = k * df                                      # абсолютные частоты, Гц

fig, axs = plt.subplots(2, 2, figsize=(15, 6.5))
draw_stem(axs[0, 0], w_k / np.pi, A_k)
axs[0, 0].set(title='Амплитудный спектр |X(k)|', xlabel=r'$\hat{\omega}_k/\pi = 2k/N$', ylabel='|X(k)|')
draw_stem(axs[0, 1], f_k, A_k)
axs[0, 1].set(title='Амплитудный спектр |X(k)|', xlabel='f_k = k f_d / N, Гц', ylabel='|X(k)|')
draw_stem(axs[1, 0], w_k / np.pi, phi_k / np.pi, color='C1')
axs[1, 0].set(title='Фазовый спектр arg X(k)', xlabel=r'$\hat{\omega}_k/\pi = 2k/N$', ylabel=r'$\arg X(k)/\pi$')
draw_stem(axs[1, 1], f_k, phi_k / np.pi, color='C1')
axs[1, 1].set(title='Фазовый спектр arg X(k)', xlabel='f_k = k f_d / N, Гц', ylabel=r'$\arg X(k)/\pi$')
plt.tight_layout()
plt.show()

for kk in np.nonzero(A_k > tol)[0]:
    print(f'k = {kk:2d}: w = {w_k[kk]/np.pi:.3f}*pi, f = {f_k[kk]:6.0f} Гц, |X| = {A_k[kk]:.2f}, '
          f'2|X|/N = {2*A_k[kk]/N:.2f}, arg X = {phi_k[kk]/np.pi:+.4f}*pi')
print(f'max |X(k)| на остальных отсчетах: {np.max(np.delete(A_k, [8, 12, 52, 56])):.1e}')
```

**Пояснения.**
- **Связь дискретного нормированного времени с дискретным временем:** $n = nT/T$ - номер отсчета равен моменту времени, деленному на период дискретизации $T = 1/f_d = 125$ мкс. Нормированная шкала безразмерна и не зависит от $f_d$; для перехода к секундам номер умножается на $T$.
- **Связь частоты $f$ (Гц) с дискретной нормированной частотой:** $\hat{\omega}_k = 2\pi f_k / f_d$; $k$-му отсчету ДПФ соответствуют $\hat{\omega}_k = 2\pi k/N$ и $f_k = k f_d/N = k \Delta f$, $\Delta f = 125$ Гц. Отсчет $k = N/2 = 32$ соответствует $\hat{\omega} = \pi$ и $f_d/2 = 4000$ Гц, правый край $k = N$ - $\hat{\omega} = 2\pi$ и $f_d = 8000$ Гц.
- **Вид амплитудного спектра.** Спектр линейчатый, ненулевые только 4 отсчета: $k = 8$ ($0.25\pi$, $f_1 = 1000$ Гц) и $k = 12$ ($0.375\pi$, $f_2 = 1500$ Гц) и их зеркальные отражения $k = 56$ и $k = 52$ ($2\pi - \hat{\omega}$, т. е. $f_d - f$: 7000 и 6500 Гц). Величины $|X(k)| = N A_i/2$: $34.56$ и $69.12$; амплитуды гармоник восстанавливаются как $2|X(k)|/N = 1.08$ и $2.16$. Остальные отсчеты равны нулю (с точностью порядка $10^{-13}$): на интервале $N$ укладывается целое число периодов каждой гармоники, поэтому утечки спектра нет. Спектр симметричен относительно $k = N/2$: $|X(N-k)| = |X(k)|$.
- **Вид фазового спектра.** $\arg X(8) = \pi/4$ и $\arg X(12) = \pi/8$ - начальные фазы гармоник; на зеркальных отсчетах фазы противоположны ($-\pi/4$, $-\pi/8$), т. е. фазовый спектр нечетен. На отсчетах с нулевой амплитудой фаза не определена (численно - шум округления), поэтому она принята равной нулю.

#### 4. ДПФ конечной последовательности

Конечная последовательность - тот же отрезок $x(n)$, $n = 0 \dots N-1$, вне которого отсчеты равны нулю. Ее непрерывный спектр $X(e^{j\hat{\omega}}) = \sum_{n=0}^{N-1} x(n) e^{-j\hat{\omega} n}$ рассчитан на густой сетке частот для сравнения с отсчетами ДПФ.

```python
x_fin = xn.copy()                                     # конечная последовательность длины N
X_fin = np.array([np.sum(x_fin * np.exp(-2j * np.pi * kk * n / N)) for kk in k])   # N-точечное ДПФ
w_dense = np.linspace(0, 2 * np.pi, 4096, endpoint=False)
X_cont = np.exp(-1j * np.outer(w_dense, n)) @ x_fin   # непрерывный спектр X(e^{jw})
print('|ДПФ конечной последовательности| = амплитудный спектр периодической:', np.allclose(np.abs(X_fin), A_k))
on_grid = np.isin(np.round(w_dense * N / (2 * np.pi), 9), k)
print('непрерывный спектр в точках w_k совпадает с ДПФ:', np.allclose(np.abs(X_cont[on_grid]), np.abs(X_fin)))

fig, ax = plt.subplots(figsize=(15, 4.2))
ax.plot(w_dense / np.pi, np.abs(X_cont), color='0.65', lw=1, label=r'$|X(e^{j\hat{\omega}})|$ конечной последовательности')
draw_stem(ax, w_k / np.pi, np.abs(X_fin), color='C0', ms=6, label='|X(k)| - модуль ДПФ конечной последовательности')
ax.plot(w_k / np.pi, A_k, 'r_', ms=14, mew=2, label='амплитудный спектр периодической последовательности')
ax.set(title='Модуль ДПФ конечной последовательности и амплитудный спектр периодической',
       xlabel=r'$\hat{\omega}/\pi$', ylabel='модуль')
ax.legend(fontsize=8, loc='upper center')
plt.show()
```

**Пояснения.** Модуль $N$-точечного ДПФ конечной последовательности длины $N$ **совпадает** с амплитудным спектром периодической последовательности с периодом $N$: периодическая последовательность - периодическое продолжение конечной, и ДПФ в обоих случаях вычисляется по одним и тем же $N$ отсчетам. Различается смысл этих отсчетов:
- у периодической последовательности спектр по своей природе дискретный, и $|X(k)|$ - это и есть весь спектр (линии на частотах $\hat{\omega}_k$);
- у конечной последовательности спектр непрерывный (серая кривая, с главными и боковыми лепестками), а ДПФ дает его отсчеты с шагом $2\pi/N$. Так как частоты гармоник кратны шагу $\Delta f$, все отсчеты, кроме $k = 8, 12, 52, 56$, попадают точно в нули непрерывного спектра, а эти четыре - в вершины главных лепестков.

#### 5. Восстановление аналогового сигнала по отсчетам ДПФ

Используется формула $x(t) = \dfrac{1}{N}\sum_{k=-N/2}^{N/2-1} X_a(k) \exp\Big(j\dfrac{2\pi}{NT}kt\Big)$, где $X_a(k) = X(N+k)$ при $-N/2 \le k \le -1$ и $X_a(k) = X(k)$ при $0 \le k \le N/2-1$; время $t \in [0;\, (N-1)T]$ с шагом $\Delta t = 0.25T$. Амплитудный спектр восстановленного сигнала - модули коэффициентов его ряда Фурье $|X_a(k)|/N$ на частотах $f_k = k/(NT)$.

```python
ka = np.arange(-N // 2, N // 2)                         # k = -N/2 ... N/2-1
Xa = np.where(ka < 0, X[(ka + N) % N], X[ka % N])       # X_a(k)
t = np.arange(4 * (N - 1) + 1) * 0.25 * T               # t = 0, 0.25T, ..., (N-1)T
x_rec = (np.exp(1j * 2 * np.pi / (N * T) * np.outer(t, ka)) @ Xa) / N
x_true = x_analog(t)                                    # исходный аналоговый сигнал
c_k = np.abs(Xa) / N                                    # амплитудный спектр восстановленного сигнала
print(f'отсчетов времени: {len(t)} (от 0 до {t[-1]*1e3:.3f} мс с шагом {0.25*T*1e6:.2f} мкс)')
print(f'max|Im x(t)| = {np.max(np.abs(x_rec.imag)):.1e};  max|x_восст(t) - x_исх(t)| = {np.max(np.abs(x_rec.real - x_true)):.1e}')
for kk in ka[c_k > 1e-9]:
    print(f'k = {kk:+3d}: f = {kk/(N*T):+6.0f} Гц, |X_a(k)|/N = {c_k[ka == kk][0]:.2f}')

fig, axs = plt.subplots(3, 2, figsize=(15, 10))
draw_stem(axs[0, 0], n, xn)
axs[0, 0].set(title='5.1. Периодическая последовательность x(n)', xlabel='n', ylabel='x(n)')
draw_stem(axs[0, 1], k, A_k)
axs[0, 1].set(title='5.1. Модуль ДПФ |X(k)|', xlabel='k', ylabel='|X(k)|')
axs[1, 0].plot(t * 1e3, x_rec.real, color='C2')
axs[1, 0].set(title='5.2. Восстановленный аналоговый сигнал x(t)', xlabel='t, мс', ylabel='x(t)')
draw_stem(axs[1, 1], ka / (N * T), c_k, color='C2')
axs[1, 1].set(title='5.2. Амплитудный спектр восстановленного сигнала |X_a(k)|/N',
              xlabel='f = k/(NT), Гц', ylabel='|X_a(k)|/N')
axs[2, 0].plot(t * 1e3, x_true, color='C3')
axs[2, 0].set(title='5.3. Исходный аналоговый сигнал x(t)', xlabel='t, мс', ylabel='x(t)')
axs[2, 1].plot(t * 1e3, x_true, color='C3', lw=3, alpha=0.5, label='исходный x(t)')
axs[2, 1].plot(t * 1e3, x_rec.real, 'k--', lw=1, label='восстановленный x(t)')
axs[2, 1].plot(n * T * 1e3, xn, 'o', ms=4, color='C0', label='отсчеты x(nT)')
axs[2, 1].set(title='Сравнение сигналов (фрагмент 0-2 мс)', xlabel='t, мс', xlim=(0, 2))
axs[2, 1].legend(fontsize=8, loc='lower right')
plt.tight_layout()
plt.show()
```

#### 6. Связь ДПФ со спектром аналогового сигнала и сравнение сигналов

- **Связь модуля ДПФ со спектром аналогового сигнала.** Восстановленный сигнал - конечный ряд Фурье периодического сигнала с периодом $NT = 8$ мс, его коэффициенты равны $X_a(k)/N$, а $k$-й член - гармоника с частотой $f_k = k/(NT) = k \Delta f$. Следовательно, модуль ДПФ, деленный на $N$, - это амплитудный спектр аналогового сигнала: линии на $\pm 1000$ Гц высотой $0.54 = A_1/2$ и на $\pm 1500$ Гц высотой $1.08 = A_2/2$; сумма пары симметричных линий дает амплитуду косинусоиды $A_i$. Перестановка $X(k) \to X_a(k)$ переносит отсчеты $k = 52, 56$ (частоты $f_d - f$ = 6500 и 7000 Гц) на отрицательные частоты $-1500$ и $-1000$ Гц, т. е. переводит спектр из диапазона $[0;\, f_d)$ в симметричный $[-f_d/2;\, f_d/2)$, где он совпадает со спектром аналогового сигнала. Такое соответствие однозначно, потому что выполнено условие теоремы Котельникова: $f_d = 8000 \ge 2 f_2 = 3000$ Гц, и наложения спектров нет.
- **Сравнение восстановленного и исходного сигналов.** Визуально сигналы совпадают на всем интервале, кривая восстановленного сигнала проходит через все отсчеты $x(nT)$ и плавно - между ними; численное расхождение $\sim 10^{-14}$ - погрешность вычислений, мнимая часть того же порядка. Восстановление точное, так как сигнал содержит только частоты ниже $f_d/2$ и на интервале $NT$ укладывается целое число периодов обеих гармоник.

#### 7. Равенство Парсеваля

```python
E_time = np.sum(np.abs(xn) ** 2)                 # энергия во временной области
E_freq = np.sum(np.abs(X) ** 2) / N              # энергия по отсчетам ДПФ
print(f'sum |x(n)|^2       = {E_time:.6f}')
print(f'(1/N) sum |X(k)|^2 = {E_freq:.6f}')
print(f'разность           = {abs(E_time - E_freq):.1e}')
print(f'средняя мощность E/N = {E_time/N:.6f};  (A1^2 + A2^2)/2 = {(A1**2 + A2**2)/2:.6f}')
for kk in (8, 12):
    print(f'доля энергии гармоники k = {kk} (с зеркальным отсчетом): {2*np.abs(X[kk])**2/N/E_freq:.4f}')
```

Равенство Парсеваля выполняется: $\sum |x(n)|^2 = \frac{1}{N}\sum |X(k)|^2 = 186.624$ (разность $\sim 10^{-13}$).

**Смысл:** энергия последовательности на периоде одинакова при вычислении во временной области и по ее спектру; величина $|X(k)|^2/N$ показывает, какая часть энергии приходится на частоту $\hat{\omega}_k$. Здесь вся энергия сосредоточена в четырех отсчетах ДПФ: гармоника $A_2$ несет 80%, гармоника $A_1$ - 20% энергии (отношение квадратов амплитуд $A_2^2 : A_1^2 = 4 : 1$). Средняя мощность $186.624/64 = 2.916$ равна сумме мощностей гармоник $(A_1^2 + A_2^2)/2$.
