## Основные теоретические положения

**Линейная дискретная система (ЛДС).** Дискретной называют систему, преобразующую входной дискретный сигнал $x(n)$ в выходной дискретный сигнал $y(n)$: $y(n)=F\{x(n)\}$. Система с линейным оператором $F$ называется линейной дискретной системой. Для линейного оператора выполняется принцип суперпозиции: сумме воздействий соответствует сумма реакций на эти воздействия.

**Импульсная характеристика (ИХ).** Импульсной характеристикой $h(n)$ называется реакция ЛДС на цифровой единичный импульс $u_0(n)$ при ННУ.

**Соотношение вход/выход.** Линейный оператор $F$ имеет две универсальные формы представления.

1. *Формула свертки.* Для воздействия длины $N_x$ и ИХ длины $N_h$ длина свертки равна $L=N_x+N_h-1$:

$$y(n)=\sum_{m=0}^{L-1}x(m)\,h(n-m).$$


2. *Разностное уравнение (РУ):*

$$y(n)=\sum_{i=0}^{N-1}b_i\,x(n-i)-\sum_{k=1}^{M-1}a_k\,y(n-k).$$

Коэффициенты $b_i$, $a_k$ — параметры ЛДС.

**Рекурсивные и нерекурсивные системы.**

- ЛДС называется **рекурсивной**, если хотя бы один из коэффициентов $a_k$ ($k\ge1$) не равен нулю. Ее реакция зависит не только от воздействия, но и от предшествующих отсчетов самой реакции. **Порядок** рекурсивной ЛДС равен максимальной задержке реакции.
- ЛДС называется **нерекурсивной**, если все $a_k=0$. Тогда $y(n)=\sum_{i=0}^{N-1}b_ix(n-i)$, а порядок равен максимальной задержке воздействия.

**Рекурсивные звенья 1-го и 2-го порядков** — простейшие рекурсивные ЛДС соответствующего порядка. Их разностные уравнения: для звена 1-го порядка

$$y(n)=b_0x(n)+b_1x(n-1)-a_1y(n-1),$$

для звена 2-го порядка

$$y(n)=b_0x(n)+b_1x(n-1)+b_2x(n-2)-a_1y(n-1)-a_2y(n-2).$$

**Конечная и бесконечная ИХ. Память.** Нерекурсивная ЛДС имеет конечную импульсную характеристику (КИХ), причем $h(i)=b_i$. Рекурсивная ЛДС имеет бесконечную импульсную характеристику (БИХ). У нерекурсивной ЛДС память конечна, а у рекурсивной из-за обратной связи любой отсчет воздействия присутствует в системе, постепенно затухая, бесконечно долго.

**Устойчивость.** ЛДС называется устойчивой, если при ограниченном воздействии реакция ограничена. Первый критерий: для устойчивости необходима и достаточна абсолютная сходимость ряда отсчетов ИХ,

$$\sum_{n=0}^{\infty}|h(n)|<\infty.$$

Нерекурсивные ЛДС устойчивы по определению, рекурсивные требуют проверки.

**Передаточная функция.** Передаточной функцией $H(z)$ называется отношение $z$-изображения реакции к $z$-изображению воздействия при ННУ; она равна $z$-изображению ИХ: $H(z)=Y(z)/X(z)=\sum_{n=0}^{\infty}h(n)z^{-n}$. Для рекурсивной ЛДС

$$H(z)=\frac{\sum_{i=0}^{N-1}b_iz^{-i}}{1+\sum_{k=1}^{M-1}a_kz^{-k}}.$$

**Нули и полюсы.** Нулями называют значения $z$, при которых $H(z)=0$, полюсами — значения $z$, при которых знаменатель $H(z)$ равен нулю. Для их нахождения числитель и знаменатель умножают на $z$ в степени, равной порядку знаменателя, и находят корни многочленов. Совокупность нулей и полюсов, изображенная на $z$-плоскости вместе с единичной окружностью, называется картой нулей и полюсов.

**Устойчивость (второй критерий).** ЛДС устойчива тогда и только тогда, когда все полюсы передаточной функции лежат внутри единичного круга: $|p_k|<1$.

**Частотные характеристики.** Частотная характеристика $H(e^{j\hat\omega})$ получается подстановкой $z=e^{j\hat\omega}$ в $H(z)$ и совпадает с ней на единичной окружности. Она показывает, как система изменяет амплитуду и фазу дискретного гармонического воздействия в установившемся режиме. Модуль

$$A(\hat\omega)=\left|H(e^{j\hat\omega})\right|$$

называется амплитудно-частотной характеристикой (АЧХ), аргумент

$$\varphi(\hat\omega)=\arg H(e^{j\hat\omega})$$

— фазочастотной характеристикой (ФЧХ). Свойства: АЧХ и ФЧХ непрерывны и периодичны с периодом $2\pi$; при вещественных коэффициентах АЧХ четная, а ФЧХ нечетная, поэтому их рассчитывают в основной полосе частот $\hat\omega\in[0;\pi]$.

**Дискретное преобразование Фурье (ДПФ)** и обратное ДПФ (ОДПФ) $N$-точечной последовательности:

$$X(k)=\sum_{n=0}^{N-1}x(n)e^{-j\frac{2\pi}{N}nk},\qquad x(n)=\frac1N\sum_{k=0}^{N-1}X(k)e^{j\frac{2\pi}{N}nk}.$$

**Равенство Парсеваля** позволяет определить энергию дискретного сигнала как по отсчетам во времени, так и по отсчетам ДПФ:

$$\sum_{n=0}^{N-1}|x(n)|^2=\frac1N\sum_{k=0}^{N-1}|X(k)|^2.$$

**1. Вычислить импульсную характеристику длины $N_1$ по заданной передаточной функции с коэффициентами $b_i, a_k$ и построить ее график. Записать аналитическую формулу импульсной характеристики рекурсивного звена 2-го порядка с учетом нулевых начальных условий. Пояснить, чему в действительности равна длина импульсной характеристики.**

Аналитическая формула импульсной характеристики рекурсивного звена 2-го порядка при нулевых начальных условиях:

$$h(n)=\begin{cases}
b_0\,r_p^{\,n}\dfrac{\sin[(n+1)\varphi_p]}{\sin\varphi_p}, & n=0;\\[3mm]
b_0\,r_p^{\,n}\dfrac{\sin[(n+1)\varphi_p]}{\sin\varphi_p}+b_1\,r_p^{\,n-1}\dfrac{\sin(n\varphi_p)}{\sin\varphi_p}, & n=1;\\[3mm]
b_0\,r_p^{\,n}\dfrac{\sin[(n+1)\varphi_p]}{\sin\varphi_p}+b_1\,r_p^{\,n-1}\dfrac{\sin(n\varphi_p)}{\sin\varphi_p}+b_2\,r_p^{\,n-2}\dfrac{\sin[(n-1)\varphi_p]}{\sin\varphi_p}, & n\ge2,
\end{cases}$$

где $r_p=\sqrt{a_2}=\sqrt{0.718}\approx0.8473$, $\ \varphi_p=\arccos\left(\dfrac{-a_1}{2r_p}\right)\approx0.8622$ рад $\approx0.2745\pi$.

```python
import numpy as np
import matplotlib.pyplot as plt

# Номер варианта
N_var = 13

# Коэффициенты числителя передаточной функции H(z) по таблице вариантов
b0 = 0.5 + 0.02 * N_var
b1 = b0 * ((-1)**(N_var + 1)) * (0.9822 + 0.0178 * N_var)
b2 = b0 * (0.8 + 0.2 * (N_var % 5))

# Коэффициенты знаменателя H(z)
a0 = 1
a1 = ((-1)**N_var) * (0.7778 + 0.025 * N_var)
a2 = 0.64 + 0.006 * N_var

b = [b0, b1, b2]
a = [a0, a1, a2]

# N1 - число отсчетов импульсной характеристики
N1 = N_var % 10 + 20

print(f"b = {b}")
print(f"a = {a}")
print(f"N1 = {N1}")

# Подаем единичный импульс длины N1 на вход фильтра
delta = np.zeros(N1)
delta[0] = 1

# Инициализируем массив импульсной характеристики h(n)
h = np.zeros(N1)

# Решение разностного уравнения при нулевых начальных условиях
for n in range(N1):
    sum_b = 0.0
    for i in range(len(b)):
        if n - i >= 0:
            sum_b += b[i] * delta[n - i]
            
    sum_a = 0.0
    for k in range(1, len(a)):
        if n - k >= 0:
            sum_a += a[k] * h[n - k]
            
    h[n] = sum_b - sum_a

# Построение графика импульсной характеристики
plt.figure(figsize=(10, 4))
plt.stem(range(N1), h, basefmt="k-")
plt.title(f"Импульсная характеристика h(n)")
plt.xlabel("n")
plt.ylabel("h(n)")
plt.grid(True)
plt.show()
```

**Длина импульсной характеристики.** Звено рекурсивное, поэтому его импульсная характеристика теоретически бесконечна: не существует номера $n$, после которого $h(n)$ тождественно равна нулю.

$$x(n) = \begin{cases} 1, & 0 \leqslant n < int(N_2/2) \\ 0, & int(N_2/2) \leqslant n \leqslant N_2 - 1 \end{cases}$$

где функция $int(t)$ — округление до ближайшего целого. Вывести графики воздействия и реакции. Пояснить:

Разностное уравнение рекурсивного звена 2-го порядка:

$$y[n] = b_0 x[n] + b_1 x[n-1] + b_2 x[n-2] - a_1 y[n-1] - a_2 y[n-2]$$

Подставляя значения коэффициентов:

$$y[n] = 0.76 \, x[n] + 0.922336 \, x[n-1] + 1.064 \, x[n-2] + 1.1028 \, y[n-1] - 0.718 \, y[n-2]$$

```python
# Реакция по разностному уравнению вычисляется на длине воздействия
Ny2 = len(x)
y2 = np.zeros(Ny2)

b0, b1, b2 = b[0], b[1], b[2]
a0, a1, a2 = a[0], a[1], a[2]

for n in range(Ny2):
    # Берем элементы входного сигнала, проверяя границы
    x_n  = x[n]
    x_n1 = x[n-1] if (n - 1 >= 0) else 0.0
    x_n2 = x[n-2] if (n - 2 >= 0) else 0.0
    
    # Берем предыдущие отсчеты выхода y2, проверяя границы
    y_n1 = y2[n-1] if (n - 1 >= 0) else 0.0
    y_n2 = y2[n-2] if (n - 2 >= 0) else 0.0
    
    # Вычисляем текущее значение отсчета реакции
    y2[n] = (b0 * x_n + b1 * x_n1 + b2 * x_n2) - (a1 * y_n1 + a2 * y_n2)

n_y2 = np.arange(Ny2)

# Построение графиков воздействия и реакции y2(n)
plt.figure(figsize=(12, 8))

plt.subplot(2, 1, 1)
plt.stem(n_x, x, basefmt="k-")
plt.title(f"Дискретное прямоугольное воздействие x(n)")
plt.xlabel("n")
plt.ylabel("x(n)")
plt.grid(True)

plt.subplot(2, 1, 2)
plt.stem(n_y2, y2, basefmt="k-")
plt.title("Реакция системы y2(n)")
plt.xlabel("n")
plt.ylabel("y2(n)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Для сравнения реакция y1 обрезается до длины воздействия N2
y1_cut = y1[:N2]
# Разность двух способов расчета реакции
diff = y1_cut - y2

plt.figure(figsize=(12, 8))

plt.subplot(2, 1, 1)
plt.stem(n_y2, y2, linefmt="C0-", markerfmt="C0o", basefmt="k-", label="y2(n), разностное уравнение")
plt.plot(n_y2, y1_cut, "rx", markersize=8, label="y1(n), свертка")
# Вертикальная линия n = N1: с этого номера для свертки не хватает отсчетов h(n)
plt.axvline(N1, color="g", linestyle="--", label=f"n = N1 = {N1}")
plt.title("Сравнение реакций y1(n) и y2(n)")
plt.xlabel("n")
plt.ylabel("y(n)")
plt.legend()
plt.grid(True)

plt.subplot(2, 1, 2)
plt.stem(n_y2, diff, basefmt="k-")
plt.axvline(N1, color="g", linestyle="--")
plt.title("Разность y1(n) - y2(n)")
plt.xlabel("n")
plt.ylabel("Δy(n)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

**4. Вычислить нули и полюсы передаточной функции в алгебраической и показательной формах, пояснить связь между ними. Выразить значения аргументов относительно $\pi$. Представить передаточную функцию в виде произведения простейших множителей.**

Передаточная функция при $N_{var} = 13$:

$$H(z)=\frac{0.76+0.922336\,z^{-1}+1.064\,z^{-2}}{1-1.1028\,z^{-1}+0.718\,z^{-2}}$$

**Связь форм.** Алгебраическая форма $z=\xi+j\eta$, показательная форма $z=re^{j\varphi}$. Радиус является модулем, а угол — аргументом комплексной переменной $z$:

$$r=|z|=\sqrt{\xi^2+\eta^2},\qquad \varphi=\arg(z)=\mathrm{arctg}\left(\frac{\eta}{\xi}\right).$$

Для комплексно-сопряженных полюсов $z_{*1,2}=r_*e^{\pm j\varphi_*}$:

$$r_*=\sqrt{a_2},\qquad \varphi_*=\arccos\left(\frac{-a_1}{2r_*}\right).$$

```python
# np.roots находит корни многочлена по списку его коэффициентов
zeros = np.roots(b) # корни b0*z^2 + b1*z + b2
poles = np.roots(a) # корни z^2 + a1*z + a2

print("Нули (алгебраическая форма):", zeros)
print("Полюсы (алгебраическая форма):", poles)

# Показательная форма: модуль и аргумент
for name, roots in (("Нули", zeros), ("Полюсы", poles)):
    for i, r in enumerate(roots, start=1):
        mod = np.abs(r)
        arg = np.angle(r)
        print(f"{name} {i}: |z| = {mod:.4f}, arg = {arg:.4f} рад = {arg / np.pi:.4f}*pi")
```

```python
# Радиус и угол комплексно-сопряженных полюсов по формулам
r_p = np.sqrt(a2)
phi_p = np.arccos(-a1 / (2 * r_p))

print(f"r* = {np.sqrt(a2):.4f}")
print(f"phi* = {phi_p:.4f} рад = {phi_p / np.pi:.4f}*pi")
```

| $p_{1,2}$ | $0.5514\pm j\,0.6434$ | $\sqrt{a_2}=\sqrt{0.718}\approx0.8473$ | $\pm0.8622$ | $\pm0.275\pi$ |

$$z_{1,2}=1.1832\,e^{\pm j\,0.671\pi},\qquad p_{1,2}=0.8473\,e^{\pm j\,0.275\pi}$$

**Передаточная функция через простейшие множители:**
Домножив числитель и знаменатель $H(z)$ на $z^2$ и вынеся $b_0$, получаем

$$H(z)=b_0\frac{z^2+\frac{b_1}{b_0}z+\frac{b_2}{b_0}}{z^2+a_1z+a_2}.$$

Квадратный трехчлен с корнями $z_1,z_2$ раскладывается как $(z-z_1)(z-z_2)$, поэтому

$$H(z)=b_0\frac{(z-z_1)(z-z_2)}{(z-p_1)(z-p_2)}=b_0\frac{(1-z_1z^{-1})(1-z_2z^{-1})}{(1-p_1z^{-1})(1-p_2z^{-1})},$$

$$H(z)=0.76\,\frac{\left(1-1.1832e^{j0.671\pi}z^{-1}\right)\left(1-1.1832e^{-j0.671\pi}z^{-1}\right)}{\left(1-0.8473e^{j0.275\pi}z^{-1}\right)\left(1-0.8473e^{-j0.275\pi}z^{-1}\right)}$$

**6. Вычислить АЧХ и ФЧХ в шкале нормированных частот $\hat\omega$ и вывести их графики. Пояснить: а) границы основной полосы частот; б) соответствие между картой нулей и полюсов и видом АЧХ; в) какому значению АЧХ соответствует скачок на $\pi$, если он имеется; г) какие частотные составляющие воздействия оказались преимущественно подавленными в реакции.**

```python
# Основная полоса [0; pi]
w = np.linspace(0, np.pi, 4001)
z = np.exp(1j * w)

# Передаточная функция H(z) при z = exp(j*w)
H = (b[0] + b[1] / z + b[2] / z**2) / (a[0] + a[1] / z + a[2] / z**2)

A = np.abs(H) # АЧХ
phi = np.angle(H) # ФЧХ, значения в (-pi; pi]

i_max = np.argmax(A)
i_min = np.argmin(A)
i_cut = np.where(A >= 1)[0][-1]
i_jump = np.where(np.abs(np.diff(phi)) > np.pi)[0]

print(f"A(0) = {A[0]:.4f}")
print(f"A(pi) = {A[-1]:.4f}")
print(f"Максимум АЧХ: {A[i_max]:.4f} при w = {w[i_max] / np.pi:.4f}*pi")
print(f"Минимум АЧХ: {A[i_min]:.4f} при w = {w[i_min] / np.pi:.4f}*pi")
print(f"A >= 1 при w <= {w[i_cut] / np.pi:.4f}*pi")

ticks = [0, np.pi / 4, np.pi / 2, 3 * np.pi / 4, np.pi]
labels = ["0", r"$\pi/4$", r"$\pi/2$", r"$3\pi/4$", r"$\pi$"]
zero_arg = np.abs(np.angle(zeros[0]))

plt.figure(figsize=(12, 8))

plt.subplot(2, 1, 1)
plt.plot(w, A)
# Вертикальные линии - частоты полюса и нуля, для сопоставления с видом АЧХ
plt.axvline(phi_p, color="g", linestyle="--", label=f"частота полюса {phi_p / np.pi:.4f}π")
plt.axvline(zero_arg, color="r", linestyle="--", label=f"частота нуля {zero_arg / np.pi:.4f}π")
plt.title("АЧХ")
plt.xlabel(r"$\hat{\omega}$, рад")
plt.ylabel(r"$A(\hat{\omega})$")
plt.xticks(ticks, labels)
plt.legend()
plt.grid(True)

plt.subplot(2, 1, 2)
# ФЧХ - это arg H в диапазоне (-pi; pi]
plt.plot(w, phi)
plt.title("ФЧХ")
plt.xlabel(r"$\hat{\omega}$, рад")
plt.ylabel(r"$\varphi(\hat{\omega})$, рад")
plt.xticks(ticks, labels)
plt.grid(True)

plt.tight_layout()
plt.show()
```

Частотная характеристика получается подстановкой $z=e^{j\hat\omega}$:

$$H(e^{j\hat\omega})=\frac{b_0+b_1e^{-j\hat\omega}+b_2e^{-j2\hat\omega}}{1+a_1e^{-j\hat\omega}+a_2e^{-j2\hat\omega}},\qquad A(\hat\omega)=\left|H(e^{j\hat\omega})\right|,\quad \varphi(\hat\omega)=\arg H(e^{j\hat\omega})$$

**а) Границы основной полосы частот.** Частотные характеристики периодичны с периодом $2\pi$. При вещественных коэффициентах АЧХ четная, а ФЧХ нечетная, поэтому их достаточно рассчитать в основной полосе $\hat\omega\in[0;\pi]$.

**б) Соответствие между картой нулей и полюсов и видом АЧХ.**

- **максимум** находится приблизительно на частоте полюса: $\varphi_p\approx0.275\pi$, $A_{max}\approx10.04$ при $\hat\omega\approx0.265\pi$. Радиус $r_*=0.847$ близок к единице, поэтому максимум достаточно крутой;
- **минимум** находится приблизительно на частоте нуля: $\varphi_z\approx0.671\pi$, $A_{min}\approx0.129$ при $\hat\omega\approx0.681\pi$. Нуль не лежит на единичной окружности ($r_z=1.183\ne1$), поэтому это минимум, а не нуль АЧХ;
- **на границах полосы** значения определяются подстановкой $z=1$ и $z=-1$:

$$A(0)=\left|\frac{b_0+b_1+b_2}{1+a_1+a_2}\right|\approx4.46,\qquad A(\pi)=\left|\frac{b_0-b_1+b_2}{1-a_1+a_2}\right|\approx0.32.$$

**в) Значение АЧХ, соответствующее скачку на $\pi$.** Скачок ФЧХ на $\pi$ соответствует нулю АЧХ, $A(\hat\omega)=0$. В такой точке амплитудная функция меняет знак. Нуль АЧХ возникает на частоте нуля передаточной функции, лежащего на единичной окружности. Здесь $|z_{1,2}|=1.183\ne1$, поэтому нулей АЧХ нет и скачков на $\pi$ нет.

**г) Подавленные частотные составляющие.** Амплитуда гармоники на выходе равна $C_y=A(\hat\omega)\,C_x$. Поэтому гармоника усиливается при $A>1$ и ослабляется при $A<1$:

- при $0\le\hat\omega<0.48\pi$ выполняется $A>1$: низкие частоты усиливаются, от $A(0)\approx4.46$ до максимума $10.04$ при $\hat\omega\approx0.265\pi$;
- при $0.49\pi<\hat\omega\le\pi$ выполняется $A<1$: высокие частоты подавляются, в области $0.6\pi\le\hat\omega\le\pi$ значения всего $0.13\dots0.32$.

Граница $A=1$ лежит между $0.48\pi$ и $0.49\pi$ (по расчету $\hat\omega\approx0.484\pi$).

В реакции преимущественно подавлены высокочастотные составляющие воздействия.

**1. Вычислить амплитудный и фазовый спектры периодической последовательности $x(n)$ с периодом $N$**:

$$x(nT)=A_1\cos(2\pi f_1nT+\pi/4)+A_2\cos(2\pi f_2nT+\pi/8)$$

Вывести графики $x(n)$ на периоде $N$ в шкале дискретного и дискретного нормированного времени.

**Параметры при $N_{var}=13$:**

- $N=64$
- $f_s=2000\,(13\bmod5+1)=2000\cdot4=8000$ Гц
- $T=1/f_s=0.125$ мс
- $A_1=1+0.01\cdot13=1.13$, $A_2=2A_1=2.26$
- $f_1=f_s/8=1000$ Гц, $f_2=1.5f_1=1500$ Гц

**Нормированные частоты:**

$$\hat\omega_1=2\pi f_1T=\frac{\pi}{4},\qquad \hat\omega_2=2\pi f_2T=\frac{3\pi}{8}$$

**Спектр.** ДПФ периодической последовательности:

$$X(k)=\sum_{n=0}^{N-1}x(n)e^{-j\frac{2\pi}{N}nk},\qquad k=0,\dots,N-1.$$

Спектр $X_p(k)=\dfrac1N X(k)$. Амплитудный спектр:

$$A(k)=\begin{cases}\dfrac1N\,|X(k)|, & k=0;\\[2mm] \dfrac2N\,|X(k)|, & k\ne0,\end{cases}$$

фазовый спектр

$$\varphi(k)=\arg\{X(k)\}.$$

```python
# Параметры периодической последовательности
N = 64
fs = 2000 * (N_var % 5 + 1)
T = 1 / fs
A1 = 1 + 0.01 * N_var
A2 = 2 * A1
f1 = fs / 8
f2 = 1.5 * f1

n_p = np.arange(N)
t_p = n_p * T
x_p = A1 * np.cos(2 * np.pi * f1 * t_p + np.pi / 4) + A2 * np.cos(2 * np.pi * f2 * t_p + np.pi / 8)

plt.figure(figsize=(12, 8))

plt.subplot(2, 1, 1)
plt.stem(t_p * 1e3, x_p, basefmt="k-")  # время в мс
plt.title("Последовательность x(nT) на периоде N (дискретное время)")
plt.xlabel("nT, мс")
plt.ylabel("x(nT)")
plt.grid(True)

plt.subplot(2, 1, 2)
plt.stem(n_p, x_p, basefmt="k-")
plt.title("Последовательность x(n) на периоде N (дискретное нормированное время)")
plt.xlabel("n")
plt.ylabel("x(n)")
plt.grid(True)

plt.tight_layout()
plt.show()

# ДПФ, без множителя 1/N
X = np.fft.fft(x_p)
k_p = np.arange(N)

# Амплитудный спектр
amp = 2 * np.abs(X) / N
amp[0] = np.abs(X[0]) / N

# Фазовый спектр
phase = np.angle(X)
phase[amp < 1e-9] = 0.0

plt.figure(figsize=(12, 8))

plt.subplot(2, 1, 1)
plt.stem(k_p, amp, basefmt="k-")
plt.title("Амплитудный спектр A(k)")
plt.xlabel("k")
plt.ylabel("A(k)")
plt.grid(True)

plt.subplot(2, 1, 2)
plt.stem(k_p, phase, basefmt="k-")
plt.title("Фазовый спектр φ(k)")
plt.xlabel("k")
plt.ylabel("arg X(k), рад")
plt.grid(True)

plt.tight_layout()
plt.show()
```

$$X_a(k)=\begin{cases}X(N+k), & -N/2\le k\le-1;\\ X(k), & 0\le k\le N/2-1,\end{cases}$$

задавая значения времени $t$ на интервале $t\in[0;(N-1)T]$ с шагом $\Delta t=0.25T$. Вывести графики: а) периодической последовательности $x(n)$ и модуля ее ДПФ; б) восстановленного аналогового сигнала $x(t)$ и его амплитудного спектра; в) исходного аналогового сигнала $x(t)=x(nT)\big|_{nT=t}$.

```python
k_a = np.arange(-N // 2, N // 2) # k = -N/2 ... N/2 - 1
X_a = X[k_a % N] # k < 0: X(N + k); k >= 0: X(k)

def restore(t_values):
    '''Восстановление x(t) по отсчетам X_a(k).'''
    x = np.zeros(len(t_values), dtype=complex)
    for i, t in enumerate(t_values): # для каждого момента времени
        s = 0
        for j, k in enumerate(k_a): # суммируем по всем гармоникам
            s += X_a[j] * np.exp(1j * 2 * np.pi / (N * T) * k * t)
        x[i] = s / N
    return x

def original(t_values):
    '''Исходный аналоговый сигнал.'''
    return (A1 * np.cos(2 * np.pi * f1 * t_values + np.pi / 4) + A2 * np.cos(2 * np.pi * f2 * t_values + np.pi / 8))

# t от 0 до (N-1)T с шагом 0.25T
dt = 0.25 * T
t_a = np.arange(4 * (N - 1) + 1) * dt
x_t_complex = restore(t_a)
# Мнимая часть равна нулю с точностью до ошибки вычислений
x_t = x_t_complex.real
x_t_orig = original(t_a)

# Амплитудный спектр восстановленного сигнала
t_full = np.arange(4 * N) * dt
x_full = restore(t_full).real
Y = np.fft.fft(x_full)
M = len(x_full)
amp_t = 2 * np.abs(Y) / M
amp_t[0] = np.abs(Y[0]) / M
f_t = np.arange(M) / (M * dt)
half = f_t <= 1 / (2 * dt)
idx = np.where((amp_t > 1e-9) & half)[0]

fig, axs = plt.subplots(3, 2, figsize=(14, 11))

axs[0, 0].stem(n_p, x_p, basefmt="k-")
axs[0, 0].set_title("а) Периодическая последовательность x(n)")
axs[0, 0].set_xlabel("n")
axs[0, 0].set_ylabel("x(n)")

axs[0, 1].stem(k_p, np.abs(X), basefmt="k-")
axs[0, 1].set_title("а) Модуль ДПФ |X(k)|")
axs[0, 1].set_xlabel("k")
axs[0, 1].set_ylabel("|X(k)|")

axs[1, 0].plot(t_a * 1e3, x_t)
axs[1, 0].set_title("б) Восстановленный аналоговый сигнал x(t)")
axs[1, 0].set_xlabel("t, мс")
axs[1, 0].set_ylabel("x(t)")

axs[1, 1].stem(f_t[half], amp_t[half], basefmt="k-")
axs[1, 1].set_title("б) Амплитудный спектр восстановленного сигнала")
axs[1, 1].set_xlabel("f, Гц")
axs[1, 1].set_ylabel("A(f)")

axs[2, 0].plot(t_a * 1e3, x_t_orig, color="C1")
axs[2, 0].set_title("в) Исходный аналоговый сигнал x(t)")
axs[2, 0].set_xlabel("t, мс")
axs[2, 0].set_ylabel("x(t)")

axs[2, 1].plot(t_a * 1e3, x_t_orig, color="C1", linewidth=3, label="исходный")
axs[2, 1].plot(t_a * 1e3, x_t, "k--", linewidth=1, label="восстановленный")
axs[2, 1].stem(t_p[:N] * 1e3, x_p, linefmt="C2-", markerfmt="C2o", basefmt=" ", label="отсчеты x(nT)")
axs[2, 1].set_xlim(0, 4)
axs[2, 1].set_title("Сравнение на начальном участке 0...4 мс")
axs[2, 1].set_xlabel("t, мс")
axs[2, 1].set_ylabel("x(t)")
axs[2, 1].legend()

for ax in axs.ravel():
    ax.grid(True)

plt.tight_layout()
plt.show()
```

**6. Исходя из результатов п. 5, объяснить: а) связь модуля ДПФ последовательности со спектром аналогового сигнала; б) результат визуального сравнения восстановленного и исходного сигналов.**

**а) Связь модуля ДПФ со спектром аналогового сигнала.** Спектр дискретного сигнала равен сумме копий спектра аналогового сигнала, сдвинутых на $N$:

$$X_p(k)=\sum_{m=-\infty}^{\infty}X_{an}(k+mN),\qquad X_p(k)=\frac1NX(k),$$

где $X_{an}(k)$ — спектр аналогового сигнала.

**б) Результат сравнения сигналов.** Визуально восстановленный и исходный сигналы совпадают, линии накладываются на всем интервале. Восстановленная кривая проходит через все отсчеты $x(nT)$ и верно воспроизводит сигнал между ними.

**7. Проверить равенство Парсеваля. Пояснить его смысл.**

```python
# Энергия во временной области
energy_time = np.sum(np.abs(x_p)**2)
# Энергия по отсчетам ДПФ
energy_freq = np.sum(np.abs(X)**2) / N

energy_harm = N * (A1**2 + A2**2) / 2

print(f"Сумма |x(n)|^2 = {energy_time:.6f}")
print(f"(1/N) Сумма |X(k)|^2 = {energy_freq:.6f}")
print(f"Разность = {abs(energy_time - energy_freq):.2e}")
print(f"N (A1^2 + A2^2) / 2 = {energy_harm:.6f}")
```

Равенство Парсеваля

$$\sum_{n=0}^{N-1}|x(n)|^2=\frac1N\sum_{k=0}^{N-1}|X(k)|^2$$

выполняется: обе части равны $204.304$, разность имеет порядок ошибки округления.

**Смысл.** Энергию дискретного сигнала можно вычислить как по отсчетам во временной области, так и по отсчетам ДПФ в частотной области, при переходе к спектру энергия не теряется и не появляется заново.
