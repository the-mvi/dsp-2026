## Основные теоретические положения

Импульсной характеристикой $h(nT)$ линейной системы называется ее реакция на цифровой единичный импульс $u_0(nT)$ при нулевых начальных условиях.

ЛДС задают разностным уравнением - правилом, по которому очередной отсчет выхода вычисляется по текущему и прошлым отсчетам входа и прошлым отсчетам выхода:
$$
y(n)=\sum^{N-1}_{i=0}b_ix(n-i) - \sum^{M-1}_{k=1}a_ky(n-k).
$$

Для рекурсивного звена второго порядка разностное уравнение имеет вид: 
$$
y(n) = b_0x(n) + b_1x(n-1)+b_2x(n-2)-a_1y(n-1)-a_2y(n-2).
$$

Аналитическая формула импульсной характеристики звена второго порядка при нулевых начальных условиях и комплексно-сопряженных полюсах $p_{1,2}=r_*e^{\pm j\varphi_*}$:
$$
h(n)=b_0r_*^n\frac{\sin((n+1)\varphi_*)}{\sin \varphi_*}+b_1r_*^{n-1}\frac{\sin(n\varphi_*)}{\sin\varphi_*}+b_2r_*^{n-2}\frac{\sin((n-1)\varphi_*)}{\sin\varphi_*}, \quad n\geq 2.
$$
При этом $h(0)=b_0, h(1)=b_1-a_1b_0$, а $h(n) = 0$ при $n<0$.

Если система линейна и стационарна, то достаточно знать ее реакцию на один единичный импульс, чтобы найти реакцию на любой сигнал.

Операция
$$
y(n)=x(n)*h(n) = \sum^\infty_{m=0}h(m)x(n-m)
$$
называется сверткой, а сама функция - формулой свертки.

Если длина импульсной характеристики $N_h$, а длина воздействия $N_x$, то длина реакции
$$
N_y=N_x+N_h-1.
$$

Передаточной функцией $H(z)$ ЛДС называется отношение изображения реакции к $z$-изображению воздействия при нулевых начальных условиях:
$$
H(z)=\frac{Y(z)}{X(z)}=\mathcal{Z}\{h(n)\}.
$$

Передаточная функция рекурсивного звена второго порядка:
$$
H(z)=\frac{b_0+b_1z^{-1}+b_2z^{-2}}{1+a_1z^{-1}+a_2z^{-2}}.
$$

Нули - значения $z=z_0$, при которых ПФ равна нулю. Особые точки (полюса) - значения $z=z_*$, при которых знаменатель ПФ равен нулю.

Комплексные нули и полюса могут быть представлены в алгебраической и показательной формах:
$$
z=\operatorname{Re}(z) + j\operatorname{Im}(z) = re^{j\varphi},
$$
где $r=|z|$, $\varphi=\arg(z)$.

Передаточная функция также может быть представлена в виде произведения множителей, определяемых ее нулями и полюсами:
$$
H(z)=K \frac{\prod_i (z-z_{0i})}{\prod_k (z-z_{*k})},
$$
где $z_{0i}$ - нули, $z_{*k}$ - полюса, $K$ - коэффициент усиления. В произведении учитываются кратности корней.

Для того чтобы ЛДС была устойчива, необходимо и достаточно, чтобы все полюса ее ПФ находились внутри круга единичного радиуса комплексной $z$-плоскости.

В установившемся режиме устойчивая стационарная ЛДС преобразует синусоидальное воздействие в синусоидальную реакцию той же частоты. Изменение амплитуды и фазы определяется частотной характеристикой системы.

Частотная характеристика определяется из передаточной функции подстановкой $z=e^{j\hat{\omega}}$:
$$
H(e^{j\hat{\omega}})=H(z)\mid_{z=e^{j\hat{\omega}}}.
$$

Амплитудно-частотная и фазочастотная характеристики:
$$
A(\hat{\omega}) = |H(e^{j\hat\omega})|, \quad \varphi(\hat\omega) = \arg H(e^{j\hat\omega}).
$$

Для дискретного сигнала важна не частота сама по себе, а ее отношение к частоте дискретизации. Поэтому вводят нормированные частоты:
$$
\hat f = \frac{f}{f_{д}}=fT, \quad \hat\omega = \omega T = 2\pi fT.
$$

Основная полоса частот в соотвествующих шкалах:
$$
f \in \left[ 0; \frac{f_{д}}{2} \right], \quad \hat\omega \in [0;\pi].
$$

Дискретным преобразованием Фурье периодической последовательности $x_p(n)$ называется пара взаимно однозначных преобразований дискретных рядов Фурье для последовательностей во временной и частотной областях.

Прямое и обратное дискретные преобразования Фурье:
$$
X(k)=\sum^{N-1}_{n=0} x(n) \exp \left(-j \frac{2\pi}{N}nk\right),
$$
$$
x(n) = \frac{1}{N} \sum^{N-1}_{k=0}X(k)\exp \left( j\frac{2\pi}{N}nk \right).
$$
Здесь $n,k=0,1,...,N-1$.

Вводится нормированное время
$$
\hat t = \frac{t}{T}.
$$

В моменты дескретизации $t=nT$ получаем $\hat t = n$.

Отсчетам ДПФ соответствют нормированные и абсолютные частоты:
$$
\hat f_k = \frac{k}{N}, \qquad f_k=\frac{kf_{д}}{N}.
$$

Амплитудный спектр дискретного сигнала - модуль Фурье-изображения.

Фазовый спектр ДС - аргумент Фурье-изображения.

Для вещественной последовательности коэффициенты ДПФ на противоположных частотах комплексно сопряжены:

$$X(-k)=\overline{X(k)},\qquad |X(-k)|=|X(k)|.$$

Индексы понимаются по модулю N. При ненулевых коэффициентах фазы противоположны по модулю $2\pi$. При нулевом коэффициенте фаза не определена.

ДПФ конечной последовательности и одного периода периодической последовательности определяется одинаковыми формулами прямого и обратного преобразований Фурье.

Периодический аналоговый сигнал может быть восстановлен по коэффициентам ДПФ периодической последовательности. Для четного $N$ используется формула:
$$
x(t) = \frac{1}{N}\sum^{N/2-1}_{k=-N/2}X_a(k)\exp\left( j\frac{2\pi}{NT}kt \right),
$$
где
$$
X_a(k) = \begin{cases}
X(N+k),& -N/2 \leq k \leq -1,\\
X(k), & 0 \leq k \leq N/2-1.
\end{cases}
$$

Равество Парсеваля:
$$
\sum^{N-1}_{n=0}|x(n)|^2 = \frac{1}{N}\sum^{N-1}_{k=0}|X(k)|^2.
$$

Равенство выражает сохранение энергии последовательности при переходи из временной области в частотную: энергия, вычисленная по отсчетам сигнала, равна соответствующей нормированной сумме квадратов модулей его ДПФ.

```python
N_var = 8

b0 = 0.5 + 0.02 * N_var
b1 = b0 * (-1)**(N_var + 1) * (0.9822 + 0.0178 * N_var)
b2 = b0 * (0.8 + 0.2 * (N_var % 5))

a0 = 1
a1 = (-1)**N_var * (0.7778 + 0.025 * N_var)
a2 = 0.64 + 0.006 * N_var

N1 = N_var % 10 + 20
N2 = N_var % 10 + 30

f_d = 1000 * N_var

variables = {
    "N_вар": N_var,
    "b0": b0,
    "b1": b1,
    "b2": b2,
    "a0": a0,
    "a1": a1,
    "a2": a2,
    "N1": N1,
    "N2": N2,
    "f_д": f_d
}

for name, value in variables.items():
    print(f"{name} = {value:.6f}".rstrip("0").rstrip("."))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import lfilter

plt.rcParams.update({
    "figure.facecolor": "#232136",
    "axes.facecolor": "#232136",
    "axes.edgecolor": "#6e6a86",
    "axes.labelcolor": "#e0def4",
    "text.color": "#e0def4",
    "xtick.color": "#908caa",
    "ytick.color": "#908caa",
    "grid.color": "#393552",
    "grid.alpha": 0.7,
    "lines.linewidth": 2,
    "font.size": 11,
    "axes.grid": True,
    "figure.dpi": 120,
})

# Коэффициенты передаточной функции
b = np.array([b0, b1, b2])
a = np.array([a0, a1, a2])

print(
    f"H(z) = ({b[0]:g} {b[1]:+g}z^-1 {b[2]:+g}z^-2) / "
    f"({a[0]:g} {a[1]:+g}z^-1 {a[2]:+g}z^-2)"
)

# Единичный импульс длины N1
n = np.arange(N1)
delta = np.zeros(N1)
delta[0] = 1

# Импульсная характеристика при нулевых начальных условиях
h = lfilter(b, a, delta)

# Полюса передаточной функции
poles = np.roots(a)

print(f"p1 = {poles[0]:.6f}")
print(f"p2 = {poles[1]:.6f}")

r = np.abs(poles[0])
phi = np.angle(poles[np.argmax(np.imag(poles))])

print(f"Радиус полюсов: r = {r:.6f}")
print(f"Аргумент полюсов: phi = {phi:.6f} рад")
print(f"Аргумент относительно pi: phi = {phi / np.pi:.6f}pi")
print(f"h(0) = {h[0]:.6f}")
print(f"h(1) = {h[1]:.6f}")

# График импульсной характеристики
fig, ax = plt.subplots(figsize=(11, 5))

markerline, stemlines, baseline = ax.stem(n ,h)
plt.setp(markerline, color="#ebbcba", markersize=5)
plt.setp(stemlines, color="#c4a7e7", linewidth=1.8)
plt.setp(baseline, color="#6e6a86", linewidth=1)

ax.set_title(f"Импульсная характеристика h(n), N1 = {N1}")
ax.set_xlabel("Номер отсчета n")
ax.set_ylabel("h(n)")
ax.set_xticks(np.arange(0, N1, 2))
ax.set_xlim(-0.5, N1 -0.5)

plt.tight_layout()
plt.show()
```

#### 1.1. Анализ импульсной характеристики

Передаточная функция рекурсивного звена второго порядка имеет вид:
$$
H(z) = \frac{0.66-0.742236z^{-1}+0.924z^{-2}}{1+0.9778z^{-1}+0.688z^{-2}}.
$$

Полюса передаточной функции образуют комплексно-сопряженную пару:
$$
p_{1,2} = -0.4889 \pm j0.670057
$$

их модуль и аргумент:
$$
r_* = 0.829458, \quad \varphi_* = 2.201139 \approx 0.700644\pi.
$$

С учетом нулевых начальных условий аналитическая формула импульсной характеристики имеет вид:
$$
h(n)=
\begin{cases}
0, & n<0,\\
0.66, & n=0,\\
-1.387584, & n=1,\\
\displaystyle
\frac{
0.66r_*^n\sin((n+1)\varphi_*)
-0.742236r_*^{n-1}\sin(n\varphi_*)
+0.924r_*^{n-2}\sin((n-1)\varphi_*)
}{\sin\varphi_*}, & n\geq 2.
\end{cases}
$$

**Длина импульсной характеристики.**

На графике представлены первые $N_1=28$ отсчетов импульсной характеристики. Однако рекурсивное звено обладает бесконечной импульсной характеристикой, поскольку текущие значения выхода зависят от предыдущих.

Так как модули полюсов $|p_{1,2}| = 0.829458<1$, система устойчива, а значения импульсной характеристики затухают и стремятся к нулю при $n\to\infty$. Следовательно, $N_1=28$ задает лишь длину вычисляемого фрагмента, тогда как действительная длина импульсной характеристики бесконечна.

```python
# Прямоугольный импульс длины N2
n_x = np.arange(N2)
x = np.zeros(N2)

N_imp = round(N2 / 2)
x[:N_imp] = 1

# Вычисление реакции по формуле свертки
y1_full = np.convolve(x, h, mode="full")
n_full = np.arange(len(y1_full))

# Ограничение реакции до длины воздействия
y1 = y1_full[:N2]
n_y1 = np.arange(N2)

print(f"Длина воздействия x: {len(x)}")
print(f"Длина прямоугольного импульса: {N_imp}")
print(f"Длина полной свертки: {len(y1_full)}")
print(f"Длина ограниченной реакции y1: {len(y1)}")

# Построение графиков
fix, axes = plt.subplots(3, 1, figsize=(11, 11))

signals = [
    (n_x, x, "Воздействие x(n)"),
    (n_full, y1_full, "Полная свертка x(n) * h(n)"),
    (n_y1, y1, "Реакция y1(n), ограниченная до N2")
]

for ax, (n, signal, title) in zip(axes, signals):
    markerline, stemlines, baseline = ax.stem(n, signal)
    plt.setp(markerline, color="#ebbcba", markersize=5)
    plt.setp(stemlines, color="#c4a7e7", linewidth=1.8)
    plt.setp(baseline, color="#6e6a86", linewidth=1)

    ax.set_title(title)
    ax.set_xlabel("Номер отсчета n")
    ax.set_ylabel("Амплитуда")
    ax.set_xlim(-0.5, len(signal) - 0.5)

plt.tight_layout()
plt.show()
```

#### 1.2. Вычисление реакции по формуле свертки

Реакция линейной дискретной системы вычисляется по формуле свертки:
$$
y_1(n) = x(n)*h(n) = \sum_{m=0}^{\infty}h(m)x(n-m).
$$

В качестве воздействия используется прямоугольный импульс:
$$
x(n)=
\begin{cases}
1, & 0 \leq n < 19,\\
0, & 19 \leq n \leq 37.
\end{cases}
$$

**Длина воздействия.**

Массив воздействия содержит 38 отсчетов, из которых первые $N_2/2 = 19$ равны единице, а остальные - нулю.

Таким образлм, длительность ненулевой части прямоугольного импульса составляет 19 отсчетов.

**Длина свертки.**

При свертке двух конечных последовательностей длины $N_1$ и $N_2$ длина результата равна:
$$
N_y = N_1 + N_2 - 1.
$$

Для заданных значений:
$$
N_y = 28 + 38 - 1 = 65.
$$

Таким образом, вычисленная полная свертка содержит 65 отсчетов, что соответствует графику.

При этом поскольку ненулевая часть воздействия содержит только 19 отсчетов, потенциально ненулевая часть вычисленной свертки имеет длину $28+19-1=46$ отсчетов. Остальные 19 отсчетом равны нулю.

**Ограничение длины реакции.**

Полученную свертку ограничивают до $N_2=38$ отсчетов, чтобы рассматривать реакцию на том же временном интервале, что и воздействие, а также обеспечить возможность дальнейшего сравнения с реакцией, рассчитанной по разностному уравнению.

Поскольку рассматрвиаемая система рекурсивная, ее действительная импульсная характеристика бесконечна. Поэтому рассчитанная свертка длиной 65 отсчетов соответсвует использованному конечному фрагменту импульсной характеристики, а не полной реакции системы.

```python
# Вычисление реакции по разностному уравнению
n_y2 = np.arange(N2)
y2 = np.zeros(N2)

for i in range(N2):
    y2[i] = b0 * x[i]

    if i >= 1:
        y2[i] += b1 * x[i-1] - a1 * y2[i-1]

    if i >= 2:
        y2[i] += b2 * x[i-2] - a2 * y2[i-2]

print(f"Длина воздействия x: {len(x)}")
print(f"Длина реакции y2: {len(y2)}")
print(f"Максимальное отличие y1 и y2: {np.max(np.abs(y1 - y2)):.6f}")
print(f"Совпадают ли первые N1 отсчетов: {np.allclose(y1[:N1], y2[:N1])}")

# Построение графиков
fig, axes = plt.subplots(3, 1, figsize=(11, 11))

signals = [
    (n_x, x, "Воздействие x(n)"),
    (n_y2, y2, "Реакция y2(n) по разностному уравнению")
]

for ax, (n, signal, title) in zip(axes[:2], signals):
    markerline, stemlines, baseline = ax.stem(n, signal)
    plt.setp(markerline, color="#ebbcba", markersize=5)
    plt.setp(stemlines, color="#c4a7e7", linewidth=1.8)
    plt.setp(baseline, color="#6e6a86", linewidth=1)

    ax.set_title(title)
    ax.set_xlabel("Номер отсчета n")
    ax.set_ylabel("Амплитуда")
    ax.set_xlim(-0.5, N2 - 0.5)

axes[2].plot(n_y2, y1, color="#c4a7e7", label="y1(n) - свертка")
axes[2].plot(n_y2, y2, color="#ebbcba", linestyle="--",
             label="y2(n) - разностное уравнение")

axes[2].set_title("Сравнение реакций y1(n) и y2(n)")
axes[2].set_xlabel("Номер отсчета n")
axes[2].set_ylabel("Амплитуда")
axes[2].set_xlim(-0.5, N2 - 0.5)
axes[2].legend()

plt.tight_layout()
plt.show()
```

#### 1.3. Вычисление реакции по разностному уравнению

Реакция рекурсивного звена второго порядка вычисляется по разностному уравнению:
$$
y_2(n) = 0.66x(n) - 0.742236x(n-1) + 0.924x(n-2) - 0.9778y_2(n-1) - 0.688y_2(n-2).
$$

Вычисления выполняются при нулевых начальных условиях:
$$
x(n)=0, \quad n < 0,
$$
$$
y_2(-1)=y_2(-2)=0.
$$

В качестве воздействия используется прямоугольный импульс $x(n)$ из пункта 2.

**Длины воздействия и реакции.**

Длина воздействия составляет $N_2=38$ отсчетов. При этом первые 19 отсчетов равны единице, а остальные 19 - нулю.

Вычисленная по разностному уравнению реакция $y_2(n)$ также содержит 38 отсчетов, поскольку рассматривается на том же временном интервале, что и воздействие.

Однако действительная реакция рекурсивной системы имеет бесконечную длительность: после окончания ненулевого воздействия система продолжает формировать затухающие выходные отсчеты.

**Сравнение реакций $y_1(n)$ и $y_2(n)$**.

На графике сравнения реакции практически совпадают.

Первые 28 отсчетов совпадают с точностью вычислений. Начиная с отсчета $n=28$, появляются небольшие различия.

Это связано с тем, что при вычислении $y_1(n)$ методом свертки использовалась усеченная импульсная характеристика длиной $N_1=28$ отсчетов.

В отличие от свертки с усеченной импульсной характеристикой, разностное уравнение учитывает влияние предыдущих выходных отсчетов на всем рассматриваемом интервале.

Таким образом, оба способа вычисления реакции согласуются между собой, а возникающие небольшие различия объясняются ограничением длины импульсной характеристики при вычислении свертки.

```python
# Вычисление нулей и полюсов предаточной функции
zeros = np.roots(b)
poles = np.roots(a)

# Вывод в алгебраической и показательной формах
for name, roots, symbol in [
    ("Нули", zeros, "z"),
    ("Полюсы", poles, "p")
]:
    print(f"\n{name}:")

    for i, root in enumerate(roots, 1):
        r = np.abs(root)
        phi = np.angle(root)

        print(f"{symbol}{i} = {root.real:.6f} {root.imag:+.6f}j")
        print(f"    = {r:.6f} * e^({phi / np.pi:+.6f}jpi)")
        print(f"Аргумент: {phi:.6f} рад = {phi / np.pi:.6f}pi")

# Представление передаточной функции через простейшие множители
K = b[0] / a[0]

num = " * ".join(
    f"(1 - {np.abs(z):.6f}e^({np.angle(z) / np.pi:+.6f}jpi)z^-1)"
    for z in zeros
)

den = " * ".join(
    f"(1 - {np.abs(p):.6f}e^({np.angle(p) / np.pi:+.6f}jpi)z^-1)"
    for p in poles
)

print(f"\nКоэффициент усиления K = {K:.6f}")
print(f"H(z) = {K:.6f} * {num} / ({den})")
```

#### 1.4. Вычисление нулей и полюсов передаточной функции

Передаточная функция рекурсивного звена второго порядка:

$$
H(z)=\frac{0.66-0.742236z^{-1}+0.924z^{-2}}
{1+0.9778z^{-1}+0.688z^{-2}}.
$$

Умножая числитель и знаменатель на $z^2$, получаем:

$$
H(z)=\frac{0.66z^2-0.742236z+0.924}
{z^2+0.9778z+0.688}.
$$

**Нули передаточной функции.**

Нули определяются корнями числителя:

$$
0.66z^2-0.742236z+0.924=0.
$$

В алгебраической форме:

$$
z_{1,2}=0.562300\pm j1.041066.
$$

В показательной форме:

$$
z_{1,2}=1.183216e^{\pm j0.342365\pi}.
$$

Модуль и аргумент нулей:

$$
|z_{01}|=|z_{02}|=1.183216,
\quad
\arg z_{01}=0.342365\pi, \arg z_{02}=-0.342365\pi.
$$

**Полюсы передаточной функции.**

Полюсы определяются корнями знаменателя:

$$
z^2+0.9778z+0.688=0.
$$

В алгебраической форме:

$$
p_{1,2}=-0.488900\pm j0.670057.
$$

В показательной форме:

$$
p_{1,2}=0.829458e^{\pm j0.700644\pi}.
$$

Модуль и аргумент полюсов:

$$
r_*=0.829458,
\quad
\varphi_*=\pm2.201139\approx\pm0.700644\pi.
$$

**Связь алгебраической и показательной форм.**

Комплексное число можно представить в двух формах:

$$
z=a+jb=re^{j\varphi},
$$

где

$$
r=\sqrt{a^2+b^2},
\quad
\varphi=\arg(z).
$$

Обе формы описывают одну и ту же точку комплексной плоскости. Алгебраическая форма задает действительную и мнимую части, а показательная - модуль и аргумент.

**Представление передаточной функции через простейшие множители.**

Передаточная функция может быть представлена через ее нули и полюсы:

$$
H(z)=K\frac{(1-z_1z^{-1})(1-z_2z^{-1})}
{(1-p_1z^{-1})(1-p_2z^{-1})},
$$

где $K=b_0/a_0=0.66$.

Подставляя вычисленные значения в показательной форме, получаем:

$$
H(z)\approx0.66\cdot
\frac{
(1-1.183216e^{j0.342365\pi}z^{-1})
(1-1.183216e^{-j0.342365\pi}z^{-1})
}{
(1-0.829458e^{j0.700644\pi}z^{-1})
(1-0.829458e^{-j0.700644\pi}z^{-1})
}.
$$

Таким образом, передаточная функция имеет два комплексно-сопряженных нуля и два комплексно-сопряженных полюса. Нули и полюсы представлены в алгебраической и показательной формах, а их аргументы выражены относительно $\pi$.

```python
# Построение карты нулей и полюсов
fig, ax = plt.subplots(figsize=(8, 8))

# Единичная окружность
phi = np.linspace(0, 2 * np.pi, 500)
ax.plot(np.cos(phi), np.sin(phi), color="#908caa", linestyle="--", label="Единичная окружность")

# Нули передаточной функции
ax.scatter(zeros.real, zeros.imag, s=100, facecolors="none", edgecolors="#ebbcba", linewidths=2, label="Нули")

# Полюсы передаточной функции
ax.scatter(poles.real, poles.imag, s=100, marker="x", color="#c4a7e7", linewidths=2, label="Полюсы")

# Подписи точек
for i, z in enumerate(zeros, 1):
    ax.annotate(f"z{i}", (z.real, z.imag), xytext=(8, 8), textcoords="offset points")

for i, p in enumerate(poles, 1):
    ax.annotate(f"p{i}", (p.real, p.imag), xytext=(8, 8), textcoords="offset points")

# Оси координат
ax.axhline(0, color="#6e6a86", linewidth=1)
ax.axvline(0, color="#6e6a86", linewidth=1)

ax.set_title("Карта нулей и полюсов")
ax.set_xlabel("Re(z)")
ax.set_ylabel("Im(z)")
ax.set_aspect("equal")
ax.set_xlim(-1.5, 1.5)
ax.set_ylim(-1.5, 1.5)
ax.legend()

plt.tight_layout()
plt.show()

# Проверка устойчивости системы
print("Модули нулей:", np.abs(zeros))
print("Модули полюсов:", np.abs(poles))

if np.all(np.abs(poles) < 1):
    print("Система устойчива: все полюсы внутри единичной окружности")
else:
    print("Система неустойчива")
```


#### 1.5. Анализ карты нулей и полюсов

На карте нулей и полюсов комплексной $z$-плоскости нули передаточной функции обозначены окружностями, полюсы - крестиками. Пунктирная окружность соответствует единичной окружности $|z|=1$.

**Устойчивость рекурсивного звена.**

Для устойчивости рекурсивного звена необходимо и достаточно, чтобы все полюсы передаточной функции находились внутри единичной окружности:

$$
|p_i| < 1.
$$

В пункте 4 были вычислены полюсы:

$$
p_{1,2}=-0.488900\pm j0.670057.
$$

Их модули равны:

$$
|z_{*1}|=|z_{*2}|=0.829458<1.
$$

Оба полюса находятся внутри единичной окружности, что также видно на построенной карте.

Следовательно, **рекурсивное звено является устойчивым**.

Карта построена по массивам zeros и poles из пункта 4, поэтому изображает те же вычисленные корни. Нули лежат справа вне единичной окружности, их модуль равен 1.183216; полюсы лежат слева внутри окружности, их модуль равен 0.829458. Это согласуется с численными координатами пункта 4.

```python

from scipy.signal import freqz

# Вычисление частотной характеристики
w_hat, H_freq = freqz(b, a, worN=np.linspace(0, np.pi, 2000))

# Амплитудно-частотная и фазочастотная характеристики
A = np.abs(H_freq)
phase = np.angle(H_freq)

# Минимум и максимум АЧХ
i_min = np.argmin(A)
i_max = np.argmax(A)

print(f"Основная полоса частот: от 0 до pi рад")
print(f"АЧХ при w = 0: {A[0]:.6f}")
print(f"АЧХ при w = pi: {A[-1]:.6f}")

print(f"\nМинимум АЧХ: {A[i_min]:.6f} при w = {w_hat[i_min] / np.pi:.6f}pi")
print(f"Максимум АЧХ: {A[i_max]:.6f} при w = {w_hat[i_max] / np.pi:.6f}pi")

# Поиск скачка ФЧХ при переходе через границу ±pi
jumps = np.where(np.abs(np.diff(phase)) > np.pi)[0]

if len(jumps) > 0:
    i = jumps[0]
    w_jump = (w_hat[i] + w_hat[i + 1]) / 2
    A_jump = (A[i] + A[i + 1]) / 2

    print(f"\nСкачок ФЧХ при w ≈ {w_jump / np.pi:.6f}pi")
    print(f"Значение АЧХ в этой точке: {A_jump:.6f}")
else:
    print("\nСкачков ФЧХ нет")

# Построение графиков
fig, axes = plt.subplots(2, 1, figsize=(11, 9), sharex=True)

axes[0].plot(w_hat, A, color="#c4a7e7")
axes[0].set_title("Амплитудно-частотная характеристика")
axes[0].set_ylabel("АЧХ")

axes[1].plot(w_hat, phase, color="#ebbcba")
axes[1].set_title("Фазочастотная характеристика")
axes[1].set_xlabel(r"Нормированная частота $\hat{\omega}$")
axes[1].set_ylabel("ФЧХ, рад")

axes[1].set_yticks([-np.pi, -np.pi/2, 0, np.pi/2, np.pi])
axes[1].set_yticklabels([r"$-\pi$", r"$-\pi/2$", "0", r"$\pi/2$", r"$\pi$"])

axes[1].set_xticks([0, np.pi/4, np.pi/2, 3*np.pi/4, np.pi])
axes[1].set_xticklabels(["0", r"$\pi/4$", r"$\pi/2$", r"$3\pi/4$", r"$\pi$"])
axes[1].set_xlim(0, np.pi)

plt.tight_layout()
plt.show()
```


#### 1.6. Вычисление АЧХ и ФЧХ в шкале нормированных частот

Частотная характеристика системы определяется подстановкой $z=e^{j\hat{\omega}}$ в передаточную функцию:

$$
H(e^{j\hat{\omega}})=
\frac{0.66-0.742236e^{-j\hat{\omega}}+0.924e^{-j2\hat{\omega}}}
{1+0.9778e^{-j\hat{\omega}}+0.688e^{-j2\hat{\omega}}}.
$$

Амплитудно-частотная и фазочастотная характеристики определяются выражениями:

$$
A(\hat{\omega})=|H(e^{j\hat{\omega}})|,
$$

$$
\varphi(\hat{\omega})=\arg H(e^{j\hat{\omega}}).
$$

Для вычисления частотных характеристик использована функция `freqz`, а для получения амплитуды и фазы - функции `np.abs` и `np.angle`.

**Границы основной полосы частот.**

Основная полоса нормированных частот находится в пределах:

$$
0\leq\hat{\omega}\leq\pi.
$$

Таким образом, нижняя граница равна $0$, а верхняя - $\pi$ рад.

**Соответствие карты нулей и полюсов виду АЧХ.**

В пункте 4 были вычислены аргументы нулей и полюсов:

$$
\arg z_{01}=0.342365\pi,\qquad \arg z_{02}=-0.342365\pi,
\qquad
\varphi_*=\pm0.700644\pi.
$$

На полученном графике АЧХ наблюдается минимум:

$$
A_{\min}\approx0.127034,
\qquad
\hat{\omega}\approx0.333\pi.
$$

Минимум расположен вблизи аргумента нулей передаточной функции. Это объясняется тем, что нули ослабляют частотные составляющие, соответствующие их угловому расположению на комплексной плоскости.

Максимум АЧХ составляет:

$$
A_{\max}\approx6.830244,
\qquad
\hat{\omega}\approx0.713\pi.
$$

Максимум находится вблизи аргумента полюсов. Поскольку полюсы расположены внутри единичной окружности, но достаточно близко к ней, возникает выраженный резонансный пик.

Таким образом, расположение нулей и полюсов соответствует наблюдаемым минимуму и максимуму АЧХ.

**Скачок ФЧХ.**

На графике ФЧХ наблюдается скачок при нормированной частоте:

$$
\hat{\omega}\approx0.509\pi.
$$

Значение АЧХ в этой точке составляет:

$$
A(\hat{\omega})\approx0.846223.
$$

Этот скачок возникает из-за представления фазы в диапазоне $[-\pi;\pi]$. При достижении значения $-\pi$ фаза переходит к $+\pi$, поэтому на графике появляется разрыв величиной $2\pi$.

При этом АЧХ не обращается в нуль. Следовательно, скачок связан с ограничением диапазона отображаемой фазы, а не с исчезновением соответствующей частотной составляющей.

**Подавление частотных составляющих воздействия.**

На нулевой нормированной частоте значение АЧХ составляет:

$$
A(0)\approx0.315764.
$$

На верхней границе основной полосы частот:

$$
A(\pi)\approx3.275466.
$$

На низких частотах коэффициент передачи преимущественно меньше единицы, тогда как в области высоких частот наблюдается значительное усиление, особенно вблизи резонансного максимума.

Следовательно, **в реакции системы преимущественно подавляются низкочастотные составляющие воздействия**, а высокочастотные составляющие в значительной части основной полосы усиливаются.


```python

# Перевод нормированной частоты в Гц
f = w_hat * f_d / (2 * np.pi)

print(f"Частота дискретизации: fд = {f_d} Гц")
print(f"Основная полоса частот: от 0 до {f_d / 2:g} Гц")

# Соответствие нормированных и абсолютных частот
print("\nСоответствие частот:")
for w in [0, np.pi/4, np.pi/2, 3*np.pi/4, np.pi]:
    print(f"w = {w / np.pi:.2f}pi -> f = {w * f_d / (2 * np.pi):.0f} Гц")

# Минимум и максимум АЧХ в Гц
print(f"\nМинимум АЧХ: {A[i_min]:.6f} при f = {f[i_min]:.2f} Гц")
print(f"Максимум АЧХ: {A[i_max]:.6f} при f = {f[i_max]:.2f} Гц")

# Построение графиков
fig, axes = plt.subplots(2, 1, figsize=(11, 9), sharex=True)

axes[0].plot(f, A, color="#c4a7e7")
axes[0].set_title("Амплитудно-частотная характеристика")
axes[0].set_ylabel("АЧХ")

axes[1].plot(f, phase, color="#ebbcba")
axes[1].set_title("Фазочастотная характеристика")
axes[1].set_xlabel("Частота f, Гц")
axes[1].set_ylabel("ФЧХ, рад")

axes[1].set_yticks([-np.pi, -np.pi/2, 0, np.pi/2, np.pi])
axes[1].set_yticklabels([r"$-\pi$", r"$-\pi/2$", "0", r"$\pi/2$", r"$\pi$"])

axes[1].set_xticks(np.linspace(0, f_d / 2, 5))
axes[1].set_xlim(0, f_d / 2)

plt.tight_layout()
plt.show()
```


#### 1.7. Вычисление АЧХ и ФЧХ в шкале абсолютных частот

Для перехода от нормированной угловой частоты $\hat{\omega}$ к абсолютной частоте $f$ (Гц) используется соотношение:

$$
\hat{\omega} = \frac{2\pi f}{f_{д}},
\qquad
f = \frac{\hat{\omega}f_{д}}{2\pi}.
$$

**Границы основной полосы частот.**

Основная полоса абсолютных частот определяется выражением:

$$
0 \leq f \leq \frac{f_{д}}{2}.
$$

Подставляя частоту дискретизации, получаем:

$$
0 \leq f \leq 4000 \text{ Гц}.
$$

Таким образом, нижняя граница основной полосы равна $0$ Гц, а верхняя - $4000$ Гц (частота Найквиста).

**Соответствие нормированных и абсолютных частот.**

Для $f_{д}=8000$ Гц нормированные и абсолютные частоты связаны следующим образом:

| Нормированная частота $\hat{\omega}$ | Абсолютная частота $f$, Гц |
|---|---|
| $0$ | $0$ |
| $\pi/4$ | $1000$ |
| $\pi/2$ | $2000$ |
| $3\pi/4$ | $3000$ |
| $\pi$ | $4000$ |

На построенных графиках АЧХ и ФЧХ имеют тот же вид, что и в пункте 6. Изменился только масштаб горизонтальной оси: вместо нормированной частоты $\hat{\omega}$ используется абсолютная частота $f$ в герцах.

Минимум АЧХ наблюдается при частоте около $1331$ Гц, а максимум - около $2853$ Гц. Эти значения соответствуют минимуму и максимуму, найденным ранее в шкале нормированных частот.

Следовательно, переход к абсолютным частотам не изменяет характеристики системы, а лишь позволяет определить, каким физическим частотам соответствуют особенности АЧХ и ФЧХ.


```python
N_var = 8

N = 64
f_d = 2000 * (N_var % 5 + 1)
T = 1 / f_d

A1 = 1 + 0.01 * N_var
A2 = 2 * A1

f1 = f_d / 8
f2 = 1.5 * f1

variables = {
    "N_вар": N_var,
    "N": N,
    "f_д": f_d,
    "T": T,
    "A1": A1,
    "A2": A2,
    "f1": f1,
    "f2": f2
}

for name, value in variables.items():
    print(f"{name} = {value:.6f}".rstrip("0").rstrip("."))
```

```python
# Отсчеты времени
n = np.arange(N)
t = n * T
t_ms = t * 1000

# Вычисление последовательности x(n)
x2 = A1 * np.cos(2 * np.pi * f1 * t + np.pi/4) + A2 * np.cos(2 * np.pi * f2 * t + np.pi/8)

print(f"Длин последовательности: {len(x2)} отсчетов")
print(f"Период N*T: {N * T * 1000:g} мс")
print(f"Период дискретизации T: {T * 1000:g} мс")

# Построение графиков
fig, axes = plt.subplots(2, 1, figsize=(11, 9))

signals = [
    (t_ms, "Дискретное время nT, мс"),
    (n, "Дискретное нормированное время")
]

for ax, (time, xlabel) in zip(axes, signals):
    markerline, stemlines, baseline = ax.stem(time, x2)

    plt.setp(markerline, color="#ebbcba", markersize=5)
    plt.setp(stemlines, color="#c4a7e7", linewidth=1.8)
    plt.setp(baseline, color="#6e6a86", linewidth=1)

    ax.set_title("Последовательность x(n)")
    ax.set_xlabel(xlabel)
    ax.set_ylabel("Амплитуда")

plt.tight_layout()
plt.show()
```


#### 2.1. Построение последовательности в шкалах дискретного и дискретного нормированного времени

Для формирования последовательности используется сумма двух гармонических составляющих:

$$
x(nT)=A_1\cos(2\pi f_1nT+\pi/4)+A_2\cos(2\pi f_2nT+\pi/8).
$$

Подставляя значения из варианта, получим:

$$
x(nT)=1.08\cos(2\pi\cdot1000\cdot nT+\pi/4)+2.16\cos(2\pi\cdot1500\cdot nT+\pi/8),
$$

где $T=0.000125$ с.

Последовательность содержит $N=64$ отсчета при частоте дискретизации $f_{д}=8000$ Гц.

Длительность рассматриваемого периода:

$$
NT = 64\cdot0.000125 = 0.008 \text{ с} = 8 \text{ мс}.
$$

Последовательность построена для $n=0,1,\ldots,63$, поэтому последний отсчет соответствует моменту $t=7.875$ мс.

Дискретное нормированное время определяется выражением:

$$
\hat{t}=\frac{t}{T}.
$$

Поскольку $t=nT$, получаем:

$$
\hat{t}=n.
$$

Таким образом, графики представляют одну и ту же последовательность, но в разных временных шкалах.

На первом графике время выражено в миллисекундах, а на втором - в нормированных единицах, соответствующих номерам отсчетов.

Форма и амплитуда последовательности при изменении шкалы времени остаются неизменными.

Первая гармоника имеет период 8 отсчетов, вторая - 16 отсчетов. Поэтому минимальный период их суммы составляет 16 отсчетов.

Следовательно, на интервале $N=64$ располагаются четыре полных повторения последовательности.


```python
# Вычисление ДПФ последовательности x(n)
X2 = np.fft.fft(x2)

# Вычисление ОДПФ
x2_restored = np.fft.ifft(X2)

# Проверка точности восстановления
error = np.max(np.abs(x2 - x2_restored))

print(f"Длина восстановленной последовательности: {len(x2_restored)}")
print(f"Максимальная ошибка восстановления: {error:.2e}")

# Построение графика
fig, ax = plt.subplots(figsize=(11, 5))

markerline, stemlines, baseline = ax.stem(n, x2_restored.real)

plt.setp(markerline, color="#ebbcba", markersize=5)
plt.setp(stemlines, color="#c4a7e7", linewidth=1.8)
plt.setp(baseline, color="#6e6a86", linewidth=1)

ax.set_title("Последовательность после ОДПФ")
ax.set_xlabel("Дискретное нормированное время")
ax.set_ylabel("Амплитуда")
ax.set_xlim(-0.5, N - 0.5)

plt.tight_layout()
plt.show()
```

#### 2.2. Вычисление ОДПФ от ДПФ последовательности

Для последовательности $x(n)$ длины $N=64$ вычисляется дискретное преобразование Фурье:

$$
X(k)=\sum_{n=0}^{N-1}x(n)e^{-j\frac{2\pi}{N}kn},
$$

где $k=0,1,\ldots,N-1$.

Для восстановления исходной последовательности используется обратное дискретное преобразование Фурье:

$$
x(n)=\frac{1}{N}\sum_{k=0}^{N-1}X(k)e^{j\frac{2\pi}{N}kn}.
$$

Вычисление ДПФ выполнено с помощью функции `np.fft.fft`, а обратного преобразования - `np.fft.ifft`.

**Анализ результата.**

На графике представлена восстановленная последовательность в шкале дискретного нормированного времени $\hat{t}=n$.

В результате применения ОДПФ к ДПФ исходной последовательности получены те же $N=64$ отсчета.

Максимальная ошибка восстановления определяется выражением:

$$
\varepsilon_{\max}=\max_{0\leq n<N}|x(n)-x_{\text{восст}}(n)|.
$$

Ошибка имеет порядок машинной точности, поскольку при вычислениях с плавающей точкой возникают небольшие погрешности округления.

Таким образом, восстановленная последовательность практически полностью совпадает с исходной, что подтверждает обратимость дискретного преобразования Фурье.

```python

# Номера отсчетов ДПФ с учетом отрицательных частот
k = np.arange(-N // 2, N // 2)

# Дискретные нормированные и абсолютные частоты
f_norm = k / N
f_abs = f_norm * f_d

# Перенос нулевой частоты в центр спектра
X2_shifted = np.fft.fftshift(X2)

# Амплитудный и фазовый спектры
amplitude = np.abs(X2_shifted)
phase = np.angle(X2_shifted)

# Отбираем ненулевые спектральные составляющие
mask = amplitude > 1e-8

print("Спектральные составляющие:")
for i in np.where(mask)[0]:
    print(
        f"k = {k[i]:3d}, "
        f"f_norm = {f_norm[i]:+.4f}, "
        f"f = {f_abs[i]:+.0f} Гц, "
        f"|X(k)| = {amplitude[i]:.4f}, "
        f"phi = {phase[i] / np.pi:+.4f}pi"
    )

# Построение графиков
fig, axes = plt.subplots(2, 2, figsize=(14, 9))

frequencies = [
    (f_norm, "Дискретная нормированная частота"),
    (f_abs, "Частота f, Гц")
]

for j, (freq, xlabel) in enumerate(frequencies):
    # Амплитудный спектр
    markerline, stemlines, baseline = axes[0, j].stem(freq, amplitude)

    plt.setp(markerline, color="#ebbcba", markersize=5)
    plt.setp(stemlines, color="#c4a7e7", linewidth=1.8)
    plt.setp(baseline, color="#6e6a86", linewidth=1)

    axes[0, j].set_title("Амплитудный спектр")
    axes[0, j].set_xlabel(xlabel)
    axes[0, j].set_ylabel("|X(k)|")
    axes[0, j].set_xlim(freq[0], -freq[0])

    # Фазовый спектр (только ненулевые составляющие)
    markerline, stemlines, baseline = axes[1, j].stem(freq[mask], phase[mask])

    plt.setp(markerline, color="#ebbcba", markersize=5)
    plt.setp(stemlines, color="#c4a7e7", linewidth=1.8)
    plt.setp(baseline, color="#6e6a86", linewidth=1)

    axes[1, j].set_title("Фазовый спектр")
    axes[1, j].set_xlabel(xlabel)
    axes[1, j].set_ylabel("Фаза, рад")
    axes[1, j].set_xlim(freq[0], -freq[0])
    axes[1, j].set_ylim(-np.pi / 2, np.pi / 2)
    axes[1, j].set_yticks([-np.pi/4, -np.pi/8, 0, np.pi/8, np.pi/4])
    axes[1, j].set_yticklabels(
        [r"$-\pi/4$", r"$-\pi/8$", "0", r"$\pi/8$", r"$\pi/4$"]
    )

plt.tight_layout()
plt.show()
```


#### 2.3. Вычисление амплитудного и фазового спектров последовательности

Амплитудный и фазовый спектры последовательности определяются по результатам дискретного преобразования Фурье:

$$
|X(k)|,\qquad \arg X(k).
$$

Для отображения отрицательных и положительных частот используется функция `np.fft.fftshift`, которая переносит нулевую частоту в центр спектра.

**Связь дискретного нормированного времени с дискретным временем.**

Как показано в пункте 1 второй части, $t=nT$ и $\hat t=n$. Далее используются частотные шкалы $\hat f=\frac{f}{f_д}$ и $f$.

**Связь абсолютной и дискретной нормированной частот.**

Нормированная частота определяется отношением абсолютной частоты к частоте дискретизации:

$$
\hat{f}=\frac{f}{f_{д}}=fT.
$$

Для отсчетов ДПФ:

$$
\hat{f}_k=\frac{k}{N},\qquad f_k=\frac{kf_{д}}{N}.
$$

$N=64$, $f_{д}=8000$ Гц. Следовательно, шаг между соседними спектральными отсчетами:

$$
\Delta f=\frac{f_{д}}{N}=\frac{8000}{64}=125\text{ Гц}.
$$

Частотам исходных гармоник соответствуют:

$$
f_1=1000\text{ Гц},\qquad \frac{f_1}{f_{д}}=\frac8{64}=0.125,
$$

$$
f_2=1500\text{ Гц},\qquad \frac{f_2}{f_{д}}=\frac{12}{64}=0.1875.
$$

**Анализ амплитудного спектра.**

На графике амплитудного спектра наблюдаются четыре выраженные спектральные составляющие на частотах $f=\pm1000$ Гц и $f=\pm1500$ Гц.

Это объясняется тем, что исходная последовательность представляет собой сумму двух косинусоидальных гармоник. Каждая гармоника вещественного сигнала образует две симметричные спектральные составляющие на положительной и отрицательной частотах.

Для косинусоиды, частота которой совпадает с отсчетом ДПФ, модуль каждой из двух составляющих равен:

$$
|X(k)|=\frac{NA}{2}.
$$

Поэтому для первой гармоники:

$$
|X(8)|=|X(-8)|=\frac{64\cdot1.08}{2}=34.56,
$$

а для второй:

$$
|X(12)|=|X(-12)|=\frac{64\cdot2.16}{2}=69.12.
$$

Таким образом, амплитудный спектр симметричен относительно нулевой частоты. Вторая гармоника имеет вдвое большую амплитуду, чем первая, что соответствует заданным параметрам последовательности.

**Анализ фазового спектра.**

Фазовый спектр содержит следующие значения:

$$
\varphi(8)=\frac{\pi}{4},\qquad \varphi(-8)=-\frac{\pi}{4},
$$

$$
\varphi(12)=\frac{\pi}{8},\qquad \varphi(-12)=-\frac{\pi}{8}.
$$

Полученные значения соответствуют начальным фазам гармоник исходной последовательности.

Для четырёх рассматриваемых ненулевых составляющих фазы на противоположных частотах равны по модулю и противоположны по знаку.

Для остальных частот модули коэффициентов ДПФ практически равны нулю, поэтому их фазы не рассматриваются.

Таким образом, полученные спектры соответствуют исходной последовательности, состоящей из двух гармоник с частотами 1000 и 1500 Гц, амплитудами 1.08 и 2.16 и начальными фазами $\pi/4$ и $\pi/8$.


```python

# ДПФ конечной последовательности длины N
X_finite = np.fft.fft(x2)

# Модуль ДПФ конечной последовательности
amplitude_finite = np.abs(np.fft.fftshift(X_finite))

# Амплитудный спектр периодической последовательности
amplitude_periodic = np.abs(np.fft.fftshift(X2))

# Проверка совпадения спектров
error = np.max(np.abs(amplitude_finite - amplitude_periodic))

print(f"Длина конечной последовательности: {len(x2)}")
print(f"Максимальное отличие спектров: {error:.2e}")

# Построение графиков
fig, axes = plt.subplots(2, 1, figsize=(11, 9))

spectra = [
    (amplitude_finite, "Модуль ДПФ конечной последовательности"),
    (amplitude_periodic, "Амплитудный спектр периодической последовательности")
]

for ax, (spectrum, title) in zip(axes, spectra):
    markerline, stemlines, baseline = ax.stem(f_norm, spectrum)

    plt.setp(markerline, color="#ebbcba", markersize=5)
    plt.setp(stemlines, color="#c4a7e7", linewidth=1.8)
    plt.setp(baseline, color="#6e6a86", linewidth=1)

    ax.set_title(title)
    ax.set_xlabel("Дискретная нормированная частота")
    ax.set_ylabel("|X(k)|")
    ax.set_xlim(-0.5, 0.5)

plt.tight_layout()
plt.show()
```


#### 2.4. Вычисление ДПФ конечной последовательности и сравнение спектров

Дискретное преобразование Фурье конечной последовательности $x(n)$ длины $N$ определяется выражением:

$$
X(k)=\sum_{n=0}^{N-1}x(n)e^{-j\frac{2\pi}{N}kn},
$$

где $k=0,1,\ldots,N-1$.

Длина последовательности составляет $N=64$ отсчета.

Для сравнения рассматриваются два случая:

- Конечная последовательность $x(n)$, заданная на интервале $0\leq n<N$ и равная нулю за его пределами.
- Периодическая последовательность, полученная бесконечным повторением этих $N$ отсчетов.

**Сравнение спектров.**

ДПФ конечной последовательности и одного периода периодической последовательности вычисляются по одинаковой формуле.

В обоих случаях используются одни и те же $N=64$ отсчета исходного сигнала. Поэтому значения ДПФ совпадают:

$$
X_{\text{кон}}(k)=X_{\text{пер}}(k).
$$

Следовательно, совпадают и их модули:

$$
|X_{\text{кон}}(k)|=|X_{\text{пер}}(k)|.
$$

На графиках амплитудный спектр определяется как $|X(k)|$, аналогично предыдущему пункту работы.

**Анализ результатов.**

На обоих графиках наблюдаются четыре выраженные спектральные составляющие на нормированных частотах:

$$
\hat f=\pm0.125,\qquad \hat f=\pm0.1875.
$$

Модули ДПФ на этих частотах составляют:

$$
|X(\pm8)|=34.56,\qquad |X(\pm12)|=69.12.
$$

Остальные значения практически равны нулю.

Графики полностью совпадают, а максимальное отличие вычисленных спектров равно нулю с точностью машинных вычислений.

При этом коэффициенты ряда Фурье периодической последовательности равны $C(k)=X(k)/N$. Поэтому их модули отличаются от ненормированного модуля ДПФ в $N$ раз.

В обоих случаях вычислено ДПФ одного и того же блока, поэтому показанные ненормированные модули совпадают. Связь с амплитудами гармоник аналогового сигнала рассматривается в пункте восстановления.

Таким образом, ДПФ конечной последовательности и одного периода периодической последовательности имеют одинаковые значения. Различие заключается в интерпретации сигналов: конечная последовательность существует на ограниченном интервале, а периодическая повторяется неограниченно. Амплитуды гармоник периодической последовательности определяются из значений ДПФ с учетом нормировки.


```python
# Временная сетка с шагом 0.25T
dt = 0.25 * T
t_analog = np.arange(4 * (N - 1) + 1) * dt
t_analog_ms = t_analog * 1000

# Номера гармоник от -N/2 до N/2 - 1
k_analog = np.arange(-N // 2, N // 2)

# Перестановка отсчетов ДПФ для учета отрицательных частот
X_a = np.fft.fftshift(X2)

# Восстановление аналогового сигнала по формуле из задания
x_recovered = np.zeros(len(t_analog), dtype=complex)

for i in range(N):
    x_recovered += X_a[i] * np.exp(
        1j * 2 * np.pi * k_analog[i] * t_analog / (N * T)
    )

x_recovered = (x_recovered / N).real

# Исходный аналоговый сигнал
x_original = (
    A1 * np.cos(2 * np.pi * f1 * t_analog + np.pi/4)
    + A2 * np.cos(2 * np.pi * f2 * t_analog + np.pi/8)
)

# Амплитудный спектр восстановленного сигнала
f_analog = k_analog * f_d / N

# Только положительные частоты с ненулевой амплитудой
mask = (f_analog > 0) & (np.abs(X_a) > 1e-8)
A_analog = 2 * np.abs(X_a[mask]) / N

# Проверка восстановления
error = np.max(np.abs(x_recovered - x_original))

print(f"Шаг восстановления: {dt * 1000:g} мс")
print(f"Количество точек: {len(t_analog)}")
print(f"Максимальная ошибка восстановления: {error:.2e}")

# Построение первых четырех графиков
fig, axes = plt.subplots(2, 2, figsize=(14, 9))

stem_plots = [
    (
        axes[0, 0], t_ms, x2,
        "Периодическая последовательность x(n)",
        "Время nT, мс", "x(n)"
    ),
    (
        axes[0, 1], f_analog, np.abs(X_a),
        "Модуль ДПФ последовательности",
        "Частота f, Гц", "|X(k)|"
    ),
    (
        axes[1, 1], f_analog[mask], A_analog,
        "Амплитудный спектр аналогового сигнала",
        "Частота f, Гц", "Амплитуда"
    )
]

for ax, axis_x, axis_y, title, xlabel, ylabel in stem_plots:
    markerline, stemlines, baseline = ax.stem(axis_x, axis_y)

    plt.setp(markerline, color="#ebbcba", markersize=5)
    plt.setp(stemlines, color="#c4a7e7", linewidth=1.8)
    plt.setp(baseline, color="#6e6a86", linewidth=1)

    ax.set_title(title)
    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)

axes[1, 0].plot(t_analog_ms, x_recovered, color="#c4a7e7")
axes[1, 0].set_title("Восстановленный аналоговый сигнал x(t)")
axes[1, 0].set_xlabel("Время t, мс")
axes[1, 0].set_ylabel("x(t)")

plt.tight_layout()
plt.show()

# График исходного аналогового сигнала
fig, ax = plt.subplots(figsize=(11, 5))

ax.plot(t_analog_ms, x_original, color="#ebbcba")

ax.set_title("Исходный аналоговый сигнал x(t)")
ax.set_xlabel("Время t, мс")
ax.set_ylabel("x(t)")

plt.tight_layout()
plt.show()
```


#### 2.5. Восстановление периодического аналогового сигнала по отсчетам ДПФ

Для восстановления периодического аналогового сигнала $x(t)$ используются отсчеты ДПФ $X(k)$ периодической последовательности $x(n)$.

Восстановление выполняется по формуле:

$$
x(t)=\frac{1}{N}\sum_{k=-N/2}^{N/2-1}X_a(k)e^{j\frac{2\pi}{NT}kt},
$$

где коэффициенты $X_a(k)$ определяются выражением:

$$
X_a(k)=
\begin{cases}
X(N+k), & -N/2\leq k\leq -1,\\
X(k), & 0\leq k\leq N/2-1.
\end{cases}
$$

Для формирования коэффициентов $X_a(k)$ используется функция `np.fft.fftshift`, которая переставляет отсчеты ДПФ так, чтобы отрицательные частоты располагались перед неотрицательными.

**Временной интервал восстановления.**

$$
N=64,\qquad T=0.000125\text{ с}.
$$

Сигнал восстанавливается на интервале:

$$
t\in[0;(N-1)T]=[0;7.875]\text{ мс}.
$$

Шаг восстановления составляет:

$$
\Delta t=0.25T=0.03125\text{ мс}.
$$

Формула задаёт непрерывную функцию времени. Для её изображения вычислены 253 значения с шагом 0.25T; на графике соседние вычисленные точки соединены отрезками.

**Амплитудный спектр восстановленного сигнала.**

Частоты гармоник определяются выражением:

$$
f_k=\frac{kf_{д}}{N}.
$$

Коэффициенты ряда Фурье восстановленного периодического сигнала связаны с отсчетами ДПФ соотношением:

$$
C(k)=\frac{X_a(k)}{N}.
$$

На графике аналогового сигнала показаны односторонние амплитуды косинусоидальных гармоник. Для рассматриваемых частот 1000 и 1500 Гц каждая гармоника образуется сопряжённой парой отсчётов ДПФ. Поэтому

$$A_1=\frac{2|X_a(8)|}{N}=1.08,\qquad A_2=\frac{2|X_a(12)|}{N}=2.16.$$

Эти амплитуды отличаются от ненормированного модуля ДПФ. Для постоянной составляющей и составляющей на частоте дискретизации, делённой на два, удвоение не применяется; в данном сигнале они отсутствуют.

Таким образом, спектр восстановленного аналогового сигнала содержит две гармоники с частотами 1000 и 1500 Гц и амплитудами 1.08 и 2.16 соответственно.

**Исходный аналоговый сигнал.**

Исходный аналоговый сигнал определяется выражением:

$$
x(t)=A_1\cos(2\pi f_1t+\pi/4)+A_2\cos(2\pi f_2t+\pi/8).
$$

$$
x(t)=1.08\cos(2\pi\cdot1000t+\pi/4)+2.16\cos(2\pi\cdot1500t+\pi/8).
$$

Для проверки восстановления вычисляется максимальная абсолютная ошибка между восстановленным и исходным аналоговыми сигналами:

$$
\varepsilon_{\max}=\max_t|x_{\text{восст}}(t)-x_{\text{исх}}(t)|.
$$

Полученная ошибка имеет порядок машинной точности. Следовательно, восстановление аналогового сигнала по отсчетам ДПФ выполнено с высокой точностью.


```python
# Сравнение модуля ДПФ с амплитудами гармоник
for f in [f1, f2]:
    k = round(f * N / f_d)
    i = np.where(k_analog == k)[0][0]

    X_module = np.abs(X_a[i])
    A = 2 * X_module / N

    print(f"Частота: {f:g} Гц")
    print(f"Модуль ДПФ: {X_module:.4f}")
    print(f"Амплитуда гармоники: {A:.4f}\n")
```

Графики восстановленного и исходного аналоговых сигналов практически полностью совпадают. Они имеют одинаковую форму, амплитуды, периоды колебаний и положения экстремумов.

Максимальная абсолютная ошибка восстановления составила:

$$
\varepsilon_{\max}=\max_t|x_{\text{восст}}(t)-x_{\text{исх}}(t)|\approx3.00\cdot10^{-14}.
$$

Такое значение соответствует погрешности машинных вычислений.

Точное восстановление стало возможным благодаря тому, что частоты исходных гармоник $f_1=1000$ Гц и $f_2=1500$ Гц меньше частоты Найквиста:

$$
f_N=\frac{f_{д}}{2}=4000\text{ Гц}.
$$

Кроме того, обе гармоники соответствуют целым номерам спектральных отсчетов ДПФ ($k_1=8$, $k_2=12$), поэтому их спектральные составляющие представлены без утечки в соседние отсчеты.

Следовательно, восстановленный по ДПФ аналоговый сигнал практически полностью совпадает с исходным не только в моменты дискретизации, но и между ними.


```python

# Энергия последовательности во временной области
E_time = np.sum(np.abs(x2)**2)

# Энергия последовательности в частотной области
E_freq = np.sum(np.abs(X2)**2) / N

# Теоретическое значение энергии двух гармоник
E_theory = N * (A1**2 + A2**2) / 2

# Проверка равенства Парсеваля
error = np.abs(E_time - E_freq)

print(f"Энергия во временной области: {E_time:.6f}")
print(f"Энергия в частотной области: {E_freq:.6f}")
print(f"Теоретическая энергия: {E_theory:.6f}")
print(f"Абсолютная разность: {error:.2e}")
print(f"Равенство Парсеваля выполняется: {np.isclose(E_time, E_freq)}")
```


#### 2.7. Проверка равенства Парсеваля

Равенство Парсеваля устанавливает связь между энергией дискретной последовательности во временной области и энергией ее спектральных составляющих.

Для последовательности $x(n)$ длины $N$ и ее ДПФ $X(k)$ равенство имеет вид:

$$
\sum_{n=0}^{N-1}|x(n)|^2 =
\frac{1}{N}\sum_{k=0}^{N-1}|X(k)|^2.
$$

Левая часть выражения представляет собой сумму квадратов модулей отсчетов последовательности, а правая - сумму квадратов модулей коэффициентов ДПФ, деленную на $N$.

**Проверка равенства.**

$N=64$ отсчета.

Энергия во временной области:

$$
E_{\text{врем}}=\sum_{n=0}^{63}|x(n)|^2=186.624.
$$

Энергия в частотной области:

$$
E_{\text{част}}=\frac{1}{64}\sum_{k=0}^{63}|X(k)|^2=186.624.
$$

Теоретическое значение энергии для суммы двух гармоник, частоты которых совпадают с отсчетами ДПФ, определяется выражением:

$$
E=\frac{N}{2}(A_1^2+A_2^2).
$$

Подставляя значения амплитуд, получаем:

$$
E=\frac{64}{2}(1.08^2+2.16^2)=186.624.
$$

Таким образом, вычисленные значения совпадают с теоретическим.


Абсолютная разность энергий составляет:

$$
\varepsilon=|E_{\text{врем}}-E_{\text{част}}|=0.
$$

Полученные значения энергий совпадают. Это подтверждает выполнение равенства Парсеваля для исследуемой последовательности.

При вычислениях с плавающей точкой возможны небольшие численные погрешности, однако в данном случае разность вычисленных энергий равна нулю.

**Физический смысл равенства Парсеваля.**

Равенство Парсеваля показывает, что при переходе от временного представления дискретной последовательности к частотному ее энергия сохраняется с учетом нормировки ДПФ.

Энергия одного периода последовательности может быть вычислена как по значениям сигнала, так и по его спектральным составляющим.

Следовательно, равенство Парсеваля для заданной последовательности выполняется с точностью до погрешности машинных вычислений.
