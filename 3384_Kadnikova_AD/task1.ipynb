## Основные теоретические положения

### Передаточная функция системы
Передаточная функция системы — $z$-изображение импульсной характеристики $h(n)$ линейной дискретной системы, которое является её основной характеристикой в $z$-области. Оно вычисляется с помощью $Z$-преобразования:

$$
\begin{aligned}
H(z) &= \mathcal{Z}\{h(n)\} = \sum_{n=0}^{\infty} h(n) z^{-n} = \frac{\sum_{i=0}^{M} b_i z^{-i}}{\sum_{k=0}^{N} a_k z^{-k}} \\
&\text{При } a_0=1:\quad H(z)=\frac{\sum_{i=0}^{M} b_i z^{-i}}{1+\sum_{k=1}^{N} a_k z^{-k}}.
\end{aligned}
$$

### Свертка 
Реакция $y(nT)$ вычисляется как дискретная свертка воздействия $x(nT)$ и импульсной характеристики $h(nT)$ по данному линейному уравнению:  
$$
y(n)=x(n)*h(n)=\sum_{m=0}^{\infty}x(m)\,h(n-m)
$$

### Линейное разностное уравнение и импульсная характеристика 

Линейное разностное уравнение связывает текущий отсчёт выхода с текущими и предыдущими отсчётами входа и выхода. Для системы с коэффициентами $b_i$ и $a_k$ оно записывается так:

$$
a_0y(n)=\sum_{i=0}^{M}b_i x(n-i)-\sum_{k=1}^{N}a_k y(n-k),\qquad a_0\ne0.
$$

При $a_0=1$ уравнение сразу задаёт $y(n)$. Рекурсивная часть определяется предыдущими отсчётами выхода; её порядок задаётся наибольшей задержкой выхода с ненулевым коэффициентом.

Импульсная характеристика $h(n)$ — выход системы при входе $x(n)=u_0(n)$ и нулевых начальных условиях. Поэтому она удовлетворяет тому же разностному уравнению:

$$
a_0h(n)=\sum_{i=0}^{M}b_i u_0(n-i)-\sum_{k=1}^{N}a_k h(n-k),
$$

где $u_0(n)=1$ при $n=0$ и $u_0(n)=0$ при $n\ne0$. Для причинной системы значения входа и выхода при $n<0$ равны нулю.

### Нули, полюса и формы записи комплексного числа 

Если $H(z)=B(z)/A(z)$, нулями передаточной функции являются корни числителя $B(z)$, а полюсами — корни знаменателя $A(z)$ после сокращения общих множителей.

Комплексное число можно задать в алгебраической и показательной формах:

$$
z=x+jy=r e^{j\varphi},\qquad r=|z|=\sqrt{x^2+y^2},\qquad \varphi=\arg z.
$$

Эквивалентная тригонометрическая запись имеет вид $z=r(\cos\varphi+j\sin\varphi)$. Аргумент $\varphi$ измеряется в радианах; его также можно представить как долю от $\pi$.

Для причинной рациональной системы критерий устойчивости выражается через импульсную характеристику и полюса:

$$
\sum_{n=0}^{\infty}|h(n)|<\infty
\quad\Longleftrightarrow\quad
|p_i|<1\ \text{для всех полюсов }p_i.
$$

То есть все полюса устойчивой системы должны находиться строго внутри единичной окружности на комплексной плоскости.

### Частотные характеристики системы 

Частотная характеристика получается подстановкой $z=e^{j\omega}$ в передаточную функцию; эквивалентно она является преобразованием Фурье импульсной характеристики:

$$
H(e^{j\omega})=\sum_{n=0}^{\infty}h(n)e^{-j\omega n}.
$$

Амплитудно-частотная характеристика и фазочастотная характеристика определяются как модуль и аргумент этой комплексной функции:

$$
A(\omega)=|H(e^{j\omega})|,\qquad
\Phi(\omega)=\arg H(e^{j\omega}).
$$

В основной полосе вещественного сигнала $0\le\omega\le\pi$. Если нормированная угловая частота задана как $\hat\omega=\omega/\pi$, то $0\le\hat\omega\le1$, а связь с частотой в герцах имеет вид

$$
f=\frac{\omega f_{\mathrm d}}{2\pi}=\frac{\hat\omega f_{\mathrm d}}{2}.
$$

### ДПФ и ОДПФ

Для одного периода последовательности длины $N$ дискретное преобразование Фурье и обратное преобразование задаются формулами

$$
X(k)=\sum_{n=0}^{N-1}x(n)e^{-j2\pi kn/N},\qquad k=0,1,\ldots,N-1,
$$

$$
x(n)=\frac{1}{N}\sum_{k=0}^{N-1}X(k)e^{j2\pi kn/N},\qquad n=0,1,\ldots,N-1.
$$

ОДПФ восстанавливает отсчёты исходной последовательности из её ДПФ. Множитель $1/N$ относится к обратному преобразованию.

### Амплитудный и фазовый спектры 

Комплексные коэффициенты ряда Фурье периодического продолжения равны $X(k)/N$. Поэтому амплитудные линии и фазы спектра задаются выражениями

$$
A(k)=\frac{|X(k)|}{N},\qquad
\Phi(k)=\arg X(k)\quad\text{при }|X(k)|\ne0.
$$

Модуль конечного ДПФ равен $|X(k)|$, то есть он в $N$ раз больше амплитуды соответствующей линии периодического сигнала. Для вещественной последовательности спектральные отсчёты сопряжённо-симметричны: $X(N-k)=X^*(k)$. Поэтому модули пар равны, а фазы противоположны с точностью до выбора ветви аргумента.

### Восстановление аналогового периодического сигнала 

Для чётного $N$ отсчёты ДПФ располагаются по индексам от $-N/2$ до $N/2-1$. Перенумерованный спектр определяется как

$$
X_{\mathrm a}(k)=
\begin{cases}
X(N+k),&-N/2\le k\le-1,\\
X(k),&0\le k\le N/2-1.
\end{cases}
$$

Периодический аналоговый сигнал восстанавливается суммой гармонических составляющих:

$$
x(t)=\frac{1}{N}\sum_{k=-N/2}^{N/2-1}X_{\mathrm a}(k)
\exp\left(j\frac{2\pi k}{NT}t\right),
$$

где $T=1/f_{\mathrm d}$ — период дискретизации. Частота составляющей с индексом $k$ равна $f_k=k/(NT)=k f_{\mathrm d}/N$.

### Равенство Парсеваля 

Для принятого определения ДПФ энергия конечной последовательности во временной и частотной областях связана равенством

$$
\sum_{n=0}^{N-1}|x(n)|^2
=\frac{1}{N}\sum_{k=0}^{N-1}|X(k)|^2.
$$

Множитель $1/N$ соответствует нормировке, при которой прямое ДПФ не делится на $N$, а ОДПФ содержит множитель $1/N$.

### 1.1. Импульсная характеристика
Вычислить импульсную характеристику длины $N_1$ по заданной передаточной функции с коэффициентами $b_i$, $a_k$ и построить её график. Записать аналитическую формулу импульсной характеристики рекурсивного звена 2-го порядка с учетом нулевых начальных условий. Пояснить, чему в действительности равна длина импульсной характеристики. 

#### Решение:
Формула заданной передаточной функции с соответствующими коэффициентами:  
$$
H(z)=\frac{\sum_{i=0}^{M}b_i z^{-i}}{1+\sum_{k=1}^{N}a_k z^{-k}} = \frac{0.62-0.67518z^{-1}+0.62z^{-2}}{1+0.9278z^{-1}+0.676z^{-2}}.
$$

Разностному уравнению с теми же коэффициентами соответствует запись:
$$
y(n)=\sum_{i=0}^{M}b_i x(n-i)-\sum_{k=1}^{N}a_k y(n-k).
$$
При импульсном входе $x(n)=u_0(n)$ выход равен импульсной характеристике $y(n)=h(n)$. Поэтому:
$$
h(n)=\sum_{i=0}^{M}b_i u_0(n-i)-\sum_{k=1}^{N}a_k h(n-k).
$$
Для заданных коэффициентов разностное уравнение принимает вид:
$$
h(n)=0.62u_0(n)-0.67518u_0(n-1)+0.62u_0(n-2)-(0.9278h(n-1)+0.676h(n-2)).
$$

При $n<0$ принимаются нулевые значения входа и выхода.  
Для данной системы при нулевых начальных условиях:
$$
\begin{aligned}
h(0) &= b_0 = 0.62, \\
h(1) &= b_1-a_1h(0)=-0.67518-0.9278\cdot0.62=-1.250416, \\
h(2) &= b_2-a_1h(1)-a_2h(0)=0.62-0.9278\cdot(-1.250416)-0.676\cdot0.62=1.3610159648.
\end{aligned}
$$
При $n\ge 3$ можно выразить как $h(n)=-0.9278h(n-1)-0.676h(n-2)$, поскольку рекурсивное звено продолжает реагировать после импульса;   
в расчёте показаны первые $N_1=26$ отсчётов.


```python
# Коэффициенты числителя и знаменателя
b = [b0, b1, b2]
a = [a0, a1, a2]

# Вычисление h(0) - h(N1-1) по разностному уравнению с использованием предыдущих значений h(n)
h = []
for n in range(N1):
    input_sum = sum(
        b[i] * float(u0(n - i))
        for i in range(len(b))
    )
    feedback_sum = sum(
        a[k] * (h[n - k] if n >= k else 0.0)
        for k in range(1, len(a))
    )
    h_n = (input_sum - feedback_sum) / a0
    h.append(h_n)

    input_substitution = " + ".join(
        f"({b[i]:g}*{float(u0(n - i)):g})"
        for i in range(len(b))
    )
    feedback_substitution = " + ".join(
        f"({a[k]:g}*{(h[n - k] if n >= k else 0.0):g})"
        for k in range(1, len(a))
    )
    print(
        f"h({n}) = ({input_substitution} - ({feedback_substitution})) = {h_n:.6f}"
    )
```

#### График импульсной характеристики

Массив $h(n)$ вычислен по разностному уравнению при входе $x(n)=u_0(n)$. На графике показаны отсчёты с номерами $n=0,1,\ldots,N_1-1$, где $N_1=26$. Теоретически импульсная характеристика рекурсивного звена бесконечна: после окончания входного импульса при $n\ge3$ она продолжает формироваться за счёт обратной связи. Хвост характеристики затухает, но в общем случае не становится равным нулю (так как ИХ рекурсивна и бесконечна из-за наличия ненулевых слагаемых $a_i$). Поэтому график показывает конечный фрагмент из 26 отсчётов, а не всю импульсную характеристику.


```python
# Первые N1 отсчётов импульсной характеристики h(n)
n_h = np.arange(N1)

fig, ax = plt.subplots(figsize=(9, 4))
ax.stem(n_h, h, basefmt=" ")
ax.axhline(0, color="black", linewidth=0.8)
ax.set_title(f"Импульсная характеристика h(n), первые {N1} отсчётов")
ax.set_xlabel("Номер отсчёта n")
ax.set_ylabel("h(n)")
ax.set_xlim(-0.5, N1 - 0.5)
ax.grid(True)
plt.tight_layout()
plt.show()
```

### 1.2. Реакция по линейной свёртке и разностному уравнению

Вычислить реакцию $y_1(n)$ по формуле свёртки. В качестве воздействия $x(n)$ длины $N_2$ выбрать дискретный прямоугольный импульс:
$$
x(n)=\begin{cases}1,&0\le n<\operatorname{int}(N_2/2),\\0,&\operatorname{int}(N_2/2)\le n\le N_2-1.\end{cases}
$$
где функция $\operatorname{int}(t)$ — округление до ближайшего целого. Вывести графики воздействия и реакции. Пояснить:  
а) чему равна длина импульса $x(n)$;  
б) чему равна длина свёртки аналитически и по графику;  
в) почему её ограничивают до длины воздействия.

#### Решение:

а) Входной массив имеет длину $N_2=36$.

$$
x(n)=\begin{cases}1,&0\le n<18,\\0,&18\le n\le35.\end{cases}
$$

Первые $\operatorname{int}(36/2)=18$ отсчётов равны единице, остальные - нулю, поэтому длина импульса $18$

б) Для линейной системы реакция на вход $x(n)$ находится свёрткой входа с импульсной характеристикой:

$$
y_1(n)=x(n)\ast h(n)=\sum_{m=0}^{\infty}x(m)\cdot h(n-m).
$$

Здесь $\ast$ обозначает свёртку, а $\cdot$ — обычное умножение. В коде свёртка вычисляется с помощью конечного массива $h(n)$, сохранённого в пункте 1.1.

Аналитическая длина свёртки двух конечных массивов равна сумме их длин минус один. Для массивов длины $N_2=36$ и $N_1=26$ она составляет 
$$
len(y)=N_2+N_1-1=36+26-1=61
$$



в) На графике полной свёртки также показаны $61$ отсчёт — от $n=0$ до $n=60$. В этой конечной свёртке значения при $n=43,\ldots,60$ равны нулю: входной массив после $n=17$ нулевой, а в расчёте использован только фрагмент $h(0),\ldots,h(25)$

В пункте 1.1 рассчитаны первые $N_1=26$ отсчётов импульсной характеристики $h(n)$. Это длина вычисленного фрагмента $h(n)$, который используется в данной свёртке; фактическая импульсная характеристика рекурсивной системы бесконечна.
  
Ограничение длины свёртки до $N_2$ сделано для сравнения вычисления свертки через ИХ и через РУ, в котором используется x, ограниченный как раз $N_2$ отсчётами

#### Расчёт $y_1(n)$ свертки

```python
# Прямоугольный импульс: x[n]=1 для 0<=n<int(N2/2), иначе 0
n_x = np.arange(N2)
x1 = (n_x < int(N2 / 2)).astype(float)

# Полная линейная свёртка 
conv_full = np.convolve(x1, h, mode="full")
y1 = conv_full[:N2]


print(f"Длина x={len(x1)}; длина полной свёртки={len(conv_full)}; длина отображаемого y1={len(y1)}")

# все отсчёты полной свёртки с подстановкой значений x и h
print("Подстановка в формулу свёртки:")
for n_conv, value in enumerate(conv_full):
    # Ненулевые отсчёты x находятся при m=0,...,17
    m_start = max(0, n_conv - (len(h) - 1))
    m_end = min(int(N2 / 2) - 1, n_conv)
    products = [
        f"({x1[m]:.0f}*{h[n_conv - m]:.6f})"
        for m in range(m_start, m_end + 1)
    ]
    substitution = " + ".join(products) if products else "0"
    print(f"y1({n_conv}) = {substitution} = {value:.6f}")
```

где при $n<0$ вход и выход полагаются равными нулю. Теоретически реакция рекурсивной системы бесконечна: обратная связь продолжает формировать выход после того, как вход стал нулевым.

#### Сравнение реакций $y_1(n)$ и $y_2(n)$
В пункте 1.2 свёртка конечных массивов $x(n)$ длины $36$ и усечённой $h(n)$ длины $26$ содержит

$$
N_y=36+26-1=61
$$

отсчёт, далее нули. Свёртка сопоставления ограничивается первыми $N_2=36$ отсчётами. После $n=25$ свёртка $y_2$ не учитывает ненулевой хвост $h(n)$, поэтому начинает отличаться от результата разностного уравнения. Нулевые значения конечной свёртки при $n=43,\ldots,60$ вызваны ограниченной длиной ненулевого воздействия $x$, в котором только первые 18 элементов не нулевые.


```python
# Реакция на том же входе, рассчитанная по разностному уравнению
n_y = np.arange(N2)
y2 = np.zeros(N2)
for n in n_y:
    x_n = x1[n]
    x_n1 = x1[n - 1] if n >= 1 else 0.0
    x_n2 = x1[n - 2] if n >= 2 else 0.0
    y_n1 = y2[n - 1] if n >= 1 else 0.0
    y_n2 = y2[n - 2] if n >= 2 else 0.0
    y2[n] = (b0*x_n + b1*x_n1 + b2*x_n2 - a1*y_n1 - a2*y_n2) / a0

fig, ax = plt.subplots(2, 1, figsize=(9, 6), sharex=True)
ax[0].stem(n_x, x1, basefmt=" ")
ax[0].set_ylabel("x[n]")
ax[0].set_title("Воздействие")
ax[1].stem(n_y, y2, basefmt=" ", label="Разностное уравнение")
ax[1].plot(n_y, y1, "o", ms=3, label="Свёртка усечённой h[n]")
ax[1].set_xlabel("Номер отсчёта n")
ax[1].set_ylabel("y[n]")
ax[1].set_title("Реакция на интервале 0 ≤ n ≤ 35")
for axis in ax:
    axis.grid(True)
ax[1].legend()
plt.tight_layout()
plt.show()
print(f"Длина входа: {len(x1)}; длина показанной реакции: {len(y2)}")
print(f"Максимальное расхождение на первых N1 отсчётах: {np.max(np.abs(y1[:N1]-y2[:N1])):.3e}")
print(f"Максимальное расхождение на первых N2 отсчётах: {np.max(np.abs(y1-y2)):.3f}")
print(f"Максимальное расхождение после первых N1 отсчётов: {np.max(np.abs(y1[N1:]-y2[N1:])):.3f}")
```

### 1.4. Нули и полюсы

Вычислить нули и полюсы передаточной функции. Представить нули и полюсы в алгебраической и показательной формах и пояснить связь между ними. Выразить значение аргумента полюса и нуля относительно $\pi$. Представить передаточную функцию в виде произведения простейших множителей с нулями и полюсами в показательной форме.

#### Решение:

Формула заданной передаточной функции с соответствующими коэффициентами:  
$$
H(z)=\frac{\sum_{i=0}^{M}b_i z^{-i}}{1+\sum_{k=1}^{N}a_k z^{-k}} = \frac{0.62-0.67518z^{-1}+0.62z^{-2}}{1+0.9278z^{-1}+0.676z^{-2}}.
$$

Нули - корни числителя, полюсы - корни знаменателя. После умножения числителя и знаменателя $H(z)$ на $z^2$ получаются соответствующие квадратные уравнения:

$$
0.62z^2-0.67518z+0.62=0,\qquad z^2+0.9278z+0.676=0.
$$

Подстановка коэффициентов в формулу корней квадратного уравнения даёт

$$
\begin{aligned}
z_{1,2}&=\frac{0.67518\pm\sqrt{(-0.67518)^2-4\cdot0.62\cdot0.62}}{2\cdot0.62}=0.5445\pm j0.83876084,\\
p_{1,2}&=\frac{-0.9278\pm\sqrt{0.9278^2-4\cdot0.676}}{2}=-0.4639\pm j0.67882015.
\end{aligned}
$$

Для комплексного числа $q=a+jb$ модуль и аргумент задают его показательную форму: $q=|q|e^{j\arg(q)}$.

$$
\begin{aligned}
z_{1,2}&=0.5445\pm j0.83876084
          =1\cdot e^{\pm j0.995003},\qquad
          \arg z_{1,2}=\pm0.316719\pi,\\
p_{1,2}&=-0.4639\pm j0.67882015
          =0.822192e^{\pm j2.170289},\qquad
          \arg p_{1,2}=\pm0.690824\pi.
\end{aligned}
$$

При разложении на простейшие множители используются найденные корни и их комплексные сопряжения:

$$
H(z)=0.62\,\frac{(1-z_1z^{-1})(1-z_2z^{-1})}{(1-p_1z^{-1})(1-p_2z^{-1})},
$$

где $z_{1,2}=e^{\pm j0.995003}$, а $p_{1,2}=0.822192e^{\pm j2.170289}$. Нули имеют единичный модуль; полюсы лежат внутри единичной окружности, поскольку $|p_{1,2}|=0.822192<1$.


```python
# Корни числителя и знаменателя передаточной функции
zeros = np.roots([b0, b1, b2])
poles = np.roots([a0, a1, a2])

for name, points in (("Нули", zeros), ("Полюсы", poles)):
    print(name)
    for q in points:
        print(f"  {q.real:+.6f}{q.imag:+.6f}j = {abs(q):.6f}·exp(j·{np.angle(q):+.6f}); "
              f"arg/π={np.angle(q)/np.pi:+.6f}")


print("Система устойчива:", bool(np.all(np.abs(poles) < 1)))
```

### 1.6. АЧХ и ФЧХ в нормированных частотах

Вычислить амплитудно-частотную (АЧХ) и фазочастотную (ФЧХ) характеристики в шкале нормированных частот $\hat\omega$ и вывести их графики. Пояснить: а) чему равны границы основной полосы частот; б) соответствие между картой нулей и полюсов и видом АЧХ; в) какому значению АЧХ соответствует скачок на $\pi$, если он имеется; г) какие частотные составляющие воздействия, низкие или высокие, преимущественно подавлены в реакции.

#### Решение:

Для вещественного сигнала основная полоса дискретных частот равна $0\le\omega\le\pi$ рад/отсчёт. В нормированной шкале $\hat\omega=\omega/\pi$ это $0\le\hat\omega\le1$.

Нуль $z=e^{\pm j0.316719\pi}$ лежит на единичной окружности, поэтому АЧХ обращается в ноль при $\hat\omega\approx0.316719$. Полюса имеют аргументы $\pm0.690824\pi$ и модуль $0.822192$; близость этой пары к единичной окружности создаёт подъём АЧХ около $\hat\omega\approx0.704$. Фаза в точке нулевой АЧХ не определена; скачок фазы на $\pi$ на графике приходится на прохождение через этот спектральный нуль, то есть на $|H(e^{j\omega})|=0$.

Это звено не является обычным фильтром, который подавляет только низкие или только высокие частоты. Оно особенно сильно подавляет составляющую около $\hat\omega=0.316719$ (при $f_{\mathrm d}=6000$ Гц это около $950$ Гц), ослабляет и постоянную составляющую ($|H(e^{j0})|\approx0.217$), а около резонансного подъёма усиливает составляющие. Поэтому по всему диапазону нельзя сказать, что подавляются только низкие либо только высокие частоты.


```python
# Положительная основная полоса: omega in [0, pi]
omega = np.linspace(0, np.pi, 2048)
z_inv = np.exp(-1j * omega)
H = (b0 + b1*z_inv + b2*z_inv**2) / (a0 + a1*z_inv + a2*z_inv**2)
mag = np.abs(H)
phase = np.angle(H)
omega_hat = omega / np.pi
freq_hz = omega * fd1 / (2*np.pi)

# АЧХ и ФЧХ в нормированной шкале omega_hat=omega/pi
fig, ax = plt.subplots(2, 1, figsize=(9, 6), sharex=True)
ax[0].plot(omega_hat, mag); ax[0].set_ylabel("|H(e^{jω})|"); ax[0].set_title("Частотные характеристики в нормированной шкале")
ax[1].plot(omega_hat, phase); ax[1].set_ylabel("arg H(e^{jω}), рад"); ax[1].set_xlabel("Нормированная частота ω̂ = ω/π")
for a in ax: a.grid(True)
plt.tight_layout(); plt.show()

# Те же характеристики по оси частот в герцах
fig, ax = plt.subplots(2, 1, figsize=(9, 6), sharex=True)
ax[0].plot(freq_hz, mag); ax[0].set_ylabel("|H(e^{jω})|"); ax[0].set_title(f"Частотные характеристики, fд={fd1:.0f} Гц")
ax[1].plot(freq_hz, phase); ax[1].set_ylabel("arg H(e^{jω}), рад"); ax[1].set_xlabel("Частота f, Гц")
for a in ax: a.grid(True)
plt.tight_layout(); plt.show()

idx_zero = np.argmin(mag)
idx_peak = np.argmax(mag)
print(f"|H(0)|={mag[0]:.6f}; |H(π)|={mag[-1]:.6f}")
print(f"Минимум сетки: ω̂={omega_hat[idx_zero]:.5f}, f={freq_hz[idx_zero]:.2f} Гц, |H|={mag[idx_zero]:.3g}")
print(f"Максимум сетки: ω̂={omega_hat[idx_peak]:.5f}, f={freq_hz[idx_peak]:.2f} Гц, |H|={mag[idx_peak]:.4f}")
```

### 2.1 Формирование периодической последовательности

Построить графики последовательности $x(n)$ на одном периоде длины $N$ в шкалах дискретного и дискретного нормированного времени. Последовательность задаётся формулой

$$
x(nT)=A_1\cos\left(2\pi f_1nT+\frac{\pi}{4}\right)+A_2\cos\left(2\pi f_2nT+\frac{\pi}{8}\right),\quad n=0,\ldots,N-1,
$$

где $N=64$, $f_{\mathrm d}=4000\,\text{Гц}$, $T=1/f_{\mathrm d}$, $A_1=1.06$, $A_2=2.12$, $f_1=500\,\text{Гц}$ и $f_2=750\,\text{Гц}$.

#### Решение:

Период дискретизации и моменты отсчётов:

$$
T=\frac{1}{4000}=0.00025\,\text{с},\qquad t_n=nT,\quad n=0,1,\ldots,63.
$$

После подстановки параметров варианта в формулу последовательность задаётся выражением

$$
x(nT)=1.06\cos\left(2\pi\cdot500\,n\cdot0.00025+\frac{\pi}{4}\right)
+2.12\cos\left(2\pi\cdot750\,n\cdot0.00025+\frac{\pi}{8}\right).
$$

В коде вычисляется эта сумма для каждого $n$. Верхний график показывает отсчёты в дискретном времени $t_n=nT$ (секунды), а нижний — те же отсчёты в нормированном дискретном времени $n=t_n/T$ (номера отсчётов). Значение $N=64$ задаёт один выбранный период последовательности.


```python
n = np.arange(N)
tn = n * T
x = A1*np.cos(2*np.pi*f1*tn + np.pi/4) + A2*np.cos(2*np.pi*f2*tn + np.pi/8)

fig, ax = plt.subplots(2, 1, figsize=(9, 6), sharey=True)
ax[0].stem(tn, x, basefmt=" "); ax[0].set_xlabel("Дискретное время tₙ, с"); ax[0].set_ylabel("x[n]"); ax[0].set_title("Один период: дискретное время tₙ=nT")
ax[1].stem(n, x, basefmt=" "); ax[1].set_xlabel("Нормированное дискретнле время n"); ax[1].set_ylabel("x[n]"); ax[1].set_title("Один период: нормированное дискретное время")
for a in ax: a.grid(True)
plt.tight_layout(); plt.show()
```

В NumPy прямое ДПФ и ОДПФ определены так:

$$
X(k)=\sum_{n=0}^{N-1}x(n)e^{-j2\pi kn/N},\qquad
x_{\mathrm{восст}}(n)=\frac{1}{N}\sum_{k=0}^{N-1}X(k)e^{j2\pi kn/N}.
$$

При $N=64$ множитель в формуле ОДПФ равен $1/64$:

$$
x_{\mathrm{восст}}(n)=\frac{1}{64}\sum_{k=0}^{63}X(k)e^{j2\pi kn/64}.
$$

На графике восстановленные отсчёты показаны в нормированном дискретном времени $n=0,1,\ldots,63$, согласованном со шкалой в пункте 2.1. Максимальное отличие от исходной последовательности, показанное программой, равно $6.661\cdot10^{-16}$; это численная погрешность округления.


```python
X = np.fft.fft(x)
x_rec = np.fft.ifft(X).real
print(f"T={T:.7f} с; f1={f1:.1f} Гц; f2={f2:.1f} Гц; max |x-ОДПФ(ДПФ(x)))|={np.max(np.abs(x-x_rec)):.3e}")
fig, ax = plt.subplots(figsize=(9, 3.5))
ax.stem(n, x_rec, basefmt=" ", label="ОДПФ(ДПФ(x))")
ax.set_xlabel("Нормированное дискретное время n"); ax.set_ylabel("Восстановленное значение")
ax.set_title("Восстановление периодической последовательности через ОДПФ"); ax.grid(True); plt.tight_layout(); plt.show()
```

### 2.3 Амплитудный и фазовый спектры

Вычислить амплитудный и фазовый спектры последовательности $x(n)$ и построить их графики в шкалах дискретных нормированных и абсолютных частот. Пояснить: а) связь дискретного нормированного времени с дискретным временем; б) связь частоты $f$ в герцах с дискретной нормированной частотой; в) вид амплитудного и фазового спектров.

#### Решение:

Дискретное время и время в секундах связаны соотношением $t_n=nT$, где

$$
T=\frac{1}{f_{\mathrm d}}=\frac{1}{4000}=0.00025\,\text{с},\qquad n=\frac{t_n}{T}.
$$

Нормированная частота в циклах на отсчёт равна $\hat f=f/f_{\mathrm d}$, поэтому $f=\hat f f_{\mathrm d}$. Для заданных частот:

$$
\hat f_1=\frac{500}{4000}=0.125,\qquad \hat f_2=\frac{750}{4000}=0.1875.
$$

Их номера в 64-точечном ДПФ равны $k_1=N\hat f_1=64\cdot0.125=8$ и $k_2=N\hat f_2=64\cdot0.1875=12$. Поэтому линии расположены при $\hat f=\pm0.125$ и $\hat f=\pm0.1875$, или при $f=\pm500$ и $f=\pm750$ Гц.

Для вещественного косинусного сигнала возникают сопряжённые пары спектральных отсчётов. После нормировки на $N$ их значения таковы:

$$
\begin{aligned}
\frac{X(8)}{64}&=\frac{1.06}{2}e^{j\pi/4}=0.53e^{j\pi/4},&\quad \frac{X(56)}{64}&=0.53e^{-j\pi/4},\\
\frac{X(12)}{64}&=\frac{2.12}{2}e^{j\pi/8}=1.06e^{j\pi/8},&\quad \frac{X(52)}{64}&=1.06e^{-j\pi/8}.
\end{aligned}
$$

Следовательно, модули линий равны $0.53$ для первой гармоники и $1.06$ для второй; фазы на отрицательных частотах противоположны фазам на соответствующих положительных частотах. В остальных узлах идеальные значения равны нулю, а малые отображаемые значения связаны с округлением.


```python
# fftshift упорядочивает частоты от отрицательных к положительным
X_shift = np.fft.fftshift(X)
f_norm = np.fft.fftshift(np.fft.fftfreq(N, d=1.0))  # циклы на отсчёт
f_abs = f_norm * fd2
amp_periodic = np.abs(X_shift) / N
phase_spectrum = np.angle(X_shift)
phase_masked = np.ma.masked_where(np.abs(X_shift) < 1e-10, phase_spectrum)

fig, ax = plt.subplots(2, 2, figsize=(12, 7))
ax[0,0].stem(f_norm, amp_periodic, basefmt=" "); ax[0,0].set_title("Амплитудный спектр, нормированная частота"); ax[0,0].set_xlabel("f̂ = f/fд, циклов/отсчёт")
ax[0,1].stem(f_abs, amp_periodic, basefmt=" "); ax[0,1].set_title("Амплитудный спектр, герцы"); ax[0,1].set_xlabel("f, Гц")
ax[1,0].stem(f_norm, phase_masked, basefmt=" "); ax[1,0].set_title("Фазовый спектр, нормированная частота"); ax[1,0].set_xlabel("f̂ = f/fд, циклов/отсчёт")
ax[1,1].stem(f_abs, phase_masked, basefmt=" "); ax[1,1].set_title("Фазовый спектр, герцы"); ax[1,1].set_xlabel("f, Гц")
for a in ax.flat: a.grid(True); a.set_ylabel("Амплитуда / фаза, рад")
plt.tight_layout(); plt.show()
```

### 2.4 ДПФ конечной последовательности

Вычислить ДПФ конечной последовательности $x(n)$ длины $N$. Построить в шкале дискретных нормированных частот графики модуля ДПФ конечной последовательности и амплитудного спектра периодической последовательности. Пояснить связь этих спектров.

#### Решение:

Конечный блок содержит один выбранный период длины $N=64$, поэтому его ДПФ совпадает с ДПФ одного периода периодической последовательности. Для конечного блока используется ненормированный модуль $|X(k)|$. Коэффициент ряда Фурье периодического продолжения равен $X(k)/N$, поэтому амплитудный спектр в принятой нормировке задаётся формулой

$$
A_{\mathrm{пер}}(k)=\frac{|X(k)|}{N}=\frac{|X(k)|}{64}.
$$

Таким образом, значения модуля ДПФ конечного блока в $64$ раза больше соответствующих амплитудных линий периодической последовательности. Положения линий на частотной оси совпадают.


```python
# ДПФ конечного блока x[n] и спектральные линии периодического продолжения
X_finite = np.fft.fft(x)
X_finite_shift = np.fft.fftshift(X_finite)
fig, ax = plt.subplots(figsize=(10, 4))
ax.stem(f_norm, np.abs(X_finite_shift), linefmt="C0-", markerfmt="C0o", basefmt=" ", label="|X[k]| конечного блока")
ax.stem(f_norm, np.abs(X_finite_shift)/N, linefmt="C1-", markerfmt="C1s", basefmt=" ", label="|X[k]|/N: амплитуда линии периодического сигнала")
ax.set_xlabel("Нормированная частота f̂=f/fд, циклов/отсчёт"); ax.set_ylabel("Спектральная амплитуда")
ax.set_title("Связь модуля ДПФ и амплитудного спектра периодического сигнала")
ax.grid(True); ax.legend(); plt.tight_layout(); plt.show()
```

### 2.5 Восстановление аналогового сигнала

Восстановить периодический аналоговый сигнал $x(t)$ по отсчётам ДПФ $X(k)$ периодической последовательности $x(n)$. Использовать формулы

$$
x(t)=\frac{1}{N}\sum_{k=-N/2}^{N/2-1}X_{\mathrm a}(k)\exp\left(j\frac{2\pi}{NT}kt\right),
$$

$$
X_{\mathrm a}(k)=\begin{cases}X(N+k),&-N/2\le k\le-1,\\X(k),&0\le k\le N/2-1.\end{cases}
$$

Задать $t\in[0,(N-1)T]$ с шагом $\Delta t=0.25T$ и построить: а) последовательность $x(n)$ и модуль её ДПФ; б) восстановленный сигнал $x(t)$ и его амплитудный спектр; в) исходный сигнал $x(t)=x(nT)|_{nT=t}$.

#### Решение:

При $N=64$ индексы восстановления равны $k=-32,-31,\ldots,31$. Для $k=-32,\ldots,-1$ берутся значения $X(64+k)$, а для $k=0,\ldots,31$ — значения $X(k)$. Период дискретизации и шаг сетки с подстановкой параметров варианта равны

$$
T=\frac{1}{4000}=0.00025\,\text{с},\qquad \Delta t=0.25T=0.0000625\,\text{с}.
$$

Конечное время равно $t_{\max}=(N-1)T=63\cdot0.00025=0.01575\,\text{с}$; на этой сетке получается $253$ точки. Формула восстановления для заданного $N$ и $T$ принимает вид

$$
x_{\mathrm{восст}}(t)=\frac{1}{64}\sum_{k=-32}^{31}X_{\mathrm a}(k)\exp\left(j\frac{2\pi k}{64\cdot0.00025}t\right).
$$

Исходный аналоговый сигнал строится по заданным гармоникам с подстановкой параметров:

$$
x_{\mathrm{исх}}(t)=1.06\cos\left(2\pi\cdot500t+\frac{\pi}{4}\right)+2.12\cos\left(2\pi\cdot750t+\frac{\pi}{8}\right).
$$

Частоты $500$ и $750$ Гц соответствуют узлам ДПФ $k=8$ и $k=12$, поэтому восстановленный сигнал совпадает с исходной суммой гармоник с точностью вычислений. Амплитуды спектральных линий ряда Фурье равны $|X_{\mathrm a}(k)|/64$.


```python
# Перестановка спектральных отсчётов в порядок k=-N/2,...,N/2-1
k_signed = np.arange(-N//2, N//2)
X_a = np.array([X[N + k] if k < 0 else X[k] for k in k_signed])

# Шаг сетки равен 0.25T, конечная точка — (N-1)T
t = np.arange(0, (N - 1) * T + 0.25 * T, 0.25 * T)
x_analog_rec = np.real(np.sum(
    X_a[:, None] * np.exp(1j * 2*np.pi*k_signed[:, None]*t[None, :] / (N*T)), axis=0
) / N)
x_analog_original = A1*np.cos(2*np.pi*f1*t + np.pi/4) + A2*np.cos(2*np.pi*f2*t + np.pi/8)

# Графики последовательности, её ДПФ и аналогового восстановления
fig, ax = plt.subplots(3, 2, figsize=(12, 10))
ax[0, 0].stem(n, x, basefmt=" ")
ax[0, 0].set_title("Отсчёты x[n]")
ax[0, 0].set_xlabel("n")
ax[0, 0].set_ylabel("Амплитуда")
ax[0, 1].stem(f_norm, np.abs(X_shift), basefmt=" ")
ax[0, 1].set_title("Модуль ДПФ |X[k]|")
ax[0, 1].set_xlabel("Нормированная частота, циклов/отсчёт")
ax[0, 1].set_ylabel("|X[k]|")
ax[1, 0].plot(t, x_analog_rec, label="Восстановленный сигнал")
ax[1, 0].set_title("Периодический сигнал, восстановленный по ДПФ")
ax[1, 0].set_xlabel("Время t, с")
ax[1, 0].set_ylabel("Амплитуда")
ax[1, 0].legend()
ax[1, 1].stem(k_signed * fd2/N, np.abs(X_a)/N, basefmt=" ")
ax[1, 1].set_title("Амплитудный спектр восстановленного сигнала")
ax[1, 1].set_xlabel("Частота, Гц")
ax[1, 1].set_ylabel("|Xa[k]|/N")
ax[2, 0].plot(t, x_analog_original, color="tab:orange")
ax[2, 0].set_title("Исходный аналоговый сигнал")
ax[2, 0].set_xlabel("Время t, с")
ax[2, 0].set_ylabel("Амплитуда")
ax[2, 1].axis("off")
for axis in ax.flat:
    if axis.axison:
        axis.grid(True)
plt.tight_layout()
plt.show()
print(f"Шаг по времени Δt={0.25*T:.8f} с")
print(f"Максимальное отклонение восстановленного сигнала: {np.max(np.abs(x_analog_rec-x_analog_original)):.3e}")
```

### 2.6 Сравнение спектра и сигналов

Исходя из результатов пункта 2.5, объяснить:  
а) связь модуля ДПФ последовательности со спектром аналогового сигнала;  
б) результат визуального сравнения восстановленного и исходного сигналов.

#### Решение:

Комплексные коэффициенты ряда Фурье восстановленного периодического сигнала равны $X_{\mathrm a}(k)/N$. Для каждой вещественной косинусоиды положительная и отрицательная частоты образуют сопряжённую пару, поэтому амплитуда косинусоиды равна удвоенному модулю одного коэффициента. В этом варианте:

$$
2\cdot\frac{1.06}{2}=1.06\quad (f=500\,\text{Гц}),\qquad
2\cdot\frac{2.12}{2}=2.12\quad (f=750\,\text{Гц}).
$$

Восстановленный сигнал визуально совпадает с исходным. Максимальное отклонение, рассчитанное программой, равно $2.975\cdot10^{-14}$, что обусловлено численной погрешностью.


```python
print(f"Максимальное отклонение восстановленного сигнала от исходного: {np.max(np.abs(x_analog_rec-x_analog_original)):.3e}")
```

### 2.7 Проверка равенства Парсеваля

Проверить равенство Парсеваля и пояснить его смысл.

#### Решение:

Для принятой нормировки ДПФ проверяется равенство Парсеваля:

$$
\sum_{n=0}^{N-1}|x(n)|^2=\frac{1}{N}\sum_{k=0}^{N-1}|X(k)|^2.
$$

При $N=64$ код вычисляет обе части. Для данной последовательности получено

$$
E_{\mathrm{врем}}=179.776000000000,\qquad
E_{\mathrm{част}}=\frac{1}{64}\sum_{k=0}^{63}|X(k)|^2=179.776000000000.
$$

Абсолютная разность равна $2.842\cdot10^{-14}$, то есть обе части совпадают с точностью округления. Равенство показывает, что энергия последовательности во временной области равна нормированной энергии её спектральных отсчётов.


```python
energy_time = np.sum(np.abs(x)**2)
energy_frequency = np.sum(np.abs(X)**2) / N
print(f"Энергия во временной области: {energy_time:.12f}")
print(f"Энергия по ДПФ (Парсеваль):   {energy_frequency:.12f}")
print(f"Абсолютная разность:           {abs(energy_time-energy_frequency):.3e}")
```
