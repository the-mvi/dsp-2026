# Основные теоретические понятия

## 1. Линейные дискретные системы (ЛДС)

**Разностное уравнение** рекурсивного звена 2-го порядка связывает вход $x(n)$ и выход $y(n)$:
$$y(n)=-a_1y(n-1)-a_2y(n-2)+b_0x(n)+b_1x(n-1)+b_2x(n-2).$$

**Импульсная характеристика** $h(n)$ - реакция системы на единичный импульс $u_0(n)$ ($u_0(0)=1$, остальные нули). Для рекурсивной системы $h(n)$ бесконечна по длине и затухает, поэтому её рассматривают как последовательность отсчётов на конечном интервале.

**Свёртка.** Реакция ЛДС-системы на произвольное воздействие равна свёртке воздействия с импульсной характеристикой:
$$y(n)=x(n)*h(n)=\sum_k x(k)\,h(n-k).$$
Длина свёртки последовательностей длин $L_1$ и $L_2$ равна $L_1+L_2-1$.

**Передаточная функция** - $z$-преобразование импульсной характеристики; для звена 2-го порядка
$$H(z)=\frac{Y(z)}{X(z)}=\frac{b_0+b_1z^{-1}+b_2z^{-2}}{1+a_1z^{-1}+a_2z^{-2}}=\frac{b_0z^2+b_1z+b_2}{z^2+a_1z+a_2}.$$

**Нули и полюсы.** Нули - корни числителя, полюсы - корни знаменателя. Комплексные корни записывают в алгебраической $z=x+jy$ и показательной $z=re^{j\varphi}$ формах, где $r=|z|$, $\varphi=\arg z$.

**Устойчивость.** Рекурсивное звено устойчиво тогда и только тогда, когда все полюсы передаточной функции лежат **внутри** единичной окружности: $|\alpha_i|<1$. В этом случае ИХ затухает.

**Частотные характеристики.** АЧХ и ФЧХ получают подстановкой $z=e^{j\hat{\omega}}$ в $H(z)$, где $\hat{\omega}$ - нормированная частота (рад/отсчёт):
$$|H(e^{j\hat{\omega}})|=\text{АЧХ},\qquad \arg H(e^{j\hat{\omega}})=\text{ФЧХ}.$$
Частотная ось нормирована: основная полоса частот для дискретного сигнала - $\hat{\omega}\in[0;\pi]$ (от $0$ до $f_{\text{д}}/2$). Связь нормированной и абсолютной частот: $\hat{\omega}=2\pi f/f_{\text{д}}$.

## 2. Дискретное преобразование Фурье (ДПФ)

**ДПФ** периодической последовательности длины $N$:
$$X(k)=\sum_{n=0}^{N-1}x(n)\,e^{-j\frac{2\pi}{N}nk},\qquad k=0,1,\dots,N-1.$$

**ОДПФ** восстанавливает последовательность:
$$x(n)=\frac{1}{N}\sum_{k=0}^{N-1}X(k)\,e^{j\frac{2\pi}{N}nk},\qquad n=0,1,\dots,N-1.$$

**Спектры.** Амплитудный спектр - $|X(k)|$; фазовый спектр - $\arg X(k)$. Нормированная частота $k$-го отсчёта - $\nu_k=k/N$ (циклов/отсчёт), абсолютная - $f_k=k f_{\text{д}}/N$. Для вещественного сигнала спектр сопряжённо-симметричен: $X(N-k)=\overline{X(k)}$.

Для вещественной косинусоиды $A\cos(2\pi f nT+\varphi)$ с частотой, точно попадающей в бин $k=fN/f_{\text{д}}$, высота пика ДПФ $|X(k)|=AN/2$, а фаза в пике $\arg X(k)=\varphi$ (учёт отрицательной частоты даёт множитель 2 при переходе к амплитуде гармоники $A=2|X(k)|/N$).

**Восстановление аналогового сигнала.** По отсчётам ДПФ периодическую аналоговую функцию восстанавливают суммой комплексных экспонент:
$$x(t)=\frac{1}{N}\sum_{k=-N/2}^{N/2-1}X_а(k)\,e^{j\frac{2\pi}{NT}kt},$$
где $X_а$ - спектр, переупорядоченный в диапазон $[-N/2,\,N/2-1]$. Если все гармоники сигнала лежат ниже $f_{\text{д}}/2$ и имеют целое число периодов на интервале, восстановление точное.

**Равенство Парсеваля** - сохранение энергии при ДПФ:
$$\sum_{n=0}^{N-1}|x(n)|^2=\frac{1}{N}\sum_{k=0}^{N-1}|X(k)|^2.$$

# Часть 1. Линейная дискретная система

## 1. Импульсная характеристика длины $N_1$.

Импульсная характеристика - реакция системы на единичный импульс $u_0(n)$.


```python
#  Решение разностного уравнения
#   y(n) = sum_k b[k]·x(n-k) - sum_{k>=1} a[k]·y(n-k)
def calculate_func(b, a, x):
    x = np.asarray(x, dtype=float)
    L = len(x)
    B = len(b) - 1          # порядок числителя
    A = len(a) - 1          # порядок знаменателя
    out = np.zeros(L)
    for n in range(L):
        acc = 0.0
        for k in range(B + 1):          # прямые связи по x
            if n - k >= 0:
                acc += b[k] * x[n - k]
        for k in range(1, A + 1):       # обратные связи по y (a[0] = 1)
            if n - k >= 0:
                acc -= a[k] * out[n - k]
        out[n] = acc
    return out


b = [b0, b1, b2]
a = [a0, a1, a2]

u0 = np.zeros(N1)
u0[0] = 1.0
h = calculate_func(b, a, u0)

n_axis = np.arange(N1)

plt.figure(figsize=(10,4))
plt.stem(n_axis, h, basefmt=" ")
plt.title("Импульсная характеристика h(n), N1=29")
plt.xlabel("n")
plt.ylabel("h(n)")
plt.grid(alpha=0.3)
plt.tight_layout(); plt.show()

print(f"Первые {N1} отсчётов импульсной характеристики:")
print(np.array2string(h, precision=4, max_line_width=100))
```

**Пояснения к п. 1.**

1. **Аналитическая формула импульсной характеристики рекурсивного звена 2-го порядка** Решается разностное уравнение
$$y(n)=-a_1y(n-1)-a_2y(n-2)+b_0x(n)+b_1x(n-1)+b_2x(n-2),\qquad y(-1)=y(-2)=0.$$
Импульсная характеристика - отклик на $x(n)=u_0(n)$.
$$h(n)=b_0\,u_0(n)+b_1\,u_0(n-1)+b_2\,u_0(n-2)-a_1h(n-1)-a_2h(n-2).$$
Поскольку полюсы комплексно-сопряжённые ( $\alpha$ и $\alpha^*$) и лежат внутри единичной окружности, дискретное решение имеет форму затухающих колебаний:
$$h(n)=\Big(\rho^n\big(C_1\cos(\varphi n)+C_2\sin(\varphi n)\big)\Big)u_1(n),$$
где $\rho=|\alpha|=\sqrt{a_2}$ — модуль полюса, $\varphi=\arg\alpha$ — его аргумент, $u_1(n)$ — единичная ступенька. В нашем случае $\rho=\sqrt{0.754}\approx 0.868$, что и даёт быстрое затухание.

2. Для рекурсивного звена импульсная характеристика бесконечна - она затухает экспоненциально, но теоретически не обращается в ноль.


## 2. Реакция $y_1(n)$ по формуле свёртки.

**Воздействие** - дискретный прямоугольный импульс длины $N_2=39$:
$$x(n)=\begin{cases}1,  & 0\leqslant n < \text{int}(N_2/2),\\ 0,   & \text{int}(N_2/2)\leqslant n \leqslant N_2-1,\end{cases}$$
При $N_2=39$ $\text{int}(39/2)=20$ единичных отсчётов.

Реакция по формуле свёртки:
$$y_1(n)=x(n)*h(n)=\sum_{k}x(k)\,h(n-k).$$


```python
def calculate_convolution(x, h):
    Nx, Nh = len(x), len(h)
    Ny = Nx + Nh - 1
    y = np.zeros(Ny)
    for n in range(Ny):
        s = 0.0
        for k in range(Nx):
            if 0 <= n - k < Nh:
                s += x[k] * h[n - k]
        y[n] = s
    return y


L = int(np.round(N2/2))
x = np.zeros(N2)
x[:L] = 1.0

y1_full = calculate_convolution(x, h)
len_y1 = len(y1_full)
print(f"Длина x = {len(x)}, Длина полной свёртки = {len_y1} = N1+N2-1 = {N1}+{N2}-1")

plt.figure(figsize=(12,4))
plt.subplot(1,2,1)
plt.stem(np.arange(N2), x, basefmt=" ")
plt.title(f"Воздействие x(n), длина N2={N2}")
plt.xlabel("n"); plt.ylabel("x"); plt.grid(alpha=.3)
plt.subplot(1,2,2)
plt.stem(np.arange(len_y1), y1_full, basefmt=" ")
plt.title(f"Реакция по свёртке y1(n), длина {len_y1}")
plt.xlabel("n"); plt.ylabel("y1"); plt.grid(alpha=.3)
plt.tight_layout(); plt.show()
```

**Пояснения к п. 2.**

- **Длина импульса $x$.** Воздействие имеет длину $N_2=39$ отсчётов, из которых первые $\text{int}(39/2)=20$ отсчётов равны 1, а остальные 19 - нули.
- **Длина свёртки.** Результат имеет длину $67=N_1+N_2-1$ отсчётов;
- **Почему её ограничивают до длины воздействия.** В практических задачах удобно работать в той же сетке, что и входной сигнал: результат «физически» относится к тем же моментам времени $n$, что и вход.

## 4. Нули и полюсы передаточной функции.

Передаточная функция
$$H(z)=\frac{b_0+b_1z^{-1}+b_2z^{-2}}{1+a_1z^{-1}+a_2z^{-2}}=\frac{b_0 z^2+b_1 z+b_2}{z^2+a_1 z+a_2}.$$

Нули - корни числителя, полюсы - корни знаменателя.


```python
zeros = np.roots([b0, b1, b2])
alpha = np.roots([a0, a1, a2])

def fmt_polar(z):
    r = np.abs(z)
    phi = np.angle(z)
    return rf"r={r:.6f},  phi={phi:.6f} рад = {phi/np.pi:.6f}·pi  (алг: {z.real:.6f}{z.imag:+.6f}j)"

print("Нули:")
for z in zeros:
    print("  ", fmt_polar(z))
print("Полюсы:")
for p in alpha:
    print("  ", fmt_polar(p))

# Проверка |alpha| < 1 => устойчивость
print("\nМодули полюсов:", np.round(np.abs(alpha),6))
print("Для всех полюсов: |полюс| < 1:", bool(np.all(np.abs(alpha) < 1)))

# Разложение на простейшие множители
# H(z) = b0 * (z-z1)(z-z2) / (z - alpha)(z - alpha*)
print("\nH(z) =", f"{b0:.4f} · (z - z1)(z - z2) / (z - alpha)(z - alpha*)")
```

**Пояснения к п. 4.**

**Нули**: корни $b_0z^2+b_1z+b_2=0$:
$$z_{1,2}=-0.6602\pm j\,1.078951,$$
в показательной форме
$$|z_{1,2}|=\sqrt{b_2/b_0}=\sqrt{1.408/0.88}\approx\sqrt{1.6}=1.2649,\qquad \arg z_1\approx 0.6748\,\pi,\quad \arg z_2\approx -0.6748\,\pi.$$

**Полюсы** (обозначим $\alpha$ и сопряжённый $\alpha^*$): корни $z^2+a_1z+a_2=0$:
$$\alpha,\;\alpha^* = 0.6264\pm j\,0.601351,$$
в показательной форме
$$|\alpha|=|\alpha^*|=\sqrt{a_2}=\sqrt{0.754}\approx 0.8683,\qquad \arg\alpha\approx 0.2435\,\pi,\quad \arg\alpha^*\approx -0.2435\,\pi.$$

**Связь форм.** Алгебраическая $z=x+jy$ и показательная $z=re^{j\varphi}$ связаны соотношениями
$$r=|z|=\sqrt{x^2+y^2},\qquad \varphi=\arg z=\operatorname{arctg}(y,x),\qquad x=r\cos\varphi,\quad y=r\sin\varphi.$$

**Аргументы относительно $\pi$**: для нуля $\pm 0.6748\,\pi$, для полюса $\pm 0.2435\,\pi$.

**Передаточная функция в виде произведения простейших множителей:**
$$H(z)=b_0\,\frac{(z-z_1)(z-z_2)}{(z-\alpha)(z-\alpha^*)}=0.88\,\frac{\big(z-1.2649e^{j0.6748\pi}\big)\big(z-1.2649e^{-j0.6748\pi}\big)}{\big(z-0.8683e^{j0.2435\pi}\big)\big(z-0.8683e^{-j0.2435\pi}\big)}.$$

## 6. АЧХ и ФЧХ в шкале нормированных частот $\hat{\omega}$.

Нормированная частота $\hat{\omega}$ (рад/отсчёт) связана с линейной частотой $f$ соотношением $\hat{\omega}=2\pi f/f_{\text{д}}$. На практике достаточно рассматривать $\hat{\omega}\in[0;\pi]$.


```python
w, H = signal.freqz(b, a, worN=2048, whole=False)  # w в [0, pi]
mag = np.abs(H)
ph  = np.unwrap(np.angle(H))

fig, ax = plt.subplots(2,1, figsize=(10,7))
ax[0].plot(w, mag, 'b', lw=1.5)
ax[0].set_title("АЧХ (нормированная частота)")
ax[0].set_xlabel(r"$\hat{\omega}$, рад/отсчёт")
ax[0].set_ylabel(r"$|H(e^{j\hat{\omega}})|$")
ax[0].grid(alpha=.3)
ax[1].plot(w, ph, 'r', lw=1.5)
ax[1].set_title("ФЧХ (нормированная частота)")
ax[1].set_xlabel(r"$\hat{\omega}$, рад/отсчёт")
ax[1].set_ylabel(r"$\arg H(e^{j\hat{\omega}})$, рад")
ax[1].grid(alpha=.3)
plt.tight_layout(); plt.show()

print(f"Границы основной полосы: 0 .. pi  (т.е. 0 .. f_д/2 = {fd1/2:.0f} Гц)")
print(f"|H(0)| = {mag[0]:.4f}")
print(f"|H(pi)| = {mag[-1]:.4f}")
print(f"max|H| = {np.max(mag):.4f}  на частоте w={w[np.argmax(mag)]:.4f} рад/отсчёт ({w[np.argmax(mag)]/np.pi:.4f}·pi)")
```

- **Соответствие карты нулей и полюсов и АЧХ.** Поскольку нули лежат вне окружности, глубина провалов АЧХ конечна; заметный подъём приходится на частоты вблизи угла полюса (~$0.24\pi$), а максимум АЧХ достигается там, где $|e^{j\omega}-\alpha|$ минимален, т.е. вблизи полюса — на частоте $w_{\max}\approx 0.24\pi$.

- **Какие составляющие подавлены.** Судя по АЧХ, к верхней границе полосы ($\hat{\omega}\to\pi$) коэффициент передачи заметно падает ($|H|$ убывает от 6.88 до 0.37, т.е. примерно в 18 раз), поэтому высокочастотные составляющие воздействия подавлены существенно сильнее, чем низкочастотные.

Нормированная частота $\hat{\omega}$ меняется от 0 до $\pi$, а $f$ - от 0 до $f_{\text{д}}/2$. По сути шкала $f$ - это шкала $\hat{\omega}$, растянутая в $f_{\text{д}}/(2\pi)\approx 3024$ раза.

# Часть 2. ДПФ и спектральный анализ периодических последовательностей

Периодическая последовательность
$$x(nT)=A_1\cos\left(2\pi f_1 nT+\frac{\pi}{4}\right)+A_2\cos\left(2\pi f_2 nT+\frac{\pi}{8}\right),\qquad n=0,1,\dots,N-1$$


```python
n = np.arange(N)                      # дискретная сетка (номер отсчёта)
tt = n*T                              # дискретное время в секундах: t = n*T
x_per = A1*np.cos(2*np.pi*f1*tt + np.pi/4) + A2*np.cos(2*np.pi*f2*tt + np.pi/8)

fig, ax = plt.subplots(1, 2, figsize=(13,4))

ax[0].stem(n, x_per, basefmt=" ")
ax[0].set_title("Дискретное время (номер отсчёта $n$)")
ax[0].set_xlabel("n (дискретное время)"); ax[0].set_ylabel("x(nT)"); ax[0].grid(alpha=.3)

n_norm = n / N
ax[1].stem(n_norm, x_per, basefmt=" ")
ax[1].set_title("Дискретное нормированное время ($\\bar n = n/N$)")
ax[1].set_xlabel(r"$\bar n = n/N$, доли периода")
ax[1].set_ylabel("x(nT)")
ax[1].grid(alpha=.3)
ax[1].set_xlim(0, 1)

plt.tight_layout(); plt.show()
```

## 2. ОДПФ от ДПФ последовательности $x(n)$.

ДПФ: $X(k)=\sum_{n=0}^{N-1}x(n)e^{-j\frac{2\pi}{N}nk}$, $k=0..N-1$. Обратное (ОДПФ): $x(n)=\tfrac{1}{N}\sum_{k=0}^{N-1}X(k)e^{j\frac{2\pi}{N}nk}$.


```python
def calculate_dft(x):
    x = np.asarray(x, dtype=float)
    N = len(x)
    X = np.zeros(N, dtype=complex)
    n = np.arange(N)
    for k in range(N):
        X[k] = np.sum(x * np.exp(-2j * np.pi * k * n / N))
    return X

def calculate_idft(X):
    X = np.asarray(X, dtype=complex)
    N = len(X)
    x = np.zeros(N, dtype=complex)
    k = np.arange(N)
    for n in range(N):
        x[n] = np.sum(X * np.exp(2j * np.pi * k * n / N)) / N
    return x



X = calculate_dft(x_per)                # ДПФ,
x_idft = calculate_idft(X)              # ОДПФ

err = np.max(np.abs(x_per - x_idft.real))
print(f"Макс. отклонение восстановленной x (ОДПФ) от исходной: {err:.2e}")

plt.figure(figsize=(9,4))
plt.plot(n, x_idft.real, 'o-', ms=4, lw=1, alpha=.8, label="ОДПФ(x(n))")
plt.plot(n, x_per, 'r--', lw=1, alpha=.8, label="исходная x(n)")
plt.title("Восстановление x(n) через ОДПФ от ДПФ")
plt.xlabel("n (дискретное нормированное время)")
plt.ylabel("x(n)"); plt.legend(); plt.grid(alpha=.3)
plt.tight_layout(); plt.show()
```

**Пояснения к п. 2.** Последовательность $x(n)$ восстановлена по формуле ОДПФ и полностью совпадает с исходной. Это подтверждает свойство обратимости пары ДПФ/ОДПФ: $\text{ОДПФ}[\text{ДПФ}[x]]=x$.


## 3. Амплитудный и фазовый спектры последовательности $x(n)$.

Для вещественного сигнала из двух косинусоид с амплитудами $A_1,A_2$ амплитудный спектр ДПФ имеет четыре ненулевых «спайка»: на частотах $\pm f_1$ и $\pm f_2$ с высотой $A_1N/2$ и $A_2N/2$ соответственно (половинные амплитуды из-за разложения косинуса на две комплексные экспоненты).


```python
k = np.arange(N)
mag2 = np.abs(X)
ph2  = np.angle(X)

mask = mag2 > 1e-9 * mag2.max()
ph2_clean = np.where(mask, ph2, 0.0)

nu = k / N
f_abs = nu * fd2

fig, ax = plt.subplots(2,2, figsize=(13,7))
ax[0,0].stem(nu, mag2, basefmt=" ")
ax[0,0].set_title("Амплитудный спектр (нормированная частота)")
ax[0,0].set_xlabel("$\\nu=k/N$, циклов/отсчёт")
ax[0,0].set_ylabel("$|X(k)|$")
ax[0,0].grid(alpha=.3)

ax[0,1].stem(nu, ph2_clean, basefmt=" ")
ax[0,1].set_title("Фазовый спектр (нормированная частота)")
ax[0,1].set_xlabel("$\\nu=k/N$")
ax[0,1].set_ylabel(r"$\arg X(k)$, рад")
ax[0,1].grid(alpha=.3)

ax[1,0].stem(f_abs, mag2, basefmt=" ")
ax[1,0].set_title("Амплитудный спектр (абсолютная частота)")
ax[1,0].set_xlabel("f, Гц"); ax[1,0].set_ylabel("$|X(k)|$")
ax[1,0].grid(alpha=.3)

ax[1,1].stem(f_abs, ph2_clean, basefmt=" ")
ax[1,1].set_title("Фазовый спектр (абсолютная частота)")
ax[1,1].set_xlabel("f, Гц")
ax[1,1].set_ylabel(r"$\arg X(k)$, рад")
ax[1,1].grid(alpha=.3)

plt.tight_layout()
plt.show()

peak_k1 = round(f1/fd2*N)
peak_k2 = round(f2/fd2*N)
print(f"Пики на k={peak_k1} (f={peak_k1/N*fd2:.0f} Гц, A1·N/2={A1*N/2:.0f}) и k={peak_k2} (f={peak_k2/N*fd2:.0f} Гц, A2·N/2={A2*N/2:.0f})")
print(f"Магнитуды: |X[{peak_k1}]|={mag2[peak_k1]:.2f}, |X[{peak_k2}]|={mag2[peak_k2]:.2f}")
print(f"Фазы в пиках: arg X[{peak_k1}]={ph2[peak_k1]:.3f} рад = {ph2[peak_k1]/np.pi:.3f}·pi,  arg X[{peak_k2}]={ph2[peak_k2]:.3f} рад = {ph2[peak_k2]/np.pi:.3f}·pi")
```

**Пояснения к п. 3.**

- **Связь дискретного нормированного и дискретного времени.** Дискретное нормированное время - это номер отсчёта $n$ (безразмерное время, выраженное в периодах дискретизации $T$); дискретное время в секундах - $t=nT$. При построении в нормированной шкале по оси абсцисс откладывают $n$, в абсолютной - $t=nT$.

- **Связь частоты $f$ (Гц) и дискретной нормированной частоты.** Нормированная частота $\nu=f/f_{\text{д}}$ (циклов/отсчёт), либо $\hat{\omega}=2\pi\nu$ (рад/отсчёт). Для $k$-го отсчёта ДПФ: $\nu_k=k/N$ и $f_k=k f_{\text{д}}/N=k/(NT)$.

- **Вид спектров.** Амплитудный спектр состоит из двух пар ненулевых отсчётов на частотах $\pm f_1$ и $\pm f_2$. $f_1=1250$ и $f_2=1875$ Гц попадают точно на целочисленные бины ($k_1=f_1N/f_{\text{д}}=8$, $k_2=12$), все остальные отсчёты равны нулю. Высота положительных пиков $|X(k_i)|=A_i N/2$: для $f_1$  $1.19\cdot 64/2=38.08$, для $f_2$  $2.38\cdot 64/2=76.16$. Фазовый спектр в пиках равен начальным фазам гармоник: $\arg X(k_1)=+\pi/4$ и $\arg X(k_2)=+\pi/8$; на отрицательных частотах фазы противоположного знака.


## 4. ДПФ конечной последовательности $x(n)$ длины $N$ и амплитудный спектр периодической.

Под «конечной последовательностью» понимается один период $x(n)$, $n=0..N-1$, «амплитудный спектр периодической последовательности» - модуль её ДПФ, нормированный на число отсчётов (высоты пиков $A_iN/2$).


```python
Xc = calculate_dft(x_per)   # ДПФ конечной последовательности (один период)
magc = np.abs(Xc)

# Амплитудный спектр периодической последовательности: амплитуды гармоник (высоты пиков)
# равны 2|X|/N с учётом пары экспонент
amp_spec = 2*magc/N

nu = k/N

plt.figure(figsize=(12,4))
plt.subplot(1,2,1)
plt.stem(nu, magc, basefmt=" ")
plt.title("Модуль ДПФ конечной последовательности $|X(k)|$")
plt.xlabel("$\\nu$ = k/N"); plt.ylabel("$|X(k)|$"); plt.grid(alpha=.3)
plt.subplot(1,2,2)
plt.stem(nu, amp_spec, basefmt=" ")
plt.title("Амплитудный спектр периодической последовательности (A=2|X|/N)")
plt.xlabel("$\\nu$ = k/N"); plt.ylabel("амплитуда гармоники"); plt.grid(alpha=.3)
plt.tight_layout(); plt.show()

print(f"Амплитуды, извлечённые из спектра: на k={peak_k1}: {amp_spec[peak_k1]:.3f} (ожид. {A1}), на k={peak_k2}: {amp_spec[peak_k2]:.3f} (ожид. {A2})")
```

## 5. Восстановление периодического аналогового сигнала $x(t)$ по отсчётам ДПФ.

По заданной формуле для $t\in[0;(N-1)T]$ с шагом $\Delta t=0.25T$:
$$x(t)=\frac{1}{N}\sum_{k=-N/2}^{N/2-1}X_а(k)\exp\left(j\frac{2\pi}{NT}kt\right),\qquad
X_а(k)=\begin{cases}X(N+k), & -N/2\leqslant k\leqslant -1\\ X(k), & 0\leqslant k\leqslant N/2-1\end{cases}$$


```python
Xa = np.fft.fftshift(X)
k_a = np.arange(-N//2, N//2)

# сетка времени t с шагом 0.25T
dT = 0.25*T
t_grid = np.arange(0, (N-1)*T + 1e-12, dT)
K = len(t_grid)

# восстановление: x(t)=1/N * sum_k Xa(k) exp(j 2pi/(NT) k t)
x_rec = np.zeros(K, dtype=complex)
for i, ti in enumerate(t_grid):
    x_rec[i] = (1.0/N) * np.sum(Xa * np.exp(1j*2*np.pi/(N*T)*k_a*ti))
x_rec = x_rec.real

# исходный аналоговый сигнал x(t) = x(nT)|_{nT=t}
x_cont = A1*np.cos(2*np.pi*f1*t_grid + np.pi/4) + A2*np.cos(2*np.pi*f2*t_grid + np.pi/8)

print(f"K={K} точек, шаг dT={dT} с, интервал [0; {(N-1)*T:.5f}] с")
print(f"Макс. отклонение восстановленного x(t) от исходного аналогового: {np.max(np.abs(x_rec - x_cont)):.2e}")

fig, ax = plt.subplots(2,2, figsize=(14,8))
ax[0,0].stem(n, x_per, basefmt=" ")
ax[0,0].set_title("Периодическая последовательность x(n)")
ax[0,0].set_xlabel("n")
ax[0,0].set_ylabel("x(n)")
ax[0,0].grid(alpha=.3)

ax[0,1].stem(nu, mag2, basefmt=" ")
ax[0,1].set_title("Модуль ДПФ |X(k)|")
ax[0,1].set_xlabel("k")
ax[0,1].set_ylabel("|X(k)|")
ax[0,1].grid(alpha=.3)

ax[1,0].plot(t_grid*1000, x_rec, 'b-', lw=1.2)
ax[1,0].set_title("Восстановленный аналоговый сигнал x(t)")
ax[1,0].set_xlabel("t, мс")
ax[1,0].set_ylabel("x(t)")
ax[1,0].grid(alpha=.3)

ax[1,1].plot(np.fft.fftshift(np.abs(np.fft.fft(x_rec, n=4096))), color='m', lw=1.2)
ax[1,1].set_title("Амплитудный спектр восстановленного x(t) (4096 отсчётов)")
ax[1,1].set_xlabel("bin")
ax[1,1].set_ylabel("|ДПФ|")
ax[1,1].grid(alpha=.3)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,4))
plt.plot(t_grid*1000, x_cont, 'g-', lw=1.4, label="x(t)=x(nT)|_{nT=t} (непрерывный)")
plt.plot(n*T*1000, x_per, 'ro', ms=4, label="отсчёты x(nT)")
plt.title("Исходный аналоговый сигнал и его отсчёты")
plt.xlabel("t, мс"); plt.ylabel("x"); plt.legend(); plt.grid(alpha=.3)
plt.tight_layout(); plt.show()

print("Визуально восстановленный и исходный сигналы накладываются друг на друга (отклонение ~1e-15).")
```

## **Пояснения к п. 6.**

- **Связь модуля ДПФ последовательности со спектром аналогового сигнала.** Модуль ДПФ $|X(k)|$ с ненулевыми отсчётами на бинах $k_1=8$, $k_2=12$ (и симметричных отрицательных) отражает амплитуды и частоты двух косинусоид аналогового сигнала. Сам аналоговый сигнал $x(t)$ - суперпозиция чистых тонов $f_1$ и $f_2$; его амплитудный спектр - две линии (дельта-функции) на $\pm f_1$, $\pm f_2$. При восстановлении $x(t)$ суммой комплексных экспонент именно эти отсчёты ДПФ дают вклад, воспроизводя оба тона с нужными амплитудами и фазами — потому что частоты попали точно в бины ДПФ.

- **Результат визуального сравнения.** Восстановленный $x(t)$ и исходный аналоговый сигнал практически неразличимы: максимальное отклонение ~$10^{-15}$ (уровень машинной точности). Поскольку сигнал состоит из гармоник с частотой ниже $f_{\text{д}}/2$ и с целым числом периодов на интервале, ДПФ с интерполяцией по формуле даёт точное восстановление аналоговой формы между отсчётами.


## 7. Проверка равенства Парсеваля.

Равенство Парсеваля для ДПФ:
$$\sum_{n=0}^{N-1}|x(n)|^2=\frac{1}{N}\sum_{k=0}^{N-1}|X(k)|^2.$$
Левая часть - энергия сигнала во временной области, правая - та же энергия, выраженная через квадраты модулей спектральных отсчётов (разделённая на $N$). Равенство означает сохранение энергии при переходе из временной области в частотную.


```python
E_t  = np.sum(x_per**2)                 # энергия во временной области
E_f  = (1.0/N) * np.sum(np.abs(X)**2)  # энергия в частотной области

print(f"Энергия (время):  sum |x|^2        = {E_t:.12f}")
print(f"Энергия (частота) 1/N sum |X|^2    = {E_f:.12f}")
print(f"Относительная разница            = {abs(E_t-E_f)/E_t:.3e}")
print()
E_theory = 0.5*(A1**2 + A2**2)*N
print(f"Теоретическая энергия 0.5(A1^2+A2^2)·N = {E_theory:.4f}")
```

**Пояснения к п. 6.**

Равенство Парсеваля выполняется с точностью до $10^{-16}$, т.е. практически точно. Энергия сигнала, посчитанная по временным отсчётам ($\sum|x(n)|^2$) и по коэффициентам ДПФ ($\tfrac1N\sum|X(k)|^2$), совпадает. Физический смысл: дискретное преобразование Фурье является унитарным и не изменяет энергию сигнала  она лишь перераспределяется между спектральными составляющими. Теоретическое значение $E=\tfrac12(A_1^2+A_2^2)N$ также совпадает.
