## Основные теоретические положения

### Линейная дискретная система

Линейная дискретная система (ЛДС) ставит в соответствие входной последовательности $x(n)$ (воздействию) выходную последовательность $y(n)$ (реакцию). Система линейна, если для неё выполняется принцип суперпозиции, и стационарна, если её параметры не меняются во времени.

**Импульсная характеристика** $h(n)$ — реакция ЛДС на цифровой единичный импульс
$$u_0(n) = \begin{cases} 1, & n = 0,\\ 0, & n \ne 0 \end{cases}$$
при нулевых начальных условиях (все отсчёты $x$ и $y$ с отрицательными номерами равны нулю).

**Формула свёртки.** При нулевых начальных условиях реакция ЛДС на произвольное воздействие равна свёртке воздействия с ИХ:
$$y(n) = \sum_{m=0}^{n} h(m)\,x(n-m) = \sum_{m=0}^{n} x(m)\,h(n-m).$$
Свёртка последовательностей длины $N_h$ и $N_x$ имеет длину $N_h + N_x - 1$.

**Разностное уравнение** связывает текущий отсчёт реакции с текущим и прошлыми отсчётами воздействия и прошлыми отсчётами реакции. Для рекурсивного звена 2-го порядка:
$$y(n) = b_0x(n) + b_1x(n-1) + b_2x(n-2) - a_1y(n-1) - a_2y(n-2).$$
Из-за обратной связи по $y$ ИХ рекурсивного звена бесконечна (БИХ-система).

### Передаточная функция, нули и полюсы

**Передаточная функция** — $Z$-преобразование ИХ, или отношение $Z$-изображений реакции и воздействия:
$$H(z) = \sum_{n=0}^{\infty} h(n)z^{-n} = \frac{b_0 + b_1z^{-1} + b_2z^{-2}}{1 + a_1z^{-1} + a_2z^{-2}}.$$
**Нули** — корни числителя, **полюсы** — корни знаменателя. Комплексное число записывается в алгебраической форме $z = \operatorname{Re}z + j\operatorname{Im}z$ или в показательной $z = re^{j\varphi}$, где $r = |z|$, $\varphi = \arg z$. При вещественных коэффициентах комплексные нули и полюсы образуют сопряжённые пары, и
$$H(z) = b_0\frac{(1 - z_1z^{-1})(1 - z_2z^{-1})}{(1 - p_1z^{-1})(1 - p_2z^{-1})}.$$

**ИХ звена 2-го порядка.** Если полюсы комплексно-сопряжённые, $p_{1,2} = r_*e^{\pm j\varphi_*}$, то $a_1 = -2r_*\cos\varphi_*$, $a_2 = r_*^2$, и с учётом нулевых начальных условий
$$h(n) = b_0\,g(n) + b_1\,g(n-1) + b_2\,g(n-2),\qquad g(n) = \begin{cases} r_*^{\,n}\dfrac{\sin\big((n+1)\varphi_*\big)}{\sin\varphi_*}, & n \ge 0,\\ 0, & n < 0, \end{cases}$$
где $g(n)$ — ИХ звена с числителем, равным 1.

**Устойчивость.** ЛДС устойчива, если при ограниченном воздействии реакция ограничена. Необходимое и достаточное условие: $\sum_{n}|h(n)| < \infty$. Это равносильно тому, что все полюсы лежат внутри единичного круга, $|p_k| < 1$.

### Частотные характеристики

**Частотная характеристика** — передаточная функция на единичной окружности:
$$H(e^{j\hat\omega}) = H(z)\big|_{z = e^{j\hat\omega}} = A(\hat\omega)\,e^{j\varphi(\hat\omega)},$$
где $A(\hat\omega) = |H(e^{j\hat\omega})|$ — АЧХ, $\varphi(\hat\omega) = \arg H(e^{j\hat\omega})$ — ФЧХ, $\hat\omega = \omega T = 2\pi f/f_д$ — нормированная частота. Частотная характеристика периодична по $\hat\omega$ с периодом $2\pi$. Для вещественной ИХ АЧХ чётная, а ФЧХ нечётная, поэтому основная полоса — $[0;\pi]$ по $\hat\omega$ или $[0;f_д/2]$ по $f$.

Геометрически $A(\hat\omega)$ пропорциональна произведению расстояний от точки $e^{j\hat\omega}$ до нулей, делённому на произведение расстояний до полюсов. Вблизи полюса АЧХ растёт, вблизи нуля падает. Если нуль лежит на окружности, АЧХ в этой точке равна нулю, а ФЧХ скачком меняется на $\pi$.

### ДПФ

Прямое и обратное ДПФ последовательности длины $N$:
$$X(k) = \sum_{n=0}^{N-1}x(n)e^{-j\frac{2\pi}{N}kn},\qquad x(n) = \frac{1}{N}\sum_{k=0}^{N-1}X(k)e^{j\frac{2\pi}{N}kn}.$$
- $n$ — дискретное нормированное время, $nT$ — дискретное время.
- $k$ — дискретная нормированная частота, $f_k = k\,f_д/N = k/(NT)$ — абсолютная частота; шаг $\Delta f = f_д/N$.
- $|X(k)|$ — амплитудный спектр, $\arg X(k)$ — фазовый спектр.
- Для вещественной $x(n)$ выполняется $X(N-k) = X^*(k)$: отсчёты $k > N/2$ соответствуют отрицательным частотам $(k - N)\Delta f$.
- Дискретная гармоника $A\cos\left(\frac{2\pi}{N}k_0n + \varphi_0\right)$ даёт ровно два ненулевых отсчёта ДПФ: $X(k_0) = \frac{AN}{2}e^{j\varphi_0}$ и $X(N-k_0) = \frac{AN}{2}e^{-j\varphi_0}$.

Для периодической последовательности ДПФ одного периода даёт её дискретный спектр. Для конечной последовательности ДПФ — это $N$ равноотстоящих отсчётов её непрерывной спектральной плотности $X(e^{j\hat\omega})$ на частотах $\hat\omega_k = 2\pi k/N$.

**Теорема Котельникова.** Аналоговый сигнал со спектром, ограниченным частотой $f_{max}$, однозначно восстанавливается по отсчётам, взятым с частотой $f_д > 2f_{max}$.

**Равенство Парсеваля** для ДПФ:
$$\sum_{n=0}^{N-1}|x(n)|^2 = \frac{1}{N}\sum_{k=0}^{N-1}|X(k)|^2.$$

### Часть 1

#### 1.1. Импульсная характеристика

ИХ — реакция звена на единичный импульс $u_0(n)$ при нулевых начальных условиях. Разностное уравнение реализую функцией `diff_eq`, она понадобится и в п. 1.3. Вычисляю $N_1 = 20$ отсчётов.

```python
def diff_eq(b, a, x):
    # y(n) = b0 x(n) + b1 x(n-1) + b2 x(n-2) - a1 y(n-1) - a2 y(n-2),
    # отсчёты с отрицательными номерами считаются нулевыми
    y = np.zeros(len(x))
    for n in range(len(x)):
        acc = sum(b[i] * x[n - i] for i in range(len(b)) if n - i >= 0)
        acc -= sum(a[k] * y[n - k] for k in range(1, len(a)) if n - k >= 0)
        y[n] = acc / a[0]
    return y

n1 = np.arange(N1)
u0 = (n1 == 0).astype(float)
h = diff_eq(b, a, u0)
print("h(n) =", h)

fig, ax = plt.subplots()
stem(ax, n1, h)
ax.set_title(f"Импульсная характеристика h(n), N1 = {N1}")
ax.set_xlabel("n"); ax.set_ylabel("h(n)")
plt.show()
```

**Аналитическая формула ИХ.** По формуле из теории нахожу модуль и аргумент полюса из соотношений $a_2 = r_*^2$, $a_1 = -2r_*\cos\varphi_*$ и проверяю, что формула даёт те же отсчёты, что и разностное уравнение.

```python
r_p = np.sqrt(a2)
phi_p = np.arccos(-a1 / (2 * r_p))
print(f"r* = {r_p:.4f},  φ* = {phi_p:.4f} рад = {phi_p/np.pi:.4f}π")

def g(n):
    n = np.asarray(n)
    return np.where(n >= 0, r_p ** n * np.sin((n + 1) * phi_p) / np.sin(phi_p), 0.0)

h_formula = b0 * g(n1) + b1 * g(n1 - 1) + b2 * g(n1 - 2)
print("max |h - h_formula| =", np.max(np.abs(h - h_formula)))
```

Подставив $r_* = 0.8367$, $\varphi_* = 0.7105\pi$, получаю
$$h(n) = 0.7\,g(n) - 0.81214\,g(n-1) + 0.56\,g(n-2),\qquad g(n) = 0.8367^{\,n}\,\frac{\sin\big(0.7105\pi(n+1)\big)}{\sin(0.7105\pi)},\ \ n \ge 0;\qquad g(n) = 0,\ n < 0.$$
Формула совпадает с расчётом по разностному уравнению с точностью до ошибок округления ($\sim10^{-16}$).

Пояснения:
- $h(0) = b_0 = 0.7$. Первые три отсчёта формируются коэффициентами числителя вместе с обратной связью. Начиная с $n = 3$, входной импульс уже не действует, и отсчёты порождаются только рекурсивной частью: $h(n) = -a_1h(n-1) - a_2h(n-2)$.
- Отсчёты почти каждый раз меняют знак. Причина в аргументе полюса $\varphi_* \approx 0.71\pi$: он близок к $\pi$, поэтому синус меняет знак примерно через отсчёт.
- Множитель $0.8367^{\,n}$ убывает, но нулём не становится ни при каком $n$. Поэтому **действительная длина ИХ бесконечна**: звено рекурсивное, и каждый отсчёт зависит от предыдущих. $N_1 = 20$ — лишь интервал, на котором ИХ вычислена и показана. К $n = 19$ отсчёты уменьшились примерно в 20 раз ($|h(19)| \approx 0.08$), но нулю не равны.

#### 1.2. Реакция по формуле свёртки

Воздействие — прямоугольный импульс в массиве длины $N_2 = 30$: единицы при $0 \le n < int(N_2/2)$, нули при $int(N_2/2) \le n \le N_2 - 1$. Число $N_2 = 30$ чётное, поэтому $int(30/2) = 15$ без неоднозначности округления. Функция `int_round` округляет к ближайшему целому (половины вверх). Встроенный `round` округлял бы половины к чётному.

Свёртку ИХ длины $N_1$ с воздействием вычисляю функцией `np.convolve`.

```python
def int_round(t):
    return int(np.floor(t + 0.5))

n2 = np.arange(N2)
L = int_round(N2 / 2)                 # число единичных отсчётов
x = (n2 < L).astype(float)

y_full = np.convolve(h, x)
print(f"единиц в импульсе: {L}, длина массива воздействия: {len(x)}")
print(f"длина полной свёртки: {len(y_full)} = N1 + N2 - 1 = {N1 + N2 - 1}")
nz = np.flatnonzero(np.abs(y_full) > 1e-12)
print(f"ненулевые отсчёты свёртки: n = {nz[0]}...{nz[-1]} ({nz[-1] + 1} шт. = N1 + L - 1)")

y1 = y_full[:N2]                      # ограничиваем длиной воздействия

fig, axs = plt.subplots(3, 1, figsize=(11, 9))
stem(axs[0], n2, x)
axs[0].set_title(f"Воздействие x(n): {L} единиц, длина массива N2 = {N2}")
stem(axs[1], np.arange(len(y_full)), y_full, color="tab:green")
axs[1].set_title(f"Полная свёртка, длина {len(y_full)}")
stem(axs[2], n2, y1, color="tab:green")
axs[2].set_title("Реакция y1(n), ограниченная длиной воздействия N2 = 30")
for ax in axs: ax.set_xlabel("n")
plt.tight_layout(); plt.show()
```

Реакция не повторяет форму импульса. Наибольшие колебания возникают сразу после переднего ($n = 0\ldots3$) и заднего ($n = 15\ldots17$) фронтов. На «полке» импульса реакция колеблется около небольшого значения, а после его окончания затухает так же, как ИХ. Причина в том, что звено слабо передаёт постоянную составляющую ($H(1) \approx 0.16$, см. п. 1.6) и выделяет быстрые изменения сигнала, то есть фронты.

**а) Длина импульса $x$.** Массив воздействия имеет длину $N_2 = 30$. Сам импульс (ненулевая часть) занимает $int(N_2/2) = 15$ отсчётов, $n = 0\ldots14$.

**б) Длина свёртки.**
- *Аналитически:* $N_1 + N_2 - 1 = 20 + 30 - 1 = 49$, отсчёты $n = 0\ldots48$.
- *По графику:* ненулевые отсчёты идут до $n = 33$, их $34 = N_1 + 15 - 1$. Последние 15 нулей воздействия ничего не добавляют, поэтому при $n = 34\ldots48$ свёртка равна нулю.

**в) Почему свёртку ограничивают длиной воздействия.**
1. Реакцию нужно сравнить с $y_2(n)$ из п. 1.3, а разностное уравнение даёт по одному выходному отсчёту на каждый входной, то есть $N_2$ отсчётов. Сравнивать можно только последовательности одной длины.
2. Хвост полной свёртки недостоверен. В нём должны участвовать отброшенные отсчёты $h(20), h(21), \ldots$, а их нет. Например, нули при $n \ge 34$ — артефакт усечения ИХ: настоящая реакция звена там продолжает затухающе колебаться.

#### 1.3. Реакция по разностному уравнению

Разностное уравнение звена с коэффициентами варианта:
$$y(n) = 0.7x(n) - 0.81214x(n-1) + 0.56x(n-2) - 1.0278y(n-1) - 0.7y(n-2).$$

```python
y2 = diff_eq(b, a, x)
print(f"длина воздействия {len(x)}, длина реакции {len(y2)}")

err = np.abs(y1 - y2)
print(f"max |y1 - y2| при n < N1: {err[:N1].max():.1e}")
print(f"max |y1 - y2| при n >= N1: {err[N1:].max():.4f} (n = {N1 + err[N1:].argmax()})")

fig, axs = plt.subplots(3, 1, figsize=(11, 9))
stem(axs[0], n2, x)
axs[0].set_title("Воздействие x(n)")
stem(axs[1], n2, y2, color="tab:purple")
axs[1].set_title("Реакция y2(n) по разностному уравнению")
axs[2].plot(n2, y1, "o-", ms=4, label="y1 — по формуле свёртки")
axs[2].plot(n2, y2, "x--", ms=6, label="y2 — по разностному уравнению")
axs[2].set_title("Сравнение реакций y1(n) и y2(n)")
axs[2].legend(loc="upper right")
for ax in axs: ax.set_xlabel("n")
plt.tight_layout(); plt.show()
```

Графики $y_1(n)$ и $y_2(n)$ при $n = 0\ldots19$ совпадают. Пока $n < N_1$, в сумму свёртки входят только известные отсчёты $h(0)\ldots h(n)$, и оба способа дают одно и то же. С $n = 20$ появляется расхождение (до $0.055$ при $n = 21$). Свёртке уже нужны отброшенные отсчёты $h(20), h(21), \ldots$, а разностное уравнение от усечения ИХ не зависит. Поэтому $y_2(n)$ — точная реакция звена на интервале $[0;N_2-1]$, а $y_1(n)$ — её приближение.

**Длины.** Воздействие — массив из $N_2 = 30$ отсчётов (из них 15 единиц). Реакция по разностному уравнению вычислена тоже на $N_2 = 30$ отсчётах: на каждый входной отсчёт приходится один выходной. Истинная реакция рекурсивного звена бесконечна, после окончания импульса оно продолжает колебаться, так что 30 — это лишь интервал вычисления.

#### 1.4. Нули и полюсы

Умножив числитель и знаменатель на $z^2$, получаю $H(z) = \dfrac{0.7z^2 - 0.81214z + 0.56}{z^2 + 1.0278z + 0.7}$. Нули — корни числителя, полюсы — корни знаменателя.

```python
zeros = np.roots(b)
poles = np.roots(a)

def describe(name, z):
    print(f"{name} = {z.real:+.4f} {z.imag:+.4f}j = {abs(z):.4f}·exp(j·{np.angle(z)/np.pi:+.4f}π)")

for i, z in enumerate(sorted(zeros, key=lambda c: -c.imag), 1): describe(f"z{i}", z)
for i, p in enumerate(sorted(poles, key=lambda c: -c.imag), 1): describe(f"p{i}", p)
```

| | алгебраическая форма | показательная форма | аргумент относительно $\pi$ |
|---|---|---|---|
| $z_{1,2}$ | $0.5801 \pm j0.6808$ | $0.8944\,e^{\pm j0.2754\pi}$ | $\pm 0.2754\pi$ |
| $p_{1,2}$ | $-0.5139 \pm j0.6602$ | $0.8367\,e^{\pm j0.7105\pi}$ | $\pm 0.7105\pi$ |

**Связь форм.** Модуль $r = \sqrt{(\operatorname{Re}z)^2 + (\operatorname{Im}z)^2}$, аргумент $\varphi = \operatorname{atan2}(\operatorname{Im}z, \operatorname{Re}z)$. Обратно: $\operatorname{Re}z = r\cos\varphi$, $\operatorname{Im}z = r\sin\varphi$ (формула Эйлера $re^{j\varphi} = r\cos\varphi + jr\sin\varphi$). Коэффициенты вещественные, поэтому нули и полюсы — комплексно-сопряжённые пары с одинаковым модулем и противоположными аргументами. Модуль полюса $0.8367 = \sqrt{a_2}$ и его аргумент $0.7105\pi$ — это те самые $r_*$ и $\varphi_*$ из формулы ИХ в п. 1.1.

**Произведение простейших множителей:**
$$H(z) = 0.7\,\frac{\left(1 - 0.8944e^{j0.2754\pi}z^{-1}\right)\left(1 - 0.8944e^{-j0.2754\pi}z^{-1}\right)}{\left(1 - 0.8367e^{j0.7105\pi}z^{-1}\right)\left(1 - 0.8367e^{-j0.7105\pi}z^{-1}\right)}.$$

#### 1.5. Карта нулей и полюсов

```python
fig, ax = plt.subplots(figsize=(6.5, 6.5))
t_c = np.linspace(0, 2 * np.pi, 500)
ax.plot(np.cos(t_c), np.sin(t_c), "k", lw=1, label="|z| = 1")
ax.plot(zeros.real, zeros.imag, "o", ms=11, mfc="none", mew=2, color="tab:blue", label="нули")
ax.plot(poles.real, poles.imag, "x", ms=12, mew=2.5, color="tab:red", label="полюсы")
for z in np.r_[zeros, poles]:
    ax.annotate(f"{abs(z):.3f}·e^(j{np.angle(z)/np.pi:+.3f}π)", (z.real, z.imag),
                xytext=(6, 8 if z.imag > 0 else -16), textcoords="offset points", fontsize=9)
ax.axhline(0, color="k", lw=0.6); ax.axvline(0, color="k", lw=0.6)
ax.set_xlim(-1.5, 1.5); ax.set_ylim(-1.5, 1.5); ax.set_aspect("equal")
ax.set_xlabel("Re z"); ax.set_ylabel("Im z"); ax.set_title("Карта нулей и полюсов")
ax.legend(loc="lower left")
plt.show()

print("модули полюсов:", np.abs(poles), "-> все внутри единичного круга:", bool(np.all(np.abs(poles) < 1)))
```

**б) Совпадение с п. 4.** Точки на карте стоят там, где их дал расчёт в п. 1.4: нули $0.894e^{\pm j0.275\pi}$ в правой полуплоскости, полюсы $0.837e^{\pm j0.711\pi}$ в левой. Значения совпадают.

#### 1.6. АЧХ и ФЧХ в шкале нормированных частот

Частотную характеристику вычисляю функцией `signal.freqz`: она подставляет $z = e^{j\hat\omega}$ в $H(z)$ на сетке частот от 0 до $\pi$.

```python
w, Hw = signal.freqz(b, a, worN=2001)
Aw = np.abs(Hw)
Fw = np.angle(Hw)
i_max, i_min = Aw.argmax(), Aw.argmin()
print(f"A(0) = {Aw[0]:.4f},  A(π) = {Aw[-1]:.4f}")
print(f"максимум A = {Aw[i_max]:.3f} при ω = {w[i_max]/np.pi:.3f}π")
print(f"минимум  A = {Aw[i_min]:.4f} при ω = {w[i_min]/np.pi:.3f}π")
print(f"A > 1 при ω > {w[np.argmax(Aw > 1)]/np.pi:.3f}π")
print(f"ФЧХ: от {Fw.min():.3f} до {Fw.max():.3f} рад")

fig, axs = plt.subplots(2, 1, figsize=(11, 7), sharex=True)
axs[0].plot(w / np.pi, Aw, lw=2)
axs[0].set_ylabel("A(ω)"); axs[0].set_title("АЧХ")
axs[1].plot(w / np.pi, Fw, lw=2, color="tab:orange")
axs[1].set_yticks([0, np.pi / 4, np.pi / 2, 3 * np.pi / 4], ["0", "π/4", "π/2", "3π/4"])
axs[1].set_ylabel("φ(ω), рад"); axs[1].set_title("ФЧХ")
axs[1].set_xlabel("ω / π")
plt.tight_layout(); plt.show()
```

**а) Границы основной полосы:** $\hat\omega \in [0;\pi]$ (в двустороннем виде $[-\pi;\pi]$). Частотная характеристика периодична с периодом $2\pi$, АЧХ чётная, ФЧХ нечётная, поэтому за пределами $[0;\pi]$ новой информации нет.

**б) Карта нулей и полюсов и вид АЧХ.** При росте $\hat\omega$ от 0 до $\pi$ точка $e^{j\hat\omega}$ обходит верхнюю половину единичной окружности от $z = 1$ до $z = -1$.
- Около $\hat\omega \approx 0.27\pi$ она проходит ближе всего к нулю $0.894e^{j0.275\pi}$. Здесь АЧХ имеет минимум $0.049$ (до нуля не падает, потому что нуль не лежит на окружности).
- Около $\hat\omega \approx 0.72\pi$ она ближе всего к полюсу $0.837e^{j0.711\pi}$. Здесь АЧХ имеет резонансный максимум $6.81$.
- При $\hat\omega = 0$ ($z = 1$) точка близка к нулям и далека от полюсов, $A(0) = 0.164$. При $\hat\omega = \pi$ ($z = -1$) — наоборот, $A(\pi) = 3.08$.

**в) Скачок ФЧХ на $\pi$** бывает на частотах, где АЧХ обращается в ноль, то есть если нуль лежит на единичной окружности: при переходе через такую точку $H(e^{j\hat\omega})$ меняет знак. У меня $|z_{1,2}| = 0.894 < 1$, АЧХ нигде не равна нулю (минимум $0.049$), поэтому **скачка на $\pi$ нет**. Скачков на $2\pi$ из-за приведения фазы к $(-\pi;\pi]$ тоже нет: ФЧХ остаётся в пределах $[0;\ 2.70]$ рад.

**г) Подавленные составляющие.** АЧХ меньше 1 при $\hat\omega < 0.53\pi$ и больше 1 выше этой частоты, максимум приходится на $0.72\pi$. Поэтому **преимущественно подавлены низкие частоты**, а высокие усилены. Это видно по реакции в п. 1.2–1.3: постоянная составляющая импульса почти не проходит ($A(0) = 0.16$), а фронты, содержащие высокочастотные составляющие, дают сильные колебания.

Переход к абсолютной частоте: $f = \dfrac{\hat\omega}{2\pi}f_д$, $f_д = 10\,000$ Гц. Значения АЧХ и ФЧХ при этом не меняются, меняется только ось.

### Часть 2

#### 2.1. Последовательность на периоде $N$

$$x(nT) = A_1\cos(2\pi f_1nT + \pi/4) + A_2\cos(2\pi f_2nT + \pi/8).$$

```python
N = 64
fs2 = 2000 * (VAR % 5 + 1)
T = 1 / fs2
A1 = 1 + 0.01 * VAR
A2 = 2 * A1
f1 = fs2 / 8
f2 = 1.5 * f1
df = fs2 / N                          # шаг по частоте

print(f"N = {N}, fд = {fs2} Гц, T = {T*1e3} мс, Δf = fд/N = {df} Гц")
print(f"A1 = {A1}, A2 = {A2}, f1 = {f1} Гц, f2 = {f2} Гц")

def x_analog(t):
    return A1 * np.cos(2 * np.pi * f1 * t + np.pi / 4) + A2 * np.cos(2 * np.pi * f2 * t + np.pi / 8)

n = np.arange(N)
xn = x_analog(n * T)

fig, axs = plt.subplots(2, 1, figsize=(11, 7))
stem(axs[0], n * T * 1e3, xn)
axs[0].set_xlabel("nT, мс"); axs[0].set_title("x(nT) — шкала дискретного времени")
stem(axs[1], n, xn)
axs[1].set_xlabel("n"); axs[1].set_title("x(n) — шкала дискретного нормированного времени")
plt.tight_layout(); plt.show()
```

Оба графика содержат одни и те же 64 отсчёта и отличаются только подписью оси: $nT$ от 0 до $(N-1)T = 31.5$ мс либо $n$ от 0 до 63. На период первой гармоники приходится $f_д/f_1 = 8$ отсчётов, второй — $f_д/f_2 = 16/3$. Сумма повторяется через 16 отсчётов, поэтому на периоде $N = 64$ видны 4 одинаковых фрагмента.

#### 2.2. ОДПФ от ДПФ

```python
X = np.fft.fft(xn)          # ДПФ
x_back = np.fft.ifft(X)     # ОДПФ

print(f"max |Im| после ОДПФ: {np.abs(x_back.imag).max():.1e}")
print(f"max |ОДПФ(ДПФ(x)) - x|: {np.abs(x_back.real - xn).max():.1e}")

fig, ax = plt.subplots()
stem(ax, n, x_back.real, color="tab:green")
ax.set_xlabel("n"); ax.set_title("Последовательность, полученная ОДПФ от ДПФ")
plt.show()
```

ОДПФ от ДПФ возвращает исходную последовательность: график совпадает с графиком $x(n)$ из п. 2.1. Отличие и мнимая часть — на уровне ошибок округления. Прямое и обратное ДПФ взаимно обратны, поэтому при переходе в частотную область и обратно информация не теряется.

#### 2.3. Амплитудный и фазовый спектры

Амплитудный спектр — $|X(k)|$, фазовый — $\arg X(k)$. Там, где $|X(k)|$ на уровне ошибок округления, фаза случайна и не имеет смысла, поэтому в этих точках принимаю её равной нулю.

```python
k = np.arange(N)
fk = k * df                              # абсолютные частоты, Гц
Xamp = np.abs(X)
Xph = np.angle(X)
Xph[Xamp < 1e-9 * Xamp.max()] = 0.0

for kk in np.flatnonzero(Xamp > 1e-9 * Xamp.max()):
    print(f"k = {kk:2d}  f = {fk[kk]:7.2f} Гц  |X| = {Xamp[kk]:6.2f}  arg X = {Xph[kk]/np.pi:+.4f}π")
print(f"A1·N/2 = {A1*N/2:.2f},  A2·N/2 = {A2*N/2:.2f}")

fig, axs = plt.subplots(2, 2, figsize=(13, 7))
stem(axs[0, 0], k, Xamp);        axs[0, 0].set_title("Амплитудный спектр |X(k)|");   axs[0, 0].set_xlabel("k")
stem(axs[0, 1], fk, Xamp);       axs[0, 1].set_title("Амплитудный спектр |X(f)|");   axs[0, 1].set_xlabel("f, Гц")
stem(axs[1, 0], k, Xph, "tab:orange");  axs[1, 0].set_title("Фазовый спектр arg X(k)"); axs[1, 0].set_xlabel("k")
stem(axs[1, 1], fk, Xph, "tab:orange"); axs[1, 1].set_title("Фазовый спектр arg X(f)"); axs[1, 1].set_xlabel("f, Гц")
for ax in axs[1]:
    ax.set_yticks([-np.pi / 4, -np.pi / 8, 0, np.pi / 8, np.pi / 4], ["-π/4", "-π/8", "0", "π/8", "π/4"])
plt.tight_layout(); plt.show()
```

**а) Связь дискретного нормированного времени с дискретным:** $n = nT/T$. Нормированное время — это номер отсчёта, безразмерная величина. Дискретное время получается умножением на период дискретизации $T = 0.5$ мс.

**б) Связь частоты $f$ с дискретной нормированной частотой $k$:** $f_k = k\,\Delta f = k\,\dfrac{f_д}{N}$, $\Delta f = 2000/64 = 31.25$ Гц. Так, $k = 8 \to 250$ Гц, $k = 12 \to 375$ Гц, $k = N/2 = 32 \to f_д/2 = 1000$ Гц, $k = N \to f_д$. Через нормированную угловую частоту: $\hat\omega_k = 2\pi k/N$.

**в) Вид спектров.**
- Амплитудный спектр линейчатый и содержит 4 ненулевых отсчёта. Линии $k = 8$ и $k = 12$ отвечают гармоникам $f_1 = 250$ Гц и $f_2 = 375$ Гц, их высоты $A_1N/2 = 35.2$ и $A_2N/2 = 70.4$. Линии $k = 56$ и $k = 52$ — зеркальные отсчёты $N - k$. Сигнал вещественный, поэтому $|X(k)|$ симметричен относительно $k = N/2$; эти линии соответствуют отрицательным частотам $-250$ и $-375$ Гц.
- Обе частоты кратны $\Delta f$: на периоде $N$ укладывается целое число периодов каждой гармоники (8 и 12). Поэтому растекания спектра нет, остальные отсчёты равны нулю.
- Фазовый спектр антисимметричен. В точках $k = 8$ и $k = 12$ он равен начальным фазам гармоник $\pi/4$ и $\pi/8$, в $k = 56$ и $k = 52$ — тем же значениям с обратным знаком ($X(N-k) = X^*(k)$). В остальных точках амплитуда нулевая и фаза не определена.

#### 2.4. ДПФ конечной последовательности

Конечная последовательность — те же 64 отсчёта $x(n)$, но вне интервала $0 \le n \le N-1$ она равна нулю, а не продолжается периодически.

```python
x_fin = xn.copy()
X_fin = np.fft.fft(x_fin)        # ДПФ конечной последовательности

fig, ax = plt.subplots(figsize=(11, 4.5))
stem(ax, k, Xamp, label="амплитудный спектр периодической последовательности")
ax.plot(k, np.abs(X_fin), "rx", ms=8, label="модуль ДПФ конечной последовательности")
ax.set_xlabel("k"); ax.set_title("Модуль ДПФ конечной и амплитудный спектр периодической последовательности")
ax.legend(loc="upper center")
plt.show()
```

Модуль ДПФ конечной последовательности **совпадает** с амплитудным спектром периодической последовательности: крестики лежат на вершинах линий. Формула ДПФ в обоих случаях одна и та же и использует одни и те же $N$ отсчётов одного периода. Разница в смысле результата:
- для периодической последовательности $|X(k)|$ — это весь её спектр: он дискретный, других частот, кроме $k\Delta f$, у неё нет;
- для конечной последовательности спектр $X(e^{j\hat\omega})$ непрерывный, и ДПФ даёт лишь $N$ его равноотстоящих отсчётов на частотах $\hat\omega_k = 2\pi k/N$.

То есть модуль ДПФ конечной последовательности — это отсчёты её спектральной плотности, и они равны амплитудному спектру периодической последовательности, полученной повторением этого отрезка с периодом $N$.

#### 2.5. Восстановление аналогового сигнала по отсчётам ДПФ

$$x(t) = \frac{1}{N}\sum_{k=-N/2}^{N/2-1}X_а(k)\exp\Big(j\frac{2\pi}{NT}kt\Big),\qquad X_а(k) = \begin{cases}X(N+k), & -N/2 \le k \le -1,\\ X(k), & 0 \le k \le N/2-1,\end{cases}$$
$t \in [0;\ (N-1)T]$, шаг $\Delta t = 0.25T$.

```python
ka = np.arange(-N // 2, N // 2)               # k = -N/2 ... N/2-1
Xa = np.where(ka < 0, X[(N + ka) % N], X[ka % N])

dt = 0.25 * T
t = np.arange(0, (N - 1) * T + dt / 2, dt)    # 0 ... (N-1)T включительно
x_rec = np.zeros(len(t), dtype=complex)
for kk, Xk in zip(ka, Xa):
    x_rec += Xk * np.exp(1j * 2 * np.pi * kk * t / (N * T))
x_rec = x_rec.real / N

x_src = x_analog(t)                           # исходный аналоговый сигнал x(t) = x(nT)|nT=t
fa = ka / (N * T)                             # частоты, Гц
amp_a = np.abs(Xa) / N                        # амплитудный спектр восстановленного сигнала

print(f"точек по времени: {len(t)}, шаг Δt = {dt*1e3} мс")
print(f"max |x_восст(t) - x_исх(t)| = {np.abs(x_rec - x_src).max():.1e}")

fig, axs = plt.subplots(3, 2, figsize=(13, 10))
stem(axs[0, 0], n, xn);                 axs[0, 0].set_title("а) периодическая последовательность x(n)"); axs[0, 0].set_xlabel("n")
stem(axs[0, 1], k, Xamp);               axs[0, 1].set_title("а) модуль ДПФ |X(k)|");                     axs[0, 1].set_xlabel("k")
axs[1, 0].plot(t * 1e3, x_rec, color="tab:green")
axs[1, 0].set_title("б) восстановленный аналоговый сигнал x(t)"); axs[1, 0].set_xlabel("t, мс")
stem(axs[1, 1], fa, amp_a, "tab:green"); axs[1, 1].set_title("б) амплитудный спектр восстановленного сигнала"); axs[1, 1].set_xlabel("f, Гц")
axs[2, 0].plot(t * 1e3, x_src, color="tab:red")
axs[2, 0].set_title("в) исходный аналоговый сигнал x(t) = x(nT)|nT=t"); axs[2, 0].set_xlabel("t, мс")
axs[2, 1].plot(t * 1e3, x_src, color="tab:red", lw=3, alpha=0.4, label="исходный")
axs[2, 1].plot(t * 1e3, x_rec, "k--", lw=1.2, label="восстановленный")
axs[2, 1].set_xlim(0, 8); axs[2, 1].set_xlabel("t, мс")
axs[2, 1].set_title("сравнение сигналов на одном периоде (8 мс)"); axs[2, 1].legend(loc="lower right")
plt.tight_layout(); plt.show()
```

#### 2.6. Анализ результатов восстановления

**а) Связь модуля ДПФ со спектром аналогового сигнала.** Восстановленный сигнал — сумма комплексных экспонент с частотами $f = k/(NT) = k\,\Delta f$, $k = -N/2\ldots N/2-1$, и комплексными амплитудами $X_а(k)/N$. Значит, его амплитудный спектр — это модуль ДПФ, делённый на $N$ и переставленный так, что отсчёты $k \ge N/2$ переходят в отрицательные частоты $(k - N)\Delta f$. Получается двусторонний спектр в основной полосе $[-f_д/2;\ f_д/2)$ с четырьмя линиями: $\pm 250$ Гц высотой $A_1/2 = 0.55$ и $\pm 375$ Гц высотой $A_2/2 = 1.1$. Каждый косинус $A\cos(2\pi ft + \varphi)$ — это пара экспонент на частотах $\pm f$ с амплитудами $A/2$, так что спектр совпадает со спектром исходного аналогового сигнала.

**б) Сравнение сигналов.** Восстановленный и исходный сигналы визуально совпадают, на графике сравнения кривые неразличимы. Разность — на уровне ошибок округления ($\sim10^{-14}$). Совпадение есть и в промежуточных точках между отсчётами, а не только при $t = nT$. Восстановление точное по двум причинам:
1. выполнено условие теоремы Котельникова: $f_{max} = f_2 = 375$ Гц $< f_д/2 = 1000$ Гц;
2. частоты обеих гармоник кратны $\Delta f$, поэтому формула восстановления содержит ровно те экспоненты, из которых состоит сигнал.

#### 2.7. Равенство Парсеваля

```python
E_t = np.sum(np.abs(xn) ** 2)
E_f = np.sum(np.abs(X) ** 2) / N

print(f"Σ|x(n)|²        = {E_t:.6f}")
print(f"(1/N)·Σ|X(k)|²  = {E_f:.6f}")
```

Обе части равенства равны $193.6$, **равенство Парсеваля выполняется**.

**Смысл:** энергия последовательности на периоде одинакова, считать ли её по отсчётам во временной области или по отсчётам ДПФ в частотной. ДПФ энергию не теряет и не добавляет. Множитель $1/N$ связан только с нормировкой прямого ДПФ. Здесь вся энергия сосредоточена в четырёх спектральных линиях: $\frac{1}{N}(2\cdot 35.2^2 + 2\cdot 70.4^2) = 193.6$.
