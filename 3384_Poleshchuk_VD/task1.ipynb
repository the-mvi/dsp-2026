## Основные теоретические положения

### Линейные дискретные системы

Линейная дискретная система преобразует входную последовательность $x(n)$ в выходную $y(n)$. В рекурсивных системах выход зависит не только от текущих и предыдущих значений входного сигнала, но и от предыдущих значений самого выхода.

Для линейного рекурсивного звена второго порядка передаточная функция имеет вид:

$$
H(z)=\frac{Y(z)}{X(z)}
=\frac{b_0+b_1z^{-1}+b_2z^{-2}}
{1+a_1z^{-1}+a_2z^{-2}}.
$$

Ей соответствует разностное уравнение:

$$
y(n)=b_0x(n)+b_1x(n-1)+b_2x(n-2)
-a_1y(n-1)-a_2y(n-2).
$$

Передаточная функция описывает связь между входом и выходом системы в области $z$-преобразования, а разностное уравнение позволяет непосредственно вычислять выходные отсчёты.

### Импульсная характеристика и свёртка

Импульсная характеристика $h(n)$ представляет собой реакцию системы на единичный импульс при нулевых начальных условиях:

$$
h(n)=Z^{-1}\{H(z)\}.
$$

Для рекурсивного звена импульсная характеристика, как правило, имеет бесконечную длительность. При этом для устойчивой системы её значения постепенно затухают.

Выход линейной стационарной системы можно определить через свёртку входной последовательности с импульсной характеристикой:

$$
y(n)=\sum_{m=-\infty}^{+\infty}x(m)h(n-m).
$$

Если свёртываются две конечные последовательности длиной $N_1$ и $N_2$, длина результата составляет:

$$
L=N_1+N_2-1.
$$

Таким образом, реакцию системы можно получить двумя способами: непосредственно из разностного уравнения или с помощью свёртки. При использовании конечного фрагмента импульсной характеристики свёртка приближает результат, полученный для системы с полной импульсной характеристикой.

### Нули, полюсы и устойчивость системы

Нулями передаточной функции называются корни её числителя, а полюсами — корни знаменателя. Для звена второго порядка передаточную функцию можно представить в факторизованной форме:

$$
H(z)=K\frac{(1-z_1z^{-1})(1-z_2z^{-1})}
{(1-p_1z^{-1})(1-p_2z^{-1})},
$$

где $z_1,z_2$ — нули, $p_1,p_2$ — полюсы, $K$ — коэффициент передачи.

Расположение нулей и полюсов на комплексной плоскости характеризует частотные свойства и устойчивость системы.

Для причинной рекурсивной системы условием устойчивости является расположение всех полюсов строго внутри единичной окружности:

$$
|p_i|<1.
$$

Комплексно-сопряжённые полюсы соответствуют колебательной составляющей импульсной характеристики. Если их модули меньше единицы, амплитуда этих колебаний постепенно уменьшается.

### Частотные характеристики системы

Частотная характеристика получается из передаточной функции заменой переменной $z$ на $e^{j\omega}$:

$$
H(e^{j\omega})=
\frac{b_0+b_1e^{-j\omega}+b_2e^{-j2\omega}}
{1+a_1e^{-j\omega}+a_2e^{-j2\omega}}.
$$

Она показывает, как система изменяет амплитуду и фазу гармонических составляющих входного сигнала.

Амплитудно-частотная характеристика (АЧХ) определяется модулем частотной характеристики:

$$
A(\omega)=|H(e^{j\omega})|.
$$

Фазочастотная характеристика (ФЧХ) определяется её аргументом:

$$
\varphi(\omega)=\arg H(e^{j\omega}).
$$

АЧХ характеризует усиление или ослабление сигнала на различных частотах, а ФЧХ — фазовый сдвиг, вносимый системой.

Связь угловой частоты $\omega$, измеряемой в радианах на отсчёт, с обычной частотой $f$ задаётся выражением:

$$
f=\frac{\omega f_d}{2\pi},
$$

где $f_d$ — частота дискретизации.

Нормированная угловая частота относительно $\pi$ определяется как:

$$
\hat{\omega}=\frac{\omega}{\pi}.
$$

Частота Найквиста равна половине частоты дискретизации:

$$
f_{\text{Н}}=\frac{f_d}{2}.
$$

### Дискретизация аналогового сигнала

Дискретизация заключается в преобразовании непрерывного сигнала $x(t)$ в последовательность отдельных отсчётов:

$$
x(n)=x(nT),
$$

где $T$ — период дискретизации, связанный с частотой дискретизации соотношением:

$$
T=\frac{1}{f_d}.
$$

Момент времени, соответствующий отсчёту с номером $n$, определяется выражением:

$$
t=nT.
$$

Согласно теореме Котельникова, для восстановления полосоограниченного аналогового сигнала частота дискретизации должна превышать удвоенную максимальную частоту сигнала:

$$
f_d>2f_{\max}.
$$

Это условие позволяет избежать наложения спектров при дискретизации и обеспечивает возможность восстановления исходного сигнала в идеальных условиях.

### Дискретное преобразование Фурье

Дискретное преобразование Фурье (ДПФ) позволяет представить конечную последовательность в виде набора комплексных спектральных составляющих:

$$
X(k)=\sum_{n=0}^{N-1}x(n)e^{-j2\pi kn/N},
$$

где $N$ — длина последовательности, а $k$ — номер частотного отсчёта.

Обратное дискретное преобразование Фурье восстанавливает исходную последовательность:

$$
x(n)=\frac{1}{N}\sum_{k=0}^{N-1}X(k)e^{j2\pi kn/N}.
$$

Модуль и аргумент комплексных коэффициентов ДПФ определяют соответственно амплитудный и фазовый спектры:

$$
A(k)=\frac{|X(k)|}{N},
\qquad
\varphi(k)=\arg X(k).
$$

### Спектр периодической последовательности

Конечную последовательность из $N$ отсчётов можно рассматривать как один период бесконечной периодической последовательности.

Такая последовательность представляется комплексным рядом Фурье:

$$
x_p(n)=\sum_{k=0}^{N-1}C(k)e^{j2\pi kn/N},
$$

где $C(k)$ — комплексные коэффициенты ряда Фурье.

Они связаны с коэффициентами ДПФ следующим образом:

$$
C(k)=\frac{X(k)}{N}.
$$

Следовательно, модуль коэффициентов равен:

$$
|C(k)|=\frac{|X(k)|}{N}.
$$

### Восстановление аналогового сигнала

Восстановление периодического аналогового сигнала по спектральным коэффициентам ДПФ выполняется с помощью обратного преобразования Фурье в непрерывном времени:

$$
x(t)=\frac{1}{N}
\sum_{k=-N/2}^{N/2-1}
X_a(k)e^{j\frac{2\pi}{NT}kt},
$$

где $X_a(k)$ — коэффициенты ДПФ, переупорядоченные от отрицательных частот к положительным.

Частоты восстановленных гармоник определяются выражением:

$$
f_k=\frac{k}{NT}.
$$

Период, задаваемый частотной сеткой восстановленного ряда, равен:

$$
T_p=NT.
$$



### Равенство Парсеваля

Равенство Парсеваля устанавливает связь между энергией сигнала во временной области и его спектральным представлением:

$$
\sum_{n=0}^{N-1}|x(n)|^2
=
\frac{1}{N}\sum_{k=0}^{N-1}|X(k)|^2.
$$

Оно показывает, что при переходе от временного представления к частотному энергия последовательности сохраняется с учётом нормировки ДПФ.

где $X(z)$ и $Y(z)$ — Z-преобразования входного и выходного сигналов соответственно.

Умножим обе части равенства на знаменатель передаточной функции:

Y(z)(a_0+a_1z^{-1}+a_2z^{-2})

Применим обратное Z-преобразование, используя свойство временного сдвига:

$$
z^{-1}X(z)\leftrightarrow x[n-1],

$$
z^{-1}Y(z)\leftrightarrow y[n-1],

y[n]=

Таким образом, текущий отсчёт выходного сигнала определяется текущими и двумя предыдущими отсчётами входного сигнала, а также двумя предыдущими отсчётами выходного сигнала.

#### Импульсная характеристика

Импульсная характеристика $h[n]$ — это отклик на $x[n]=\delta[n]$ при нулевых начальных условиях ($y[-1]=y[-2]=0$)

```python
def lfilter_df2(b, a, x):
    b = np.asarray(b, float); a = np.asarray(a, float)
    x = np.asarray(x, float)
    y = np.zeros(len(x))
    for n in range(len(x)):
        acc = b[0] * x[n]
        if n - 1 >= 0: acc += b[1] * x[n-1]
        if n - 2 >= 0: acc += b[2] * x[n-2]
        if n - 1 >= 0: acc -= a[1] * y[n-1]
        if n - 2 >= 0: acc -= a[2] * y[n-2]
        y[n] = acc / a[0]
    return y

delta = np.zeros(N1); delta[0] = 1.0
h = lfilter_df2(b, a, delta)

print("Импульсная характеристика:")
for i in range(len(h)):
    print(f"h[{i}] = {h[i]:.5f}")
```

```python
n = np.arange(N1)
plt.figure(figsize=(9, 4))
plt.stem(n, h)
plt.xlabel("n, отсчёты")
plt.ylabel("h[n]")
plt.title(f"Импульсная характеристика, первые N1 = {N1} отсчётов")
plt.grid(True)
plt.show()
```

#### Аналитическая формула импульсной характеристики

Коэффициенты передаточной функции равны:

$$
b_0=0.8,\quad b_1=0.99936,\quad b_2=0.64,
$$

$$
a_0=1,\quad a_1=-1.1528,\quad a_2=0.73.
$$

Передаточная функция рекурсивного звена второго порядка имеет вид:

$$
H(z)=\frac{0.8+0.99936z^{-1}+0.64z^{-2}}
{1-1.1528z^{-1}+0.73z^{-2}}.
$$

Импульсная характеристика определяется как обратное Z-преобразование передаточной функции:

$$
h(n)=Z^{-1}\{H(z)\}.
$$

Для нахождения аналитического выражения необходимо определить полюсы системы. Они находятся из уравнения знаменателя:

$$
z^2-1.1528z+0.73=0.
$$

Корни уравнения:

$$
p_{1,2}=0.5764\pm j0.6307.
$$

Представим полюсы в показательной форме:

$$
p_{1,2}=\rho e^{\pm j\theta},
$$

где модуль полюсов:

$$
\rho=\sqrt{0.73}\approx 0.8544,
$$

а аргумент:

$$
\theta=\arccos\left(\frac{1.1528}{2\sqrt{0.73}}\right)
\approx 0.83034.
$$

Поскольку полюсы являются комплексно-сопряжёнными, импульсная характеристика имеет вид:

$$
h(n)=\rho^n\left(C\cos(n\theta)+D\sin(n\theta)\right),
\quad n\geq 0.
$$

Для определения коэффициентов $C$ и $D$ воспользуемся разностным уравнением системы с учётом нулевых начальных условий:

$$
h(n)=1.1528h(n-1)-0.73h(n-2)
+0.8\delta(n)+0.99936\delta(n-1)+0.64\delta(n-2).
$$

При $n=0$ получаем:

$$
h(0)=0.8.
$$

При $n=1$:

$$
h(1)=1.1528\cdot h(0)+0.99936=1.9216.
$$

Теперь подставим полученные начальные значения в общее аналитическое выражение импульсной характеристики.

Для нахождения коэффициента $C$ положим $n=0$:

$$
h(0)=\rho^0(C\cos(0)+D\sin(0)).
$$

Учитывая, что $\rho^0=1$, $\cos(0)=1$ и $\sin(0)=0$, получаем:

$$
h(0)=C.
$$

Следовательно:

$$
{C=0.8}.
$$

Для нахождения коэффициента $D$ подставим $n=1$:

$$
h(1)=\rho(C\cos\theta+D\sin\theta).
$$

Выразим $D$ из полученного равенства:

$$
D=\frac{h(1)/\rho-C\cos\theta}{\sin\theta}.
$$

Подставляя известные значения, получаем:

$$
D=\frac{1.9216/0.8544-0.8\cos(0.83034)}
{\sin(0.83034)}
\approx 2.31571.
$$

Таким образом, аналитическая формула импульсной характеристики принимает вид:

$$
{
h(n)=0.8544^n
\left(
0.8\cos(0.83034n)+
2.31571\sin(0.83034n)
\right),\quad n\geq 0.
}
$$

При $n<0$ импульсная характеристика равна нулю:

$$
h(n)=0,\quad n<0.
$$

#### Связь алгебраической и показательной форм

Корень $q=x+jy$ записывается в показательной форме $q=re^{j\alpha}$, где $r=|q|=\sqrt{x^2+y^2}$, $\alpha=\arg q$ (с учётом квадранта), и обратно $x=r\cos\alpha$, $y=r\sin\alpha$ (формула Эйлера).

Коэффициенты $b_i,a_k$ вещественны, поэтому нули и полюса образуют комплексно-сопряжённые пары: у пары одинаковые модули и противоположные по знаку аргументы, $q_{1,2}=re^{\pm j\alpha}$. Для такой пары $q_1q_2=r^2$, поэтому модуль равен корню из свободного члена (теорема Виета):

$$|z_0|=\sqrt{b_2/b_0}=\sqrt{0.8}\approx0.8944,\qquad |p|=\sqrt{a_2/a_0}=\sqrt{0.73}\approx0.8544.$$

Аргумент находится из коэффициента при первой степени: $q_1+q_2=2r\cos\alpha$, то есть $\cos\theta=-\dfrac{b_1}{2b_0|z_0|}$ для нулей и $\cos\varphi=-\dfrac{a_1}{2|p|}$ для полюсов.

| | алгебраическая форма | показательная форма |
|---|---|---|
| нули | $-0.6246\pm0.6402\,j$ | $0.8944\,e^{\pm j\,0.7461\pi}$ |

Каждая гармоника даёт две линии (при $k$ и $N-k$), поскольку косинус содержит две комплексные экспоненты по $A_i/2$. Спектр симметричен: $|X(k)|=|X(N-k)|$.
- Фазовый спектр: в линиях фаза равна начальным фазам гармоник, $+\pi/4$ и $+\pi/8$ при положительных частотах, и $-\pi/4$, $-\pi/8$ при $N-k$ (нечётная симметрия, так как $x(n)$ вещественна). В остальных точках $|X(k)|=0$, фаза не определена (на графике там пусто).

```python
X = np.fft.fft(x)                     
k = np.arange(N)
w_hat = 2*np.pi*k/N                   

mod_dft = np.abs(X)                  
c = X / N                           
amp_per = np.abs(c)                   

k1 = int(round(f1*N/fd)); k2 = int(round(f2*N/fd))
c_th = np.zeros(N, dtype=complex)
c_th[k1]     = A1/2*np.exp(1j*np.pi/4);  c_th[N-k1] = np.conj(c_th[k1])
c_th[k2]     = A2/2*np.exp(1j*np.pi/8);  c_th[N-k2] = np.conj(c_th[k2])
print(f"k1 = {k1}, k2 = {k2}")
print("max |X/N - c_теор| =", np.max(np.abs(c - c_th)))
print(f"|X(k1)| = {mod_dft[k1]:.2f},  |X(k1)|/N = {amp_per[k1]:.4f} = A1/2 = {A1/2:.4f}")
print(f"|X(k2)| = {mod_dft[k2]:.2f},  |X(k2)|/N = {amp_per[k2]:.4f} = A2/2 = {A2/2:.4f}")

fig, ax = plt.subplots(2, 1, figsize=(11, 7), sharex=True)

ax[0].stem(w_hat/np.pi, mod_dft, basefmt=" ")
ax[0].set_title("Модуль ДПФ конечной последовательности |X(k)|")
ax[0].set_ylabel("|X(k)|"); ax[0].grid(True)

ax[1].stem(w_hat/np.pi, amp_per, basefmt=" ")
ax[1].set_title("Амплитудный спектр периодической последовательности |c_k| = |X(k)|/N")
ax[1].set_xlabel(r"$\hat{\omega}/\pi$"); ax[1].set_ylabel(r"$|c_k|$"); ax[1].grid(True)

plt.tight_layout(); plt.show()
```

ДПФ последовательности $x(n)$, $n=0,\dots,N-1$, $N=64$:

$$X(k)=\sum_{n=0}^{N-1}x(n)e^{-j\frac{2\pi}{N}kn},\qquad \hat\omega_k=\frac{2\pi k}{N}$$

Периодическая последовательность $\tilde x(n)$ с периодом $N$, составленная из повторений $x(n)$, раскладывается в ряд Фурье:

$$\tilde x(n)=\sum_{k=0}^{N-1}c_k\,e^{j\frac{2\pi}{N}kn},\qquad c_k=\frac1N\sum_{n=0}^{N-1}\tilde x(n)e^{-j\frac{2\pi}{N}kn}$$

Сумма для $c_k$ берётся по одному периоду, где $\tilde x(n)=x(n)$, поэтому

$${c_k=\frac{X(k)}{N}}\qquad\Longrightarrow\qquad |c_k|=\frac{|X(k)|}{N}$$

То есть модуль ДПФ конечной последовательности отличается от амплитудного спектра периодической последовательности только масштабом $N$

$$ \sum_{n=0}^{N-1}|x(n)|^2=\frac1N\sum_{k=0}^{N-1}|X(k)|^2 $$

```python
energy_time = np.sum(np.abs(x)**2)
energy_freq = np.sum(np.abs(X)**2) / N
print(f"Энергия во временной области : sum|x(n)|^2       = {energy_time:.6f}")
print(f"Энергия в частотной области  : (1/N) sum|X(k)|^2 = {energy_freq:.6f}")
```

Смысл равенства Парсеваля - это закон сохранения энергии при переходе из временной области в частотную: энергия последовательности за период одинакова, считаем ли мы её по отсчётам $x(n)$ или по спектру $|X(k)|^2$. ДПФ только перераспределяет энергию между представлениями и ничего не теряет, а $|X(k)|^2$ показывает, как энергия распределена по частотам. Множитель $1/N$ связан с нормировкой ДПФ.
