## Основные теоретические положения

**Дискретный сигнал** $x(n)$ — последовательность отсчётов с целочисленным индексом $n$. Интервал между соседними отсчётами равен $T=1/f_d$, где $f_d$ — частота дискретизации.

**Линейная стационарная дискретная система (ЛДС)** удовлетворяет принципу суперпозиции (аддитивности и однородности), а сдвиг воздействия приводит к такому же сдвигу реакции.

**Единичный дискретный импульс**:

$$\delta(n)=\begin{cases}1,&n=0,\\0,&n\neq0.\end{cases}$$

**Импульсная характеристика** $h(n)$ — реакция ЛДС на $\delta(n)$ при нулевых начальных условиях. Для причинной системы $h(n)=0$ при $n<0$.

**Разностное уравнение рекурсивного звена второго порядка** при $a_0=1$:

$$y(n)=\sum_{i=0}^{2}b_i x(n-i)-a_1y(n-1)-a_2y(n-2).$$

**Передаточная функция** — отношение Z-изображений реакции и воздействия при нулевых начальных условиях:

$$H(z)=\frac{Y(z)}{X(z)}=\frac{b_0+b_1z^{-1}+b_2z^{-2}}{1+a_1z^{-1}+a_2z^{-2}}.$$

**Свёртка** связывает воздействие и реакцию ЛДС:

$$y(n)=(x*h)(n)=\sum_{m=0}^{\infty}h(m)x(n-m).$$

Для устойчивой причинной ЛДС импульсная характеристика абсолютно суммируема: $\sum_{n=0}^{\infty}|h(n)|<\infty$. Эквивалентное условие для рационального рекурсивного звена - все полюсы находятся строго внутри единичной окружности.

**Дискретное преобразование Фурье (ДПФ)** представляет последовательность из $N$ отсчётов в виде $N$ комплексных частотных коэффициентов:

$$X(k)=\sum_{n=0}^{N-1}x(n)e^{-j2\pi kn/N},\qquad k=0,\ldots,N-1.$$

**Обратное дискретное преобразование Фурье (ОДПФ)** восстанавливает исходные отсчёты по коэффициентам ДПФ:

$$x(n)=\frac{1}{N}\sum_{k=0}^{N-1}X(k)e^{j2\pi kn/N},\qquad n=0,\ldots,N-1.$$

**Амплитудный спектр** показывает амплитуды частотных составляющих сигнала. Для двустороннего спектра периодической последовательности амплитуды вычисляются по формуле: $A(k)=|X(k)|/N$. При построении одностороннего спектра действительного сигнала амплитуды положительных частот удваиваются, постоянная составляющая не удваивается и при четном N последний отсчёт одностороннего спектра также не удваивается.

**Фазовый спектр** показывает начальные фазы частотных составляющих сигнала: $\varphi(k)=\arg X(k)$. Если амплитуда частотной составляющей равна нулю $|X(k)|=0$, то её фаза не определена.

**Частотные отсчёты ДПФ** определяют частоты, которым соответствуют значения спектра: 
$$f_k=\frac{kf_{\text{д}}}{N},
\qquad
\widehat{\omega}_k=\frac{2\pi k}{N},$$

где k — номер частотного отсчёта, N — длина последовательности,  $f_{\text{д}}$ — частота дискретизации. При чётном $N$ отсчёты с индексами $k>N/2$ соответствуют отрицательным частотам в двустороннем представлении спектра.

**Равенство Парсеваля для ДПФ** устанавливает связь между энергией конечной последовательности и её ДПФ:

$$\sum_{n=0}^{N-1}|x(n)|^2=\frac{1}{N}\sum_{k=0}^{N-1}|X(k)|^2.$$

### Часть 1. Линейные дискретные системы

#### 1. Вычисление импульсной характеристики

```python
# Вариант 3
variant = 3
b0 = 0.5 + 0.02 * variant
b1 = b0 * (-1)**(variant + 1) * (0.9822 + 0.0178 * variant)
b2 = b0 * (0.8 + 0.2 * (variant % 5))
a0 = 1.0
a1 = (-1)**variant * (0.7778 + 0.025 * variant)
a2 = 0.64 + 0.006 * variant
N1 = variant % 10 + 20
N2 = variant % 10 + 30
fd1 = 1000 * variant

b = np.array([b0, b1, b2], dtype=float)
a = np.array([a0, a1, a2], dtype=float)

print(f"Вариант: {variant}")
print(f"b = {b}")
print(f"a = {a}")
print(f"N1 = {N1}, N2 = {N2}, fд = {fd1} Гц")
```

При $N_{\mathrm{var}}=3$ получены:

$$b_0=0.56,\quad b_1=0.579936,\quad b_2=0.784,$$
$$a_0=1,\quad a_1=-0.8528,\quad a_2=0.658,\quad N_1=23.$$

Передаточная функция рекурсивного звена второго порядка:

$$H(z)=\frac{0.56+0.579936z^{-1}+0.784z^{-2}}{1-0.8528z^{-1}+0.658z^{-2}}.$$

Соответствующее разностное уравнение:

$$y(n)=0.56x(n)+0.579936x(n-1)+0.784x(n-2)+0.8528y(n-1)-0.658y(n-2).$$

Для определения импульсной характеристики подаётся единичный дискретный импульс $x(n)=\delta(n)$ при нулевых начальных условиях $h(-1)=h(-2)=0$. Тогда:

$$h(n)=0.56\delta(n)+0.579936\delta(n-1)+0.784\delta(n-2)+0.8528h(n-1)-0.658h(n-2).$$

Первые отсчёты:

$$h(0)=0.56,\qquad h(1)=0.579936+0.8528\cdot0.56=1.057504,$$

$$h(2)=0.784+0.8528\cdot1.057504-0.658\cdot0.56\approx1.317359.$$

```python
x_impulse = np.zeros(N1)
x_impulse[0] = 1.0

h = signal.lfilter(b, a, x_impulse)
n = np.arange(N1)

print("Первые пять отсчётов h(n):")
for i in range(5):
    print(f"h({i}) = {h[i]:.6f}")

fig, ax = plt.subplots(figsize=(10, 4))
ax.stem(n, h)
ax.set(title="Импульсная характеристика h(n)",
       xlabel="Номер отсчёта n", ylabel="h(n)")
ax.set_xticks(n)
ax.grid(alpha=0.35)
plt.tight_layout()
plt.show()
```

#### Аналитическое выражение импульсной характеристики

Для нахождения аналитического выражения используем обратное Z-преобразование передаточной функции:

$$h(n)=\mathcal{Z}^{-1}\{H(z)\},\qquad H(z)=\frac{0.56+0.579936z^{-1}+0.784z^{-2}}{1-0.8528z^{-1}+0.658z^{-2}}.$$

Передаточная функция заданного рекурсивного звена второго порядка представляется произведением:

$$
H(z)=\underbrace{(0.56+0.579936z^{-1}+0.784z^{-2})}_{B(z)}\,
\underbrace{\frac{1}{1-0.8528z^{-1}+0.658z^{-2}}}_{G(z)}.
$$

Сначала найдём импульсную характеристику вспомогательного звена $G(z)$ с единичным числителем. Определим корни знаменателя (полюсы):

$$z^2-0.8528z+0.658=0, \qquad p_{1,2}=0.4264\pm0.69006017j.$$

Полюсы представляются в показательной форме $p_{1,2}=re^{\pm j\varphi}$, где их модуль и аргумент равны:

$$r=\sqrt{0.658}\approx0.811172,\qquad \varphi=\operatorname{atan2}(0.69006017,\,0.4264)\approx1.017307\ \text{рад}.$$


Для рекурсивного звена второго порядка с комплексно-сопряжёнными полюсами аналитическое выражение импульсной характеристики при нулевых начальных условиях имеет вид:

$$
\begin{aligned}
h(n)={}&0.56\,r^n\frac{\sin((n+1)\varphi)}{\sin\varphi}\\
&+0.579936\,r^{n-1}\frac{\sin(n\varphi)}{\sin\varphi}\\
&+0.784\,r^{n-2}\frac{\sin((n-1)\varphi)}{\sin\varphi},
\end{aligned}
$$

где $r\approx0.811172$, $\varphi\approx1.017307$, первое слагаемое учитывается при $n\geq0$, второе — при $n\geq1$, третье — при $n\geq2$. При $n<0$ импульсная характеристика равна нулю.

Для упрощения записи аналитического выражения введена вспомогательная последовательность $g(n)$, соответствующая звену $G(z)$

$$
g(n)=\begin{cases}
0, & n<0,\\[3pt]
r^n\dfrac{\sin\big((n+1)\varphi\big)}{\sin\varphi}, & n\geq0.
\end{cases}
$$

Последовательность $g(n)$ относится к $G(z)$, а не ко всей передаточной функции $H(z)$. Умножение $G(z)$ на $B(z)$ соответствует сумме трёх сдвинутых последовательностей, поэтому аналитическая формула для заданной системы имеет вид:

$$h(n)=0.56\,g(n)+0.579936\,g(n-1)+0.784\,g(n-2)$$

При $n<0$ все слагаемые равны нулю. Проверка начальных отсчётов:

$$g(0)=1,\quad g(1)=0.8528,\quad g(2)=0.8528^2-0.658,$$

$$h(0)=0.56,\qquad h(1)=0.56\cdot0.8528+0.579936=1.057504,$$

$$h(2)=0.56\cdot g(2)+0.579936\cdot g(1)+0.784=1.3173594112.$$

```python
poles = np.roots(a)
p = poles[np.imag(poles) > 0][0]
r = np.abs(p)
phi = np.angle(p)

def g(k):
    if k < 0:
        return 0.0
    return r**k * np.sin((k + 1) * phi) / np.sin(phi)


h_analytic = np.array([
    b[0] * g(i) + b[1] * g(i - 1) + b[2] * g(i - 2)
    for i in n
])

max_error = np.max(np.abs(h - h_analytic))
print("Полюсы:", poles)
print(f"r = {r:.6f}")
print(f"phi = {phi:.8f} рад = {phi / np.pi:.8f}π")
print(f"h(0) = {h_analytic[0]:.9f}")
print(f"h(1) = {h_analytic[1]:.9f}")
print(f"h(2) = {h_analytic[2]:.9f}")
print(f"Максимальное расхождение = {max_error:.3e}")
assert np.allclose(h, h_analytic, rtol=1e-12, atol=1e-12)

fig, ax = plt.subplots(figsize=(10, 4))
ax.stem(n, h, linefmt="C0-", markerfmt="C0o",
        basefmt="k-", label="Численный расчёт")
ax.plot(n, h_analytic, "C1x", markersize=8,
        label="Аналитическая формула")
ax.set(title="Сравнение способов вычисления h(n)",
       xlabel="Номер отсчёта n", ylabel="h(n)")
ax.set_xticks(n)
ax.grid(alpha=0.35)
ax.legend()
plt.tight_layout()
plt.show()
```

**Анализ результатов** 

По заданной передаточной функции вычислены первые $N_1=23$ отсчёта импульсной характеристики рекурсивного звена второго порядка при нулевых начальных условиях. 

Полученная импульсная характеристика имеет затухающий колебательный характер. На графике видно, что значения отсчётов изменяют знак, а амплитуда колебаний постепенно уменьшается. Такой характер импульсной характеристики обусловлен наличием комплексно-сопряжённых полюсов, модуль которых меньше единицы: $r\approx0.811172<1$.

Фактическая длина импульсной характеристики бесконечна, поскольку при вычислении новых выходных отсчётов используются предыдущие значения выходного сигнала. Поэтому даже после прекращения входного воздействия система продолжает формировать ненулевые значения, которые постепенно уменьшаются. При этом $N_1=23$ - это только количество вычисленных и отображённых отсчётов.

#### 2. Вычисление реакции по формуле свёртки


Для расчёта реакции линейной стационарной дискретной системы используется линейная свёртка входного воздействия $x(n)$ и импульсной характеристики $h(n)$:

$$y_1(n)=(x*h)(n)=\sum_{m=0}^{\infty}h(m)x(n-m).$$

В численном расчёте используется фрагмент импульсной характеристики из пункта 1 длиной $N_1=23$ и воздействие длиной $N_2=33$. Входная последовательность представляет собой прямоугольный импульс:

$$
x(n)=\begin{cases}
1, & 0\le n < \operatorname{int}(N_2/2),\\
0, & \operatorname{int}(N_2/2)\le n\le N_2-1.
\end{cases}
$$

$\operatorname{int}$ - округление до ближайшего целого.

Для двух конечных последовательностей длины $N_1$ и $N_2$ полная линейная свёртка содержит

$$N_y=N_1+N_2-1=23+33-1=55$$

отсчётов.

```python
active_length = int(np.floor(N2 / 2 + 0.5))

x = np.zeros(N2)
x[:active_length] = 1.0
n_x = np.arange(N2)

y1_full = np.convolve(x, h, mode="full")
n_y1_full = np.arange(len(y1_full))

y1 = y1_full[:N2]

print(f"Длина входной последовательности: {len(x)} отсчёта")
print(f"Количество единичных отсчётов: {active_length}")
print(f"Число нулевых отсчётов: {N2 - active_length}")
print(f"Полная длина свёртки: {len(y1_full)} отсчётов")
print(f"Длина ограниченной реакции: {len(y1)} отсчёта")
assert len(y1_full) == N1 + N2 - 1
assert len(y1) == N2

fig, ax = plt.subplots(figsize=(10, 3.6))
ax.stem(n_x, x)
ax.set(title="Прямоугольный импульс x(n)",
       xlabel="Номер отсчёта n", ylabel="x(n)")
ax.set_xticks(np.arange(0, N2, 2))
ax.set_ylim(-0.12, 1.22)
ax.grid(alpha=0.35)
plt.tight_layout()
plt.show()

fig, axes = plt.subplots(2, 1, figsize=(11, 7))
axes[0].stem(n_y1_full, y1_full)
axes[0].axvline(N2 - 0.5, color="C3", linestyle="--",
                label="Граница первых 33 отсчётов")
axes[0].set(title="Полная свёртка конечных последовательностей (55 отсчётов)",
            xlabel="Номер отсчёта n", ylabel="y₁(n)")
axes[0].legend()
axes[0].grid(alpha=0.35)

axes[1].stem(np.arange(N2), y1)
axes[1].set(title="Реакция y₁(n), ограниченная длиной воздействия (33 отсчёта)",
            xlabel="Номер отсчёта n", ylabel="y₁(n)")
axes[1].grid(alpha=0.35)
plt.tight_layout()
plt.show()
```

**Анализ результатов**

В качестве входного воздействия сформирован дискретный прямоугольный импульс длиной $N_2=33$ отсчёта, из которых 17 единичных отсчётов ($n=0,\ldots,16$) и 16 нулевых ($n=17,\ldots,32$).

Реакция системы $y_1(n)$ вычислена методом линейной свёртки входного воздействия $x(n)$ и первых $N_1=23$ отсчётов импульсной характеристики $h(n)$.

Длина полной свёртки конечных последовательностей составляет $N_1+N_2-1=23+33-1=55$ отсчётов, что соответствует диапазону номеров отсчётов от 0 до 54 на графике.

Для дальнейшего анализа свёртка ограничена первыми $N_2=33$ отсчётами. Это позволяет рассматривать реакцию на интервале входного воздействия и сравнить её с результатом вычисления по разностному уравнению в следующем задании. При этом полученная свёрткой реакция является приближённой, так как для расчёта взяты лишь первые $N_1=23$ отсчёта.

#### 3. Вычисление реакции по разностному уравнению

Реакция $y_2(n)$ на то же воздействие $x(n)$ вычисляется непосредственно по разностному уравнению рекурсивного звена второго порядка:

$$y_2(n)=b_0x(n)+b_1x(n-1)+b_2x(n-2)-a_1y_2(n-1)-a_2y_2(n-2).$$

То есть:

$$\begin{aligned}
y_2(n)={}&0.56x(n)+0.579936x(n-1)+0.784x(n-2)\\
&+0.8528y_2(n-1)-0.658y_2(n-2).
\end{aligned}$$

Используются нулевые начальные условия: $x(n)=0$ и $y_2(n)=0$ при $n<0$. В качестве входного воздействия берётся прямоугольный импульс из предыдущего пункта длиной $N_2=33$ отсчёта, также для сравнения используются первые 33 отсчёта свёртки $y_1(n)$.

```python
y2 = signal.lfilter(b, a, x)
n_y2 = np.arange(N2)

difference = y2 - y1
max_difference = np.max(np.abs(difference))
first_part_difference = np.max(np.abs(difference[:N1]))

print(f"Длина воздействия: {len(x)} отсчёта")
print(f"Длина реакции по разностному уравнению: {len(y2)} отсчёта")
print(f"Максимальное расхождение на первых {N1} отсчётах: "
      f"{first_part_difference:.3e}")
print(f"Максимальное расхождение на всех {N2} отсчётах: "
      f"{max_difference:.6f}")
print(f"Номер отсчёта максимального расхождения: "
      f"{np.argmax(np.abs(difference))}")

assert len(y2) == N2
assert np.allclose(y1[:N1], y2[:N1], atol=1e-12, rtol=1e-12)

fig, axes = plt.subplots(2, 1, figsize=(10, 7))
axes[0].stem(n_x, x)
axes[0].set(title="Входное воздействие x(n)",
            xlabel="Номер отсчёта n", ylabel="x(n)")
axes[0].set_ylim(-0.12, 1.22)
axes[0].grid(alpha=0.35)

axes[1].stem(n_y2, y2)
axes[1].set(title="Реакция y₂(n) по разностному уравнению",
            xlabel="Номер отсчёта n", ylabel="y₂(n)")
axes[1].grid(alpha=0.35)
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(10, 4.5))
ax.plot(n_y2, y1, "o-", markersize=4,
        label="y₁(n): свёртка с 23 отсчётами h(n)")
ax.plot(n_y2, y2, "x--", markersize=5,
        label="y₂(n): разностное уравнение")
ax.axvline(N1 - 0.5, color="C3", linestyle=":",
           label="Граница первых 23 отсчётов")
ax.set(title="Сравнение реакций y₁(n) и y₂(n)",
       xlabel="Номер отсчёта n", ylabel="Реакция")
ax.grid(alpha=0.35)
ax.legend()
plt.tight_layout()
plt.show()
```

#### 4. Вычисление нулей и полюсов передаточной функции

Нули передаточной функции — корни её числителя, а полюсы — корни знаменателя (при отсутствии общих корней). Для звена второго порядка:

$$
H(z)=\frac{0.56+0.579936z^{-1}+0.784z^{-2}}{1-0.8528z^{-1}+0.658z^{-2}}.
$$

Для нахождения корней числитель и знаменатель умножим на $z^2$. Получим:

$$
\begin{aligned}
0.56z^2+0.579936z+0.784&=0 &&\text{(нули)},\\
z^2-0.8528z+0.658&=0 &&\text{(полюсы)}.
\end{aligned}
$$

Нули и полюсы передаточной функции представляются в алгебраической $z=a+jb$ и показательной $z=re^{j\varphi}$ формах, где модуль и аргумент комплексного числа определяются по формулам:

$$
r=\sqrt{a^2+b^2},\qquad \varphi=\operatorname{atan2}(b,a).
$$

Для комплексно-сопряжённых корней модули равны, а аргументы противоположны по знаку.

```python
zeros = np.roots(b)
poles_task4 = np.roots(a)

zeros = zeros[np.argsort(-zeros.imag)]
poles_task4 = poles_task4[np.argsort(-poles_task4.imag)]

def print_roots(name, roots):
    print(name)
    for i, z in enumerate(roots, start=1):
        print(f"  {i}: {z.real:+.6f}{z.imag:+.6f}j; "
              f"r = {abs(z):.6f}; "
              f"phi = {np.angle(z):+.6f} рад = "
              f"{np.angle(z) / np.pi:+.6f}π")

print_roots("Нули:", zeros)
print_roots("Полюсы:", poles_task4)

K = b[0] / a[0]
b_recovered = K * np.poly(zeros)
a_recovered = np.poly(poles_task4)

print(f"\nМножитель K = {K:.6f}")
print("Коэффициенты числителя:", np.real_if_close(b_recovered))
print("Коэффициенты знаменателя:", np.real_if_close(a_recovered))

assert np.allclose(b, b_recovered)
assert np.allclose(a, a_recovered)
print("Проверка разложения на множители выполнена успешно")
```

#### 6. Вычисление АЧХ и ФЧХ в шкале нормированных частот

**Частотная характеристика** линейной стационарной системы определяется подстановкой $z=e^{j\widehat\omega}$ в передаточную функцию:

$$
H(e^{j\widehat\omega})=\frac{b_0+b_1e^{-j\widehat\omega}+b_2e^{-2j\widehat\omega}}{a_0+a_1e^{-j\widehat\omega}+a_2e^{-2j\widehat\omega}}.
$$

**Амплитудно-частотная характеристика (АЧХ)** — модуль частотной характеристики:

$$ A(\widehat\omega)=|H(e^{j\widehat\omega})|. $$

**Фазочастотная характеристика (ФЧХ)** — её аргумент:

$$ \varphi(\widehat\omega)=\arg H(e^{j\widehat\omega}). $$

Нормированная угловая частота $\widehat\omega$ измеряется в радианах на отсчёт. Основная полоса частот: $-\pi\le\widehat\omega\le\pi$, для действительной системы АЧХ симметрична относительно нуля, поэтому отдельно можно рассматривать $0\le\widehat\omega\le\pi$.

При построении ФЧХ используется главный аргумент комплексного числа в диапазоне от $-\pi$ до $\pi$. Переход от $+\pi$ к $-\pi$ (или наоборот) может выглядеть как скачок на графике, хотя сама частотная характеристика меняется непрерывно.

```python
omega = np.linspace(-np.pi, np.pi, 2001)
omega, H = signal.freqz(b, a, worN=omega)

amplitude = np.abs(H)
phase = np.angle(H)

fig, axes = plt.subplots(2, 1, figsize=(10, 7), sharex=True)
axes[0].plot(omega / np.pi, amplitude)
axes[0].set_title("Амплитудно-частотная характеристика")
axes[0].set_ylabel(r"$|H(e^{j\hat{\omega}})|$")
axes[0].grid(True)

axes[1].plot(omega / np.pi, phase)
axes[1].set_title("Фазочастотная характеристика")
axes[1].set_xlabel(r"Нормированная частота $\hat{\omega}/\pi$")
axes[1].set_ylabel("Фаза, рад")
axes[1].set_yticks([-np.pi, -np.pi/2, 0, np.pi/2, np.pi])
axes[1].set_yticklabels([r"$-\pi$", r"$-\pi/2$", "0", r"$\pi/2$", r"$\pi$"])
axes[1].grid(True)
plt.tight_layout()
plt.show()

omega_check = np.array([0, np.pi / 4, np.pi / 2, 3 * np.pi / 4, np.pi])
_, H_check = signal.freqz(b, a, worN=omega_check)
print("АЧХ на характерных нормированных частотах:")
for w, value in zip(omega_check, H_check):
    print(f"omega/pi = {w / np.pi:.2f}: A = {abs(value):.6f}")

mask = omega >= 0
w_pos, A_pos = omega[mask], amplitude[mask]
print(f"Максимум АЧХ: {A_pos.max():.6f} при omega/pi = {w_pos[A_pos.argmax()]/np.pi:.4f}")
print(f"Минимум АЧХ: {A_pos.min():.6f} при omega/pi = {w_pos[A_pos.argmin()]/np.pi:.4f}")

jumps = np.where(np.abs(np.diff(phase)) > np.pi)[0]
for index in jumps:
    print(f"Разрыв отображения фазы возле omega/pi = {omega[index] / np.pi:.4f}; "
          f"АЧХ = {amplitude[index]:.6f}")
```

**Анализ результатов**

Основная полоса нормированных угловых частот составляет $[-\pi;\pi]$ рад/отсчёт.

На графике АЧХ наблюдается максимум примерно при $\widehat\omega\approx0.307\pi$, что связано с расположением комплексно-сопряжённых полюсов под углами $\pm0.323819\pi$. Вблизи этих частот АЧХ увеличивается. В области расположения нулей $\widehat\omega\approx0.644\pi$ АЧХ уменьшается, но не достигает нуля, поскольку нули находятся вне единичной окружности.

На графике ФЧХ наблюдаются скачки около $\widehat\omega\approx\pm0.503\pi$. Они связаны с переходом фазы от $-\pi$ к $+\pi$ при отображении графика. 
В этих точках значение АЧХ составляет примерно $0.655$, то есть частотная характеристика не обращается в ноль.

Значения АЧХ на низких частотах больше, чем на высоких: $A(0)\approx2.3894$, $A(\pi/2)\approx0.6766$, $A(\pi)\approx0.3043$. Следовательно, система преимущественно подавляет высокочастотные составляющие воздействия.

### Часть 2. Дискретное преобразование Фурье

#### 1. Формирование дискретной периодической последовательности

$$
x(nT)=A_1\cos(2\pi f_1nT+\pi/4)+A_2\cos(2\pi f_2nT+\pi/8).
$$

Параметры: $N=64$, $f_{\text{д}}=8000$ Гц, $A_1=1.03$, $A_2=2.06$, $f_1=1000$ Гц, $f_2=1500$ Гц.

Период дискретизации равен $T=1/f_{\text{д}}=0.000125$ с. Получаем дискретную последовательность:

$$
x(n)=1.03\cos\left(2\pi\cdot1000\frac{n}{8000}+\frac{\pi}{4}\right)
+2.06\cos\left(2\pi\cdot1500\frac{n}{8000}+\frac{\pi}{8}\right),\quad n=0,\ldots,63.
$$

```python
N = 64
fd2 = 8000
T = 1 / fd2
A1, A2 = 1.03, 2.06
f1, f2 = 1000, 1500
phi1, phi2 = np.pi / 4, np.pi / 8

n2 = np.arange(N)
t2 = n2 * T
n_norm = n2 / N
x_periodic = (A1 * np.cos(2 * np.pi * f1 * t2 + phi1)
              + A2 * np.cos(2 * np.pi * f2 * t2 + phi2))

period_samples = 16
period_s = period_samples * T

print(f"N = {N}, fд = {fd2} Гц, T = {T:.6f} с")
print(f"Период сигнала: {period_s:.4f} с ({period_samples} отсчётов)")
print(f"Длина блока: {N} отсчётов ({N // period_samples} периода)")

fig, axes = plt.subplots(2, 1, figsize=(10, 7))

axes[0].stem(n2, x_periodic, basefmt=" ")
axes[0].set_title("Последовательность в шкале дискретного времени")
axes[0].set_xlabel("Дискретное время n, отсчёты")
axes[0].set_ylabel("x(nT)")
axes[0].grid(True, alpha=0.3)

axes[1].stem(n_norm, x_periodic, basefmt=" ")
axes[1].set_title("Последовательность в шкале нормированного дискретного времени")
axes[1].set_xlabel("Нормированное дискретное время n/N")
axes[1].set_ylabel("x(nT)")
axes[1].grid(True, alpha=0.3)
axes[1].set_xlim(-0.02, 1.02)

plt.tight_layout()
plt.show()
```

**Анализ результатов**

Cформирована последовательность из $N=64$ отсчётов с частотой дискретизации $8000$ Гц. Она состоит из двух косинусоид с частотами $1000$ и $1500$ Гц и амплитудами $1.03$ и $2.06$.

Первый график показывает значения последовательности в зависимости от номера отсчёта $n$, а второй - в зависимости от нормированного времени $n/N$.

Общий период сигнала равен $0.002$ с, или 16 отсчётам. Следовательно, на участке из 64 отсчётов содержатся четыре полных периода.

#### 2. Вычисление ОДПФ от ДПФ

Дискретное преобразование Фурье (ДПФ):

$$
X(k)=\sum_{n=0}^{N-1}x(nT)e^{-j2\pi kn/N}, \qquad k=0,\ldots,N-1.
$$

Обратное ДПФ (ОДПФ):

$$
x_{\text{восст}}(nT)=\frac{1}{N}\sum_{k=0}^{N-1}X(k)e^{j2\pi kn/N}.
$$

```python
X_dft = np.fft.fft(x_periodic)
x_reconstructed_complex = np.fft.ifft(X_dft)
x_reconstructed = x_reconstructed_complex.real

max_reconstruction_error = np.max(np.abs(x_reconstructed - x_periodic))
max_imaginary_part = np.max(np.abs(x_reconstructed_complex.imag))

print(f"Количество коэффициентов ДПФ: {len(X_dft)}")
print(f"Количество восстановленных отсчётов: {len(x_reconstructed)}")
print(f"Максимальная ошибка восстановления: {max_reconstruction_error:.3e}")
print(f"Максимальная мнимая часть после ОДПФ: {max_imaginary_part:.3e}")

fig, ax = plt.subplots(figsize=(10, 4))
ax.stem(n_norm, x_reconstructed, basefmt=" ")
ax.set_title("Последовательность, восстановленная с помощью ОДПФ")
ax.set_xlabel("Нормированное дискретное время n/N")
ax.set_ylabel("xвосст(nT)")
ax.set_xlim(-0.02, 1.02)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

assert np.allclose(x_reconstructed, x_periodic, atol=1e-12)
```

**Анализ результатов**

Вычислено ДПФ последовательности из 64 отсчётов, после чего обратное ДПФ восстановило 64 значения сигнала. График восстановленной последовательности повторяет график исходной последовательности из пункта 1.

Максимальное отклонение между исходными и восстановленными отсчётами очень мало и связано с округлением. Следовательно, обратное ДПФ восстанавливает исходную последовательность с высокой точностью.

#### 3. Вычисление амплитудного и фазового спектров

Амплитудный спектр показывает амплитуды частотных составляющих, а фазовый — их начальные фазы.

Дискретная нормированная частота определяется как $f_{\text{норм}}=k/N$ (циклов на отсчёт), а абсолютная частота - как $f=kf_{\text{д}}/N$ (Гц). Шаг по частоте равен $\Delta f=f_{\text{д}}/N=125$ Гц.

Для действительной последовательности достаточно рассмотреть неотрицательные частоты. В одностороннем амплитудном спектре значения $|X(k)|$ для внутренних положительных частот умножаются на $2/N$, а для нулевой и крайней частоты - на $1/N$. 

Фазовый спектр вычисляется как $\varphi(k)=\arg X(k)$. На графиках отображаются фазы всех частотных отсчётов, включая те, где амплитуда практически равна нулю.

```python
X_spectrum = np.fft.rfft(x_periodic)
k_spectrum = np.arange(len(X_spectrum))
f_norm_spectrum = k_spectrum / N
f_hz_spectrum = k_spectrum * fd2 / N

amp_spectrum = np.abs(X_spectrum) / N
amp_spectrum[1:-1] *= 2

phase_spectrum = np.angle(X_spectrum)
threshold = 1e-8 * np.max(amp_spectrum)

fig, axes = plt.subplots(2, 2, figsize=(12, 7), constrained_layout=True)

axes[0, 0].stem(f_norm_spectrum, amp_spectrum, basefmt=" ")
axes[0, 0].set_title("Амплитудный спектр: нормированная частота")
axes[0, 0].set_xlabel(r"$k/N$, циклы/отсчёт")
axes[0, 0].set_ylabel("Амплитуда")

axes[0, 1].stem(f_hz_spectrum, amp_spectrum, basefmt=" ")
axes[0, 1].set_title("Амплитудный спектр: абсолютная частота")
axes[0, 1].set_xlabel("Частота, Гц")
axes[0, 1].set_ylabel("Амплитуда")

axes[1, 0].stem(f_norm_spectrum, phase_spectrum, basefmt=" ")
axes[1, 0].set_title("Фазовый спектр: нормированная частота")
axes[1, 0].set_xlabel(r"$k/N$, циклы/отсчёт")
axes[1, 0].set_ylabel("Фаза, рад")

axes[1, 1].stem(f_hz_spectrum, phase_spectrum, basefmt=" ")
axes[1, 1].set_title("Фазовый спектр: абсолютная частота")
axes[1, 1].set_xlabel("Частота, Гц")
axes[1, 1].set_ylabel("Фаза, рад")

for ax in axes.flat:
    ax.grid(alpha=0.3)
axes[0, 0].set_xlim(0, 0.5)
axes[1, 0].set_xlim(0, 0.5)
axes[0, 1].set_xlim(0, fd2 / 2)
axes[1, 1].set_xlim(0, fd2 / 2)
plt.show()

print(f"Шаг по частоте: {fd2/N:.0f} Гц")
for k in np.flatnonzero(amp_spectrum > threshold):
    print(f"k={k:2d}, k/N={f_norm_spectrum[k]:.5f}, "
          f"f={f_hz_spectrum[k]:.0f} Гц, "
          f"A={amp_spectrum[k]:.6f}, φ={phase_spectrum[k]:.6f} рад")
```

**Анализ результатов**

Дискретная нормированная частота связана с номером спектрального отсчёта соотношением $f_{\text{норм}}=k/N$. При $N=64$ соседние частотные отсчёты отличаются на $1/64$ цикла/отсчёт.

Абсолютная частота вычисляется по формуле $f=f_{\text{норм}}f_{\text{д}}$. При $f_{\text{д}}=8000$ Гц шаг по частоте составляет $125$ Гц. Поэтому $k=8$ соответствует $1000$ Гц, а $k=12$ - $1500$ Гц.

На амплитудном спектре видны два пика: $A=1.03$ при $1000$ Гц и $A=2.06$ при $1500$ Гц. На фазовом спектре в этих точках получены фазы $\pi/4\approx0.7854$ рад и $\pi/8\approx0.3927$ рад. Эти значения совпадают с параметрами исходных косинусоид.

На всех четырёх графиках представлены одни и те же частотные отсчёты. В каждой паре графиков изменяется только шкала горизонтальной оси.

На остальных частотах амплитуды практически равны нулю, поэтому соответствующие значения фазы преимущественно определяются численными погрешностями и не характеризуют реальные частотные составляющие сигнала.

#### 4. Вычисление ДПФ конечной последовательности

Для конечной последовательности из $N=64$ отсчётов вычисляется ДПФ:

$$
X(k)=\sum_{n=0}^{N-1}x(n)e^{-j2\pi kn/N},\qquad k=0,\ldots,N-1.
$$

Модуль ДПФ равен $|X(k)|$ и показывает величину коэффициентов преобразования без нормировки.

Для периодического продолжения этой последовательности коэффициенты ряда Фурье равны $C(k)=X(k)/N$. Поэтому двусторонний амплитудный спектр периодической последовательности определяется как

$$
|C(k)|=\frac{|X(k)|}{N}.
$$

На графиках отображаются все 64 частотных отсчёта в шкале нормированных частот $k/N$, от $0$ до $63/64$. При этом вторая половина шкалы соответствует отрицательным частотам, записанным в порядке коэффициентов ДПФ.

В отличие от одностороннего спектра в пункте 3, здесь амплитуды не удваиваются. Отрицательные частоты отображаются в правой половине графика из-за периодичности ДПФ.

```python
X_finite = np.fft.fft(x_periodic, n=N)
k_full = np.arange(N)
f_norm_full = k_full / N
module_finite = np.abs(X_finite)

amp_periodic_full = module_finite / N

fig, axes = plt.subplots(2, 1, figsize=(11, 7), sharex=True)
axes[0].stem(f_norm_full, module_finite, basefmt=" ")
axes[0].set_title("Модуль ДПФ конечной последовательности")
axes[0].set_ylabel(r"$|X(k)|$")

axes[1].stem(f_norm_full, amp_periodic_full, basefmt=" ")
axes[1].set_title("Двусторонний амплитудный спектр периодической последовательности")
axes[1].set_xlabel(r"Нормированная частота $k/N$, циклы/отсчёт")
axes[1].set_ylabel(r"$|C(k)|=|X(k)|/N$")

for ax in axes:
    ax.grid(True, alpha=0.3)
    ax.set_xlim(0, 1)
    ax.axvline(0.5, color="gray", linestyle="--", linewidth=1, alpha=0.7)
axes[1].set_xticks(np.arange(0, 1.01, 0.125))
plt.tight_layout()
plt.show()

amp_one_sided_check = amp_periodic_full[:N // 2 + 1].copy()
amp_one_sided_check[1:-1] *= 2
max_spectrum_difference = np.max(np.abs(amp_one_sided_check - amp_spectrum))
print(f"Максимальное расхождение с односторонним спектром пункта 3: {max_spectrum_difference:.2e}")
for k in (8, 12, 52, 56):
    print(f"k={k:2d}, k/N={k/N:.4f}: "
          f"|X(k)|={module_finite[k]:.4f}, |C(k)|={amp_periodic_full[k]:.4f}")
assert np.allclose(amp_one_sided_check, amp_spectrum, atol=1e-12)
assert np.allclose(module_finite[1:N // 2], module_finite[-1:N // 2:-1], atol=1e-12)
```

**Анализ результатов**

На обоих графиках видны четыре пика при нормированных частотах $k/N=0.125$, $0.1875$, $0.8125$ и $0.875$. Первые два соответствуют частотам $+1000$ и $+1500$ Гц, а последние два — частотам $-1500$ и $-1000$ Гц. Симметрия связана с тем, что исходная последовательность действительная.

В модуле ДПФ высоты пиков составляют $32.96$ для частот $\pm1000$ Гц и $65.92$ для $\pm1500$ Гц. В двустороннем амплитудном спектре периодической последовательности значения в $N=64$ раза меньше: $0.515$ и $1.03$ соответственно, поскольку $|C(k)|=|X(k)|/N$.

В пункте 3 построен односторонний амплитудный спектр с пиками $1.03$ и $2.06$. В пункте 4 показан полный спектр, где каждой частоте соответствуют два симметричных пика с вдвое меньшими амплитудами. Таким образом, результаты обоих пунктов согласуются.

Таким образом, модуль ДПФ конечной последовательности и амплитудный спектр её периодического продолжения имеют пики на одинаковых частотах, но отличаются нормировкой.

#### 5. Восстановление периодического аналогового сигнала

Периодический аналоговый сигнал восстанавливается по коэффициентам ДПФ $X(k)$ с помощью суммы:

$$
x_{\text{восст}}(t)=\frac{1}{N}\sum_{k=-N/2}^{N/2-1}X_a(k)\exp\left(j\frac{2\pi k}{NT}t\right),
$$

где $N=64$, $T=1/8000$ с. Для перехода от обычного порядка коэффициентов ДПФ к отрицательным и положительным частотам используется соответствие

$$
X_a(k)=\begin{cases}
X(N+k), & -N/2\le k\le -1,\\
X(k), & 0\le k\le N/2-1.
\end{cases}
$$

Время $t$ изменяется от $0$ до $(N-1)T$ с шагом $\Delta t=0.25T$. Восстановление позволяет получить значения сигнала между исходными отсчётами. Для проверки восстановленный сигнал сравнивается с исходной аналоговой функцией:

$$
x_{\text{исх}}(t)=1.03\cos\left(2\pi\cdot1000t+\frac{\pi}{4}\right)
+2.06\cos\left(2\pi\cdot1500t+\frac{\pi}{8}\right).
$$

```python
k_signed = np.arange(-N // 2, N // 2)
X_a = np.fft.fftshift(X_dft)
f_signed = k_signed / (N * T)

dt = 0.25 * T
t_fine = np.arange(4 * (N - 1) + 1) * dt

exponents = np.exp(1j * 2 * np.pi * np.outer(t_fine, f_signed))
x_analog_complex = (exponents @ X_a) / N
x_analog_reconstructed = x_analog_complex.real

x_analog_original = (
    A1 * np.cos(2 * np.pi * f1 * t_fine + phi1)
    + A2 * np.cos(2 * np.pi * f2 * t_fine + phi2)
)

analog_amplitude_spectrum = np.abs(X_a) / N
max_analog_error = np.max(np.abs(x_analog_reconstructed - x_analog_original))
max_sample_error = np.max(np.abs(x_analog_reconstructed[::4] - x_periodic))
max_analog_imag = np.max(np.abs(x_analog_complex.imag))

print(f"Шаг восстановления: Δt = {dt:.8f} с = {dt/T:.2f}T")
print(f"Временной интервал: [0; {(N-1)*T:.6f}] с")
print(f"Число точек восстановленного сигнала: {len(t_fine)}")
print(f"Максимальное отличие от исходной аналоговой функции: {max_analog_error:.3e}")
print(f"Максимальное отличие в точках дискретизации: {max_sample_error:.3e}")
print(f"Максимальная мнимая часть результата: {max_analog_imag:.3e}")

fig, axes = plt.subplots(2, 1, figsize=(11, 7))
axes[0].stem(n2, x_periodic, basefmt=" ")
axes[0].set_title("Исходная периодическая последовательность")
axes[0].set_xlabel("Номер отсчёта n")
axes[0].set_xlim(0, N - 1)
axes[0].set_ylabel("x(nT)")
axes[1].stem(k_full / N, module_finite, basefmt=" ")
axes[1].set_title("Модуль ДПФ исходной последовательности")
axes[1].set_xlabel("Нормированная частота k/N, циклы/отсчёт")
axes[1].set_ylabel("|X(k)|")
for ax in axes:
    ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

fig, axes = plt.subplots(2, 1, figsize=(11, 7))
axes[0].plot(t_fine * 1000, x_analog_reconstructed, label="Восстановленный сигнал")
axes[0].plot(t2 * 1000, x_periodic, "o", markersize=3,
             label="Исходные дискретные отсчёты")
axes[0].set_title("Восстановленный аналоговый сигнал")
axes[0].set_xlabel("Время t, мс")
axes[0].set_ylabel("xвосст(t)")
axes[0].legend()

axes[1].stem(f_signed, analog_amplitude_spectrum, basefmt=" ")
axes[1].set_title("Амплитудный спектр восстановленного аналогового сигнала")
axes[1].set_xlabel("Частота f, Гц")
axes[1].set_ylabel("|X_a(k)|/N")
for ax in axes:
    ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(11, 4))
ax.plot(t_fine * 1000, x_analog_original,
        label="Исходный аналоговый сигнал", linewidth=2)
ax.plot(t_fine * 1000, x_analog_reconstructed, "--",
        label="Восстановленный сигнал", linewidth=1.5)
ax.set_title("Сравнение исходного и восстановленного аналоговых сигналов")
ax.set_xlabel("Время t, мс")
ax.set_ylabel("x(t)")
ax.grid(True, alpha=0.3)
ax.legend()
plt.tight_layout()
plt.show()

assert len(t_fine) == 4 * (N - 1) + 1
assert np.allclose(x_analog_reconstructed[::4], x_periodic, atol=1e-12)
assert np.allclose(x_analog_reconstructed, x_analog_original, atol=1e-12)
```

**Анализ результатов**

Восстановление выполнено на интервале $0\le t\le(N-1)T$ с шагом $\Delta t=0.25T$. Получено 253 значения восстановленного сигнала вместо 64 исходных отсчётов.

На первом наборе графиков показаны исходная последовательность и модуль её ДПФ. 

На втором - восстановленный аналоговый сигнал и его двусторонний амплитудный спектр. Пики спектра находятся на частотах $\pm1000$ и $\pm1500$ Гц, их высоты равны $0.515$ и $1.03$ соответственно. Это половины амплитуд исходных косинусоид, поскольку положительные и отрицательные частоты показаны отдельно.

На третьем графике исходная и восстановленная аналоговые функции практически полностью совпадают. В точках $t=nT$ восстановленный сигнал также совпадает с исходными отсчётами. Небольшие расхождения связаны с точностью вычислений. Таким образом, по коэффициентам ДПФ удалось восстановить не только исходные отсчёты, но и значения сигнала между ними.

Поэтому эти составляющие представлены в ДПФ без спектральной утечки, а их сумма позволяет восстановить исходные косинусоиды.

#### 7. Проверка равенства Парсеваля

Равенство Парсеваля показывает, что энергию конечной последовательности можно вычислить по отсчётам сигнала и по коэффициентам его ДПФ.

$$
E_{\text{сигн}}=\sum_{n=0}^{N-1}|x(n)|^2,
\qquad
E_{\text{ДПФ}}=\frac{1}{N}\sum_{k=0}^{N-1}|X(k)|^2.
$$

```python
energy_time = np.sum(np.abs(x_periodic) ** 2)
energy_frequency = np.sum(np.abs(X_dft) ** 2) / N

parseval_difference = abs(energy_time - energy_frequency)
relative_difference = parseval_difference / energy_time

print(f"Энергия по отсчётам сигнала: {energy_time:.6f}")
print(f"Энергия по коэффициентам ДПФ: {energy_frequency:.6f}")
```

Энергия последовательности, вычисленная по исходным отсчётам, равна $169.744$. При расчёте по коэффициентам ДПФ получено то же значение - $169.744$.

В данном расчёте разница между суммами при используемой точности вычислений равна нулю.

Таким образом, равенство Парсеваля выполняется - энергию сигнала можно определить как по его отсчётам во времени, так и по частотным коэффициентам ДПФ.
