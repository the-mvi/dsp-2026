## Основные теоретические положения

### Дискретные сигналы и нормирование

Дискретный сигнал $x(nT)$ получается взятием отсчетов аналогового сигнала $x(t)$ с периодом дискретизации $T = 1/f_d$, где $f_d$ - частота дискретизации. Нормированное время $\hat{t} = t/T$; при $t = nT$ получаем $\hat{t} = n$ и $x(nT) \equiv x(n)$.

Нормированные частоты: $\hat{f} = f/f_d = fT$, $\hat{\omega} = \omega T = 2\pi f / f_d$. Основная полоса частот: $f \in [0;\, f_d/2]$, $\hat{f} \in [0;\, 0.5]$, $\hat{\omega} \in [0;\, \pi]$. По теореме Котельникова непрерывный сигнал, частота которого ограничена сверху значением $f_B$, полностью определяется своими отсчетами, взятыми с частотой $f_d \geq 2 f_B$; частота Найквиста $f_N = f_d/2$ - максимальная частота, которую можно воспроизвести без искажений.

Функция $int(t)$ - округление до ближайшего целого.

Энергия последовательности длины $N$: $E_x = \sum_{n=0}^{N-1} x^2(n)$; средняя мощность $\overline{P}_x = E_x / N$.

Цифровой единичный импульс $u_0(n)$ равен 1 при $n = 0$ и 0 при $n \neq 0$; цифровой единичный скачок $u_1(n)$ равен 1 при $n \geq 0$ и 0 при $n < 0$.

### Линейные дискретные системы (ЛДС)

Линейная система обладает свойствами аддитивности и однородности; стационарная (инвариантная) система - еще и свойством инвариантности к сдвигу. Нулевые начальные условия: $x(n-i) = 0$ и $y(n-k) = 0$ при $n-i<0$, $n-k<0$.

**Импульсная характеристика (ИХ)** $h(n)$ - реакция ЛДС на $u_0(n)$ при нулевых начальных условиях.

**Формула свертки** (реакция на произвольное воздействие):
$$ y(n) = \sum_{m=0}^{\infty} h(m)\,x(n-m) = \sum_{m=0}^{\infty} h(n-m)\,x(m). $$
Для последовательностей конечной длины $N_x$ и $N_h$ длина линейной свертки $N_y = N_x + N_h - 1$.

**Разностное уравнение (РУ)**:
$$ y(n) = \sum_{i=0}^{N-1} b_i\,x(n-i) - \sum_{k=1}^{M-1} a_k\,y(n-k). $$
Если все $a_k = 0$, система нерекурсивная (КИХ); если хотя бы один $a_k \neq 0$ - рекурсивная (БИХ, ИХ бесконечной длины).

### Z-преобразование и передаточная функция

$Z$-преобразование: $X(z) = \sum_{n=0}^{\infty} x(n) z^{-n}$; свойства: линейность, $\mathscr{Z}\{x(n-m)\} = z^{-m}X(z)$, $\mathscr{Z}\{x_1 * x_2\} = X_1(z)X_2(z)$.

**Передаточная функция (ПФ)** - отношение $z$-изображений реакции и воздействия при нулевых начальных условиях, $H(z) = Y(z)/X(z) = \mathscr{Z}\{h(n)\}$. Для РУ:
$$ H(z) = \frac{\sum_{i=0}^{N-1} b_i z^{-i}}{1 + \sum_{k=1}^{M-1} a_k z^{-k}}. $$
Для рекурсивного звена 2-го порядка:
$$ H(z) = \frac{b_0 + b_1 z^{-1} + b_2 z^{-2}}{1 + a_1 z^{-1} + a_2 z^{-2}} = \frac{b_0 z^2 + b_1 z + b_2}{z^2 + a_1 z + a_2}. $$

**Нули** - значения $z = z_{\circ}$, при которых ПФ равна нулю; **полюсы** - значения $z = z_{*}$, при которых знаменатель ПФ равен нулю. При комплексно-сопряженной паре полюсов $z_{*1,2} = r_* e^{\pm j\varphi_*}$ знаменатель раскладывается как
$$ 1 + a_1 z^{-1} + a_2 z^{-2} = (1 - r_* e^{j\varphi_*} z^{-1})(1 - r_* e^{-j\varphi_*} z^{-1}),\quad a_1 = -2 r_* \cos\varphi_*,\quad a_2 = r_*^2, $$
и справедливо соответствие
$$ r_*^n \frac{\sin\big((n+1)\varphi_*\big)}{\sin\varphi_*}\,u_1(n) \;\leftrightarrow\; \frac{1}{1 + a_1 z^{-1} + a_2 z^{-2}}. $$

**Устойчивость.** ЛДС устойчива, если ограниченное воздействие дает ограниченную реакцию. Критерий во временной области: $\sum_{n=0}^{\infty} |h(n)| < \infty$. Критерий в $z$-области: для устойчивости ЛДС необходимо и достаточно, чтобы все полюсы ее ПФ, представленной в виде $H(z) = A_0 + \sum_{k=1}^{M-1} \frac{A_k}{1 - \alpha_k z^{-1}}$, находились внутри круга единичного радиуса $|z| < 1$. Карта нулей и полюсов - их изображение на $z$-плоскости (нули - «o», полюсы - «x») вместе с единичной окружностью.

### Частотные характеристики ЛДС

Отображение $z = e^{pT}$ переводит отрезок мнимой оси $\omega \in (-\pi/T;\, \pi/T]$ в единичную окружность $z = e^{j\hat{\omega}}$. Частотная характеристика:
$$ H(e^{j\hat{\omega}}) = \sum_{n=0}^{\infty} h(n) e^{-j\hat{\omega} n} = H(z)\big|_{z = e^{j\hat{\omega}}} = A(\hat{\omega})\, e^{j\varphi(\hat{\omega})}, $$
где $A(\hat{\omega}) = |H(e^{j\hat{\omega}})|$ - АЧХ (Амплитудная Частотная Характеристика), $\varphi(\hat{\omega}) = \arg H(e^{j\hat{\omega}})$ - ФЧХ (Фазовая Частотная Характеристика). Частотные характеристики периодичны с периодом $2\pi$ по $\hat{\omega}$ (с периодом $\omega_d = 2\pi/T$ по $\omega$); для вещественной ИХ АЧХ - четная, ФЧХ - нечетная функция частоты.

### Дискретное преобразование Фурье

Для $N$-точечной последовательности (одного периода периодической последовательности) ДПФ и ОДПФ (Обратное ДПФ):
$$ X(k) = \sum_{n=0}^{N-1} x(n) W_N^{nk},\quad x(n) = \frac{1}{N}\sum_{k=0}^{N-1} X(k) W_N^{-nk},\quad W_N^{nk} = e^{-j\frac{2\pi}{N}nk}. $$
Отсчет $X(k)$ соответствует частоте $\hat{\omega}_k = 2\pi k/N$ и абсолютной частоте $f_k = k f_d/N$ (шаг по частоте $\Delta f = f_d/N = 1/(NT)$). Амплитудный спектр - $|X(k)|$, фазовый - $\arg X(k)$.

Свойства ДПФ: периодичность по $k$ с периодом $N$; линейность; для вещественной $x(n)$ выполняется $|X(k)| = |X(N-k)|$, $\arg X(k) = -\arg X(N-k)$. ДПФ конечной последовательности - дискретизация ее непрерывного спектра $X(e^{j\hat{\omega}})$ с шагом $2\pi/N$ (теорема Котельникова в частотной области).

**Равенство Парсеваля** для ДПФ:
$$ \sum_{n=0}^{N-1} |x(n)|^2 = \frac{1}{N}\sum_{k=0}^{N-1} |X(k)|^2. $$

### Восстановление аналогового сигнала по ДПФ

Непрерывная периодическая функция с периодом $T_s$ представляется рядом Фурье $x_p(t) = \sum_{k=-\infty}^{\infty} c_k e^{jk\Delta\omega t}$, $\Delta\omega = 2\pi/T_s$. Для сигнала с периодом $T_s = NT$, спектр которого лежит в основной полосе ($f_d \geq 2f_B$), в ряду остаются $N$ членов с $-N/2 \le k \le N/2-1$, а коэффициенты выражаются через ДПФ: $c_k = X_a(k)/N$. Отсюда
$$ x(t) = \frac{1}{N}\sum_{k=-N/2}^{N/2-1} X_a(k)\, e^{j\frac{2\pi}{NT}kt}. $$
По свойству периодичности ДПФ $X(k) = X(k+N)$, поэтому для отрицательных $k$ берется $X_a(k) = X(N+k)$.

### Равенство Парсеваля

Для принятого определения ДПФ энергия конечной последовательности во временной и частотной областях связана равенством $$\sum_{n=0}^{N-1}\left|x(n)\right|^2=\frac1N\sum_{k=0}^{N-1}\left|X(k)\right|^2$$

Множитель $1/N$ есть нормировка при которой ДПФ не делится на $N$, а ОДПФ содержит множитель $1/N$

```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({'figure.figsize': (11, 3.6), 'axes.grid': True, 'grid.alpha': 0.35,
                     'font.size': 10, 'axes.titlesize': 11})
np.set_printoptions(precision=4, suppress=True, linewidth=110)

def stem(ax, n, v, label=None, color='C0', ms=4):
    # Дискретный график (отсчеты) в едином стиле
    ml, sl, bl = ax.stem(n, v, linefmt=color + '-', markerfmt=color + 'o', basefmt='k-', label=label)
    plt.setp(ml, markersize=ms); plt.setp(sl, linewidth=1); plt.setp(bl, linewidth=0.8)

def w_axis(ax, N, title):
    # Нижняя ось - номер отсчета ДПФ k, верхняя - дискретная нормированная частота w_k = 2*pi*k/N (в долях pi)
    sec = ax.secondary_xaxis('top', functions=(lambda kk: 2 * kk / N, lambda ww: ww * N / 2))
    sec.set_xlabel(r'$\hat{\omega}_k/\pi = 2k/N$', fontsize=9)
    ax.set_title(title, pad=8)
    ax.set_xlabel('k')

def r_int(t):
    # int(t) - округление до ближайшего целого (половина округляется вверх)
    return int(np.floor(t + 0.5))
```

#### 1. Импульсная характеристика

ИХ вычисляется как реакция звена на цифровой единичный импульс $u_0(n)$ длины $N_1$ при нулевых начальных условиях (по разностному уравнению, см. п. 3).

```python
def diff_eq(b, a, x):
    # Реакция по разностному уравнению y(n) = sum b_i x(n-i) - sum a_k y(n-k), a[0] = 1,
    # нулевые начальные условия: x(n-i) = 0, y(n-k) = 0 при отрицательных индексах
    y = np.zeros(len(x))
    for n in range(len(x)):
        acc = sum(b[i] * x[n - i] for i in range(len(b)) if n - i >= 0)
        acc -= sum(a[k] * y[n - k] for k in range(1, len(a)) if n - k >= 0)
        y[n] = acc / a[0]
    return y

n1 = np.arange(N1)
u0 = (n1 == 0).astype(float)          # цифровой единичный импульс
h = diff_eq(b, a, u0)                 # импульсная характеристика

fig, ax = plt.subplots()
stem(ax, n1, h)
ax.set(title=f'Импульсная характеристика h(n), N1 = {N1}', xlabel='n', ylabel='h(n)')
plt.show()
print('h(n) =', h)
```

**Аналитическая формула ИХ.** Знаменатель Передаточной Функции $z^2 + a_1 z + a_2$ имеет дискриминант $a_1^2 - 4a_2 = 0.9078 - 2.728 < 0$, поэтому полюсы комплексно-сопряженные: $z_{*1,2} = r_* e^{\pm j\varphi_*}$, где $r_* = \sqrt{a_2}$, $\cos\varphi_* = -a_1/(2 r_*)$. Используя соответствие $r_*^n \sin((n+1)\varphi_*)/\sin\varphi_* \leftrightarrow 1/(1 + a_1 z^{-1} + a_2 z^{-2})$, свойство линейности и теорему о задержке, при нулевых начальных условиях получаем:
$$ h(n) = \frac{1}{\sin\varphi_*}\Big[ b_0 r_*^{n} \sin\big((n+1)\varphi_*\big)u_1(n) + b_1 r_*^{n-1} \sin\big(n\varphi_*\big)u_1(n-1) + b_2 r_*^{n-2} \sin\big((n-1)\varphi_*\big)u_1(n-2) \Big]. $$
Множители $u_1(n-1)$, $u_1(n-2)$ существенны: без них, например, третье слагаемое при $n = 0$ дало бы ненулевое значение $-b_2/r_*^2$, что противоречит нулевым начальным условиям.

```python
r_p = np.sqrt(a2)                          # радиус полюсов
phi_p = np.arccos(-a1 / (2 * r_p))         # аргумент полюса
print(f'r* = {r_p:.5f}, phi* = {phi_p:.5f} рад = {phi_p/np.pi:.5f}*pi')

def h_analytic(n):
    n = np.asarray(n, dtype=float)
    u1 = lambda m: (m >= 0).astype(float)  # цифровой единичный скачок
    s = np.sin(phi_p)
    return (b0 * r_p**n * np.sin((n + 1) * phi_p) * u1(n)
            + b1 * r_p**(n - 1) * np.sin(n * phi_p) * u1(n - 1)
            + b2 * r_p**(n - 2) * np.sin((n - 1) * phi_p) * u1(n - 2)) / s

h_an = h_analytic(n1)
print('max |h_РУ - h_аналит| =', np.max(np.abs(h - h_an)))

# Огибающая ИХ определяется множителем r*^n; сумма |h| на длинном интервале
h_long = diff_eq(b, a, (np.arange(400) == 0).astype(float))
print(f'r*^N1 = {r_p**N1:.4f};  |h(N1-1)| = {abs(h[-1]):.4f};  sum|h(n)|, n<400: {np.abs(h_long).sum():.4f}')

# Числовая запись формулы с округленными r* = 0.8258, phi* = 0.3043*pi
g = lambda m: np.where(m >= 0, 0.8258**m * np.sin((m + 1) * 0.3043 * np.pi) / np.sin(0.3043 * np.pi), 0.0)
h_num = 0.64 * g(n1) + 0.708352 * g(n1 - 1) + 0.768 * g(n1 - 2)
print(f'max |h_РУ - h_числ.формула| = {np.max(np.abs(h - h_num)):.1e} (следствие округления r* и phi*)')
```

Значения ИХ, вычисленные по разностному уравнению и по аналитической формуле, совпадают с точностью до погрешности вычислений с плавающей точкой.

Для варианта 7 ($r_* = 0.8258$, $\varphi_* = 0.3043\pi$, $\sin\varphi_* = 0.8169$) аналитическая формула ИХ в числовом виде:
$$ h(n) = 0.64\,g(n) + 0.708352\,g(n-1) + 0.768\,g(n-2),\qquad g(n) = 0.8258^{\,n}\,\frac{\sin\big((n+1)\,0.3043\pi\big)}{\sin(0.3043\pi)}\,u_1(n). $$
При округленных до четырех знаков $r_*$ и $\varphi_*$ она отличается от точного расчета не более чем на $4 \cdot 10^{-4}$. Первые отсчеты легко проверить вручную: $h(0) = b_0 = 0.64$, $h(1) = b_1 - a_1 h(0) = 1.3181$, $h(2) = b_2 - a_1 h(1) - a_2 h(0) = 1.5874$.

**Длина ИХ.** Звено рекурсивное ($a_1, a_2 \neq 0$), поэтому ИХ теоретически имеет **бесконечную длину** (БИХ-система): $h(n) \neq 0$ сколь угодно далеко. Длина $N_1 = 27$ - лишь искусственно выбранное число отсчетов для расчета и графика. ИХ представляет собой затухающее колебание с огибающей $\sim r_*^n = 0.826^n$ и нормированной частотой колебаний $\varphi_* \approx 0.304\pi$ (период $\approx$ $2\pi/\varphi_* \approx 6.6$ отсчета). К $n = N_1 - 1 = 26$ значения ИХ уменьшились примерно в 200 раз по сравнению с максимумом ($|h(26)| \approx 0.008$), поэтому практически ИХ можно считать затухшей, хотя формально она не обрывается.

#### 2. Реакция по формуле свертки

Воздействие - дискретный прямоугольный импульс длины $N_2 = 37$: $x(n) = 1$ при $0 \le n < int(N_2/2)$ и $x(n) = 0$ при $int(N_2/2) \le n \le N_2 - 1$. Функция $int(t)$ - округление до ближайшего целого $int(37/2) = int(18.5) = 19$.

Воздействие $x(n)$ - дискретный прямоугольный импульс: $$x(n)=\begin{cases}1, \ 0 \leq n < int(N_2/2) \\ 0, \ int(N_2/2) \leq n \leq N_2-1 \end{cases}$$

```python
n2 = np.arange(N2)
L_imp = r_int(N2 / 2)                       # int(18.5) = 19
x = (n2 < L_imp).astype(float)              # прямоугольный импульс

def conv(x, h):
    # Линейная свертка y(n) = sum_m h(m) x(n-m); длина len(x) + len(h) - 1
    y = np.zeros(len(x) + len(h) - 1)
    for n in range(len(y)):
        y[n] = sum(h[m] * x[n - m] for m in range(len(h)) if 0 <= n - m < len(x))
    return y

y_full = conv(x, h)
y1 = y_full[:N2]                            # ограничение длиной воздействия
print(f'Длина импульса (число единичных отсчетов): {int(x.sum())}; длина последовательности x: {len(x)}')
print(f'Длина полной свертки: {len(y_full)} = N2 + N1 - 1 = {N2 + N1 - 1}')
print('Проверка с numpy.convolve:', np.allclose(y_full, np.convolve(x, h)))

fig, axs = plt.subplots(1, 3, figsize=(14, 3.6))
stem(axs[0], n2, x); axs[0].set(title='Воздействие x(n)', xlabel='n', ylabel='x(n)')
stem(axs[1], np.arange(len(y_full)), y_full, color='C2')
axs[1].axvline(N2 - 0.5, color='r', ls='--', lw=1, label='граница N2')
axs[1].set(title=f'Полная свертка, длина {len(y_full)}', xlabel='n'); axs[1].legend()
stem(axs[2], n2, y1, color='C1'); axs[2].set(title=f'Реакция y1(n), ограниченная до N2 = {N2}', xlabel='n', ylabel='y1(n)')
plt.tight_layout(); plt.show()
```

**Пояснения.**
- Длина последовательности воздействия $x(n)$ равна $N_2 = 37$ отсчетов, а длина самого импульса (ненулевой части) - $int(N_2/2) = 19$ отсчетов ($n = 0 \dots 18$).
- Аналитически длина линейной свертки последовательностей длины $N_2$ и $N_1$ равна $N_2 + N_1 - 1 = 37 + 27 - 1 = 63$ (массив полной свертки - $n = 0 \dots 62$). По графику ненулевые отсчеты занимают $n = 0 \dots 44$, т. е. $45 = 19 + 27 - 1$ отсчетов: нулевая часть воздействия ($n = 19 \dots 36$) вклада в свертку не дает, поэтому фактическая длина определяется длиной самого импульса.
- Свертку ограничивают длиной воздействия $N_2$, чтобы реакция была определена на той же сетке $n = 0 \dots N_2-1$, что и воздействие, и ее можно было отсчет в отсчет сравнить с реакцией по разностному уравнению (п. 3), которое дает ровно $N_2$ отсчетов. Кроме того, хвост полной свертки ($n \ge N_1$) вычислен по усеченной ИХ и не совпадает с истинной реакцией БИХ-звена, поэтому дальний хвост не несет достоверной информации.

#### 4. Нули и полюсы передаточной функции

Нули - корни числителя $b_0 z^2 + b_1 z + b_2 = 0$, полюсы - корни знаменателя $z^2 + a_1 z + a_2 = 0$.

```python
zeros = np.roots(b)      # корни b0 z^2 + b1 z + b2
poles = np.roots(a)      # корни z^2 + a1 z + a2

def describe(name, v):
    for i, c in enumerate(v, 1):
        print(f'{name}{i}: алгебр. {c.real:+.5f} {c.imag:+.5f}j  |  показат. {abs(c):.5f}*exp(j*{np.angle(c):+.5f}) '
              f'= {abs(c):.5f}*exp(j*{np.angle(c)/np.pi:+.5f}*pi)')
describe('z0_', zeros); describe('z*_', poles)

# Связь форм и коэффициентов: r^2 = коэффициент при z^0, -2 r cos(phi) = коэффициент при z^1
r_z, phi_z = abs(zeros[0]), abs(np.angle(zeros[0]))
print(f'\nПроверка: r_z^2 = {r_z**2:.5f} vs b2/b0 = {b2/b0:.5f};  -2 r_z cos(phi_z) = {-2*r_z*np.cos(phi_z):.5f} vs b1/b0 = {b1/b0:.5f}')
print(f'Проверка: r*^2 = {r_p**2:.5f} vs a2 = {a2:.5f};  -2 r* cos(phi*) = {-2*r_p*np.cos(phi_p):.5f} vs a1 = {a1:.5f}')
```

**Связь алгебраической и показательной форм.** Комплексное число $z = \xi + j\eta$ (алгебраическая форма) записывается как $z = r e^{j\varphi}$ (показательная форма), где $r = |z| = \sqrt{\xi^2 + \eta^2}$, $\varphi = \arg z = \mathrm{atan2}(\eta, \xi)$; обратный переход: $\xi = r\cos\varphi$, $\eta = r\sin\varphi$. Так как коэффициенты ПФ вещественны, нули и полюсы образуют комплексно-сопряженные пары (одинаковые модули, аргументы противоположного знака), а модули и аргументы однозначно связаны с коэффициентами: $r^2 = b_2/b_0$, $-2r\cos\varphi = b_1/b_0$ для нулей и $r_*^2 = a_2$, $-2r_*\cos\varphi_* = a_1$ для полюсов.

Получено (аргументы выражены относительно $\pi$):

| | Алгебраическая форма | Показательная форма |
|:--|:--|:--|
| Нули $z_{\circ 1,2}$ | $-0.5534 \pm 0.9454j$ | $1.0954\, e^{\pm j\,0.6686\pi}$ |
| Полюсы $z_{*1,2}$ | $0.4764 \pm 0.6746j$ | $0.8258\, e^{\pm j\,0.3043\pi}$ |

**ПФ в виде произведения простейших множителей:**
$$ H(z) = b_0\,\frac{(1 - z_{\circ 1} z^{-1})(1 - z_{\circ 2} z^{-1})}{(1 - z_{*1} z^{-1})(1 - z_{*2} z^{-1})} = 0.64\,\frac{\big(1 - 1.0954\,e^{j0.6686\pi} z^{-1}\big)\big(1 - 1.0954\,e^{-j0.6686\pi} z^{-1}\big)}{\big(1 - 0.8258\,e^{j0.3043\pi} z^{-1}\big)\big(1 - 0.8258\,e^{-j0.3043\pi} z^{-1}\big)}. $$

```python
# Проверка разложения: значение ПФ в произвольной точке двумя способами
z_test = 0.7 + 0.3j
H_poly = np.polyval(b, z_test) / np.polyval(a, z_test)
H_fact = b0 * np.prod(1 - zeros / z_test) / np.prod(1 - poles / z_test)
print('H(z) по коэффициентам:', np.round(H_poly, 10), '\nH(z) по разложению:  ', np.round(H_fact, 10))
```

#### 6. АЧХ и ФЧХ в шкале нормированных частот

Частотная характеристика вычисляется подстановкой $z = e^{j\hat{\omega}}$ в ПФ: $H(e^{j\hat{\omega}}) = \dfrac{b_0 + b_1 e^{-j\hat{\omega}} + b_2 e^{-j2\hat{\omega}}}{1 + a_1 e^{-j\hat{\omega}} + a_2 e^{-j2\hat{\omega}}}$.

```python
def freq_resp(b, a, w):
    # H(e^{jw}) = sum b_i e^{-jwi} / sum a_k e^{-jwk}
    E = lambda c: sum(ci * np.exp(-1j * w * i) for i, ci in enumerate(c))
    return E(b) / E(a)

w = np.linspace(0, np.pi, 2001)          # основная полоса [0, pi]
Hw = freq_resp(b, a, w)
A_w, phi_w = np.abs(Hw), np.angle(Hw)

i_max, i_min = A_w.argmax(), A_w.argmin()
fig, axs = plt.subplots(1, 2, figsize=(14, 3.8))
axs[0].plot(w / np.pi, A_w)
axs[0].plot(w[i_max] / np.pi, A_w[i_max], 'r^'); axs[0].plot(w[i_min] / np.pi, A_w[i_min], 'gv')
for v, c in [(phi_p, 'C3'), (phi_z, 'C0')]:
    axs[0].axvline(v / np.pi, color=c, ls='--', lw=1)
axs[0].set(title=r'АЧХ $A(\hat{\omega})$', xlabel=r'$\hat{\omega}/\pi$', ylabel=r'$A(\hat{\omega})$')
axs[0].legend([r'$A(\hat{\omega})$', 'максимум', 'минимум', 'arg полюса', 'arg нуля'], fontsize=8)
axs[1].plot(w / np.pi, phi_w / np.pi, label=r'$\varphi(\hat{\omega})$ (главное значение)')
axs[1].plot(w / np.pi, np.unwrap(phi_w) / np.pi, '--', label=r'$\varphi(\hat{\omega})$ без скачков $2\pi$')
axs[1].set(title=r'ФЧХ $\varphi(\hat{\omega})$', xlabel=r'$\hat{\omega}/\pi$', ylabel=r'$\varphi/\pi$'); axs[1].legend(fontsize=8)
plt.tight_layout(); plt.show()

jumps = np.where(np.abs(np.diff(phi_w)) > np.pi)[0]
print(f'A(0) = {A_w[0]:.4f};  A(pi) = {A_w[-1]:.4f};  A(pi)/A(0) = {A_w[-1]/A_w[0]:.3f}')
print(f'максимум A = {A_w[i_max]:.4f} при w = {w[i_max]/np.pi:.4f}*pi (arg полюса {phi_p/np.pi:.4f}*pi)')
print(f'минимум  A = {A_w[i_min]:.4f} при w = {w[i_min]/np.pi:.4f}*pi (arg нуля {phi_z/np.pi:.4f}*pi)')
print('скачки ФЧХ:', [f'w = {w[k]/np.pi:.4f}*pi, dphi = {(phi_w[k+1]-phi_w[k])/np.pi:+.3f}*pi, A = {A_w[k]:.3f}' for k in jumps])
```

**Пояснения.**
- **Границы основной полосы частот** в шкале нормированных частот: $\hat{\omega} \in [0;\, \pi]$. Вне ее частотные характеристики повторяются (период $2\pi$), а на $[-\pi; 0]$ АЧХ симметрична (четная), ФЧХ - антисимметрична (нечетная), т. к. ИХ вещественна.
- **Соответствие карте нулей и полюсов.** Значение АЧХ на частоте $\hat{\omega}$ пропорционально отношению произведения расстояний от точки $e^{j\hat{\omega}}$ на единичной окружности до нулей к произведению расстояний до полюсов. Полюс вблизи окружности ($r_* = 0.826$) под углом $0.3043\pi$ дает **резонансный максимум** $A \approx 5.99$ при $\hat{\omega} \approx 0.291\pi$; нуль под углом $0.6686\pi$ дает **провал** $A \approx 0.060$ при $\hat{\omega} \approx 0.6715\pi$. Поскольку нули лежат не на окружности, АЧХ нигде не обращается в ноль.
- **Скачок ФЧХ на $\pi$** возникает, когда АЧХ проходит через ноль (нуль ПФ на единичной окружности), и соответствует значению $A = 0$. В данном варианте нули не лежат на окружности ($|z_\circ| \neq 1$), поэтому **скачков на $\pi$ нет**. Единственный скачок при $\hat{\omega} \approx 0.55\pi$ равен $2\pi$ и является следствием приведения фазы к главному значению $(-\pi;\, \pi]$ (при $A \approx 0.40 \neq 0$); пунктир - та же ФЧХ без этого скачка.
- **Подавленные составляющие.** $A(0) \approx 2.90$, $A(\pi) \approx 0.27$, т. е. высокочастотные составляющие (выше $\approx$ $0.5\pi$, особенно в окрестности $0.67\pi$) ослабляются примерно в 10 раз сильнее низкочастотных. Следовательно, в реакции преимущественно подавлены **высокие** частоты: это видно и по реакции на прямоугольный импульс - резкие фронты (высокочастотные составляющие) сглажены, а постоянная составляющая усилена в $\approx$ 2.9 раза.

#### 1. Последовательность на периоде N

```python
n = np.arange(N)
xs = lambda t: A1 * np.cos(2 * np.pi * f1 * t + np.pi / 4) + A2 * np.cos(2 * np.pi * f2 * t + np.pi / 8)
xn = xs(n * T)                       # x(nT)

fig, axs = plt.subplots(1, 2, figsize=(14, 3.6))
stem(axs[0], n * T * 1e3, xn); axs[0].set(title='x(nT): шкала дискретного времени', xlabel='nT, мс', ylabel='x(nT)')
stem(axs[1], n, xn, color='C1'); axs[1].set(title='x(n): шкала дискретного нормированного времени', xlabel='n', ylabel='x(n)')
plt.tight_layout(); plt.show()
```

Графики в двух шкалах совпадают по форме: дискретное нормированное время $n = nT/T$ - это номер отсчета, а дискретное время $nT$ получается умножением номера на период дискретизации $T \approx 166.7$ мкс (интервал $[0;\,(N-1)T] = [0;\, 10.5]$ мс). Так как $f_1/f_d = 1/8$ и $f_2/f_d = 3/16$, наименьший период последовательности равен 16 отсчетам, и на интервале $N = 64$ укладывается ровно 4 периода (8 периодов первой и 12 периодов второй гармоники).

#### 2. ОДПФ от ДПФ

```python
k = np.arange(N)
W = np.exp(-2j * np.pi * np.outer(k, n) / N)     # матрица ДПФ W_N^{nk}
X = W @ xn                                        # ДПФ по определению
x_rec = (np.conj(W) @ X) / N                      # ОДПФ по определению
print('ДПФ по определению совпадает с np.fft.fft:', np.allclose(X, np.fft.fft(xn)))
print(f'max|x(n) - ОДПФ(ДПФ(x))| = {np.max(np.abs(xn - x_rec)):.2e};  max|Im| = {np.max(np.abs(x_rec.imag)):.2e}')

fig, ax = plt.subplots()
stem(ax, n, x_rec.real, color='C2', label='ОДПФ{ДПФ{x(n)}}')
ax.plot(n, xn, 'k.', ms=4, label='исходная x(n)')
ax.set(title='Последовательность, полученная ОДПФ', xlabel='n', ylabel='x(n)'); ax.legend(fontsize=8)
plt.show()
```

Последовательность, полученная ОДПФ от ДПФ, совпадает с исходной (расхождение порядка $10^{-14}$ обусловлено арифметикой с плавающей точкой, мнимая часть пренебрежимо мала): ДПФ и ОДПФ - пара взаимно однозначных преобразований.

#### 3. Амплитудный и фазовый спектры

```python
A_k = np.abs(X)
tol = 1e-9 * A_k.max()
phi_k = np.where(A_k > tol, np.angle(X), 0.0)     # фаза отсчетов с нулевой амплитудой не определена -> 0
fk = k * fd2 / N                                    # абсолютные частоты, Гц

fig, axs = plt.subplots(2, 2, figsize=(14, 6.4))
stem(axs[0, 0], k, A_k); w_axis(axs[0, 0], N, 'Амплитудный спектр |X(k)|'); axs[0, 0].set_ylabel('|X(k)|')
stem(axs[1, 0], k, phi_k / np.pi, color='C1'); w_axis(axs[1, 0], N, 'Фазовый спектр arg X(k)'); axs[1, 0].set_ylabel(r'$\varphi/\pi$')
stem(axs[0, 1], fk, A_k); axs[0, 1].set(title='Амплитудный спектр |X(f)|', xlabel='f, Гц', ylabel='|X|')
stem(axs[1, 1], fk, phi_k / np.pi, color='C1'); axs[1, 1].set(title='Фазовый спектр arg X(f)', xlabel='f, Гц', ylabel=r'$\varphi/\pi$')
plt.tight_layout(); plt.show()

nz = np.where(A_k > tol)[0]
for kk in nz:
    print(f'k = {kk:2d}: f = {fk[kk]:7.2f} Гц, w = {2*kk/N:.4f}*pi, |X| = {A_k[kk]:.2f}, 2|X|/N = {2*A_k[kk]/N:.3f}, arg = {phi_k[kk]/np.pi:+.4f}*pi')
```

**Пояснения.**
- **Связь дискретного нормированного времени с дискретным временем:** $n = nT/T$, т. е. номер отсчета получается делением момента времени на период дискретизации ($T = 1/6000$ с). Шкала $n$ безразмерна и не зависит от $f_d$.
- **Связь частоты $f$ (Гц) с дискретной нормированной частотой** (на левых графиках нижняя шкала - номер отсчета $k$, верхняя - $\hat{\omega}_k/\pi$): $k$-му отсчету ДПФ соответствуют нормированная частота $\hat{\omega}_k = 2\pi k/N$ (или $\hat{f}_k = k/N$) и абсолютная частота $f_k = k\,f_d/N = k\,\Delta f$, где $\Delta f = f_d/N = 1/(NT) = 93.75$ Гц - шаг по частоте. Отсчет $k = N/2 = 32$ соответствует $f_d/2 = 3000$ Гц, отсчет $k = N$ - частоте $f_d$.
- **Вид амплитудного спектра:** спектр линейчатый, ненулевые только 4 отсчета: $k = 8$ ($f_1 = 750$ Гц) и $k = 12$ ($f_2 = 1125$ Гц), а также их зеркальные отражения $k = N-8 = 56$ и $k = N-12 = 52$ (частоты $f_d - f$). Значения $|X(k)| = N A_i / 2$: 34.24 и 68.48, т. е. амплитуды гармоник восстанавливаются как $2|X(k)|/N$ = 1.07 и 2.14. Утечки спектра нет, так как на интервале $N$ укладывается целое число периодов каждой гармоники. Амплитудный спектр - четная функция относительно $k = N/2$ ($|X(k)| = |X(N-k)|$).
- **Вид фазового спектра:** $\arg X(8) = \pi/4$ и $\arg X(12) = \pi/8$ - начальные фазы гармоник; на зеркальных отсчетах фазы противоположны ($-\pi/4$, $-\pi/8$) - фазовый спектр нечетен. На отсчетах с нулевой амплитудой фаза не определена (численно - шум округления), поэтому принята равной нулю.

#### 4. ДПФ конечной последовательности

```python
x_fin = xn.copy()                         # конечная последовательность длины N (один отрезок, вне его - нули)
X_fin = np.fft.fft(x_fin)                 # N-точечное ДПФ конечной последовательности
print('|ДПФ конечной| совпадает с амплитудным спектром периодической:', np.allclose(np.abs(X_fin), A_k))

# Непрерывный спектр конечной последовательности |X(exp(jw))| (ДПФ с дополнением нулями до 4096 точек)
M = 4096
X_cont = np.fft.fft(x_fin, M)
w_c = 2 * np.pi * np.arange(M) / M

fig, ax = plt.subplots(figsize=(14, 4))
ax.plot(w_c / (2 * np.pi) * N, np.abs(X_cont), color='0.7', lw=1, label=r'$|X(e^{j\hat{\omega}})|$ конечной посл. (непрерывный)')
stem(ax, k, np.abs(X_fin), color='C0', ms=6, label='|X(k)| конечной последовательности')
ax.plot(k, A_k, 'r_', ms=14, mew=2, label='амплитудный спектр периодической')
w_axis(ax, N, 'Модуль ДПФ конечной последовательности и амплитудный спектр периодической')
ax.set_ylabel('модуль'); ax.legend(fontsize=8)
plt.show()
```

**Пояснения.** Модуль $N$-точечного ДПФ конечной последовательности длины $N$ **совпадает** с амплитудным спектром периодической последовательности с периодом $N$: периодическая последовательность - периодическое продолжение конечной, а ее ДПФ вычисляется по одному периоду по той же формуле. Серая кривая - непрерывный спектр $|X(e^{j\hat{\omega}})|$ конечной последовательности; отсчеты ДПФ являются его дискретизацией с шагом $2\pi/N$ (теорема Котельникова в частотной области). Для конечной последовательности непрерывный спектр имеет боковые лепестки, но все отсчеты ДПФ, кроме $k = 8, 12, 52, 56$, попадают в его нули, так как частоты гармоник кратны $\Delta f$. Для периодической же последовательности спектр по своей природе дискретный (линейчатый) и определен только на частотах $f_k$.

#### 5. Восстановление аналогового сигнала по отсчетам ДПФ

Используется формула $x(t) = \dfrac{1}{N}\sum_{k=-N/2}^{N/2-1} X_a(k) \exp\Big(j\dfrac{2\pi}{NT}kt\Big)$, где $X_a(k) = X(N+k)$ при $-N/2 \le k \le -1$ и $X_a(k) = X(k)$ при $0 \le k \le N/2-1$; $t \in [0;\, (N-1)T]$ с шагом $\Delta t = 0.25T$.

```python
ka = np.arange(-N // 2, N // 2)                       # k = -N/2 ... N/2-1
Xa = np.where(ka < 0, X[(N + ka) % N], X[ka % N])     # X_a(k): X(N+k) при k<0, X(k) при k>=0
t = np.arange(0, (N - 1) * T + 1e-12, 0.25 * T)       # t in [0; (N-1)T], шаг 0.25T
x_t = (np.exp(1j * 2 * np.pi / (N * T) * np.outer(t, ka)) @ Xa) / N
x_orig = xs(t)                                        # исходный аналоговый сигнал x(t) = x(nT)|_{nT=t}
print(f'Отсчетов времени: {len(t)}; max|Im x(t)| = {np.max(np.abs(x_t.imag)):.2e}; '
      f'max|x_восст(t) - x_исх(t)| = {np.max(np.abs(x_t.real - x_orig)):.2e}')

fig, axs = plt.subplots(3, 2, figsize=(14, 9.5))
stem(axs[0, 0], n, xn); axs[0, 0].set(title='Периодическая последовательность x(n)', xlabel='n')
stem(axs[0, 1], k, A_k); w_axis(axs[0, 1], N, 'Модуль ДПФ |X(k)|')
axs[1, 0].plot(t * 1e3, x_t.real, color='C2'); axs[1, 0].set(title='Восстановленный сигнал x(t)', xlabel='t, мс')
stem(axs[1, 1], ka / (N * T), np.abs(Xa), color='C2'); axs[1, 1].set(title=r'Амплитудный спектр восстановленного сигнала $|X_a(k)|$', xlabel='f = k/(NT), Гц')
axs[2, 0].plot(t * 1e3, x_orig, color='C3'); axs[2, 0].set(title='Исходный аналоговый сигнал x(t)', xlabel='t, мс')
axs[2, 1].plot(t * 1e3, x_orig, color='C3', lw=2.5, alpha=0.6, label='исходный')
axs[2, 1].plot(t * 1e3, x_t.real, 'k--', lw=1, label='восстановленный')
axs[2, 1].plot(n * T * 1e3, xn, 'o', ms=3, color='C0', label='отсчеты x(nT)')
axs[2, 1].set(title='Сравнение (фрагмент 0-3 мс)', xlabel='t, мс', xlim=(0, 3)); axs[2, 1].legend(fontsize=8)
plt.tight_layout(); plt.show()
```

#### 6. Связь ДПФ со спектром аналогового сигнала и сравнение сигналов

- **Связь модуля ДПФ со спектром аналогового сигнала.** Отсчеты $X_a(k)$ - это коэффициенты ряда Фурье периодического аналогового сигнала с периодом $NT$ (умноженные на $N$): $k$-й отсчет соответствует гармонике с частотой $f_k = k/(NT)$, ее амплитуда равна $2|X(k)|/N$. Перестановка $X(k) \to X_a(k)$ переносит отсчеты верхней половины ($k \ge N/2$, частоты $f_d - f$) на отрицательные частоты, т. е. переводит спектр из диапазона $[0;\, f_d)$ в симметричный диапазон $[-f_d/2;\, f_d/2)$. Амплитудный спектр восстановленного сигнала содержит линии на $\pm 750$ Гц и $\pm 1125$ Гц - это и есть линейчатый спектр исходного аналогового сигнала. Связь однозначна при выполнении условия теоремы Котельникова: $f_d = 6000 \ge 2 f_B = 2250$ Гц, т. е. все составляющие лежат в основной полосе и наложения спектров нет.
- **Сравнение восстановленного и исходного сигналов.** Визуально восстановленный и исходный аналоговые сигналы совпадают; максимальное расхождение порядка $10^{-14}$ (погрешность вычислений). Точное восстановление объясняется тем, что сигнал ограничен по спектру частотами ниже $f_d/2$ и на интервале $NT$ укладывается целое число периодов гармоник. Между отсчетами $x(nT)$ восстановленная кривая проходит гладко и сохраняет форму исходного сигнала.

#### 7. Равенство Парсеваля

```python
E_time = np.sum(np.abs(xn) ** 2)
E_freq = np.sum(np.abs(X) ** 2) / N
print(f'sum |x(n)|^2         = {E_time:.6f}')
print(f'(1/N) sum |X(k)|^2   = {E_freq:.6f}')
print(f'разность             = {abs(E_time - E_freq):.2e}')
print(f'средняя мощность E/N = {E_time / N:.6f};  (A1^2 + A2^2)/2 = {(A1**2 + A2**2) / 2:.6f}')
```

Равенство Парсеваля выполняется: $\sum |x(n)|^2 = \frac{1}{N}\sum |X(k)|^2 = 183.184$. **Смысл:** энергия последовательности на периоде, вычисленная во временной области, равна энергии, вычисленной в частотной области по ее ДПФ, т. е. ДПФ (с точностью до множителя $1/N$) сохраняет энергию, а $|X(k)|^2/N$ показывает распределение энергии по частотам. Здесь вся энергия сосредоточена в четырех отсчетах спектра; средняя мощность $183.184/64 = 2.8623$ равна сумме мощностей гармоник $(A_1^2 + A_2^2)/2$.
