## Основные теоретические положения

#### 1. Дискретные сигналы и представление по дельта-импульсам
Дискретный сигнал представляет собой последовательность отсчетов (чисел), полученных в результате измерения непрерывного сигнала с периодом дискретизации $T$. 

Базовым элементом пространства дискретных сигналов является единичный дельта-импульс Кронекера
$$\delta(n) = \begin{cases} 1, & n = 0, \\ 0, & n \neq 0. \end{cases}$$

Любой дискретный сигнал $x(n)$ длины $N$ раскладывается по базису сдвинутых единичных импульсов:
$$x(n) = \sum_{k=0}^{N-1} x(k) \cdot \delta(n - k)$$

#### 2. Линейные дискретные системы, импульсная характеристика и свёртка
Линейная дискретная стационарная система (ЛДС) преобразует входную последовательность $x(n)$ в выходную последовательность (реакцию) $y(n)$. 

Реакция системы на единичный импульс $\delta(n)$ при нулевых начальных условиях называется импульсной характеристикой $h(n)$.

По характеру импульсной характеристики системы делятся на два класса:
* Системы с конечной импульсной характеристикой (КИХ) - нерекурсивные системы, импульсная характеристика которых содержит конечное число отсчетов.
* Системы с бесконечной импульсной характеристикой (БИХ) - рекурсивные системы, имеющие теоретически бесконечную импульсную характеристику из-за наличия обратных связей по выходу.

Реакция линейной системы $y(n)$ на произвольное воздействие $x(n)$ вычисляется с помощью свёртки:
$$y(n) = x(n) \ast h(n) = \sum_{m} x(m) h(n - m)$$

Если последовательность $h(n)$ имеет длину $N_1$ отсчетов, а воздействие $x(n)$ - длину $N_2$ отсчетов, то аналитическая длина свёртки составляет:
$$L = N_1 + N_2 - 1$$

#### 3. Разностные уравнения рекурсивного звена 2-го порядка
Временное описание рекурсивного звена 2-го порядка задается линейным разностным уравнением с коэффициентами $b_i$ и $a_k$:
$$y(n) = b_0 x(n) + b_1 x(n-1) + b_2 x(n-2) - a_1 y(n-1) - a_2 y(n-2)$$

#### 4. Z-преобразование, передаточная функция и устойчивость
Z-преобразование переводит дискретную последовательность $x(n)$ из временной области в комплексную Z-область:
$$X(z) = \sum_{n=-\infty}^{\infty} x(n) z^{-n}$$
где $z$ - комплексная переменная.

Операции задержки сигнала на $k$ отсчетов $x(n-k)$ соответствует умножение его Z-образа на $z^{-k}$. 

Передаточная функция $H(z)$ представляет собой отношение Z-изображения выхода $Y(z)$ к Z-изображению входа $X(z)$ при нулевых начальных условиях:
$$H(z) = \frac{Y(z)}{X(z)} = \frac{b_0 + b_1 z^{-1} + b_2 z^{-2}}{1 + a_1 z^{-1} + a_2 z^{-2}} = \frac{b_0 z^2 + b_1 z + b_2}{z^2 + a_1 z + a_2}$$

* Нули ($z_k$) - корни полинома числителя: $b_0 z^2 + b_1 z + b_2 = 0$.
* Полюсы ($p_k$) - корни полинома знаменателя: $z^2 + a_1 z + a_2 = 0$.

Нули и полюсы записываются в алгебраической ($z = a + jb$) и показательной ($z = R e^{j\theta}$) формах, где $R = \sqrt{a^2 + b^2}$ — модуль, а $\theta = \operatorname{atan2}(b, a)$ - аргумент (угол).

Критерий устойчивости: Рекурсивная дискретная система является устойчивой тогда и только тогда, когда все полюсы её передаточной функции расположены строго внутри единичного круга на Z-плоскости:
$$|p_k| < 1$$

#### 5. Частотные характеристики (АЧХ и ФЧХ)
Подстановка $z = e^{j\hat{\omega}}$ в передаточную функцию $H(z)$ задает комплексную частотную характеристику $H(e^{j\hat{\omega}}) = H_r(\hat{\omega}) + j H_i(\hat{\omega})$:

* Амплитудно-частотная характеристика (АЧХ):
  $$A(\hat{\omega}) = |H(e^{j\hat{\omega}})| = \sqrt{H_r(\hat{\omega})^2 + H_i(\hat{\omega})^2}$$
* Фазочастотная характеристика (ФЧХ):
  $$\Phi(\hat{\omega}) = \arg H(e^{j\hat{\omega}}) = \operatorname{atan2}(H_i(\hat{\omega}), H_r(\hat{\omega}))$$

Связь между нормированной частотой $\hat{\omega}$ (рад/отсчет) и абсолютной частотой $f$ (Гц) при частоте дискретизации $f_д$ описывается формулой:
$$\hat{\omega} = \frac{2\pi f}{f_д}$$
Граница основной полосы частот (частота Найквиста) в нормированной шкале равна $\hat{\omega} = \pi$, а в шкале реальных частот - $f = \frac{f_д}{2}$

#### 6. Дискретное преобразование Фурье (ДПФ) и равенство Парсеваля
Дискретное преобразование Фурье сопоставляет дискретному сигналу $x(n)$ длины $N$ его комплексный спектр $X(k)$:
$$X(k) = \sum_{n=0}^{N-1} x(n) e^{-j \frac{2\pi}{N} kn}, \quad k = 0, 1, \dots, N-1$$

Обратное дискретное преобразование Фурье (ОДПФ) восстанавливает дискретный сигнал по его спектру:
$$x(n) = \frac{1}{N} \sum_{k=0}^{N-1} X(k) e^{j \frac{2\pi}{N} kn}, \quad n = 0, 1, \dots, N-1$$

* Амплитудный спектр: $|X(k)| = \sqrt{\operatorname{Re}(X(k))^2 + \operatorname{Im}(X(k))^2}$.
* Фазовый спектр: $\arg X(k) = \operatorname{atan2}(\operatorname{Im}(X(k)), \operatorname{Re}(X(k)))$.

Равенство Парсеваля утверждает равенство энергии сигнала во временной и частотной областях:
$$\sum_{n=0}^{N-1} |x(n)|^2 = \frac{1}{N} \sum_{k=0}^{N-1} |X(k)|^2$$

#### 7. Теорема отсчетов Котельникова
Согласно теореме Котельникова (Найквиста - Шеннона), аналоговый сигнал с ограниченным спектром может быть восстановлен по своим отсчетам однозначно и без потерь, если частота дискретизации $f_д = \frac{1}{T}$ строго превышает удвоенную верхнюю частоту спектра $f_{\text{макс}}$:
$$f_д > 2 f_{\text{макс}}$$

### Пункт 1. Импульсная характеристика

Рекурсивное звено второго порядка описывается разностным уравнением
$$y(n) = b_0 x(n) + b_1 x(n-1) + b_2 x(n-2) - a_1 y(n-1) - a_2 y(n-2)$$

При воздействии $x(n) = \delta(n)$ и нулевых начальных условиях $y(-1) = y(-2) = 0$ реакция совпадает с импульсной характеристикой $h(n)$. Аналитическая формула имеет вид
$$h(0) = b_0, \quad h(1) = b_1 - a_1 h(0), \quad h(2) = b_2 - a_1 h(1) - a_2 h(0),$$
$$h(n) = -a_1 h(n-1) - a_2 h(n-2), \quad n \geqslant 3.$$


```python
h = np.zeros(N1)
h[0] = b0
h[1] = b1 - a1 * h[0]
h[2] = b2 - a1 * h[1] - a2 * h[0]
for n in range(3, N1):
    h[n] = -a1 * h[n - 1] - a2 * h[n - 2]
```

```python
print(len(h))
print(h)
```

```python
plt.figure()
plt.stem(range(len(h)), h)
plt.xlabel("n")
plt.ylabel("h(n)")
plt.grid(True)
plt.show()
```

### Выводы по пункту:
* В действительности длина импульсной характеристики рекурсивного звена бесконечна, потому что каждое значение зависит от предыдущих. Система относится к классу БИХ-систем. График имеет затухающий колебательный характер. К концу интервала усечения характеристика в ноль не обращается, поэтому усечение до $N_1$ отбрасывает остаток бесконечной последовательности.

Реакция вычисляется по формуле линейной свертки усеченной характеристики $h(n)$ с воздействием $x(n)$

### Пункт 3. Реакция по разностному уравнению

Реакция вычисляется напрямую из разностного уравнения при нулевых начальных условиях $y(-1) = y(-2) = 0$
$$y_2(n) = 0.86 x(n) - 1.120236 x(n-1) + 1.204 x(n-2) - 1.2278 y_2(n-1) - 0.748 y_2(n-2)$$

### Пункт 4. Нули и полюсы

Передаточная функция звена имеет вид
$$H(z) = {{b_0 z^2 + b_1 z + b_2} \over {z^2 + a_1 z + a_2}}$$
$$H(z) = {{0.86 z^2 - 1.120236 z + 1.204} \over {z^2 + 1.2278 z + 0.748}}$$
Нули являются корнями числителя, полюсы являются корнями знаменателя
$$b_0 z^2 + b_1 z + b_2 = 0, \quad z^2 + a_1 z + a_2 = 0$$

Алгебраическая и показательная формы связаны соотношениями $R = \sqrt{a^2 + b^2}$ и $\theta = \mathrm{atan2}(b, a)$
$$a + jb = R e^{j \theta}$$

Передаточная функция представляется в виде произведения простейших множителей
$$H(z) = b_0 \frac{(z - q_1)(z - q_2)}{(z - p_1)(z - p_2)}$$


```python
q = np.roots([b0, b1, b2])
p = np.roots([1, a1, a2])
Rq = np.abs(q)
Rp = np.abs(p)
Aq = np.angle(q)
Ap = np.angle(p)
Aq_pi = np.angle(q) / np.pi
Ap_pi = np.angle(p) / np.pi
```

```python
print("q =", q)
print("p =", p)
print("Rq =", Rq)
print("Rp =", Rp)
print("Aq (theta) =", Aq)
print("Ap (theta) =", Ap)
print("Aq (theta/pi) =", Aq_pi)
print("Ap (theta/pi) =", Ap_pi)
```

### Пункт 6. АЧХ и ФЧХ в нормированной шкале

Частотная характеристика получается подстановкой $z = e^{j \hat{\omega}}$ при $\hat{\omega} \in [0, \pi]$
$$H(e^{j \hat{\omega}}) = \frac{b_0 + b_1 e^{-j \hat{\omega}} + b_2 e^{-j 2 \hat{\omega}}}{1 + a_1 e^{-j \hat{\omega}} + a_2 e^{-j 2 \hat{\omega}}}$$
АЧХ равна модулю, ФЧХ равна аргументу
$$A = |H|, \quad \Phi = \arg H$$


```python
w = np.linspace(0, np.pi, 10001)
e1 = np.exp(-1j * w)
e2 = np.exp(-2j * w)
H = (b0 + b1 * e1 + b2 * e2) / (1 + a1 * e1 + a2 * e2)
A = np.abs(H)
Ph = np.angle(H)
```

```python
print(H)
```

```python
print("A0 =", A[0])
print("Api =", A[-1])
print("Amin =", np.min(A))
print("wmin/pi =", w[np.argmin(A)] / np.pi)
print("Amax =", np.max(A))
print("wmax/pi =", w[np.argmax(A)] / np.pi)
```

```python
plt.rc("figure", figsize=(10, 4))
plt.figure()
plt.plot(w, A)
plt.xlabel("w")
plt.ylabel("A")
plt.grid(True)
plt.show()
plt.figure()
plt.plot(w, Ph)
plt.xlabel("w")
plt.ylabel("phi")
plt.grid(True)
plt.show()
```

### Выводы по пункту:

### Часть 2. Пункт 1. Гармонический сигнал

### Часть 2. Пункт 2. Обращение ДПФ

Прямое и обратное ДПФ связаны соотношениями
$$X(k) = \sum_{n = 0}^{N - 1} x(n) e^{-j 2 \pi k n / N}, \quad x(n) = \frac{1}{N} \sum_{k = 0}^{N - 1} X(k) e^{j 2 \pi k n / N}$$


```python
X = np.fft.fft(x)
xr = np.fft.ifft(X)
```

### Часть 2. Пункт 3. Амплитудный и фазовый спектры

Дискретное время равно $t_n = nT$, нормированное время равно $n = t_n / T$. Нормированная частота равна $k / N$, абсолютная частота равна $f = k f_д / N$ с шагом $f_д / N$. Радианная частота равна $\hat{\omega} = 2 \pi f / f_д$, частота Найквиста $f_д / 2$ соответствует $\pi$.

Амплитудный и фазовый спектры определяются модулем и аргументом отсчетов ДПФ
$$|X(k)|, \quad \arg X(k)$$

Для вещественного сигнала модуль симметричен относительно $N / 2$, фаза антисимметрична. При совпадении гармоник с сеткой ДПФ высоты пиков равны $N A / 2$, фазы равны начальным фазам гармоник.


```python
k = np.arange(N)
Ak = np.abs(X)
Pk = np.angle(X)
kn = k / N
fk = k * f_d2 / N
```

```python
print(Ak[8])
print(Ak[12])
print(Pk[8] / np.pi)
print(Pk[12] / np.pi)
```

```python
plt.figure()
plt.stem(kn, Ak)
plt.xlabel("k/N")
plt.ylabel("|X|")
plt.grid(True)
plt.show()
plt.figure()
plt.stem(fk, Ak)
plt.xlabel("f")
plt.ylabel("|X|")
plt.grid(True)
plt.show()
plt.figure()
plt.stem(kn, Pk)
plt.xlabel("k/N")
plt.ylabel("phi")
plt.grid(True)
plt.show()
plt.figure()
plt.stem(fk, Pk)
plt.xlabel("f")
plt.ylabel("phi")
plt.grid(True)
plt.show()
```

* Частота связана с дискретной нормированной частотой соотношением $f = (k / N) f_д$.
* Амплитудный спектр имеет линейчатый вид с пиками $37.76$ на $k = 8$ ($1000$ Гц) и $75.52$ на $k = 12$ ($1500$ Гц), симметричными относительно $N / 2$. ### Часть 2. Пункт 4. Спектр конечной последовательности

ДПФ конечной последовательности длины $N$ вычисляется по той же формуле
$$X(k) = \sum_{n = 0}^{N - 1} x(n) e^{-j 2 \pi k n / N}$$

ДПФ формально продолжает отрезок периодически. Модуль пропорционален длине кадра, физический спектр получается нормировкой $A = 2 |X| / N$.

При целом числе периодов на интервале модуль совпадает с амплитудным спектром периодической последовательности.


```python
Xf = np.fft.fft(x)
Af = np.abs(Xf)
An = 2 * Af / N
```

```python
print(np.max(np.abs(Af - Ak)))
print(An[8])
print(An[12])
```

```python
plt.figure()
plt.stem(kn, Ak)
plt.stem(kn, Af, markerfmt=".")
plt.xlabel("k/N")
plt.ylabel("|X|")
plt.legend(["per", "fin"])
plt.grid(True)
plt.show()
```

### Часть 2. Пункт 5. Восстановление аналогового сигнала

Сигнал восстанавливается по отсчетам ДПФ на интервале $t \in [0, (N - 1) T]$ с шагом $0.25 T$
$$x(t) = \frac{1}{N} \sum_{k = -N / 2}^{N / 2 - 1} X_a(k) e^{j 2 \pi k t / (N T)}$$

$$X_a(k) = X(N + k), \; k < 0, \quad X_a(k) = X(k), \; k \geqslant 0$$


```python
k2 = np.arange(-N // 2, N // 2)  # от -32 до 31
Xa = np.concatenate([X[N // 2 :], X[: N // 2]])
t = np.arange(0, (N - 1) * 4 + 1) / 4 * T
xr2 = np.exp(1j * 2 * np.pi * np.outer(t, k2) / (N * T)).dot(Xa) / N
xo = A1 * np.cos(2 * np.pi * f1 * t + np.pi / 4) + A2 * np.cos(
    2 * np.pi * f2 * t + np.pi / 8
)
```

```python
print(len(t))
print(np.max(np.abs(xr2.real - xo)))
```
