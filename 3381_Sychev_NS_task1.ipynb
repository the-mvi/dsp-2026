## Основные теоретические положения

#### 1 часть
1) Определение импульсной характеристики, передаточной функции, линейное разностное уравнение

Определение. Линейное разностное уравнение - уравнение вида:
$$
y(n) = \sum_{i = 0}^{N - 1} b_i x(n - i) - \sum_{k = 1}^{M - 1} a_k y(n - k),
$$
где порядок разностного уравнения равен $\max(N-1, M-1)$. В работе рассматривается рекурсивное звено 2-го порядка:
$$
y(n) = b_0 x(n) + b_1 x(n-1) + b_2 x(n-2) - a_1 y(n-1) - a_2 y(n-2).
$$

Определение. Импульсная характеристика - реакция системы на цифровой единичный импульс $u_0(n)$ ($u_0(0) = 1$, $u_0(n) = 0$ при $n \ne 0$) при нулевых начальных условиях. Для линейной системы:
$$
h(n) = \sum_{i = 0}^{N - 1} b_i u_{0}(n - i) - \sum_{k = 1}^{M - 1} a_k h(n - k).
$$

Определение. Передаточная функция - импульсная характеристика в Z области:
$$
\mathcal{Z}\{h(n)\} = H(z) = \sum_{n = 0}^{+\infty} h(n) z^{-n}.
$$
Для звена 2-го порядка (по свойству линейности и теореме о задержке):
$$
H(z) = \frac{Y(z)}{X(z)} = \frac{b_0 + b_1 z^{-1} + b_2 z^{-2}}{1 + a_1 z^{-1} + a_2 z^{-2}}.
$$

2) Формула свертки, реакция через свертку, БИХ

Определение. Сверткой двух сигналов называется следующая операция:
$$
y(n) = x(n) * h(n) = \sum_{m = 0}^{+\infty} h(m)x(n - m).
$$
Свертка входного сигнала с импульсной характеристикой системы равна реакции системы. Свертка последовательностей длины $N_h$ и $N_x$ имеет длину $N_h + N_x - 1$.

Определение. Импульсная характеристика системы называется бесконечной (БИХ), если выход системы зависит не только от текущих и прошлых входных данных, но и от предыдущих значений самого выхода.

3) Нули, полюсы ПФ, алгебраическая форма записи, показательная форма записи

Определение. Нуль передаточной функции - значение $z = z_0$, при котором передаточная функция принимает нулевое значение.

Определение. Полюс передаточной функции - значение $z = z_p$ такое, что $\lim_{z \rightarrow z_p} H(z) = \infty$.

Через нули и полюсы передаточная функция записывается как произведение простейших множителей:
$$
H(z) = K\,\frac{\prod_i \left(z - z_0^{(i)}\right)}{\prod_k \left(z - z_p^{(k)}\right)}.
$$

Связь форм записи комплексного числа.

Алгебраическая форма записи: $z = x + jy$ в декартовой системе координат.

Тригонометрическая форма записи: $z = |z|(\cos\phi + j\sin\phi)$ в полярных координатах.

Показательная форма записи: $z = |z|e^{j\phi}$, где $|z| = \sqrt{x^2 + y^2}$, $\phi = \arg z = \operatorname{atan2}(y, x)$, обратно $x = |z|\cos\phi$, $y = |z|\sin\phi$.

4) Критерий устойчивости по значениям полюсов

Критерий. ЛДС устойчива $\Leftrightarrow$ $\sum_{n = 0}^{\infty} |h(n)| < \infty$.

Критерий. ЛДС устойчива $\Leftrightarrow$ все полюсы ПФ находятся внутри круга единичного радиуса: $\forall k \;\; |z_p^{(k)}| < 1$.

5) АЧХ, ФЧХ

Определение. Амплитудно-частотной характеристикой системы называется $A(\hat{\omega}) = \left|\sum_{n = 0}^{\infty} h(n)e^{-j \hat{\omega} n}\right| = |H(e^{j\hat\omega})|$.

Определение. Фазочастотной характеристикой системы называется $\Phi(\hat{\omega}) = \arg\left\{\sum_{n = 0}^{\infty} h(n)e^{-j \hat{\omega} n}\right\} = \arg H(e^{j\hat\omega})$.

Здесь $\hat\omega = \omega T = 2\pi f / f_{\text{д}}$ - нормированная частота.

#### 2 часть
1) ДПФ, ОДПФ

Определение. Дискретным преобразованием Фурье периодической последовательности длины $N$ называется:
$$
\mathcal{F}\{x(n)\} = X(k) = \sum_{n = 0}^{N - 1} x(n) e^{-j \frac{2 \pi k}{N} n}.
$$

Определение. Обратным дискретным преобразованием Фурье периодической последовательности длины $N$ называется:
$$
\mathcal{F}^{-1}\{X(k)\} = x(n) = \frac{1}{N} \sum_{k = 0}^{N - 1} X(k) e^{j \frac{2 \pi k}{N} n}.
$$

2) Амплитудный, фазовый спектры

Определение. Амплитудный спектр периодического сигнала длины $N$ равен $A(k) = \dfrac{1}{N}|X(k)|$.

Определение. Фазовый спектр периодического сигнала длины $N$ равен $\Phi(k) = \arg\{X(k)\}$ для тех гармоник, у которых $|X(k)| \ne 0$.

3) Формула для восстановления сигнала из последовательности

Аналоговый сигнал:
$$
x(t) = \frac{1}{N} \sum_{k = -N/2}^{N/2 - 1} X_a(k) e^{j \frac{2 \pi k}{NT} t}, \qquad
X_a(k) = \begin{cases} X(N+k), & -N/2 \le k \le -1, \\ X(k), & 0 \le k \le N/2 - 1. \end{cases}
$$

4) Равенство Парсеваля

Равенство Парсеваля:
$$
\sum_{n = 0}^{N - 1} |x(n)|^{2} = \frac{1}{N}\sum_{k = 0}^{N-1} |X(k)|^{2}.
$$

#### 1) Вычислить импульсную характеристику длины $N_1$ по заданной передаточной функции с коэффициентами $b_i$, $a_k$ и построить её график. Записать аналитическую формулу импульсной характеристики рекурсивного звена 2-го порядка с учетом нулевых начальных условий. Пояснить, чему в действительности равна длина импульсной характеристики.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

n_var = 16

b0 = 0.5 + 0.02 * n_var
b1 = b0 * (-1) ** (n_var + 1) * (0.9822 + 0.0178 * n_var)
b2 = b0 * (0.8 + 0.2 * (n_var % 5))
a1 = (-1) ** n_var * (0.7778 + 0.025 * n_var)
a2 = 0.64 + 0.006 * n_var
b = np.array([b0, b1, b2])
a = np.array([1, a1, a2])

N1 = n_var % 10 + 20
N2 = n_var % 10 + 30
fd = 1000 * n_var

print('b =', b)
print('a =', a)
print('N1 =', N1, ' N2 =', N2, ' fd =', fd)
```

ИХ считается по разностному уравнению при $x(n) = u_0(n)$:

$$h(n) = b_0 u_0(n) + b_1 u_0(n-1) + b_2 u_0(n-2) - a_1 h(n - 1) - a_2 h(n-2), \qquad h(-1) = h(-2) = 0.$$

$h(0) = b_0 = 0.82$

$h(1) = b_1 - a_1 h(0) = -1.03894 - 1.1778 \cdot 0.82 = -2.004736$

$h(2) = b_2 - a_1 h(1) - a_2 h(0) = 0.82 + 1.1778 \cdot 2.004736 - 0.736 \cdot 0.82 = 2.577658$

Остальные значения посчитаны программно.

```python
def system_response(b, a, x):
    y = np.zeros(len(x))
    for n in range(len(x)):
        for i in range(len(b)):
            if n - i >= 0:
                y[n] += b[i] * x[n - i]
        for k in range(1, len(a)):
            if n - k >= 0:
                y[n] -= a[k] * y[n - k]
    return y


u0 = np.zeros(N1)
u0[0] = 1
h = system_response(b, a, u0)
print('h(n) =', np.round(h, 4))

plt.figure(figsize=(10, 4))
plt.stem(range(N1), h)
plt.title('Импульсная характеристика h(n)')
plt.xlabel('n')
plt.grid()
plt.show()
```

Аналитическая формула. Характеристическое уравнение (знаменатель ПФ):

$$\lambda^2 + 1.1778\lambda + 0.736 = 0, \qquad D = 1.38721 - 2.944 = -1.55679,$$

$$\lambda_{1,2} = \frac{-1.1778 \pm j\,1.24771}{2} = -0.5889 \pm j\,0.6239 = r e^{\pm j\theta},$$

где $r = \sqrt{0.5889^2 + 0.6239^2} = 0.8579$, $\theta = \pi - \operatorname{arctg}\frac{0.6239}{0.5889} = 0.7408\pi$.

При комплексных корнях решение имеет вид $r^n(C_1\cos\theta n + C_2\sin\theta n)$. Однородное уравнение $h(n) + 1.1778h(n-1) + 0.736h(n-2) = 0$ выполняется при $n \geqslant 3$, начальные значения для него $h(1)$ и $h(2)$, поэтому решение записывается со сдвигом на единицу:

$$h(n) = r^{n-1}\left(C_1\cos\theta(n-1) + C_2\sin\theta(n-1)\right), \quad n \geqslant 1.$$

$C_1$ и $C_2$ находятся из $h(1)$ и $h(2)$:

$$\begin{cases} n = 1: & C_1 = -2.004736 \\ n = 2: & -0.5889\,C_1 + 0.62386\,C_2 = 2.577658 \end{cases} \quad\Rightarrow\quad C_2 = \frac{2.577658 - 1.180589}{0.62386} = 2.2394.$$

Аналитическая формула ИХ:

$$h(n) = \begin{cases} 0.82, & n = 0 \\ 0.8579^{\,n-1}\left(-2.0047\cos(0.7408\pi(n-1)) + 2.2394\sin(0.7408\pi(n-1))\right), & n \geqslant 1 \end{cases}$$

```python
lam = np.roots(a)
r = abs(lam[0])
theta = abs(np.angle(lam[0]))
C1 = h[1]
C2 = (h[2] - r * np.cos(theta) * C1) / (r * np.sin(theta))
print('r =', round(r, 4), ' theta =', round(theta / np.pi, 4), 'pi')
print('C1 =', round(C1, 4), ' C2 =', round(C2, 4))

n = np.arange(1, N1)
h_formula = r ** (n - 1) * (C1 * np.cos(theta * (n - 1)) + C2 * np.sin(theta * (n - 1)))
print('формула совпадает с расчетом:', np.allclose(h_formula, h[1:]))
```

Длина ИХ бесконечна: множитель $0.8579^{n-1}$ стремится к нулю, но нулю не равен ни при каком $n$. $N_1 = 26$ - только число посчитанных отсчетов.

#### 6) Вычислить амплитудно-частотную (АЧХ) и фазочастотную (ФЧХ) характеристики в шкале нормированных частот $\hat{\omega}$ и вывести их графики.

Пояснить: чему равны границы основной полосы частот; соответствие между картой нулей и полюсов и видом АЧХ; какому значению АЧХ соответствует скачок на $\pi$, если он имеется; какие частотные составляющие воздействия, низкие или высокие, оказались преимущественно подавленными в реакции.

В прошлом пункте ПФ разложена на простые множители: $$H(z) = K \dfrac{\prod_{i} (z - z_{0}^{(i)})}{\prod_{k} (z - z_{p}^{(k)})}$$

Для АЧХ и ФЧХ подставляем $z = e^{j\hat{\omega}}$ и смотрим модуль и аргумент:

$$H(e^{j\hat{\omega}}) = K \dfrac{\prod_{i} (e^{j\hat{\omega}} - z_{0}^{(i)})}{\prod_{k} (e^{j\hat{\omega}} - z_{p}^{(k)})}$$

АЧХ: $A(\hat{\omega}) = |H(e^{j\hat{\omega}})| = |K| \dfrac{\prod_{i} |e^{j\hat{\omega}} - z_{0}^{(i)}|}{\prod_{k} |e^{j\hat{\omega}} - z_{p}^{(k)}|}$

ФЧХ: $\phi(\hat{\omega}) = \arg H(e^{j\hat{\omega}})$

Так как коэффициенты вещественные, АЧХ четная, а ФЧХ нечетная, и все периодично с периодом $2\pi$. Поэтому достаточно смотреть $0 \le \hat{\omega} \le \pi$.

```python
w = np.linspace(0, np.pi, 4096)
_, H = signal.freqz(b, a, worN=w)
A = np.abs(H)
phi = np.angle(H)

print('A(0) =', round(A[0], 4))
print('A(pi) =', round(A[-1], 4))
print('максимум A =', round(A.max(), 4), 'при w =', round(w[np.argmax(A)] / np.pi, 4), 'pi')
print('минимум A =', round(A.min(), 4), 'при w =', round(w[np.argmin(A)] / np.pi, 4), 'pi')

fig, ax = plt.subplots(2, 1, figsize=(10, 7), sharex=True)
ax[0].plot(w / np.pi, A)
ax[0].axvline(np.angle(zeros[0]) / np.pi, color='C2', ls='--', label='arg нуля')
ax[0].axvline(np.angle(poles[0]) / np.pi, color='C1', ls='--', label='arg полюса')
ax[0].legend()
ax[0].set_title('АЧХ')
ax[0].grid()
ax[1].plot(w / np.pi, phi)
ax[1].set_title('ФЧХ, рад')
ax[1].set_xlabel('w / pi')
ax[1].grid()
plt.tight_layout()
plt.show()
```

Основная полоса частот от $0$ до $\pi$.

Нули лежат на единичной окружности, поэтому на частоте $0.2816\pi$ АЧХ падает до нуля. Около аргумента полюсов ($0.7408\pi$) АЧХ максимальная, примерно 11.4.

Скачок ФЧХ на $\pi$ бывает там, где АЧХ равна нулю. Здесь он есть на $0.2816\pi$, это видно на графике ФЧХ.

Подавляются низкие частоты, высокие наоборот усиливаются: $A(0) \approx 0.21$, а $A(\pi) \approx 4.8$.

#### 1) Вывести графики последовательности $x(n)$ на периоде $N$ в шкале дискретного и дискретного нормированного времени:

$$
x(nT) = A_1 \cos(2\pi f_1 nT + \pi/4) + A_2 \cos(2\pi f_2 nT + \pi/8)
$$

```python
N = 64
fd2 = 2000 * (n_var % 5 + 1)
T = 1 / fd2
A1 = 1 + 0.01 * n_var
A2 = 2 * A1
f1 = fd2 / 8
f2 = 1.5 * f1
print('N =', N, ' fd =', fd2, 'Гц  T =', T, 'с')
print('A1 =', A1, ' A2 =', A2, ' f1 =', f1, 'Гц  f2 =', f2, 'Гц')


def x_analog(t):
    return A1 * np.cos(2 * np.pi * f1 * t + np.pi / 4) + A2 * np.cos(2 * np.pi * f2 * t + np.pi / 8)


n = np.arange(N)
xn = x_analog(n * T)

fig, ax = plt.subplots(2, 1, figsize=(10, 7))
ax[0].stem(n * T * 1000, xn)
ax[0].set_title('x(nT), дискретное время')
ax[0].set_xlabel('nT, мс')
ax[0].grid()
ax[1].stem(n, xn)
ax[1].set_title('x(n), дискретное нормированное время')
ax[1].set_xlabel('n')
ax[1].grid()
plt.tight_layout()
plt.show()
```

По формуле Эйлера каждый косинус раскладывается на две экспоненты:

$$x(nT) = \dfrac{A_1}{2}\left(e^{j (2\pi f_1 nT + \pi / 4)} + e^{-j (2\pi f_1 nT + \pi / 4)}\right) + \dfrac{A_2}{2}\left(e^{j (2\pi f_2 nT + \pi / 8)} + e^{-j (2\pi f_2 nT + \pi / 8)}\right).$$

Значит на частотах $\pm 500$ Гц должна быть амплитуда $A_1/2 = 0.58$ и фаза $\pm \pi / 4$, а на $\pm 750$ Гц амплитуда $A_2/2 = 1.16$ и фаза $\pm \pi / 8$.

После подстановки $T = 1/4000$:
$$x(n) = 1.16 \cos\!\left(\tfrac{\pi}{4} n + \tfrac{\pi}{4}\right) + 2.32 \cos\!\left(\tfrac{3\pi}{8} n + \tfrac{\pi}{8}\right).$$

#### 2) Вычислить ОДПФ от ДПФ последовательности $x(n)$ и вывести график полученной последовательности в шкале дискретного нормированного времени.

```python
def dft(x, inv=False):
    N = len(x)
    X = np.zeros(N, dtype=complex)
    sign = 1 if inv else -1
    for k in range(N):
        for m in range(N):
            X[k] += x[m] * np.exp(sign * 2j * np.pi * k * m / N)
    if inv:
        X = X / N
    return X


X = dft(xn)
x_back = dft(X, inv=True)
print('максимальная разница с x(n):', np.max(np.abs(x_back - xn)))

plt.figure(figsize=(10, 4))
plt.stem(n, x_back.real)
plt.title('ОДПФ от ДПФ x(n)')
plt.xlabel('n')
plt.grid()
plt.show()
```

ОДПФ от ДПФ восстанавливает исходную последовательность, разница на уровне ошибок округления.
