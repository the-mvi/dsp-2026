## Основные теоретические положения
Дискретный сигнал — это последовательность значений непрерывного сигнала, взятых в определённые моменты времени. В общем случае он записывается как $ x(n) $, где $ n $ — целочисленный номер отсчёта.


$ u_0(n)= \begin{cases} 1, & n=0,\\ 0, & n\neq 0. \end{cases} $ - Цифровой единичный импульс

Импульсная характеристика $ h(n) $ — это реакция системы на единичный импульс

Амплитудно-частотная характеристика (АЧХ) — показывает, как система изменяет амплитуду сигнала в зависимости от частоты: $ A(w)=|H(e^{jw})|. $

Фазочастотная характеристика (ФЧХ) — показывает изменение фазы сигнала при прохождении через систему: $\varphi(w)=arg(H(e^{jw})) $.

Дискретным преобразованием Фурье (ДПФ) периодической последовательности $ x_{p}(n) $ называется пара взаимно однозначных преобразований дискретных рядов Фурье для последовательностей во временной и частотной областях, где
* прямое преобразование:

$ X_{p}(k) = \sum_{n=0}^{N - 1}(x_{p}(n)e^{-j\frac{2\pi}{N}nk}) , k = 0,1,...,N-1$

* обратное преобразование (ОДПФ): 

$ x_{p}(n) = \frac{1}{N}\sum_{k=0}^{N - 1}(X_{p}(k)e^{j\frac{2\pi}{N}nk}) , n = 0,1,...,N-1$


Амплитудный спектр определяется как: $ A(k)=|X(k)|= \sqrt{Re^{2}X(k)+Im^{2}X(k)} $.

Фазовый спектр определяется как: $ \varphi(k)=\arg X(k) $.

Импульсной характеристикой h(n) называется её реакция на цифровой единичный импульс $ u_{0}(n) $ при нулевых начальных условиях

Передаточная функция $ H(z) = \frac{Y(z)}{X(z)} $

Из формулы для передаточной функции можно получить разностное уравнение $ y(n) = \sum_{i=0}^{n-1} b_{i}  x(n-i) - \sum_{k=1}^{m-1}a_{k}  y(n-k) $

Тогда для нахождения импульсной характеристики достаточно расчитать реакцию системы на цифровой единичный импульс $ u_{0}(n) $ :

 $ h(n) = b_{0} u_{0}(n) + b_{1} u_{0}(n-1) + b_{2} u_{0}(n-2) - a_{1}h(n-1) - a_{2}h(n-2) $

Из нулевых начальных условий : $ h(-1) = 0 $ и $ h(-2) = 0 $

Для нахождения аналитической формулы необходимо записать передаточную функцию в общем виде. Для заданного варианта передаточная функция $ H(z) = \frac{ b_{0} + b_{1} z^{-1} + b_{2} z^{-2} }{1 + a_{1} z^{-1} + a_{2} z^{-2} } $ 

Применив обратное z-преобразование получаем: $ h(n) = b_{0} r_{*}^{n} \frac{sin((n+1)\varphi_{*})}{sin(\varphi_{*})}u_{0}(n) + b_{1} r_{*}^{n-1} \frac{sin(n\varphi_{*})}{sin(\varphi_{*})}u_{0}(n-1) + b_{2} r_{*}^{n-2} \frac{sin((n-1)\varphi_{*})}{sin(\varphi_{*})}u_{0}(n-2) $ , где $ r_{*} = \sqrt{a_{2}} $ , $ \varphi_{*} = arccos(- \frac{a_{1}}{2r_{*}}) $

Поскольку $ a_{1} \neq 0 $ и $ a_{2} \neq 0 $ , то система является рекурсивной. Рекурсивная система является системой с бесконечной импульсной характеристикой, поэтому длина импульсной характеристики в данном случае бесконечна.   

```python
import matplotlib.pyplot as plt

# Определение параметров
N = 11
N1 = N % 10 + 20

b0 = 0.5 + 0.02 * N 
b1 = b0 * (-1) ** (N + 1) * (0.9822 + 0.0178 * N)
b2 = b0 * (0.8 + 0.2 * (N % 5))

a0 = 1
a1 = (-1) ** N * (0.7778 + 0.025 * N)
a2 = 0.64 + 0.006 * N

print(b0, b1, b2)
print(a1, a2)

def u0(n): # Цифровой единичный импульс
    if n == 0:
        return 1
    else:
        return 0

def h(n): # Импульсная характеристика
    if n < 0:
        return 0
    
    hn = b0 * u0(n) + b1 * u0(n-1) + b2 * u0(n-2) - a1 * h(n-1) - a2 * h(n-2)

    return hn

hs = []

for i in range(0, N1):
    hi = h(i)
    hs.append(hi)
    print(f"h({i}) = {hi}")


plt.stem(range(N1), hs)
plt.xlabel('n'); plt.ylabel('h[n]'); plt.grid(True)
plt.title('Импульсная характеристика')
plt.show()
```

Из полученного графика видно, что импульсная характеристика затухает со временем, что соответствует ожидаемым результатам

Представить передаточную функцию в виде произведения простейших множителей с нулями и полюсами в показательной форме.

Нули передаточной функции - значения $ z = z_{0} $, при которых передаточная функция равна нулю

Полюса передаточной функции - значения $ z = z_{0} $, при которых знаменатель передаточной функции равен нулю 

Передаточная функция задаётся по формуле $ H(z) = \frac{ b_{0} + b_{1} z^{-1} + b_{2} z^{-2} }{1 + a_{1} z^{-1} + a_{2} z^{-2} } $. Для удобства числитель и знаменатель можно домножить на $ z^{2} $.

Тогда передаточная функция принимает вид $ H(z) = \frac{ b_{0} z^{2} + b_{1} z + b_{2}}{z^{2} + a_{1} z + a_{2}} $. Нулями являютася значения z, при которых  $ b_{0} z^{2} + b_{1} z + b_{2} = 0 $. Полюсами являютася значения z, при которых $ z^{2} + a_{1} z + a_{2} = 0 $

Поскольку $ b_{i}, a_{i} $ вещественные, то корни уравенений комплексно-сопряженные. Для перевода из алгебраической формы в показательную необходимо расчитать модуль по формуле: $ |z_{0}| = \sqrt{Im(z_{0})^{2} + Re(z_{0})^{2}} $ и аргумент по формуле: $\varphi_{0} = arctg(\frac{Im(z_{0})}{Re(z_{0})}) $. Тогда комплексное число в показательной форме равно $ |z_{0}|e^{j\varphi_{0}}$

Представление передаточной функции в виде произведения простейших множетелей с нулями и полюсами в показательной форме в общем виде:

 $ H(z) = \frac{ b_{0} z^{2} + b_{1} z + b_{2}}{z^{2} + a_{1} z + a_{2}} = b_{0} \frac{(z - z_{01})(z - z_{02})}{(z - p_{1})(z - p_{2})} = b_{0}\frac{(z - |z_{0}|e^{j\varphi_{0}})(z - |z_{0}|e^{-j\varphi_{0}})}{(z - |p|e^{j\varphi_{p}})(z - |p|e^{-j\varphi_{p}})}$

```python
import math

descreminant_zer = (b1 ** 2 - 4 * b0 * b2) ** 0.5
zeros = [(-b1 - descreminant_zer) / (2 * b0), (-b1 + descreminant_zer) / (2 * b0)]

descreminant_pol = (a1 ** 2 - 4 * a2) ** 0.5
polus = [(-a1 - descreminant_pol) / 2, (-a1 + descreminant_pol) / 2]
print(f"Нули передаточной функции в алгебраической формуле: {zeros}") 
print(f"Полюса передаточной функции в алгебраической формуле: {polus}") 

for zero in zeros: # Перевод нулей в показательную форму 
    module = (zero.real ** 2 + zero.imag ** 2) ** 0.5 
    arg = math.atan2(zero.imag, zero.real)
    arg_str = "exp(j * " + str(arg) + ")"
    print(f"Нуль {zero} в показательной форме: {str(module) + arg_str} ")
    print(f"Аргумент нуля {zero} относительно pi: {arg / math.pi} pi")


for pol in polus: # Перевод полюсов в показательную форму 
    module = (pol.real ** 2 + pol.imag ** 2) ** 0.5 
    arg = math.atan2(pol.imag, pol.real)
    arg_str = "exp(j * " + str(arg) + ")"
    print(f"Полюс {pol} в показательной форме: {str(module) + arg_str}")
    print(f"Аргумент полюса {pol} относительно pi: {arg / math.pi} pi")
```

Поскольку полюсы находятся внутри единичной окружности, то система будет устойчивой.

6. Вычислить амплитудно-частотную (АЧХ) и фазочастотную (ФЧХ) характеристики в шкале нормированных частот $ \hat{ω} $ и вывести их графики. Пояснить:
* чему равны границы основной полосы частот;
* соответствие между картой нулей и полюсов и видом АЧХ;
* какому значению АЧХ соответствует скачок на π, если он имеется;
* какие частотные составляющие воздействия, низкие или высокие, оказались преимущественно подавленными в реакции.

Нормированная частота определяется по формуле: $ \hat{w} = \frac{w}{\pi} = 2 \frac{f}{f_{d}} $, где $ f $ - частота сигнала, $ f_{d} $ - частота дискретизации. 

Амплитудно частотной характеристикой (АЧХ) называется модуль частотной характеристики $ H(e^{j\hat{w}})$ : $ A(\hat{w}) = |H(e^{j\hat{w}})| $

Фазочастотной характеристикой (ФЧХ) называется аргумент частотной характеристики $ H(e^{j\hat{w}})$ : $ \varphi(\hat{w}) = arg(H(e^{j\hat{w}})) $

Передаточная функция задаётся по формуле: $ H(z) = \frac{b_{0} + b_{1} z^{-1} + b_{2} z^{-2}}{1 + a_{1} z^{-1} + a_{2} z^{-2}} $

Основная полоса частот ограничивается частотой Найксвита, поэтому : $ 0 \le w \le \pi $

Тогда нормированная частота находится в диапазоне: $ 0 \le \hat{w} \le 1 $

```python

def H(z): # Передаточная функция
    chisl = b0 + b1 * (z ** -1) + b2 * (z ** -2)
    znam = 1 + a1 * (z ** -1) + a2 * (z ** - 2)
    return chisl / znam 

w_norm = np.linspace(0, 1, 1000)

w = np.pi * w_norm


# АЧХ
def ACH(he):
    return (((he.real ** 2) + (he.imag ** 2)) ** 0.5)

# ФЧХ
def FCH(he):
    return math.atan2(he.imag, he.real) / math.pi

A = []
phi = []
for i in range(0, len(w)):
    he = H(np.exp(1j * w[i]))
    A.append(ACH(he))
    phi.append(FCH(he))


plt.figure(figsize=(10, 4))
plt.plot(w_norm, A)
plt.xlabel("w^")
plt.ylabel("A(w^)")
plt.title('Амплитудно-частотная характеристика')
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 4))
plt.plot(w_norm, phi)
plt.xlabel("w^")
plt.ylabel("phi(w^)")
plt.title('Фазочастотная характеристика')
plt.grid(True)
plt.show()
```

Из построенной АЧХ видно, что вблизи частоты, соответствующей аргументу полюсов относительно $ \pi $ (0.284491), наблюдается выраженный максимум.

На частоте, соответствующей аргументу нулей относительно $ \pi $ (0.700478), АЧХ обращается в ноль. Это объясняется тем, что нули передаточной функции расположены на единичной окружности.

В точке, где АЧХ обращается в ноль, фаза передаточной функции не определена. При прохождении через эту частоту наблюдается скачок фазы на $ \pi $.

По графику АЧХ видно, что на высоких частотах коэффициент передачи значительно уменьшается, поэтому высокочастотные составляющие воздействия преимущественно подавляются.

$ x(nT)=A_{1}cos(2 \pi f_{1} nT+\pi /4)+A_{2}cos(2 \pi f_{2} nT+ \pi /8) $

Дискретное время определяется по формуле: $ t = nT = \frac{n}{f_{d}}$

Нормированное время - это время относительно длины последовательности. $ t_{norm} = \frac{n}{N_{d}} $

```python
Nd = 64
A1 = 1 + 0.01 * N
A2 = 2 * A1
fd = 2000 * (N % 5 + 1)
f1 = fd / 8
f2 = 1.5 * f1
T = 1 / fd

xs = []
ts = []
tnorm = []
for n in range(0, Nd):
    t = n * T
    x = A1 * math.cos(2 * math.pi * f1 * t + math.pi / 4) + A2 * math.cos(2 * math.pi * f2 * t + math.pi / 8)
    xs.append(x)
    ts.append(t) 
    tnorm.append(n / Nd)

plt.stem(ts, xs)
plt.xlabel('nT, с'); plt.ylabel('x(nT)'); plt.grid(True)
plt.title('x(n) в шкале дискретного времени')
plt.show()

plt.stem(tnorm, xs)
plt.xlabel('n/N'); plt.ylabel('x(n)'); plt.grid(True)
plt.title('x(n) в шкале нормированного времени')
plt.show()
```

2. Вычислить ОДПФ от ДПФ последовательности $ x(n) $ и вывести график полученной последовательности в шкале дискретного нормированного временах.

Дискретным преобразованием Фурье (ДПФ) периодиеческой последовательности $ x_{p}(n) $ называется пара взаимно однозначных преобразований дискретных рядов Фурье для последовательностей во временной и частотной областях, где
* прямое преобразование:

$ X_{p}(k) = \sum_{n=0}^{N - 1}(x_{p}(n)e^{-j\frac{2\pi}{N}nk}) , k = 0,1,...,N-1$

* обратное преобразование (ОДПФ): 

$ x_{p}(n) = \frac{1}{N}\sum_{k=0}^{N - 1}(X_{p}(k)e^{j\frac{2\pi}{N}nk}) , n = 0,1,...,N-1$

```python
df = [] # Значения ДПФ

for k in range(0, Nd):
    k_val = 0
    for n in range(0, Nd):
        angle = -2 * math.pi * n * k / Nd
        k_val += xs[n] * complex(math.cos(angle), math.sin(angle))
    df.append(k_val)

print(df)

odpf = [] # Значения ОДПФ

for n in range(0, Nd):
    n_val = 0
    for k in range(0, Nd):
        angle = 2 * math.pi * n * k / Nd
        n_val += df[k] * complex(math.cos(angle), math.sin(angle))
    odpf.append(n_val / Nd)
    
print(odpf)
print(xs)

odpf_real = [val.real for val in odpf]
plt.stem(tnorm, odpf_real)
plt.xlabel('n/N'); plt.ylabel('x(n)'); plt.grid(True)
plt.title('Восстановленный сигнал (ОДПФ от ДПФ)')
plt.show()
```

Полученная последовательность после выполнения обратного дискретного преобразования Фурье полностью совпадает с исходной последовательностью из задания 1. Небольшие погрешности в вычисленных значениях обусловлены особенностями представления чисел с плавающей точкой.

3. Вычислить амплитудный и фазовый спектры последовательности x(n) и вывести их графики в шкалах дискретных нормированных и абсолютных частот. Пояснить:

Частота отдельного отсчёта определяется как $ f_{k} = \frac{kf_{d}}{N_{d}}$

Амплитудным спектром дискретного сигнала называется модуль его фурье-изображения: $ |X(e^{jwT})| $

Фазовым спектром дискретного сигнала называется аргумент фурье-изображения: $ arg(X(e^{jwT})) $

```python
ampl_vals = []
faz_vals = []

for val in df:
    ampl_vals.append((val.real ** 2 + val.imag ** 2) ** 0.5)
    faz_vals.append(math.atan2(val.imag, val.real))

plt.stem(tnorm, ampl_vals)
plt.xlabel('n/N'); plt.ylabel('|X(exp(jwT))|'); plt.grid(True)
plt.title('Амплитудный спектр в шкале дискретных нормированных частот')
plt.show()

plt.stem(tnorm, faz_vals)
plt.xlabel('n/N'); plt.ylabel('arg(X(exp(jwT)))'); plt.grid(True)
plt.title('Фазовый спектр в шкале дискретных нормированных частот')
plt.show()


f_abs = np.arange(Nd) * (fd / Nd)

plt.stem(f_abs, ampl_vals)
plt.xlabel('f, Гц'); plt.ylabel('|X(f)|'); plt.grid(True)
plt.title('Амплитудный спектр в шкале абсолютных частот')
plt.show()

plt.stem(f_abs, faz_vals)
plt.xlabel('f, Гц'); plt.ylabel('arg(X(f))'); plt.grid(True)
plt.title('Фазовый спектр в шкале абсолютных частот')
plt.show()
```

4. Вычислить ДПФ конечной последовательности x(n) длины N. Вывести графики в шкале дискретных нормированных частот модуля ДПФ конечной последовательности и амплитудного спектра периодической последовательности. Пояснить связь модуля ДПФ конечной последовательности с амплитудным спектром периодической последовательности.

ДПФ конечной последовательности длины $ N_{d} $ определяется формулой:

$ X(k) = \sum_{n=0}^{N_d-1} x(n)e^{-j\frac{2\pi}{N_d}nk} $, где $ k=0,1,...,N_d-1 $.

Модуль ДПФ определяется как: $|X(k)|=\sqrt{Re^{2}(X(k)) + Im^{2}(X(k))} $ 

Амплитудный спектр периодической последовательности: $ C(k)=\frac{X(k)}{N_d} $ 

Амплитуда гармонической составляющей определяется выражением: $ A_k=\frac{2|X(k)|}{N_d} $

При $ k = 0 $ и $ k = N_{d}/2 $ удвоение не производится. 

```python
mod_dft = []
for val in df:
    mod_dft.append(((val.real ** 2) + (val.imag ** 2)) ** 0.5)

amp_spec = []

for k in range(Nd):
    if k == 0 or k == Nd // 2:
        amp_spec.append(mod_dft[k] / Nd)
    else:
        amp_spec.append(2 * mod_dft[k] / Nd)

k_norm = np.arange(Nd) / Nd

plt.figure(figsize=(10, 4))
plt.stem(k_norm, mod_dft)
plt.xlabel('k/N'); plt.ylabel('|X(k)|')
plt.title('Модуль ДПФ конечной последовательности')
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 4))
plt.stem(k_norm, amp_spec)
plt.xlabel('k/N'); plt.ylabel('A_k')
plt.title('Амплитудный спектр периодической последовательности')
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 5))
plt.stem(k_norm, mod_dft, linefmt='b-', markerfmt='bo', label='|X(k)| (ДПФ)')
plt.stem(k_norm, amp_spec, linefmt='r--', markerfmt='rx', label='A_k (спектр периодич.)')
plt.xlabel('k/N'); plt.ylabel('Амплитуда'); plt.legend()
plt.title('Сравнение модуля ДПФ и амплитудного спектра')
plt.grid(True)
plt.show()
```

Из построенных графиков видно, что модуль ДПФ конечной последовательности и амплитудный спектр периодической последовательности имеют одинаковое расположение спектральных составляющих, однако отличаются масштабом значений.

5. Восстановить периодический аналоговый сигнал $ x(t) $ по отсчётам ДПФ $ X(k) $ периодической последовательности x(n). Для вычисления значений сигнала x(t) использовать формулу:

$ x(t) = \frac{1}{N}\sum_{k=-\frac{N}{2}}^{\frac{N}{2}-1}(X_{a}(k)e^{j\frac{2\pi}{NT}kt}) $, где отсчеты $ X_{a}(k) $ связаны с отсчетами ДПФ $ X(k) $ соотношением:

$ X_{a}(k) = X(N + k), -\frac{N}{2} \le k \le -1 $

$ X_{a}(k) = X(k), 0 \le k \le \frac{N}{2} - 1$

задавая значения времени t на интервале $ t∈[0;(N−1)T] $ с шагом $ Δt=0.25T $. Вывести графики:

* периодической последовательности x(n) и модуля ее ДПФ;

* исходного аналогового сигнала $ x(t)=x(nT)∣_{nT=t} $.

```python
def Xa(k):
    if k <= -1 and k >= - Nd // 2:
        return df[N+k]
    elif k >= 0 and k <= N // 2 - 1:
        return df[k]
    else:
        return 0

def x_analog(t):
    res = 0
    for k in range(- Nd // 2, Nd // 2):
        angle = 2 * math.pi * k * t / (Nd * T)
        res += Xa(k) * complex(math.cos(angle), math.sin(angle))
    return (res / Nd).real

ts = []
xts = []
for t in np.arange(0, Nd * T, 0.25 * T): 
    ts.append(t)
    xts.append(x_analog(t))

f_abs = np.arange(Nd // 2) * (fd / Nd) 

t_orig = np.linspace(0, (Nd - 1) * T, 500)
x_orig = [A1 * math.cos(2*math.pi*f1*ti + math.pi/4) + A2 * math.cos(2*math.pi*f2*ti + math.pi/8) for ti in t_orig]

plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.stem(tnorm, xs)
plt.xlabel('n/N'); plt.ylabel('x(n)'); plt.title('x(n)')

plt.subplot(1, 2, 2)
plt.stem(np.arange(Nd)/Nd, mod_dft)
plt.xlabel('k/N'); plt.ylabel('|X(k)|'); plt.title('Модуль ДПФ')
plt.tight_layout(); plt.show()

plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(ts, xts, 'bo-', label='Восстановленный x(t)')
plt.xlabel('t, с'); plt.ylabel('x(t)'); plt.title('Восстановленный аналоговый x(t)')
plt.grid(True)

plt.subplot(1, 2, 2)
plt.stem(f_abs, mod_dft[:Nd//2])
plt.xlabel('f, Гц'); plt.ylabel('|X(k)|'); plt.title('Амплитудный спектр сигнала')
plt.tight_layout(); plt.show()

plt.figure(figsize=(8, 4))
plt.plot(t_orig, x_orig, 'r-', linewidth=2, label='Исходный аналоговый x(t)')
plt.plot(ts, xts, 'b--o', label='Восстановленный x(t)')
plt.xlabel('t, с'); plt.ylabel('x(t)'); plt.title('Сравнение исходного и восстановленного сигналов')
plt.legend(); plt.grid(True)
plt.show()
```

7. Проверить равенство Парсеваля. Пояснить его смысл.

Равенство Парсеваля устанавливает связь между энергией дискретной последовательности во временной области и её представлением в частотной области.

Для последовательности $ x(n) $ длины $ N $ равенство Парсеваля:

$ \sum_{n=0}^{N-1}|x(n)|^2 = \frac{1}{N}\sum_{k=0}^{N-1}|X(k)|^2 $

Равенство Парсеваля утверждает, что энергия сигнала не изменяется при переходе из временной области в частотную. Она лишь по-разному представляется в зависимости от выбранной области.

```python
E_time = 0
for n in range(Nd):
    E_time += xs[n] ** 2  

E_freq_sum = 0
for k in range(Nd):
    mod_sq = df[k].real ** 2 + df[k].imag ** 2 
    E_freq_sum += mod_sq

E_freq = E_freq_sum / Nd 

print(f"Энергия по времени: {E_time:.5f}")
print(f"Энергия по частоте: {E_freq:.5f}")
print(f"Разница: {abs(E_time - E_freq):.10f}")
```

Исходя из полученных значений равенство выполняется.
