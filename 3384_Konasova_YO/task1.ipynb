## Основные теоретические положения

Линейная дискретная система преобразует входную последовательность $x[n]$ в выходную последовательность $y[n]$. Для линейной стационарной системы выполняются свойства линейности и инвариантности к сдвигу.

Поведение системы может быть описано с помощью разностного уравнения, которое связывает текущие и предыдущие значения входного и выходного сигналов:

$$
\sum_{i=0}^{M}a_i\,y[n-i]=\sum_{i=0}^{L}b_i\,x[n-i].
$$

Импульсная характеристика $h[n]$ - это реакция системы на единичный импульс:

$$
\delta[n]=
\begin{cases}
1, & n=0,\\[4pt]
0, & n\neq 0.
\end{cases}
$$

Для линейной стационарной системы выходной сигнал определяется свёрткой входного сигнала с импульсной характеристикой:

$$
y[n]=\sum_{k=-\infty}^{+\infty}x[k]\,h[n-k].
$$

Длина линейной свёртки двух конечных последовательностей определяется выражением:

$$
L_y=L_x+L_h-1.
$$

Передаточная функция описывает систему в области $Z$-преобразования и представляет собой отношение $Z$-преобразования выходного сигнала к $Z$-преобразованию входного сигнала при нулевых начальных условиях:

$$
H(z)=\frac{Y(z)}{X(z)}.
$$

Для рекурсивного звена второго порядка:

$$
H(z)=\frac{b_0+b_1z^{-1}+b_2z^{-2}}{a_0+a_1z^{-1}+a_2z^{-2}}.
$$

Корни числителя передаточной функции называются нулями, а корни знаменателя - полюсами.

Для причинной системы устойчивость определяется расположением полюсов: система устойчива, если все полюса находятся внутри единичной окружности комплексной плоскости:

$$
|p_i|<1.
$$

Частотные характеристики позволяют исследовать реакцию системы на гармонические сигналы различных частот.

Амплитудно-частотная характеристика (АЧХ) определяется модулем передаточной функции:

$$
A(\omega)=\left|H\!\left(e^{\,j\omega}\right)\right|.
$$

Фазочастотная характеристика (ФЧХ) определяется аргументом передаточной функции:

$$
\varphi(\omega)=\arg H\!\left(e^{\,j\omega}\right).
$$

АЧХ показывает, как система изменяет амплитуду входного сигнала на разных частотах, а ФЧХ характеризует изменение его фазы.

Дискретное преобразование Фурье (ДПФ) используется для представления конечной последовательности в частотной области.

Прямое ДПФ определяется формулой:

$$
X[k]=\sum_{n=0}^{N-1}x[n]\,e^{-j\frac{2\pi}{N}kn},
$$

где $N$ - количество отсчётов, $n$ - номер отсчёта во временной области, $k$ - номер частотного отсчёта.

Обратное дискретное преобразование Фурье (ОДПФ) позволяет восстановить исходную последовательность:

$$
x[n]=\frac{1}{N}\sum_{k=0}^{N-1}X[k]\,e^{\,j\frac{2\pi}{N}kn}.
$$

Результат ДПФ является комплексной последовательностью. Для анализа сигнала используются её модуль и аргумент.

Амплитудный спектр:

$$
A[k]=\frac{|X[k]|}{N}.
$$

Фазовый спектр:

$$
\varphi[k]=\arg X[k].
$$

Амплитудный спектр показывает амплитуды частотных составляющих сигнала, а фазовый спектр - их начальные фазы.

Для вещественной последовательности спектр обладает свойством комплексной симметрии:

$$
X[N-k]=X^*[k],
$$

где звёздочка обозначает комплексное сопряжение.

Периодический аналоговый сигнал может быть восстановлен по отсчётам ДПФ с помощью обратного представления в виде суммы гармонических составляющих:

$$
x(t)=\frac{1}{N}\sum_{k=-N/2}^{N/2-1}X_a[k]\,e^{\,j\frac{2\pi kt}{NT}},
$$

где:
- $X_a[k]$ - отсчёты ДПФ, расположенные в диапазоне отрицательных и положительных частот;
- $N$ - количество отсчётов;
- $T$ - период дискретизации;
- $t$ - аналоговое время.

Период восстановленного сигнала равен $NT$.

Равенство Парсеваля устанавливает связь между энергией последовательности во временной области и энергией её спектра:

$$
\sum_{n=0}^{N-1}|x[n]|^2=\frac{1}{N}\sum_{k=0}^{N-1}|X[k]|^2.
$$

Оно показывает, что при переходе от временного представления сигнала к частотному его энергия сохраняется. Это свойство позволяет проверять правильность вычисления ДПФ и ОДПФ.

Коэффициенты числителя

$b_0 = 0.5 + 0.02 \cdot 9 = 0.68$

$b_1 = b_0 \cdot (-1)^{9+1} \cdot (0.9822 + 0.0178 \cdot 9) = 0.776832$

$b_2 = b_0 \cdot (0.8 + 0.2 \cdot (9 \bmod 5)) = 1.088$

Коэффициенты знаменателя

$a_0 = 1$

$a_1 = (-1)^9 \cdot (0.7778 + 0.025 \cdot 9) = -1.0028$

$a_2 = 0.64 + 0.006 \cdot 9 = 0.694$

Для рекурсивного звена второго порядка передаточная функция имеет вид:
$$
H(z)=\frac{b_0+b_1z^{-1}+b_2z^{-2}}{a_0+a_1z^{-1}+a_2z^{-2}}
$$

Подставляем найденные коэффициенты:
$$
\boxed{
H(z)=
\frac{0.68+0.776832\,z^{-1}+1.088\,z^{-2}}
{1-1.0028\,z^{-1}+0.694\,z^{-2}}
}
$$

Определение передаточной функции системы:
$$
H(z)=\frac{Y(z)}{X(z)}
$$

Отсюда имеем:
$$
Y(z)\left(1-1.0028\,z^{-1}+0.694\,z^{-2}\right)
=
X(z)\left(0.68+0.776832\,z^{-1}+1.088\,z^{-2}\right)
$$

Переход к последовательностям:
$$
\boxed{
\begin{aligned}
y[k]={}&1.0028\,y[k-1]-0.694\,y[k-2]+0.68\,x[k]+0.776832\,x[k-1]+1.088\,x[k-2]
\end{aligned}
}
$$

Длина импульсной характеристики $N_1 = 29$

Будем подставлять импульс в разностное уравнение

```python
import numpy as np
import matplotlib.pyplot as plt

b = [0.68, 0.776832, 1.088]
a = [1, -1.0028, 0.694]
N1 = 29

x = np.zeros(N1) # Единичный импульс
x[0] = 1

imp_h = np.zeros(N1)

for k in range(N1): # Вычисление импульсной характеристики
    for i in range(len(b)):  # Входные сигналы
        if k - i >= 0:
            imp_h[k] += b[i] * x[k - i]

    for i in range(1, len(a)): # Выходные сигналы
        if k - i >= 0:
            imp_h[k] -= a[i] * imp_h[k - i]
    imp_h[k] /= a[0]

print("Импульсная характеристика:")
for k in range(N1):
    print(f"imp_h[{k}] = {imp_h[k]:.6f}")

plt.figure(figsize=(10, 5))
plt.stem(range(N1), imp_h)
plt.title("Импульсная характеристика")
plt.xlabel("Номер отсчёта k")
plt.ylabel("Значение импульсной характеристики")
plt.grid(True)

plt.show()
```

Начиная с k=3, входные составляющие становятся равны нулю, поэтому остаётся однородное разностное уравнение: $$h[k]=1.0028\,h[k-1]-0.694\,h[k-2]$$

Его характеристическое уравнение: $$r^2-1.0028\,r+0.694=0$$

Корни уравнения комплексно-сопряженные, аналитическая формула импульсной характеристики для k>=2 имеет вид:
$$
h[k]=\rho^{\,k-2}\left(C\cos\big((k-2)\theta\big)+D\sin\big((k-2)\theta\big)\right)
$$
где:
$$
\rho=\sqrt{0.694}\approx 0.833067
$$
$$
\theta=\arccos\!\left(\frac{1.0028}{2\sqrt{0.694}}\right)\approx 0.924952
$$
$$
C=2.078900
$$
$$
D\approx 0.045091
$$
Таким образом, аналитическая формула:
$$
\boxed{
\begin{aligned}
h[k]={}&0.833067^{\,k-2}\Big(2.078900\cos\big(0.924952\,(k-2)\big)+0.045091\sin\big(0.924952\,(k-2)\big)\Big)
\end{aligned}
}
$$

для $k\geq 2$, а первые два значения задаются отдельно: $h[0]=0.68$, $h[1]=1.458736$.

Так как в нашем случае используется рекурсивное звено второго порядка, каждое новое выходное значение зависит от двух предыдущих выходных значений. Поэтому даже после окончания входного импульса выходной сигнал продолжает изменяться. Поскольку коэффициенты знаменателя обеспечивают затухание, значения импульсной характеристики постепенно приближаются к нулю. Но теоретически они не становятся точно равными нулю за конечное количество отсчётов. Следовательно, действительная длина импульсной характеристики бесконечна.

#### Задание 3
Разностное уравнение см. в пункте 1.

```python
import numpy as np
import matplotlib.pyplot as plt

y_conv = np.convolve(x, imp_h)[:N2]

y_diff = np.zeros(N2)

for n in range(N2):
    for i in range(len(b)):
        if n - i >= 0:
            y_diff[n] += b[i] * x[n - i]

    for i in range(1, len(a)):
        if n - i >= 0:
            y_diff[n] -= a[i] * y_diff[n - i]

print("Реакция по разностному уравнению:")
for n in range(N2):
    print(f"y_diff[{n}] = {y_diff[n]:.6f}")

print("\nМаксимальное отличие результатов:",
      np.max(np.abs(y_conv - y_diff)))

plt.figure(figsize=(10, 7))

plt.subplot(2, 1, 1)
plt.stem(range(N2), x)
plt.title("Входное воздействие")
plt.xlabel("Номер отсчёта n")
plt.ylabel("Амплитуда")
plt.grid(True)

plt.subplot(2, 1, 2)
plt.stem(range(N2), y_diff, linefmt="b-", markerfmt="bo", basefmt="k-", label="По разностному уравнению")
plt.plot(range(N2), y_conv, "o", color="orange", markersize=5, label="По свёртке")
plt.title("Сравнение реакций системы")
plt.xlabel("Номер отсчёта n")
plt.ylabel("Амплитуда")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

Длина входного воздействия составляет 39 отсчётов, из которых первые 20 имеют значение 1, а остальные 19 - значение 0. Длина рассчитанной реакции также составляет 39 отсчётов. При этом реакция системы продолжается после окончания входного воздействия, что обусловлено наличием обратной связи в рекурсивном звене.

Реакция системы была рассчитана двумя способами: по разностному уравнению и с использованием свёртки входного сигнала с импульсной характеристикой. Максимальное отличие полученных результатов составило $1{,}33 \cdot 10^{-15}$, что свидетельствует об их практически полном совпадении. Таким образом, оба способа расчёта дают одинаковый результат.

Вычислим программно нули и полюсы, а также изобразим карту нулей и полюсов.

```python

import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

b = [0.68, 0.776832, 1.088]
a = [1, -1.0028, 0.694]

zeros, poles, k = signal.tf2zpk(b, a) # Автоматический расчет нулей и полюсов

print("Нули:")
for z in zeros:
    print(f"{z:.6f}, модуль = {abs(z):.6f}")

print("\nПолюсы:")
for p in poles:
    print(f"{p:.6f}, модуль = {abs(p):.6f}")

if all(abs(p) < 1 for p in poles):
    print("\nРекурсивное звено устойчиво")
else:
    print("\nРекурсивное звено неустойчиво")


t = np.linspace(0, 2 * np.pi, 500)# Построение карты нулей и полюсов
plt.figure(figsize=(8, 8))
plt.plot(np.cos(t), np.sin(t), 'k--', label='Единичная окружность')
plt.scatter(zeros.real, zeros.imag, marker='o', s=100, facecolors='none', edgecolors='blue', linewidths=2, label='Нули')
plt.scatter(poles.real, poles.imag, marker='x', s=100, color='red', linewidths=2, label='Полюсы')
plt.axhline(0, color='black', linewidth=0.7)
plt.axvline(0, color='black', linewidth=0.7)
plt.xlabel('Re(z)')
plt.ylabel('Im(z)')
plt.title('Карта нулей и полюсов')
plt.grid()
plt.axis('equal')
plt.legend()
plt.show()
```

По построенной карте нулей и полюсов видно, что рекурсивное звено является устойчивым, поскольку оба полюса находятся внутри единичной окружности (их модули равны 0,833067 и меньше единицы). Нули передаточной функции расположены за пределами единичной окружности, их модули составляют 1,264911.
```python
import numpy as np
import matplotlib.pyplot as plt

N = 64
fd = 10000
T = 1 / fd
A1 = 1.09
A2 = 2.18
f1 = 1250
f2 = 1875

n = np.arange(N)
t = n * T
x = A1 * np.cos(2 * np.pi * f1 * t + np.pi / 4) + A2 * np.cos(2 * np.pi * f2 * t + np.pi / 8)

fig, axes = plt.subplots(2, 1, figsize=(10, 7), constrained_layout=True)

axes[0].stem(t, x, basefmt=" ")
axes[0].set_title("Последовательность в шкале дискретного времени")
axes[0].set_xlabel("t = nT, с")
axes[0].set_ylabel("x(nT)")
axes[0].grid(True)

axes[1].stem(n, x, basefmt=" ")
axes[1].set_title("Последовательность в шкале дискретного нормированного времени")
axes[1].set_xlabel("n = t/T, отсчёты")
axes[1].set_ylabel("x[n]")
axes[1].set_xticks(np.arange(0, N, 4))
axes[1].grid(True)

plt.show()
```

```python
# ДПФ и обратное ДПФ
X = np.fft.fft(x)
x_recovered = np.fft.ifft(X).real

plt.figure(figsize=(10, 4.8))
plt.stem(n, x_recovered, basefmt=" ")
plt.title("Последовательность после ОДПФ")
plt.xlabel("n, номер отсчёта")
plt.ylabel("x_восст[n]")
plt.xticks(np.arange(0, N, 4))
plt.grid(True)
plt.tight_layout()
plt.show()

print(f"Максимальная абсолютная погрешность восстановления: {np.max(np.abs(x - x_recovered)):.3e}")
```

Было вычислено ДПФ исходной последовательности, а затем выполнено ОДПФ для восстановления исходных данных. #### Задание 5

```python
n = np.arange(N)
tn = n * T
x_n = A1 * np.cos(2 * np.pi * f1 * tn + np.pi / 4) + \
      A2 * np.cos(2 * np.pi * f2 * tn + np.pi / 8)
k = np.arange(-N // 2, N // 2)
X_a = np.concatenate((X[N // 2:], X[:N // 2])) # Перестановка отсчетов ДПФ

dt = 0.25 * T
t = np.arange(0, (N - 1) * T + dt / 2, dt)

x_rec_complex = np.zeros(t.size, dtype=complex)
for ki, Xki in zip(k, X_a): # Восстановление периодического аналогового сигнала
    x_rec_complex += Xki * np.exp(1j * 2 * np.pi * ki * t / (N * T))
x_rec = (x_rec_complex / N).real

x_original = A1 * np.cos(2 * np.pi * f1 * t + np.pi / 4) + \
             A2 * np.cos(2 * np.pi * f2 * t + np.pi / 8)

fig, axes = plt.subplots(2, 1, figsize=(10, 7), constrained_layout=True)
axes[0].stem(n, x_n, basefmt=" ")
axes[0].set_title("Периодическая последовательность x(n)")
axes[0].set_xlabel("n, номер отсчёта")
axes[0].set_ylabel("x(n)")
axes[0].grid(True)

freq_bins = np.fft.fftfreq(N, d=1)
freq_bins_shifted = np.fft.fftshift(freq_bins)
X_mag_shifted = np.fft.fftshift(X_mag)
axes[1].stem(freq_bins_shifted, X_mag_shifted, basefmt=" ")
axes[1].set_title("Модуль ДПФ |X(k)|")
axes[1].set_xlabel("ν = k/N, циклов/отсчёт")
axes[1].set_ylabel("|X(k)|")
axes[1].grid(True)
plt.show()

fig, axes = plt.subplots(2, 1, figsize=(10, 7), constrained_layout=True)
axes[0].plot(t, x_rec)
axes[0].set_title("Восстановленный аналоговый сигнал x(t)")
axes[0].set_xlabel("t, с")
axes[0].set_ylabel("x(t)")
axes[0].grid(True)

freq_hz = k / (N * T)
amp_spectrum = np.abs(X_a) / N
axes[1].stem(freq_hz, amp_spectrum, basefmt=" ")
axes[1].set_title("Амплитудный спектр восстановленного сигнала")
axes[1].set_xlabel("f, Гц")
axes[1].set_ylabel("|Xₐ(k)| / N")
axes[1].grid(True)
plt.show()

plt.figure(figsize=(10, 4))
plt.plot(t, x_original)
plt.title("Исходный аналоговый сигнал x(t)")
plt.xlabel("t, с")
plt.ylabel("x(t)")
plt.grid(True)
plt.tight_layout()
plt.show()

print(f"Количество точек времени: {len(t)}")
print(f"Шаг времени Δt = {dt:.8f} с = 0.25T")
print(f"Максимальное отклонение восстановленного сигнала от исходного: "
      f"{np.max(np.abs(x_rec - x_original)):.3e}")
```

В результате восстановления аналогового сигнала по отсчётам ДПФ получено $253$ значения на заданном временном интервале с шагом $0.25\,T$. Сравнение восстановленного сигнала с исходным показало максимальное отклонение $2.731\cdot 10^{-14}$. Таким образом, восстановление выполнено с высокой точностью, а полученный результат подтверждает корректность реализации формулы.

#### Задание 7
Равенство Парсеваля устанавливает связь между энергией дискретной последовательности и энергией её спектра, полученного с помощью ДПФ. Оно имеет следующий вид:
$$
\sum_{n=0}^{N-1}|x(n)|^2=\frac{1}{N}\sum_{k=0}^{N-1}|X(k)|^2
$$


```python
energy_x = np.sum(np.abs(x)**2) # Энергия исходной последовательности 
energy_X = np.sum(np.abs(X)**2) / N # Энергия спектра
difference = abs(energy_x - energy_X)
print(f"Энергия последовательности: {energy_x:.10f}")
print(f"Энергия спектра: {energy_X:.10f}")
print(f"Разность: {difference:.3e}")
```

Равенство Парсеваля означает, что энергия сигнала не изменяется при переходе из временной области в частотную. Энергия, рассчитанная по исходным отсчётам последовательности, равна энергии, рассчитанной по спектральным отсчётам ДПФ с учётом коэффициента 1/N. Полученные результаты подтверждают выполнение равенства Парсеваля. Энергия исходной последовательности составляет $190.096$, энергия спектра также равна $190.096$. Разность между ними составляет $2.842\cdot 10^{-14}$, что практически равно нулю и объясняется погрешностью вычислений с плавающей точкой.

В первой части работы была рассчитана импульсная характеристика рекурсивного звена второго порядка, определена его аналитическая формула и исследована реакция системы на входной сигнал. Реакция была вычислена двумя способами: с помощью свёртки и разностного уравнения. Полученные результаты практически полностью совпали, максимальное отличие составило $1{,}33\cdot 10^{-15}$. Также были найдены нули и полюсы передаточной функции аналитически и программно. По их расположению на комплексной плоскости установлено, что система является устойчивой, поскольку оба полюса находятся внутри единичной окружности. Дополнительно были построены амплитудно-частотная и фазочастотная характеристики, определены границы основной полосы частот. 

Во второй части работы была сформирована дискретная последовательность из двух гармонических составляющих с частотами 1250 и 1875 Гц. Для неё было вычислено ДПФ, исследованы амплитудный и фазовый спектры, а также выполнено обратное преобразование Фурье. Максимальная погрешность восстановления исходной последовательности составила $8{,}882\cdot 10^{-16}$. Затем было выполнено восстановление периодического аналогового сигнала по отсчётам ДПФ. Полученный сигнал практически полностью совпал с исходным, максимальное отклонение составило $2{,}731\cdot 10^{-14}$. В завершение была проведена проверка равенства Парсеваля: энергия исходной последовательности и её спектра составила 190,096, что подтвердило выполнение данного равенства.

Таким образом, в ходе работы были изучены основные способы анализа линейных дискретных систем, освоены применение ДПФ и ОДПФ, исследованы частотные характеристики системы и проверены основные свойства преобразования Фурье. Полученные численные результаты подтвердили корректность выполненных вычислений.
