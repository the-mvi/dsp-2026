## Основные теоретические положения


### Линейные дискретные системы

Дискретный сигнал $x(n)$ представляет собой последовательность отсчётов, где $n$ - индекс дискретного времени. Линейная стационарная дискретная система преобразует входное воздействие $x(n)$ в выходную реакцию $y(n)$.

Единичный импульс и единичный скачок определяются следующим образом:

$$
u_0(n)=
\begin{cases}
1, & n=0,\\
0, & n\neq0,
\end{cases}
\qquad
u_1(n)=
\begin{cases}
1, & n\geq0,\\
0, & n<0
\end{cases}
$$

Импульсная характеристика $h(n)$ - реакция системы на $u_0(n)$ при нулевых начальных условиях.

Рекурсивное звено второго порядка описывается разностным уравнением:

$$
y(n)=\sum_{i=0}^{2}b_i x(n-i)-\sum_{k=1}^{2}a_k y(n-k)
$$

Передаточная функция при $a_0=1$:

$$
H(z)=\frac{Y(z)}{X(z)}
=\frac{b_0+b_1z^{-1}+b_2z^{-2}}
{1+a_1z^{-1}+a_2z^{-2}}
$$

Выходную реакцию линейной стационарной системы можно вычислить с помощью свёртки:

$$
y(n)=\sum_{k=-\infty}^{+\infty}x(k)h(n-k)
$$

Для конечных последовательностей длины $N_1$ и $N_2$ длина полной свёртки равна $N_1+N_2-1$.

### Нули, полюсы и частотные характеристики

Нули передаточной функции - корни числителя, полюсы - корни знаменателя после сокращения общих множителей. Комплексные корни можно записать в алгебраической форме $z=x+jy$ или показательной форме $z=re^{j\varphi}$.

Причинная рекурсивная система устойчива, если все её полюсы находятся внутри единичной окружности:

$$
|p_k|<1.
$$

Комплексная частотная характеристика определяется подстановкой $z=e^{j\widehat{\omega}}$ в $H(z)$.

Амплитудно-частотная (АЧХ) и фазочастотная (ФЧХ) характеристики:

$$
A(\widehat{\omega})=|H(e^{j\widehat{\omega}})|,
\qquad
\varphi(\widehat{\omega})=\arg H(e^{j\widehat{\omega}})
$$

Связь нормированной угловой частоты с абсолютной:

$$
\widehat{\omega}=\frac{2\pi f}{f_{\text{d}}},
\qquad
T=\frac{1}{f_{\text{d}}}
$$

Основная двусторонняя полоса частот соответствует интервалу $[-\pi;\pi]$, или $[-f_{\text{d}}/2;f_{\text{d}}/2]$ в герцах.

### Дискретное преобразование Фурье

Дискретное преобразование Фурье (ДПФ) позволяет определить частотный состав последовательности длины $N$:

$$
X(k)=\sum_{n=0}^{N-1}x(n)e^{-j2\pi kn/N}
$$

Обратное ДПФ (ОДПФ) восстанавливает исходную последовательность:

$$
x(n)=\frac{1}{N}\sum_{k=0}^{N-1}X(k)e^{j2\pi kn/N}
$$

Амплитудный спектр периодической последовательности характеризуется модулями коэффициентов ДПФ, фазовый спектр - их аргументами. Для двустороннего представления коэффициенты спектра нормируются на $N$:

$$
A(k)=\frac{|X(k)|}{N},
\qquad
\varphi(k)=\arg X(k)
$$

Абсолютные частоты отсчётов ДПФ определяются соотношением:

$$
f_k=\frac{kf_{\text{d}}}{N}
$$

Периодический аналоговый сигнал можно восстановить по коэффициентам ДПФ:

$$
x(t)=\frac{1}{N}
\sum_{k=-N/2}^{N/2-1}
X_a(k)e^{j2\pi kt/(NT)},
$$

где $X_a(k)=X(N+k)$ при отрицательных $k$ и $X_a(k)=X(k)$ при неотрицательных $k$.

### Равенство Парсеваля

Равенство Парсеваля выражает сохранение энергии сигнала при переходе от временного представления к частотному:

$$
\sum_{n=0}^{N-1}|x(n)|^2
=\frac{1}{N}\sum_{k=0}^{N-1}|X(k)|^2
$$

#### Задание 1. Импульсная характеристика

Для варианта №3 передаточная функция рекурсивного звена второго порядка имеет вид:

$$
H(z)=\frac{0.56+0.579936z^{-1}+0.784z^{-2}}
{1-0.8528z^{-1}+0.658z^{-2}}
$$

Импульсная характеристика $h(n)$ - реакция системы на единичный импульс $u_0(n)$ при нулевых начальных условиях.

Для данного звена она определяется уравнением:

$$
h(n)=0.56u_0(n)+0.579936u_0(n-1)+0.784u_0(n-2)
+0.8528h(n-1)-0.658h(n-2)
$$

Начальные условия: $h(-1)=h(-2)=0$

Первые отсчёты вычисляются по формулам:

$$
h(0)=b_0,\qquad
h(1)=b_1-a_1h(0),
$$

$$
h(2)=b_2-a_1h(1)-a_2h(0)
$$

Для $n\geq3$:

$$
h(n)=-a_1h(n-1)-a_2h(n-2)
$$



Аналитическая формула импульсной характеристики:

$$
h(n)=b_0g(n)+b_1g(n-1)+b_2g(n-2),
$$

где

$$
g(n)=
\begin{cases}
r^n\dfrac{\sin((n+1)\varphi)}{\sin\varphi}, & n\geq0,\\
0, & n<0,
\end{cases}
$$

$$
r=\sqrt{a_2},\qquad
\varphi=\arccos\left(\frac{-a_1}{2r}\right)
$$

$r\approx0.8112$, $\varphi\approx1.017$ рад
 

```python

import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

# Параметры варианта 3
N1 = 23
b = [0.56, 0.579936, 0.784]
a = [1, -0.8528, 0.658]

# Единичный импульс длиной N1
u0 = np.zeros(N1)
u0[0] = 1

# Вычисляем импульсную характеристику
h = signal.lfilter(b, a, u0)

# Строим график
n = np.arange(N1)
plt.figure(figsize=(9, 4))
plt.stem(n, h)
plt.title("Импульсная характеристика")
plt.xlabel("n")
plt.ylabel("h(n)")
plt.grid(True)
plt.show()

print("Первые 5 отсчётов:", np.round(h[:5], 6))
```


**Результат.** Вычислены первые $N_1=23$ отсчёта импульсной характеристики рекурсивного звена второго порядка.

График показывает затухающие колебания. Начальные значения импульсной характеристики: $h(0)=0.56$, $h(1)=1.057504$, $h(2)\approx1.317359$.

**Чему в действительности равна длина импульсной характеристики?**

Импульсная характеристика данного рекурсивного звена имеет бесконечную длину, поскольку выход системы зависит от предыдущих выходных значений. Число $N_1=23$ задаёт только количество вычисленных отсчётов, а не полную длину характеристики.



#### Задание 2. Вычисление реакции по формуле свёртки

В качестве входного воздействия используется прямоугольный импульс длины $N_2=33$:

$$
x(n)=
\begin{cases}
1, & 0\leq n<17,\\
0, & 17\leq n\leq32
\end{cases}
$$

Реакция системы вычисляется по формуле дискретной свёртки:

$$
y_1(n)=\sum_{k=0}^{n}x(k)h(n-k)
$$

Для вычисления используется импульсная характеристика $h(n)$ из первого задания.


```python

# Прямоугольный импульс
N2 = 33
n2 = np.arange(N2)

x = np.zeros(N2)
x[:17] = 1

# Вычисляем свёртку
y_full = np.convolve(x, h)
y1 = y_full[:N2]

print("Длина воздействия:", len(x))
print("Число единичных отсчётов:", int(np.sum(x)))
print("Длина полной свёртки:", len(y_full))
print("Длина ограниченной реакции:", len(y1))

# Графики воздействия и полной свёртки
fig, axs = plt.subplots(2, 1, figsize=(10, 7))

axs[0].stem(n2, x)
axs[0].set_title("Входное воздействие x(n)")
axs[0].set_ylabel("x(n)")

axs[1].stem(np.arange(len(y_full)), y_full)
axs[1].set_title("Реакция y1(n) по формуле свёртки")
axs[1].set_xlabel("n")
axs[1].set_ylabel("y1(n)")

for ax in axs:
    ax.grid(True)

plt.tight_layout()
plt.show()
```

Длина входной последовательности составляет $N_2=33$ отсчёта. При этом ненулевая часть прямоугольного импульса имеет длину 17 отсчётов, от $n=0$ до $n=16$.

Длина полной свёртки двух конечных последовательностей:

$$
N_y=N_1+N_2-1=23+33-1=55.
$$

По графику полная свёртка содержит 55 отсчётов ($n=0,\ldots,54$). Однако последние 16 отсчётов равны нулю, поскольку входной импульс содержит нули в конце. Ненулевая часть свёртки занимает 39 отсчётов.

**3. Почему свёртку ограничивают до длины воздействия?**

Свёртку ограничивают первыми $N_2=33$ отсчётами, чтобы сравнить её с реакцией, вычисленной по разностному уравнению в следующем задании. Оба способа должны давать выходные последовательности одинаковой длины.

#### Задание 6. АЧХ и ФЧХ в шкале нормированных частот

Комплексная частотная характеристика получается из передаточной функции при $z=e^{j\widehat{\omega}}$

$$
H(e^{j\widehat{\omega}})=
\frac{0.56+0.579936e^{-j\widehat{\omega}}+0.784e^{-2j\widehat{\omega}}}
{1-0.8528e^{-j\widehat{\omega}}+0.658e^{-2j\widehat{\omega}}}
$$

АЧХ и ФЧХ:

$$
A(\widehat{\omega})=|H(e^{j\widehat{\omega}})|
$$

$$
\varphi(\widehat{\omega})=\arg H(e^{j\widehat{\omega}})
$$


```python

# Частотные характеристики
w, H = signal.freqz(b, a, worN=4097, include_nyquist=True)

A = np.abs(H)
phi = np.angle(H)

fig, axs = plt.subplots(2, 1, figsize=(10, 7))

axs[0].plot(w / np.pi, A)
axs[0].set_title("АЧХ")
axs[0].set_ylabel("A(ω)")

axs[1].plot(w / np.pi, phi)
axs[1].set_title("ФЧХ")
axs[1].set_xlabel("Нормированная частота ω / π")
axs[1].set_ylabel("φ(ω), рад")

for ax in axs:
    ax.grid(True)

plt.tight_layout()
plt.show()

print("Минимум АЧХ:", round(A.min(), 4),
      "при ω =", round(w[A.argmin()] / np.pi, 4), "π")
print("Максимум АЧХ:", round(A.max(), 4),
      "при ω =", round(w[A.argmax()] / np.pi, 4), "π")
print("A(0) =", round(A[0], 4))
print("A(π) =", round(A[-1], 4))
```

### Часть 2. Дискретное преобразование Фурье

#### Задание 1. Графики периодической последовательности

Последовательность состоит из двух гармонических сигналов:

$$
x(nT)=A_1\cos(2\pi f_1nT+\pi/4)
+A_2\cos(2\pi f_2nT+\pi/8)
$$

Период дискретизации:

$$
T=\frac{1}{8000}=0.000125\text{ с}
$$


```python

# Параметры второй части
N = 64
fd2 = 8000
T = 1 / fd2

A1, A2 = 1.03, 2.06
f1, f2 = 1000, 1500

# Дискретное время
n = np.arange(N)
t = n * T

# Исходная последовательность
x2 = (A1 * np.cos(2 * np.pi * f1 * t + np.pi / 4)
      + A2 * np.cos(2 * np.pi * f2 * t + np.pi / 8))

# Графики
fig, axs = plt.subplots(2, 1, figsize=(10, 7))

axs[0].stem(t, x2)
axs[0].set_title("Дискретное время")
axs[0].set_xlabel("t, с")
axs[0].set_ylabel("x(nT)")

axs[1].stem(n, x2)
axs[1].set_title("Нормированное время")
axs[1].set_xlabel("n")
axs[1].set_ylabel("x(n)")

for ax in axs:
    ax.grid(True)

plt.tight_layout()
plt.show()
```


Построены графики последовательности из 64 отсчётов в двух шкалах времени

Форма сигналов одинаковая, отличается только горизонтальная ось: на первом графике время указано в секундах, на втором - в номерах отсчётов



#### Задание 2. Обратное дискретное преобразование Фурье

Сначала вычисляется ДПФ исходной последовательности:

$$
X(k)=\sum_{n=0}^{N-1}x(n)e^{-j2\pi kn/N}
$$

Затем выполняется обратное ДПФ:

$$
x_{\text{восст}}(n)=\frac{1}{N}\sum_{k=0}^{N-1}X(k)e^{j2\pi kn/N}
$$


```python

# ДПФ и обратное ДПФ
X = np.fft.fft(x2)
x_restored = np.fft.ifft(X).real

# Погрешность восстановления
error = np.max(np.abs(x2 - x_restored))
print("Максимальная погрешность:", error)

plt.figure(figsize=(10, 4))
plt.stem(n, x_restored)
plt.title("Последовательность после обратного ДПФ")
plt.xlabel("n")
plt.ylabel("x(n)")
plt.grid(True)
plt.show()
```


После выполнения ДПФ и ОДПФ полученная последовательность совпадает с исходной

Максимальная погрешность восстановления близка к нулю и связана с округлением чисел при вычислениях

Фазовый спектр:

$$

Нормированная и абсолютная частоты:

$$
\widehat{f}_k=\frac{k}{N},\qquad f_k=\frac{kf_{\text{д}}}{N}

#### Задание 5. Восстановление аналогового сигнала

Восстановление периодического аналогового сигнала по коэффициентам ДПФ:

$$
x(t)=\frac{1}{N}\sum_{k=-N/2}^{N/2-1}
X_a(k)e^{j\frac{2\pi}{NT}kt}
$$

Коэффициенты $X_a(k)$:

$$
X_a(k)=
\begin{cases}
X(N+k), & -N/2\leq k\leq-1\\
X(k), & 0\leq k\leq N/2-1
\end{cases}
$$

Шаг восстановления:

$$
\Delta t=0.25T=0.00003125\text{ с}
$$


```python

# Время с шагом 0.25T
t_new = np.arange(0, (N - 1) * T + T / 8, T / 4)

# Коэффициенты ДПФ с отрицательными частотами
k_full = np.arange(-N // 2, N // 2)
Xa = np.fft.fftshift(X)

# Восстановленный сигнал
x_rest = np.zeros(len(t_new), dtype=complex)

for i in range(N):
    x_rest += Xa[i] * np.exp(1j * 2 * np.pi * k_full[i] * t_new / (N * T))

x_rest = (x_rest / N).real

# Исходный аналоговый сигнал
x_original = (A1 * np.cos(2 * np.pi * f1 * t_new + np.pi / 4)
              + A2 * np.cos(2 * np.pi * f2 * t_new + np.pi / 8))

# Амплитудный спектр восстановленного сигнала
amplitude_full = np.abs(Xa) / N
freq_full = k_full / (N * T)

fig, axs = plt.subplots(3, 2, figsize=(13, 10))

axs[0, 0].stem(n, x2)
axs[0, 0].set_title("Периодическая последовательность x(n)")
axs[0, 0].set_xlabel("n")

axs[0, 1].stem(np.arange(N), np.abs(X))
axs[0, 1].set_title("Модуль ДПФ |X(k)|")
axs[0, 1].set_xlabel("k")

axs[1, 0].plot(t_new, x_rest)
axs[1, 0].set_title("Восстановленный сигнал")
axs[1, 0].set_xlabel("t, с")

axs[1, 1].stem(freq_full, amplitude_full)
axs[1, 1].set_title("Амплитудный спектр восстановленного сигнала")
axs[1, 1].set_xlabel("f, Гц")

axs[2, 0].plot(t_new, x_original)
axs[2, 0].set_title("Исходный аналоговый сигнал")
axs[2, 0].set_xlabel("t, с")

axs[2, 1].axis("off")

for ax in axs.flat:
    ax.grid(True)

plt.tight_layout()
plt.show()

print("Максимальное отличие:",
      np.max(np.abs(x_rest - x_original)))
```

#### Задание 7. Проверка равенства Парсеваля

Равенство Парсеваля для ДПФ:

$$
\sum_{n=0}^{N-1}|x(n)|^2=\frac{1}{N}\sum_{k=0}^{N-1}|X(k)|^2
$$

Левая часть - энергия последовательности во временной области, правая - та же энергия, вычисленная по коэффициентам ДПФ


```python

# Энергия во временной области
E_time = np.sum(np.abs(x2) ** 2)

# Энергия в частотной области
E_freq = np.sum(np.abs(X) ** 2) / N

print(f"Энергия во временной области: {E_time:.6f}")
print(f"Энергия в частотной области: {E_freq:.6f}")
print(f"Разница: {abs(E_time - E_freq):.2e}")

print("Равенство Парсеваля выполняется:",
      np.allclose(E_time, E_freq))
```

Энергия последовательности во временной и частотной областях равна 169.744. Разница между вычисленными значениями близка к нулю

**Смысл равенства Парсеваля**

Равенство Парсеваля показывает, что ДПФ сохраняет энергию сигнала с учётом нормировочного множителя $1/N$. Энергию можно вычислить как по отсчётам сигнала, так и по его спектру
