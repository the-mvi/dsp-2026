Передаточной функцией $H(z)$ ЛДС называется отношение z-преобразования реакции к z-изображению воздействия при нулевых начальных условиях:

$$
H(z) = \frac{Y(z)}{X(z)}
$$

ПФ общего вида:

$$
H(z) = \frac{\sum_{i=0}^{N-1} b_i z^{-i}}{1 + \sum_{k=1}^{M-1} a_k z^{-k}}
$$

Порядок передаточной функции: $\max\{(M - 1), (N - 1)\}$

По определению, импульсная характеристика $h[n]$ — это реакция на цифровой единичный импульс при нулевых начальных условиях

$$
x(n) = u_0(n) = 
\begin{cases} 
1, & n = 0 \\ 
0, & n \neq 0 
\end{cases}
$$

При нулевых начальных условиях ($h[-1] = 0, h[-2] = 0$) разностное уравнение фильтра 2-го порядка для $h[n]$ записывается как:

$$
h(n) = b_0u_0(n) + b_1u_0(n-1) + b_2u_0(n-2) - a_1h(n-1) - a_2h(n-2)
$$

Длина импульсной характеристики: Из-за наличия обратных связей ($a_1 \neq 0, a_2 \neq 0$) текущий отсчет $h(n)$ постоянно опирается на значения прошлых отсчетов $h(n - 1)$ и $h(n - 2)$. Происходит бесконечная рекурсия. Длина математически равна бесконечности ($L = \infty$).

#### Задание 2

Операция дискретной свёртки:

$$
y(n) = x(n) * h(n) = \sum_{k=-\infty}^{\infty} x(k) \cdot h(n - k)
$$

Для сигналов конечной длины верхний предел ограничивается шагом времени $n$.

Длина полной последовательности, получаемой в результате линейной свёртки двух дискретных сигналов конечной длительности вычисляется как:

$$
L = N_1 + N_2 - 1
$$

В формуле свертки, когда входной сигнал $x(n)$ выключается и далее на всем интервале до $N_2 - 1$ равен нулю, выходной сигнал не исчезает мгновенно — продолжает вычислять новые значения, опираясь на свои же предыдущие результаты.

Полная свёртка рассчитывает реакцию до тех пор, пока сигналы полностью не разойдутся. Ограничение до длины $N_2$ (усечение массива) производится для того, чтобы  исключить избыточные расчеты затухания.

#### Задание 4

Передаточную функцию $H(z)$ цифрового фильтра 2-го порядка можно представить в виде отношения двух полиномов от комплексной переменной $z$:

$$
H(z) = \frac{b_0 + b_1 z^{-1} + b_2 z^{-2}}{a_0 + a_1 z^{-1} + a_2 z^{-2}} = \frac{b_0 z^2 + b_1 z + b_2}{a_0 z^2 + a_1 z + a_2}
$$

Нули ($z_k$): — это корни полинома числителя ($b_0 z^2 + b_1 z + b_2 = 0$).  
Полюса ($p_k$): — это корни полинома знаменателя ($a_0 z^2 + a_1 z + a_2 = 0$).

Любое комплексное число на Z-плоскости можно записать в двух эквивалентных формах:

1. Алгебраическая форма (декартовы координаты):

$$
z = x + jy
$$

где $x = \text{Re}(z)$ — вещественная часть, $y = \text{Im}(z)$ — мнимая часть.

2. Показательная форма (полярные координаты):

$$
z = r \cdot e^{j\varphi}
$$

где $r$ — модуль числа, $\varphi$ — его аргумент (угол наклона).

Математическая связь между формами:

Из алгебраической в показательную: Модуль (радиус) находится по теореме Пифагора: $r = \sqrt{x^2 + y^2}$. Аргумент: $\varphi = \text{arctg}(y/x)$.

Из показательной в алгебраическую: $x = r \cdot \cos(\varphi)$, $y = r \cdot \sin(\varphi)$.


Зная корни квадратных уравнений числителя $(z_1, z_2)$ и знаменателя $(p_1, p_2)$, полином можно разложить на линейные множители. Передаточная функция $H(z)$ представляется в виде:

$$
H(z) = \frac{b_0}{a_0} \cdot \frac{(z - z_1)(z - z_2)}{(z - p_1)(z - p_2)} = \frac{b_0}{a_0} \cdot \frac{(1 - z_1 z^{-1})(1 - z_2 z^{-1})}{(1 - p_1 z^{-1})(1 - p_2 z^{-1})}
$$

Подстановка нулей и полюсов в показательной форме $(z_k = r_z \cdot e^{j\varphi_z}, p_k = r_p \cdot e^{j\varphi_p})$:

$$
H(z) = \frac{b_0}{a_0} \cdot \frac{(1 - r_{z1} e^{j\varphi_{z1}} z^{-1})(1 - r_{z2} e^{j\varphi_{z2}} z^{-1})}{(1 - r_{p1} e^{j\varphi_{p1}} z^{-1})(1 - r_{p2} e^{j\varphi_{p2}} z^{-1})}
$$

Критерий устойчивости:   
Дискретная линейная рекурсивная система является устойчивой тогда и только тогда, когда все её полюса строго лежат внутри единичной окружности.

Модули всех полюсов $p_k$ должны быть меньше единицы:

$$
|p_k| < 1
$$

Расположение нулей на устойчивость системы никак не влияет. Нули могут находиться как внутри, так и вне единичной окружности.

#### Задание 6

Частотная характеристика

Комплексный коэффициент передачи $H(e^{j\hat{\omega}})$ описывает изменение амплитуды и фазы гармонических сигналов при их прохождении через линейную дискретную систему.

Математически он получается из передаточной функции $H(z)$ путем подстановки комплексной переменной $z = e^{j\hat{\omega}}$ (переход из Z-области на единичную окружность):

$$
H(e^{j\hat{\omega}}) = \frac{b_0 + b_1e^{-j\hat{\omega}} + b_2e^{-j2\hat{\omega}}}{a_0 + a_1e^{-j\hat{\omega}} + a_2e^{-j2\hat{\omega}}}
$$

где $\hat{\omega}$ — нормированная частота, выраженная в радианах на отсчет (рад/отсчет).

Амплитудно-частотная (АЧХ) и фазочастотная (ФЧХ) характеристики

Комплексную функцию $H(e^{j\hat{\omega}})$ можно представить в полярной форме как комплексное число:

$$
H(e^{j\hat{\omega}}) = A(\hat{\omega}) \cdot e^{j\phi(\hat{\omega})}
$$

Амплитудно-частотная характеристика (АЧХ) — $A(\hat{\omega})$: Это модуль комплексного коэффициента передачи ($|H(e^{j\hat{\omega}})|$). Она показывает, во сколько раз изменяется амплитуда входной синусоиды с частотой $\hat{\omega}$ после прохождения через фильтр.

Фазочастотная характеристика (ФЧХ) — $\phi(\hat{\omega})$: Это аргумент (полярный угол) комплексного коэффициента передачи ($\text{arg } H(e^{j\hat{\omega}})$). Она определяет фазовый сдвиг (временное смещение), который фильтр вносит в гармонический сигнал на частоте $\hat{\omega}$.

Спектр дискретных сигналов и частотные характеристики цифровых фильтров всегда являются периодическими функциями с периодом $2\pi$. Это напрямую следует из периодичности комплексной экспоненты: $e^{-j(\hat{\omega}+2\pi)} = e^{-j\hat{\omega}}$.

Границы основной полосы частот (интервал уникальности спектра) лежат в пределах от $0$ до $\pi$ радиан на отсчет.

Геометрическая связь карты нулей/полюсов с формой АЧХ:

Расчет частотной характеристики геометрически эквивалентен обходу единичной окружности на комплексной Z-плоскости. Коэффициент передачи фильтра в любой точке частоты $\hat{\omega}$ обратно пропорционален расстоянию от этой точки до полюсов системы и прямо пропорционален расстоянию до её нулей:

Влияние полюсов: Чем ближе текущая точка единичной окружности (под углом $\hat{\omega}$) подходит к полюсу системы, тем меньше становится знаменатель передаточной функции. Это приводит к возникновению резонансного пика (подъема) на графике АЧХ.

Влияние нулей: Чем ближе единичная окружность подходит к нулю системы, тем сильнее уменьшается числитель передаточной функции. В этой частотной области на графике АЧХ возникает провал (минимум). Если нуль лежит прямо на окружности ($r = 1$), АЧХ падает до абсолютного нуля.

Если график АЧХ в какой-то точке частоты касается нуля ($A(\hat{\omega}) = 0$), на графике ФЧХ в этой же точке возникает вертикальный скачок (разрыв) на $\pi$ радиан $(180^\circ)$.

#### Задание 7


Частота дискретизации $f_\text{д}$ (Гц) — количество отсчетов, фиксируемых системой за одну секунду. Период дискретизации (интервал времени между двумя соседними отсчетами) равен $T = 1/f_\text{д}$.

Связь между нормированной частотой $\hat{\omega}$ и физической частотой $f$ (в Гц) задается фундаментальным соотношением:

$$
\hat{\omega} = \omega T = \frac{2\pi \cdot f}{f_\text{д}}
$$

Отсюда физическая частота $f$ в Герцах выражается как:

$$
f = \frac{\hat{\omega} \cdot f_\text{д}}{2\pi}
$$

В шкале нормированных частот область уникальности спектра лежит в пределах от 0 до $\pi$ радиан на отсчет.
Подставляя граничное значение $\hat{\omega} = \pi$ в формулу связи, получаем верхнюю границу:

$$
f_{max} = \frac{\pi \cdot f_\text{д}}{2\pi} = \frac{f_\text{д}}{2}
$$

Таким образом, в шкале физических частот границы основной полосы частот всегда строго равны интервалу от 0 до $f_\text{д}/2$ Гц.

T = \frac{1}{f_\text{д}}

#### Задание 2

ДПФ и ОДПФ:

Дискретное преобразование Фурье (ДПФ) переводит дискретный сигнал из временной области в комплексную частотную область, раскладывая его на сумму комплексных экспонент:

$$
X(k) = \sum_{n=0}^{N-1} x(n)e^{-j\frac{2\pi}{N}kn}, \quad k=0,...,N-1
$$

Обратное ДПФ (ОДПФ) — полностью восстанавливает исходную временную последовательность отсчетов по её комплексному спектру $X(k)$:

$$
x(n) = \frac{1}{N} \sum_{k=0}^{N-1} X(k)e^{j\frac{2\pi}{N}kn}, \quad n=0,...,N-1
$$

ДПФ и ОДПФ представляют собой линейные взаимно-однозначные взаимно-обратные ортогональные преобразования. Это означает, что при прямом и последующем обратном преобразовании информация о сигнале не теряется и не искажается.

#### Задание 3

Дискретное нормированное время n — это абстрактный безразмерный целочисленный шаг ($0, 1, 2...N - 1$).

Дискретное время $t_n = nT$ (выраженное в секундах) — это реальное физическое время, жестко привязанное к физическому миру через период дискретизации $T = 1/f_\text{д}$.

Перейти от абстрактной шкалы к абсолютной можно обычным умножением номера отсчета на период дискретизации: $t = n \cdot T$.

Дискретная нормированная частота k — это номер спектральной составляющей (индекс спектрального отсчета от $0$ до N - 1.

Абсолютная частота f указывает на реальную физическую частоту анализируемой гармоники.

Шаг сетки ДПФ по частоте равен $\Delta f = f_д/N$. Соответственно, любая физическая частота f линейно пересчитывается через спектральный индекс k по формуле:

$$
f = k \cdot \Delta f = \frac{k \cdot f_д}{N}
$$

Поскольку исходный сигнал $x(n)$ является строго вещественным, его амплитудный спектр строго симметричен относительно частоты $f_\text{д}/2$, отсчет $N/2$, а фазовый спектр нечетен относительно этого же значения. Первая половина графиков содержит реальные частоты сигнала, а вторая половина — их зеркальное комплексное отображение.

#### Задание 7

Теорема Парсеваля:
Для дискретных последовательностей конечной длины $N$ равенство Парсеваля связывает энергию сигнала во временной области с его энергетическим спектром в частотной области и записывается в виде:

$$
\sum_{n=0}^{N-1} |x(n)|^2 = \frac{1}{N} \sum_{k=0}^{N-1} |X(k)|^2
$$

где $x(n)$ — временные отсчеты сигнала, а $X(k)$ — комплексные отсчеты его ДПФ.

Теорема Парсеваля выражает закон сохранения энергии сигнала при его переходе из одной области представления в другую. Она утверждает, что полная энергия сигнала не зависит от способа её измерения: рассчитываем ли мы её как сумму мгновенных мощностей отсчетов во временной области или как сумму мощностей всех частотных гармоник (спектральных составляющих) в частотной области.

Вычислить импульсную характеристику длины $N_1$ по заданной передаточной функции с коэффициентами $b_i$, $a_k$ и построить её график. Записать аналитическую формулу импульсной характеристики рекурсивного звена 2-го порядка с учетом нулевых начальных условий. Пояснить, чему в действительности равна длина импульсной характеристики.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

N_var = 17
b0 = 0.5 + 0.02 * N_var
b1 = b0 * ((-1)**(N_var + 1)) * (0.9822 + 0.0178 * N_var)
b2 = b0 * (0.8 + 0.2 * (N_var % 5))
a0 = 1
a1 = ((-1)**N_var) * (0.7778 + 0.025 * N_var)
a2 = 0.64 + 0.006 * N_var
N1 = (N_var % 10) + 20

print(f"коэффициенты числителя (b): [{b0:.4f}, {b1:.6f}, {b2:.4f}]")
print(f"коэффициенты знаменателя (a): [{a0}, {a1:.4f}, {a2:.4f}]")
print(f"длина N1: {N1}\n")

#вычисление импульсной характеристики
x = np.zeros(N1)
x[0] = 1.0

#y(n) = (b0 * x(n) + b1 * x(n-1) + b2 * x(n-2) - a1 * y(n-1) - a2 * y(n-2)) / a0
h = np.zeros(N1)

for n in range(N1):
    x_n   = x[n]
    x_n_1 = x[n-1] if (n - 1) >= 0 else 0.0
    x_n_2 = x[n-2] if (n - 2) >= 0 else 0.0

    y_n_1 = h[n-1] if (n - 1) >= 0 else 0.0
    y_n_2 = h[n-2] if (n - 2) >= 0 else 0.0
    
    h[n] = (b0 * x_n + b1 * x_n_1 + b2 * x_n_2 - a1 * y_n_1 - a2 * y_n_2) / a0
    
print("Значения отсчетов импульсной характеристики h(n):")
for n, val in enumerate(h):
    print(f"h[{n}] = {val:10.6f}")

#построение графика
plt.figure(figsize=(10, 5))
plt.stem(range(N1), h, basefmt="b-", linefmt='b-', markerfmt='bo')
plt.title('импульсная характеристика')
plt.xlabel('отсчеты n')
plt.ylabel('h(n)')
plt.grid(True)
plt.xticks(range(0, N1, 2))
plt.show()
```

Аналитическая формула импульсной характеристики:

С учетом ННУ и того, что воздействием является единичный импульс $x(n) = u_0(n)$, разностное уравнение рекурсивного звена 2-го порядка принимает вид:

$$
h(n) = 0.84u_0(n) + 1.079232u_0(n - 1) + 1.008u_0(n - 2) + 1.2028h(n - 1) - 0.742h(n - 2)
$$

В действительности длина импульсной характеристики данного фильтра бесконечна ($L = \infty$).

Из-за наличия обратных связей (коэффициенты знаменателя $a_1 = -1.2028$ и $a_2 = 0.742$ отличны от нуля) каждый новый отсчет $h[n]$ рассчитывается рекурсивно на основе двух предыдущих выходных значений. Из-за этого процесс математически продолжается бесконечно.

####  Задача 2

Вычислить реакцию $y_1(n)$ по формуле свёртки. В качестве воздействия $x(n)$ длины $N_2$ выбрать дискретный прямоугольный импульс: 
$$
x(n) = 
\begin{cases}
1, & 0 \leqslant n < \text{int}(N_2/2) \\
0, & \text{int}(N_2/2) \leqslant n \leqslant N_2 - 1
\end{cases}
$$
где функция int(t) – округление до ближайшего целого. Вывести графики воздействия и реакции. Пояснить:  
a. чему равна длина импульса x;  
b. чему равна длина свёртки аналитически и по графику;  
c. почему её ограничивают до длины воздействия.

```python
N2 = (N_var % 10) + 30 

x = np.zeros(N2)
border = int(np.round(N2 / 2)) 
x[0:border] = 1.0

L_full = N1 + N2 - 1
y1_full = np.zeros(L_full)

print(f"длина вектора N2 = {N2}")
print(f"граница активной части (border) = {border}")
print(f"длина свертки = {L_full}\n")

#расчет формулы свёртки
for n in range(L_full):
    sum_val = 0.0
    for k in range(N2):
        h_idx = n - k
        if 0 <= h_idx < N1:
            sum_val += x[k] * h[h_idx]
            
    y1_full[n] = sum_val

y1_limited = y1_full[:N2]

for n_idx, val in enumerate(y1_full):
    print(f"y1[{n_idx:2d}] = {val:10.6f}")

plt.figure(figsize=(12, 10))
#график воздействия x(n)
plt.subplot(3, 1, 1)
plt.stem(range(N2), x, basefmt="g-", linefmt='g-', markerfmt='go')
plt.title(f'воздействие x(n)')
plt.xlabel('n')
plt.ylabel('x(n)')
plt.grid(True)

#график полной реакции y1(n)
plt.subplot(3, 1, 2)
plt.stem(range(L_full), y1_full, basefmt="b-", linefmt='b-', markerfmt='bo')
plt.title(f'полная реакция y1(n)')
plt.xlabel('n')
plt.ylabel('y1(n)')
plt.grid(True)

#график ограниченной реакции
plt.subplot(3, 1, 3)
plt.stem(range(N2), y1_limited, basefmt="r-", linefmt='r-', markerfmt='ro')
plt.title(f'ограниченная реакция y1(n)')
plt.xlabel('n')
plt.ylabel('y1(n)')
plt.grid(True)

plt.tight_layout()
plt.show()
```

a. Полная длина вектора воздействия $x(n)$ равна 37 отсчетов, при этом длительность прямоугольного импульса составляет 19 отсчетов.  
b. Полная длина линейной свёртки равна 63 отсчетам.

#### Задача 4

Вычислить нули и полюсы передаточной функции. Нули и полюсы в алгебраической и показательной формах и пояснить связь между ними. Выразить значение аргумента полюса и нуля относительно $\pi$. Представить передаточную функцию в виде произведения простейших множителей с нулями и полюсами в показательной форме.

```python
numerator_coeffs = [b0, b1, b2]
denominator_coeffs = [a0, a1, a2]

zeros = np.roots(numerator_coeffs)
poles = np.roots(denominator_coeffs)

def analyze_complex_points(points, label):
    for i, p in enumerate(points, 1):
        real_part = np.real(p)
        imag_part = np.imag(p)
        
        r = np.abs(p)
        phi = np.angle(p)
        
        phi_over_pi = phi / np.pi
        
        print(f"{label} {i}:")
        print(f"алгебраическая форма: {real_part:.4f} + ({imag_part:.4f})j")
        print(f"показательная форма:  {r:.4f} * exp(j * ({phi_over_pi:.4f}*π))")
        print(f"модуль (r) = {r:.4f}, аргумент (φ) = {phi_over_pi:.4f}*π рад.\n")

analyze_complex_points(zeros, "ноль (z)")
analyze_complex_points(poles, "полюс (p)")
```

Нули передаточной функции (корни числителя):  
В алгебраической форме: $z_{1,2} = -0.6424 \pm 0.8873j$.  
В показательной форме: $z_{1,2} = 1.0954 \cdot e^{\pm j0.6995\pi}$.   

Полюсы передаточной функции (корни знаменателя):  
В алгебраической форме: $p_{1,2} = 0.6014 \pm 0.6167j$.  
В показательной форме: $p_{1,2} = 0.8614 \cdot e^{\pm j0.2540\pi}$.  

Связь между алгебраической и показательной формами корней:  
Вычисленные нули и полюса представляют собой комплексно-сопряженные пары чисел. Переход между формами осуществляется на основе тождества $(z = x + jy = r \cdot e^{j\varphi})$:

Алгебраическая форма ($x + jy$) определяет положение корней в декартовых координатах на комплексной плоскости, где вещественная часть ($Re$) задает положение по горизонтальной оси, а мнимая ($Im$) — по вертикальной. Для полюсов это $0.6014 \pm 0.6167j$, для нулей — $-0.6424 \pm 0.8873j$.

Показательная форма ($r \cdot e^{j\varphi}$) задает положение тех же точек в полярных координатах. Модуль ($r$) — это расстояние от начала координат до корня. Для полюсов $r = 0.8614$ (они находятся внутри единичного круга, что говорит об устойчивости), а для нулей $r = 1.0954$ (выходят за пределы круга).

Математическая связь: Модуль рассчитывается через гипотенузу $r = \sqrt{x^2 + y^2}$, а аргумент (угол $\varphi$) — через арктангенс $\varphi = \text{arctg}(y/x)$ с учетом знаков четверти плоскости.

Аргумент комплексного корня имеет смысл нормированной частоты $\hat{\omega}$, которая на комплексной плоскости изменяется от 0 до $\pi$.

Для полюсов $(p_{1,2})$: Значение аргумента составляет $\varphi = \pm 0.2540\pi$ рад. Наличие знака $\pm$ обусловлено симметрией относительно вещественной оси.

Для нулей $(z_{1,2})$: Значение аргумента составляет $\varphi = \pm 0.6995\pi$ рад. 

На основе корней, передаточная функция рекурсивного звена 2-го порядка раскладывается на простейшие множители следующей формулой:

$$
H(z) = 0.8400 \cdot \frac{(1 - 1.0954e^{j0.6995\pi}z^{-1})(1 - 1.0954e^{-j0.6995\pi}z^{-1})}{(1 - 0.8614e^{j0.2540\pi}z^{-1})(1 - 0.8614e^{-j0.2540\pi}z^{-1})}
$$

Можно также записать в чисто вещественном виде через тригонометрические функции косинуса:

$$
H(z) = 0.8400 \cdot \frac{1 - 2 \cdot 1.0954 \cdot \cos(0.6995\pi)z^{-1} + 1.0954^2 z^{-2}}{1 - 2 \cdot 0.8614 \cdot \cos(0.2540\pi)z^{-1} + 0.8614^2 z^{-2}}
$$

#### Задача 6


Вычислить амплитудно-частотную (АЧХ) и фазочастотную (ФЧХ) характеристики в шкале нормированных частот $\hat{\omega}$ и вывести их графики. Пояснить:  
a. чему равны границы основной полосы частот;  
b. соответствие между картой нулей и полюсов и видом АЧХ;  
c. какому значению АЧХ соответствует скачок на $\pi$, если он имеется;  
d. какие частотные составляющие воздействия, низкие или высокие, оказались преимущественно подавленными в реакции.

```python
num_points = 1000
w = np.linspace(0, np.pi, num_points)

h_freq = np.zeros(num_points, dtype=complex)

for idx, w_val in enumerate(w):
    #вычисление комплексных экспонент экспоненциальной задержки
    e_jw1 = np.exp(-1j * w_val)
    e_jw2 = np.exp(-1j * 2 * w_val)
    
    # комплексный числитель и знаменатель
    numerator = b0 + b1 * e_jw1 + b2 * e_jw2
    denominator = a0 + a1 * e_jw1 + a2 * e_jw2
    
    #значение H(e^jw)
    h_freq[idx] = numerator / denominator

achx = np.abs(h_freq)
fchx = np.unwrap(np.angle(h_freq))

plt.figure(figsize=(12, 8))
#график АЧХ
plt.subplot(2, 1, 1)
plt.plot(w, achx, 'b-', linewidth=2)
plt.title('АЧХ')
plt.xlabel(r'$\hat{\omega}$ (рад/отсчет)')
plt.ylabel(r'$|H(\hat{\omega})|$')
plt.xticks([0, np.pi/4, np.pi/2, 3*np.pi/4, np.pi], ['0', r'$\pi/4$', r'$\pi/2$', r'$3\pi/4$', r'$\pi$'])
plt.grid(True)
#график ФЧХ
plt.subplot(2, 1, 2)
plt.plot(w, fchx, 'r-', linewidth=2)
plt.title('ФЧХ')
plt.xlabel(r'$\hat{\omega}$ (рад/отсчет)')
plt.ylabel(r'$\phi(\hat{\omega})$ (рад)')
plt.xticks([0, np.pi/4, np.pi/2, 3*np.pi/4, np.pi], ['0', r'$\pi/4$', r'$\pi/2$', r'$3\pi/4$', r'$\pi$'])
plt.grid(True)

plt.tight_layout()
plt.show()
```

а. Границы основной полосы частот дискретной системы лежат в пределах от 0 до $\pi$ радиан на отсчет.  
Из-за периодичности спектра дискретного сигнала (с периодом $2\pi$), вся уникальная информация о частотных свойствах фильтра содержится на интервале $[0, \pi]$. Все частоты за пределами этого интервала являются зеркальным повторением основного спектра.
b. Положение полюсов задает резонансные пики (подъемы) на АЧХ, а положение нулей определяет провалы (минимумы) АЧХ.  

Физический смысл частотной характеристики на Z-плоскости — это движение по дуге единичной окружности под углом $\hat{\omega }$. В п.5 угол полюса равен $\varphi_p = \pm 0.2540\pi$. При прохождении частоты $\hat{\omega} \approx 0.254\pi$ единичная окружность сближается с полюсом, расстояние до него минимально, что математически вызывает сильный подъем АЧХ. Угол нуля равен $\varphi_z = \pm 0.6995\pi$. В области этой частоты окружность приближается к нулю, вызывая затухание АЧХ.    

c. Скачок фазы на $\pi$ радиан происходит строго в тех точках частоты, где АЧХ падает точно до нуля.  
Поскольку нули фильтра лежат вне единичной окружности, траектория частотного отклика на комплексной плоскости лишь проходит рядом с началом координат, но не пересекает точку $(0,0)$. Из-за этого на частоте минимума $\hat{\omega}$ синяя линия АЧХ опускается к оси x, но не становится равной нулю, поэтому фаза остается без скачка.  

d. Преимущественно подавленными оказались высокие частоты.  
Так как полюса сосредоточены в низкочастотной зоне $(\approx 0.254\pi)$, а нули сдвинуты в высокочастотную часть $(\approx 0.699\pi)$, график АЧХ имеет форму, близкую к фильтру нижних частот (ФНЧ) или полосовому низкочастотному фильтру. Коэффициент передачи фильтра на частотах, близких к $\pi$, значительно меньше, чем на частотах вблизи нуля, что и приводит к эффективному подавлению высокочастотных компонент прямоугольного импульса.

#### Задача 7

Вывести графики последовательности $x(n)$ на периоде $N$ в шкале дискретного и дискретного нормированного времени:

$$
 x(nT) = A_1 \cos(2\pi f_1 nT + \pi/4) + A_2 \cos(2\pi f_2 nT + \pi/8)

```python
N_var = 17
N = 64

fd = 2000 * (N_var % 5 + 1)  
T = 1 / fd  

A1 = 1 + 0.01 * N_var 
A2 = 2 * A1

f1 = fd / 8 
f2 = 1.5 * f1

print(f"N = {N}")
print(f"fd = {fd} Гц")
print(f"T = {T:.6f} c")
print(f"амплитуды: A1 = {A1:.2f}, A2 = {A2:.2f}")
print(f"частоты: f1 = {f1:.1f} Гц, f2 = {f2:.1f} Гц\n")

n = np.arange(N)

#x(nT)=A1cos(2*pi*f1*n*T+pi/4)+A2cos(2*pi*f2*n*T+pi/8) <=> x(nT)=A1cos(2*pi*f1/fд+pi/4)+A2cos(2*pi*f2/fд+pi/8)
x_n = A1 * np.cos(2 * np.pi * f1 * n * T + np.pi/4) + A2 * np.cos(2 * np.pi * f2 * n * T + np.pi/8)

for idx, val in enumerate(x_n):
    t_val = idx * T  # Физическое время t = n*T
    print(f"n = {idx} | t = nT = {t_val:10.6f} c | x(nT) = {val:10.6f}")
print("\n")

plt.figure(figsize=(12, 10))
#график в шкале дискретного нормированного времени
plt.subplot(2, 1, 1)
plt.stem(n, x_n, basefmt="b-", linefmt='b-', markerfmt='bo')
plt.title('последовательность x(n) в шкале дискретного нормированного времени')
plt.xlabel('нормированное время n (отсчеты)')
plt.ylabel('x(n)')
plt.grid(True)
#график в шкале дискретного времени (в физических секундах t = n*T)
plt.subplot(2, 1, 2)
plt.stem(n * T, x_n, basefmt="m-", linefmt='m-', markerfmt='mo')
plt.title('последовательность x(t) в шкале дискретного времени')
plt.xlabel('физическое время t (секунды)')
plt.ylabel('x(nT)')
plt.grid(True)

plt.tight_layout()
plt.show()
```

#### Задача 2

Вычислить ОДПФ от ДПФ последовательности $x(n)$ и вывести график полученной последовательности в шкале дискретного нормированного временах.

```python
#X(k) = sum_{n=0}^{N-1} x(n) * exp(-j * 2 * pi * k * n / N)
X_k = np.zeros(N, dtype=complex)

for k in range(N):
    sum_val = 0.0 + 0.0j
    for idx_n in range(N):
        #экспоненциальный множитель поворота
        angle = -2 * np.pi * k * idx_n / N
        sum_val += x_n[idx_n] * (np.cos(angle) + 1j * np.sin(angle))
    X_k[k] = sum_val

# x_rec(n) = (1 / N) * sum_{k=0}^{N-1} X(k) * exp(j * 2 * pi * k * n / N)
x_rec = np.zeros(N, dtype=complex)

for idx_n in range(N):
    sum_val = 0.0 + 0.0j
    for k in range(N):
        angle = 2 * np.pi * k * idx_n / N
        sum_val += X_k[k] * (np.cos(angle) + 1j * np.sin(angle))
    x_rec[idx_n] = sum_val / N

x_rec_real = np.real(x_rec)

for n_idx in range(N):
    print(f"n = {n_idx:2d} | Исходный x = {x_n[n_idx]:10.6f} | Восстановленный x_rec = {x_rec_real[n_idx]:10.6f}")
print(f"максимальное расхождение векторов: {np.max(np.abs(x_n - x_rec_real)):.2e}\n")

plt.figure(figsize=(12, 5))
plt.stem(range(N), x_rec_real, basefmt="b-", linefmt='b-', markerfmt='bo')
plt.title('последовательность, восстановленная с помощью ОДПФ')
plt.xlabel('нормированное время n (отсчеты)')
plt.ylabel('x_rec(n)')
plt.grid(True)
plt.show()
```

Полученный график последовательности после выполнения ОДПФ абсолютно идентичен первому графику исходного сигнала $x(n)$, значения амплитуд и фаз совпадают.

Максимальная разность между отсчетами исходного массива $x_n$ и массива после ОДПФ составляет около $10^{-14}$, что вызвано погрешностью округления чисел с плавающей точкой.

#### Задача 5

Восстановить периодический аналоговый сигнал $x(t)$ по отсчётам ДПФ $X(k)$ периодической последовательности $x(n)$. Для вычисления значений сигнала $x(t)$ использовать формулу:

$$
x(t) = \frac{1}{N} \sum_{k=-N/2}^{N/2-1} X_a(k) \exp \left( j \frac{2\pi}{NT} kt \right),
$$

где отсчеты \(X_a(k)\) связаны с отсчетами ДПФ $X(k)$ соотношением:

$$
X_a(k) = 
\begin{cases} 
X(N + k), & -N/2 \leqslant k \leqslant -1 \\ 
X(k), & 0 \leqslant k \leqslant N/2 - 1 
\end{cases}
$$

задавая значения времени \(t\) на интервале $t \in [0; (N - 1)T]$ с шагом $\Delta t = 0.25T$. Вывести графики:  
а. периодической последовательности $x(n)$ и модуля ее ДПФ;  
b. восстановленного аналогового сигнала $x(t)$ и его амплитудного спектра;  
c. исходного аналогового сигнала $x(t) = x(nT)|_{nT=t}$.

```python
dt = 0.25 * T
t_max = (N - 1) * T
t_axis = np.arange(0, t_max + dt/2, dt)

#формирование смещенного спектра X_a(k) для k от -N/2 до N/2 - 1
half_N = N // 2
X_a = np.zeros(N, dtype=complex)
k_shifted = np.arange(-half_N, half_N)

for idx, k in enumerate(k_shifted):
    #вторая половина ДПФ переходит в область отрицательных частот
    if -half_N <= k <= -1:
        X_a[idx] = X_k[N + k]   
    elif 0 <= k <= half_N - 1:
        X_a[idx] = X_k[k]

#восстановление аналогового сигнала x(t)
x_recovered = np.zeros(len(t_axis), dtype=complex)

for idx_t, t_val in enumerate(t_axis):
    sum_val = 0.0 + 0.0j
    for idx_k, k in enumerate(k_shifted):
        #аргумент комплексной экспоненты: j * (2 * pi * k * t) / (N * T)
        angle = (2 * np.pi * k * t_val) / (N * T)
        sum_val += X_a[idx_k] * (np.cos(angle) + 1j * np.sin(angle))
    x_recovered[idx_t] = sum_val / N

x_recovered_real = np.real(x_recovered)

#расчет исходного непрерывного аналогового сигнала
x_original = A1 * np.cos(2 * np.pi * f1 * t_axis + np.pi/4) + A2 * np.cos(2 * np.pi * f2 * t_axis + np.pi/8)

plt.figure(figsize=(14, 12))
#периодическая последовательность x(n) и модуль ее ДПФ
plt.subplot(3, 2, 1)
plt.stem(range(N), x_n, basefmt="b-", linefmt='b-', markerfmt='bo')
plt.title(r'периодическая последовательность $x(n)$')
plt.xlabel('n (отсчеты)')
plt.grid(True)
plt.subplot(3, 2, 2)
plt.stem(range(N), np.abs(X_k), basefmt="g-", linefmt='g-', markerfmt='go')
plt.title(r'модуль ДПФ $|X(k)|$')
plt.xlabel('k (отсчеты частоты)')
plt.grid(True)
#восстановленный аналоговый сигнал x(t) и его амплитудный спектр
plt.subplot(3, 2, 3)
plt.plot(t_axis, x_recovered_real, 'm-', linewidth=2)
plt.title(r'восстановленный аналоговый сигнал $x(t)$ ($\Delta t = 0.25T$)')
plt.xlabel('t (секунды)')
plt.grid(True)

plt.subplot(3, 2, 4)
#для амплитудного спектра x(t) выводим масштабированные коэффициенты |X_a(k)| / N
plt.stem(k_shifted * (fd / N), np.abs(X_a) / N, basefmt="r-", linefmt='r-', markerfmt='ro')
plt.title('амплитудный спектр восстановленного сигнала (в Гц)')
plt.xlabel('f (Гц)')
plt.grid(True)

#график исходного непрерывного аналогового сигнала
plt.subplot(3, 2, 5)
plt.plot(t_axis, x_original, 'c--', linewidth=2, label='исходный аналоговый')
plt.plot(t_axis, x_recovered_real, 'm:', linewidth=1.5, label='восстановленный')
plt.title(r'исходный аналоговый сигнал $x(t) = x(nT)|_{nT=t}$')
plt.xlabel('t (секунды)')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

#### Задача 6

Исходя из результатов п. 5, объяснить:  
a. связь модуля ДПФ последовательности со спектром аналогового сигнала;  
b. результат визуального сравнения восстановленного и исходного сигналов.

a. Отсчеты модуля ДПФ дискретного сигнала $|X(k)|$ пропорциональны длине выборки N.  ДПФ вычисляет сумму (накопленный сигнал), а аналоговый спектр — среднее значение, поэтому и появляется множитель 1/N. В то же время комплексный амплитудный спектр аналогового сигнала оперирует физическими средними значениями амплитуд $|C_k|$. Связь между ними выражается через масштабный множитель $1/N$:

$$
|C_k| = \frac{|X(k)|}{N}
$$

Поскольку исходный косинусоидальный сигнал раскладывается на две комплексные экспоненты (с положительной и отрицательной частотой), амплитуда спектральной линии $|C_k|$ на графике ровно в 2 раза меньше истинной амплитуды A аналоговой гармоники ($|C_k| = A/2$). На графиках из п. 5 это подтверждается: пик на частоте 750 Гц имеет высоту 0.585 (что равно $1.17 / 2$), а пик на частоте 1125 Гц равен 1.17 (что равно $2.34 / 2$).

Каждому дискретному частотному шагу k соответствует строго определенная физическая частота f в Герцах, зависящая от частоты дискретизации $f_\text{д}$ и числа точек N:

$$
f = \frac{k \cdot f_\text{д}}{N}
$$

b. График непрерывного аналогового сигнала $x(t)$, идеально восстановленного из спектральных точек с мелким шагом $\Delta t = 0.25T$, и график исходного аналитического сигнала, рассчитанного теоретически по исходной формуле двух гармоник (пункт c), идентичны.

Это объясняется тем, что частоты исходных гармоник (750 Гц и 1125 Гц) попали точно в узлы частотной сетки ДПФ. В спектре в окно наблюдения укладывается целое число периодов сигнала. Кроме того, условие теоремы Котельникова выполнено с запасом ($6000 > 2 \times 1125$), что гарантирует восстановление непрерывной функции из дискретных коэффициентов Фурье.

#### Задача 7

Проверить равенство Парсеваля. Пояснить его смысл.

```python
#вычисление энергии сигнала во временной облости
energy_time = np.sum(np.abs(x_n)**2)

#вычисление энергии сигнала в частотной области
energy_freq = np.sum(np.abs(X_k)**2) / N

print(f"Энергия во временной области (E_time) = {energy_time:.6f}")
print(f"Энергия в частотной области  (E_freq) = {energy_freq:.6f}")

#нахождение разности
diff = np.abs(energy_time - energy_freq)
print(f"Абсолютное расхождение: {diff:.2e}")

if np.isclose(energy_time, energy_freq, atol=1e-10):
    print("\nравенство выполняется")
else:
    print("\nобнаружено расхождение")
```

Результат проверки равенства:
Выполненный программный расчет подтверждает справедливость теоремы. Полная энергия дискретной последовательности во временной области $E_{time}$ и суммарная энергия её спектральных компонент $E_{freq}$ совпали до 13 знака после запятой.

Численная оценка погрешности:
Абсолютное расхождение между двумя вычисленными значениями энергий находится в пределах $10^{-13} - 10^{-14}$. Наличие этой разницы обусловлено аппаратными погрешностями.

Теорема выражает фундаментальный закон сохранения энергии сигнала при переходе из временной области в частотную. Она означает, что полная энергия сигнала не зависит от способа её представления: будь то сумма мощностей мгновенных отсчетов во времени или сумма мощностей всех спектральных гармоник. Преобразование Фурье лишь перераспределяет энергию между отдельными частотами, но не создает и не уничтожает её.
