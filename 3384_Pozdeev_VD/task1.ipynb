## Основные теоретические сведения

**Линейная дискретная система (ЛДС)** — система, преобразующая входную последовательность $x(n)$ в выходную $y(n)$ и обладающая свойствами линейности, стационарности (инвариантности к сдвигу) и физической реализуемости (отклик не опережает воздействие).

**Импульсная характеристика (ИХ)** $h(n)$ — реакция ЛДС на цифровой единичный импульс $u_0(n)$ при нулевых начальных условиях. ИХ рекурсивной ЛДС — бесконечно затухающая последовательность, если полюсы системы лежат внутри единичной окружности.

**Формула свёртки** — реакция ЛДС с известной ИХ на произвольное воздействие:
$$y(n) = \sum_{k=0}^{n} x(k)\,h(n-k).$$
Свёртка последовательностей конечных длин $N_1$ и $N_2$ имеет длину $N_1+N_2-1$.

**Разностное уравнение** рекурсивного звена 2-го порядка:
$$a_0\,y(n) + a_1\,y(n-1) + a_2\,y(n-2) = b_0\,x(n) + b_1\,x(n-1) + b_2\,x(n-2).$$

**Передаточная функция** — Z-преобразование ИХ при нулевых начальных условиях:
$$H(z) = \frac{b_0 + b_1 z^{-1} + b_2 z^{-2}}{a_0 + a_1 z^{-1} + a_2 z^{-2}}.$$

**Нули и полюсы** — корни числителя $B(z)$ и знаменателя $A(z)$. Комплексное число задаётся в алгебраической форме $z=\mathrm{Re}\,z+j\,\mathrm{Im}\,z$ или показательной $z=|z|e^{j\varphi}$. ЛДС устойчива, когда все полюсы лежат внутри единичной окружности ($|z_p|<1$).

**Частотная характеристика** $H(e^{j\hat\omega})$ — передаточная функция на единичной окружности, $\hat\omega = 2\pi f/f_д$ — нормированная частота. **АЧХ** — $|H(e^{j\hat\omega})|$, **ФЧХ** — $\arg H(e^{j\hat\omega})$. Для вещественной системы АЧХ симметрична и ФЧХ антисимметрична относительно $\hat\omega=\pi$, поэтому основная полоса — $[0;\pi]$, а в шкале $f$ (Гц) — $[0;f_д/2]$.

**ДПФ** последовательности конечной длины $N$ и **ОДПФ**:
$$X(k) = \sum_{n=0}^{N-1} x(n)\,e^{-j\frac{2\pi}{N}kn}, \qquad
x(n) = \frac{1}{N}\sum_{k=0}^{N-1} X(k)\,e^{j\frac{2\pi}{N}kn},\qquad k,n=0\ldots N-1.$$

**Спектры периодической последовательности** с периодом $N$: амплитудный $|X(k)|$ и фазовый $\arg X(k)$ — отсчёты ДПФ одного периода; спектр дискретен (линейчатый) и периодичен по $k$ с периодом $N$; для вещественной $x(n)$ он эрмитово-симметричен: $X(N-k)=X^*(k)$.

**Равенство Парсеваля** — равенство энергий во временной и частотной областях:
$$\sum_{n=0}^{N-1}|x(n)|^2 = \frac{1}{N}\sum_{k=0}^{N-1}|X(k)|^2.$$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import lfilter  # контрольная сверка ручных расчётов

plt.rcParams['figure.figsize'] = [10, 4.2]
plt.rcParams['figure.dpi'] = 100
plt.rcParams['font.size'] = 10
np.set_printoptions(precision=4, suppress=True, linewidth=110)

def cplx(z):
    'Алгебраическая форма комплексного числа'
    return f'{z.real:+.4f} {"+" if z.imag >= 0 else "-"} j{abs(z.imag):.4f}'

def expo(z):
    'Показательная форма: |z|·e^{jφ}, аргумент в долях π'
    return f'{abs(z):.4f}·exp(j{np.angle(z)/np.pi:+.4f}·π)'
```

## Параметры варианта

Единственная величина, определяющая вариант, — номер $N_{вар}$. **Для смены варианта измените значение `N_var` в следующей ячейке и перезапустите блокнот.**

```python

N_var = 14

b0 = 0.5 + 0.02 * N_var
b1 = b0 * (-1) ** (N_var + 1) * (0.9822 + 0.0178 * N_var)
b2 = b0 * (0.8 + 0.2 * (N_var % 5))
a0 = 1
a1 = (-1) ** N_var * (0.7778 + 0.025 * N_var)
a2 = 0.64 + 0.006 * N_var
N1  = N_var % 10 + 20      # длина импульсной характеристики
N2  = N_var % 10 + 30      # длина воздействия
fd1 = 1000 * N_var         # частота дискретизации (часть 1), Гц

b = np.array([b0, b1, b2])   # числитель B(z)
a = np.array([a0, a1, a2])   # знаменатель A(z)

# ---------- Часть 2: периодическая последовательность ----------
N   = 64                          # период (длина) последовательности
fd2 = 2000 * (N_var % 5 + 1)      # частота дискретизации (часть 2), Гц
T   = 1 / fd2                     # период дискретизации, с
A1  = 1 + 0.01 * N_var            # амплитуды дискретных гармоник
A2  = 2 * A1
f1  = fd2 / 8                     # частоты дискретных гармоник, Гц
f2  = 1.5 * f1

print(f'Часть 1:  b0 = {b0:.4f}, b1 = {b1:+.4f}, b2 = {b2:.4f}')
print(f'          a0 = {a0},   a1 = {a1:+.4f}, a2 = {a2:.4f}')
print(f'          N1 = {N1}, N2 = {N2}, fд = {fd1} Гц')
print(f'Часть 2:  N = {N}, fд = {fd2} Гц, T = {T} с')
print(f'          A1 = {A1:.2f}, A2 = {A2:.2f}, f1 = {f1:.1f} Гц, f2 = {f2:.1f} Гц')
```

## Часть 1, п. 1. Импульсная характеристика $h(n)$

Импульсная характеристика — реакция звена на цифровой единичный импульс $u_0(n)$ при нулевых начальных условиях. Расчёт — прямой пересчёт разностного уравнения
$$y(n) = -a_1 y(n-1) - a_2 y(n-2) + b_0 x(n) + b_1 x(n-1) + b_2 x(n-2),\qquad x(n)=u_0(n).$$

**Аналитическая формула ИХ рекурсивного звена 2-го порядка** при нулевых начальных условиях и различных полюсах $p_1\ne p_2$:
$$h(n) = b_0\,u_0(n) + \big(K_1\,p_1^{\,n-1} + K_2\,p_2^{\,n-1}\big)\,u_1(n-1),$$
где $p_{1,2}$ — полюсы (корни $A(z)$), $u_1(n)$ — цифровой единичный скачок, а коэффициенты разложения
$$K_{1,2} = \frac{(b_1-b_0a_1)\,p_{1,2} + (b_2-b_0a_2)}{p_{1,2}-p_{2,1}}.$$
При комплексно-сопряжённых полюсах $p_{1,2}=|p|e^{\pm j\varphi_p}$ сумма в скобках — затухающая дискретная гармоника:
$$h(n) = b_0\,u_0(n) + 2|K_1|\,|p|^{\,n-1}\cos\!\big((n-1)\varphi_p + \arg K_1\big)\,u_1(n-1).$$

```python
u0 = np.zeros(N1); u0[0] = 1.0

h = np.zeros(N1)
for n in range(N1):
    acc = 0.0
    for k in (1, 2):
        if n >= k:
            acc -= a[k] * h[n - k]              # рекурсивная часть
    for k in (0, 1, 2):
        if n >= k:
            acc += b[k] * u0[n - k]          # нерекурсивная часть
    h[n] = acc

assert np.allclose(h, lfilter(b, a, u0), atol=1e-12)

poles = np.roots(a)                              
p1, p2 = poles
K1 = ((b1 - b0 * a1) * p1 + (b2 - b0 * a2)) / (p1 - p2)
K2 = ((b1 - b0 * a1) * p2 + (b2 - b0 * a2)) / (p2 - p1)
n_ = np.arange(N1)
h_an = np.zeros(N1)
h_an[0] = b0
h_an[1:] = np.real(K1 * p1 ** (n_[1:] - 1) + K2 * p2 ** (n_[1:] - 1))
assert np.allclose(h, h_an, atol=1e-10)

print(f'Полюсы: p0 = {cplx(poles[0])}  =  {expo(poles[0])}')
print(f'        p1 = {cplx(poles[1])}  =  {expo(poles[1])}')
print(f'Коэффициенты разложения: K0 = {cplx(K1)}, K1 = {cplx(K2)}')
print(f'Совпадение рекурсии и аналитической формулы: max|h − h_анал| = {np.max(np.abs(h - h_an)):.2e}')
print(f'|h(N1−1)| = {abs(h[-1]):.3e},  max|h| = {np.max(np.abs(h)):.3f},  |p| = {abs(p1):.4f}')

plt.figure(figsize=(10, 3.8))
plt.stem(np.arange(N1), h)
plt.xlabel('n'); plt.ylabel('h(n)')
plt.title(f'Импульсная характеристика, N1 = {N1} отсчётов')
plt.grid(alpha=0.4); plt.tight_layout(); plt.show()
```

**Пояснение.** Импульсная характеристика рекурсивного звена — бесконечно убывающая последовательность: каждый полюс $p$ внутри единичной окружности даёт слагаемое $|p|^{\,n-1}\to 0$, так что строго $h(n)\ne0$ при любом $n$. **Фактическая** (практически значимая) длина ограничена: при $|p|<1$ отсчёты быстро затухают, и уже при $n=N_1-1$ ИХ существенно меньше своего максимума — для данного варианта примерно на два порядка (см. выведенные значения $|h(N_1-1)|$, $\max|h|$ и $|p|$). Поэтому в расчётах ИХ представляют конечным числом отсчётов $N_1$; погрешность такого усечения уменьшается с ростом $N_1$.

## Часть 1, п. 2. Реакция $y_1(n)$ по формуле свёртки

Воздействие — дискретный прямоугольный импульс длины $N_2$:
$$x(n)=\begin{cases}1, & 0\le n< int(N_2/2),\\[2pt] 0, & int(N_2/2)\le n\le N_2-1,\end{cases}
\qquad int(t)\ \text{— округление до ближайшего целого}.$$

Реакция по формуле свёртки с ИХ $h(n)$ длины $N_1$:
$$y_1(n)=\sum_{k=0}^{n}x(k)\,h(n-k),\qquad L_y = N_1+N_2-1\ \text{— длина полной свёртки}.$$

Длина воздействия $N_2$, длина ИХ $N_1$ — их значения и длина свёртки выведены ниже.

```python
nlim = int(round(N2 / 2))                 
n_x = np.arange(N2)
x = np.where(n_x < nlim, 1.0, 0.0)       

y1_full = np.convolve(x, h)               
y1 = y1_full[:N2]                       

print(f'Длина импульса: {nlim} единичных отсчётов из {N2} (int(N2/2) = {nlim})')
print(f'Длина полной свёртки: {len(y1_full)} = N1 + N2 − 1 = {N1} + {N2} − 1')

fig, ax = plt.subplots(1, 3, figsize=(13, 3.4))
ax[0].stem(n_x, x)
ax[0].set_title(f'Воздействие x(n), N2 = {N2}'); ax[0].set_xlabel('n'); ax[0].set_ylabel('x(n)')
ax[1].stem(np.arange(len(y1_full)), y1_full)
ax[1].axvline(N2 - 1, color='r', ls='--', lw=1, label=f'n = N2−1 = {N2-1}')
ax[1].set_title(f'Полная свёртка x∗h, {N1+N2-1} отсчётов'); ax[1].set_xlabel('n'); ax[1].legend()
ax[2].stem(n_x, y1)
ax[2].set_title(f'Реакция y1(n), усечённая до N2 = {N2}'); ax[2].set_xlabel('n'); ax[2].set_ylabel('y1(n)')
for a_ in ax: a_.grid(alpha=0.4)
plt.tight_layout(); plt.show()
```

**Пояснение.**
- **Длина импульса $x$:** $int(N_2/2)$ единичных отсчётов (см. вывод ячейки), полная длина воздействия — $N_2$.
- **Длина свёртки** (аналитически): $N_1+N_2-1$; по графику это видно так: реакция начинается в $n=0$ вместе с воздействием и длится ещё $N_1-1$ отсчётов после окончания импульса — пока ИХ «выносит» последний единичный отсчёт за пределы воздействия.
- **Почему свёртку ограничивают до длины воздействия:** на интервале $0\le n\le N_2-1$ реакция определяется воздействием и «вынужденной» частью ИХ; отсчёты полной свёртки при $n\ge N_2$ — это свободные затухающие колебания системы после окончания импульса. Для сравнения с реакцией по разностному уравнению $y_2(n)$ (п. 3), вычисляемой на интервале воздействия, свёртку ограничивают до $N_2$ отсчётов.

## Часть 1, п. 3. Реакция $y_2(n)$ по разностному уравнению

С тем же воздействием $x(n)$ решается разностное уравнение звена (написано с численными коэффициентами в выводе ячейки):
$$y_2(n) = -a_1\,y_2(n-1) - a_2\,y_2(n-2) + b_0\,x(n) + b_1\,x(n-1) + b_2\,x(n-2).$$

```python
print(f'Разностное уравнение с заданными коэффициентами:')
print(f'  y(n) = ({-a1:+.4f})·y(n−1) + ({-a2:+.4f})·y(n−2) + {b0:.4f}·x(n) + ({b1:+.4f})·x(n−1) + ({b2:+.4f})·x(n−2)')

y2 = np.zeros(N2)
for n in range(N2):
    acc = 0.0
    for k in (1, 2):
        if n >= k:
            acc -= a[k] * y2[n - k]
    for k in (0, 1, 2):
        if n >= k:
            acc += b[k] * x[n - k]
    y2[n] = acc

assert np.allclose(y2, lfilter(b, a, x), atol=1e-12)

diff = y1 - y2
n_match = min(N1, N2)
print(f'Длина воздействия N2 = {N2}, длина реакции y2(n) также N2 = {len(y2)}')
print(f'max|y1 − y2| при n = 0…{n_match-1}: {np.max(np.abs(diff[:n_match])):.2e} (машинная точность)')
print(f'max|y1 − y2| при n = {n_match}…{N2-1}: {np.max(np.abs(diff[n_match:])):.2e} (усечение ИХ в свёртке)')

fig, ax = plt.subplots(1, 2, figsize=(12, 3.8))
ax[0].stem(n_x, y1, basefmt='C0-', label='y1: свёртка')
ax[0].stem(n_x, y2, markerfmt='C1x', linefmt='C1-', basefmt='C1-', label='y2: разностное уравнение')
ax[0].set_title(f'Реакции y1(n) и y2(n), N2 = {N2}'); ax[0].set_xlabel('n'); ax[0].legend()
ax[1].stem(n_x, diff)
ax[1].axvline(N1 - 1, color='r', ls='--', lw=1, label=f'n = N1−1 = {N1-1}')
ax[1].set_title('Разность y1(n) − y2(n)'); ax[1].set_xlabel('n'); ax[1].legend()
for a_ in ax: a_.grid(alpha=0.4)
plt.tight_layout(); plt.show()
```

**Пояснение.** $y_1(n)$ и $y_2(n)$ совпадают: свёртка и разностное уравнение — два эквивалентных способа описания одной и той же ЛДС. На отсчётах $0\le n\le N_1-1$ совпадение машинное, поскольку свёртка использует только те отсчёты ИХ, которые к этому моменту уже известны. Небольшое расхождение в конце интервала ($n>N_1-1$) вызвано тем, что свёртка выполнена с усечённой ИХ ($N_1$ отсчётов) и не учитывает малые значения $h(n)$ при $n\ge N_1$; с ростом $N_1$ оно стремится к нулю.

**Длины:** воздействие — $N_2$ отсчётов; реакция по разностному уравнению вычисляется на каждом $n$ того же интервала, поэтому её длина также $N_2$: при нулевых начальных условиях реакция не начинается раньше воздействия и рассматривается на его длине.

## Часть 1, п. 4. Нули и полюсы передаточной функции

$$H(z)=\frac{b_0+b_1z^{-1}+b_2z^{-2}}{a_0+a_1z^{-1}+a_2z^{-2}}=b_0\,\frac{(z-z_0)(z-z_1)}{(z-p_0)(z-p_1)},$$
где $z_0,z_1$ — нули (корни $B(z)$), $p_0,p_1$ — полюсы (корни $A(z)$).

**Связь алгебраической и показательной форм:** $z=\mathrm{Re}\,z+j\,\mathrm{Im}\,z=|z|e^{j\varphi}$, где $|z|=\sqrt{(\mathrm{Re}\,z)^2+(\mathrm{Im}\,z)^2}$, $\varphi=\operatorname{arctg}(\mathrm{Im}\,z/\mathrm{Re}\,z)$ — угол поворота вектора $z$ на комплексной плоскости; ниже аргументы выражены в долях $\pi$. Комплексно-сопряжённые пары имеют равные модули и противоположные аргументы — на карте нулей и полюсов они симметричны относительно вещественной оси.

```python
zeros = np.roots(b)
poles = np.roots(a)
print('алгебраическая форма                показательная форма')
for i in range(2):
    print(f'z{i} = {cplx(zeros[i])}   |   z{i} = {expo(zeros[i])}')
    print(f'p{i} = {cplx(poles[i])}   |   p{i} = {expo(poles[i])}')
print()
print(f'H(z) = {b0:.4f} · (z − z0)(z − z1) / [(z − p0)(z − p1)]')
print(f'     = {b0:.4f} · (z − {abs(zeros[0]):.4f}·e^(j±{abs(np.angle(zeros[0])/np.pi):.4f}·π)) '
      f'· (z − {abs(zeros[1]):.4f}·e^(j±{abs(np.angle(zeros[1])/np.pi):.4f}·π)) '
      f'/ (z − {abs(poles[0]):.4f}·e^(j±{abs(np.angle(poles[0])/np.pi):.4f}·π)) '
      f'· (z − {abs(poles[1]):.4f}·e^(j±{abs(np.angle(poles[1])/np.pi):.4f}·π))')
print()
print('Контроль сопряжённости:')
print(f'  |z0| = {abs(zeros[0]):.4f}, |z1| = {abs(zeros[1]):.4f}; '
      f'arg z0 + arg z1 = {(np.angle(zeros[0]) + np.angle(zeros[1]))/np.pi:+.2f}·π (сумма аргументов пары = 0)')
print(f'  |p0| = {abs(poles[0]):.4f}, |p1| = {abs(poles[1]):.4f}; '
      f'arg p0 + arg p1 = {(np.angle(poles[0]) + np.angle(poles[1]))/np.pi:+.2f}·π')
```

## Часть 1, п. 5. Карта нулей и полюсов. Устойчивость звена

Карта нулей и полюсов — расположение нулей (○) и полюсов (×) на комплексной $z$-плоскости с единичной окружностью $|z|=1$. Звено устойчиво, если **все полюсы строго внутри** единичной окружности.

```python
theta = np.linspace(0, 2 * np.pi, 300)
fig, ax = plt.subplots(figsize=(6, 6))
ax.plot(np.cos(theta), np.sin(theta), 'k--', lw=1, label='единичная окружность |z| = 1')
ax.plot(zeros.real, zeros.imag, 'o', ms=13, mfc='none', mec='tab:blue', mew=2, label='нули')
ax.plot(poles.real, poles.imag, 'x', ms=13, c='tab:red', mew=2.5, label='полюсы')
for z, s in [(zeros[0], 'z0'), (zeros[1], 'z1'), (poles[0], 'p0'), (poles[1], 'p1')]:
    ax.annotate(f'{s} = {abs(z):.3f}·e^(j{np.angle(z)/np.pi:+.3f}π)',
                (z.real, z.imag), textcoords='offset points', xytext=(9, 6), fontsize=9)
ax.axhline(0, color='gray', lw=0.5); ax.axvline(0, color='gray', lw=0.5)
ax.set_aspect('equal'); ax.set_xlabel('Re z'); ax.set_ylabel('Im z')
ax.set_title('Карта нулей и полюсов'); ax.grid(alpha=0.4)
ax.legend(loc='upper left', fontsize=9)
plt.tight_layout(); plt.show()

stable = np.all(np.abs(poles) < 1)
print(f'Модули полюсов: |p0| = {abs(poles[0]):.4f}, |p1| = {abs(poles[1]):.4f} → все < 1: {stable}')
print('Звено УСТОЙЧИВО' if stable else 'Звено НЕУСТОЙЧИВО')
print(f'Модули нулей: |z0| = |z1| = {abs(zeros[0]):.4f} — '
      + ('вне единичной окружности (нуль не создаёт нуля АЧХ)' if abs(zeros[0]) > 1
         else 'на единичной окружности (АЧХ обращается в нуль на аргументе нуля)'))
```

**Пояснение.** Значения на карте полностью совпадают с п. 4: та же пара нулей и та же пара полюсов (модули и аргументы подписаны у точек). Оба полюса лежат **внутри** единичной окружности ($|p|<1$ — выведено в ячейке), поэтому звено **устойчиво**: свободные составляющие затухают как $|p|^n$, что подтверждается графиком ИХ в п. 1. Нули на устойчивость не влияют — они определяют лишь глубину подавления АЧХ на соответствующих частотах.

## Часть 1, п. 6. АЧХ и ФЧХ в шкале нормированных частот $\hat\omega$

Частотная характеристика получается подстановкой $z=e^{j\hat\omega}$:
$$H(e^{j\hat\omega}) = \frac{b_0+b_1e^{-j\hat\omega}+b_2e^{-2j\hat\omega}}{1+a_1e^{-j\hat\omega}+a_2e^{-2j\hat\omega}},\qquad
\text{АЧХ}=|H(e^{j\hat\omega})|,\quad \text{ФЧХ}=\arg H(e^{j\hat\omega}).$$
Основная полоса нормированных частот: $\hat\omega\in[0;\pi]$ (для вещественной системы вне её АЧХ симметрична, ФЧХ антисимметрична).

```python
w_hat = np.linspace(0, np.pi, 2000)                 # основная полоса [0; π]
E = np.exp(-1j * w_hat)
H = (b[0] + b[1] * E + b[2] * E**2) / (a[0] + a[1] * E + a[2] * E**2)
ACH, FCH = np.abs(H), np.angle(H)

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].plot(w_hat / np.pi, ACH); ax[0].set_title('АЧХ |H(e^{jω̂})|')
ax[0].set_xlabel('ω̂ / π'); ax[0].set_ylabel('|H|')
ax[1].plot(w_hat / np.pi, FCH); ax[1].set_title('ФЧХ arg H(e^{jω̂})')
ax[1].set_xlabel('ω̂ / π'); ax[1].set_ylabel('φ, рад')
for a_ in ax: a_.grid(alpha=0.4)
plt.tight_layout(); plt.show()

i_min = np.argmin(ACH)
print(f'АЧХ(ω̂=0) = {ACH[0]:.4f};  АЧХ(ω̂=π) = {ACH[-1]:.4f}')
print(f'Минимум АЧХ = {ACH[i_min]:.4f} при ω̂ = {w_hat[i_min]/np.pi:+.4f}·π '
      f'(аргумент нуля: ±{abs(np.angle(zeros[0])/np.pi):.4f}·π — близко к нему)')
print(f'Максимум АЧХ = {np.max(ACH):.4f} при ω̂/π = {w_hat[np.argmax(ACH)]/np.pi:+.4f} '
      f'(аргумент полюса: ±{abs(np.angle(poles[0])/np.pi):.4f}·π)')
print(f'Подавлены составляющие вблизи ω̂ ≈ {w_hat[i_min]/np.pi:.2f}·π '
      + ('— нижние/средние частоты; высокие проходят с усилением'
         if ACH[0] < ACH[-1] else '— высокие частоты; низкие проходят'))
jumps = np.where(np.abs(np.diff(FCH)) > np.pi / 2)[0]
if len(jumps) == 0:
    print('Скачков ФЧХ на основной полосе нет (нули не лежат на единичной окружности, фаза непрерывна)')
for j in jumps:
    dj = FCH[j + 1] - FCH[j]
    kind = '≈ ±π' if abs(abs(dj) - np.pi) < 0.3 else '≈ ±2π (перенос главного значения)'
    print(f'Скачок ФЧХ {dj:+.2f} рад при ω̂/π ≈ {w_hat[j + 1]/np.pi:.4f}, АЧХ там = {0.5*(ACH[j]+ACH[j+1]):.4f} → {kind}')
```

**Пояснение.**
- **Границы основной полосы:** $\hat\omega\in[0;\pi]$ — за пределами АЧХ периодична ($\hat\omega$ и $\hat\omega\pm2\pi$ эквивалентны), а внутри $[0;2\pi]$ она симметрична относительно $\pi$.
- **Связь с картой нулей и полюсов:** АЧХ мала там, где близко к единичной окружности проходит вектор нуля (знаменатель векторной диаграммы — произведение расстояний до полюсов, числитель — до нулей), и велика вблизи полюсов. Минимум АЧХ (см. вывод) находится вблизи аргумента нулей, максимум — вблизи аргумента полюсов (небольшое смещение создаёт зависимость знаменателя от частоты), что соответствует карте п. 5.
- **Скачок ФЧХ на $\pi$** соответствует **нулевому (минимальному) значению АЧХ**: при переходе вектора $H$ через нуль его аргумент меняется на $\pi$. Такой скачок возникает, если ноль лежит на единичной окружности; для текущего варианта нули вне окружности, поэтому АЧХ в ноль не обращается и скачков на $\pi$ нет — фаза непрерывна (переносы главного значения на $\pm2\pi$, если есть, отмечены в выводе и не связаны с АЧХ).
- **Какие составляющие подавлены:** согласно выведенным значениям, подавлены частоты вблизи минимума АЧХ (низкочастотные/средние — при $\mathrm{АЧХ}(0)<\mathrm{АЧХ}(\pi)$), а верхние частоты полосы проходят с усилением; в реакции $y_1,y_2$ по сравнению с воздействием $x$ (пп. 2–3) это проявляется в сглаживании фронта и колебательном «звоне» — следствие подавления части спектра импульса.

## Часть 1, п. 7. АЧХ и ФЧХ в шкале частот $f$ (Гц)

Связь абсолютной и нормированной частот: $\hat\omega = 2\pi f/f_д \Rightarrow f = \hat\omega\,f_д/2\pi$. Основная полоса $\hat\omega\in[0;\pi]$ соответствует $f\in[0;f_д/2]$.

```python
f_hz = w_hat * fd1 / (2 * np.pi)                    
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].plot(f_hz, ACH); ax[0].set_title('АЧХ в шкале частот f')
ax[0].set_xlabel('f, Гц'); ax[0].set_ylabel('|H|')
ax[1].plot(f_hz, FCH); ax[1].set_title('ФЧХ в шкале частот f')
ax[1].set_xlabel('f, Гц'); ax[1].set_ylabel('φ, рад')
for a_ in ax: a_.grid(alpha=0.4)
plt.tight_layout(); plt.show()

print(f'fд = {fd1} Гц; основная полоса: 0 … fд/2 = {fd1/2:.0f} Гц')
print(f'Границы шкалы: ω̂ = 0 → f = 0 Гц;  ω̂ = π → f = fд/2 = {fd1/2:.0f} Гц')
print(f'Минимум АЧХ: ω̂ = {w_hat[i_min]/np.pi:.4f}·π → f = {f_hz[i_min]:.1f} Гц')
print(f'Максимум АЧХ: ω̂/π = {w_hat[np.argmax(ACH)]/np.pi:.4f} → f = {f_hz[np.argmax(ACH)]:.1f} Гц')
```

**Пояснение.** Основная полоса в шкале $f$: $0\le f\le f_д/2$ (частота Найквиста). Соответствие частот линейно: $f = \hat\omega f_д/2\pi$, поэтому границе $\hat\omega=\pi$ отвечает $f=f_д/2$, а частоты минимума/максимума АЧХ просто масштабируются этим коэффициентом (выведены в ячейке). Графики повторяют п. 6 с заменой оси: обе шкалы описывают одну и ту же частотную характеристику.

# ЧАСТЬ 2. ДПФ периодических и конечных последовательностей

## Часть 2, п. 1. Периодическая последовательность и её спектры

Периодическая последовательность с периодом $N$:
$$x(nT) = A_1\cos(2\pi f_1 nT + \pi/4) + A_2\cos(2\pi f_2 nT + \pi/8),\qquad n=0\ldots N-1.$$
Дискретное нормированное время: $\hat t = n/N$ (доля периода); связь с дискретным временем: $t = nT \Rightarrow \hat t = t/(NT)$.
Амплитудный спектр — $|X(k)|$, фазовый — $\arg X(k)$, где $X(k)$ — ДПФ одного периода (вычислено здесь, построено в п. 3).

```python
n2 = np.arange(N)
x2 = A1 * np.cos(2 * np.pi * f1 * n2 * T + np.pi / 4) \
   + A2 * np.cos(2 * np.pi * f2 * n2 * T + np.pi / 8)

X2 = np.fft.fft(x2)                       # ДПФ одного периода — спектр периодической x(nT)
k1 = int(round(f1 * N / fd2))             # номер бина 1-й гармоники
k2 = int(round(f2 * N / fd2))             # номер бина 2-й гармоники
print(f'Период N = {N};  f1 = {f1:.1f} Гц → бин k1 = {k1};  f2 = {f2:.1f} Гц → бин k2 = {k2}')
print(f'f1·N/fд = {f1*N/fd2:.1f}, f2·N/fд = {f2*N/fd2:.1f} — целые: гармоники точно в бинах ДПФ')

fig, ax = plt.subplots(1, 2, figsize=(12, 3.8))
ax[0].stem(n2, x2)
ax[0].set_title(f'x(nT) в шкале дискретного времени n (N = {N})'); ax[0].set_xlabel('n'); ax[0].set_ylabel('x(nT)')
ax[1].stem(n2 / N, x2)
ax[1].set_title('x(nT) в шкале дискретного нормированного времени n/N'); ax[1].set_xlabel('n/N'); ax[1].set_ylabel('x(nT)')
for a_ in ax: a_.grid(alpha=0.4)
plt.tight_layout(); plt.show()
```

**Пояснение.** Обе гармоники укладываются в период целое число раз: $f_1 N/f_д = 8$ и $f_2 N/f_д = 12$ периодов (выведено в ячейке), поэтому последовательность строго периодична с периодом $N = 64$. Оба графика показывают одни и те же отсчёты в разных шкалах времени: слева — в шкале дискретного времени $n$ (номера отсчётов $0\ldots N-1$), справа — в шкале дискретного нормированного времени $\hat t = n/N$ (доля периода, от 0 до 1); форма кривых одинакова, меняется только разметка оси. Сумма двух гармоник разных частот даёт периодическую модуляцию огибающей (биения): за период последовательности укладывается ровно четыре цикла огибающей — $(f_2 - f_1)\,NT = 4$.

## Часть 2, п. 2. ОДПФ от ДПФ

Обратное ДПФ от полученного спектра должно восстановить исходную последовательность:
$$x(n) = \frac{1}{N}\sum_{k=0}^{N-1} X(k)\,e^{j\frac{2\pi}{N}kn}.$$
График строится в шкале дискретного нормированного времени $\hat t = n/N$.

```python
x2_idft = np.real(np.fft.ifft(X2))
err = np.max(np.abs(x2_idft - x2))
print(f'max|ОДПФ(ДПФ x) − x| = {err:.2e} → последовательность восстановлена точно')

plt.figure(figsize=(10, 3.6))
plt.stem(n2 / N, x2_idft, basefmt='C0-')
plt.xlabel('n/N'); plt.ylabel('x(n)')
plt.title('Результат ОДПФ от ДПФ (шкала дискретного нормированного времени)')
plt.grid(alpha=0.4); plt.tight_layout(); plt.show()
```

**Пояснение.** ОДПФ от ДПФ восстановило исходную последовательность с машинной точностью (расхождение выведено в ячейке). Это проверка обратимости пары ДПФ/ОДПФ: преобразование не теряет и не добавляет информации — $N$ отсчётов сигнала взаимно однозначно соответствуют $N$ комплексным отсчётам спектра, меняется лишь представление (время ↔ частота). График построен в шкале дискретного нормированного времени $\hat t = n/N$ и повторяет правый график п. 1.

## Часть 2, п. 3. Спектры в шкалах нормированных и абсолютных частот

Дискретная нормированная частота: $\hat\omega_k = 2\pi k/N$; абсолютная частота бина: $f_k = k\,f_д/N = k/(NT)$.
**Связи:** дискретное нормированное время $\hat t = n/N = t/(NT)$ — время, нормированное к периоду; частота $f$ (Гц) связана с нормированной как $f = \hat\omega f_д/2\pi$, а с номером бина — $f = k f_д/N$. Для вещественной $x(n)$ спектр эрмитово-симметричен: $|X(k)|=|X(N-k)|$, $\arg X(N-k) = -\arg X(k)$.

```python
AMPS = np.abs(X2)
PH = np.angle(X2)
PH[AMPS < 1e-9] = 0.0                     

w_k = 2 * np.pi * np.arange(N) / N        
f_k = np.arange(N) * fd2 / N             

fig, ax = plt.subplots(2, 2, figsize=(12, 7))
ax[0, 0].stem(w_k / np.pi, AMPS); ax[0, 0].set_title('Амплитудный спектр |X(k)|')
ax[0, 0].set_xlabel('ω̂k / π = 2k/N'); ax[0, 0].set_ylabel('|X(k)|')
ax[0, 1].stem(f_k, AMPS); ax[0, 1].set_title('Амплитудный спектр |X(k)|')
ax[0, 1].set_xlabel('f_k = k·fд/N, Гц'); ax[0, 1].set_ylabel('|X(k)|')
ax[1, 0].stem(w_k / np.pi, PH); ax[1, 0].set_title('Фазовый спектр arg X(k)')
ax[1, 0].set_xlabel('ω̂k / π = 2k/N'); ax[1, 0].set_ylabel('φ(k) / π')
ax[1, 1].stem(f_k, PH); ax[1, 1].set_title('Фазовый спектр arg X(k)')
ax[1, 1].set_xlabel('f_k = k·fд/N, Гц'); ax[1, 1].set_ylabel('φ(k) / π')
for a_ in ax.flat: a_.grid(alpha=0.4)
plt.tight_layout(); plt.show()

print('       k |  ω̂k/π  |  f_k, Гц  |  |X(k)|  |  arg X(k)/π   (ненулевые бины)')
for k in (k1, k2, N - k2, N - k1):
    print(f'      {k:2d} |  {k/(N/2):.3f}  |  {f_k[k]:7.1f}  |  {AMPS[k]:6.2f}  |     {PH[k]/np.pi:+.3f}')
print(f'Ожидаемо: |X(k1)| = A1·N/2 = {A1*N/2:.2f}, фаза = +1/4·π (π/4);  |X(k2)| = A2·N/2 = {A2*N/2:.2f}, фаза = +1/8·π (π/8)')
```

**Пояснение.** Спектр периодической последовательности — **линейчатый**: всего четыре ненулевых бина — $\pm k_1$ и $\pm k_2$ (в диапазоне $0\ldots N-1$ это $k_1$, $k_2$ и зеркальные $N-k_1$, $N-k_2$), значения выведены в таблице и совпадают с теоретическими: амплитуды $A_1N/2$ и $A_2N/2$, фазы $+\pi/4$ и $+\pi/8$ (у зеркальных — противоположные). **Связи шкал:** дискретное нормированное время $\hat t=n/N$ — это дискретное время $nT$, отнесённое к периоду $NT$; частота $f$ (Гц) связана с нормированной $\hat\omega=2\pi f/f_д$, а её отсчёты (бины) стоят через $f_д/N$ Гц — поэтому одна и та же гармоника видна в обеих шкалах на эквивалентных позициях (выведенные $f_k$ получаются из $\hat\omega_k$ умножением на $f_д/2\pi$). Растекания спектра нет, так как частоты гармоник кратны разрешению $f_д/N$.

## Часть 2, п. 4. ДПФ конечной последовательности и амплитудный спектр

Конечная последовательность $x(n)$ длины $N$ — один период $x(nT)$ из п. 1. Её ДПФ сравнивается с амплитудным спектром периодической последовательности $|X(k)|$; график ниже — в шкале дискретных нормированных частот $\hat\omega_k = 2\pi k/N$.

```python
x_fin = x2.copy()               
X_fin = np.fft.fft(x_fin)
print(f'Длина конечной последовательности: {len(x_fin)} = N')
print(f'max||ДПФ x_fin| − |X(k)|| = {np.max(np.abs(np.abs(X_fin) - AMPS)):.2e} → графики совпадают')

kk = np.arange(N)
wk = 2 * np.pi * kk / N                 
plt.figure(figsize=(10, 3.8))
plt.stem(wk / np.pi, AMPS, basefmt='C0-', label='амплитудный спектр периодической x(nT)')
plt.plot(wk / np.pi, np.abs(X_fin), 'rx', ms=8, mew=1.5, label='|ДПФ конечной x(n)|')
plt.xlabel('ω̂k / π = 2k/N'); plt.ylabel('|X(k)|'); plt.title('Модуль ДПФ конечной последовательности и амплитудный спектр')
plt.legend(); plt.grid(alpha=0.4); plt.tight_layout(); plt.show()
```

**Пояснение.** Модуль ДПФ конечной последовательности длины $N$ **совпадает** с амплитудным спектром периодической последовательности (выведенное расхождение — машинный нуль): ДПФ и есть спектральный анализ одного периода, а периодическая последовательность получается бесконечным повторением этого периода. Совпадение точное потому, что конечная последовательность содержит целое число периодов обеих гармоник ($k_1$ и $k_2$ — целые), т. е. её периодическое продолжение непрерывно: разрывов на стыке периода нет, и растекание спектра отсутствует. Если бы длина отсчётов не была кратна периоду гармоник, ДПФ конечного фрагмента «размазалось» бы по соседним бинам и прямого совпадения не было бы.

## Часть 2, п. 5. Восстановление аналогового сигнала по отсчётам ДПФ

$$x(t) = \frac{1}{N}\sum_{k=-N/2}^{N/2-1} X_а(k)\,\exp\!\Big(j\frac{2\pi}{NT}\,k\,t\Big),\qquad
X_а(k)=\begin{cases}X(N+k), & -N/2\le k\le -1,\\ X(k), & 0\le k\le N/2-1,\end{cases}$$
где переиндексация переносит бины $k\ge N/2$ на отрицательные частоты (верхняя половина ДПФ — зеркальные составляющие). Время $t\in[0;(N-1)T]$ с шагом $\Delta t=0.25T$. Спектр восстановленного сигнала — те же отсчёты $X_а(k)/N$ на частотах $f_k=k/(NT)$, включая отрицательные.

```python
K = np.arange(-N // 2, N // 2)            
Xa = X2[K % N]                           
assert np.allclose(Xa[:N // 2], X2[N // 2:]) and np.allclose(Xa[N // 2:], X2[:N // 2])

dt = 0.25 * T
t = np.arange(0, (N - 1) * T + dt / 2, dt)          
x_rest = np.real(Xa @ np.exp(1j * 2 * np.pi * np.outer(K, t) / (N * T))) / N

t_n = n2 * T
x_rest_n = np.real(Xa @ np.exp(1j * 2 * np.pi * np.outer(K, t_n) / (N * T))) / N
assert np.allclose(x_rest_n, x2, atol=1e-10)
print(f'Отсчётов времени: {len(t)} (t от 0 до {t[-1]:.4f} с, шаг Δt = 0.25T = {dt:.2e} с)')
print(f'Проверка на сетке отсчётов: max|x_rest(nT) − x(nT)| = {np.max(np.abs(x_rest_n - x2)):.2e}')

x_analog = A1 * np.cos(2 * np.pi * f1 * t + np.pi / 4) + A2 * np.cos(2 * np.pi * f2 * t + np.pi / 8)
print(f'Сравнение с исходным аналоговым сигналом: max|x_rest(t) − x_анал(t)| = {np.max(np.abs(x_rest - x_analog)):.2e}')

fig = plt.figure(figsize=(12, 9))
gs = fig.add_gridspec(3, 2)

ax1 = fig.add_subplot(gs[0, 0])
ax1.stem(n2, x2, basefmt='C0-'); ax1.set_title('Последовательность x(nT)'); ax1.set_xlabel('n')
ax2 = fig.add_subplot(gs[0, 1])
ax2.stem(np.arange(N), AMPS, basefmt='C0-'); ax2.set_title('Её ДПФ |X(k)|'); ax2.set_xlabel('k')

ax3 = fig.add_subplot(gs[1, 0])
ax3.plot(t / T, x_rest, 'C1-', lw=1.2, label='восстановленный x(t)')
ax3.stem(n2, x2, basefmt='C0-', linefmt='C0-', markerfmt='C0o', label='отсчёты x(nT)')
ax3.set_title('Восстановленный сигнал'); ax3.set_xlabel('t / T'); ax3.legend()

ax4 = fig.add_subplot(gs[1, 1])
mask = np.abs(Xa) > 1e-9
ax4.stem(K[mask] / (N * T), np.abs(Xa[mask]) / N, basefmt='C1-')
ax4.set_title('Спектр восстановленного сигнала |Xа(k)|/N'); ax4.set_xlabel('f = k/(NT), Гц')
ax4.set_xlim(-fd2 / 2, fd2 / 2)

ax5 = fig.add_subplot(gs[2, :])
ax5.plot(t / T, x_analog, 'C2-', lw=1.2, label='исходный аналоговый x(t) = x(nT)|_{nT=t}')
ax5.plot(t / T, x_rest, 'C1--', lw=1, label='восстановленный x(t)')
ax5.stem(n2, x2, basefmt='C0-', linefmt='C0-', markerfmt='C0o', label='отсчёты')
ax5.set_title('Исходный и восстановленный сигналы (визуальное сравнение)'); ax5.set_xlabel('t / T')
ax5.legend(); ax5.grid(alpha=0.4)
for a_ in (ax1, ax2, ax3, ax4): a_.grid(alpha=0.4)
plt.tight_layout(); plt.show()
```

## Часть 2, п. 6. Связь модуля ДПФ со спектром аналогового сигнала. Сравнение сигналов

**Связь.** Отсчёты $X_а(k)$ после переиндексации — комплексные амплитуды гармоник аналогового периодического сигнала на частотах $f_k = k/(NT)$, включая отрицательные: слагаемое суммы с индексом $k$ имеет амплитуду $|X_а(k)|/N$ и частоту $f_k$. Пара бинов $\pm k$ даёт вещественную гармонику с амплитудой $2|X_а(k)|/N$ и фазой $\arg X_а(k)$ — в нашем случае $2\cdot(A_1N/2)/N = A_1$ и $2\cdot(A_2N/2)/N = A_2$, фазы $\pi/4$ и $\pi/8$, т. е. **модуль ДПФ — это (с точностью до масштаба $N$ и переиндексации бинов) спектр аналогового периодического сигнала**, дискретизированный по частоте с шагом $f_д/N$.

**Сравнение.** Восстановленный сигнал визуально неотличим от исходного, а численно совпадает с ним с точностью до машинной погрешности (значения выведены в п. 5). Причина: обе частоты гармоник кратны $f_д/N$, поэтому все составляющие попадают точно в бины ДПФ, формула восстановления (конечный ряд Фурье) представляет сигнал без остатка, а интервал $[0;(N-1)T]$ содержит целое число периодов. При несоблюдении условия кратности восстановление по отсчётам ДПФ вернуло бы периодическое продолжение дискретизованного сигнала со «спектральным» искажением частот (растекание), и совпадение было бы лишь приблизительным.

## Часть 2, п. 7. Равенство Парсеваля

$$\sum_{n=0}^{N-1}|x(n)|^2 = \frac{1}{N}\sum_{k=0}^{N-1}|X(k)|^2$$

```python
E_time = np.sum(np.abs(x2) ** 2)          # энергия во временной области
E_freq = np.sum(np.abs(X2) ** 2) / N      # энергия в частотной области
print(f'Σ|x(n)|²            = {E_time:.6f}')
print(f'(1/N)·Σ|X(k)|²      = {E_freq:.6f}')
print(f'Абсолютная разность = {abs(E_time - E_freq):.3e},  относительная = {abs(E_time - E_freq) / E_time:.3e}')
```

**Пояснение.** Равенство Парсеваля выполняется с машинной точностью. **Смысл:** ДПФ — унитарное (с точностью до масштаба $1/\sqrt{N}$) преобразование, сохраняющее энергию: суммарная энергия последовательности, вычисленная по её отсчётам во времени, равна сумме энергий её спектральных составляющих, делённой на $N$. Иначе говоря, энергия сигнала лишь «перераспределяется» между бинами спектра, и ни временно́е, ни частотное представление не теряет информации о ней — это обосновывает анализ сигналов в частотной области.
