# Краткие теоретические сведения

## Дискретизация и виды сигналов

Дискретный сигнал состоит из отсчётов $x(n)=x(nT)$, взятых через период $T=1/f_d$, где $f_d$ - частота дискретизации. Номер отсчёта $n$ представляет нормированное дискретное время, а $nT$ - время в секундах. Нормированная круговая частота связана с частотой в герцах формулой

$$
\widehat\omega=2\pi fT=\frac{2\pi f}{f_d}.
$$

Единичный импульс $\delta(n)$ равен единице при $n=0$ и нулю при остальных $n$. Прямоугольный импульс содержит подряд несколько единичных отсчётов; в заданном массиве за ними следуют нули. Косинусоида после дискретизации имеет вид $A\cos(2\pi fnT+\varphi)$, где $A$ - амплитуда, $f$ - частота, $\varphi$ - начальная фаза.

## Линейная система и её импульсная характеристика

Линейная система подчиняется принципу суперпозиции: реакция на сумму входов равна сумме их реакций. Если задержка входа вызывает такую же задержку выхода, система инвариантна во времени. Для такой системы реакция на единичный импульс $h(n)$ при нулевых начальных условиях определяет поведение при любом входе.

Выход можно вычислить свёрткой входа с импульсной характеристикой:

$$
y(n)=\sum_m x(m)h(n-m).
$$

Для конечных массивов длины $L_x$ и $L_h$ полная вычисленная свёртка содержит $L_x+L_h-1$ элементов. Если $h(n)$ на самом деле бесконечна, расчёт с её конечным фрагментом является приближением. Разность между ним и точной реакцией появляется после того, как отброшенные отсчёты $h(n)$ начинают участвовать в сумме.

## Разностное уравнение, $z$-преобразование и передаточная функция

Рекурсивное звено второго порядка при $a_0=1$ описывается разностным уравнением

$$
y(n)=b_0x(n)+b_1x(n-1)+b_2x(n-2)
-a_1y(n-1)-a_2y(n-2).
$$

Расчёт выполняется при нулевых начальных условиях $y(-1)=y(-2)=0$. До начала воздействия входной сигнал также равен нулю: $x(n)=0$ при $n<0$.

При $z$-преобразовании задержке сигнала на один отсчёт соответствует умножение на $z^{-1}$, а на два отсчёта - на $z^{-2}$. Применяя это свойство к разностному уравнению, получаем

$$
Y(z)(1+a_1z^{-1}+a_2z^{-2})
=
X(z)(b_0+b_1z^{-1}+b_2z^{-2}).
$$

Передаточная функция равна отношению преобразования выхода к преобразованию входа:

$$
H(z)=\frac{Y(z)}{X(z)}
=\frac{b_0+b_1z^{-1}+b_2z^{-2}}
{1+a_1z^{-1}+a_2z^{-2}}.
$$

Она также является $z$-преобразованием импульсной характеристики: $H(z)=\mathcal Z\{h(n)\}$.

После умножения числителя и знаменателя на $z^2$ получаем

$$
H(z)=\frac{b_0z^2+b_1z+b_2}{z^2+a_1z+a_2}.
$$

Корни числителя называют нулями, корни знаменателя - полюсами. Комплексный корень можно записать в алгебраической форме $a+jb$ или в показательной форме $re^{j\varphi}$, где $r$ - модуль, а $\varphi$ - аргумент.

Система устойчива, если ограниченный вход вызывает ограниченный выход. Для линейной системы это равносильно условию

$$
\sum_{n=0}^{\infty}|h(n)|<\infty.
$$

У причинного звена устойчивость можно проверить по полюсам передаточной функции: после сокращения общих множителей все они должны находиться внутри единичной окружности, то есть $|p_i|<1$. Положение нулей само по себе устойчивость не определяет.

## Частотная характеристика

Подстановка $z=e^{j\widehat\omega}$ даёт частотную характеристику $H(e^{j\widehat\omega})$. Её модуль называют АЧХ, аргумент - ФЧХ. Основная полоса нормированных частот задаётся интервалом $-\pi\le\widehat\omega<\pi$; в герцах ей соответствует $-f_d/2\le f<f_d/2$.

Когда точка $e^{j\widehat\omega}$ на единичной окружности приближается к полюсу, АЧХ обычно возрастает; при приближении к нулю уменьшается. Скачок графика ФЧХ может быть связан с выбором диапазона отображения фазы, поэтому его нужно рассматривать вместе со значением АЧХ.

## ДПФ и спектры

ДПФ последовательности длины $N$ и обратное ДПФ определяются формулами

$$
X(k)=\sum_{n=0}^{N-1}x(n)e^{-j2\pi kn/N},\qquad
x(n)=\frac1N\sum_{k=0}^{N-1}X(k)e^{j2\pi kn/N}.
$$

Отсчёт $k$ соответствует частоте $f_k=kf_d/N$; расстояние между соседними отсчётами равно $\Delta f=f_d/N$. В двустороннем спектре старшие индексы ДПФ изображают на отрицательных частотах. Для периодического сигнала амплитуда линии равна $|X(k)|/N$, а её фаза равна $\arg X(k)$.

У действительной косинусоиды две линии на частотах $\pm f_0$ с амплитудами $A/2$ и противоположными фазами. Если частота точно попадает на сетку ДПФ, энергия сосредоточена в этих линиях; иначе она распределяется между соседними отсчётами. ДПФ конечного массива и спектр его периодического продолжения используют те же числа $X(k)$, но коэффициенты периодического спектра равны $X(k)/N$.

## Восстановление сигнала и равенство Парсеваля

По двусторонним отсчётам ДПФ можно построить периодический непрерывный сигнал:

$$
x_{\mathrm{восст}}(t)=\frac1N
\sum_{k=-N/2}^{N/2-1}X_a(k)
\exp\!\left(j\frac{2\pi k}{NT}t\right).
$$

В моменты $t=nT$ он совпадает с исходными отсчётами. Если исходные гармоники лежат на частотной сетке ДПФ и ниже частоты Найквиста, восстановление совпадает с исходным сигналом и между отсчётами.

Равенство Парсеваля связывает энергию отсчётов и их ДПФ:

$$
\sum_{n=0}^{N-1}|x(n)|^2
=\frac1N\sum_{k=0}^{N-1}|X(k)|^2.
$$

Для периодической последовательности эта формула описывает энергию одного периода. Совпадение обеих сумм служит проверкой вычислений и нормировки ДПФ.

## Задание 1. Импульсная характеристика

Импульсная характеристика вычисляется как реакция на $x(n)=\delta(n)$ при нулевых начальных условиях. Используется разностное уравнение

$$
h(n)=0{,}86\delta(n)-1{,}120236\delta(n-1)+1{,}204\delta(n-2)
-1{,}2278h(n-1)-0{,}748h(n-2).
$$

В явном виде при $n\ge1$ характеристика выражается через полюса $p_1$ и $p_2$:

$$
h(0)=0{,}86,\qquad h(n)=C_1p_1^{n-1}+C_2p_2^{n-1},
$$

$$
C_1=\frac{h(2)-p_2h(1)}{p_1-p_2},\qquad
C_2=\frac{p_1h(1)-h(2)}{p_1-p_2},
$$

где $h(1)=-2{,}176144$, $h(2)=3{,}232590$. Эта запись учитывает первые три коэффициента числителя и нулевые начальные условия.

Первые восемь отсчётов:

| $n$ | 0 | 1 | 2 | 3 | 4 | 5 | 6 | 7 |
|---:|---:|---:|---:|---:|---:|---:|---:|---:|
| $h(n)$ | 0,860000 | -2,176144 | 3,232590 | -2,341218 | 0,456570 | 1,190654 | -1,803400 | 1,323605 |


На рисунке показаны $N_1=28$ отсчётов. Действительная импульсная характеристика рекурсивной системы бесконечна: каждый новый отсчёт зависит от двух предыдущих, поэтому после исчезновения входного импульса реакция не обрывается. Следовательно, система является БИХ-системой. Затухание колебаний на графике согласуется с устойчивостью системы.

```python
delta = np.zeros(N1)
delta[0] = 1
h = recursive_filter(b, a, delta)

fig, ax = plt.subplots(figsize=(10, 4.2))
stem(ax, np.arange(N1), h)
ax.set(title="Импульсная характеристика: первые $N_1=28$ отсчётов", xlabel="$n$", ylabel="$h(n)$")
finish_axes(ax)
fig.tight_layout()
plt.show()

print("Первые 8 отсчётов h(n):", np.round(h[:8], 6))
```

## Задание 2. Реакция $y_1(n)$ по формуле свёртки

Входной прямоугольный импульс имеет длину $N_2=38$. Поскольку $\operatorname{int}(N_2/2)=19$,

$$
x(n)=
\begin{cases}
1,&0\le n<19,\\
0,&19\le n\le37.
\end{cases}
$$

То есть сигнал содержит 19 единичных и 19 нулевых отсчётов.

Реакция рассчитывается по формуле линейной свёртки:

$$
y_1(n)=\sum_{m=0}^{37}x(m)h(n-m).
$$


Длина входного импульса - **38 отсчётов**, а не 19: нулевые отсчёты также входят в заданную последовательность. Длина используемого фрагмента импульсной характеристики - 28 отсчётов. Поэтому аналитическая длина полной свёртки равна

$$
L_{y_1}=N_2+N_1-1=38+28-1=65.
$$

Это подтверждается графиком: отсчёты имеют индексы от $0$ до $64$ включительно.

При сравнении с реакцией $y_2(n)$ свёртку ограничивают первыми $N_2=38$ отсчётами, чтобы обе реакции были заданы на одинаковом интервале. Ненулевые отсчёты рассчитанной свёртки заканчиваются при $n=45$: отсчёты с $n=46$ по $n=64$ равны нулю из-за усечения $h(n)$. Реальная реакция рекурсивной системы продолжается и после $n=45$.

```python
x1 = np.zeros(N2)
x1[:round(N2 / 2)] = 1
y1_full = np.convolve(x1, h)
y1 = y1_full[:N2]

fig, axes = plt.subplots(2, 1, figsize=(10, 7))
stem(axes[0], np.arange(N2), x1)
axes[0].set(title="Входной прямоугольный импульс", xlabel="$n$", ylabel="$x(n)$", ylim=(-0.15, 1.2))
stem(axes[1], np.arange(len(y1_full)), y1_full, color="C1")
axes[1].axvline(N2 - 1, color="C3", linestyle="--", label="граница первых $N_2$ отсчётов")
axes[1].set(title="Полная линейная свёртка $y_1(n)=x(n)*h(n)$", xlabel="$n$", ylabel="$y_1(n)$")
axes[1].legend()
for ax in axes:
    finish_axes(ax)
fig.tight_layout()
plt.show()

print(f"Длина x: {len(x1)}; единичных отсчётов: {int(x1.sum())}")
print(f"Длина полной свёртки: {len(y1_full)} = {N2} + {N1} - 1")
```

## Задание 3. Реакция $y_2(n)$ по разностному уравнению

Разностное уравнение рекурсивного звена с заданными коэффициентами:

$$
y(n)=0{,}86x(n)-1{,}120236x(n-1)+1{,}204x(n-2)
-1{,}2278y(n-1)-0{,}748y(n-2)
$$

при $x(n)=y(n)=0$ для $n<0$. Длина рассчитанной реакции - 38 отсчётов.


До $n=27$ включительно графики совпадают с максимальной абсолютной ошибкой $8{,}33\cdot10^{-16}$, то есть с точностью до округления чисел двойной точности.

Начиная с $n=28$, видно небольшое расхождение. Его причина - использование в свёртке только 28 отсчётов бесконечной импульсной характеристики. Разностное уравнение учитывает всю предыдущую историю рекурсивно, а усечённая свёртка отбрасывает $h(28),h(29),\ldots$. Максимальная разница на интервале $0\le n<38$ составляет $0{,}052165$. Это не ошибка разностного уравнения, а следствие усечения $h(n)$.

```python
y2 = recursive_filter(b, a, x1)
convolution_error = np.max(np.abs(y1 - y2))
initial_error = np.max(np.abs(y1[:N1] - y2[:N1]))

fig, axes = plt.subplots(2, 1, figsize=(10, 7))
stem(axes[0], np.arange(N2), y2, color="C2")
axes[0].set(title="Реакция по разностному уравнению", xlabel="$n$", ylabel="$y_2(n)$")
axes[1].plot(np.arange(N2), y1, "o-", markersize=3, label="$y_1(n)$ - свёртка")
axes[1].plot(np.arange(N2), y2, "x--", markersize=4, label="$y_2(n)$ - разностное уравнение")
axes[1].set(title="Сравнение реакций на интервале $0\\leq n<N_2$", xlabel="$n$", ylabel="$y(n)$")
axes[1].legend()
for ax in axes:
    finish_axes(ax)
fig.tight_layout()
plt.show()

print(f"Максимальная ошибка на первых N1 отсчётах: {initial_error:.3e}")
print(f"Максимальная ошибка на всём интервале N2: {convolution_error:.6f}")
np.testing.assert_allclose(y1[:N1], y2[:N1], rtol=1e-12, atol=1e-12)
```

## Задание 4. Нули и полюса передаточной функции

После умножения числителя и знаменателя $H(z)$ на $z^2$:

$$
H(z)=\frac{0{,}86z^2-1{,}120236z+1{,}204}
{z^2+1{,}2278z+0{,}748}.
$$

Нули определяются корнями числителя, полюса - корнями знаменателя.

| Элемент | Алгебраическая форма | Модуль | Аргумент, рад | Аргумент/$\pi$ |
|---|---|---:|---:|---:|
| $z_1$ | $0{,}651300+j0{,}987830$ | $1{,}183216$ | $0{,}987894$ | $0{,}314457$ |
| $z_2$ | $0{,}651300-j0{,}987830$ | $1{,}183216$ | $-0{,}987894$ | $-0{,}314457$ |
| $p_1$ | $-0{,}613900+j0{,}609202$ | $0{,}864870$ | $2{,}360036$ | $0{,}751223$ |
| $p_2$ | $-0{,}613900-j0{,}609202$ | $0{,}864870$ | $-2{,}360036$ | $-0{,}751223$ |

В показательной форме:

$$
\begin{aligned}
z_{1,2}&=1{,}183216e^{\pm j0{,}987894},\\
p_{1,2}&=0{,}864870e^{\pm j2{,}360036}.
\end{aligned}
$$

Нули и полюса образуют комплексно-сопряжённые пары, поскольку все коэффициенты передаточной функции действительны.

Представление $H(z)$ произведением простейших множителей:

$$
H(z)=0{,}86
\frac{(z-z_1)(z-z_2)}{(z-p_1)(z-p_2)},
$$

или с явным использованием показательной формы:

$$
H(z)=0{,}86
\frac{(z-1{,}183216e^{j0{,}987894})(z-1{,}183216e^{-j0{,}987894})}
{(z-0{,}864870e^{j2{,}360036})(z-0{,}864870e^{-j2{,}360036})}.
$$

Эквивалентная запись через $z^{-1}$ использует множители $(1-z_i z^{-1})$ и $(1-p_i z^{-1})$.

```python
zeros = np.roots([b0, b1, b2])
poles = np.roots([a0, a1, a2])
stable = np.all(np.abs(poles) < 1)

for name, values in [("Нули", zeros), ("Полюса", poles)]:
    for index, value in enumerate(values, 1):
        print(f"{'Нуль' if name == 'Нули' else 'Полюс'} {index}: {value:.6f}; |z| = {abs(value):.6f}; "
              f"arg(z) = {np.angle(value):.6f} рад = {np.angle(value) / np.pi:.6f}pi")
print("Система устойчива:", stable)
assert stable
```

## Задание 2. ДПФ и ОДПФ последовательности

Прямое и обратное преобразования:

$$
X(k)=\sum_{n=0}^{63}x(n)e^{-j2\pi kn/64},
$$

$$
\widetilde x(n)=\frac1{64}\sum_{k=0}^{63}X(k)e^{j2\pi kn/64}.

Полученная последовательность совпадает с исходной. Максимальная абсолютная ошибка:

$$
\max_n|\widetilde x(n)-x(n)|=1{,}337\cdot10^{-15}.

Это численная ошибка округления, аналитически ОДПФ от ДПФ точно восстанавливает исходную последовательность.
