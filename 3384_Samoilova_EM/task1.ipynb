### Теоретические сведения

**1. Передаточная функция дискретной системы**

Определение передаточной функции через Z-преобразование входного и выходного сигналов:
$H(z)=\frac{Y(z)}{X(z)}$.

Для рекурсивного звена второго порядка:
$$
H(z)=\frac{b_0+b_1z^{-1}+b_2z^{-2}}{a_0+a_1z^{-1}+a_2z^{-2}}
$$

Здесь $b_i$ - коэффициенты числителя, $a_i$ - коэффициенты знаменателя.

**2. Импульсная характеристика**

Импульсная характеристика $(h(n))$ — это выходная последовательность линейной дискретной системы при воздействии на её вход единичным импульсом при нулевых начальных условиях.

Единичный импульс:
$
\delta(n)=
\begin{cases}
1,&n=0,\\
0,&n\neq0.
\end{cases}
$

Импульсная характеристика связана с передаточной функцией обратным Z-преобразованием:
$h(n)=\mathcal{Z}^{-1}\{H(z)\}.$

**3. Аналитическая формула для рекурсивного звена второго порядка**

Характеристическое уравнение:

$$
a_0z^2+a_1z+a_2=0.
$$

Если корни комплексно-сопряжённые:

$$
p_{1,2}=\rho e^{\pm j\theta},
$$

то импульсная характеристика при $n\geq1$ имеет вид:

$$
h(n)=\rho^{n-1}
\left[
C\cos((n-1)\theta)+
D\sin((n-1)\theta)
\right].
$$

Константы $C$ и $D$ определяются коэффициентами передаточной функции и начальными условиями.

**4. Связь корней характеристического уравнения с устойчивостью системы.**

Модуль корней характеристического уравнения определяет характер изменения импульсной характеристики при увеличении номера отсчёта:

- Если $\rho<1$, импульсная характеристика затухает, поскольку множитель $\rho^{n-1}$ стремится к нулю. Полюса находятся внутри единичной окружности, система устойчива.
- Если $\rho=1$, амплитуда колебаний не затухает. Полюса находятся на единичной окружности, система не является асимптотически устойчивой.
- Если $\rho>1$, амплитуда колебаний возрастает. Полюса находятся вне единичной окружности, система неустойчива.

Таким образом, для устойчивости причинного рекурсивного звена второго порядка необходимо, чтобы все корни характеристического уравнения находились строго внутри единичной окружности:

$$
|p_1|<1,\qquad |p_2|<1.
$$

**5. Дискретное преобразование Фурье**

Дискретное преобразование Фурье (ДПФ) используется для представления конечной последовательности отсчётов в частотной области. Оно позволяет определить частотный состав дискретного сигнала и исследовать его гармонические составляющие.

Для последовательности $x(n)$ длины $N$ прямое дискретное преобразование Фурье определяется выражением:

$$
X(k)=\sum_{n=0}^{N-1}x(n)e^{-j\frac{2\pi}{N}kn},
$$

где $n=0,\ldots,N-1$ — номер отсчёта во временной области, $k=0,\ldots,N-1$ — номер частотного отсчёта, $N$ — длина последовательности, $j$ — мнимая единица.

Обратное дискретное преобразование Фурье позволяет восстановить исходную последовательность по её спектральным отсчётам:

$$
x(n)=\frac{1}{N}\sum_{k=0}^{N-1}X(k)e^{j\frac{2\pi}{N}kn}.
$$

Таким образом, ДПФ устанавливает взаимно однозначное соответствие между конечной последовательностью отсчётов и её спектральным представлением.

**6.Амплитудный и фазовый спектры**

Результатом ДПФ является комплексная последовательность $X(k)$, которую можно представить в показательной форме:

$$
X(k)=|X(k)|e^{j\varphi(k)}.
$$

Модуль комплексного спектра определяет амплитудный спектр, а аргумент — фазовый спектр:

$$
A(k)=|X(k)|,
$$

$$
\varphi(k)=\arg X(k).
$$

**7. Дискретное время и дискретная частота:**

- Дискретное время: $t_n=nT$, где $T=1/f_д$ — период дискретизации, $f_д$ — частота дискретизации.
- Дискретное нормированное время: $n=t_n/T$ — целочисленный номер отсчёта.
- Индекс частотного отсчёта ДПФ: $k$, где $k=0,1,\ldots,N-1$.
- Дискретная нормированная частота: $\hat f_k=k/N$ — отношение частоты $k$-го отсчёта ДПФ к частоте дискретизации.
- Абсолютная дискретная частота: 

$
f_k=\frac{k f_{\text{д}}}{N}=\frac{k}{NT}.
$

Шаг по частоте (частотное разрешение ДПФ) определяется выражением:

$
\Delta f=\frac{f_{\text{д}}}{N}=\frac{1}{NT}.
$

Для получения нормированного двустороннего амплитудного спектра модуль ДПФ делится на количество отсчётов $N$:

$
A(k)=\frac{|X(k)|}{N}.
$

При построении одностороннего амплитудного спектра амплитуды ненулевых частотных составляющих, кроме постоянной составляющей и составляющей на частоте Найквиста (при чётном $N$), дополнительно умножаются на два.

**8. Равенство Парсеваля**

Равенство Парсеваля устанавливает связь между энергией дискретного сигнала во временной и частотной областях:

$$
\sum_{n=0}^{N-1}|x(n)|^2
=
\frac{1}{N}\sum_{k=0}^{N-1}|X(k)|^2.
$$

Оно показывает, что суммарная энергия сигнала не изменяется при переходе от временного представления к частотному. Данное свойство может использоваться для проверки правильности вычисления дискретного преобразования Фурье.

1. Вычислить импульсную характеристику длины $N_1$ по заданной передаточной функции с коэффициентами $b_i$, $a_k$ и построить её график. Записать аналитическую формулу импульсной характеристики рекурсивного звена 2-го порядка с учетом нулевых начальных условий. Пояснить, чему в действительности равна длина импульсной характеристики.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import lfilter

b = [b0, b1, b2]
a = [a0, a1, a2]

# Единичный импульс
delta = np.zeros(N1)
delta[0] = 1

# Вычисление импульсной характеристики
h = lfilter(b, a, delta)

# Построение графика
n = np.arange(N1)

plt.figure(figsize=(10, 5))
plt.stem(n, h)
plt.title('Импульсная характеристика системы')
plt.xlabel('n')
plt.ylabel('h(n)')
plt.grid(True)
plt.show()

print("Коэффициенты b:", b)
print("Коэффициенты a:", a)
print("Импульсная характеристика:", h)
```

Для получения аналитической формулы импульсной характеристики подставим единичный импульс $x[n]=\delta[n]$ в разностное уравнение системы. При этом $y[n]=h[n]$:

$h[n]=1.22\delta[n]-1.98006\delta[n-1]+1.22\delta[n-2]-1.6778h[n-1]-0.856h[n-2]$.

При нулевых начальных условиях $h[-1]=h[-2]=0$. Первые три значения импульсной характеристики определяются подстановкой $(n=0,1,2)$ в полученное уравнение. Их численные значения вычислены программно.

При $n\geq3$ все слагаемые с единичным импульсом равны нулю. Поэтому получаем однородное разностное уравнение:

$ \boxed{ h[n]+1.6778h[n-1]+0.856h[n-2]=0 } $

```python
import numpy as np

# Начальные значения импульсной характеристики
h0 = b[0]
h1 = b[1] - a[1] * h0
h2 = b[2] - a[1] * h1 - a[2] * h0

# Корни характеристического уравнения
roots = np.roots(a)

# Модуль и аргумент корней
rho = np.abs(roots[0])
theta = np.angle(roots[0])

# Коэффициенты аналитической формулы
C = h1
D = (h2 / rho - C * np.cos(theta)) / np.sin(theta)

print(f"h[0] = {h0:.10f}")
print(f"h[1] = {h1:.10f}")
print(f"h[2] = {h2:.10f}")

print(f"Корни: {roots}")
print(f"Модуль корней rho = {rho:.10f}")
print(f"Аргумент корней theta = {theta:.10f} рад")

print(f"C = {C:.10f}")
print(f"D = {D:.10f}")
```

Было получено два комплексно сопряженных корня:

$\lambda_1 = -0.8389+0.39018815j$

$\lambda_2 = -0.8389-0.39018815j$

Корни необходимо привести к показательной форме:
Комплексный корень представляется как: $\lambda=\rho e^{j\theta}$, где $\rho$ — модуль корня, $\theta$ — его аргумент.

По результатам вычислений:
$\rho=0.925203,\qquad \theta=2.706237.$

Подставляем все найденные параметры:

$ \boxed{ \begin{aligned} h[n]=0.925203^{n-1}\big[& -4.026976\cos(2.706237(n-1))\\ &+9.108196\sin(2.706237(n-1)) \big]. \end{aligned} } $

Формула действует при $n\geq1$. Для нулевого отсчёта отдельно:

$ h[0]=1.22 $

Модуль обоих полюсов:
$ |\lambda_1|=|\lambda_2|=0.925203. $
Поскольку:
$ 0.925203<1, $
полюса находятся внутри единичной окружности.

Следовательно, система устойчива, а импульсная характеристика будет затухать при увеличении $n$.

Теоретически импульсная характеристика рекурсивного звена имеет бесконечную длину, поскольку выходная последовательность не становится строго нулевой за конечное число отсчётов.


2. Вычислить реакцию $y_1(n)$ по формуле свёртки. В качестве воздействия $x(n)$
длины $N_2$ выбрать дискретный прямоугольный импульс:

$$x(n) = \begin{cases} 1, & 0 \le n < int(N_2/2) \\ 0, & int(N_2/2) \le n \le N_2 - 1 \end{cases}$$

 где функция $int(t)$ – округление до ближайшего целого. Вывести графики воздействия и реакции. Пояснить:

 - чему равна длина свёртки аналитически и по графику;
 - почему её ограничивают до длины воздействия.

```python
import numpy as np
import matplotlib.pyplot as plt

# Формирование прямоугольного импульса
x = np.zeros(N2)
x[:int(N2 / 2)] = 1

# Полная линейная свёртка
y_full = np.convolve(x, h)

# Ограничение реакции длиной воздействия
y1 = y_full[:N2]

# Построение графиков
n_x = np.arange(N2)
n_y = np.arange(N2)

fig, axes = plt.subplots(2, 1, figsize=(10, 7))

axes[0].stem(n_x, x)
axes[0].set_title('Входное воздействие x(n)')
axes[0].set_xlabel('n')
axes[0].set_ylabel('x(n)')
axes[0].grid(True)

axes[1].stem(n_y, y1)
axes[1].set_title('Реакция системы y1(n)')
axes[1].set_xlabel('n')
axes[1].set_ylabel('y1(n)')
axes[1].grid(True)

plt.tight_layout()
plt.show()

print("Длина воздействия:", len(x))
print("Длина прямоугольной части:", np.count_nonzero(x))
print("Длина полной свёртки:", len(y_full))
print("Длина ограниченной реакции:", len(y1))
```

$
y_1(n)=\sum_{k=0}^{n}x(k)h(n-k).
$

Поскольку длина импульсной характеристики составляет 26 отсчётов, а длина входного воздействия — 36 отсчётов, длина полной линейной свёртки равна:

$
N_y=N_1+N_2-1=26+36-1=61.
$

В соответствии с условием работы полученная реакция была ограничена первыми 36 отсчётами. Это позволяет рассматривать реакцию на интервале определения входного воздействия и впоследствии сравнить её с результатом вычисления по разностному уравнению.

4. Вычислить нули и полюсы передаточной функции нули и полюсы в алгебраической и показательной формах и пояснить связь между ними. Выразить значение аргумента полюса и нуля относительно π. Представить передаточную функцию в виде произведения простейших множителей с нулями и полюсами в показательной форме.

```python
zeros = np.roots([b0, b1, b2])    # корни числителя: b0*z^2 + b1*z + b2 = 0
poles = np.roots([1.0, a1, a2])   # корни знаменателя: z^2 + a1*z + a2 = 0

def show_roots(r, name):
    print(name)
    for z in r:
        rho, theta = abs(z), np.angle(z)
        print(f"  z = {z.real:.4f} {z.imag:+.4f}j")
        print(f"    |z| = {rho:.4f},  arg z = {theta:.4f} рад = {theta/np.pi:.4f} * pi")

show_roots(zeros, "Нули:")
show_roots(poles, "Полюсы:")
```

**Связь между формами.** Алгебраическая форма $z = \mathrm{Re}\,z + j\,\mathrm{Im}\,z$ и показательная $z = \rho e^{j\theta}$ описывают одну и ту же точку комплексной плоскости: $\rho = |z| = \sqrt{(\mathrm{Re}\,z)^2 + (\mathrm{Im}\,z)^2}$ — модуль (расстояние до начала координат), $\theta = \arg z = \operatorname{arctg}\frac{\mathrm{Im}\,z}{\mathrm{Re}\,z}$ — аргумент (угол от положительной вещественной оси). Так как все коэффициенты фильтра вещественные, корни образуют комплексно-сопряжённые пары.

Нули лежат **на единичной окружности** ($|z_{1,2}| = 1$), так как числитель симметричен: $b_0 = b_2$. Аргумент нулей $\theta_z = \pm 0{,}1987\pi$ задаёт частоту идеального подавления сигнала. Полюсы находятся **внутри единичной окружности** ($|p_{1,2}| = 0{,}9252 < 1$), их аргумент $\theta_p = \pm 0{,}8614\pi$ близок к $\pi$, что указывает на высокочастотный характер резонанса звена.

 **Передаточная функция в виде произведения простейших множителей:**

3. Вычислить амплитудный и фазовый спектры последовательности x(n) и вывести их графики в шкалах дискретных нормированных и абсолютных частот. 

Пояснить:
 - связь дискретного нормированного времени с дискретным временем;
 - связь частоты f (Гц) с дискретной нормированной частотой;
 - вид амплитудного и фазового спектров.

```python
# Частотные оси
fd = 1 / T                         # Частота дискретизации, Гц
freq = np.fft.fftfreq(N, d=T)       # Абсолютная частота, Гц
freq_norm = freq / fd              # Дискретная нормированная частота

# Центрирование спектра
X_shifted = np.fft.fftshift(X)
freq_shifted = np.fft.fftshift(freq)
freq_norm_shifted = np.fft.fftshift(freq_norm)

# Амплитудный и фазовый спектры
amplitude = np.abs(X_shifted) / N
phase = np.angle(X_shifted)

# Фаза в точках с практически нулевой амплитудой не определена
threshold = 1e-10 * amplitude.max()
phase[amplitude < threshold] = np.nan

# Построение графиков
fig, axes = plt.subplots(2, 2, figsize=(14, 9))

# Амплитудный спектр в нормированных частотах
axes[0, 0].stem(freq_norm_shifted, amplitude)
axes[0, 0].set_title('Амплитудный спектр')
axes[0, 0].set_xlabel('Нормированная частота')
axes[0, 0].set_ylabel('|X(k)| / N')
axes[0, 0].grid(True)

# Амплитудный спектр в герцах
axes[0, 1].stem(freq_shifted, amplitude)
axes[0, 1].set_title('Амплитудный спектр')
axes[0, 1].set_xlabel('Частота, Гц')
axes[0, 1].set_ylabel('|X(k)| / N')
axes[0, 1].grid(True)

# Фазовый спектр в нормированных частотах
axes[1, 0].stem(freq_norm_shifted, phase)
axes[1, 0].set_title('Фазовый спектр')
axes[1, 0].set_xlabel('Нормированная частота')
axes[1, 0].set_ylabel('Фаза, рад')
axes[1, 0].grid(True)

# Фазовый спектр в герцах
axes[1, 1].stem(freq_shifted, phase)
axes[1, 1].set_title('Фазовый спектр')
axes[1, 1].set_xlabel('Частота, Гц')
axes[1, 1].set_ylabel('Фаза, рад')
axes[1, 1].grid(True)

plt.tight_layout()
plt.show()
```

**Связь дискретного времени с дискретным нормированным временем:**

Дискретное время определяется как $t=nT$, где $n$ — номер отсчёта, $T$ — период дискретизации. Нормированное время получается делением дискретного времени на период дискретизации:

$$
\hat{t}=\frac{t}{T}=n.
$$

**Связь абсолютной частоты с дискретной нормированной частотой:**

Дискретная нормированная частота определяется как отношение абсолютной частоты к частоте дискретизации:

$$
\hat{f}=\frac{f}{f_{\text{д}}}=fT.
$$

Для частотных отсчётов ДПФ:

$$
f_k=k\frac{f_{\text{д}}}{N}, \qquad
\hat{f}_k=\frac{k}{N},
$$

где $k$ — номер частотного отсчёта, $N$ — количество отсчётов сигнала, $f_{\text{д}}$ — частота дискретизации.

**Вид амплитудного и фазового спектра:**

Амплитудный спектр имеет дискретный характер: основная часть его значений равна нулю, а на частотах, соответствующих двум гармоническим составляющим исходного сигнала, наблюдаются четыре выраженных пика. Пики расположены симметрично относительно нулевой частоты, поскольку исходный сигнал является суммой косинусоид. На положительных и отрицательных частотах амплитуды соответствующих составляющих одинаковы. 

Фазовый спектр также имеет дискретный характер. На частотах, соответствующих ненулевым составляющим амплитудного спектра, наблюдаются значения фаз, определяемые начальными фазами исходных косинусоид. Для вещественной последовательности фазовый спектр является нечётным с точностью до $2\pi$: значения фаз на отрицательных частотах противоположны значениям на соответствующих положительных частотах. В точках, где амплитуда спектра равна нулю, фаза не определена.

Вывести графики в шкале дискретных нормированных частот модуля ДПФ конечной последовательности и амплитудного спектра периодической последовательности. Пояснить связь модуля ДПФ конечной последовательности с амплитудным спектром периодической последовательности.

```python
# ДПФ конечной последовательности
X = np.fft.fft(x)

# Дискретная нормированная частота
freq_norm = np.fft.fftfreq(N, d=1)

# Центрирование спектров
freq_norm_shifted = np.fft.fftshift(freq_norm)
X_shifted = np.fft.fftshift(X)

# Модуль ДПФ и амплитудный спектр периодической последовательности
dft_magnitude = np.abs(X_shifted)
amplitude_spectrum = dft_magnitude / N

# Построение графиков
fig, axes = plt.subplots(2, 1, figsize=(12, 8))

axes[0].stem(freq_norm_shifted, dft_magnitude)
axes[0].set_title('Модуль ДПФ конечной последовательности')
axes[0].set_xlabel('Дискретная нормированная частота')
axes[0].set_ylabel('|X(k)|')
axes[0].grid(True)

axes[1].stem(freq_norm_shifted, amplitude_spectrum)
axes[1].set_title('Амплитудный спектр периодической последовательности')
axes[1].set_xlabel('Дискретная нормированная частота')
axes[1].set_ylabel('|X(k)| / N')
axes[1].grid(True)

plt.tight_layout()
plt.show()
```

**Связь модуля ДПФ конечной последовательности с амплитудным спектром периодической последовательности**

Амплитудный спектр периодической последовательности получается из модуля дискретного преобразования Фурье конечной последовательности путём деления его значений на длину последовательности $N$:

$$
A(k)=\frac{|X(k)|}{N}.
$$

Таким образом, модуль ДПФ и амплитудный спектр имеют одинаковое расположение спектральных составляющих и одинаковую форму, но отличаются масштабом по вертикальной оси. Значения амплитудного спектра в $N$ раз меньше соответствующих значений модуля ДПФ. Следовательно, модуль ДПФ характеризует ненормированные спектральные составляющие конечной последовательности, а деление на $N$ позволяет получить амплитуды гармоник периодического сигнала.

5. Восстановить периодический аналоговый сигнал x(t) по отсчётам ДПФ X(k) периодической последовательности x(n). Для вычисления значений сигнала x(t) использовать формулу:
$$x(t) = \frac{1}{N}\sum_{k=-N/2}^{N/2-1} X_а(k)\, e^{j\frac{2\pi}{NT}kt}$$

где отсчеты X_а(k) связаны с отсчетами ДПФ X(k) соотношением:
$$\qquad X_а(k) = \begin{cases} X(N+k), & -N/2 \le k \le -1 \\ X(k), & 0 \le k \le N/2-1 \end{cases}$$
задавая значения времени $t$ на интервале $t ∈ [0;(N−1)T]$ с шагом $Δt=0.25T$.

Вывести графики:
 - периодической последовательности $x(n)$ и модуля ее ДПФ;
 - восстановленного аналогового сигнала x(t) и его амплитудного спектра;
 - исходного аналогового сигнала $x(t)=x(nT)∣_nT=t$.

```python
# Индексы частот от -N/2 до N/2 - 1
k = np.arange(-N // 2, N // 2)

# Спектральные отсчёты X_a(k)
X_a = X[k % N]

# Временная сетка с шагом 0.25T
dt = 0.25 * T
t = np.arange(0, (N - 1) * T + dt / 2, dt)

# Восстановление аналогового сигнала
x_restored = np.zeros(len(t), dtype=complex)

for i, time in enumerate(t):
    x_restored[i] = np.sum(
        X_a * np.exp(1j * 2 * np.pi * k * time / (N * T))
    ) / N

# Действительная часть
x_restored = np.real(x_restored)

# Амплитудный спектр восстановленного сигнала
amplitude_spectrum = np.abs(X_a) / N

# Частоты спектральных составляющих, Гц
freq = k / (N * T)

# Построение графиков
fig, axes = plt.subplots(3, 2, figsize=(14, 13))

# 1. Периодическая последовательность x(n)
axes[0, 0].stem(np.arange(N), x)
axes[0, 0].set_title('Периодическая последовательность x(n)')
axes[0, 0].set_xlabel('Номер отсчёта n')
axes[0, 0].set_ylabel('x(n)')
axes[0, 0].grid(True)

# 2. Модуль ДПФ
axes[0, 1].stem(k, np.abs(X_a))
axes[0, 1].set_title('Модуль ДПФ |X(k)|')
axes[0, 1].set_xlabel('Частотный индекс k')
axes[0, 1].set_ylabel('|X(k)|')
axes[0, 1].grid(True)

# 3. Восстановленный аналоговый сигнал
axes[1, 0].plot(t, x_restored)
axes[1, 0].set_title('Восстановленный аналоговый сигнал x(t)')
axes[1, 0].set_xlabel('Время, с')
axes[1, 0].set_ylabel('x(t)')
axes[1, 0].grid(True)

# 4. Амплитудный спектр восстановленного сигнала
axes[1, 1].stem(freq, amplitude_spectrum)
axes[1, 1].set_title('Амплитудный спектр восстановленного сигнала')
axes[1, 1].set_xlabel('Частота, Гц')
axes[1, 1].set_ylabel('Амплитуда')
axes[1, 1].grid(True)

# 5. Исходный аналоговый сигнал
x_original = (
    A1 * np.cos(2 * np.pi * f1 * t + np.pi / 4)
    + A2 * np.cos(2 * np.pi * f2 * t + np.pi / 8)
)

axes[2, 0].plot(t, x_original)
axes[2, 0].set_title('Исходный аналоговый сигнал x(t)')
axes[2, 0].set_xlabel('Время, с')
axes[2, 0].set_ylabel('x(t)')
axes[2, 0].grid(True)

axes[2, 1].axis('off')

plt.tight_layout()
plt.show()
```

7. Проверить равенство Парсеваля. Пояснить его смысл.

```python
# Вычисление левой и правой частей равенства Парсеваля

left_part = np.sum(np.abs(x) ** 2)
right_part = np.sum(np.abs(X) ** 2) / N

print("Левая часть:", left_part)
print("Правая часть:", right_part)
print("Абсолютная погрешность:", abs(left_part - right_part))

if np.isclose(left_part, right_part):
    print("Равенство Парсеваля выполняется.")
else:
    print("Равенство Парсеваля не выполняется.")
```

Смысл равенства Парсеваля заключается в том, что энергия конечной последовательности может быть вычислена как во временной области, так и в частотной области. При этом суммарная энергия сигнала не изменяется при переходе от представления последовательности к её спектральному представлению.
