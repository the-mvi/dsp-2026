## Основные теоретические положения



### 1.1. Разностное уравнение и передаточная функция
Рекурсивное звено 2-го порядка описывается линейным разностным уравнением:
$
 y[n] = \sum_{i=0}^{2} b_i x[n-i] - \sum_{k=1}^{2} a_k y[n-k]
$

Применяя Z-преобразование при нулевых начальных условиях, получаем передаточную функцию:
$
 H(z) = \frac{Y(z)}{X(z)} = \frac{b_0 + b_1 z^{-1} + b_2 z^{-2}}{a_0 + a_1 z^{-1} + a_2 z^{-2}}
$

### 1.2. Импульсная и переходная характеристики
*   **Импульсная характеристика (ИХ)** $h[n]$ — это реакция системы на единичный импульс $u_0[n]$ при нулевых начальных условиях. Аналитически она вычисляется через обратное Z-преобразование $H(z)$ или рекурсивно через разностное уравнение.
*   **Переходная характеристика** $g[n]$ — это реакция на единичный скачок $u_1[n]$. Она связана с ИХ соотношением: $g[n] = \sum_{k=0}^{n} h[k]$.

### 1.3. Линейная свертка
Реакция линейной стационарной системы $y[n]$ на произвольное воздействие $x[n]$ вычисляется как свертка этого воздействия с импульсной характеристикой:
$
 y[n] = x[n] * h[n] = \sum_{m=0}^{\infty} x[m] h[n-m]
$
Для двух конечных последовательностей длин $N_1$ и $N_2$ длина результата линейной свертки равна $L = N_1 + N_2 - 1$.

### 1.4. Нули, полюсы и устойчивость
Корни числителя $H(z)$ называются **нулями**, а корни знаменателя — **полюсами**.
Цифровой фильтр **устойчив**, если все его полюсы лежат строго внутри единичной окружности на Z-плоскости ($|p_i| < 1$).
Аргумент полюса $\varphi$ связан с нормированной резонансной частотой $\hat{\omega}$ соотношением $\varphi = \hat{\omega}$. Выражение аргумента в долях $\pi$ показывает положение полюса на комплексной плоскости.

### 1.5. Частотные характеристики (АЧХ и ФЧХ)
Частотная характеристика получается подстановкой $z = e^{j\hat{\omega}}$ в передаточную функцию, где $\hat{\omega} \in [0, \pi]$ — нормированная круговая частота.
*   **Амплитудно-частотная характеристика (АЧХ):** $A(\hat{\omega}) = |H(e^{j\hat{\omega}})|$
*   **Фазочастотная характеристика (ФЧХ):** $\Phi(\hat{\omega}) = \arg(H(e^{j\hat{\omega}}))$
Связь нормированной частоты $\hat{\omega}$ с физической частотой $f$ (в Гц) и частотой дискретизации $f_d$: $\hat{\omega} = 2\pi \frac{f}{f_d}$.

---

### 2.1. Дискретизация и шкалы времени
Аналоговый сигнал $x(t)$ дискретизируется с шагом $T = 1/f_d$, образуя последовательность $x(n) = x(nT)$.
*   **Дискретное время:** $n = 0, 1, 2, \dots, N-1$.
*   **Дискретное нормированное время:** $\hat{t} = \frac{n}{N}$, где $N$ — период.

### 2.2. Дискретное преобразование Фурье (ДПФ)
Для перехода в частотную область используется прямое ДПФ:
$
 X(k) = \sum_{n=0}^{N-1} x(n) \exp\left(-j \frac{2\pi}{N} kn\right), \quad k = 0, \dots, N-1
$
Обратное ДПФ (ОДПФ) полностью восстанавливает исходную последовательность:
$
 x(n) = \frac{1}{N} \sum_{k=0}^{N-1} X(k) \exp\left(j \frac{2\pi}{N} kn\right)
$

### 2.3. Спектры и связь частот
*   **Амплитудный спектр:** $|X(k)|$.
*   **Фазовый спектр:** $\arg(X(k))$.
Номер отсчета $k$ связан с абсолютной частотой $f$ (Гц) и нормированной частотой $\hat{f}$:
$
 f = k \frac{f_d}{N}, \quad \hat{f} = \frac{k}{N}
$
Модуль ДПФ $|X(k)|$ связан с амплитудами гармоник ряда Фурье $|C_k|$ периодического сигнала как $|X(k)| = N \cdot |C_k|$.

### 2.4. Восстановление аналогового сигнала
Если выполняется теорема Котельникова ($f_d \ge 2f_{\max}$), непрерывный сигнал можно восстановить в любой момент времени $t$ (а не только в точках отсчета). Для периодического сигнала восстановление выполняется через ряд Фурье с использованием центрированных коэффициентов ДПФ $X_a(k)$:
$
 x(t) = \frac{1}{N} \sum_{k=-N/2}^{N/2-1} X_a(k) \exp\left(j \frac{2\pi}{NT} kt\right)
$
Сдвиг спектра $X_a(k)$ формируется из стандартного ДПФ так, чтобы отрицательные частоты шли первыми.

### 2.5. Равенство Парсеваля
Закон сохранения энергии. Энергия сигнала во временной области равна энергии в частотной области:
$
 E_{\text{время}} = \sum_{n=0}^{N-1} |x(n)|^2 = \frac{1}{N} \sum_{k=0}^{N-1} |X(k)|^2 = E_{\text{частота}}
$
ДПФ является унитарным преобразованием: оно не создает и не уничтожает энергию, а лишь перераспределяет её по гармоникам.

### 4. Нули и полюсы передаточной функции

**Передаточная функция:**
$$H(z) = \frac{0.84 + 1.079232 z^{-1} + 1.008 z^{-2}}{1 - 1.2028 z^{-1} + 0.742 z^{-2}}$$

Умножим числитель и знаменатель на $z^2$:
$$H(z) = \frac{0.84 z^2 + 1.079232 z + 1.008}{z^2 - 1.2028 z + 0.742} = \frac{b_0(z - z_{1})(z - z_{2})}{(z - p_1)(z - p_2)}$$

**Алгебраическая форма** (комплексные числа $a + jb$):
Нули $z_0$ (корни числителя):
$$z_{1, 2} = \frac{-1.079232 \pm \sqrt{1.079232^2 - 4 \cdot 0.84 \cdot 1.008}}{2 \cdot 0.84}$$
Полюсы $p$ (корни знаменателя):
$$p_{1, 2} = \frac{1.2028 \pm \sqrt{(-1.2028)^2 - 4 \cdot 1 \cdot 0.742}}{2}$$

**Показательная форма** ($|z| e^{j\varphi}$): модуль и аргумент.

**Связь аргумента с $\pi$:** $\varphi / \pi$ показывает долю числа $\pi$.

**Произведение простейших множителей:**
$$H(z) = b_0 \frac{(1 - z_{1} z^{-1})(1 - z_{2} z^{-1})}{(1 - p_1 z^{-1})(1 - p_2 z^{-1})} = 0.84 \frac{(1 - z_{1} z^{-1})(1 - z_{2} z^{-1})}{(1 - p_1 z^{-1})(1 - p_2 z^{-1})}$$



```python
import numpy as np
import matplotlib.pyplot as plt

b = np.array([0.84, 1.079232, 1.008])
a = np.array([1.0, -1.2028, 0.742])

# Вычисление корней
zeros = np.roots(b)
poles = np.roots(a)

def format_complex(z):
    # Возвращает алгебраическую и показательную формы
    alg = f"{z.real: .4f} {'+' if z.imag >= 0 else '-'} j{abs(z.imag):.4f}"
    mod = np.abs(z)
    arg = np.angle(z)
    exp_form = f"{mod:.4f} * e^(j * {arg:.4f})"
    arg_pi = arg / np.pi
    return alg, exp_form, arg_pi

print("НУЛИ")
for i, z in enumerate(zeros, 1):
    alg, exp_form, arg_pi = format_complex(z)
    print(f"z0_{i}: Алг. {alg} | Показ. {exp_form} | arg/pi = {arg_pi:.4f}")

print("\nПОЛЮСА")
for i, p in enumerate(poles, 1):
    alg, exp_form, arg_pi = format_complex(p)
    print(f"p_{i}:  Алг. {alg} | Показ. {exp_form} | arg/pi = {arg_pi:.4f}")
```

**Результаты:**
- **Нули** комплексно-сопряжённые, лежат вне единичной окружности.
- **Полюсы** комплексно-сопряжённые, их модуль $|p| = \sqrt{0.742} \approx 0.861 < 1$.
- Аргумент полюса $\varphi_p \approx \pm 0.2540 \pi$. Связь с $\pi$ показывает, что полюс находится во второй/третьей четверти, что соответствует резонансной частоте.
- Передаточная функция в виде множителей:
$$H(z) = 0.84 \frac{(1 - z_{01} z^{-1})(1 - z_{02} z^{-1})}{(1 - p_1 z^{-1})(1 - p_2 z^{-1})}$$

### 5. Карта нулей и полюсов

### 6. АЧХ и ФЧХ в нормированных частотах
Нормированная частота $\hat{\omega} = \omega T_d = 2\pi f / f_d$ изменяется в пределах $[0, \pi]$.
$$H(e^{j\hat{\omega}}) = \frac{0.84 + 1.079232 e^{-j\hat{\omega}} + 1.008 e^{-j2\hat{\omega}}}{1 - 1.2028 e^{-j\hat{\omega}} + 0.742 e^{-j2\hat{\omega}}}$$

```python
w_hat = np.linspace(0, np.pi, 500)
z = np.exp(1j * w_hat)
H = (b[0] + b[1]*z**-1 + b[2]*z**-2) / (a[0] + a[1]*z**-1 + a[2]*z**-2)

AFC = np.abs(H)
FFC = np.angle(H)

fig, ax = plt.subplots(2, 1, figsize=(8, 6))
ax[0].plot(w_hat/np.pi, AFC)
ax[0].set_title("АЧХ (нормированные частоты)")
ax[0].set_ylabel("|H|"); ax[0].grid(alpha=0.4)

ax[1].plot(w_hat/np.pi, FFC)
ax[1].set_title("ФЧХ (нормированные частоты)")
ax[1].set_xlabel("w_шапка / pi"); ax[1].set_ylabel("Фаза (рад)"); ax[1].grid(alpha=0.4)

plt.tight_layout()
plt.show()
```

**Пояснения:**

а. Границы основной полосы частот: от $\hat{\omega} = 0$ до $\hat{\omega} = \pi$ (от 0 до половины частоты дискретизации).

б. **Карта и АЧХ:** Полюсы (крестики), находящиеся близко к единичной окружности, дают «всплеск» на АЧХ на соответствующих частотах ($\hat{\omega} \approx 0.715\pi$). Нули (кружочки), лежащие на или рядом с единичной окружностью, создают «провалы».

в. **Скачок на $\pi$ (переворот фазы):** возникает, если нули (или полюсы) лежат на действительной оси (мнимая часть = 0). Если нуль лежит на единичной окружности (в точке $z = -1$), ФЧХ испытывает скачок на $\pi$. В нашем случае нули не на окружности, но могут быть близко к действительной оси.

г. **Подавление:** Поскольку полюсы ближе к единичной окружности в верхней/нижней полуплоскости, звено преимущественно подавляет **высокие** частоты (вблизи $\pi$) и/или постоянную составляющую (вблизи 0), в зависимости от положения нулей.

### 7. Проверка равенства Парсеваля

Для дискретной последовательности длины \(N\) равенство Парсеваля утверждает, что полная энергия сигнала, вычисленная во временной области, равна полной энергии, вычисленной в частотной области (через отсчеты ДПФ):
$E_{\text{время}} = \sum_{n=0}^{N-1} |x(n)|^2 = \frac{1}{N} \sum_{k=0}^{N-1} |X(k)|^2 = E_{\text{частота}}$

Исходный сигнал представляет собой сумму двух гармоник:
$x(n) = A_1 \cos(\omega_1 n + \varphi_1) + A_2 \cos(\omega_2 n + \varphi_2)$
где $\omega_1 = 2\pi \frac{f_1}{f_d} = 2\pi \frac{1}{8}$ и $\omega_2 = 2\pi \frac{f_2}{f_d} = 2\pi \frac{3}{16}$.

**1. Аналитический расчет энергии во временной области:**
Возведем сигнал в квадрат и просуммируем по периоду $N=64$:
$E_{\text{время}} = \sum_{n=0}^{N-1} \big( A_1 \cos(\omega_1 n + \varphi_1) + A_2 \cos(\omega_2 n + \varphi_2) \big)^2$
Раскрывая квадрат суммы, получим три слагаемых:
$E_{\text{время}} = A_1^2 \sum_{n=0}^{N-1} \cos^2(\omega_1 n + \varphi_1) + A_2^2 \sum_{n=0}^{N-1} \cos^2(\omega_2 n + \varphi_2) + 2A_1 A_2 \sum_{n=0}^{N-1} \cos(\omega_1 n + \varphi_1)\cos(\omega_2 n + \varphi_2)$

*   Первое слагаемое: так как $\omega_1 = 2\pi/8$, за период $N=64$ укладывается целое число периодов (8 циклов). Сумма квадратов косинуса за целое число периодов равна $\frac{N}{2}$:
$A_1^2 \cdot \frac{N}{2} = A_1^2 \cdot 32$
*   Второе слагаемое: аналогично, $\omega_2 = 2\pi \cdot 3/16$, за \(N=64\) укладывается 12 полных циклов (4 периода по 16 отсчетов). Сумма также равна $\frac{N}{2}$:
$A_2^2 \cdot \frac{N}{2} = A_2^2 \cdot 32$
*   Третье слагаемое (перекрестное): так как частоты \omega_1 и \omega_2 ортогональны на периоде, сумма произведений двух косинусов за период равна **нулю**.

Итоговая энергия во времени:
$E_{\text{время}} = 32 \cdot (A_1^2 + A_2^2) = 32 \cdot (1.17^2 + 2.34^2) = 32 \cdot (1.3689 + 5.4756) = 32 \cdot 6.8445 = 219.024$

**2. Аналитический расчет энергии в частотной области:**
Спектр ДПФ суммы двух гармоник состоит из четырех ненулевых отсчетов (две частоты и их зеркальные отображения на отрицательных частотах).
Для гармоники \(A_1\), коэффициенты ДПФ на частотах \(k_1\) и \(N-k_1\) равны по модулю \(\frac{N A_1}{2}\). Их суммарная энергия:
$\frac{1}{N} \left( \left|X(k_1)\right|^2 + \left|X(N-k_1)\right|^2 \right) = \frac{1}{N} \left( \left(\frac{N A_1}{2}\right)^2 + \left(\frac{N A_1}{2}\right)^2 \right) = \frac{1}{N} \cdot \frac{N^2 A_1^2}{2} = \frac{N}{2} A_1^2$
Аналогично для второй гармоники \(A_2\):
$\frac{N}{2} A_2^2$
Суммируя, получаем:
$E_{\text{частота}} = \frac{N}{2} (A_1^2 + A_2^2) = 32 \cdot (A_1^2 + A_2^2) = 219.024$

**Вывод:**
Аналитические расчеты показывают, что $E_{\text{время}} = E_{\text{частота}} = 219.024$.
Это подтверждает равенство Парсеваля: ДПФ сохраняет полную энергию сигнала. Переход из временной области в частотную не приводит к потере или генерации энергии, а лишь раскладывает исходный сигнал на ортогональные спектральные составляющие.

### Сложности, возникшие в процессе работы

**1. Разная длина массивов при сравнении свёртки и рекурсии**

**2. Восстановление аналогового сигнала в произвольные моменты времени**
*   **Проблема:** Формула обратного ряда Фурье для восстановления \(x(t)\) использует непрерывное время \(t\). Использование циклов `for` для расчёта каждого момента времени \(t\) с шагом \(0.25T\) оказалось крайне медленным и громоздким.
*   **Решение:** Была применена векторизация вычислений. С помощью функции `np.outer` было сформировано внешнее произведение вектора времени и вектора частот, что позволило создать матрицу фазовых множителей. Это дало возможность вычислить экспоненты для всех моментов времени одной матричной операцией, что значительно ускорило работу кода и упростило его вид.
