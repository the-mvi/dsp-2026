## Основные теоретические положения

### 1. Линейные дискретные системы

Линейная дискретная система (ЛДС) описывается разностным уравнением, которое связывает текущий отсчёт реакции $y(n)$ с текущим и прошлыми отсчётами воздействия $x(n)$ и прошлыми отсчётами реакции. В общем виде:
$$y(n) = \sum_{i=0}^{N-1} b_i x(n-i) - \sum_{k=1}^{M-1} a_k y(n-k)$$

Для рекурсивного звена 2-го порядка разностное уравнение имеет вид:
$$y(n) = b_0 x(n) + b_1 x(n-1) + b_2 x(n-2) - a_1 y(n-1) - a_2 y(n-2)$$

Применяя Z-преобразование при нулевых начальных условиях, получаем передаточную функцию (ПФ) — отношение Z-изображений реакции и воздействия:
$$H(z) = \frac{Y(z)}{X(z)} = \frac{b_0 + b_1 z^{-1} + b_2 z^{-2}}{1 + a_1 z^{-1} + a_2 z^{-2}}$$

**Импульсная характеристика** $h(n)$ — это реакция системы на цифровой единичный импульс $u_0(n)$ при нулевых начальных условиях, $H(z) = Z\{h(n)\}$. Для рекурсивных (БИХ) систем она бесконечна.

**Свёртка** позволяет найти реакцию системы на произвольное воздействие:
$$y(n) = \sum_{m=0}^{n} h(m)\, x(n-m)$$
Длина линейной свёртки двух последовательностей длиной $N_1$ и $N_2$ равна $N_1 + N_2 - 1$.

**Нули и полюсы** — это корни числителя и знаменателя передаточной функции соответственно. Они могут быть представлены в алгебраической ($z = \xi + j\eta$) и показательной ($z = r e^{j\varphi}$) формах, где $r = \sqrt{\xi^2 + \eta^2}$, $\xi = r\cos\varphi$, $\eta = r\sin\varphi$.

**Устойчивость.** ЛДС устойчива, если при ограниченном воздействии реакция также ограничена, что равносильно условию $\sum_{n=0}^{\infty} |h(n)| < \infty$. Для этого необходимо и достаточно, чтобы все полюсы передаточной функции лежали строго внутри единичной окружности ($|z_*| < 1$).

**Частотные характеристики** (АЧХ и ФЧХ) определяются подстановкой $z = e^{j\hat{\omega}}$ в передаточную функцию:
$$H(e^{j\hat\omega}) = |H(e^{j\hat\omega})|\, e^{j \arg H(e^{j\hat\omega})}$$
АЧХ $A(\hat\omega) = |H(e^{j\hat\omega})|$ показывает изменение амплитуды на разных частотах, ФЧХ $\varphi(\hat\omega) = \arg H(e^{j\hat\omega})$ — фазовый сдвиг. Нормированная частота связана с частотой $f$ (Гц) соотношением $\hat\omega = \omega T = 2\pi f / f_d$, где $f_d$ — частота дискретизации. Частотные характеристики периодичны с периодом $2\pi$ ($f_d$), основная полоса частот: $\hat\omega \in [0;\ \pi]$, $f \in [0;\ f_d/2]$.

### 2. Дискретное преобразование Фурье (ДПФ)

**ДПФ** применяется для анализа частотного состава периодических последовательностей и последовательностей конечной длины $N$. Прямое ДПФ переводит сигнал из временной области в частотную:
$$X(k) = \sum_{n=0}^{N-1} x(n) e^{-j \frac{2\pi}{N} kn}, \quad k = 0, 1, \dots, N-1$$

**Обратное ДПФ (ОДПФ)** восстанавливает сигнал по его спектру:
$$x(n) = \frac{1}{N} \sum_{k=0}^{N-1} X(k) e^{j \frac{2\pi}{N} kn}, \quad n = 0, 1, \dots, N-1$$

**Шкалы времени и частоты.** Дискретное время — $nT$, дискретное нормированное время — $\hat t = nT/T = n$. Дискретная нормированная частота — номер отсчёта $k$; ему соответствуют частота $f_k = k \frac{f_d}{N}$ (Гц) и нормированная частота $\hat\omega_k = \frac{2\pi k}{N}$.

**Спектры периодической последовательности.** Периодическая последовательность с периодом $N$ представляется дискретным рядом Фурье с коэффициентами
$$c_k = \frac{1}{N} \sum_{n=0}^{N-1} x(n) e^{-j \frac{2\pi}{N} kn} = \frac{X(k)}{N}$$
* **Амплитудный спектр** $|c_k|$ показывает амплитуды гармоник;
* **Фазовый спектр** $\arg c_k = \arg X(k)$ показывает их начальные фазы.

Для вещественной последовательности $|X(k)| = |X(N-k)|$, $\arg X(k) = -\arg X(N-k)$. Модуль ДПФ конечной последовательности связан с амплитудным спектром её периодического продолжения соотношением $|X(k)| = N |c_k|$. Если на периоде $N$ укладывается целое число периодов гармоник, спектр не испытывает «растекания».

**Восстановление аналогового сигнала** по отсчётам ДПФ выполняется с помощью ряда Фурье. Согласно теореме Котельникова, если частота дискретизации больше удвоенной максимальной частоты сигнала, непрерывный сигнал может быть восстановлен по отсчётам без потерь.

**Равенство Парсеваля** отражает закон сохранения энергии: энергия сигнала во временной области равна энергии в частотной области:
$$\sum_{n=0}^{N-1} |x(n)|^2 = \frac{1}{N} \sum_{k=0}^{N-1} |X(k)|^2$$

### Пункт 1.1 Импульсная характеристика

```python
n_h = np.arange(N1)
delta = np.zeros(N1); delta[0] = 1.0

h = np.zeros(N1)
for n in range(N1):
    h[n] = b0 * delta[n]
    if n >= 1: h[n] += b1 * delta[n-1] - a1 * h[n-1]
    if n >= 2: h[n] += b2 * delta[n-2] - a2 * h[n-2]


plt.figure()
plt.stem(n_h, h, basefmt="k-")
plt.xlabel("n"); plt.ylabel("h(n)")
plt.title(f"Импульсная характеристика h(n), N1 = {N1}")
plt.grid(True)
plt.show()

print("Первые 10 отсчётов h(n):")
for i in range(10):
    print(f"h({i}) = {h[i]:.6f}")
```

```python
r = np.sqrt(a2)
phi = np.arccos(-a1 / (2 * r))
u1 = lambda m: (m >= 0).astype(float)      

h_an = (b0 * r**n_h * np.sin((n_h + 1) * phi)
        + b1 * r**(n_h - 1.0) * np.sin(n_h * phi) * u1(n_h - 1)
        + b2 * r**(n_h - 2.0) * np.sin((n_h - 1) * phi) * u1(n_h - 2)) / np.sin(phi)

print(f"r = sqrt(a2) = {r:.4f}, phi = arccos(-a1/(2r)) = {phi:.4f} рад = {phi/np.pi:.4f}·π")
print("Расхождение аналитической формулы с расчётом по разностному уравнению:", np.max(np.abs(h_an - h)))
```

Для БИХ-звена полная импульсная характеристика бесконечна, поэтому $N_1$ здесь задаёт длину вычисленного фрагмента. Аналитическая формула через полюса проверяется в следующей ячейке.

### Пункт 1.2 Реакция по формуле свёртки

```python
n_x = np.arange(N2)
x = np.zeros(N2)
x[:int(np.floor(N2 / 2 + 0.5))] = 1.0          

y1_full = np.convolve(h, x)                     
n_y1_full = np.arange(len(y1_full))

y1 = y1_full[:N2]                             
n_y1 = np.arange(N2)

fig, axes = plt.subplots(2, 1, figsize=(10, 6))
axes[0].stem(n_x, x, basefmt="k-")
axes[0].set_title("Воздействие x(n) — прямоугольный импульс")
axes[0].set_xlabel("n"); axes[0].set_ylabel("x(n)"); axes[0].grid(True)

axes[1].stem(n_y1_full, y1_full, basefmt="k-", label="полная свёртка")
axes[1].stem(n_y1, y1, basefmt="k-", linefmt="r-", markerfmt="ro", label="y1(n) = первые N2 отсчётов")
axes[1].set_title("Реакция y1(n) по формуле свёртки")
axes[1].set_xlabel("n"); axes[1].set_ylabel("y1(n)"); axes[1].legend(); axes[1].grid(True)
plt.tight_layout(); plt.show()

print(f"Длина импульса (число единичных отсчётов): {int(x.sum())}")
print(f"Длина x: {len(x)}")
print(f"Длина h: {len(h)}")
print(f"Длина полной свёртки (аналитически): {len(x) + len(h) - 1}")
print(f"Длина полной свёртки (по графику): {len(y1_full)}")
print(f"Последний ненулевой отсчёт полной свёртки: n = {np.nonzero(np.abs(y1_full) > 1e-12)[0][-1]}")
```

На вход подан конечный прямоугольный импульс длины $N_2$. Реакция рассчитана свёрткой с усечённым фрагментом $h[0],\ldots,h[N_1-1]$. Поэтому полная длина массива свёртки равна $N_1+N_2-1$, а ненулевая часть заканчивается раньше из-за конечной длины входного импульса.

### Пункт 1.3 Реакция по разностному уравнению

```python
y2 = np.zeros(N2)
for n in range(N2):
    y2[n] = b0 * x[n]
    if n >= 1: y2[n] += b1 * x[n-1] - a1 * y2[n-1]
    if n >= 2: y2[n] += b2 * x[n-2] - a2 * y2[n-2]
n_y2 = np.arange(N2)

print("Проверка с lfilter, max|разности| =", np.max(np.abs(y2 - lfilter(b, a, x))))

fig, axes = plt.subplots(2, 1, figsize=(10, 6))
axes[0].stem(n_x, x, basefmt="k-")
axes[0].set_title("Воздействие x(n)")
axes[0].set_xlabel("n"); axes[0].set_ylabel("x(n)"); axes[0].grid(True)

axes[1].stem(n_y2, y2, basefmt="k-")
axes[1].set_title("Реакция y2(n) по разностному уравнению")
axes[1].set_xlabel("n"); axes[1].set_ylabel("y2(n)"); axes[1].grid(True)
plt.tight_layout(); plt.show()

plt.figure()
plt.stem(n_y1, y1, basefmt="k-", linefmt="b-", markerfmt="bo", label="y1 (свёртка)")
plt.stem(n_y2, y2, basefmt="k-", linefmt="r--", markerfmt="rx", label="y2 (разностное ур.)")
plt.axvline(N1 - 0.5, color="gray", ls=":", label="n = N1")
plt.title("Сравнение y1(n) и y2(n)")
plt.xlabel("n"); plt.ylabel("y(n)"); plt.legend(); plt.grid(True)
plt.show()

d = np.abs(y1 - y2)
print(f"Максимальное расхождение |y1 - y2| на первых N2 отсчётах: {d.max():.3e}")

# Для честного сравнения на всём интервале входа пересчитываем h длиной N2.
h_check = np.zeros(N2)
for n in range(N2):
    h_check[n] = b0 * (1.0 if n == 0 else 0.0)
    if n >= 1: h_check[n] += b1 * (1.0 if n == 1 else 0.0) - a1 * h_check[n-1]
    if n >= 2: h_check[n] += b2 * (1.0 if n == 2 else 0.0) - a2 * h_check[n-2]
y_check = np.zeros(N2)
for n in range(N2):
    y_check[n] = sum(x[k] * h_check[n-k] for k in range(n+1))
print(f"Максимальное расхождение при h длиной N2: {np.max(np.abs(y_check - y2)):.3e}")
```

Теперь тот же вход рассчитан непосредственно по разностному уравнению. На первых $N_1$ отсчётах результат совпадает со свёрткой с машинной точностью. Хвост отличается, потому что в первой свёртке использован только фрагмент импульсной характеристики. Дополнительная проверка ниже строит $h$ длины $N_2$ и показывает совпадение с рекурсивным расчётом на всём интервале входа.

### Пункт 1.4 Нули и полюсы

```python
zeros, poles, k_gain = tf2zpk(b, a)

print("Нули (z0_i):")
for z in zeros:
    rz = np.abs(z); ph = np.angle(z)
    print(f"  z0 = {z.real:+.6f} {z.imag:+.6f}j | |z0| = {rz:.6f}, arg = {ph:+.6f} рад = {ph/np.pi:+.6f}·π")

print("\nПолюсы (z*_i):")
for p in poles:
    rp = np.abs(p); ph = np.angle(p)
    print(f"  z* = {p.real:+.6f} {p.imag:+.6f}j | |z*| = {rp:.6f}, arg = {ph:+.6f} рад = {ph/np.pi:+.6f}·π")

print(f"\nКоэффициент усиления k = {k_gain:.6f}")
print("Остаток числителя:", np.max(np.abs(k_gain * np.poly(zeros) - b)))
print("Остаток знаменателя:", np.max(np.abs(np.poly(poles) - a)))
```

На карте отмечены нули, полюсы и единичная окружность. Полюса лежат внутри неё, поэтому звено устойчиво. Дополнительно проверяю восстановление коэффициентов полиномов по найденным корням.

### Пункт 1.6 АЧХ и ФЧХ в нормированных частотах

```python
w_norm = np.linspace(0, np.pi, 2000)            
_, H = freqz(b, a, worN=w_norm)                 
A = np.abs(H)
Phi = np.unwrap(np.angle(H))

fig, axes = plt.subplots(2, 1, figsize=(10, 6))
axes[0].plot(w_norm/np.pi, A)
axes[0].axvline(np.angle(poles[0])/np.pi, color='r', ls=':', label="аргумент полюса")
axes[0].axvline(np.angle(zeros[0])/np.pi, color='b', ls=':', label="аргумент нуля")
axes[0].set_xlabel(r"$\hat{\omega}/\pi$"); axes[0].set_ylabel("|H|")
axes[0].set_title("АЧХ в нормированных частотах"); axes[0].legend(); axes[0].grid(True)

axes[1].plot(w_norm/np.pi, Phi/np.pi)
axes[1].set_xlabel(r"$\hat{\omega}/\pi$"); axes[1].set_ylabel(r"arg H / $\pi$")
axes[1].set_title("ФЧХ в нормированных частотах"); axes[1].grid(True)
plt.tight_layout(); plt.show()

i_max, i_min = np.argmax(A), np.argmin(A)
print(f"A(0)  = {A[0]:.4f}")
print(f"A(pi) = {A[-1]:.4f}")
print(f"Максимум АЧХ: A = {A[i_max]:.4f} при w = {w_norm[i_max]/np.pi:.4f}·π")
print(f"Минимум АЧХ:  A = {A[i_min]:.4f} при w = {w_norm[i_min]/np.pi:.4f}·π")
```

АЧХ и ФЧХ построены по подстановке $z=e^{j\hat\omega}$ для $0\le\hat\omega\le\pi$. Провожу на графике аргументы одного нуля и одного полюса: рядом с полюсом появляется максимум, а рядом с нулём — провал.

### Пункт 1.7 АЧХ и ФЧХ в Гц

```python
f_hz = w_norm * fd1 / (2*np.pi)   

fig, axes = plt.subplots(2, 1, figsize=(10, 6))
axes[0].plot(f_hz, A)
axes[0].set_xlabel("f, Гц"); axes[0].set_ylabel("|H|")
axes[0].set_title(f"АЧХ в Гц (f_д = {fd1:.0f} Гц)"); axes[0].grid(True)

axes[1].plot(f_hz, Phi/np.pi)
axes[1].set_xlabel("f, Гц"); axes[1].set_ylabel(r"arg H / $\pi$")
axes[1].set_title("ФЧХ в Гц"); axes[1].grid(True)
plt.tight_layout(); plt.show()

print(f"Основная полоса: [{f_hz[0]:.0f}; {f_hz[-1]:.0f}] Гц")
print(f"Максимум АЧХ при f = {f_hz[i_max]:.0f} Гц, минимум при f = {f_hz[i_min]:.0f} Гц")
```

Для практической шкалы частот нормированную частоту перевёл в герцы по формуле $f=\hat\omega f_d/(2\pi)$, используя частоту дискретизации первой части.

### Пункт 2.1 Периодическая последовательность

```python
n = np.arange(N)
t_abs = n * T               

x_per = A1*np.cos(2*np.pi*f1*n*T + np.pi/4) + A2*np.cos(2*np.pi*f2*n*T + np.pi/8)

fig, axes = plt.subplots(2, 1, figsize=(10, 6))
axes[0].stem(t_abs * 1e3, x_per, basefmt="k-")
axes[0].set_title("x(nT) в шкале дискретного времени nT")
axes[0].set_xlabel("nT, мс"); axes[0].set_ylabel("x(nT)"); axes[0].grid(True)

axes[1].stem(n, x_per, basefmt="k-")
axes[1].set_title("x(n) в шкале дискретного нормированного времени n = nT/T")
axes[1].set_xlabel("n"); axes[1].set_ylabel("x(n)"); axes[1].grid(True)
plt.tight_layout(); plt.show()

print(f"Номера гармоник: k1 = f1·N/fd = {f1*N/fd2:.0f}, k2 = f2·N/fd = {f2*N/fd2:.0f}")
```

Последовательность на периоде $N$ задана суммой двух косинусоид с фазами $\pi/4$ и $\pi/8$. Номера гармоник проверяются отдельно: на периоде должны укладываться целые числа периодов обеих составляющих.

### Пункт 2.2 ОДПФ от ДПФ

```python
X = np.fft.fft(x_per)         # ДПФ
x_rec = np.fft.ifft(X)        # ОДПФ

plt.figure()
plt.stem(n, x_rec.real, basefmt="k-")
plt.title("Последовательность после ОДПФ(ДПФ(x)) в шкале дискретного нормированного времени")
plt.xlabel("n"); plt.ylabel("x_rec(n)"); plt.grid(True)
plt.show()

print(f"Максимальная мнимая часть ОДПФ: {np.max(np.abs(x_rec.imag)):.3e}")
print(f"Максимальная ошибка восстановления: {np.max(np.abs(x_rec.real - x_per)):.3e}")
```

Сначала вычислил ДПФ, затем применил ОДПФ. Восстановленная последовательность совпадает с исходной; в выводе приведены максимальная мнимая часть и максимальная ошибка.

### Пункт 2.3 Амплитудный и фазовый спектры

```python
k = np.arange(N)              
f_k = k * fd2 / N             

c = X / N                                             
X_mag = np.abs(c)                                    
X_phase = np.where(np.abs(X) > 1e-6, np.angle(X), 0) 

fig, axes = plt.subplots(2, 2, figsize=(14, 8))

axes[0,0].stem(k, X_mag, basefmt="k-")
axes[0,0].set_title("Амплитудный спектр |X(k)|/N (дискр. норм. частота)")
axes[0,0].set_xlabel("k"); axes[0,0].grid(True)

axes[0,1].stem(f_k, X_mag, basefmt="k-")
axes[0,1].set_title("Амплитудный спектр |X(k)|/N (Гц)")
axes[0,1].set_xlabel("f, Гц"); axes[0,1].grid(True)

axes[1,0].stem(k, X_phase / np.pi, basefmt="k-")
axes[1,0].set_title("Фазовый спектр arg X(k) (дискр. норм. частота)")
axes[1,0].set_xlabel("k"); axes[1,0].set_ylabel(r"arg / $\pi$"); axes[1,0].grid(True)

axes[1,1].stem(f_k, X_phase / np.pi, basefmt="k-")
axes[1,1].set_title("Фазовый спектр arg X(k) (Гц)")
axes[1,1].set_xlabel("f, Гц"); axes[1,1].set_ylabel(r"arg / $\pi$"); axes[1,1].grid(True)
plt.tight_layout(); plt.show()

peaks = np.where(np.abs(X) > 1e-6)[0]
print("Ненулевые отсчёты спектра:")
for p in peaks:
    print(f"  k={p:2d}, f={f_k[p]:7.1f} Гц, |X|={np.abs(X[p]):.3f}, |X|/N={X_mag[p]:.3f}, arg={X_phase[p]:+.4f} рад = {X_phase[p]/np.pi:+.4f}·π")
```

Для полученного ДПФ построены амплитудный спектр $|X(k)|/N$ и фазовый спектр. Оба спектра показаны по номеру отсчёта и в герцах. Для вещественной последовательности положительные и отрицательные частоты образуют симметричные пары.

### Пункт 2.4 ДПФ конечной последовательности

```python
x_fin = x_per.copy()                    
X_fin = np.fft.fft(x_fin)

fig, axes = plt.subplots(2, 1, figsize=(10, 7))
axes[0].stem(k, np.abs(X_fin), basefmt="k-")
axes[0].set_title("Модуль ДПФ конечной последовательности |X(k)|")
axes[0].set_xlabel("k"); axes[0].grid(True)

axes[1].stem(k, X_mag, basefmt="k-", linefmt="r-", markerfmt="ro")
axes[1].set_title("Амплитудный спектр периодической последовательности |c_k| = |X(k)|/N")
axes[1].set_xlabel("k"); axes[1].grid(True)
plt.tight_layout(); plt.show()

print("Проверка |X(k)| = N·|c_k|, max|разности| =", np.max(np.abs(np.abs(X_fin) - N * X_mag)))
```

Здесь сравниваются ДПФ конечного массива и амплитудный спектр периодического продолжения. Для выбранных частот на периоде укладывается целое число колебаний, поэтому энергия сосредоточена в нескольких спектральных отсчётах без заметного растекания.

### Пункт 2.5 и 2.6 Восстановление аналогового сигнала

```python
half = N // 2
k_a = np.arange(-half, half)                
X_a = X[k_a % N]                             

# Время: t ∈ [0; (N−1)T] с шагом Δt = 0.25T
dt = 0.25 * T
t = np.arange(0, (N - 1) * T + dt/2, dt)

# Восстановленный сигнал по формуле x(t) = 1/N · Σ X_a(k) exp(j 2π k t / (NT))
x_rec_t = (np.exp(1j * 2 * np.pi * np.outer(t, k_a) / (N * T)) @ X_a) / N
print(f"Максимальная мнимая часть x(t): {np.max(np.abs(x_rec_t.imag)):.3e}")
x_rec_t = x_rec_t.real

# Исходный аналоговый сигнал x(t) = x(nT)|nT=t
x_orig_t = A1*np.cos(2*np.pi*f1*t + np.pi/4) + A2*np.cos(2*np.pi*f2*t + np.pi/8)

f_a = k_a / (N * T)                          

fig, axes = plt.subplots(2, 2, figsize=(14, 8))
axes[0,0].stem(n, x_per, basefmt="k-")
axes[0,0].set_title("Периодическая последовательность x(n)")
axes[0,0].set_xlabel("n"); axes[0,0].grid(True)

axes[0,1].stem(k, np.abs(X), basefmt="k-")
axes[0,1].set_title("Модуль ДПФ |X(k)|")
axes[0,1].set_xlabel("k"); axes[0,1].grid(True)

axes[1,0].plot(t * 1e3, x_rec_t, 'b-')
axes[1,0].plot(t_abs * 1e3, x_per, 'k.', ms=5)
axes[1,0].set_title("Восстановленный x(t) (точки — отсчёты x(nT))")
axes[1,0].set_xlabel("t, мс"); axes[1,0].grid(True)

axes[1,1].stem(f_a, np.abs(X_a) / N, basefmt="k-")
axes[1,1].set_title("Амплитудный спектр восстановленного сигнала |X_a(k)|/N")
axes[1,1].set_xlabel("f, Гц"); axes[1,1].grid(True)
plt.tight_layout(); plt.show()

plt.figure(figsize=(12, 4))
plt.plot(t * 1e3, x_orig_t, 'r-', lw=3, alpha=0.5, label="исходный x(t)")
plt.plot(t * 1e3, x_rec_t, 'b--', lw=1, label="восстановленный x(t)")
plt.title("Восстановленный и исходный аналоговые сигналы")
plt.xlabel("t, мс"); plt.legend(); plt.grid(True)
plt.show()

print(f"Максимальное расхождение восстановленного и исходного сигналов: {np.max(np.abs(x_rec_t - x_orig_t)):.3e}")
```

По отсчётам ДПФ восстановил аналоговый сигнал рядом Фурье на сетке с шагом $T/4$. На одном рисунке показаны отсчёты, модуль ДПФ, двусторонний спектр и сравнение исходного и восстановленного сигналов.

### Пункт 2.7 Равенство Парсеваля

```python
E_time = np.sum(np.abs(x_per)**2)
E_freq = (1.0/N) * np.sum(np.abs(X)**2)

print(f"Энергия во времени:  Σ|x(n)|²       = {E_time:.6f}")
print(f"Энергия в частоте:  (1/N)·Σ|X(k)|² = {E_freq:.6f}")
print(f"Относительное расхождение: {abs(E_time - E_freq)/E_time:.3e}")
print(f"Равенство Парсеваля выполняется: {np.isclose(E_time, E_freq)}")
E_analytic = N * (A1**2 + A2**2) / 2
print(f"Аналитическая энергия: {E_analytic:.6f}")
print(f"Разность с аналитической: {abs(E_time - E_analytic):.3e}")
print(f"Средняя мощность: {E_time / N:.6f}")
```

В конце проверил равенство Парсеваля двумя способами: сравнил энергии во временной и частотной областях и сопоставил их с аналитической энергией суммы двух косинусоид. Разность должна быть только из-за округления.
