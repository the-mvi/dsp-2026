## Основные теоретические положения

### Обозначения и нормирование частоты

Номер отсчёта $n\in\mathbb{Z}$ безразмерен.

Моменты наблюдения определяются соотношением $t_n=nT$, где $T$ обозначает период дискретизации, $T>0$.

Частота дискретизации — $f_{\mathrm{д}}=1/T$.

Физическая угловая частота — $\omega=2\pi f$.

Нормированная частота — $\hat f=f/f_{\mathrm{д}}=fT$ (циклов/отсчёт).

Нормированная угловая частота — $\hat\omega=\omega T=2\pi\hat f$ (рад/отсчёт).

Частота Найквиста — $f_N=f_{\mathrm{д}}/2$.

Основной интервал нормированных угловых частот — $-\pi\le\hat\omega<\pi$.

### Цифровой единичный импульс

$$
u_0(n)=\begin{cases}1,&n=0,\\0,&n\ne 0.\end{cases}
$$

### Импульсная характеристика

Импульсной характеристикой $h(nT)$ линейной дискретной системы (ЛДС) называется её реакция на цифровой единичный импульс $u_0(nT)$ при нулевых начальных условиях.

В расчётах воздействие и реакция до начала наблюдения равны нулю
$$
x((n-i)T)\big|_{n-i<0}=0,\quad i=1,2,\ldots,
$$
$$
y((n-k)T)\big|_{n-k<0}=0,\quad k=1,2,\ldots.
$$

### Свёртка

Для линейной системы, инвариантной во времени, реакция при нулевых начальных условиях определяется свёрткой
$$
y(n)=x(n)*h(n)=\sum_{m=0}^{\infty}h(m)x(n-m)=\sum_{m=0}^{\infty}h(n-m)x(m).
$$

Для двух конечных последовательностей длины $N_h$ и $N_x$ длина полной линейной свёртки равна
$$
N_y=N_h+N_x-1.
$$

Системы с бесконечной импульсной характеристикой называют БИХ-системами. При расчёте можно ограничиться конечным фрагментом $h(n)$, но сама импульсная характеристика остаётся бесконечной.

### Разностное уравнение

$$
y(n)=\sum_{i=0}^{N-1}b_i x(n-i)-\sum_{k=1}^{M-1}a_k y(n-k),
$$
где $x(n)$ — входное воздействие, $y(n)$ — реакция системы, $b_i$ и $a_k$ — коэффициенты разностного уравнения.

Линейная система называется рекурсивной, если хотя бы один из коэффициентов $a_k\neq 0$, $k=1,2,\ldots$.

Звеном второго порядка называют ЛДС, у которой в разностном уравнении запаздывания воздействия и реакции не превышают двух отсчётов
$$
y(n)=b_0 x(n)+b_1 x(n-1)+b_2 x(n-2)-a_1 y(n-1)-a_2 y(n-2).
$$

### Передаточная функция

$Z$-преобразование импульсной характеристики называется передаточной функцией (ПФ)
$$
H(z)=\mathcal{Z}\{h(n)\}=\sum_{n=0}^{\infty}h(n)z^{-n}.
$$

Передаточной функцией ЛДС также называют отношение $Z$-изображения реакции к $Z$-изображению воздействия при нулевых начальных условиях
$$
H(z)=\frac{Y(z)}{X(z)}.
$$

Общий вид
$$
H(z)=\frac{\sum_{i=0}^{N-1}b_i z^{-i}}{1+\sum_{k=1}^{M-1}a_k z^{-k}}.
$$

Для звена второго порядка
$$
H(z)=\frac{b_0+b_1 z^{-1}+b_2 z^{-2}}{1+a_1 z^{-1}+a_2 z^{-2}}.
$$

### Нули и полюса

Нули — значения $z=z_0$, при которых ПФ равна нулю.

Полюса — значения $z=z_k$, при которых знаменатель ПФ равен нулю после сокращения общих множителей числителя и знаменателя.

Для звена второго порядка нули и полюса определяются соответственно из уравнений
$$
b_0z^2+b_1z+b_2=0,\qquad z^2+a_1z+a_2=0.
$$

Алгебраическая форма — $z=\xi+j\eta$.

Показательная форма — $z=re^{j\varphi}$, где
$$
r=\sqrt{\xi^2+\eta^2},\qquad \varphi=\arg(z),\qquad -\pi<\varphi\le\pi.
$$

### Последовательность, заданная отношением синусов

Для пары комплексно-сопряжённых полюсов обозначим их модуль через $r_x>0$, а положительный аргумент через $0<\varphi_x<\pi$. Последовательность имеет вид

$$
x(n)=\begin{cases}r_x^n\dfrac{\sin((n+1)\varphi_x)}{\sin\varphi_x},&n\ge0,\\0,&n<0.\end{cases}
$$

Её $Z$-изображение
$$
X(z)=\frac{1}{1-2r_x\cos\varphi_x\,z^{-1}+r_x^2z^{-2}},\qquad |z|>r_x.
$$

Связь с коэффициентами знаменателя
$$
a_1=-2r_x\cos\varphi_x,\qquad a_2=r_x^2.
$$

При указанных условиях полюса звена второго порядка равны $r_x e^{\pm j\varphi_x}$.

Импульсная характеристика звена второго порядка (оригинал ПФ)
$$
h(n)=b_0 r_x^n\frac{\sin((n+1)\varphi_x)}{\sin\varphi_x}
+b_1 r_x^{n-1}\frac{\sin(n\varphi_x)}{\sin\varphi_x}
+b_2 r_x^{n-2}\frac{\sin((n-1)\varphi_x)}{\sin\varphi_x}.
$$

Формула записана для $n\ge0$. Второе слагаемое учитывается только при $n\ge1$, третье — только при $n\ge2$. До этих моментов соответствующие слагаемые принимаются равными нулю. При $n<0$ импульсная характеристика равна нулю.

### Устойчивость

Линейная система называется устойчивой, если при ограниченном входном сигнале выходной сигнал тоже ограничен.

При $h(n)=0$ для $n<0$ условие устойчивости ЛДС записывается в виде
$$
\sum_{n=0}^{\infty}|h(n)|<\infty.
$$

Если $h(n)=0$ при $n<0$, то система с рациональной передаточной функцией устойчива тогда и только тогда, когда все полюса её передаточной функции лежат строго внутри единичного круга $|z|<1$.

### Частотная характеристика

Для устойчивой системы частотная характеристика равна значению передаточной функции на единичной окружности
$$
H(e^{j\hat\omega})=\sum_m h(m)e^{-j\hat\omega m}=H(z)\big|_{z=e^{j\hat\omega}}.
$$

Частотную характеристику можно выразить через коэффициенты разностного уравнения
$$
H(e^{j\hat\omega})=\frac{\sum_{i=0}^{N-1}b_i e^{-j\hat\omega i}}{1+\sum_{k=1}^{M-1}a_k e^{-j\hat\omega k}}.
$$

Амплитудный спектр — $|X(e^{j\hat\omega})|$.

Фазовый спектр — $\arg X(e^{j\hat\omega})$.

АЧХ системы — $|H(e^{j\hat\omega})|$.

ФЧХ системы — $\arg H(e^{j\hat\omega})$.

### Гармонический сигнал

Дискретный гармонический сигнал (косинусоида)
$$
x(n)=A\cos(2\pi f n T)=A\cos(\omega n T).
$$
С начальной фазой $\varphi_0$
$$
x(n)=A\cos(\hat\omega n+\varphi_0),\qquad \hat\omega=\omega T.
$$

Дискретный комплексный гармонический сигнал
$$
x(n)=A\exp(j\omega n T).
$$

### Дискретное преобразование Фурье

Поворачивающий множитель
$$
W_N=e^{-j2\pi/N}.
$$

Прямое ДПФ
$$
X(k)=\sum_{n=0}^{N-1}x(n)W_N^{kn},\qquad k=0,\ldots,N-1.
$$

Обратное ДПФ
$$
x(n)=\frac1N\sum_{k=0}^{N-1}X(k)W_N^{-kn},\qquad n=0,\ldots,N-1.
$$

Частота $k$-го отсчёта спектра
$$
\Delta f=\frac{f_{\mathrm{д}}}{N},\qquad f_k=k\Delta f,\qquad \hat\omega_k=2\pi\frac{k}{N}.
$$

Отсчёты с индексами $k>N/2$ соответствуют отрицательным частотам $f_k-f_{\mathrm{д}}$. При чётном $N$ отсчёт $k=N/2$ соответствует частоте Найквиста. Частоты $f_{\mathrm{д}}/2$ и $-f_{\mathrm{д}}/2$ эквивалентны для дискретной последовательности.

Для вещественной косинусоиды
$$
x(n)=A\cos\!\left(2\pi k_0\frac{n}{N}+\varphi\right),\qquad 0<k_0<\frac{N}{2}
$$
при целом $k_0$ ДПФ имеет два ненулевых отсчёта
$$
X(k_0)=N\frac{A}{2}e^{j\varphi},\qquad X(N-k_0)=N\frac{A}{2}e^{-j\varphi}.
$$

Амплитуда косинусоиды определяется по отсчёту ДПФ на положительной частоте
$$
A=\frac{2|X(k_0)|}{N}.
$$
Двусторонний спектр содержит составляющие как на положительных, так и на отрицательных частотах. Каждой косинусоиде соответствуют две спектральные линии на частотах $+f_0$ и $-f_0$. Для каждой линии модуль коэффициента ряда Фурье равен $A/2$, что объясняет множитель $2$ в формуле амплитуды. Для постоянной составляющей и частоты Найквиста множитель $2$ не используется.

Для вещественной последовательности выполняется свойство сопряжённой симметрии спектра
$$
X((N-k)\bmod N)=\overline{X(k)}.
$$
Поэтому отрицательные частоты несут ту же информацию, что и положительные.

### Конечная запись и периодическое продолжение

Для конечной записи преобразование Фурье непрерывно по частоте
$$
X_L(e^{j\hat\omega})=\sum_{n=0}^{N-1}x(n)e^{-j\hat\omega n}.
$$
ДПФ задаёт его отсчёты при $\hat\omega_k=2\pi k/N$. При периодическом продолжении той же записи коэффициенты дискретного ряда Фурье равны $C_k=X(k)/N$. Следовательно, амплитуды линий двустороннего спектра периодической последовательности равны $|X(k)|/N$.

### Восстановление периодического аналогового сигнала

Для чётного $N$ отсчёты ДПФ записываются с учётом отрицательных частот
$$
X_c(k)=\begin{cases}X(N+k),&-N/2\le k<0,\\X(k),&0\le k<N/2.\end{cases}
$$
Периодический сигнал восстанавливается по формуле
$$
z(t)=\frac1N\sum_{k=-N/2}^{N/2-1}X_c(k)\exp\!\left(j\frac{2\pi k}{NT}t\right).
$$
В моменты $t=nT$ эта формула переходит в ОДПФ, поэтому $z(nT)=x(n)$. Для рассматриваемой суммы косинусоид восстановление точно и между отсчётами. Обе частоты строго ниже $f_{\mathrm{д}}/2$ и совпадают с частотами $k/(NT)$.

### Равенство Парсеваля

Энергия записи из $N$ отсчётов одинакова во временном и частотном представлениях
$$
\sum_{n=0}^{N-1}|x(n)|^2=\frac1N\sum_{k=0}^{N-1}|X(k)|^2.
$$

### Часть 1. Линейная дискретная система
#### 1. Параметры


```python
import numpy as np
import matplotlib.pyplot as plt

# --- константы части 1 ---
V = 16                                          # N_вар — номер варианта
b0 = 0.5 + 0.02 * V                             # b0 — коэфф. числителя ПФ
b1 = b0 * (-1) ** (V + 1) * (0.9822 + 0.0178 * V)  # b1 — коэфф. числителя ПФ
b2 = b0 * (0.8 + 0.2 * (V % 5))                 # b2 — коэфф. числителя ПФ
a0 = 1.0                                        # a0 — коэфф. знаменателя ПФ
a1 = (-1) ** V * (0.7778 + 0.025 * V)           # a1 — коэфф. знаменателя ПФ
a2 = 0.64 + 0.006 * V                           # a2 — коэфф. знаменателя ПФ
b = np.array([b0, b1, b2])                      # вектор коэффициентов числителя
a = np.array([a0, a1, a2])                      # вектор коэффициентов знаменателя
N1 = V % 10 + 20                                # N1 — длина импульсной характеристики
N2 = V % 10 + 30                                # N2 — длина воздействия
fd = 1000 * V                                   # fд — частота дискретизации, Гц
print('b =', b)
print('a =', a)
print('N1 =', N1, 'N2 =', N2, 'fd =', fd)


def stem(t, y, title, xlabel='n', ylabel='Амплитуда'):
    plt.figure(figsize=(9, 3))
    plt.stem(t, y)
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.grid()
    plt.tight_layout()
    plt.show()
```

#### Импульсная характеристика


```python
def filt(x):
    # собираем реакцию y[n] по разностному уравнению 2-го порядка
    # y[n] = b0 x[n] + b1 x[n-1] + b2 x[n-2] - a1 y[n-1] - a2 y[n-2]
    # начальные условия нулевые — x[n]=y[n]=0 при n<0
    y = np.zeros(len(x))
    for n in range(len(x)):
        yn = 0.0
        for i in range(3):
            if n >= i:
                yn += b[i] * x[n - i]   # вклад воздействия
        for i in (1, 2):
            if n >= i:
                yn -= a[i] * y[n - i]   # вклад прошлой реакции
        y[n] = yn
    return y


# --- импульсная характеристика длины N1 ---
delta = np.zeros(N1)              # массив под единичный импульс
delta[0] = 1                      # u0[n] — 1 при n=0
h = filt(delta)                   # численная ИХ

poles = np.roots(a)               # полюса ПФ (корни знаменателя)
r = abs(poles[0])                 # модуль полюса
phi = abs(np.angle(poles[0]))     # аргумент полюса


def g_sin(n):
    # g[n] = r^n * sin((n+1)phi)/sin(phi) при n>=0, иначе 0
    n = np.asarray(n)
    return np.where(n >= 0, r ** n * np.sin((n + 1) * phi) / np.sin(phi), 0.0)


ns = np.arange(N1)                # индексы отсчётов 0 … N1-1
h_an = sum(b[i] * g_sin(ns - i) for i in range(3))  # аналитическая ИХ
assert np.allclose(h, h_an, atol=1e-12)

stem(ns, h, 'Первые 26 отсчётов h[n]')
print('r =', r, 'phi/pi =', phi / np.pi)
```

По таблице $Z$-преобразований последовательности $g[n]$ соответствует изображение $1/(1+a_1z^{-1}+a_2z^{-2})$. Умножение изображения на $z^{-i}$ означает задержку последовательности на $i$ отсчётов. Поэтому импульсная характеристика имеет вид
$$g[n]=r^n\frac{\sin((n+1)\varphi)}{\sin\varphi}u_1[n],\quad h[n]=b_0g[n]+b_1g[n-1]+b_2g[n-2].$$
Здесь $u_1[n]=1$ при $n\ge0$ и $u_1[n]=0$ при $n<0$. Параметры определяются по коэффициентам знаменателя как $r=\sqrt{a_2}$ и $\varphi=\arccos(-a_1/(2r))$. Последовательности $g[n-1]$ и $g[n-2]$ равны нулю при $n<1$ и $n<2$ соответственно. Начальные условия нулевые, поэтому $h[-1]=h[-2]=0$.

Импульсная характеристика бесконечна, поскольку полюса передаточной функции не сокращаются нулями. В расчёте взяты только первые 26 отсчётов. На графике видны затухающие колебания. Их затухание определяется модулем полюсов $r\approx0{,}858<1$, а частота колебаний определяется аргументом $\varphi\approx0{,}741\pi$.

#### 2. Воздействие и реакции (свёртка и рекурсия)


```python
# --- п.2 — прямоугольный импульс и две реакции ---
x = (np.arange(N2) < int(N2 / 2)).astype(float)  # воздействие — 1 при 0 <= n < int(N2/2)
y_conv = np.convolve(x, h)   # реакция свёрткой с усечённой ИХ
y_rec = filt(x)              # реакция по разностному уравнению

stem(np.arange(N2), x, 'Воздействие — 36 отсчётов, 18 единиц')
stem(np.arange(len(y_conv)), y_conv, 'Полная свёртка с усечённой характеристикой')

plt.figure(figsize=(9, 3))
plt.plot(y_conv[:N2], 'o-', label='Свёртка, первые 36')
plt.plot(y_rec, '.--', label='Рекурсия')
plt.legend()
plt.xlabel('n')
plt.ylabel('y[n]')
plt.grid()
plt.show()

# проверка совпадения реакций при 36 отсчётах ИХ
h_long = filt(np.r_[1.0, np.zeros(36 - 1)])  # 36 отсчётов ИХ
y_full = np.convolve(x, h_long)[:N2]      # свёртка с ИХ из 36 отсчётов, первые N2 значений
assert np.allclose(y_full, y_rec, atol=1e-12)
assert np.allclose(y_conv[:N1], y_rec[:N1], atol=1e-12)  # совпадение до усечения h

print('Длина массива свёртки:', len(y_conv))
print('Максимальное расхождение при 26 отсчётах ИХ:', np.max(np.abs(y_conv[:N2] - y_rec)))
print('Максимальное расхождение при 36 отсчётах ИХ:', np.max(np.abs(y_full - y_rec)))
```

Для свёртки используются $N_1=26$ отсчётов импульсной характеристики, поэтому результат содержит $N_1+N_2-1=61$ отсчёт. Ненулевая часть занимает $18+26-1=43$ отсчёта, от $n=0$ до $n=42$. Последние 18 отсчётов равны нулю.

По разностному уравнению вычислены первые 36 отсчётов реакции. На этом интервале сравниваются оба способа расчёта. Сама реакция БИХ-системы продолжается и после $n=35$.

Результаты свёртки $y_1[n]$ и расчёта по разностному уравнению $y_2[n]$ совпадают при $0\le n\le25$. Максимальное расхождение на интервале сравнения составляет примерно $0{,}05555$. При использовании 36 отсчётов импульсной характеристики результаты совпадают в пределах машинной точности.

При включении и выключении прямоугольного импульса возникают затухающие колебания. После окончания импульса выходной сигнал продолжает изменяться, поскольку зависит от своих предыдущих значений.

#### 3. Нули, полюса и карта


```python
# --- нули и полюса ---
zeros = np.roots(b)               # нули ПФ (корни числителя)
for name, roots in [('Нули', zeros), ('Полюса', poles)]:
    for z in roots:
        # алгебраическая и показательная формы
        print(f'{name}: {z:.9f}, {abs(z):.9f} exp(j*pi*{np.angle(z) / np.pi:.9f})')

assert np.all(np.abs(poles) < 1)  # критерий устойчивости

# карта нулей/полюсов и единичная окружность
theta = np.linspace(0, 2 * np.pi, 500)  # угол для единичной окружности
plt.figure(figsize=(5, 5))
plt.plot(np.cos(theta), np.sin(theta), '--', color='gray')  # просто рисуем кружок
plt.scatter(zeros.real, zeros.imag, facecolors='none', edgecolors='blue', label='Нули')
plt.scatter(poles.real, poles.imag, marker='x', label='Полюса')
plt.axis('equal')
plt.grid()
plt.legend()
plt.xlabel('Re z')
plt.ylabel('Im z')
plt.show()
```

Нули и полюса образуют комплексно-сопряжённые пары. В каждой паре модули равны, а аргументы противоположны по знаку. По формуле Эйлера $z=re^{j\varphi}=r\cos\varphi+jr\sin\varphi$. Поэтому $\operatorname{Re}z=r\cos\varphi$ и $\operatorname{Im}z=r\sin\varphi$. Обе формы записи задают одни и те же точки на карте нулей и полюсов.

Передаточную функцию можно записать через её нули и полюса
$$H(z)=0{,}82\frac{(1-z_1z^{-1})(1-z_2z^{-1})}{(1-p_1z^{-1})(1-p_2z^{-1})},$$
где $z_{1,2}=e^{\pm j\arccos(1{,}03894/1{,}64)}$, $p_{1,2}=\sqrt{0{,}736}e^{\pm j\arccos(-1{,}1778/(2\sqrt{0{,}736}))}$. Оба полюса лежат внутри единичного круга, поэтому система устойчива.

#### 4. АЧХ и ФЧХ


```python
# --- частотные характеристики ---
omega_hat = np.linspace(-np.pi, np.pi, 2005)  # сетка нормированных частот ω̂
num = sum(b[i] * np.exp(-1j * omega_hat * i) for i in range(3))  # числитель H
den = sum(a[i] * np.exp(-1j * omega_hat * i) for i in range(3))  # знаменатель H
H = num / den  # H(e^{jω̂})

# две шкалы — нормированная ω̂/π и физическая f, Гц
scales = [
    (omega_hat / np.pi, r'$\hat\omega/\pi$'),
    (omega_hat * fd / (2 * np.pi), 'f, Гц'),
]
for axis, label in scales:
    fig, ax = plt.subplots(2, 1, figsize=(9, 5), sharex=True)
    ax[0].plot(axis, np.abs(H))
    ax[0].set_ylabel('|H|')
    phase = np.angle(H)                 # ФЧХ
    phase[np.abs(H) < 1e-10] = np.nan  # фаза в нулях АЧХ не определена
    ax[1].plot(axis, phase)
    ax[1].set_ylabel('arg H, рад')
    ax[1].set_xlabel(label)
    for q in ax:
        q.grid()
    plt.tight_layout()
    plt.show()

print('Частоты нулей, Гц:', np.angle(zeros) * fd / (2 * np.pi))
# |H(0)| — АЧХ на нулевой частоте (середина сетки ω̂), |H(π)| — на частоте Найквиста (конец сетки)
print('|H(0)|, |H(pi)|:', np.abs(H[len(omega_hat) // 2]), np.abs(H[-1]))
```

Нули передаточной функции на единичной окружности задают частоты, на которых АЧХ равна нулю. При переходе через эти частоты фаза изменяется скачком на $\pi$. В самих точках нулей фаза не определена. Аргументы полюсов равны примерно $\pm0{,}741\pi$. Вблизи этих частот модуль знаменателя частотной характеристики мал, поэтому АЧХ возрастает. Низкочастотные составляющие ослабляются сильнее высокочастотных.

На графике фаза $H$ показана в интервале $(-\pi,\pi]$. Значения фазы, различающиеся на $2\pi$, описывают одно и то же комплексное число. Поэтому при переходе через границу интервала на графике появляется скачок на $2\pi$.

Частотная характеристика повторяется с периодом $2\pi$ по нормированной угловой частоте и с периодом $f_d$ по частоте в герцах. Достаточно рассмотреть интервал $-\pi\le\hat\omega<\pi$, которому соответствует $-8000\le f<8000$ Гц. Шкалы связаны формулами $f=\hat\omega f_d/(2\pi)$ и $\hat f=f/f_d$. На горизонтальной оси первого графика отложено $\hat\omega/\pi$. Границы $-1$ и $1$ соответствуют частотам $-8000$ и $8000$ Гц. Значения характеристики на этих границах совпадают из-за её периодичности.

АЧХ равна нулю при $f\approx\pm2253$ Гц, то есть при $\hat\omega\approx\pm0{,}282\pi$. На нулевой частоте её значение составляет примерно $0{,}206$, а при $\hat\omega=\pi$ оно равно $4{,}799$. Значение АЧХ больше единицы означает усиление гармоники на этой частоте.

### Часть 2. ДПФ
#### Параметры


```python
# --- константы части 2 ---
N = 64                                          # N — период (длина) последовательности
fd2 = 2000 * (V % 5 + 1)                        # fд — частота дискретизации, Гц
T = 1 / fd2                                     # T — период дискретизации, с
A1 = 1 + 0.01 * V                               # A1 — амплитуда 1-й дискретной гармоники
A2 = 2 * A1                                     # A2 — амплитуда 2-й дискретной гармоники
f1 = fd2 / 8                                    # f1 — частота 1-й дискретной гармоники, Гц
f2 = 1.5 * f1                                   # f2 — частота 2-й дискретной гармоники, Гц
print('N, fd, T, A1, A2, f1, f2:', N, fd2, T, A1, A2, f1, f2)


def x_cont(t):
    # сумма двух косинусоид
    return (
        A1 * np.cos(2 * np.pi * f1 * t + np.pi / 4)
        + A2 * np.cos(2 * np.pi * f2 * t + np.pi / 8)
    )
```

#### 5. Период, ДПФ и ОДПФ


```python
# --- п.5 — сигнал, ДПФ, ОДПФ ---
n = np.arange(N)              # индексы отсчётов периода
x = x_cont(n * T)             # один период дискретной последовательности
X = np.fft.fft(x)             # прямое ДПФ — x[n] -> X[k]
x_back = np.fft.ifft(X)       # обратное ДПФ — X[k] -> x[n]

stem(n, x, 'Период последовательности, нормированное время')
stem(n * T, x, 'Период последовательности, физическое время', 't, с')
stem(n, x_back.real, 'ОДПФ — восстановленные отсчёты')
assert np.max(np.abs(x_back - x)) < 1e-10  # ОДПФ(ДПФ(x)) = x
```

ОДПФ восстанавливает исходные отсчёты. Значения $x_{\mathrm{back}}[n]$ и $x[n]$ совпадают в пределах машинной точности.

Наименьшие периоды первой и второй гармоник равны 8 и 16 отсчётам. Наименьший период их суммы равен 16 отсчётам. На этом интервале укладываются два периода первой гармоники и три периода второй. Для ДПФ взяты $N=64$ отсчёта, поэтому на графике показаны четыре наименьших периода сигнала. Число 64 также является периодом последовательности.

#### 6. Амплитудный и фазовый спектры


```python
# --- п.6 — амплитудный и фазовый спектры ---
k = np.arange(-N // 2, N // 2)           # номера гармоник от -N/2 до N/2-1 (отрицательные слева)
X_shift = np.fft.fftshift(X)             # тот же спектр X, но с отрицательными частотами слева (= X_c)
amp = np.abs(X_shift) / N                # двусторонний амплитудный спектр
alive = amp > 1e-10                      # маска ненулевых гармоник

for axis, label in [(k / N, 'f/fd'), (k * fd2 / N, 'f, Гц')]:
    stem(axis, amp, 'Двусторонний амплитудный спектр', label)
    stem(axis[alive], np.angle(X_shift[alive]), 'Фазы ненулевых линий', label, 'arg X, рад')

# сравнение |X| и |X|/N
plt.figure(figsize=(9, 3))
plt.stem(k / N, np.abs(X_shift), label='|X|')
plt.stem(k / N, amp, linefmt='C1-', markerfmt='C1o', label='|X|/N')
plt.legend()
plt.xlabel('f/fd')
plt.ylabel('Модуль / амплитуда')
plt.grid()
plt.show()

bins = np.flatnonzero(np.abs(X) > 1e-10)  # индексы ненулевых отсчётов ДПФ
print('Отсчёты k:', bins)
print('Модули:', np.abs(X)[bins])
assert np.array_equal(bins, [8, 12, 52, 56])
assert np.allclose(np.abs(X[[8, 12]]), N * np.array([A1, A2]) / 2)
assert np.allclose(np.angle(X[[8, 12, 52, 56]]), [np.pi / 4, np.pi / 8, -np.pi / 8, -np.pi / 4])
```

Нормированное и физическое время связаны соотношением $n=t/T$. На графиках двустороннего спектра номера гармоник идут от $-N/2$ до $N/2-1$. Для них $f_k=kf_d/N$, $\hat f_k=k/N=f_k/f_d$ и $\hat\omega_k=2\pi k/N$.

В исходном порядке отсчётов ДПФ индексы идут от $0$ до $N-1$. Индексам $k>N/2$ соответствуют отрицательные частоты $f_k=(k-N)f_d/N$. Частоты и фазы ненулевых отсчётов приведены в таблице.

| Индекс $k$ | Частота, Гц | Фаза |
| --- | --- | --- |
| 8 | 500 | $\pi/4$ |
| 12 | 750 | $\pi/8$ |
| 52 | $-750$ | $-\pi/8$ |
| 56 | $-500$ | $-\pi/4$ |

При нулевой амплитуде фаза не определена, поэтому такие отсчёты не показаны на графике фазового спектра.

ДПФ конечной записи длины $N$ и ДПФ одного периода её периодического продолжения дают одинаковые значения. Двусторонний амплитудный спектр получается делением модуля ДПФ на $N$, то есть $|C_k|=|X(k)|/64$. На частотах $\pm500$ Гц его значения равны $0{,}58$, а на частотах $\pm750$ Гц они равны $1{,}16$.

Для этого сигнала односторонний спектр содержит линии на частотах $500$ и $750$ Гц. Их амплитуды вычисляются по формуле $2|X(k)|/N$ и равны $1{,}16$ и $2{,}32$. Эти значения совпадают с амплитудами исходных косинусоид.

Шаг частотной сетки равен $\Delta f=f_d/N=62{,}5$ Гц. На интервале $NT$ укладываются 8 периодов первой гармоники и 12 периодов второй. Частоты обеих гармоник совпадают с частотами отсчётов ДПФ, поэтому спектральной утечки нет. Остальные отсчёты ДПФ равны нулю в пределах машинной точности.

Сигнал вещественный, поэтому его амплитудный спектр симметричен относительно нулевой частоты. Фазы на частотах $+f$ и $-f$ равны по модулю и противоположны по знаку.

#### 7. Восстановление аналогового сигнала


```python
# --- п.7 — восстановление непрерывного сигнала ---
# шаг T/4. X_shift соответствует X_c[k] из формулы
t = np.arange(4 * (N - 1) + 1) * T / 4   # сетка времени 0 … 63T с шагом T/4
rest = (X_shift[:, None] * np.exp(1j * 2 * np.pi * k[:, None] * t[None, :] / (N * T))).sum(axis=0) / N  # z(t)

z_n = rest[::4].real                    # z(nT) в моменты дискретизации
diff_n = np.abs(z_n - x)                 # |z(nT) - x[n]|

assert np.max(np.abs(rest.imag)) < 1e-10
assert np.max(np.abs(rest.real - x_cont(t))) < 1e-10
assert np.max(diff_n) < 1e-10

# графики x[n] и модуля ДПФ
stem(n, x, 'Период x[n]')
stem(k / N, np.abs(X_shift), 'Модуль ДПФ |X|', 'f/fd')

fig, ax = plt.subplots(3, 1, figsize=(10, 8))
ax[0].plot(t, rest.real, label='Восстановленный z(t)')
ax[0].plot(t, x_cont(t), '--', label='Исходный')
ax[0].scatter(n * T, x, s=12, label='Отсчёты x[n]')
ax[0].legend()
ax[0].set_xlabel('t, с')
ax[0].set_ylabel('x(t), z(t)')
ax[0].grid()
ax[1].stem(k * fd2 / N, amp)
ax[1].set_xlabel('f, Гц')
ax[1].set_ylabel('Амплитуды аналогового сигнала')
ax[1].grid()
ax[2].stem(n, diff_n)
ax[2].set_xlabel('n')
ax[2].set_ylabel('|z(nT) - x[n]|')
ax[2].set_title('Разность в моменты дискретизации')
ax[2].grid()
plt.tight_layout()
plt.show()
```

Сигнал восстановлен на интервале $0\le t\le63T$ с шагом $T/4$. Восстановленный сигнал $z(t)$ совпадает с исходным аналоговым сигналом в пределах машинной точности. В моменты $t=nT$ значения $z(nT)$ совпадают с отсчётами $x[n]$, поэтому график разности $|z(nT)-x[n]|$ близок к нулю.

Двусторонний спектр восстановленного сигнала содержит линии на частотах $\pm500$ и $\pm750$ Гц. Их амплитуды равны $|X(k)|/N$. На частотах $\pm500$ Гц получено значение $0{,}58$, а на частотах $\pm750$ Гц получено $1{,}16$. Таким образом, амплитудный спектр восстановленного сигнала получается делением модуля ДПФ на $N$.

Обе частоты косинусоид ниже частоты Найквиста $f_d/2=2000$ Гц и совпадают с частотами отсчётов ДПФ. Поэтому при восстановлении сохраняются частоты, амплитуды и начальные фазы. Максимальная абсолютная погрешность на выбранной сетке времени составляет примерно $3{,}22\cdot10^{-14}$.

#### 8. Равенство Парсеваля


```python
# --- п.8 — равенство Парсеваля ---
E_t = np.sum(np.abs(x) ** 2)           # энергия во времени
E_f = np.sum(np.abs(X) ** 2) / N       # энергия по спектру
print('Энергия во времени:', E_t)
print('Энергия по спектру:', E_f)
print('max |z(nT)-x[n]|:', np.max(diff_n))
print('Ошибка восстановления:', np.max(np.abs(rest.real - x_cont(t))))
assert np.isclose(E_t, E_f, atol=1e-10)
```

Равенство Парсеваля выполняется. Расчёт по отсчётам сигнала и расчёт по ДПФ с нормировкой на $N$ дают одинаковую энергию в пределах машинной точности. Это энергия последовательности из $N=64$ отсчётов, которая содержит четыре наименьших периода сигнала.

Получено значение $E\approx215{,}296$. Его также можно найти по амплитудам косинусоид с помощью формулы $E=N(A_1^2+A_2^2)/2=215{,}296$.
