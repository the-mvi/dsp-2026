## Основные теоретические положения

_def:_ **Нулевые начальные условия** - отсутствие реакции при отсутствии воздействия:

$$
\begin{cases}
\left. x((n-i)T) \right|_{n-i<0} = 0, \quad i=1,2,\ldots \\
\left. y((n-k)T) \right|_{n-k<0} = 0, \quad k=1,2,\ldots
\end{cases}
$$

_def:_ **Импульсной характеристикой (ИХ) $h(nT)$** линейной системы называется её реакция на цифровой единичный импульс $U_0(nT)$ при нулевых начальных условиях:

$$
U_0(n)=\begin{cases}1, & n=0,\\0, & n\ne 0.\end{cases}
$$

_def:_ **Передаточной функцией (ПФ) $H(z)$ ЛДС** называется отношение $Z$-изображения реакции к $Z$-изображению воздействия при нулевых начальных условиях:

$$
H(z)=\frac{Y(z)}{X(z)}=\frac{\mathcal{Z}\{h(n)\}}{\mathcal{Z}\{U_0(n)\}}=\mathcal{Z}\{h(n)\},\qquad \mathcal{Z}\{U_0(n)\}=1.
$$


_def:_ **Свертка**:

$$
y(n) = x(n) \ast h(n)
$$

_def:_ **Сверточная функция**:

$$
y(n)=\sum_{m=0}^{n}h(n-m)x(m) = \sum_{m=0}^{n}h(m)x(n-m),\qquad n\geq 0.
$$

_def:_ **Разностное уравнение** при $a_0=1$:

$$
y(n)=\sum_{i=0}^{N-1}b_i x(n-i)-\sum_{k=1}^{M-1}a_k y(n-k).
$$

_def:_ Линейная система называется **нерекурсивной**, если $a_k = 0$ для любых $k = 1, 2, ..., M-1$

_def:_ Линейная система называется **рекурсивной**, если хотя бы один из коэффициентов $a_k \neq 0$, $k = 1, 2, ..., M-1$

_def:_ Линейная система называется **устойчивой**, если при ограниченном входном сигнале $x(n)$

$$
\underset{n}{\max} |x(n)| <= C_x, C_x \in R_{>0}
$$

выходной сигнал $y(n)$ также будет ограничен

$$
\underset{n}{\max} |y(n)| <= C_y, C_y \in R_{>0}
$$

_Th:_ Для того, чтобы линейная система с импульсной характеристикой вида

$$
h(n) = A_0 U_0(n) + \sum_{k=1}^{M-1} A_k \alpha_k^n, \quad A_k \neq 0, \quad k = 0, \dots, M-1
$$

была устойчива, необходимо и достаточно, чтобы выполнялось условие абсолютной сходимости ряда

$$
\sum_{n=0}^{\infty} |h(n)| < \infty
$$

_def:_ **ПФ общего вида:**

$$
H(z)=\frac{\sum_{i=0}^{N-1}b_i z^{-i}}{1+\sum_{k=1}^{M-1}a_k z^{-k}}
=\frac{\sum_{i=0}^{N-1}b_i z^{L-i}}{z^L+\sum_{k=1}^{M-1}a_k z^{L-k}},\qquad L=\max\{M-1,N-1\}.
$$

**Порядок передаточной функции:** $L$, если общие множители числителя и знаменателя не сокращаются.

_def:_ **Нули** — значения $z = z_0$, при которых ПФ равна 0:

$$
\sum_{i=0}^{N-1}b_i z^{L-i}=0.
$$

_def:_ **Полюса** — значения $z = z_0$, при которых знаменатель ПФ равен 0:

$$
z^L+\sum_{k=1}^{M-1}a_k z^{L-k}=0.
$$

_Th:_ Для того, чтобы ЛДС была устойчива, необходимо и достаточно, чтобы все полюса её передаточной функции:

$$
H(z) = A_0 + \sum_{k=1}^{M-1} \frac{A_k}{1 - \alpha_k z^{-1}}
$$

находились внутри круга единичного радиуса комплексной $z$-плоскости.

**ИХ рекурсивного звена 2-го порядка.** Для

$$
H(z)=\frac{b_0+b_1z^{-1}+b_2z^{-2}}{1+a_1z^{-1}+a_2z^{-2}}
$$

при $x(n)=U_0(n)$ и $h(-1)=h(-2)=0$:

$$
\begin{aligned}
h(0)&=b_0,\\
h(1)&=b_1-a_1h(0),\\
h(2)&=b_2-a_1h(1)-a_2h(0),\\
h(n)&=-a_1h(n-1)-a_2h(n-2),\qquad n\geq 3.
\end{aligned}
$$


_def:_ **Свёртка конечных последовательностей.** При дополнении последовательностей нулями за пределами заданных массивов длина полной линейной свёртки равна

$$
N_y=N_x+N_h-1.
$$


_def:_ **Дискретное преобразование Фурье (ДПФ):**
Периодической последовательности $x_p(n)$ определяет пара взаимно однозначных преобразований дискретных рядов Фурье для последовательностей во временной и частотной областях:

*   прямое преобразование:

$$
X_p(k) = \sum_{n=0}^{N-1} x_p(n) \exp\left(-j \frac{2\pi}{N} nk\right), \qquad k = 0, 1, \dots, (N-1).
$$

*   обратное преобразование (ОДПФ):

$$
x_p(n) = \frac{1}{N} \sum_{k=0}^{N-1} X_p(k) \exp\left(j \frac{2\pi}{N} nk\right), \qquad n = 0, 1, \dots, (N-1).
$$


_def:_ **Непрерывный спектр $X(e^{j\omega T})$ конечного сигнала** полностью определяется последовательностью своих отсчетов в частотной области, взятых с интервалом $\Delta\omega$, где:

$$
\Delta\omega \leq \frac{2\pi}{T_c},
$$

где $T_c$ — период сигнала.

_def:_ **Дискретным преобразованием Фурье (ДПФ)** наз. пара взаимно однозначных преобразований:

*   прямое преобразование:

$$
X(k) = \sum_{n=0}^{N-1} x(n) W_N^{nk}, \qquad k = 0, 1, \dots, (N-1).
$$

*   обратное преобразование (ОДПФ):

$$
x(n) = \frac{1}{N} \sum_{k=0}^{N-1} X(k) W_N^{-nk}, \qquad n = 0, 1, \dots, (N-1).
$$

#### 1. Импульсная характеристика

Передаточная функция:
$$
H(z) = \frac{0.62 - 0.67518z^{-1} + 0.62z^{-2}}{1 + 0.9278z^{-1} + 0.676z^{-2}}
$$

На вход подается единичный импульс и вычисляются первые 26 отсчётов ИХ: $0 \leq n \leq 25$.

```python
import numpy as np
import matplotlib.pyplot as plt

N_var = 6
b0 = 0.5 + 0.02 * N_var
b = np.array([
    b0,
    b0 * (-1) ** (N_var + 1) * (0.9822 + 0.0178 * N_var),
    b0 * (0.8 + 0.2 * (N_var % 5)),
])
a = np.array([1.0, (-1) ** N_var * (0.7778 + 0.025 * N_var),
              0.64 + 0.006 * N_var])
N1 = N_var % 10 + 20

n = np.arange(N1)
impulse = np.zeros(N1)
impulse[0] = 1.0

h = np.zeros(N1)

for i in range(N1):
    h[i] = b[0] * impulse[i]
    if i >= 1:
        h[i] += b[1] * impulse[i - 1] - a[1] * h[i - 1]
    if i >= 2:
        h[i] += b[2] * impulse[i - 2] - a[2] * h[i - 2]

fig, ax = plt.subplots(figsize=(8, 3))
markers, stems, baseline = ax.stem(n, h, linefmt='C0-', markerfmt='C0o', basefmt='k-')
plt.setp(markers, markersize=4)
plt.setp(stems, linewidth=1)
plt.setp(baseline, linewidth=0.7)
ax.set(xlabel='$n$', ylabel='$h(n)$',
       title=f'Импульсная характеристика: первые {N1} отсчётов',
       xticks=np.arange(0, N1, 5))
ax.grid(axis='y', alpha=0.25)
fig.tight_layout()
plt.show()
```

**Расчёт импульсной характеристики.** ИХ — это реакция системы на единичный импульс $x(n)=\delta(n)$ при нулевых начальных условиях: $h(n)=0$ при $n<0$. Здесь $\delta(0)=1$, а при $n\ne0$ вход равен нулю. Из заданной передаточной функции получаем разностное уравнение:

$$
h(n)=0.62\delta(n)-0.67518\delta(n-1)+0.62\delta(n-2)
-0.9278h(n-1)-0.676h(n-2).
$$

Поэтому первые отсчёты равны

$$
\begin{aligned}
h(0)&=0.62,\\
h(1)&=-0.67518-0.9278\cdot0.62=-1.250416,\\
h(2)&=0.62-0.9278h(1)-0.676h(0)=1.3610159648.
\end{aligned}
$$

Начиная с $n=3$, входной импульс и его задержанные копии равны нулю, но система продолжает реагировать за счёт предыдущих значений выхода:

$$
h(n)=-0.9278h(n-1)-0.676h(n-2),\qquad n\geq3.
$$

**Аналитическая формула рекурсивного звена 2-го порядка.** Корни характеристического уравнения $p^2+0.9278p+0.676=0$ имеют вид $p_{1,2}=re^{\pm j\theta}$, где

$$
r=\sqrt{0.676}\approx0.822192,\qquad
\theta=\arccos\left(-\frac{0.9278}{2r}\right)\approx2.170289.
$$

Для обратной величины знаменателя передаточной функции обратное $Z$-преобразование даёт

$$
g(n)=\begin{cases}
\displaystyle r^n\frac{\sin((n+1)\theta)}{\sin\theta},&n\geq0,\\
0,&n<0.
\end{cases}
$$

Умножение на $z^{-1}$ и $z^{-2}$ соответствует задержкам на один и два отсчёта. Таким образом, аналитическая формула ИХ с учётом числителя имеет вид

$$
h(n)=0.62g(n)-0.67518g(n-1)+0.62g(n-2)
$$

Условие $g(n)=0$ при $n<0$ учитывает причинность и нулевые начальные условия. В явном виде для $n\geq2$:

$$
h(n)=\frac{0.62r^n\sin((n+1)\theta)
-0.67518r^{n-1}\sin(n\theta)
+0.62r^{n-2}\sin((n-1)\theta)}{\sin\theta},
$$

а значения $h(0)$ и $h(1)$ приведены выше.

**График и действительная длина ИХ.** В программе рассчитаны и показаны первые $N_1=26$ отсчётов, то есть $n=0,\ldots,25$. График представляет затухающие колебания: комплексно-сопряжённые корни задают колебательный характер, а их модуль $r<1$ — затухание.

Фактическая длина ИХ данного рекурсивного звена **бесконечна**. Если бы её хвост стал тождественно нулевым, то из рекурсии при $a_2=0.676\ne0$ два последовательных нуля потребовали бы нулевых предыдущих отсчётов, что противоречит найденным $h(1)$ и $h(2)$. Поэтому $h(n)$ стремится к нулю при $n\to\infty$, но не обрывается после конечного числа отсчётов. Значение $N_1=26$ определяет лишь длину рассчитанного и изображённого фрагмента, а не полную длину ИХ.

#### 2. Реакция по формуле свёртки

Используем 26 отсчётов ИХ из п. 1. Полная линейная свёртка вычисляется по формуле

y_{1,\mathrm{full}}(n)=\sum_{m=0}^{35}x(m)h_{26}(n-m)

Для наблюдения на интервале воздействия $y_1(n)=y_{1,\mathrm{full}}(n)$ при $0\leq n\leq35$


```python
N2 = N_var % 10 + 30
L = round(N2 / 2)
n_x = np.arange(N2)
x = np.zeros(N2)
x[:L] = 1.0

# Полная линейная свёртка с фрагментом ИХ из п. 1.
y1_full = np.convolve(x, h, mode='full')
y1 = y1_full[:N2]
n_y = np.arange(len(y1_full))

fig, axes = plt.subplots(2, 1, figsize=(8, 5), sharex=True)
for ax, indices, values, label in (
    (axes[0], n_x, x, '$x(n)$'),
    (axes[1], n_y, y1_full, r'$y_{1,\mathrm{full}}(n)$'),
):
    markers, stems, baseline = ax.stem(
        indices, values, linefmt='C0-', markerfmt='C0o', basefmt='k-')
    plt.setp(markers, markersize=3)
    plt.setp(stems, linewidth=1)
    plt.setp(baseline, linewidth=0.7)
    ax.set_ylabel(label)
    ax.grid(axis='y', alpha=0.25)

axes[0].set_title('Прямоугольный импульс и реакция по свёртке')
axes[0].set_yticks([0, 1])
axes[1].axvline(N2 - 0.5, color='0.4', linestyle='--', linewidth=1,
                label='Граница окна: n = 35')
axes[1].legend(frameon=False, fontsize=9)
axes[1].set(xlabel='$n$', xlim=(-1, len(y1_full)),
            xticks=[0, 10, 17, 25, 35, 42, 50, 60])
fig.tight_layout()
plt.show()
```

c) Первые 36 отсчётов выделяют, чтобы рассматривать воздействие и реакцию на одном временном интервале. Отсечение хвоста — выбор окна наблюдения: после окончания импульса система продолжает реагировать за счёт накопленного состояния.


#### 3. Реакция по разностному уравнению

При заданных коэффициентах разностное уравнение имеет вид

$$
\begin{aligned}
y_2(n)={}&0.62x(n)-0.67518x(n-1)+0.62x(n-2)
-0.9278y_2(n-1)-0.676y_2(n-2).
\end{aligned}
$$

Начальные условия: $y_2(-1)=y_2(-2)=0$, $x(n)=0$ при $n<0$. Воздействие $x(n)$ задано в п. 2. Интервал расчёта: $0\leq n\leq35$.


```python
# Два начальных нуля соответствуют отсчётам n = -2 и n = -1.
x_pad = np.pad(x, (2, 0))
y_pad = np.zeros(N2 + 2)
for i in range(2, N2 + 2):
    y_pad[i] = (b[0] * x_pad[i] + b[1] * x_pad[i - 1]
                + b[2] * x_pad[i - 2]
                - a[1] * y_pad[i - 1] - a[2] * y_pad[i - 2])
y2 = y_pad[2:]
delta_y = y2 - y1

fig, axes = plt.subplots(3, 1, figsize=(8, 7), sharex=True)
for ax, values, label in (
    (axes[0], x, '$x(n)$'),
    (axes[1], y2, '$y(n)$'),
    (axes[2], delta_y, '$y_2(n)-y_1(n)$'),
):
    markers, stems, baseline = ax.stem(
        n_x, values, linefmt='C0-', markerfmt='C0o', basefmt='k-')
    plt.setp(markers, markersize=4)
    plt.setp(stems, linewidth=1)
    plt.setp(baseline, linewidth=0.7)
    ax.set_ylabel(label)
    ax.grid(axis='y', alpha=0.25)

axes[0].set_title('Воздействие и реакция по разностному уравнению')
axes[0].set_yticks([0, 1])
axes[1].plot([], [], 'C0o', markersize=4, label='$y_2$: разностное уравнение')
axes[1].plot(n_x, y1, 'C1x', markersize=5, label='$y_1$: свёртка с 26 отсчётами ИХ')
axes[1].legend(frameon=False, fontsize=9)
axes[2].axvline(N1 - 0.5, color='0.4', linestyle='--', linewidth=1)
axes[2].ticklabel_format(axis='y', style='sci', scilimits=(0, 0))
axes[2].set(xlabel='$n$', xlim=(-1, N2), xticks=[0, 5, 10, 17, 20, 25, 30, 35])
fig.tight_layout()
plt.show()

print(f'Длины x и y2: {len(x)} и {len(y2)} отсчётов')
print(f'max |y2 - y1| при 0 <= n < {N1}: {np.max(np.abs(delta_y[:N1])):.2e}')
print(f'max |y2 - y1| при 0 <= n < {N2}: {np.max(np.abs(delta_y)):.6f}')
```

**Сравнение реакций.** При $0\leq n\leq25$ реакции $y_1(n)$ и $y_2(n)$ совпадают с точностью машинного округления. При $26\leq n\leq35$ возникает расхождение из-за усечения ИХ в п. 2:

$$
y_2(n)-y_1(n)=\sum_{k=26}^{n}h(k)x(n-k).
$$

На интервале $0\leq n\leq35$ максимальное абсолютное расхождение составляет **$0.008792$** при $n=27$. Разностное уравнение учитывает рекурсивный хвост ИХ без усечения. Свёртка с первыми 36 отсчётами ИХ даёт ту же реакцию на этом интервале с точностью машинного округления.

**Длины последовательностей.** Массивы $x$ и $y_2$ содержат по 36 отсчётов. Ненулевая часть воздействия длится 18 отсчётов. После выключения входа реакция содержит затухающие колебания; её фактическая длина бесконечна. Значение $N_2=36$ определяет только длину рассчитанного фрагмента реакции.

=e^{\pm j\varphi_z},\\
p_{1,2}&=-0.4639\pm j\sqrt{0.676-0.4639^2}
=r_p e^{\pm j\varphi_p},
\end{aligned}
$$

где

$$
\begin{aligned}
r_p&=\sqrt{0.676}\approx0.822192,\\
\varphi_z&=\arccos(0.5445)\approx0.995003\approx0.316719\pi,\\
\varphi_p&=\arccos\!\left(-\frac{0.4639}{\sqrt{0.676}}\right)
\approx2.170289\approx0.690824\pi.
\end{aligned}
$$

Обе формы описывают одни и те же корни: действительная и мнимая части равны $\rho\cos\varphi$ и $\rho\sin\varphi$. Вещественные коэффициенты обусловливают сопряжённость $z_2=z_1^*$ и $p_2=p_1^*$. Общих корней у числителя и знаменателя нет.

**Передаточная функция в виде произведения множителей:**

$$
H(z)=0.62\,
\frac{(1-e^{j\varphi_z}z^{-1})(1-e^{-j\varphi_z}z^{-1})}
{(1-r_p e^{j\varphi_p}z^{-1})(1-r_p e^{-j\varphi_p}z^{-1})}.
$$

С округлением параметров:

$$
H(z)\approx0.62\,
\frac{(1-e^{j\,0.316719\pi}z^{-1})(1-e^{-j\,0.316719\pi}z^{-1})}
{(1-0.822192e^{j\,0.690824\pi}z^{-1})(1-0.822192e^{-j\,0.690824\pi}z^{-1})}.
$$


#### 5. Карта нулей и полюсов. Устойчивость

Координаты точек карты соответствуют корням из п. 4. Пунктирная линия обозначает единичную окружность $|z|=1$.


```python
phi = np.linspace(0, 2 * np.pi, 400)
fig, ax = plt.subplots(figsize=(5, 5))
ax.plot(np.cos(phi), np.sin(phi), '--', color='0.5', linewidth=1)
ax.axhline(0, color='0.7', linewidth=0.7)
ax.axvline(0, color='0.7', linewidth=0.7)
ax.plot(zeros.real, zeros.imag, 'o', color='C0', fillstyle='none',
        markersize=8, markeredgewidth=1.5, label='Нули')
ax.plot(poles.real, poles.imag, 'x', color='C1',
        markersize=8, markeredgewidth=1.5, label='Полюсы')
for prefix, roots in (('z', zeros), ('p', poles)):
    for i, root in enumerate(roots, start=1):
        ax.annotate(f'${prefix}_{i}$', (root.real, root.imag),
                    xytext=(8, 5), textcoords='offset points')
ax.set(xlabel='Re z', ylabel='Im z',
       title='Карта нулей и полюсов', xlim=(-1.2, 1.2), ylim=(-1.2, 1.2),
       xticks=[-1, -0.5, 0, 0.5, 1], yticks=[-1, -0.5, 0, 0.5, 1])
ax.set_aspect('equal', adjustable='box')
ax.legend(loc='upper left', frameon=False, fontsize=9)
fig.tight_layout()
plt.show()

print(f'Модули нулей: {np.abs(zeros)}')
print(f'Модули полюсов: {np.abs(poles)}')
print(f'Все полюсы строго внутри единичной окружности: {np.all(np.abs(poles) < 1)}')
```

**Устойчивость.** Оба полюса расположены строго внутри единичной окружности:

$$
|p_1|=|p_2|=\sqrt{0.676}\approx0.822192<1.
$$

Следовательно, причинное рекурсивное звено устойчиво. Область сходимости $|z|>\sqrt{0.676}$ содержит единичную окружность. Затухание ИХ определяется множителем $(\sqrt{0.676})^n$, что обеспечивает её абсолютную суммируемость. Нули на единичной окружности не нарушают устойчивость.

**Соответствие п. 4.** На карте представлены нули $z_{1,2}\approx0.544500\pm j\,0.838761$ и полюсы $p_{1,2}\approx-0.463900\pm j\,0.678820$. Их координаты совпадают с результатами п. 4; обе пары симметричны относительно действительной оси.


#### 6. АЧХ и ФЧХ в шкале нормированных частот

При $z=e^{j\hat\omega}$ частотная характеристика равна

$$
H(e^{j\hat\omega})=
\frac{0.62-0.67518e^{-j\hat\omega}+0.62e^{-j2\hat\omega}}
{1+0.9278e^{-j\hat\omega}+0.676e^{-j2\hat\omega}}
=\frac{1.24\cos\hat\omega-0.67518}
{1.676\cos\hat\omega+0.9278+j\,0.324\sin\hat\omega}.
$$

На графиках абсцисса равна $\hat\omega/\pi$, фаза выражена в долях $\pi$. Точки $-\pi$ и $\pi$ соответствуют одной частоте вследствие периодичности. Фаза в нулях АЧХ не определена и на графике пропущена.


```python
from scipy.signal import freqz

omega_zero = np.arccos(-b[1] / (2 * b[0]))
omega_hat = np.unique(np.r_[np.linspace(-np.pi, np.pi, 4097),
                           -omega_zero, omega_zero])
omega_hat, H_freq = freqz(b, a, worN=omega_hat)
amplitude = np.abs(H_freq)
phase = np.angle(H_freq)
phase[amplitude < 1e-12] = np.nan

fig, axes = plt.subplots(2, 1, figsize=(8, 5), sharex=True)
axes[0].plot(omega_hat / np.pi, amplitude, color='C0', linewidth=1.2)
axes[1].plot(omega_hat / np.pi, phase / np.pi, color='C0', linewidth=1.2)
for ax in axes:
    for sign in (-1, 1):
        ax.axvline(sign * omega_zero / np.pi, color='0.6',
                   linestyle='--', linewidth=0.8)
    ax.grid(alpha=0.2)
axes[0].set(title='Амплитудно-частотная характеристика', ylabel=r'$A(\hat\omega)$',
            ylim=(0, None))
axes[1].set(title='Фазочастотная характеристика', ylabel=r'$\varphi(\hat\omega)/\pi$',
            xlabel=r'$\hat\omega/\pi$', xlim=(-1, 1), ylim=(-1.05, 1.05),
            xticks=np.linspace(-1, 1, 9), yticks=[-1, -0.5, 0, 0.5, 1])
fig.tight_layout()
plt.show()

edge_amplitudes = np.abs(freqz(b, a, worN=[0, np.pi])[1])
print(f'Частоты нулевой АЧХ: ±{omega_zero / np.pi:.6f} pi')
print(f'A(0) = {edge_amplitudes[0]:.6f}; A(pi) = {edge_amplitudes[1]:.6f}')
positive_indices = np.flatnonzero(omega_hat >= 0)
peak_index = positive_indices[np.argmax(amplitude[positive_indices])]
print(f'Максимум на сетке: A = {amplitude[peak_index]:.6f}, '
      f'частота = {omega_hat[peak_index] / np.pi:.6f} pi')
```

**Основная полоса.** Частотная характеристика задана на $-\pi\leq\hat\omega<\pi$ рад/отсчёт. Для вещественной системы достаточно неотрицательной половины $0\leq\hat\omega\leq\pi$.

**Связь с картой нулей и полюсов.** Нули $z_{1,2}\approx e^{\pm j\,0.316719\pi}$ лежат на единичной окружности, поэтому $A(\hat\omega)=0$ при $\hat\omega\approx\pm0.316719\pi$. Полюсы с модулем $0.822192$ и аргументами $\pm0.690824\pi$ обусловливают резонансные подъёмы вблизи этих частот. Максимум АЧХ находится примерно при $|\hat\omega|\approx0.704\pi$; его положение зависит от расстояний до всех нулей и полюсов и не обязано совпадать с аргументом полюса.

**Скачок фазы.** При переходе через каждый нуль АЧХ вещественный числитель частотной характеристики меняет знак. Односторонние пределы фазы различаются на $\pi$; в самой точке скачка $A=0$, а фаза не определена. Скачок **на** $\pi$ означает величину изменения фазы, а не частоту $\hat\omega=\pi$.

**Подавление составляющих.** $A(0)\approx0.216921$, $A(\pi)\approx2.559717$. Низкочастотные составляющие преимущественно подавлены, высокочастотные вблизи резонанса усилены. Гармоника на частоте $\hat\omega\approx0.316719\pi$ полностью подавляется в установившемся режиме. Для конечного прямоугольного импульса границы включения и выключения возбуждают затухающие колебания, наблюдаемые в реакции из п. 3.


#### 7. АЧХ и ФЧХ в шкале частот, Гц

Для варианта 6 частота дискретизации $f_{\text{д}}=6000$ Гц. Соответствие шкал:

$$
f=3000\frac{\hat\omega}{\pi}\;\text{Гц},\qquad
\hat\omega=\frac{\pi f}{3000}\;\text{рад/отсчёт}.
$$

Частотная характеристика:

$$
H_f(f)=\frac{0.62-0.67518e^{-j2\pi f/6000}+0.62e^{-j4\pi f/6000}}
{1+0.9278e^{-j2\pi f/6000}+0.676e^{-j4\pi f/6000}}.
$$


```python
fd = 1000 * N_var
f_hz = omega_hat * fd / (2 * np.pi)
f_hz, H_hz = freqz(b, a, worN=f_hz, fs=fd)
amplitude_hz = np.abs(H_hz)
phase_hz = np.angle(H_hz)
phase_hz[amplitude_hz < 1e-12] = np.nan
f_zero = omega_zero * fd / (2 * np.pi)

fig, axes = plt.subplots(2, 1, figsize=(8, 5), sharex=True)
axes[0].plot(f_hz, amplitude_hz, color='C0', linewidth=1.2)
axes[1].plot(f_hz, phase_hz / np.pi, color='C0', linewidth=1.2)
for ax in axes:
    for sign in (-1, 1):
        ax.axvline(sign * f_zero, color='0.6', linestyle='--', linewidth=0.8)
    ax.grid(alpha=0.2)
axes[0].set(title='Амплитудно-частотная характеристика', ylabel='$A(f)$',
            ylim=(0, None))
axes[1].set(title='Фазочастотная характеристика', ylabel=r'$\varphi(f)/\pi$',
            xlabel='$f$, Гц', xlim=(-fd / 2, fd / 2), ylim=(-1.05, 1.05),
            xticks=np.arange(-fd / 2, fd / 2 + 1, 1000),
            yticks=[-1, -0.5, 0, 0.5, 1])
fig.tight_layout()
plt.show()

print(f'Частота дискретизации: {fd} Гц; частота Найквиста: {fd / 2:g} Гц')
print(f'Частоты нулевой АЧХ: ±{f_zero:.3f} Гц')
print(f'Положительная частота максимума АЧХ на сетке: {f_hz[peak_index]:.3f} Гц')
```

**Основная полоса.** Интервал $-\pi\leq\hat\omega<\pi$ соответствует $-3000\leq f<3000$ Гц. Неотрицательная половина — $0\leq f\leq3000$ Гц. Точки $f=-3000$ и $f=3000$ Гц эквивалентны вследствие периодичности с периодом 6000 Гц.

**Соответствие шкал.** Значения $\hat\omega=0$, $\pi/2$, $\pi$ соответствуют частотам $f=0$, $1500$, $3000$ Гц. Нули АЧХ расположены при $f\approx\pm950.158$ Гц; резонансные максимумы — около $\pm2113$ Гц. Фаза в нулях АЧХ не определена, поэтому соответствующие точки пропущены.

Форма характеристик совпадает с п. 6: изменён только масштаб оси частот. АЧХ безразмерна, фаза на графике выражена в долях $\pi$.


### Часть 2

#### 1. Последовательность в двух шкалах времени

$$
x(nT)=1.06\cos\left(2\pi\cdot500nT+\frac{\pi}{4}\right)
+2.12\cos\left(2\pi\cdot750nT+\frac{\pi}{8}\right).
$$

В шкале нормированного времени:

$$
x(n)=1.06\cos\left(\frac{\pi n}{4}+\frac{\pi}{4}\right)
+2.12\cos\left(\frac{3\pi n}{8}+\frac{\pi}{8}\right),\qquad n=0,\ldots,63.
$$


```python
N = 64
fd_part2 = 2000 * (N_var % 5 + 1)
T = 1 / fd_part2
A1 = 1 + 0.01 * N_var
A2 = 2 * A1
f1 = fd_part2 / 8
f2 = 1.5 * f1

n_part2 = np.arange(N)
t_part2 = n_part2 * T
x_part2 = (A1 * np.cos(2 * np.pi * f1 * t_part2 + np.pi / 4)
           + A2 * np.cos(2 * np.pi * f2 * t_part2 + np.pi / 8))

fig, axes = plt.subplots(2, 1, figsize=(8, 5), sharey=True)
for ax, time, xlabel in (
    (axes[0], 1000 * t_part2, '$nT$, мс'),
    (axes[1], n_part2, r'$n=t_n/T$'),
):
    markers, stems, baseline = ax.stem(
        time, x_part2, linefmt='C0-', markerfmt='C0o', basefmt='k-')
    plt.setp(markers, markersize=3)
    plt.setp(stems, linewidth=1)
    plt.setp(baseline, linewidth=0.7)
    ax.set(xlabel=xlabel, ylabel='$x$', ylim=(-3.4, 3.4))
    ax.grid(axis='y', alpha=0.25)
axes[0].set(title='Дискретное время', xlim=(-0.25, 16),
            xticks=[0, 4, 8, 12, 15.75])
axes[1].set(title='Дискретное нормированное время', xlim=(-1, 64),
            xticks=[0, 16, 32, 48, 63])
fig.tight_layout()
plt.show()

print(f'N = {N}; T = {T * 1000:g} мс; NT = {N * T * 1000:g} мс')
print(f'Моменты отсчётов: 0 … {t_part2[-1] * 1000:g} мс')
```

**Шкалы времени.** Обе диаграммы содержат одни и те же 64 отсчёта. Шаг дискретного времени равен $T=0.25$ мс, шаг нормированного времени — единице. Интервал $0\leq n<64$ соответствует $0\leq t<16$ мс; последний отсчёт находится при $t_{63}=15.75$ мс. Точка $n=64$ относится к следующему периоду и в массив не входит.

**Периодичность.** Нормированные частоты гармоник равны $f_1T=1/8$ и $f_2T=3/16$. Их основные периоды составляют 8 и 16 отсчётов, основной период суммы — 16 отсчётов (4 мс). Заданное $N=64$ также является периодом, но не основным: на графиках представлены четыре повторения последовательности. За 64 отсчёта первая гармоника совершает 8 колебаний, вторая — 12.


#### 2. Восстановление последовательности посредством ОДПФ

Для последовательности из п. 1 при $N=64$:

$$
X(k)=\sum_{n=0}^{63}x(n)e^{-j2\pi kn/64},\qquad
x_{\text{восст}}(n)=\frac{1}{64}\sum_{k=0}^{63}X(k)e^{j2\pi kn/64}.
$$
Абсцисса графика — дискретное нормированное время $n=t_n/T$.


```python
X_part2 = np.fft.fft(x_part2)
x_restored = np.fft.ifft(X_part2)
reconstruction_error = np.max(np.abs(x_restored - x_part2))
imaginary_residual = np.max(np.abs(x_restored.imag))

fig, ax = plt.subplots(figsize=(8, 3))
markers, stems, baseline = ax.stem(
    n_part2, x_restored.real, linefmt='C0-', markerfmt='C0o', basefmt='k-')
plt.setp(markers, markersize=3)
plt.setp(stems, linewidth=1)
plt.setp(baseline, linewidth=0.7)
ax.set(title='Последовательность после ДПФ и ОДПФ',
       xlabel=r'$n=t_n/T$', ylabel=r'$x_{\mathrm{восст}}(n)$',
       xlim=(-1, N), xticks=[0, 16, 32, 48, 63], ylim=(-3.4, 3.4))
ax.grid(axis='y', alpha=0.25)
fig.tight_layout()
plt.show()

print(f'Максимальная ошибка восстановления: {reconstruction_error:.3e}')
print(f'Максимальный модуль мнимой части: {imaginary_residual:.3e}')
```

**Результат.** Восстановленная последовательность содержит 64 отсчёта и совпадает с исходной с точностью машинного округления:

$$
\varepsilon_{\max}=\max_{0\leq n<64}|x_{\text{восст}}(n)-x(n)|\approx 8.951\cdot10^{-16}.
$$

Для вещественной исходной последовательности точное ОДПФ также вещественно. Численный остаток мнимой части не превышает $2.220\cdot10^{-16}$.


#### 3. Амплитудный и фазовый спектры

Двусторонние спектры определяются коэффициентами $C_k=X(k\bmod64)/64$. При $N=64$ и $f_{\text{д}}=4000$ Гц:

$$
\Delta f=62.5\;\text{Гц},\qquad \Delta\hat\omega=\frac{\pi}{32},\qquad
k_1=\frac{f_1N}{f_{\text{д}}}=8,\quad k_2=12.
$$

Нормированная шкала графиков — $\hat\omega_k/\pi=2k/N$, абсолютная — $f_k$ в герцах.

```python
k_part2 = np.arange(-N // 2, N // 2)
C_part2 = np.fft.fftshift(X_part2) / N
omega_spectrum = 2 * np.pi * k_part2 / N
f_spectrum = k_part2 * fd_part2 / N
amplitude_spectrum = np.abs(C_part2)

spectrum_tolerance = 1e-12 * amplitude_spectrum.max()
nonzero_spectrum = amplitude_spectrum > spectrum_tolerance
phase_spectrum = np.full(N, np.nan)
phase_spectrum[nonzero_spectrum] = np.angle(C_part2[nonzero_spectrum])

fig, axes = plt.subplots(2, 2, figsize=(10, 5), sharey='row')
for column, frequencies, xlabel, limits, ticks in (
    (0, omega_spectrum / np.pi, r'$\hat\omega_k/\pi$', (-1, 1),
     [-1, -0.5, 0, 0.5, 1]),
    (1, f_spectrum, '$f_k$, Гц', (-fd_part2 / 2, fd_part2 / 2),
     [-2000, -1000, 0, 1000, 2000]),
):
    for row, selected, values in (
        (0, np.ones(N, dtype=bool), amplitude_spectrum),
        (1, nonzero_spectrum, phase_spectrum / np.pi),
    ):
        ax = axes[row, column]
        markers, stems, baseline = ax.stem(
            frequencies[selected], values[selected],
            linefmt='C0-', markerfmt='C0o', basefmt='k-')
        plt.setp(markers, markersize=3)
        plt.setp(stems, linewidth=1)
        plt.setp(baseline, linewidth=0.7)
        ax.set(xlabel=xlabel, xlim=limits, xticks=ticks)
        ax.grid(alpha=0.2)
    axes[0, column].set(title='Амплитудный спектр', ylim=(0, 1.2))
    axes[1, column].set(title='Фазовый спектр', ylim=(-0.3, 0.3),
                        yticks=[-0.25, -0.125, 0, 0.125, 0.25])
axes[0, 0].set_ylabel('$|C_k|$')
axes[1, 0].set_ylabel(r'$\varphi_k/\pi$')
fig.tight_layout()
plt.show()
```

**Ненулевые составляющие:**

| $k$ | $\hat\omega_k/\pi$ | $f_k$, Гц | $A_k$ | $\varphi_k$ |
|---:|---:|---:|---:|---:|
| $-12$ | $-3/8$ | $-750$ | $1.06$ | $-\pi/8$ |
| $-8$ | $-1/4$ | $-500$ | $0.53$ | $-\pi/4$ |
| $8$ | $1/4$ | $500$ | $0.53$ | $\pi/4$ |
| $12$ | $3/8$ | $750$ | $1.06$ | $\pi/8$ |

**Связь шкал.** Дискретное время $t_n=nT$ и нормированное время $n=t_n/T$ связаны через $T=0.25$ мс. Для частот $\hat\omega_k=2\pi f_kT$, поэтому $f_k=2000\hat\omega_k/\pi$ Гц.

**Вид спектров.** Каждая косинусоида образует пару составляющих на положительной и отрицательной частотах. Их амплитуды равны половине амплитуды соответствующей косинусоиды: $A_1/2=0.53$ и $A_2/2=1.06$. Положительным частотам соответствуют начальные фазы $\pi/4$ и $\pi/8$, отрицательным — противоположные значения. Амплитудный спектр симметричен относительно нуля.

Частоты гармоник совпадают с узлами сетки: $k_1=8$, $k_2=12$. Остальные коэффициенты, включая постоянную составляющую, теоретически равны нулю; спектральное растекание отсутствует. Малые численные остатки обусловлены округлением. Их фазы не имеют физического смысла и на графике отсутствуют.


#### 4. ДПФ конечной последовательности

Конечная последовательность длины $N=64$ совпадает с отсчётами из п. 1 при $0\leq n\leq63$ и равна нулю вне этого интервала. Её ДПФ:

$$
X_N(k)=\sum_{n=0}^{63}x(n)e^{-j2\pi kn/64}.
$$

На графиках представлена центрированная сетка $k=-32,\ldots,31$ с нормированной частотой $\hat\omega_k/\pi=2k/64$.


```python
x_finite = x_part2.copy()
X_finite = np.fft.fft(x_finite, n=N)
dft_magnitude = np.abs(np.fft.fftshift(X_finite))

fig, axes = plt.subplots(2, 1, figsize=(8, 5), sharex=True)
for ax, values, title, ylabel in (
    (axes[0], dft_magnitude, 'Модуль ДПФ конечной последовательности', '$|X_N(k)|$'),
    (axes[1], amplitude_spectrum, 'Амплитудный спектр периодической последовательности', '$A_k$'),
):
    markers, stems, baseline = ax.stem(
        omega_spectrum / np.pi, values,
        linefmt='C0-', markerfmt='C0o', basefmt='k-')
    plt.setp(markers, markersize=3)
    plt.setp(stems, linewidth=1)
    plt.setp(baseline, linewidth=0.7)
    ax.set(title=title, ylabel=ylabel, ylim=(0, 1.15 * values.max()))
    ax.grid(alpha=0.2)
axes[1].set(xlabel=r'$\hat\omega_k/\pi$', xlim=(-1, 1),
            xticks=[-1, -0.5, -0.25, 0, 0.25, 0.5, 1])
fig.tight_layout()
plt.show()

print(f'Модуль ДПФ при k = ±8: {dft_magnitude[k_part2 == 8][0]:.2f}')
print(f'Модуль ДПФ при k = ±12: {dft_magnitude[k_part2 == 12][0]:.2f}')
print(f'max ||X_N| / N - A_k| = {np.max(np.abs(dft_magnitude / N - amplitude_spectrum)):.3e}')
```

**Сравнение.** ДПФ конечной последовательности численно совпадает с ДПФ одного периода из п. 2: обе суммы содержат одни и те же 64 отсчёта. Различие относится к продолжению сигнала за пределами этого интервала: нулевому для конечной последовательности и периодическому для периодической.

Между модулем ДПФ и двусторонним амплитудным спектром выполняется соотношение

$$
|X_N(k)|=64A_k.
$$

| Индексы $k$ | $\hat\omega_k/\pi$ | Модуль ДПФ | $A_k$ |
|---:|---:|---:|---:|
| $\pm8$ | $\pm1/4$ | $33.92$ | $0.53$ |
| $\pm12$ | $\pm3/8$ | $67.84$ | $1.06$ |

Положение составляющих и отношение их высот совпадают; вертикальные масштабы графиков различаются в 64 раза. Остальные отсчёты ДПФ теоретически равны нулю, поскольку обе гармоники совпадают с узлами частотной сетки.


#### 5. Восстановление аналогового сигнала

При $N=64$, $T=0.25$ мс:

$$
x_{\text{восст}}(t)=\frac{1}{64}\sum_{k=-32}^{31}X_a(k)e^{j2\pi kt/0.016},
\qquad 0\leq t\leq0.01575\;\text{с}.
$$

Шаг $\Delta t=T/4=0.0625$ мс. Число точек — $4(N-1)+1=253$, включая обе границы интервала. Перестановка fftshift формирует $X_a(k)$ в порядке $k=-32,\ldots,31$.

Исходный аналоговый сигнал:

$$
x_{\text{исх}}(t)=1.06\cos(2\pi\cdot500t+\pi/4)
+2.12\cos(2\pi\cdot750t+\pi/8).
$$


```python
k_analog = np.arange(-N // 2, N // 2)
X_a = np.fft.fftshift(X_part2)
dt_analog = T / 4
t_analog = np.arange(4 * (N - 1) + 1) * dt_analog
f_analog = k_analog / (N * T)

exponentials = np.exp(2j * np.pi * np.outer(t_analog, f_analog))
x_analog_complex = exponentials @ X_a / N
x_analog = x_analog_complex.real
x_analog_original = (A1 * np.cos(2 * np.pi * f1 * t_analog + np.pi / 4)
                     + A2 * np.cos(2 * np.pi * f2 * t_analog + np.pi / 8))
analog_amplitude = np.abs(X_a) / N

fig = plt.figure(figsize=(11, 8))
grid = fig.add_gridspec(3, 2)
ax_samples = fig.add_subplot(grid[0, 0])
ax_dft = fig.add_subplot(grid[0, 1])
ax_restored = fig.add_subplot(grid[1, 0])
ax_spectrum = fig.add_subplot(grid[1, 1])
ax_original = fig.add_subplot(grid[2, :])

for ax, coordinates, values in (
    (ax_samples, t_part2 * 1000, x_part2),
    (ax_dft, f_analog, np.abs(X_a)),
    (ax_spectrum, f_analog, analog_amplitude),
):
    markers, stems, baseline = ax.stem(
        coordinates, values, linefmt='C0-', markerfmt='C0o', basefmt='k-')
    plt.setp(markers, markersize=3)
    plt.setp(stems, linewidth=1)
    plt.setp(baseline, linewidth=0.7)

ax_restored.plot(t_analog * 1000, x_analog, color='C0', linewidth=1.1)
ax_original.plot(t_analog * 1000, x_analog_original, color='C0', linewidth=1.1)
ax_samples.set(title='Периодическая последовательность', xlabel='$nT$, мс', ylabel='$x(n)$')
ax_restored.set(title='Восстановленный сигнал', xlabel='$t$, мс', ylabel=r'$x_{\mathrm{восст}}(t)$')
ax_original.set(title='Исходный аналоговый сигнал', xlabel='$t$, мс', ylabel=r'$x_{\mathrm{исх}}(t)$')
for ax in (ax_samples, ax_restored, ax_original):
    ax.set(xlim=(-0.2, 16), ylim=(-3.4, 3.4), xticks=[0, 4, 8, 12, 15.75])
ax_dft.set(title='Модуль ДПФ последовательности', ylabel='$|X_a(k)|$', ylim=(0, 75))
ax_spectrum.set(title='Амплитудный спектр восстановленного сигнала', ylabel='$|C_k|$', ylim=(0, 1.2))
for ax in (ax_dft, ax_spectrum):
    ax.set(xlabel='$f_k$, Гц', xlim=(-2000, 2000), xticks=[-2000, -1000, 0, 1000, 2000])
for ax in fig.axes:
    ax.grid(alpha=0.2)
fig.tight_layout()
plt.show()

sample_error = np.max(np.abs(x_analog_complex[::4] - x_part2))
print(f'Шаг: {dt_analog * 1000:g} мс; число точек: {len(t_analog)}')
print(f'Ошибка в исходных узлах дискретизации: {sample_error:.3e}')
print(f'Максимальный модуль мнимого остатка: {np.max(np.abs(x_analog_complex.imag)):.3e}')
```

**Построение.** Восстановленный и исходный сигналы представлены на заданной сетке с шагом $T/4$; линии соединяют рассчитанные значения. Каждый четвёртый узел этой сетки совпадает с исходным моментом дискретизации $nT$.

Амплитудный спектр восстановленного сигнала — двусторонний линейчатый спектр с составляющими на $\pm500$ и $\pm750$ Гц. На графике показаны модули коэффициентов ряда Фурье.

#### 7. Равенство Парсеваля

Для последовательности длины $N=64$ проверяется соотношение

$$
\sum_{n=0}^{63}|x(n)|^2=\frac{1}{64}\sum_{k=0}^{63}|X(k)|^2.
$$


```python
energy_time = np.sum(np.abs(x_part2) ** 2)
energy_frequency = np.sum(np.abs(X_part2) ** 2) / N
parseval_error = abs(energy_time - energy_frequency)
parseval_relative_error = parseval_error / energy_time
power_time = energy_time / N
power_frequency = np.sum(np.abs(C_part2) ** 2)

print(f'Сумма квадратов отсчётов: {energy_time:.12f}')
print(f'Сумма квадратов модулей ДПФ / N: {energy_frequency:.12f}')
print(f'Абсолютное расхождение: {parseval_error:.3e}')
print(f'Относительное расхождение: {parseval_relative_error:.3e}')
print(f'Средняя мощность во временной области: {power_time:.12f}')
print(f'Средняя мощность в частотной области: {power_frequency:.12f}')
print(f'Равенство выполнено: {np.isclose(energy_time, energy_frequency, rtol=1e-12, atol=1e-12)}')
```

**Аналитическая проверка.** На интервале из 64 отсчётов гармоники ортогональны и содержат целое число периодов. Поэтому

$$
E_N=\frac{N}{2}(A_1^2+A_2^2)
=\frac{64}{2}(1.06^2+2.12^2)=179.776.
$$

В ДПФ ненулевыми являются две пары коэффициентов с модулями $33.92$ и $67.84$:

$$
\frac{1}{64}\sum_{k=0}^{63}|X(k)|^2
=\frac{2\cdot33.92^2+2\cdot67.84^2}{64}=179.776.
$$

Средняя мощность периодической последовательности:

$$
P=\frac{179.776}{64}=2.809
=2\cdot0.53^2+2\cdot1.06^2.
$$

**Смысл результата.** Равенство Парсеваля подтверждено аналитически и численно с точностью округления. Переход от отсчётов к спектральным коэффициентам не изменяет энергию при указанной нормировке. Значение $179.776$ относится к конечному фрагменту из 64 отсчётов; для бесконечного периодического продолжения характеристикой служит средняя мощность $2.809$.
