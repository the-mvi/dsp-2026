## Основные теоретические положения

### 1. Линейные дискретные системы

Линейная дискретная система (ЛДС) описывается разностным уравнением, которое связывает текущий выходной сигнал $y(n)$ с текущими и прошлыми значениями входного сигнала $x(n)$ и прошлыми значениями выходного сигнала. Для рекурсивного звена 2-го порядка разностное уравнение имеет вид:
$$y(n) = b_0 x(n) + b_1 x(n-1) + b_2 x(n-2) - a_1 y(n-1) - a_2 y(n-2)$$

Применяя Z-преобразование при нулевых начальных условиях, получается передаточная функция:
$$H(z) = \frac{b_0 + b_1 z^{-1} + b_2 z^{-2}}{1 + a_1 z^{-1} + a_2 z^{-2}}$$

**Импульсная характеристика** $h(n)$ - это реакция системы на единичный импульс. Для рекурсивных (БИХ) систем она теоретически бесконечна, но на практике быстро затухает.

**Свёртка** позволяет найти реакцию системы на произвольный сигнал:
$$y(n) = \sum_{k} x(k) h(n-k)$$
Длина линейной свёртки двух последовательностей длиной $N_1$ и $N_2$ равна $N_1 + N_2 - 1$.

**Нули и полюсы** - это корни числителя и знаменателя передаточной функции соответственно. Они могут быть представлены в алгебраической ($z = x + jy$) и показательной ($z = |z|e^{j\phi}$) формах. 
**Устойчивость** системы гарантируется, если все полюсы лежат строго внутри единичной окружности на комплексной плоскости ($|p| < 1$).

**Частотные характеристики** (АЧХ и ФЧХ) определяются путем подстановки $z = e^{j\hat{\omega}}$ в передаточную функцию. АЧХ показывает изменение амплитуды на разных частотах, а ФЧХ - фазовый сдвиг. Физическая частота $f$ в Герцах связана с нормированной частотой $\hat{\omega}$ соотношением $f = \frac{\hat{\omega}}{2\pi} f_д$, где $f_д$ — частота дискретизации.

### 2. Дискретное преобразование Фурье (ДПФ)

**ДПФ** применяется для анализа частотного состава периодических последовательностей и последовательностей конечной длины $N$. Прямое ДПФ переводит сигнал из временной области в частотную:
$$X(k) = \sum_{n=0}^{N-1} x(n) e^{-j \frac{2\pi}{N} kn}$$

**Обратное ДПФ (ОДПФ)** восстанавливает сигнал по его спектру:
$$x(n) = \frac{1}{N} \sum_{k=0}^{N-1} X(k) e^{j \frac{2\pi}{N} kn}$$

**Спектры сигнала**:
* **Амплитудный спектр** $|X(k)|$ показывает амплитуды гармоник.
* **Фазовый спектр** $\arg X(k)$ показывает их начальные фазы.
Модуль ДПФ конечной последовательности связан с амплитудным спектром периодической последовательности соотношением $|X(k)| = N \cdot |c_k|$, где $c_k$ - коэффициенты ряда Фурье. Если частоты гармоник кратны основной частоте, спектр не испытывает «растекания».

**Восстановление аналогового сигнала** по отсчетам ДПФ выполняется с помощью ряда Фурье. Согласно теореме Котельникова, если частота дискретизации больше удвоенной максимальной частоты сигнала, непрерывный сигнал может быть восстановлен без потерь.

**Равенство Парсеваля** отражает закон сохранения энергии: энергия сигнала во временной области равна энергии в частотной области:
$$\sum_{n=0}^{N-1} |x(n)|^2 = \frac{1}{N} \sum_{k=0}^{N-1} |X(k)|^2$$

### Пункт 1.4 Нули и полюсы

```python
zeros, poles, k_gain = tf2zpk(b, a)

print("Нули (z_i):")
for z in zeros:
    r = np.abs(z); phi = np.angle(z)
    print(f"  z = {z.real:+.6f} {z.imag:+.6f}j | |z| = {r:.6f}, arg = {phi:.6f} рад = {phi/np.pi:.6f}·π")

print("\nПолюсы (p_i):")
for p in poles:
    r = np.abs(p); phi = np.angle(p)
    print(f"  p = {p.real:+.6f} {p.imag:+.6f}j | |p| = {r:.6f}, arg = {phi:.6f} рад = {phi/np.pi:.6f}·π")

print(f"\nКоэффициент усиления k = {k_gain:.6f}")
```

Передаточная функция системы определяется как отношение полиномов:
$$H(z) = \frac{b_0 z^2 + b_1 z + b_2}{z^2 + a_1 z + a_2}$$

* **Нули** - это корни числителя. Они определяют частоты, на которых система подавляет сигнал.
* **Полюсы** - это корни знаменателя. Они определяют частоты резонанса и скорость затухания переходных процессов.

**1. Формы записи нулей и полюсов:**
В ходе вычислений были получены комплексно-сопряженные пары нулей и полюсов. Комплексное число можно записать двумя способами:
1. **Алгебраическая форма:** $z = x + jy$ (где $x$ - действительная часть, $y$ - мнимая).
2. **Показательная (полярная) форма:** $z = |z| e^{j\phi}$, где $|z| = \sqrt{x^2 + y^2}$ - модуль (радиус), а $\phi = \arctan(y/x)$ - аргумент (угол).

**Связь между формами:**
Модуль и аргумент связаны с действительной и мнимой частями через формулу Эйлера: $e^{j\phi} = \cos\phi + j\sin\phi$. То есть $x = |z|\cos\phi$, $y = |z|\sin\phi$.

**2. Результаты вычислений для нашего варианта:**
Вычисленные значения аргументов удобно выражать в долях числа $\pi$.

| Тип | Алгебраическая форма | Модуль ($\| \cdot \|$) | Аргумент (рад) | Аргумент ($\cdot \pi$) |
| :--- | :--- | :--- | :--- | :--- |
| **Нуль \(z_1\)** | $-0.5534 + j0.9454$ | $1.0954$ | $2.1004$ | $0.6686 \pi$ |
| **Нуль \(z_2\)** | $-0.5534 - j0.9454$ | $1.0954$ | $-2.1004$ | $-0.6686 \pi$ |
| **Полюс \(p_1\)** | $+0.4764 + j0.6746$ | $0.8258$ | $0.9559$ | $0.3043 \pi$ |
| **Полюс \(p_2\)** | $+0.4764 - j0.6746$ | $0.8258$ | $-0.9559$ | $-0.3043 \pi$ |

*Примечание: Нули и полюсы всегда образуют комплексно-сопряженные пары (отличаются только знаком перед мнимой частью), так как коэффициенты фильтра — действительные числа.*

**3. Представление передаточной функции в виде произведения простейших множителей:**
Используя найденные нули и полюсы в показательной форме, передаточную функцию можно записать как:
$$H(z) = k \frac{(1 - z_1 z^{-1})(1 - z_2 z^{-1})}{(1 - p_1 z^{-1})(1 - p_2 z^{-1})}$$
Подставляем наши значения ($k = b_0 = 0.64$):
$$H(z) = 0.64 \frac{(1 - 1.0954 e^{j0.6686\pi} z^{-1})(1 - 1.0954 e^{-j0.6686\pi} z^{-1})}{(1 - 0.8258 e^{j0.3043\pi} z^{-1})(1 - 0.8258 e^{-j0.3043\pi} z^{-1})}$$

**1. Границы основной полосы частот:**
Графики построены в шкале нормированных частот $\hat{\omega}$, которая связана с физической частотой $f$ соотношением $\hat{\omega} = 2\pi f / f_д$. 
Основная полоса частот ограничена интервалом $\hat{\omega} \in [0, \pi]$. Значение $\hat{\omega} = \pi$ соответствует половине частоты дискретизации ($f_д/2$). За пределами этого интервала АЧХ начинает зеркально повторяться из-за эффекта наложения спектров (алиасинга).

**2. Соответствие карты нулей и полюсов и вида АЧХ:**

**3. Скачок на $\pi$:**
Скачок ФЧХ ровно на $\pi$ (180 градусов) происходит в тех точках, где АЧХ обращается в ноль (когда полюс/нуль лежит точно на единичной окружности). В нашем случае АЧХ не опускается до абсолютного нуля (так как нули лежат не на окружности), поэтому явного скачка на $\pi$ нет. ФЧХ меняется непрерывно (с учетом `unwrap`), общий набег фазы составляет около $-2\pi$, что характерно для фильтра 2-го порядка.

**4. Подавление частот:**
Судя по графику АЧХ, система преимущественно **подавляет высокие частоты** (справа от пика, особенно в районе $0.7\pi - 1.0\pi$). При этом низкие и средние частоты (в районе $0 - 0.4\pi$) система **усиливает**. Таким образом, данное рекурсивное звено ведет себя как полосовой фильтр с доминирующим резонансом в области низких/средних частот.

### Пункт 2.1 Периодическая последовательность

```python
n = np.arange(N)

t_norm = n / N
t_abs = n * T

x_per = A1*np.cos(2*np.pi*f1*n*T + np.pi/4) + A2*np.cos(2*np.pi*f2*n*T + np.pi/8)

fig, axes = plt.subplots(2, 1, figsize=(10, 6))
axes[0].stem(n, x_per, basefmt="k-")
axes[0].set_title("x(n) в шкале дискретного времени n")
axes[0].set_xlabel("n"); axes[0].set_ylabel("x(n)"); axes[0].grid(True)

axes[1].stem(t_norm, x_per, basefmt="k-")
axes[1].set_title("x(n) в шкале дискретного нормированного времени n/N")
axes[1].set_xlabel("n/N"); axes[1].set_ylabel("x(n)"); axes[1].grid(True)
plt.tight_layout(); plt.show()
```

$$x(nT) = A_1 \cos(2\pi f_1 n T + \pi/4) + A_2 \cos(2\pi f_2 n T + \pi/8)$$

Согласно заданию варианта 7, параметры сигнала следующие:

* Период дискретизации: $T = 1/f_д$
* Амплитуды: $A_1 = 1.07$, $A_2 = 2.14$
* Частоты: \(f_1 = 750\) Гц, $f_2 = 1125$ Гц

**Пояснение к графикам:**
На рисунке представлены два графика одного и того же сигнала:
1. **В шкале дискретного времени $n$** (верхний график): по оси X откладываются номера отсчетов от 0 до 63. Показывает, сколько раз были произведены замеры.
2. **В шкале дискретного нормированного времени $n/N$** (нижний график): по оси X откладывается относительное время от 0 до 1. 

**Связь дискретного нормированного времени с дискретным временем:**
Нормированное время вычисляется как $t_{norm} = n/N$, где $n$ - номер текущего отсчета, а \(N\) - период (длина) последовательности. Ось от 0 до 1 показывает ровно один полный период сигнала. Нормированное время удобно тем, что позволяет абстрагироваться от конкретного количества точек \(N\) и частоты дискретизации, показывая чисто относительные временные интервалы.

### Пункт 2.2 ОДПФ от ДПФ

```python
X = np.fft.fft(x_per)
x_rec = np.fft.ifft(X).real   # ОДПФ

plt.figure()
plt.stem(t_norm, x_rec, basefmt="k-")
plt.title("Последовательность после ОДПФ(ДПФ(x)) — совпадает с исходной")
plt.xlabel("n/N"); plt.ylabel("x_rec(n)"); plt.grid(True)
plt.show()

print(f"Максимальная ошибка восстановления: {np.max(np.abs(x_rec - x_per)):.3e}")
```

В данном пункте проверяется фундаментальное свойство дискретного преобразования Фурье: прямое и обратное преобразования являются взаимно однозначными.

**Теоретическое обоснование:**
* **ДПФ (Дискретное преобразование Фурье)** переводит сигнал из временной области в частотную, раскладывая его на гармоники (вычисление спектра).
* **ОДПФ (Обратное дискретное преобразование Фурье)** выполняет противоположную операцию: по известным амплитудам и фазам гармоник восстанавливает исходный временной сигнал.

С помощью функций `np.fft.fft()` и `np.fft.ifft()` в Python были последовательно вычислены ДПФ и ОДПФ:
$$x_{rec}(n) = \text{ОДПФ}\{ \text{ДПФ}\{ x(n) \} \}$$

**Анализ результатов:**
1. **Визуальное сравнение:** График восстановленной последовательности $x_{rec}(n)$ в шкале нормированного времени $n/N$ полностью совпадает с исходным графиком \(x(n)\), построенным в Пункте 2.1.
2. **Численная проверка:** Максимальное абсолютное расхождение между исходным и восстановленным сигналами составило $\mathbf{8.882 \cdot 10^{-16}}$. Это значение является **машинным нулем** (пределом точности вычислений чисел с плавающей точкой). 

Результат подтверждает, что преобразование Фурье не теряет информацию о сигнале. Ошибка порядка \(10^{-16}\) вызвана лишь погрешностью округления компьютера и на практике считается равной нулю.

**1. Связь частоты $f$ (Гц) с дискретной нормированной частотой:**

В нашем варианте $f_д = 6000$ Гц. Следовательно, каждый шаг по оси $k/N$ (на $\Delta k = 1$) соответствует сдвигу по частоте на $6000 / 64 = 93.75$ Гц.

* Зеркальные пики: $k=56$ ($f = 5250$ Гц) и $k=52$ ($f = 4875$ Гц). Они возникают из-за свойства симметрии ДПФ для действительных сигналов. Частота $3000$ Гц ($f_д/2$) является осью симметрии.

**3. Вид фазового спектра:**
Фазовый спектр $\arg X(k)$ имеет физический смысл только в точках, где амплитуда отлична от нуля. Для наших пиков фазы составляют:

Эти значения в точности совпадают с начальными фазами, заданными в уравнении сигнала $x(nT)$. В остальных точках (где амплитуда равна нулю) график фазового спектра представляет собой «шум» из-за отсутствия физического смысла фазы.


### Пункт 2.4 ДПФ конечной последовательности

```python
X_fin = np.fft.fft(x_per)
k_norm = np.arange(N) / N

fig, axes = plt.subplots(2, 1, figsize=(10, 6))
axes[0].stem(k_norm, np.abs(X_fin), basefmt="k-")
axes[0].set_title("Модуль ДПФ конечной последовательности |X(k)|")
axes[0].set_xlabel("k/N"); axes[0].grid(True)

axes[1].stem(k_norm, X_mag, basefmt="k-", linefmt="r-", markerfmt="ro")
axes[1].set_title("Амплитудный спектр периодической последовательности")
axes[1].set_xlabel("k/N"); axes[1].grid(True)
plt.tight_layout(); plt.show()
```

В данном пункте сравниваются два понятия:
1. **ДПФ конечной последовательности** - операция, которую выполняет компьютер над массивом из \(N\) отсчетов (от \(0\) до \(N-1\)). За пределами этого массива сигнал считается нулевым.
2. **Амплитудный спектр периодической последовательности** - математическая модель, где этот же массив из \(N\) отсчетов повторяется бесконечно.

**Связь модуля ДПФ конечной последовательности с амплитудным спектром периодической последовательности:**
Модуль ДПФ конечной последовательности \(|X(k)|\) связан с коэффициентами дискретного ряда Фурье периодической последовательности \(c_k\) следующим соотношением:
$$|X(k)| = N \cdot |c_k|$$
где \(N\) - длина периода (в нашем случае \(N=64\)).

**Анализ графиков:**
На рисунке представлены графики модуля ДПФ (синий) и амплитудного спектра периодической последовательности (красный). Они **полностью совпадают**. 
Это объясняется тем, что исходный сигнал \(x(n)\) состоит из гармоник, частоты которых кратны основной частоте \(f_0 = f_д / N\) (то есть укладываются в период \(N\) целое число раз). В этом случае ДПФ конечного фрагмента не испытывает «растекания спектра» (spectral leakage), и его значения в точках \(k\) в точности равны амплитудам гармоник бесконечного периодического сигнала (умноженным на \(N\)). 

Таким образом, ДПФ конечной последовательности позволяет абсолютно точно определить амплитудный спектр периодического сигнала, если на интервале анализа укладывается целое число периодов.

### Пункт 2.5 и 2.6 Восстановление аналогового сигнала

```python
# Отсчёты X_a(k): сдвиг на N/2
X_a = np.zeros(N, dtype=complex)
half = N // 2
X_a[:half] = X[half:]       # k = -N/2 ... -1  →  X(N+k)
X_a[half:] = X[:half]       # k = 0 ... N/2-1  →  X(k)
# Но проще: X_a(k) для k = 0..N-1 соответствует X((k+N/2) mod N)
X_a = np.roll(X, -half)     # X_a[0] = X(N/2), ..., X_a[N/2] = X(0), ...

# Время: t = m·Δt, Δt = 0.25T, t ∈ [0, (N−1)T]
dt = 0.25 * T
t = np.arange(0, (N - 1) * T + dt/2, dt)

# Восстановленный сигнал по формуле
x_rec_t = np.zeros_like(t, dtype=complex)
for i, ti in enumerate(t):
    s = 0.0 + 0.0j
    for kk in range(-half, half):
        idx = kk % N
        s += X[idx] * np.exp(1j * 2 * np.pi * kk * ti / (N * T))
    x_rec_t[i] = s / N
x_rec_t = x_rec_t.real

# Исходный аналоговый сигнал
x_orig_t = A1*np.cos(2*np.pi*f1*t + np.pi/4) + A2*np.cos(2*np.pi*f2*t + np.pi/8)

fig, axes = plt.subplots(3, 1, figsize=(10, 9))
axes[0].stem(n, x_per, basefmt="k-")
axes[0].set_title("Периодическая последовательность x(n)")
axes[0].set_xlabel("n"); axes[0].grid(True)

axes[1].stem(np.arange(N)/N, np.abs(X), basefmt="k-")
axes[1].set_title("Модуль ДПФ |X(k)|")
axes[1].set_xlabel("k/N"); axes[1].grid(True)

axes[2].plot(t, x_rec_t, 'b-', label="восстановленный x(t)")
axes[2].plot(t, x_orig_t, 'r--', label="исходный x(t)")
axes[2].set_title("Восстановленный и исходный аналоговые сигналы")
axes[2].set_xlabel("t, с"); axes[2].legend(); axes[2].grid(True)
plt.tight_layout(); plt.show()

# Спектр восстановленного (амплитудный)
A_rec = np.abs(np.fft.fft(x_rec_t)) / len(t)
f_rec = np.fft.fftfreq(len(t), d=dt)

plt.figure()
plt.stem(f_rec[:len(f_rec)//2], 2*A_rec[:len(f_rec)//2], basefmt="k-")
plt.title("Амплитудный спектр восстановленного сигнала")
plt.xlabel("f, Гц"); plt.ylabel("Амплитуда"); plt.grid(True)
plt.show()
```

В данном пункте выполняется обратная задача: по дискретным отсчетам ДПФ восстанавливается непрерывный (аналоговый) сигнал \(x(t)\). Это демонстрирует, что ДПФ сохраняет полную информацию о сигнале.

**1. Пояснение к формуле и сдвигу индексов:**
Для восстановления используется формула обратного ряда Фурье:
$$x(t) = \frac{1}{N} \sum_{k=-N/2}^{N/2-1} X_a(k) \exp\left(j \frac{2\pi}{N T} k t\right)$$
Стандартное ДПФ выдает спектр в диапазоне от \(0\) до \(N-1\). Для удобства математического суммирования (чтобы учесть отрицательные частоты) массив сдвигается: индексы от \(N/2\) до \(N-1\) переносятся в начало (становятся отрицательными частотами от \(-N/2\) до \(-1\)). В коде это реализовано функцией `np.roll(X, -half)`.

**2. Связь модуля ДПФ со спектром аналогового сигнала:**
На графике амплитудного спектра восстановленного сигнала (третий график) четко видны два пика.
* Первый пик находится на частоте \(750\) Гц, его амплитуда равна **\(1.07\)**.
* Второй пик на частоте \(1125\) Гц имеет амплитуду **\(2.14\)**.
Эти значения в точности совпадают с амплитудами \(A_1\) и \(A_2\), заданными в исходной формуле сигнала. Таким образом, модуль ДПФ (делённый на \(N\)) напрямую отражает реальные амплитуды гармоник аналогового сигнала.

**3. Результат визуального сравнения:**
На нижнем графике (время от 0 до \(\approx 0.01\) с) наложены два сигнала:
* Синяя сплошная линия — восстановленный по отсчетам ДПФ сигнал \(x(t)\) с шагом дискретизации \(\Delta t = 0.25T\).
* Красная пунктирная линия — исходный аналоговый сигнал, рассчитанный по точной формуле.
Видно, что графики **полностью совпадают**. Это доказывает, что восстановление прошло успешно: по 64 дискретным точкам удалось идеально восстановить непрерывную форму волны между ними. Погрешность отсутствует, так как частота дискретизации (6000 Гц) удовлетворяет теореме Котельникова для наших частот (750 и 1125 Гц).

### Пункт 2.7 Равенство Парсеваля

```python
E_time = np.sum(np.abs(x_per)**2)
E_freq = (1.0/N) * np.sum(np.abs(X)**2)

print(f"Энергия во времени:  Σ|x(n)|²      = {E_time:.6f}")
print(f"Энергия в частоте:  (1/N)·Σ|X(k)|² = {E_freq:.6f}")
print(f"Относительное расхождение: {abs(E_time - E_freq)/E_time:.3e}")
print(f"Равенство Парсеваля выполняется: {np.isclose(E_time, E_freq)}")
```

**Равенство Парсеваля** - это фундаментальный закон сохранения энергии для сигналов. Оно гласит, что полная энергия сигнала, вычисленная во временной области, равна полной энергии сигнала, вычисленной в частотной области. 

Математически для ДПФ это записывается так:
$$\sum_{n=0}^{N-1} |x(n)|^2 = \frac{1}{N} \sum_{k=0}^{N-1} |X(k)|^2$$

Смысл равенства заключается в том, что при переходе от временного представления сигнала к частотному (и наоборот) преобразование Фурье не теряет и не добавляет энергию. Сколько энергии было в исходных отсчетах, столько же ее будет содержаться в спектре.

**Результаты вычислений в Python:**
* Энергия во временной области: $E_{time} = \sum |x(n)|^2 = \mathbf{183.184000}$
* Энергия в частотной области: $E_{freq} = \frac{1}{N} \sum |X(k)|^2 = \mathbf{183.184000}$

**Относительное расхождение** составило $4.655 \cdot 10^{-16}$. Это значение является **машинным нулем** (пределом точности вычислений с плавающей точкой). Такая крошечная разница возникает из-за погрешностей округления компьютера при вычислении экспонент в ДПФ и является абсолютно нормальной.

Функция `np.isclose()` подтвердила, что равенство выполняется (`True`). Это доказывает, что наши спектральные преобразования выполнены корректно, и энергия сигнала сохраняется при переходе в частотную область.
