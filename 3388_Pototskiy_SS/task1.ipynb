## Основные теоретические положения

### 1. Линейные дискретные системы

**Разностное уравнение.** Для причинного рекурсивного звена второго порядка
при $a_0=1$ текущая реакция определяется текущим и прошлыми значениями входа
и прошлыми значениями выхода:

$$y(n)=b_0x(n)+b_1x(n-1)+b_2x(n-2)-a_1y(n-1)-a_2y(n-2).$$

Нулевые начальные условия означают $x(n)=y(n)=0$ при $n<0$.
Записи $x(n)$ и $x[n]$ равнозначны.

**Передаточная функция** — отношение $Z$-изображения реакции к $Z$-изображению
воздействия при нулевых начальных условиях:

$$H(z)=\frac{Y(z)}{X(z)}=
\frac{b_0+b_1z^{-1}+b_2z^{-2}}{1+a_1z^{-1}+a_2z^{-2}}.$$

**Импульсная характеристика** $h(n)$ — реакция на цифровой единичный импульс
$u_0(n)$, равный единице при $n=0$ и нулю при остальных $n$.
Её $Z$-преобразование равно $H(z)$.
Для рассматриваемого рекурсивного звена без сокращения полюсов нулями
характеристика бесконечна; конечный массив содержит только её начальный участок.

**Свёртка.** Реакция причинной линейной стационарной системы определяется формулой

$$y(n)=\sum_{m=0}^{n}h(m)x(n-m).$$

Для двух конечных массивов длины $N_1$ и $N_2$ полная линейная свёртка
содержит $N_1+N_2-1$ элементов, включая возможные нули на краях.

**Нули и полюсы** определяются как корни числителя и знаменателя передаточной
функции после сокращения общих множителей. Для звена второго порядка:

$$b_0z^2+b_1z+b_2=0,\qquad z^2+a_1z+a_2=0.$$

Алгебраическая и показательная формы связаны формулой Эйлера:

$$q=\xi+j\eta=re^{j\varphi},\qquad
r=\sqrt{\xi^2+\eta^2},\quad \varphi=\operatorname{atan2}(\eta,\xi).$$

При вещественных коэффициентах комплексные корни образуют сопряжённые пары.

**Устойчивость.** Причинная система устойчива по ограниченному входу и выходу,
если $\sum_{n=0}^{\infty}|h(n)|<\infty$.
Для причинной рациональной системы это равносильно условию $|p_i|<1$
для всех полюсов передаточной функции.

**Частотные характеристики** получаются подстановкой $z=e^{j\widehat\omega}$:

$$H(e^{j\widehat\omega})=
\frac{\sum_{i=0}^{2}b_i e^{-ji\widehat\omega}}
{\sum_{k=0}^{2}a_k e^{-jk\widehat\omega}}.$$

АЧХ равна $A(\widehat\omega)=|H|$, ФЧХ — $\Phi(\widehat\omega)=\arg H$.
Связь с физической частотой:

$$\widehat\omega=\omega T=\frac{2\pi f}{f_д}.$$

Частотная характеристика периодична с периодом $2\pi$ по $\widehat\omega$,
или $f_д$ по $f$. Полная основная полоса может быть выбрана как
$[-\pi;\pi)$; для вещественных коэффициентов достаточно показать
положительную половину $[0;\pi]$, соответствующую $[0;f_д/2]$ Гц.
Нули около единичной окружности создают провалы АЧХ, полюсы — резонансные
подъёмы. Если АЧХ равна нулю, фаза в этой точке не определена.

### 2. Дискретное преобразование Фурье

**ДПФ** переводит блок из $N$ отсчётов во временной области в $N$ спектральных
коэффициентов:

$$X(k)=\sum_{n=0}^{N-1}x(n)e^{-j2\pi kn/N}.$$

**Обратное ДПФ** восстанавливает исходные отсчёты:

$$x(n)=\frac1N\sum_{k=0}^{N-1}X(k)e^{j2\pi kn/N}.$$

**Шкалы времени.** Дискретное физическое время равно $t_n=nT$,
а дискретное нормированное время — $\widehat t_n=t_n/T=n$.

**Шкалы частоты.** Отсчёт спектра имеет номер $k$. Ему соответствуют

$$\Delta f=\frac{f_д}{N},\qquad f_k=k\Delta f,$$

$$\widehat f_k=\frac{k}{N},\qquad \widehat\omega_k=\frac{2\pi k}{N}.$$

Таким образом, шкала $k$ и шкала нормированных частот связаны линейно.
В стандартном порядке ДПФ $k=0,\ldots,N-1$.
Для $k>N/2$ частоты $kf_д/N$ эквивалентны отрицательным частотам
$(k-N)f_д/N$. На центрированных графиках используются индексы
$-N/2,\ldots,N/2-1$.

**Спектры периодической последовательности.** Коэффициенты дискретного ряда
Фурье равны $c_k=X(k)/N$. Амплитуда двусторонней линии равна $|c_k|$,
её фаза — $\arg c_k=\arg X(k)$.
Для косинуса амплитуды $A$ и начальной фазы $\varphi_0$ получаются линии
высоты $A/2$ на частотах $\pm f_0$ с фазами $\pm\varphi_0$.
Амплитуда косинуса определяется по положительной частоте как $2|X(k_0)|/N$.
Постоянная составляющая и частота Найквиста не удваиваются.

Для вещественной последовательности выполняется сопряжённая симметрия
$X((N-k)\bmod N)=\overline{X(k)}$.
При нулевой амплитуде фаза не определена.
Модуль ДПФ конечной записи связан с амплитудным спектром её периодического
продолжения соотношением

$$|X(k)|=N|c_k|.$$

Если частоты гармоник совпадают с узлами частотной сетки, растекания по
другим отсчётам ДПФ нет.

**Восстановление периодического аналогового сигнала** выполняется по сумме

$$z(t)=\frac1N\sum_{k=-N/2}^{N/2-1}X_a(k)e^{j2\pi kt/(NT)},$$

где $X_a(k)=X(N+k)$ при $k<0$ и $X_a(k)=X(k)$ при $k\geq0$.
В моменты $t=nT$ сумма совпадает с ОДПФ. Для данной суммы косинусоид,
частоты которых ниже $f_д/2$ и совпадают с сеткой ДПФ, восстановление точно
также между узлами.

**Равенство Парсеваля** при выбранной нормировке ДПФ:

$$\sum_{n=0}^{N-1}|x(n)|^2=\frac1N\sum_{k=0}^{N-1}|X(k)|^2.$$

Равенство выражает сохранение энергии блока при переходе между представлениями.
Для бесконечной периодической последовательности рассматриваются энергия
выбранного блока и средняя мощность $P=E/N$.

## Выполнение работы

### Часть 1. Линейная дискретная система

#### Исходные данные и функции расчёта

```python
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline

plt.rcParams.update({
    "figure.figsize": (9, 3.5), "figure.dpi": 115, "font.size": 10,
    "axes.titlesize": 11, "grid.alpha": 0.3,
    "axes.spines.top": False, "axes.spines.right": False,
})

VARIANT = 13
b0 = 0.5 + 0.02*VARIANT
b1 = b0*(-1)**(VARIANT+1)*(0.9822+0.0178*VARIANT)
b2 = b0*(0.8+0.2*(VARIANT % 5))
a0 = 1.0
a1 = (-1)**VARIANT*(0.7778+0.025*VARIANT)
a2 = 0.64+0.006*VARIANT
b = np.array([b0, b1, b2])
a = np.array([a0, a1, a2])
N1 = VARIANT % 10+20
N2 = VARIANT % 10+30
fs1 = 1000*VARIANT
print('Вариант:', VARIANT)
print('b =', b, '; a =', a)
print('N1 =', N1, '; N2 =', N2, '; fд,1 =', fs1, 'Гц')

def difference_response(x, b, a):
    # y(n) = Σ b_i x(n-i) − Σ a_k y(n-k), нулевые начальные условия.
    y = np.zeros(len(x))
    for n in range(len(x)):
        for i in range(len(b)):
            if n >= i:
                y[n] += b[i]*x[n-i]
        for k in range(1, len(a)):
            if n >= k:
                y[n] -= a[k]*y[n-k]
        y[n] /= a[0]
    return y

def linear_convolution(x, h):
    # Полная линейная свёртка по определению.
    y = np.zeros(len(x)+len(h)-1)
    for n in range(len(y)):
        for m in range(len(x)):
            if 0 <= n-m < len(h):
                y[n] += x[m]*h[n-m]
    return y

def stem(ax, xx, yy, title, xlabel='n', ylabel='Амплитуда', color='C0'):
    markers, lines, _ = ax.stem(xx, yy, basefmt=' ')
    plt.setp(markers, color=color, markersize=4)
    plt.setp(lines, color=color, linewidth=1)
    ax.axhline(0, color='gray', lw=0.6)
    ax.set(title=title, xlabel=xlabel, ylabel=ylabel)
    ax.grid(True)
```

#### Пункт 1.1. Импульсная характеристика

Для варианта 13 передаточная функция равна

$$H(z)=\frac{0{,}76+0{,}922336z^{-1}+1{,}064z^{-2}}
{1-1{,}1028z^{-1}+0{,}718z^{-2}}.$$

При $x(n)=u_0(n)$ и нулевых начальных условиях

$$h(n)=0{,}76u_0(n)+0{,}922336u_0(n-1)+1{,}064u_0(n-2)$$
$$\phantom{h(n)=}+1{,}1028h(n-1)-0{,}718h(n-2).$$

По табличной паре $Z$-преобразования из теоретической части:

$$r=\sqrt{0{,}718},\qquad \varphi=\arccos\frac{1{,}1028}{2r},$$

$$v(n)=r^n\frac{\sin((n+1)\varphi)}{\sin\varphi}\,u_1(n),$$

где $u_1(n)=1$ при $n\geq0$ и 0 при $n<0$.
Аналитическая формула импульсной характеристики:

$$h(n)=0{,}76v(n)+0{,}922336v(n-1)+1{,}064v(n-2).$$

Задержанные слагаемые принимаются равными нулю до момента включения;
$h(-1)=h(-2)=0$. Первые значения:
$h(0)=0{,}76$, $h(1)=1{,}760464$, $h(2)=2{,}4597596992$.

```python
impulse = np.r_[1.0, np.zeros(N1-1)]
h = difference_response(impulse, b, a)
r = np.sqrt(a2)
theta = np.arccos(-a1/(2*r))

def v(n):
    n = np.asarray(n)
    values = np.zeros(n.shape)
    active = n >= 0
    values[active] = r**n[active]*np.sin((n[active]+1)*theta)/np.sin(theta)
    return values

nh = np.arange(N1)
h_analytic = sum(b[i]*v(nh-i) for i in range(3))
h_formula_error = np.max(np.abs(h-h_analytic))
assert np.allclose(h, h_analytic, atol=1e-12, rtol=0)
print(f'r = {r:.9f}; φ/π = {theta/np.pi:.9f}')
print(f'Ошибка аналитической формулы: {h_formula_error:.3e}')
print('Первые шесть отсчётов h(n):', h[:6])

fig, ax = plt.subplots()
stem(ax, nh, h, 'Рисунок 1. Первые 23 отсчёта импульсной характеристики', ylabel='h(n)')
ax.plot(nh, h_analytic, 'x', color='C1', label='Аналитическая формула')
ax.legend()
fig.tight_layout()
plt.show()
```

Характеристика представляет собой затухающие колебания. Их огибающая определяется
степенью $r<1$, а частота колебаний — аргументом полюсов $\theta$.
**Истинная длина $h[n]$ бесконечна**: рекурсивная часть продолжает формировать
отклик после прекращения входа. Значение $N_1=23$ задаёт только длину вычисленного
начального участка; последний показанный отсчёт не означает завершения реакции.

#### Пункт 1.2. Реакция по формуле свёртки

Прямоугольное воздействие задаётся на $N_2=33$ отсчётах:

$$x(n)=\begin{cases}1,&0\leq n<M,\\0,&M\leq n\leq32.\end{cases}$$

В условии $int(t)$ обозначает округление до ближайшего целого.
Для положительных полуцелых значений принято правило «0,5 вверх»:

$$M=\left\lfloor\frac{N_2}{2}+0{,}5\right\rfloor
=\lfloor17\rfloor=17.$$

Это правило реализовано явно; преобразование Python `int(N2/2)` само по себе
означает усечение и дало бы другое значение.
Длина массива воздействия равна 33, число единичных отсчётов — 17,
их индексы — от 0 до 16. За пределами массива воздействие равно нулю.
Реакция $y_1$ вычисляется по формуле линейной свёртки с первыми 23 отсчётами $h(n)$.

```python
M = int(np.floor(N2/2+0.5))
nx = np.arange(N2)
x = (nx < M).astype(float)
y1_full = linear_convolution(x, h)
assert np.allclose(y1_full, np.convolve(x, h), atol=1e-12, rtol=0)
y1 = y1_full[:N2]
conv_nonzero = np.flatnonzero(np.abs(y1_full) > 1e-12)
print(f"Длина массива x: {len(x)}; число единичных отсчётов: {M}")
print(f"Длина полной свёртки: {len(y1_full)} = {N2}+{N1}-1")
print(f"Последний ненулевой индекс свёртки: {conv_nonzero[-1]}")
print(f"Длина сравниваемого начального участка y1: {len(y1)}")

fig, axs = plt.subplots(3, 1, figsize=(10, 8.5))
stem(axs[0], nx, x, "Рисунок 2а. Прямоугольное воздействие",
     "n, отсчёт", "x[n]")
stem(axs[1], np.arange(y1_full.size), y1_full,
     "Рисунок 2б. Полная свёртка с h₂₃[n] — 55 элементов",
     "n, отсчёт", "y₁,full[n]")
axs[1].axvline(N2-0.5, color="C3", ls="--", label="Граница участка из 33 отсчётов")
axs[1].legend()
stem(axs[2], nx, y1, "Рисунок 2в. Первые 33 отсчёта свёртки",
     "n, отсчёт", "y₁[n]")
fig.tight_layout()
plt.show()
```

**Длины последовательностей.** Полная свёртка массивов содержит
$N_2+N_1-1=33+23-1=55$ элементов с индексами от 0 до 54.
Из-за нулевого хвоста воздействия ненулевая часть занимает только
$17+23-1=39$ позиций с индексами от 0 до 38. Последние 16 элементов равны нулю.
Рисунок 2б показывает как полную длину массива, так и его ненулевой участок.

**Ограничение до длины воздействия.** Первые $N_2=33$ значения выбираются
для сопоставления с реакцией по разностному уравнению на том же интервале.
Это выбор окна наблюдения: выход БИХ-системы продолжает изменяться и после
окончания воздействия. Усечение массива результата и усечение самой
импульсной характеристики — разные операции.

**Вид реакции.** Пока на вход подаются единицы, выход равен накопленной сумме
импульсной характеристики. На включении и выключении импульса возникают
затухающие колебания. Хвост свёртки с усечённой $h(n)$ представляет приближение
к истинному отклику системы.

#### Пункт 1.3. Реакция по разностному уравнению

Для заданных коэффициентов разностное уравнение имеет вид

$$y_2(n)=0{,}76x(n)+0{,}922336x(n-1)+1{,}064x(n-2)$$
$$\phantom{y_2(n)=}+1{,}1028y_2(n-1)-0{,}718y_2(n-2).$$

При вычислении отрицательные индексы не используются: соответствующие
значения входа и реакции равны нулю. Оба способа сравниваются на первых
$N_2$ отсчётах.

```python
y2 = difference_response(x, b, a)
truncation_error = np.max(np.abs(y1-y2))

assert np.allclose(y1[:N1], y2[:N1], atol=1e-12, rtol=0)
print(f"max |y1(h23) − y2| = {truncation_error:.12f}")

fig, axs = plt.subplots(3, 1, figsize=(10, 8.5))
stem(axs[0], nx, x, "Рисунок 3а. Воздействие для разностного уравнения",
     "n, отсчёт", "x[n]")
axs[1].plot(nx, y2, "o-", ms=4, label="y₂: разностное уравнение")
axs[1].plot(nx, y1, "s--", ms=3, label="y₁: свёртка с h₂₃")
axs[1].axvline(N1-0.5, color="0.5", ls=":")
axs[1].set(title="Рисунок 3б. Сравнение реакций", xlabel="n, отсчёт", ylabel="y[n]")
axs[1].grid(True)
axs[1].legend()
stem(axs[2], nx, y1-y2, "Рисунок 3в. Ошибка усечения характеристики до 23 отсчётов",
     "n, отсчёт", "y₁[n] − y₂[n]", color="C3")
fig.tight_layout()
plt.show()
```

Реакции совпадают до $n=22$ включительно: для этих индексов все необходимые
отсчёты характеристики входят в $h_{23}$. При $n\geq23$ в $y_1$ отсутствуют

#### Пункт 1.4. Нули и полюсы передаточной функции

Умножение числителя и знаменателя на $z^2$ даёт

$$H(z)=\frac{0{,}76z^2+0{,}922336z+1{,}064}{z^2-1{,}1028z+0{,}718}.$$

Для перехода от алгебраической формы $q=\alpha+j\beta$ к показательной
используются $|q|=\sqrt{\alpha^2+\beta^2}$ и $\arg q=\operatorname{atan2}(\beta,\alpha)$.

```python
zeros = np.roots(b)
poles = np.roots(a)
gain = b0/a0
for label, roots in [('Нуль', zeros), ('Полюс', poles)]:
    for i, root in enumerate(roots, 1):
        print(f'{label} {i}: {root.real:.9f}{root.imag:+.9f}j; '
              f'{abs(root):.9f} exp(jπ·{np.angle(root)/np.pi:+.9f})')
assert np.allclose(np.poly(zeros)*gain, b, atol=1e-12, rtol=0)
assert np.allclose(np.poly(poles), a, atol=1e-12, rtol=0)
```

Получены пары:

$$z_{1,2}=-0{,}6068\pm j1{,}015772494
=1{,}183215957e^{\pm j0{,}671406444\pi},$$

$$p_{1,2}=0{,}5514\pm j0{,}643395710
=0{,}847348807e^{\pm j0{,}274460650\pi}.$$

Числа в каждой паре комплексно сопряжены: их модули одинаковы, а аргументы
противоположны. Корни числителя и знаменателя между собой не совпадают.
Факторизация через показательные формы:

$$H(z)=0{,}76\frac{
(z-1{,}183215957e^{j0{,}671406444\pi})
(z-1{,}183215957e^{-j0{,}671406444\pi})}{
(z-0{,}847348807e^{j0{,}274460650\pi})
(z-0{,}847348807e^{-j0{,}274460650\pi})}.$$

Для сохранения точности вычисления используют исходные коэффициенты,
а не округлённые значения корней из этой записи.

Эквивалентная форма через задержки:

$$H(z)=0{,}76\frac{(1-z_1z^{-1})(1-z_2z^{-1})}
{(1-p_1z^{-1})(1-p_2z^{-1})},$$

где $z_i$ и $p_i$ подставляются в показательной форме, приведённой выше.

#### Пункт 1.6. АЧХ и ФЧХ в нормированных частотах

Частотная характеристика вычисляется по формуле

$$H(e^{j\widehat\omega})=
\frac{0{,}76+0{,}922336e^{-j\widehat\omega}+1{,}064e^{-j2\widehat\omega}}
{1-1{,}1028e^{-j\widehat\omega}+0{,}718e^{-j2\widehat\omega}}.$$

Графики построены для положительной половины основной полосы:
$0\leq\widehat\omega\leq\pi$. По горизонтальной оси отложено
$\widehat\omega/\pi$. На АЧХ отмечены аргументы полюса и нуля;
ФЧХ показана в главных значениях и с развёрткой фазы.

```python
w = np.linspace(0, np.pi, 65537)
numerator = sum(b[i]*np.exp(-1j*w*i) for i in range(3))
denominator = sum(a[k]*np.exp(-1j*w*k) for k in range(3))
H = numerator/denominator
magnitude = np.abs(H)
phase_principal = np.angle(H)

# Развёртка устраняет переходы между ветвями аргумента.
phase_unwrapped = np.unwrap(phase_principal)
center = np.argmin(np.abs(w))
phase_unwrapped -= 2*np.pi*np.round(phase_unwrapped[center]/(2*np.pi))
positive = np.flatnonzero(w >= 0)
peak_idx = positive[np.argmax(magnitude[positive])]
notch_idx = positive[np.argmin(magnitude[positive])]

fig, axs = plt.subplots(3, 1, figsize=(9, 8), sharex=True)
axs[0].plot(w/np.pi, magnitude)
axs[0].axvline(theta/np.pi, color='C3', ls=':', label='Аргумент полюса')
axs[0].axvline(abs(np.angle(zeros[0]))/np.pi, color='C2', ls=':', label='Аргумент нуля')
axs[0].set(title='Рисунок 5а. АЧХ', ylabel='|H|')
axs[0].legend()
axs[1].plot(w/np.pi, phase_principal)
axs[1].set(title='Рисунок 5б. Главное значение ФЧХ', ylabel='arg H, рад')
axs[2].plot(w/np.pi, phase_unwrapped)
axs[2].set(title='Рисунок 5в. Развёрнутая ФЧХ', ylabel='Фаза, рад',
           xlabel='Нормированная угловая частота ω̂ / π')
for ax in axs:
    ax.set_xlim(0, 1)
    ax.grid(True)
fig.tight_layout()
plt.show()

H_dc = np.sum(b)/np.sum(a)
H_nyquist = (b0-b1+b2)/(a0-a1+a2)
print(f'|H(0)| = {abs(H_dc):.9f}; |H(π)| = {abs(H_nyquist):.9f}')
print(f'Максимум при ω̂/π ≈ {w[peak_idx]/np.pi:.6f}: {magnitude[peak_idx]:.6f}')
print(f'Минимум при ω̂/π ≈ {w[notch_idx]/np.pi:.6f}: {magnitude[notch_idx]:.6f}')
```

**Границы полосы.** Положительная половина основной полосы равна $[0;\pi]$.
Полную основную полосу можно выбрать как $[-\pi;\pi)$.
При вещественных коэффициентах АЧХ чётная, поэтому положительная половина
содержит всю независимую информацию о модуле частотной характеристики.

**Карта нулей и полюсов и вид АЧХ.** Для точки $z=e^{j\widehat\omega}$
на единичной окружности модуль передаточной функции определяется расстояниями
до нулей и полюсов:

$$A(\widehat\omega)=0{,}76\frac{|e^{j\widehat\omega}-z_1|
|e^{j\widehat\omega}-z_2|}{|e^{j\widehat\omega}-p_1|
|e^{j\widehat\omega}-p_2|}.$$

Полюс с аргументом $0{,}27446\pi$ объясняет резонансный подъём в близкой

Экстремумы зависят от всех нулей и полюсов, поэтому их частоты не обязаны

### Часть 2. Дискретное преобразование Фурье

#### Исходные данные

$N=64$, $f_{д,2}=8000$ Гц, $T=0{,}000125$ с,
$A_1=1{,}13$, $A_2=2{,}26$, $f_1=1000$ Гц, $f_2=1500$ Гц.

#### Пункт 2.1. Периодическая последовательность

$$x(nT)=A_1\cos(2\pi f_1nT+\pi/4)+A_2\cos(2\pi f_2nT+\pi/8).$$

Для варианта 13:

$$x(n)=1{,}13\cos(\pi n/4+\pi/4)+2{,}26\cos(3\pi n/8+\pi/8).$$

Графики строятся в шкале дискретного времени $t_n=nT$ и дискретного
нормированного времени $\widehat t_n=t_n/T=n$.
Заданный период равен $NT=0{,}008$ с, последний отсчёт имеет время
$(N-1)T=0{,}007875$ с.
Минимальные периоды гармоник равны 8 и 16 отсчётам; минимальный период
суммы — 16. Блок из 64 отсчётов содержит четыре минимальных периода.

```python
N = 64
fs2 = 2000*(VARIANT % 5+1)
T = 1/fs2
A1 = 1+0.01*VARIANT
A2 = 2*A1
f1 = fs2/8
f2 = 1.5*f1

def x_cont(t):
    return (A1*np.cos(2*np.pi*f1*t+np.pi/4)
            + A2*np.cos(2*np.pi*f2*t+np.pi/8))

n = np.arange(N)
tn = n*T
x_period = x_cont(tn)
assert np.allclose(x_period[16:], x_period[:-16], atol=1e-12, rtol=0)
print(f'fд,2={fs2} Гц; T={T} с; Δf={fs2/N} Гц')
print(f'A1={A1}; A2={A2}; f1={f1} Гц; f2={f2} Гц')

fig, axs = plt.subplots(2, 1, figsize=(9, 6))
stem(axs[0], tn*1000, x_period, 'Рисунок 7а. Дискретное физическое время',
     'tₙ = nT, мс', 'x(n)')
stem(axs[1], n, x_period, 'Рисунок 7б. Дискретное нормированное время',
     'tₙ / T = n', 'x(n)')
fig.tight_layout()
plt.show()
```

В обоих графиках показаны одинаковые 64 значения. Шаг физического времени
равен 0,125 мс, шаг нормированного времени — единице.
Связь шкал: $t_n=nT$; обратный переход получается делением на $T$.
Номера гармоник равны $k_1=f_1N/f_{д,2}=8$ и $k_2=f_2N/f_{д,2}=12$.

#### Пункт 2.2. ОДПФ от ДПФ последовательности

```python
X = np.fft.fft(x_period)
x_idft = np.fft.ifft(X)
idft_error = np.max(np.abs(x_idft-x_period))
assert np.allclose(x_idft, x_period, atol=1e-12, rtol=0)
print(f"max |ОДПФ(ДПФ(x)) − x| = {idft_error:.3e}")
print(f"Максимальная мнимая часть ОДПФ: {np.max(np.abs(x_idft.imag)):.3e}")

fig, axs = plt.subplots(2, 1, figsize=(9, 6), sharex=True)
stem(axs[0], n, x_idft.real, "Рисунок 8а. Последовательность после ОДПФ",
     "n = tₙ/T", "xОДПФ[n]")
axs[0].plot(n, x_period, "x", color="C1", ms=5, label="Исходные отсчёты")
axs[0].legend()
stem(axs[1], n, x_idft.real-x_period, "Рисунок 8б. Ошибка восстановления отсчётов",
     "n = tₙ/T", "Ошибка", color="C3")
axs[1].ticklabel_format(axis="y", style="sci", scilimits=(0, 0))
fig.tight_layout()
plt.show()
```

ОДПФ восстанавливает все исходные отсчёты. Остаточная ошибка порядка
$10^{-15}$ связана с арифметикой с плавающей точкой. Мнимая часть также имеет
порядок машинного округления и не представляет отдельную компоненту сигнала.

#### Пункт 2.3. Амплитудный и фазовый спектры

Фазы практически нулевых коэффициентов не показываются, поскольку

#### Пункт 2.4. ДПФ конечной последовательности и спектр периодической

Тот же блок теперь рассматривается как конечная последовательность, равная нулю
вне $0\leq n<64$. ДПФ вычисляется непосредственно по сумме определения.
Это позволяет сопоставить прямое вычисление с FFT и проверить нормировку.

```python
k = np.arange(N)
dft_matrix = np.exp(-2j*np.pi*np.outer(k, n)/N)
X_finite = dft_matrix @ x_period
direct_dft_error = np.max(np.abs(X_finite-X))
assert np.allclose(X_finite, X, atol=1e-10, rtol=0)
print(f"max |ДПФ по определению − FFT| = {direct_dft_error:.3e}")
print(f"max |X| на 1000 Гц: {abs(X[8]):.6f} = {N} × {abs(X[8])/N:.6f}")
print(f"max |X| на 1500 Гц: {abs(X[12]):.6f} = {N} × {abs(X[12])/N:.6f}")

fig, axs = plt.subplots(2, 1, figsize=(10, 6.5), sharex=True)
stem(axs[0], k_fft, np.abs(X_finite),
     "Рисунок 10а. Модуль ДПФ конечной последовательности", "Номер гармоники k", "|Xкон[k]|")
stem(axs[1], k_fft, amplitude_fft,
     "Рисунок 10б. Амплитудный спектр периодического продолжения", "Номер гармоники k", "|C[k]| = |X[k]|/N")
for ax in axs:
    ax.set_xlim(0, N)
fig.tight_layout()
plt.show()
```

Для совпадающих номеров гармоник модуль ДПФ конечного блока в $N=64$ раза больше
амплитуды соответствующей двусторонней линии периодического сигнала:

$$|X_{\mathrm{кон}}[k]|=N|C[k]|.$$

Поэтому значения 36,16 и 72,32 переходят в 0,565 и 1,13 после деления на 64.
ДПФ обоих представлений блока численно одинакова; отличается интерпретация.
У конечного блока непрерывная по частоте ДВПФ, а показанная ДПФ содержит только
её 64 равномерных отсчёта. У периодического продолжения спектр линейчатый.

#### Пункт 2.5. Восстановление периодического аналогового сигнала

Коэффициенты переставляются в порядок $k=-32,\ldots,31$ функцией `fftshift`.
Сумма восстановления вычисляется непосредственно, без интерполяционных функций.
Время задаётся на отрезке $[0;(N-1)T]$ с шагом $\Delta t=T/4=31{,}25$ мкс,
поэтому сетка содержит 253 точки и включает оба конца отрезка.

Исходный аналоговый сигнал для сравнения:

$$x_{\mathrm{исх}}(t)=1{,}13\cos(2\pi1000t+\pi/4)
+2{,}26\cos(2\pi1500t+\pi/8).$$

Амплитудный спектр восстановленного аналогового сигнала показывается как
веса линий $|X_a[k]|/N$ в герцах, а не как повторная ДПФ усечённой временной сетки.

```python
t_dense = np.arange(4*(N-1)+1)*T/4
synthesis = np.exp(2j*np.pi*np.outer(k_center, t_dense)/(N*T))
x_analog_complex = (X_center @ synthesis)/N
x_analog = x_analog_complex.real
x_original = (A1*np.cos(2*np.pi*f1*t_dense + np.pi/4)
              + A2*np.cos(2*np.pi*f2*t_dense + np.pi/8))
analog_error = np.max(np.abs(x_analog-x_original))
sample_error = np.max(np.abs(x_analog[::4]-x_period))
imag_residual = np.max(np.abs(x_analog_complex.imag))
assert t_dense.size == 253
assert np.isclose(t_dense[-1], (N-1)*T)
assert np.allclose(x_analog, x_original, atol=1e-11, rtol=0)
assert np.allclose(x_analog[::4], x_period, atol=1e-11, rtol=0)
print(f"Число точек: {t_dense.size}; Δt={T/4:.8f} с")
print(f"Максимальная ошибка аналогового восстановления: {analog_error:.3e}")
print(f"Ошибка в исходных отсчётах: {sample_error:.3e}")
print(f"Максимальная мнимая часть: {imag_residual:.3e}")

fig, axs = plt.subplots(3, 2, figsize=(12, 10))
stem(axs[0, 0], tn*1000, x_period, "Рисунок 11а. Периодическая последовательность",
     "tₙ, мс", "x[n]")
stem(axs[0, 1], f, np.abs(X_center), "Рисунок 11б. Модуль её ДПФ", "f, Гц", "|X[k]|")
axs[1, 0].plot(t_dense*1000, x_analog, label="Сумма по коэффициентам ДПФ")
axs[1, 0].plot(tn*1000, x_period, "o", ms=3, label="Исходные отсчёты")
axs[1, 0].set(title="Рисунок 11в. Восстановленный аналоговый сигнал", xlabel="t, мс", ylabel="xвосст(t)")
axs[1, 0].legend(fontsize=8)
stem(axs[1, 1], f[significant], amplitude[significant],
     "Рисунок 11г. Амплитудный спектр аналогового сигнала", "f, Гц", "Вес линии |C[k]|")
axs[2, 0].plot(t_dense*1000, x_original, color="C2", label="Исходная формула")
axs[2, 0].plot(t_dense*1000, x_analog, "--", color="C1", label="Восстановление")
axs[2, 0].set(title="Рисунок 11д. Исходный аналоговый сигнал и сравнение", xlabel="t, мс", ylabel="x(t)")
axs[2, 0].legend(fontsize=8)
axs[2, 1].plot(t_dense*1000, x_analog-x_original, color="C3")
axs[2, 1].set(title="Рисунок 11е. Ошибка аналогового восстановления", xlabel="t, мс", ylabel="Ошибка")
axs[2, 1].ticklabel_format(axis="y", style="sci", scilimits=(0, 0))
axs[0, 1].set_xlim(-fs2/2, fs2/2)
axs[1, 1].set_xlim(-fs2/2, fs2/2)
for ax in axs.flat:
    ax.grid(True)
fig.tight_layout()
plt.show()
```

#### Пункт 2.6. Интерпретация восстановления

ДПФ задаёт частоты аналоговых спектральных линий по правилу $f_k=k/(NT)$.
Комплексный вес линии равен $C[k]=X_a[k]/N$; её амплитуда — $|C[k]|$,
а фаза — $\arg C[k]$. Это объясняет различие вертикальных шкал рисунков 11б и 11г.
В строгом смысле спектр периодического аналогового сигнала является суммой
дельта-линий. Высота маркера на рисунке обозначает вес линии, а не конечную
высоту дельта-функции.

Восстановленный и исходный сигналы визуально совпадают. Их частоты 1000 и 1500 Гц
ниже частоты Найквиста 4000 Гц и точно представлены сеткой ДПФ с шагом 125 Гц.
Алиасинг отсутствует. На четырёхкратно более плотной временной сетке ошибка имеет
порядок $10^{-14}$, а исходные отсчёты воспроизводятся с машинной точностью.
В частности, восстановление между отсчётами проверено сопоставлением с исходной
аналитической формулой, а не только совпадением в узлах сетки.

#### Пункт 2.7. Равенство Парсеваля

При ненормированном прямом ДПФ проверяется соотношение

$$E_t=\sum_{n=0}^{N-1}|x(n)|^2,\qquad E_f=\frac1N\sum_{k=0}^{N-1}|X(k)|^2.$$

На блоке из 64 отсчётов гармоники ортогональны: сумма их произведений равна нулю,
а сумма квадратов каждого косинуса — $N/2$. Поэтому

$$E_t=\frac{N}{2}(A_1^2+A_2^2)=32(1{,}13^2+2{,}26^2)=204{,}304.$$

В спектре находятся две пары линий с модулями $NA_1/2$ и $NA_2/2$:

$$E_f=\frac2N\left[\left(\frac{NA_1}{2}\right)^2+
\left(\frac{NA_2}{2}\right)^2\right]=204{,}304.$$

Средняя мощность равна $P=E/N=(A_1^2+A_2^2)/2$.

```python
energy_time = float(np.sum(np.abs(x_period)**2))
energy_freq = float(np.sum(np.abs(X)**2)/N)
parseval_error = abs(energy_time-energy_freq)
power_time = energy_time/N
power_theory = (A1**2+A2**2)/2
assert np.isclose(energy_time, energy_freq, atol=1e-10, rtol=0)
assert np.isclose(power_time, power_theory, atol=1e-12, rtol=0)

print(f'Энергия по отсчётам: {energy_time:.12f}')
print(f'Энергия по ДПФ:     {energy_freq:.12f}')
print(f'Невязка Парсеваля:  {parseval_error:.3e}')
print(f'Средняя мощность:   {power_time:.12f}')

results = {
    "variant": VARIANT,
    "b": b.tolist(), "a": a.tolist(), "N1": N1, "N2": N2,
    "fs1": fs1, "N": N, "fs2": fs2, "input_nonzero_length": M,
    "h_formula_error": float(h_formula_error),
    "truncated_convolution_error": float(truncation_error),
    "stable": stable, "idft_error": float(idft_error),
    "direct_dft_error": float(direct_dft_error),
    "analog_error": float(analog_error), "analog_imag_residual": float(imag_residual),
    "energy_time": energy_time, "energy_frequency": energy_freq,
    "parseval_error": float(parseval_error), "average_power": power_time,
}
print("Все численные проверки выполнены успешно.")
```

Получено равенство $204{,}304\approx204{,}304$ с невязкой порядка машинного округления.
Средняя мощность равна $3{,}19225$ и совпадает с $(A_1^2+A_2^2)/2$.
ДПФ перераспределяет представление сигнала по частотным составляющим,
сохраняя энергию при учёте нормирующего множителя $1/N$.
