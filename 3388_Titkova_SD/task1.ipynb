## Основные теоретические положения

### 1. Линейные дискретные системы

Линейная дискретная система (ЛДС) описывается разностным уравнением, которое связывает текущий отсчёт реакции $y(n)$ с текущим и прошлыми отсчётами воздействия $x(n)$ и прошлыми отсчётами реакции. В общем виде:
$$y(n) = \sum_{i=0}^{N-1} b_i x(n-i) - \sum_{k=1}^{M-1} a_k y(n-k)$$

Для рекурсивного звена 2-го порядка разностное уравнение имеет вид:
$$y(n) = b_0 x(n) + b_1 x(n-1) + b_2 x(n-2) - a_1 y(n-1) - a_2 y(n-2)$$

Применяя Z-преобразование при нулевых начальных условиях, получаем передаточную функцию (ПФ) — отношение Z-изображений реакции и воздействия:
$$H(z) = \frac{Y(z)}{X(z)} = \frac{b_0 + b_1 z^{-1} + b_2 z^{-2}}{1 + a_1 z^{-1} + a_2 z^{-2}}$$

**Импульсная характеристика** $h(n)$ — это реакция системы на цифровой единичный импульс $u_0(n)$ при нулевых начальных условиях, $H(z) = Z\{h(n)\}$. Для рекурсивных (БИХ) систем она бесконечна.

**Свёртка** позволяет найти реакцию системы на произвольное воздействие:
$$y(n) = \sum_{m=0}^{n} h(m)\, x(n-m)$$
Длина линейной свёртки двух последовательностей длиной $N_1$ и $N_2$ равна $N_1 + N_2 - 1$.

**Нули и полюсы** — это корни числителя и знаменателя передаточной функции соответственно. Они могут быть представлены в алгебраической ($z = \xi + j\eta$) и показательной ($z = r e^{j\varphi}$) формах, где $r = \sqrt{\xi^2 + \eta^2}$, $\xi = r\cos\varphi$, $\eta = r\sin\varphi$.

**Устойчивость.** ЛДС устойчива, если при ограниченном воздействии реакция также ограничена, что равносильно условию $\sum_{n=0}^{\infty} |h(n)| < \infty$. Для этого необходимо и достаточно, чтобы все полюсы передаточной функции лежали строго внутри единичной окружности ($|z_*| < 1$).

**Частотные характеристики** (АЧХ и ФЧХ) определяются подстановкой $z = e^{j\hat{\omega}}$ в передаточную функцию:
$$H(e^{j\hat\omega}) = |H(e^{j\hat\omega})|\, e^{j \arg H(e^{j\hat\omega})}$$
АЧХ $A(\hat\omega) = |H(e^{j\hat\omega})|$ показывает изменение амплитуды на разных частотах, ФЧХ $\varphi(\hat\omega) = \arg H(e^{j\hat\omega})$ — фазовый сдвиг. Нормированная частота связана с частотой $f$ (Гц) соотношением $\hat\omega = \omega T = 2\pi f / f_d$, где $f_d$ — частота дискретизации. Частотные характеристики периодичны с периодом $2\pi$ ($f_d$), основная полоса частот: $\hat\omega \in [0;\ \pi]$, $f \in [0;\ f_d/2]$.

### 2. Дискретное преобразование Фурье (ДПФ)

**ДПФ** применяется для анализа частотного состава периодических последовательностей и последовательностей конечной длины $N$. Прямое ДПФ переводит сигнал из временной области в частотную:
$$X(k) = \sum_{n=0}^{N-1} x(n) e^{-j \frac{2\pi}{N} kn}, \quad k = 0, 1, \dots, N-1$$

**Обратное ДПФ (ОДПФ)** восстанавливает сигнал по его спектру:
$$x(n) = \frac{1}{N} \sum_{k=0}^{N-1} X(k) e^{j \frac{2\pi}{N} kn}, \quad n = 0, 1, \dots, N-1$$

**Шкалы времени и частоты.** Дискретное время — $nT$, дискретное нормированное время — $\hat t = nT/T = n$. Дискретная нормированная частота — номер отсчёта $k$; ему соответствуют частота $f_k = k \frac{f_d}{N}$ (Гц) и нормированная частота $\hat\omega_k = \frac{2\pi k}{N}$.

**Спектры периодической последовательности.** Периодическая последовательность с периодом $N$ представляется дискретным рядом Фурье с коэффициентами
$$c_k = \frac{1}{N} \sum_{n=0}^{N-1} x(n) e^{-j \frac{2\pi}{N} kn} = \frac{X(k)}{N}$$
* **Амплитудный спектр** $|c_k|$ показывает амплитуды гармоник;
* **Фазовый спектр** $\arg c_k = \arg X(k)$ показывает их начальные фазы.

Для вещественной последовательности $|X(k)| = |X(N-k)|$, $\arg X(k) = -\arg X(N-k)$. Модуль ДПФ конечной последовательности связан с амплитудным спектром её периодического продолжения соотношением $|X(k)| = N |c_k|$. Если на периоде $N$ укладывается целое число периодов гармоник, спектр не испытывает «растекания».

**Восстановление аналогового сигнала** по отсчётам ДПФ выполняется с помощью ряда Фурье. Согласно теореме Котельникова, если частота дискретизации больше удвоенной максимальной частоты сигнала, непрерывный сигнал может быть восстановлен по отсчётам без потерь.

**Равенство Парсеваля** отражает закон сохранения энергии: энергия сигнала во временной области равна энергии в частотной области:
$$\sum_{n=0}^{N-1} |x(n)|^2 = \frac{1}{N} \sum_{k=0}^{N-1} |X(k)|^2$$

### Пункт 1.1 Импульсная характеристика

```python
n_h = np.arange(N1)
delta = np.zeros(N1); delta[0] = 1.0

h = np.zeros(N1)
for n in range(N1):
    h[n] = b0 * delta[n]
    if n >= 1: h[n] += b1 * delta[n-1] - a1 * h[n-1]
    if n >= 2: h[n] += b2 * delta[n-2] - a2 * h[n-2]


plt.figure()
plt.stem(n_h, h, basefmt="k-")
plt.xlabel("n"); plt.ylabel("h(n)")
plt.title(f"Импульсная характеристика h(n), N1 = {N1}")
plt.grid(True)
plt.show()

print("Первые 10 отсчётов h(n):")
for i in range(10):
    print(f"h({i}) = {h[i]:.6f}")
```

Рассчитана импульсная характеристика $h(n)$ длины $N_1 = 28$ для заданных коэффициентов $b = [0.86,\ -1.120236,\ 1.204]$ и $a = [1,\ 1.2278,\ 0.748]$. Отсчёты вычислены по разностному уравнению при $x(n) = u_0(n)$ и нулевых начальных условиях $h(-1) = h(-2) = 0$:

ИХ имеет вид затухающих колебаний, знаки соседних отсчётов почти чередуются. Затухание определяется модулем полюсов $r = 0.8649 < 1$, а высокая частота колебаний — аргументом полюсов $\varphi \approx 0.75\pi$ (период около $2\pi/\varphi \approx 2.7$ отсчёта).

**Аналитическая формула:**
Полюсы звена комплексно-сопряжённые, $z_{*1,2} = r e^{\pm j\varphi}$, поэтому знаменатель ПФ можно записать как
$$1 + a_1 z^{-1} + a_2 z^{-2} = 1 - 2r\cos\varphi\, z^{-1} + r^2 z^{-2}, \qquad r = \sqrt{a_2} = 0.8649, \quad \cos\varphi = -\frac{a_1}{2r} \Rightarrow \varphi = 0.7512\pi.$$

По таблице соответствий $Z^{-1}\left\{\frac{1}{1 + a_1 z^{-1} + a_2 z^{-2}}\right\} = r^n \frac{\sin((n+1)\varphi)}{\sin\varphi}$, а по теореме о задержке множитель $z^{-i}$ сдвигает оригинал на $i$ отсчётов. С учётом нулевых начальных условий:
$$h(n) = b_0 r^n \frac{\sin((n+1)\varphi)}{\sin\varphi} + b_1 r^{n-1} \frac{\sin(n\varphi)}{\sin\varphi} u_1(n-1) + b_2 r^{n-2} \frac{\sin((n-1)\varphi)}{\sin\varphi} u_1(n-2), \quad n \ge 0,$$
где $u_1(n)$ — цифровой единичный скачок. Для варианта 18:
$$h(n) = 0.86 \cdot 0.8649^{n} \frac{\sin(0.7512\pi(n+1))}{\sin 0.7512\pi} - 1.1202 \cdot 0.8649^{n-1} \frac{\sin(0.7512\pi n)}{\sin 0.7512\pi} u_1(n-1) + 1.204 \cdot 0.8649^{n-2} \frac{\sin(0.7512\pi(n-1))}{\sin 0.7512\pi} u_1(n-2)$$

Формула совпадает с расчётом по разностному уравнению (расхождение порядка $10^{-15}$).

В действительности длина импульсной характеристики бесконечна: звено рекурсивное (БИХ-система), каждый отсчёт $h(n)$ вычисляется через предыдущие, и $h(n) \ne 0$ при сколь угодно большом $n$. Так как система устойчива ($r = 0.8649 < 1$), отсчёты затухают, но в ноль не обращаются. Длина $N_1 = 28$ — это лишь длина вычисленного фрагмента: к $n = 27$ огибающая уменьшается до $0.8649^{27} \approx 0.02$ от начального значения, поэтому фрагмент хорошо, но не полностью описывает систему.

После окончания импульса ($n \ge 19$) второй фронт снова вызывает колебания, которые затем затухают.

### Пункт 1.3 Реакция по разностному уравнению

```python
y2 = np.zeros(N2)
for n in range(N2):
    y2[n] = b0 * x[n]
    if n >= 1: y2[n] += b1 * x[n-1] - a1 * y2[n-1]
    if n >= 2: y2[n] += b2 * x[n-2] - a2 * y2[n-2]
n_y2 = np.arange(N2)

print("Проверка с lfilter, max|разности| =", np.max(np.abs(y2 - lfilter(b, a, x))))

fig, axes = plt.subplots(2, 1, figsize=(10, 6))
axes[0].stem(n_x, x, basefmt="k-")
axes[0].set_title("Воздействие x(n)")
axes[0].set_xlabel("n"); axes[0].set_ylabel("x(n)"); axes[0].grid(True)

axes[1].stem(n_y2, y2, basefmt="k-")
axes[1].set_title("Реакция y2(n) по разностному уравнению")
axes[1].set_xlabel("n"); axes[1].set_ylabel("y2(n)"); axes[1].grid(True)
plt.tight_layout(); plt.show()

plt.figure()
plt.stem(n_y1, y1, basefmt="k-", linefmt="b-", markerfmt="bo", label="y1 (свёртка)")
plt.stem(n_y2, y2, basefmt="k-", linefmt="r--", markerfmt="rx", label="y2 (разностное ур.)")
plt.axvline(N1 - 0.5, color="gray", ls=":", label="n = N1")
plt.title("Сравнение y1(n) и y2(n)")
plt.xlabel("n"); plt.ylabel("y(n)"); plt.legend(); plt.grid(True)
plt.show()

d = np.abs(y1 - y2)
print(f"Максимальное расхождение |y1 - y2| при n < N1:  {d[:N1].max():.3e}")
print(f"Максимальное расхождение |y1 - y2| при n >= N1: {d[N1:].max():.3e}")
```

В отличие от свёртки, вычисление реакции по разностному уравнению имитирует реальную работу цифрового фильтра: на каждом шаге новый отсчёт реакции вычисляется по текущему и прошлым отсчётам воздействия и прошлым отсчётам реакции.

**Разностное уравнение рекурсивного звена 2-го порядка:**
$$y(n) = b_0 x(n) + b_1 x(n-1) + b_2 x(n-2) - a_1 y(n-1) - a_2 y(n-2)$$
Подставляем коэффициенты варианта 18 ($b = [0.86,\ -1.120236,\ 1.204]$, $a = [1,\ 1.2278,\ 0.748]$), учитывая знак «минус» перед $a_k$:
$$y(n) = 0.86\, x(n) - 1.120236\, x(n-1) + 1.204\, x(n-2) - 1.2278\, y(n-1) - 0.748\, y(n-2)$$

**Анализ длин сигналов:**
* **Длина воздействия** $x(n)$ равна **38** отсчётам.
* **Длина реакции** $y_2(n)$ также равна **38** отсчётам: разностное уравнение на каждый отсчёт воздействия вычисляет ровно один отсчёт реакции.

**Сравнение реакций $y_1(n)$ и $y_2(n)$:**
На нижнем графике наложены две реакции: синие кружки — результат свёртки ($y_1$), красные крестики — результат разностного уравнения ($y_2$).
* При $n < N_1 = 28$ реакции **совпадают**: расхождение порядка $10^{-16}$.
* При $n \ge 28$ появляется небольшое расхождение (до $\approx 0.05$). Свёртка использует ИХ, усечённую до 28 отсчётов, а разностное уравнение учитывает всю бесконечную ИХ. Например, при $n = 28$ в свёртке отсутствует слагаемое $h(28) x(0)$.

Таким образом, оба метода описывают одну и ту же систему и эквивалентны при полной ИХ. Разностное уравнение даёт точную реакцию рекурсивного звена, а свёртка с усечённой ИХ — приближённую, что подтверждает бесконечную длину ИХ.

### Пункт 1.4 Нули и полюсы

```python
zeros, poles, k_gain = tf2zpk(b, a)

print("Нули (z0_i):")
for z in zeros:
    rz = np.abs(z); ph = np.angle(z)
    print(f"  z0 = {z.real:+.6f} {z.imag:+.6f}j | |z0| = {rz:.6f}, arg = {ph:+.6f} рад = {ph/np.pi:+.6f}·π")

print("\nПолюсы (z*_i):")
for p in poles:
    rp = np.abs(p); ph = np.angle(p)
    print(f"  z* = {p.real:+.6f} {p.imag:+.6f}j | |z*| = {rp:.6f}, arg = {ph:+.6f} рад = {ph/np.pi:+.6f}·π")

print(f"\nКоэффициент усиления k = {k_gain:.6f}")
```

Умножив числитель и знаменатель ПФ на $z^2$, получаем отношение полиномов по положительным степеням $z$:
$$H(z) = \frac{b_0 z^2 + b_1 z + b_2}{z^2 + a_1 z + a_2}$$

* **Нули** — корни числителя $0.86 z^2 - 1.120236 z + 1.204 = 0$, т.е. значения $z$, при которых $H(z) = 0$.
* **Полюсы** — корни знаменателя $z^2 + 1.2278 z + 0.748 = 0$, т.е. значения $z$, при которых знаменатель ПФ обращается в ноль.

Оба уравнения имеют отрицательный дискриминант, поэтому корни комплексные.

**1. Формы записи нулей и полюсов:**
1. **Алгебраическая форма:** $z = \xi + j\eta$, где $\xi$ — действительная часть, $\eta$ — мнимая.
2. **Показательная форма:** $z = r e^{j\varphi}$, где $r = \sqrt{\xi^2 + \eta^2}$ — модуль, $\varphi = \arg z$ — аргумент.

**Связь между формами:** по формуле Эйлера $e^{j\varphi} = \cos\varphi + j\sin\varphi$, т.е. $\xi = r\cos\varphi$, $\eta = r\sin\varphi$. При вычислении аргумента нужно учитывать четверть: у полюсов $\xi < 0$, $\eta > 0$ (II четверть), поэтому $\varphi = \pi - \text{arctg}\,|\eta/\xi| = \pi - 0.2488\pi = 0.7512\pi$.

**2. Результаты вычислений для варианта 18:**

| Тип | Алгебраическая форма | Модуль $r$ | Аргумент (рад) | Аргумент ($\cdot \pi$) |
| :--- | :--- | :--- | :--- | :--- |
| **Нуль $z_{01}$** | $0.6513 + j0.9878$ | $1.1832$ | $0.9879$ | $0.3145\pi$ |
| **Нуль $z_{02}$** | $0.6513 - j0.9878$ | $1.1832$ | $-0.9879$ | $-0.3145\pi$ |
| **Полюс $z_{*1}$** | $-0.6139 + j0.6092$ | $0.8649$ | $2.3600$ | $0.7512\pi$ |
| **Полюс $z_{*2}$** | $-0.6139 - j0.6092$ | $0.8649$ | $-2.3600$ | $-0.7512\pi$ |

Нули и полюсы образуют **комплексно-сопряжённые пары** (одинаковые модули, аргументы противоположного знака), так как коэффициенты ПФ — действительные числа.

**3. Представление передаточной функции в виде произведения простейших множителей:**
$$H(z) = b_0 \frac{(1 - z_{01} z^{-1})(1 - z_{02} z^{-1})}{(1 - z_{*1} z^{-1})(1 - z_{*2} z^{-1})}$$
Подставляем значения ($k = b_0 = 0.86$):
$$H(z) = 0.86 \frac{(1 - 1.1832 e^{j0.3145\pi} z^{-1})(1 - 1.1832 e^{-j0.3145\pi} z^{-1})}{(1 - 0.8649 e^{j0.7512\pi} z^{-1})(1 - 0.8649 e^{-j0.7512\pi} z^{-1})}$$

### Пункт 1.6 АЧХ и ФЧХ в нормированных частотах

```python
w_norm = np.linspace(0, np.pi, 2000)            
_, H = freqz(b, a, worN=w_norm)                 
A = np.abs(H)
Phi = np.unwrap(np.angle(H))

fig, axes = plt.subplots(2, 1, figsize=(10, 6))
axes[0].plot(w_norm/np.pi, A)
axes[0].axvline(np.angle(poles[0])/np.pi, color='r', ls=':', label="аргумент полюса")
axes[0].axvline(np.angle(zeros[0])/np.pi, color='b', ls=':', label="аргумент нуля")
axes[0].set_xlabel(r"$\hat{\omega}/\pi$"); axes[0].set_ylabel("|H|")
axes[0].set_title("АЧХ в нормированных частотах"); axes[0].legend(); axes[0].grid(True)

axes[1].plot(w_norm/np.pi, Phi/np.pi)
axes[1].set_xlabel(r"$\hat{\omega}/\pi$"); axes[1].set_ylabel(r"arg H / $\pi$")
axes[1].set_title("ФЧХ в нормированных частотах"); axes[1].grid(True)
plt.tight_layout(); plt.show()

i_max, i_min = np.argmax(A), np.argmin(A)
print(f"A(0)  = {A[0]:.4f}")
print(f"A(pi) = {A[-1]:.4f}")
print(f"Максимум АЧХ: A = {A[i_max]:.4f} при w = {w_norm[i_max]/np.pi:.4f}·π")
print(f"Минимум АЧХ:  A = {A[i_min]:.4f} при w = {w_norm[i_min]/np.pi:.4f}·π")
```

**Амплитудно-частотная характеристика (АЧХ)** показывает, как система изменяет амплитуду гармонических составляющих разных частот. **Фазочастотная характеристика (ФЧХ)** показывает их фазовый сдвиг.

**1. Границы основной полосы частот:**
Графики построены в шкале нормированных частот $\hat\omega = \omega T = 2\pi f / f_d$. Основная полоса частот: $\hat\omega \in [0;\ \pi]$. Значение $\hat\omega = \pi$ соответствует половине частоты дискретизации ($f_d/2$). Частотные характеристики дискретной системы периодичны с периодом $2\pi$, а при действительных коэффициентах АЧХ — чётная, ФЧХ — нечётная функция частоты. Поэтому полоса $[0;\ \pi]$ содержит всю информацию о частотных свойствах звена.

**2. Соответствие карты нулей и полюсов и вида АЧХ:**
АЧХ можно записать через нули и полюсы:
$$A(\hat\omega) = b_0 \frac{|e^{j\hat\omega} - z_{01}|\,|e^{j\hat\omega} - z_{02}|}{|e^{j\hat\omega} - z_{*1}|\,|e^{j\hat\omega} - z_{*2}|}$$
При изменении $\hat\omega$ от $0$ до $\pi$ точка $e^{j\hat\omega}$ движется по единичной окружности от $z = 1$ к $z = -1$, и АЧХ определяется расстояниями от неё до нулей и полюсов:
* **Резонансный пик** $A \approx 14.76$ на частоте $\hat\omega \approx 0.758\pi$ обусловлен **полюсом** с аргументом $\varphi = 0.7512\pi$: вблизи него знаменатель мал. Пик высокий и острый, так как полюс близок к единичной окружности.
* **Провал** $A \approx 0.13$ на частоте $\hat\omega \approx 0.306\pi$ обусловлен **нулём** с аргументом $\varphi = 0.3145\pi$: вблизи него мал числитель. Нуль лежит не на окружности, поэтому АЧХ в ноль не обращается.

**3. Скачок на $\pi$:**
Скачок ФЧХ ровно на $\pi$ происходит в тех точках, где АЧХ обращается в ноль, т.е. когда нуль лежит точно на единичной окружности; такому скачку соответствует значение **АЧХ, равное нулю**. В нашем случае нули лежат вне окружности, АЧХ нигде не равна нулю, поэтому **скачка на $\pi$ нет**. ФЧХ (с учётом `unwrap`) меняется непрерывно, общий набег фазы на основной полосе составляет $-2\pi$. Без `unwrap` на графике появился бы разрыв на $2\pi$ при $\hat\omega \approx 0.54\pi$, но он вызван только приведением фазы к интервалу $(-\pi;\ \pi]$, а не свойствами системы.

**4. Подавление частот:**
На нулевой частоте $A(0) \approx 0.32$, при $\hat\omega < 0.51\pi$ АЧХ меньше единицы, а на высоких частотах — резонанс около $0.75\pi$ и $A(\pi) \approx 6.12$. Следовательно, система преимущественно **подавляет низкие частоты** и **усиливает высокие**, т.е. ведёт себя как резонансный фильтр верхних частот. Это видно и на графиках реакции: постоянная составляющая прямоугольного импульса почти не проходит (реакция на «полке» около $0.32$), а на фронтах импульса возникают высокочастотные колебания с частотой, близкой к резонансной.

В данном пункте формируется периодическая последовательность $x(n)$ — сумма двух дискретных гармонических сигналов:
$$x(nT) = A_1 \cos(2\pi f_1 n T + \pi/4) + A_2 \cos(2\pi f_2 n T + \pi/8)$$

Параметры сигнала для варианта 18:
* Период последовательности: $N = 64$
* Частота дискретизации: $f_d = 8000$ Гц
* Период дискретизации: $T = 1/f_d = 0.125$ мс
* Амплитуды: $A_1 = 1.18$, $A_2 = 2.36$
* Частоты: $f_1 = 1000$ Гц, $f_2 = 1500$ Гц

**Пояснение к графикам:**
1. **В шкале дискретного времени $nT$** (верхний график): по оси абсцисс отложены моменты взятия отсчётов в миллисекундах, от $0$ до $(N-1)T = 7.875$ мс. Период последовательности равен $NT = 8$ мс.
2. **В шкале дискретного нормированного времени $n$** (нижний график): по оси абсцисс отложены номера отсчётов от 0 до 63.

**Связь дискретного нормированного времени с дискретным временем:**
Нормированное время определяется как $\hat t = t / T$. Для дискретных моментов $t = nT$ получаем $\hat t = nT/T = n$, т.е. дискретное нормированное время — это номер отсчёта. Переход от шкалы $n$ к шкале $nT$ выполняется умножением на период дискретизации $T$, обратный переход — делением на $T$. Нормированное время удобно тем, что не зависит от частоты дискретизации.

На периоде $N = 64$ укладывается целое число периодов каждой гармоники: $f_1 N / f_d = 8$ и $f_2 N / f_d = 12$, поэтому последовательность действительно периодична с периодом $N$.

### Пункт 2.2 ОДПФ от ДПФ

В данном пункте проверяется свойство дискретного преобразования Фурье: прямое и обратное преобразования являются взаимно однозначными.

**Теоретическое обоснование:**
* **ДПФ** переводит последовательность из временной области в частотную, раскладывая её на дискретные гармоники.
* **ОДПФ** выполняет противоположную операцию: по отсчётам спектра восстанавливает исходную последовательность.

Были последовательно вычислены ДПФ и ОДПФ:
$$x_{rec}(n) = \text{ОДПФ}\{\text{ДПФ}\{x(n)\}\}$$

**Анализ результатов:**
1. **Визуальное сравнение:** график восстановленной последовательности $x_{rec}(n)$ в шкале дискретного нормированного времени $n$ полностью совпадает с исходным графиком $x(n)$ из пункта 2.1.
2. **Численная проверка:** максимальное расхождение между исходной и восстановленной последовательностями, как и мнимая часть результата ОДПФ, — порядка $10^{-15}$, т.е. машинный ноль.

Результат подтверждает, что ДПФ не теряет информацию о последовательности.

Для анализа частотного состава последовательности вычислено ДПФ $X(k)$. Амплитудный спектр периодической последовательности — модули коэффициентов дискретного ряда Фурье $|c_k| = |X(k)|/N$, фазовый спектр — $\arg X(k)$. Спектры построены в двух шкалах: дискретных нормированных частот $k$ и абсолютных частот $f$ (Гц).

**1. Связь дискретного нормированного времени с дискретным временем:**
Как показано в пункте 2.1, $\hat t = nT/T = n$: дискретное нормированное время — номер отсчёта, а дискретное время $nT$ получается умножением на период дискретизации $T = 0.125$ мс.

**2. Связь частоты $f$ (Гц) с дискретной нормированной частотой:**
Дискретная нормированная частота — номер отсчёта спектра $k$. Связь с абсолютной частотой:
$$f_k = k \frac{f_d}{N} = k \Delta f, \qquad \Delta f = \frac{f_d}{N} = \frac{1}{NT}$$
В нашем варианте $\Delta f = 8000 / 64 = 125$ Гц, т.е. каждый шаг по $k$ соответствует сдвигу по частоте на 125 Гц. Соответствующая нормированная частота $\hat\omega_k = 2\pi k / N$; отсчёт $k = N/2 = 32$ соответствует $\hat\omega = \pi$ и $f = f_d/2 = 4000$ Гц.

**3. Вид амплитудного спектра:**
Спектр **линейчатый**: на нём **четыре ненулевых отсчёта** (две гармоники и их симметричные отражения):
* $k = 8$ ($f = 1000$ Гц) и $k = 12$ ($f = 1500$ Гц) — это в точности частоты $f_1$ и $f_2$ исходного сигнала.
* $k = 56$ ($f = 7000$ Гц) и $k = 52$ ($f = 6500$ Гц) — симметричные отсчёты, соответствующие отрицательным частотам $-f_1$ и $-f_2$. Они возникают из-за свойства симметрии ДПФ вещественной последовательности $|X(k)| = |X(N-k)|$; осью симметрии является $k = N/2$ ($f_d/2 = 4000$ Гц).
* *Пояснение по амплитуде:* $|X(k)| = A N / 2$, т.е. $|X(8)| = 1.18 \cdot 64 / 2 = 37.76$ и $|X(12)| = 2.36 \cdot 64 / 2 = 75.52$. Соответственно, амплитудный спектр $|c_k| = A/2$: $0.59$ и $1.18$. Каждая гармоника делит свою амплитуду пополам между положительной и отрицательной частотами, поэтому $2|c_k| = A_i$.

Растекания спектра нет, так как на периоде $N$ укладывается целое число периодов гармоник (8 и 12).

**4. Вид фазового спектра:**
Фазовый спектр имеет смысл только в точках, где амплитуда отлична от нуля:
* на частоте $1000$ Гц ($k = 8$) фаза равна $+0.785$ рад $= \pi/4$;
* на частоте $1500$ Гц ($k = 12$) фаза равна $+0.393$ рад $= \pi/8$;
* в симметричных точках $k = 56$ и $k = 52$ фазы противоположны по знаку: $-\pi/4$ и $-\pi/8$ (свойство $\arg X(k) = -\arg X(N-k)$).

Эти значения совпадают с начальными фазами гармоник исходного сигнала. В остальных точках амплитуда равна нулю, и их фаза принята равной нулю.

В данном пункте сравниваются:

2. **Амплитудный спектр периодической последовательности** — модули коэффициентов дискретного ряда Фурье последовательности, в которой эти же $N$ отсчётов повторяются бесконечно.

**Связь модуля ДПФ конечной последовательности с амплитудным спектром периодической последовательности:**
$$|X(k)| = N \cdot |c_k|$$
где $N = 64$ — длина последовательности (период).

**Анализ графиков:**
Графики модуля ДПФ (верхний) и амплитудного спектра (нижний) **совпадают по форме** и отличаются только масштабом в $N = 64$ раз: $|X(8)| = 37.76 = 64 \cdot 0.59$, $|X(12)| = 75.52 = 64 \cdot 1.18$. Это объясняется тем, что обе величины вычисляются по одним и тем же $N$ отсчётам, а различаются только множителем $1/N$ в определении коэффициентов ряда Фурье.

Так как на длине $N$ укладывается целое число периодов гармоник, «растекания спектра» нет и в обоих случаях ненулевыми оказываются только 4 отсчёта ($k = 8, 12, 52, 56$).

Таким образом, ДПФ конечной последовательности длины $N$ с точностью до множителя $N$ даёт амплитудный спектр её периодического продолжения с периодом $N$.

### Пункт 2.5 и 2.6 Восстановление аналогового сигнала

```python
half = N // 2
k_a = np.arange(-half, half)                
X_a = X[k_a % N]                             

# Время: t ∈ [0; (N−1)T] с шагом Δt = 0.25T
dt = 0.25 * T
t = np.arange(0, (N - 1) * T + dt/2, dt)

# Восстановленный сигнал по формуле x(t) = 1/N · Σ X_a(k) exp(j 2π k t / (NT))
x_rec_t = (np.exp(1j * 2 * np.pi * np.outer(t, k_a) / (N * T)) @ X_a) / N
print(f"Максимальная мнимая часть x(t): {np.max(np.abs(x_rec_t.imag)):.3e}")
x_rec_t = x_rec_t.real

# Исходный аналоговый сигнал x(t) = x(nT)|nT=t
x_orig_t = A1*np.cos(2*np.pi*f1*t + np.pi/4) + A2*np.cos(2*np.pi*f2*t + np.pi/8)

f_a = k_a / (N * T)                          

fig, axes = plt.subplots(2, 2, figsize=(14, 8))
axes[0,0].stem(n, x_per, basefmt="k-")
axes[0,0].set_title("Периодическая последовательность x(n)")
axes[0,0].set_xlabel("n"); axes[0,0].grid(True)

axes[0,1].stem(k, np.abs(X), basefmt="k-")
axes[0,1].set_title("Модуль ДПФ |X(k)|")
axes[0,1].set_xlabel("k"); axes[0,1].grid(True)

axes[1,0].plot(t * 1e3, x_rec_t, 'b-')
axes[1,0].plot(t_abs * 1e3, x_per, 'k.', ms=5)
axes[1,0].set_title("Восстановленный x(t) (точки — отсчёты x(nT))")
axes[1,0].set_xlabel("t, мс"); axes[1,0].grid(True)

axes[1,1].stem(f_a, np.abs(X_a) / N, basefmt="k-")
axes[1,1].set_title("Амплитудный спектр восстановленного сигнала |X_a(k)|/N")
axes[1,1].set_xlabel("f, Гц"); axes[1,1].grid(True)
plt.tight_layout(); plt.show()

plt.figure(figsize=(12, 4))
plt.plot(t * 1e3, x_orig_t, 'r-', lw=3, alpha=0.5, label="исходный x(t)")
plt.plot(t * 1e3, x_rec_t, 'b--', lw=1, label="восстановленный x(t)")
plt.title("Восстановленный и исходный аналоговые сигналы")
plt.xlabel("t, мс"); plt.legend(); plt.grid(True)
plt.show()

print(f"Максимальное расхождение восстановленного и исходного сигналов: {np.max(np.abs(x_rec_t - x_orig_t)):.3e}")
```

В данном пункте решается обратная задача: по дискретным отсчётам ДПФ восстанавливается непрерывный (аналоговый) сигнал $x(t)$.

**1. Пояснение к формуле и сдвигу индексов:**
Для восстановления используется ряд Фурье:
$$x(t) = \frac{1}{N} \sum_{k=-N/2}^{N/2-1} X_a(k) \exp\left(j \frac{2\pi}{N T} k t\right), \qquad X_a(k) = \begin{cases} X(N+k), & -N/2 \le k \le -1, \\ X(k), & 0 \le k \le N/2 - 1. \end{cases}$$
ДПФ вычисляется для $k = 0 \dots N-1$. Отсчёты с $k = N/2 \dots N-1$ соответствуют отрицательным частотам, поэтому они переносятся в начало и получают индексы $-N/2 \dots -1$. В коде это реализовано выражением `X[k_a % N]`. Время задано на интервале $t \in [0;\ (N-1)T]$ с шагом $\Delta t = 0.25T$.

**2. Связь модуля ДПФ со спектром аналогового сигнала:**
На графике амплитудного спектра восстановленного сигнала видны четыре линии на частотах $f_k = k / (NT)$:
* на частотах $\pm 1000$ Гц высотой $|X_a|/N = 0.59 = A_1/2$;
* на частотах $\pm 1500$ Гц высотой $|X_a|/N = 1.18 = A_2/2$.

Положительная и отрицательная частоты вместе дают гармонику амплитудой $A_1 = 1.18$ и $A_2 = 2.36$ — в точности как в исходном сигнале. Таким образом, модуль ДПФ на основной полосе $[-f_d/2;\ f_d/2)$, делённый на $N$, совпадает с амплитудным спектром аналогового сигнала. Это выполняется потому, что спектр сигнала ограничен ($f_{max} = 1500$ Гц $< f_d/2 = 4000$ Гц), т.е. соблюдено условие теоремы Котельникова и наложения спектров нет.

**3. Результат визуального сравнения:**
На нижнем графике (время от 0 до $7.875$ мс) наложены два сигнала:
* синяя пунктирная линия — сигнал, восстановленный по отсчётам ДПФ;
* красная линия — исходный аналоговый сигнал $x(t) = x(nT)|_{nT = t}$, рассчитанный по формуле.

Графики **полностью совпадают** (расхождение порядка $10^{-14}$), а восстановленный сигнал проходит через все отсчёты $x(nT)$. По 64 отсчётам удалось точно восстановить непрерывную форму сигнала между ними, потому что:
* частоты гармоник меньше $f_d/2$ (теорема Котельникова);
* на интервале $NT$ укладывается целое число периодов гармоник, поэтому спектр не растекается;
* отсчёт на частоте $-f_d/2$ ($k = -N/2$) равен нулю, поэтому восстановленный сигнал вещественный.

### Пункт 2.7 Равенство Парсеваля

```python
E_time = np.sum(np.abs(x_per)**2)
E_freq = (1.0/N) * np.sum(np.abs(X)**2)

print(f"Энергия во времени:  Σ|x(n)|²       = {E_time:.6f}")
print(f"Энергия в частоте:  (1/N)·Σ|X(k)|² = {E_freq:.6f}")
print(f"Относительное расхождение: {abs(E_time - E_freq)/E_time:.3e}")
print(f"Равенство Парсеваля выполняется: {np.isclose(E_time, E_freq)}")
```

**Равенство Парсеваля** выражает закон сохранения энергии для сигналов: энергия последовательности, вычисленная во временной области, равна энергии, вычисленной в частотной области.

Для ДПФ оно записывается так:
$$\sum_{n=0}^{N-1} |x(n)|^2 = \frac{1}{N} \sum_{k=0}^{N-1} |X(k)|^2$$

Смысл равенства заключается в том, что при переходе от временного представления сигнала к частотному (и обратно) ДПФ не теряет и не добавляет энергию. Энергию можно вычислять как сумму энергий отдельных спектральных составляющих.

**Результаты вычислений:**
* Энергия во временной области: $E_{time} = \sum |x(n)|^2 = \mathbf{222.784}$
* Энергия в частотной области: $E_{freq} = \frac{1}{N} \sum |X(k)|^2 = \mathbf{222.784}$

Относительное расхождение — порядка $10^{-16}$ (машинный ноль), функция `np.isclose()` подтвердила выполнение равенства.

**По результатам выполнения второй части (Дискретное преобразование Фурье) установлено:**
1. Прямое и обратное ДПФ являются взаимно обратными операциями: последовательность восстановлена без потерь (погрешность порядка $10^{-15}$).
2. Амплитудный и фазовый спектры точно определили частотный состав сигнала: обнаружены линии на частотах $1000$ Гц и $1500$ Гц ($k = 8$ и $k = 12$) с амплитудами $A_1/2$ и $A_2/2$ и начальными фазами $\pi/4$ и $\pi/8$, а также симметричные им линии на отрицательных частотах.
3. Модуль ДПФ конечной последовательности совпадает по форме с амплитудным спектром периодической последовательности и отличается от него в $N$ раз.
4. Восстановление аналогового сигнала по $64$ отсчётам ДПФ с помощью ряда Фурье показало полное совпадение с исходным аналоговым сигналом, что подтверждает выполнение условия теоремы Котельникова при $f_d = 8000$ Гц.
5. Проверка равенства Парсеваля подтвердила сохранение энергии: энергия во временной и частотной областях одинакова и равна $222.784$.
