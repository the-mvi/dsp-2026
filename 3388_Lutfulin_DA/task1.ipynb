## Основные теоретические положения

### Z-преобразование дискретного сигнала

Z-преобразование дискретного сигнала определяется выражением:

$$
X(z)=\mathcal{Z}\{x(n)\}
=\sum_{n=-\infty}^{+\infty}x(n)z^{-n}.
$$


### Передаточная функция

Для линейной стационарной системы выходной сигнал является свёрткой входного сигнала с импульсной характеристикой:

$$
y(n)=\sum_{k=0}^{+\infty}h(k)x(n-k).
$$

В Z-области свёртка преобразуется в произведение:

$$
Y(z)=H(z)X(z),
$$

где $X(z)$ и $Y(z)$ — Z-преобразования входного и выходного сигналов. Передаточная функция связана с импульсной характеристикой соотношением:

$$
H(z)=\mathcal{Z}\{h(n)\}=\frac{Y(z)}{X(z)}.
$$

В общем случае передаточная функция записывается в виде:

$$
H(z)=
\frac{\displaystyle\sum_{i=0}^{N-1}b_i z^{-i}}
{1+\displaystyle\sum_{k=1}^{M-1}a_k z^{-k}},
$$

где $a_k$ и $b_i$ — коэффициенты фильтра, $N$ и $M$ определяют количество коэффициентов числителя и знаменателя соответственно. Порядок фильтра определяется максимальной степенью $z^{-1}$ в знаменателе и равен $M-1$.

### Разностное уравнение

Разностное уравнение имеет вид:

$$
y(n)=\sum_{i=0}^{N-1}b_i x(n-i)
-\sum_{k=1}^{M-1}a_k y(n-k).
$$

Для фильтра второго порядка передаточная функция записывается следующим образом:

$$
H(z)=
\frac{b_0+b_1z^{-1}+b_2z^{-2}}
{1+a_1z^{-1}+a_2z^{-2}}.
$$

Соответствующее разностное уравнение имеет вид:

$$
\begin{aligned}
y(n)=b_0x(n)+b_1x(n-1)+b_2x(n-2) -a_1y(n-1)-a_2y(n-2).
\end{aligned}
$$

### Частотные характеристики

Амплитудно-частотная характеристика определяется модулем передаточной функции, а фазочастотная — её аргументом:

$$
A(\omega)=|H(e^{j\omega})|,
\qquad
\varphi(\omega)=\arg H(e^{j\omega}).
$$


### Нули и полюса передаточной функции

Передаточная функция может быть представлена в виде:

$$
H(z)=\frac{B(z)}{A(z)}.
$$

Нули определяются корнями уравнения:

$$
B(z)=0,
$$

а полюса — корнями уравнения:

$$
A(z)=0.
$$

Условием устойчивости ЛДС является расположение всех полюсов её передаточной функции строго внутри единичной окружности:

$$
|p_k|<1.
$$

Передаточная функция в форме произведения множителей:

$$
H(z)=K
\frac{\prod_i(1-z_i z^{-1})}
{\prod_k(1-p_k z^{-1})},
$$

где $z_i$ — нули, $p_k$ — полюса, $K$ — постоянный коэффициент.

### Дискретное преобразование Фурье

Дискретное преобразование Фурье последовательности длины $N$:

$$
X(k)=\sum_{n=0}^{N-1}
x(n)e^{-j\frac{2\pi}{N}kn}.
$$

Обратное дискретное преобразование Фурье:

$$
x(n)=\frac{1}{N}
\sum_{k=0}^{N-1}
X(k)e^{j\frac{2\pi}{N}kn}.
$$

### Амплитудный и фазовый спектры

Амплитудный спектр:
$$
|X(k)| =
\sqrt{
\left(\operatorname{Re}X(k)\right)^2 +
\left(\operatorname{Im}X(k)\right)^2}
$$

Фазовый спектр:
$$
\varphi(k) =
\operatorname{arg}X(e^{j\omega T}) =
\operatorname{arctg}\big(\frac{
\operatorname{Im}X(k)}
{\operatorname{Re}X(k)}\big)
$$

Амплитудный спектр сигнала получается делением модуля ДПФ на $N$: $A(k)=\frac{|X(k)|}{N}$, при этом фазовый спектр остаётся неизменным: $\varphi(k)=\arg X(k)$.

### Восстановление аналогового сигнала

Для восстановления аналогового сигнала по спектральным отсчётам используется выражение:

$$
x(t)=\frac{1}{N}
\sum_{k=-N/2}^{N/2-1}
X_a(k)e^{j\frac{2\pi}{NT}kt}.
$$

Спектральные отсчёты:

$$
X_a(k)=
\begin{cases}
X(N+k),&-\frac{N}{2}\leq k\leq-1,\\
X(k),&0\leq k\leq\frac{N}{2}-1.
\end{cases}
$$

В точках дискретизации восстановленный сигнал совпадает с исходными отсчётами:

$$
x(nT)=x(n).
$$

Условие восстановления аналогового сигнала определяется теоремой Котельникова:

$$
f_d>2f_{\max}.
$$

Где $f_{\max}$ - максимальная частота сигнала

### Равенство Парсеваля

Для ДПФ и ОДПФ равенство Парсеваля имеет вид:
$$
\sum_{n=0}^{N-1}|x(n)|^2
=
\frac{1}{N}
\sum_{k=0}^{N-1}|X(k)|^2.
$$

```python
b0 = 0.5 + 0.02 * N_VAR

b = np.array([
    b0,
    b0 * (-1)**(N_VAR+1) * (0.9822 + 0.0178*N_VAR),
    b0 * (0.8 + 0.2*(N_VAR % 5))
])

a = np.array([
    1.,
    (-1)**N_VAR * (0.7778 + 0.025*N_VAR),
    0.64 + 0.006*N_VAR
])

N1, N2, fs1 = N_VAR % 10 + 20, N_VAR % 10 + 30, 1000*N_VAR

# Импульсная характеристика
imp = np.r_[1., np.zeros(N1-1)]
h = signal.lfilter(b, a, imp)

print(f'b  = {b}\na  = {a}\nN1 = {N1}\nN2 = {N2}\nfд = {fs1} Гц')

fig, ax = plt.subplots()
markerline, stemlines, baseline = ax.stem(np.arange(N1), h)
baseline.set_visible(False)
ax.set_title('Импульсная характеристика')
ax.set_xlabel('n')
ax.set_ylabel('h(n)')
ax.grid(True)

plt.show()
```

Связь между ними определяется формулами:

$r=\sqrt{x^2+y^2} \qquad \varphi=\arg(z)=\arctg\left(\frac{y}{x}\right), \qquad x=r\cos\varphi,\qquad y=r\sin\varphi$

Передаточная функция:

$H(z) = 0.680\cdot\frac{(1 - 1.265\cdot e^{j\cdot (0.649)\cdot \pi}\cdot z^{-1})(1 - 1.265\cdot e^{j\cdot (-0.649)\cdot \pi}\cdot z^{-1})}{(1 - 0.833\cdot e^{j\cdot (0.294)\cdot \pi}\cdot z^{-1})(1 - 0.833\cdot e^{j\cdot (-0.294)\cdot \pi}\cdot z^{-1})}$

Заметим, что т.к. коэффициенты $a_i$ и  $b_i$ - вещественные, полюса и нули получились комплексно-сопряжёнными

#### 6. Частотные характеристики в нормированной шкале

- Вычислить амплитудно-частотную (АЧХ) и фазочастотную (ФЧХ) характеристики в шкале нормированных частот $ \hat{\omega} $ и вывести их графики. Пояснить:
    - чему равны границы основной полосы частот;
    - соответствие между картой нулей и полюсов и видом АЧХ;
    - какому значению АЧХ соответствует скачок на $ \pi $, если он имеется;
    - какие частотные составляющие воздействия, низкие или высокие, оказались преимущественно подавленными в реакции.

```python
# Вычисление частотной характеристики фильтра
# worN=2048 — количество точек по частоте
# whole=True — расчёт на полном диапазоне от 0 до 2π
w, H = signal.freqz(b, a, worN=2048, whole=True)
# Перенос нулевой частоты в центр массива (от -π до π а не от 0 до 2π)
w = np.fft.fftshift(w)
H = np.fft.fftshift(H)
# Преобразование частот из диапазона [0, 2π) в диапазон [-π, π)
w = np.where(w >= np.pi, w - 2*np.pi, w)

fig, ax = plt.subplots(2, 1, figsize=(10, 6))
ax[0].plot(w/np.pi, abs(H))
ax[0].set(xlabel='ω̂ / π', ylabel='|H|', title='АЧХ')
ax[1].plot(w/np.pi, np.angle(H))
ax[1].set(xlabel='ω̂ / π', ylabel='φ, рад', title='Главное значение ФЧХ')
plt.tight_layout()
plt.show()

# Фаза частотной характеристики
phi = np.angle(H)

# Вычисление разности фаз между соседними отсчётами
dphi = np.diff(phi)

# Поиск скачков фазы, превышающих π
idx = np.where(np.abs(dphi) > np.pi)[0]

# Вывод частоты, фазы и значения АЧХ в точках скачка
for i in idx:
    print(f'Частота ω/π = {w[i]/np.pi:.6f}')
    print(f'Фаза до скачка φ = {phi[i]:.6f} рад')
    print(f'Фаза после скачка φ = {phi[i+1]:.6f} рад')
    print(f'АЧХ до скачка = {abs(H[i]):.6f}')
    print(f'АЧХ после скачка = {abs(H[i+1]):.6f}')
    print()
```

Основная полоса нормированных угловых частот: $-\pi\leqslant\hat\omega<\pi$. На графике изображено $\hat\omega / \pi$, поэтому диапазон $[-1,1)$

Полюсы фильтра расположены внутри единичной окружности и имеют аргументы приблизительно $\pm0.294\pi$. Вблизи этих углов на АЧХ наблюдаются выраженные максимумы, поскольку полюсы способствуют увеличению модуля передаточной функции. Нули расположены за пределами единичной окружности, их аргументы приблизительно равны $\pm0.649\pi$. Вблизи соответствующих частот наблюдаются минимумы АЧХ. Поскольку нули не лежат непосредственно на единичной окружности, полное подавление этих частот не происходит.

Скачки фазы происходят на частотах $\omega/\pi=-0.456055$ и $\omega/\pi=0.455078$. В первой точке фаза изменяется с $-3.139561$ рад до $3.139537$ рад, а значение АЧХ изменяется с $1.326492$ до $1.340288$. Во второй точке фаза изменяется с $-3.139537$ рад до $3.139561$ рад, а значение АЧХ изменяется с $1.340288$ до $1.326492$. Низкие частоты соответствуют медленным изменениям сигнала и располагаются вблизи $\omega=0$. Высокие частоты соответствуют быстрым изменениям сигнала и располагаются вблизи $\omega=\pi$ в одностороннем частотном диапазоне цифрового сигнала. Амплитудно-частотная характеристика (АЧХ) показывает, как фильтр изменяет амплитуду составляющих сигнала на разных частотах.

#### 1. Периодическая последовательность и её спектр

- Вычислить амплитудный и фазовый спектры периодической последовательности $ x(n) $ с периодом $ N $: $$ x(nT) = A_1\cos(2\pi f_1nT + \pi/4) + A_2\cos(2\pi f_2nT + \pi/8) $$ Вывести графики последовательности $ x(n) $ на периоде $ N $ в шкале дискретного и дискретного нормированного времени.

```python
N = 64
fs = 2000*(N_VAR%5+1)
T = 1/fs
A1 = 1+0.01*N_VAR
A2 = 2*A1
f1, f2 = fs/8, 1.5*fs/8

n = np.arange(N)

tn = n * T # Дискретное время
x1 = A1 * np.cos(2 * np.pi * f1 * tn + np.pi / 4) # Первая гармоника
x2 = A2 * np.cos(2 * np.pi * f2 * tn + np.pi / 8) # Вторая гармоника
x = x1 + x2 # Суммарный сигнал

fig, ax = plt.subplots(2, 2, figsize=(12, 7))

stem(ax[0, 0], tn, x1,
     'Первая гармоника — дискретное время',
     't = nT, с', 'x1(nT)')

stem(ax[0, 1], tn, x,
     'Вторая гармоника — дискретное время',
     't = nT, с', 'x(nT)')

stem(ax[1, 0], tn, x,
     'Суммарный сигнал — дискретное время',
     't = nT, с', 'x(nT)')

stem(ax[1, 1], n, x,
     'Суммарный сигнал — дискретное нормированное время',
     'n', 'x(n)')

plt.tight_layout()
plt.show()

print(f"fs = {fs}. T = {T:.4f}, f1 = {f1:.2f} Гц, f2 = {f2:.2f} Гц, A1 = {A1:.6f}, A2 = {A2:.6f}")
```

#### 3. Амплитудный и фазовый спектры

  - Вычислить амплитудный и фазовый спектры последовательности $ x(n) $ и вывести их графики в шкалах дискретных нормированных и абсолютных частот. Пояснить:
    - связь дискретного нормированного времени с дискретным временем;
    - связь частоты $ f $ (Гц) с дискретной нормированной частотой;
    - вид амплитудного и фазового спектров.

```python
# Получаем частоты спектральных отсчётов в Гц
freq = np.fft.fftfreq(N, d=T)

# Переносим нулевую частоту в центр
freq = np.fft.fftshift(freq)

# Вычисляем нормированную угловую частоту
normfreq = 2 * np.pi * freq / fs

# Центрируем спектр ДПФ в том же порядке, что и частотную ось
Xs = np.fft.fftshift(X)

# Вычисляем амплитудный спектр

amp = np.abs(Xs) / N
mask = amp > 1e-10
amp_plot = np.where(mask, amp, np.nan)
phase = np.where(mask, np.angle(Xs), np.nan)

# Вычисляем фазовый спектр в радианах
fig, ax = plt.subplots(2, 2, figsize=(12, 7))

for j, (axis, label) in enumerate([
    (normfreq, 'ω, рад/отсчёт'),
    (freq, 'f, Гц')
]):
    # Амплитудный спектр
    stem(
        ax[0, j], axis, amp_plot,
        'Амплитудный спектр',
        label, '|A(k) = X(k)/N|'
    )

    # Фазовый спектр
    stem(
        ax[1, j], axis, phase,
        'Фазовый спектр',
        label, 'φ, рад'
    )

plt.tight_layout()
plt.show()

print('Ненулевые линии:')
mask = amp > 1e-10

for f, aa, pp in zip(freq[mask], amp[mask], phase[mask]):
    print(
        f'частота = {f:9.2f} Гц, '
        f'амплитуда = {aa:.6f}, '
        f'фаза = {pp:+.6f} рад = {pp / np.pi:+.4f}π'
    )
```

- Дискретное время задаётся моментами $t=nT$, где $T$ — период дискретизации, а нормированное время — целочисленным номером отсчёта $n$. $x(nT)$ - в дискретном, $x(n)$ - в дискретном нормированном

- Нормированная частота получается делением абсолютной частоты на частоту дискретизации с учётом множителя $2\pi$: $\hat{\omega}=2\pi f/f_s$.
  - Абсолютная частота (Гц): имеет диапазон

  $$f(k) = \frac{k f_s}{N} = \frac{k}{NT_s}, \quad -\frac{f_s}{2}\leq f < \frac{f_s}{2}$$

  - Нормированная частота (рад/отсчёт):
  $$\hat{\omega}(k) = \frac{2\pi f}{f_s} = 2\pi fT_s = \frac{2\pi k}{N}, \quad -\pi\leq\hat{\omega}<\pi$$

  где $T_s$ - период дискретизации, $f_s$ - частота дискретизации
  $$\hat{\omega}=\frac{2\pi f}{f_s} $$


По формуле Эйлера каждая косинусоида раскладывается на две комплексные экспоненты: с положительной и отрицательной частотами.
$$
\cos(\hat{\omega} n+\varphi)=
\frac{1}{2}e^{j(\hat{\omega} n+\varphi)}
+\frac{1}{2}e^{-j(\hat{\omega} n+\varphi)}
$$

$$
\begin{aligned}
A_1\cos(2\pi f_1nT+\pi/4)
={}&\frac{A_1}{2}e^{j\pi/4}e^{j2\pi f_1nT} +\frac{A_1}{2}e^{-j\pi/4}e^{-j2\pi f_1nT}
\end{aligned}
$$

$$
\begin{aligned}
A_2\cos(2\pi f_2nT+\pi/8)
={}&\frac{A_2}{2}e^{j\pi/8}e^{j2\pi f_2nT}+\frac{A_2}{2}e^{-j\pi/8}e^{-j2\pi f_2nT}
\end{aligned}
$$

Таким образом, у сигнала 4 комплексных составляющие:
| Частота $\hat{\omega}$ | Комплексный коэффициент | Амплитуда | Фаза |
|---|---|---|---|
| $-\hat{\omega}_2=-2\pi f_2T$ | $\frac{A_2}{2}e^{-j\pi/8}$ | $\frac{A_2}{2}$ | $-\frac{\pi}{8}$ |
| $-\hat{\omega}_1=-2\pi f_1T$ | $\frac{A_1}{2}e^{-j\pi/4}$ | $\frac{A_1}{2}$ | $-\frac{\pi}{4}$ |
| $+\hat{\omega}_1=+2\pi f_1T$ | $\frac{A_1}{2}e^{j\pi/4}$ | $\frac{A_1}{2}$ | $+\frac{\pi}{4}$ |
| $+\hat{\omega}_2=+2\pi f_2T$ | $\frac{A_2}{2}e^{j\pi/8}$ | $\frac{A_2}{2}$ | $+\frac{\pi}{8}$ |

- Амплитуда спектра равна амплитуде исходной косинусоиды, делённой пополам (по формуле Эйлера). Она в $N$ раз больше модуля ДПФ
- Фаза равна начальной фазы исходной косинусоиды. Знак зависит от направления частоты.
- Частота берётся из частот исходных косинусоид $f_1$ и $f_2$ и переводится в нормированную частоту по формуле $\hat{\omega}=2\pi fT$. Каждая косинусоида даёт две составляющие: с положительной и отрицательной частотами.

#### 4. ДПФ конечной последовательности

  - Вычислить ДПФ конечной последовательности $ x(n) $ длины $ N $. Вывести графики в шкале дискретных нормированных частот модуля ДПФ конечной последовательности и амплитудного спектра периодической последовательности. Пояснить связь модуля ДПФ конечной последовательности с амплитудным спектром периодической последовательности.

```python
fig, ax = plt.subplots(2, 1, figsize=(10, 6))
stem(
    ax[0], normfreq, np.abs(np.fft.fftshift(X)),
    'Модуль ДПФ конечной последовательности',
    'ω, рад/отсчёт', '|X(k)|'
)
stem(
    ax[1], normfreq, amp,
    'Амплитудный спектр периодической последовательности',
    'ω, рад/отсчёт', 'A(k) = |X(k)|/N'
)
plt.tight_layout()
plt.show()
print(
    'Максимальная ошибка нормировки:',
    np.max(np.abs(np.abs(np.fft.fftshift(X)) / N - amp))
)
```

Связь между графиками заключается в том, что амплитудный спектр периодической последовательности получается делением модуля ДПФ на количество отсчётов $N$:

$$A(k)=\frac{|X(k)|}{N}$$

#### 5. Восстановление аналогового сигнала

- Восстановить периодический аналоговый сигнал $ x(t) $ по отсчётам ДПФ $ X(k) $ периодической последовательности $ x(n) $. Для вычисления значений сигнала $ x(t) $ использовать формулу: $$ x(t) = \frac{1}{N}\sum_{k = -N/2}^{N/2-1}X_а(k)\exp\Big(\,j\frac{2\pi}{NT}kt\Big), $$ где отсчеты $ X_а(k) $ связаны с отсчетами ДПФ $ X(k) $ соотношением: $$ X_а(k) = \begin{cases}X(N+k), &-N/2 \leqslant k \leqslant -1 \\ X(k), &0 \leqslant k \leqslant N/2 - 1\end{cases} $$ задавая значения времени $ t $ на интервале $ t \in [0; (N-1)T] $ с шагом $ \Delta t = 0.25T $. Вывести графики:
    - периодической последовательности $ x(n) $ и модуля ее ДПФ;
    - восстановленного аналогового сигнала $ x(t) $ и его амплитудного спектра;
    - исходного аналогового сигнала $ x(t) = x(nT)\big|_{nT = t} $.

```python
# Индексы k от -N/2 до N/2-1
# Перестановка спектральных отсчётов относительно нулевой частоты
k = np.arange(-N//2, N//2)
Xa = np.fft.fftshift(X)

# Создание более частой временной сетки
tfine = np.arange(4*N) * 0.25 * T

# Восстановление аналогового сигнала по спектральным коэффициентам ДПФ
xanalog = (
    np.exp(1j * 2 * np.pi * np.outer(tfine, k) / (N * T)) @ Xa
) / N

# Вычисление исходного аналогового сигнала для сравнения
xoriginal = (
    A1 * np.cos(2 * np.pi * f1 * tfine + np.pi/4)
    + A2 * np.cos(2 * np.pi * f2 * tfine + np.pi/8)
)

# Повторное ДПФ восстановленного аналогового сигнала
Xanalog = np.fft.fft(xanalog.real)

# Перенос нулевой частоты в центр спектра
Xanalog_shift = np.fft.fftshift(Xanalog)

# Вычисление амплитудного спектра восстановленного сигнала
# Нормировка на количество отсчётов восстановленного сигнала
Aanalog = np.abs(Xanalog_shift) / len(xanalog)

# Формирование частотной оси для повторного ДПФ
fanalog = np.fft.fftshift(
    np.fft.fftfreq(len(xanalog), d=0.25*T)
)

fig, ax = plt.subplots(3, 2, figsize=(13, 9))

stem(
    ax[0, 0], tn, x,
    'Периодическая последовательность',
    't, с', 'x[n]'
)

stem(
    ax[0, 1], freq, abs(Xa),
    'Модуль исходного ДПФ',
    'f, Гц', '|X[k]|'
)

ax[1, 0].plot(tfine, xanalog.real)
ax[1, 0].set(
    title='Восстановленный аналоговый сигнал',
    xlabel='t, с',
    ylabel='x(t)'
)

stem(
    ax[1, 1], fanalog, Aanalog,
    'Спектр восстановленного сигнала (повторное ДПФ)',
    'f, Гц', 'Амплитуда'
)

ax[2, 0].plot(tfine, xoriginal, label='Исходный')
ax[2, 0].plot(tfine, xanalog.real, '--', label='Восстановленный')
ax[2, 0].set(
    title='Исходный и восстановленный сигналы',
    xlabel='t, с',
    ylabel='x(t)'
)
ax[2, 0].legend()
ax[2, 1].axis('off')
plt.tight_layout()
plt.show()

print(
    'Максимальное расхождение исходного и восстановленного:',
    np.max(np.abs(xanalog.real - xoriginal))
)
print(
    'Максимальная мнимая часть:',
    np.max(np.abs(xanalog.imag))
)
```

#### 7. Проверка равенства Парсеваля
- Проверить равенство Парсеваля. Пояснить его смысл

```python
E_time = np.sum(np.abs(x)**2)
E_freq = np.sum(np.abs(X)**2)/N
print(f'Энергия за период во времени: {E_time:.12f}')
print(f'Энергия по спектру:           {E_freq:.12f}')
print(f'Абсолютное расхождение:        {abs(E_time-E_freq):.3e}')
assert np.isclose(E_time, E_freq, rtol=1e-12, atol=1e-12)
```

**Равенство Парсеваля:**
$$
\sum_{n=0}^{N-1}|x(n)|^2
=
\frac{1}{N}
\sum_{k=0}^{N-1}|X(k)|^2.
$$

Равенство Парсеваля выполняется: энергия периодической последовательности, рассчитанная во временной области, совпадает с энергией, рассчитанной по спектру ДПФ. Это подтверждает корректность вычисления ДПФ и согласованность результатов во временной и частотной областях.
