## Основные теоретические положения

Линейная дискретная система (ЛДС) преобразует входную последовательность $x(n)$ в выходную $y(n)$ и подчиняется принципу суперпозиции. Стационарная ЛДС полностью описывается импульсной характеристикой $h(n)$, то есть реакцией на цифровой единичный импульс $u_0(n)$ при нулевых начальных условиях.

Реакция на произвольное воздействие находится по формуле свёртки
$$y(n) = \sum_{m=0}^{n} h(m)\,x(n-m).$$
Свёртка последовательностей длины $N_1$ и $N_2$ имеет длину $N_1 + N_2 - 1$.

Рекурсивное звено 2-го порядка описывается разностным уравнением
$$y(n) = b_0x(n) + b_1x(n-1) + b_2x(n-2) - a_1y(n-1) - a_2y(n-2)$$
и передаточной функцией
$$H(z) = \frac{Y(z)}{X(z)} = \frac{b_0 + b_1z^{-1} + b_2z^{-2}}{1 + a_1z^{-1} + a_2z^{-2}}
= b_0\frac{(1 - z_1z^{-1})(1 - z_1^*z^{-1})}{(1 - p_1z^{-1})(1 - p_1^*z^{-1})},$$
где $z_1, z_1^*$ - нули, $p_1, p_1^*$ - полюсы. При вещественных коэффициентах они образуют комплексно-сопряжённые пары. Звено устойчиво, если все полюсы лежат внутри единичной окружности, $|p_k| < 1$.

Частотная характеристика получается подстановкой $z = e^{j\hat\omega}$, где $\hat\omega = 2\pi f/f_d$:
$$H(e^{j\hat\omega}) = |H(e^{j\hat\omega})|\,e^{j\varphi(\hat\omega)}.$$
Модуль $|H(e^{j\hat\omega})|$ называют АЧХ, аргумент $\varphi(\hat\omega)$ называют ФЧХ. Характеристика периодична с периодом $2\pi$, основная полоса $\hat\omega\in[-\pi,\pi]$, или $f\in[-f_d/2,\,f_d/2]$. Нуль на единичной окружности обращает АЧХ в ноль на частоте, равной его аргументу. Полюс вблизи окружности даёт максимум АЧХ около частоты, равной его аргументу.

Прямое и обратное ДПФ последовательности длины $N$:
$$X(k) = \sum_{n=0}^{N-1} x(n)\,e^{-j\frac{2\pi}{N}kn},\qquad x(n) = \frac{1}{N}\sum_{k=0}^{N-1} X(k)\,e^{j\frac{2\pi}{N}kn}.$$
Здесь $n = nT/T$ - дискретное нормированное время, $k$ - дискретная нормированная частота, которой соответствует частота $f_k = k\,f_d/N$. Для периодической последовательности ДПФ одного периода даёт амплитудный спектр $|X(k)|$ и фазовый спектр $\arg X(k)$. Для вещественной последовательности $X(N-k) = X^*(k)$, поэтому амплитудный спектр симметричен относительно $k = N/2$, а фазовый антисимметричен.

Равенство Парсеваля для ДПФ:
$$\sum_{n=0}^{N-1}|x(n)|^2 = \frac{1}{N}\sum_{k=0}^{N-1}|X(k)|^2 .$$

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.signal as signal

plt.rcParams.update({'figure.figsize': (12, 4), 'axes.grid': True, 'grid.alpha': 0.4,
                     'font.size': 10})
```

### Часть 1. Линейные дискретные системы

```python
N_var = 1

# коэффициенты передаточной функции
b0 = 0.5 + 0.02 * N_var
b1 = b0 * (-1) ** (N_var + 1) * (0.9822 + 0.0178 * N_var)
b2 = b0 * (0.8 + 0.2 * (N_var % 5))
a0 = 1.0
a1 = (-1) ** N_var * (0.7778 + 0.025 * N_var)
a2 = 0.64 + 0.006 * N_var

N1 = N_var % 10 + 20       # длина импульсной характеристики
N2 = N_var % 10 + 30       # длина воздействия
fs1 = 1000 * N_var         # частота дискретизации, Гц

b = np.array([b0, b1, b2])
a = np.array([a0, a1, a2])
print(f'b = [{b0:.4f}, {b1:.4f}, {b2:.4f}]')
print(f'a = [{a0:.4f}, {a1:.4f}, {a2:.4f}]')
print(f'N1 = {N1}, N2 = {N2}, fд = {fs1} Гц')
```

#### 1. Импульсная характеристика

Импульсная характеристика вычисляется как реакция звена на единичный импульс $u_0(n)$ длины $N_1$. Для проверки она также вычисляется по аналитической формуле.

Полюсы звена равны $p_{1,2} = r e^{\pm j\varphi}$, где $r = \sqrt{a_2}$ и $\cos\varphi = -a_1/(2r)$. Знаменатель $1/(1 + a_1z^{-1} + a_2z^{-2})$ при нулевых начальных условиях имеет импульсную характеристику
$$g(n) = r^{\,n}\,\frac{\sin\big((n+1)\varphi\big)}{\sin\varphi},\quad n \geqslant 0,\qquad g(n) = 0,\quad n < 0.$$
Числитель добавляет к ней задержанные копии с весами $b_i$:
$$h(n) = b_0\,g(n) + b_1\,g(n-1) + b_2\,g(n-2).$$

```python
n1 = np.arange(N1)
u0 = np.zeros(N1)
u0[0] = 1.0
h = signal.lfilter(b, a, u0)               # реакция на единичный импульс

# аналитическая формула через полюсы
r = np.sqrt(a2)
phi = np.arccos(-a1 / (2 * r))

def g(m):
    return np.where(m >= 0, r ** m * np.sin((m + 1) * phi) / np.sin(phi), 0.0)

h_an = b0 * g(n1) + b1 * g(n1 - 1) + b2 * g(n1 - 2)

print(f'r = {r:.4f},  φ = {phi/np.pi:.4f}π')
print(f'max|h(n) - h_ан(n)| = {np.max(np.abs(h - h_an)):.2e}')
print('h(n) =', np.round(h, 4))

plt.figure(figsize=(12, 4))
plt.stem(n1, h, basefmt='k-', label='h(n), lfilter')
plt.plot(n1, h_an, 'rx', ms=7, label='аналитическая формула')
plt.title(f'Импульсная характеристика, N1 = {N1}')
plt.xlabel('n')
plt.ylabel('h(n)')
plt.legend()
plt.show()
```

Для варианта 1 получаем $r = 0.8037$, $\varphi = 0.3335\pi$ и
$$h(n) = 0.52\,\big[g(n) + g(n-1) + g(n-2)\big],\qquad g(n) = 0.8037^{\,n}\,\frac{\sin\big((n+1)\,0.3335\pi\big)}{\sin(0.3335\pi)}.$$
Значения по формуле совпадают с вычисленными рекурсивно с точностью до машинной погрешности.

Звено рекурсивное, поэтому его импульсная характеристика бесконечна: $h(n)$ затухает как $0.8037^n$, но в ноль не обращается. Значение $N_1 = 21$ задаёт только длину вычисленного фрагмента.

#### 2. Реакция по формуле свёртки

Воздействие - дискретный прямоугольный импульс длины $N_2$: $x(n) = 1$ при $0 \leqslant n < \mathrm{int}(N_2/2)$, иначе $0$. Здесь $\mathrm{int}(31/2) = \mathrm{int}(15.5) = 16$.

```python
n2 = np.arange(N2)
L = int(np.floor(N2 / 2 + 0.5))             # int(t) - округление до ближайшего целого
x = np.where(n2 < L, 1.0, 0.0)

y1_full = np.convolve(h, x)                  # полная линейная свёртка
y1 = y1_full[:N2]                            # ограничение до длины воздействия
nz = np.nonzero(np.abs(y1_full) > 1e-12)[0]

print(f'длина импульса x: {L} отсчётов (ненулевых), длина последовательности x: {N2}')
print(f'длина свёртки аналитически: N1 + N2 - 1 = {N1 + N2 - 1}, len(np.convolve) = {len(y1_full)}')
print(f'ненулевые отсчёты свёртки: n = {nz[0]} ... {nz[-1]}, всего {nz[-1] + 1}')

fig, ax = plt.subplots(1, 2, figsize=(13, 4))
ax[0].stem(n2, x, basefmt='k-')
ax[0].set(title='Воздействие x(n)', xlabel='n', ylabel='x(n)')
n_full = np.arange(len(y1_full))
ax[1].stem(n_full, y1_full, basefmt='k-', linefmt='C0-', markerfmt='C0o', label='полная свёртка')
ax[1].axvspan(N2 - 0.5, len(y1_full) - 0.5, color='grey', alpha=0.2, label='отбрасывается (n ≥ N2)')
ax[1].set(title='Реакция y1(n) по формуле свёртки', xlabel='n', ylabel='y1(n)')
ax[1].legend()
plt.tight_layout()
plt.show()
```

Импульс содержит 16 единичных отсчётов ($n = 0\ldots15$), а вся последовательность $x(n)$ вместе с нулями имеет длину $N_2 = 31$.

Аналитически длина свёртки равна $N_1 + N_2 - 1 = 21 + 31 - 1 = 51$ отсчёт. На графике ненулевые отсчёты занимают $n = 0\ldots35$, то есть $36 = 16 + 21 - 1$. Нулевой хвост $x(n)$ вклада не даёт, поэтому фактическая длина определяется длиной самого импульса.

Реакцию ограничивают длиной воздействия, чтобы рассматривать её на том же интервале, на котором задан вход, и сравнивать с реакцией $y_2(n)$ по разностному уравнению, у которой тоже $N_2$ отсчётов. Отсчёты свёртки при $n \geqslant N_1$ к тому же искажены усечением $h(n)$ до $N_1$ отсчётов.

#### 3. Реакция по разностному уравнению

Разностное уравнение рекурсивного звена 2-го порядка с заданными коэффициентами:
$$y(n) = 0.52\,x(n) + 0.52\,x(n-1) + 0.52\,x(n-2) + 0.8028\,y(n-1) - 0.646\,y(n-2).$$

```python
def diff_eq(b, a, x):
    # Рекурсивное звено 2-го порядка, нулевые начальные условия
    y = np.zeros(len(x))
    for n in range(len(x)):
        acc = 0.0
        for i in range(len(b)):
            if n - i >= 0:
                acc += b[i] * x[n - i]
        for k in range(1, len(a)):
            if n - k >= 0:
                acc -= a[k] * y[n - k]
        y[n] = acc / a[0]
    return y

y2 = diff_eq(b, a, x)
print(f'длина воздействия: {len(x)}, длина реакции y2: {len(y2)}')
d = np.abs(y1 - y2)
print(f'max|y1 - y2| при n < N1: {d[:N1].max():.2e}')
print(f'max|y1 - y2| при n ≥ N1: {d[N1:].max():.4f} (n = {N1 + np.argmax(d[N1:])})')

fig, ax = plt.subplots(1, 3, figsize=(15, 4))
ax[0].stem(n2, x, basefmt='k-')
ax[0].set(title='Воздействие x(n)', xlabel='n', ylabel='x(n)')
ax[1].stem(n2, y2, basefmt='k-')
ax[1].set(title='Реакция y2(n) по разностному уравнению', xlabel='n', ylabel='y2(n)')
ax[2].plot(n2, y1, 'o-', ms=4, label='y1(n), свёртка')
ax[2].plot(n2, y2, 'x--', ms=6, label='y2(n), разностное уравнение')
ax[2].stem(n2, y1 - y2, linefmt='r-', markerfmt='r.', basefmt='k-', label='y1 − y2')
ax[2].axvline(N1 - 0.5, color='g', ls=':', label='n = N1')
ax[2].set(title='Сравнение реакций', xlabel='n')
ax[2].legend(fontsize=8)
plt.tight_layout()
plt.show()
```

Длина воздействия $N_2 = 31$, длина вычисленной реакции $y_2(n)$ тоже 31 отсчёт: рекурсия даёт по одному выходному отсчёту на каждый входной. Теоретически реакция БИХ-звена бесконечна и после окончания импульса затухает как $0.8037^n$.

При $n < N_1 = 21$ реакции $y_1(n)$ и $y_2(n)$ совпадают с точностью до машинной погрешности. При $n \geqslant N_1$ появляется расхождение до 0.018, потому что свёртка использует импульсную характеристику, усечённую до 21 отсчёта, а разностное уравнение такого усечения не содержит. Точной является реакция $y_2(n)$.

#### 4. Нули и полюсы передаточной функции

```python
zeros = np.roots(b)
poles = np.roots(a)

def show(name, v):
    for i, z in enumerate(v, 1):
        print(f'{name}{i} = {z.real:+.4f} {z.imag:+.4f}j   =  {abs(z):.4f} · exp(j·{np.angle(z)/np.pi:+.4f}π)'
              f'   (arg = {np.angle(z):+.4f} рад)')

show('z', zeros)
show('p', poles)
```

Нули: $z_{1,2} = -0.5 \pm j0.8660 = e^{\pm j\frac{2\pi}{3}} = e^{\pm j0.6667\pi}$.

Полюсы: $p_{1,2} = 0.4014 \pm j0.6963 = 0.8037\,e^{\pm j0.3335\pi}$.

Алгебраическая форма $z = \mathrm{Re}\,z + j\,\mathrm{Im}\,z$ переводится в показательную $z = |z|e^{j\arg z}$ по формулам $|z| = \sqrt{\mathrm{Re}^2 z + \mathrm{Im}^2 z}$, $\arg z = \mathrm{arctg}(\mathrm{Im}\,z / \mathrm{Re}\,z)$ с учётом квадранта. Обратный переход: $\mathrm{Re}\,z = |z|\cos\arg z$, $\mathrm{Im}\,z = |z|\sin\arg z$. Коэффициенты вещественные, поэтому нули и полюсы образуют сопряжённые пары с равными модулями и противоположными аргументами. В варианте 1 $b_0 = b_1 = b_2$, числитель равен $b_0(1 + z^{-1} + z^{-2})$, и нули лежат точно на единичной окружности под углами $\pm 2\pi/3$.

Передаточная функция в виде произведения простейших множителей:
$$H(z) = 0.52\,\frac{\left(1 - e^{\,j\frac{2\pi}{3}}z^{-1}\right)\left(1 - e^{-j\frac{2\pi}{3}}z^{-1}\right)}{\left(1 - 0.8037\,e^{\,j0.3335\pi}z^{-1}\right)\left(1 - 0.8037\,e^{-j0.3335\pi}z^{-1}\right)}.$$

#### 6. АЧХ и ФЧХ в шкале нормированных частот $\hat\omega$

```python
w = np.linspace(-np.pi, np.pi, 4001)
_, H = signal.freqz(b, a, worN=w)
A = np.abs(H)
phase = np.angle(H)

i_max = np.argmax(A * (w >= 0))
print(f'|H| при ω = 0: {A[np.argmin(abs(w))]:.4f},  при ω = π: {A[-1]:.4f}')
print(f'максимум АЧХ: {A[i_max]:.4f} при ω = {w[i_max]/np.pi:.4f}π (аргумент полюса {np.angle(poles).max()/np.pi:.4f}π)')
print(f'минимум АЧХ на [0, π]: {A[w >= 0].min():.2e} при ω = {w[w >= 0][np.argmin(A[w >= 0])]/np.pi:.4f}π')
jumps = np.nonzero(np.abs(np.diff(phase)) > 2)[0]
for j in jumps:
    print(f'скачок ФЧХ на {phase[j+1]-phase[j]:+.4f} рад при ω ≈ {w[j]/np.pi:+.4f}π, |H| там = {A[j:j+2].min():.2e}')

fig, ax = plt.subplots(1, 2, figsize=(13, 4))
ax[0].plot(w / np.pi, A)
for v in np.angle(zeros):
    ax[0].axvline(v / np.pi, color='b', ls=':', lw=1)
for v in np.angle(poles):
    ax[0].axvline(v / np.pi, color='r', ls='--', lw=1)
ax[0].set(title='АЧХ (синие - arg нулей, красные - arg полюсов)', xlabel='ω / π', ylabel='|H(e^(jω))|')
ax[1].plot(w / np.pi, phase / np.pi)
ax[1].set(title='ФЧХ', xlabel='ω / π', ylabel='φ(ω) / π')
for a_ in ax: a_.set_xlim(-1, 1)
plt.tight_layout()
plt.show()
```

Основная полоса частот $\hat\omega \in [-\pi,\,\pi]$ рад. АЧХ чётная, ФЧХ нечётная, поэтому достаточно полосы $[0,\pi]$. За её пределами характеристики повторяются с периодом $2\pi$.

АЧХ в точке $\hat\omega$ пропорциональна отношению произведения расстояний от точки $e^{j\hat\omega}$ до нулей к произведению расстояний до полюсов. Полюсы $0.8037e^{\pm j0.3335\pi}$ расположены рядом с окружностью, поэтому около $\hat\omega \approx 0.316\pi$ у АЧХ есть резонансный максимум 3.49. Нули лежат на окружности под углом $\pm 2\pi/3$, и при $\hat\omega = \pm 2\pi/3$ АЧХ равна нулю. Между этим нулём и $\pi$ АЧХ немного растёт, но остаётся малой: $|H(e^{j\pi})| = 0.21$ против $|H(e^{j0})| = 1.85$ на нулевой частоте.

Скачок ФЧХ на $\pi$ наблюдается при $\hat\omega = \pm 2\pi/3$, где АЧХ равна нулю: множитель $(1 - e^{j2\pi/3}z^{-1})$ при переходе через нуль на окружности меняет знак.

Звено работает как фильтр нижних частот с резонансом около $0.32\pi$ (158 Гц). Низкие частоты проходят с усилением ($|H| = 1.85$ при $\hat\omega = 0$ и до 3.49 в резонансе). Составляющая на частоте $2\pi/3$ (333 Гц) подавляется полностью, более высокие частоты ослаблены до $|H| \leqslant 0.21$.

#### 7. АЧХ и ФЧХ в шкале частот $f$ (Гц)

Связь шкал: $f = \dfrac{\hat\omega\,f_d}{2\pi}$, $f_d = 1000$ Гц.

```python
f = w * fs1 / (2 * np.pi)
fig, ax = plt.subplots(1, 2, figsize=(13, 4))
ax[0].plot(f, A)
ax[0].axvline(fs1 * np.angle(zeros).max() / (2*np.pi), color='b', ls=':', lw=1, label='нуль')
ax[0].axvline(fs1 * np.angle(poles).max() / (2*np.pi), color='r', ls='--', lw=1, label='arg полюса')
ax[0].set(title=f'АЧХ, fд = {fs1} Гц', xlabel='f, Гц', ylabel='|H|', xlim=(-fs1/2, fs1/2))
ax[0].legend(fontsize=8)
ax[1].plot(f, phase / np.pi)
ax[1].set(title='ФЧХ', xlabel='f, Гц', ylabel='φ / π', xlim=(-fs1/2, fs1/2))
plt.tight_layout()
plt.show()

print('граница полосы:', fs1 / 2, 'Гц')
print('нуль АЧХ:', round(fs1 / 3, 1), 'Гц')
print('аргумент полюса:', round(np.angle(poles).max() * fs1 / (2 * np.pi), 1), 'Гц')
print('максимум АЧХ:', round(w[i_max] * fs1 / (2 * np.pi), 1), 'Гц')
```

Основная полоса частот $f \in [-f_d/2,\, f_d/2] = [-500,\,500]$ Гц.

Шкалы связаны формулой $\hat\omega = 2\pi f/f_d$ и отличаются только множителем $f_d/(2\pi)$, поэтому графики одинаковы по форме. Границе $\hat\omega = \pi$ соответствует частота Найквиста $f_d/2 = 500$ Гц, нулю АЧХ $\hat\omega = 2\pi/3$ соответствует 333.3 Гц, максимуму АЧХ $0.316\pi$ соответствует примерно 158 Гц, аргументу полюса $0.3335\pi$ соответствует примерно 167 Гц.

### Часть 2. Дискретное преобразование Фурье

```python
N = 64
fs = 2000 * (N_var % 5 + 1)       # Гц
T = 1 / fs
A1 = 1 + 0.01 * N_var
A2 = 2 * A1
f1 = fs / 8
f2 = 1.5 * f1

n = np.arange(N)
xn = A1 * np.cos(2 * np.pi * f1 * n * T + np.pi / 4) + A2 * np.cos(2 * np.pi * f2 * n * T + np.pi / 8)

print(f'N = {N}, fд = {fs} Гц, T = {T*1e3:.3f} мс')
print(f'A1 = {A1}, A2 = {A2}, f1 = {f1} Гц, f2 = {f2} Гц')
print(f'шаг по частоте df = {fs/N} Гц, k1 = {f1*N/fs:g}, k2 = {f2*N/fs:g}')
print(f'на периоде N укладывается {f1*N*T:g} периодов первой и {f2*N*T:g} периодов второй гармоники')
```

#### 1. Последовательность $x(n)$ на периоде $N$

Графики строятся в шкале дискретного времени $nT$ (мс) и дискретного нормированного времени $n$.

```python
fig, ax = plt.subplots(2, 1, figsize=(13, 6.5))
ax[0].stem(n * T * 1e3, xn, basefmt='k-')
ax[0].set(title='x(nT) - шкала дискретного времени', xlabel='nT, мс', ylabel='x(nT)')
ax[1].stem(n, xn, basefmt='k-')
ax[1].set(title='x(n) - шкала дискретного нормированного времени', xlabel='n', ylabel='x(n)')
plt.tight_layout()
plt.show()
```

#### 2. ОДПФ от ДПФ последовательности

```python
X = np.fft.fft(xn)
x_rec = np.fft.ifft(X)
print(f'max|ОДПФ{{ДПФ{{x}}}} - x| = {np.max(np.abs(x_rec.real - xn)):.2e}')

fig, ax = plt.subplots(figsize=(13, 3.5))
ax.stem(n, x_rec.real, basefmt='k-', label='ОДПФ{ДПФ{x(n)}}')
ax.plot(n, xn, 'rx', ms=6, label='исходная x(n)')
ax.set(title='Последовательность, полученная ОДПФ', xlabel='n', ylabel='x(n)')
ax.legend(fontsize=8)
plt.show()
```

ОДПФ от ДПФ восстанавливает исходную последовательность: пара ДПФ/ОДПФ взаимно обратна.

#### 3. Амплитудный и фазовый спектры

```python
k = np.arange(N)
fk = k * fs / N
amp = np.abs(X)
ph = np.angle(X)
ph[amp < 1e-9 * amp.max()] = 0.0      # фаза нулевых (на уровне шума) отсчётов не определена

idx = np.nonzero(amp > 1e-9 * amp.max())[0]
print(' k   f, Гц    |X(k)|   arg X(k)')
for i in idx:
    print(f'{i:2d}  {fk[i]:6.1f}  {amp[i]:8.3f}  {ph[i]/np.pi:+.4f}π')
print(f'N·A1/2 = {N*A1/2:.3f},  N·A2/2 = {N*A2/2:.3f}')

fig, ax = plt.subplots(2, 2, figsize=(14, 7))
ax[0, 0].stem(k, amp, basefmt='k-')
ax[0, 0].set(title='Амплитудный спектр |X(k)|', xlabel='k', ylabel='|X(k)|')
ax[1, 0].stem(k, ph / np.pi, basefmt='k-')
ax[1, 0].set(title='Фазовый спектр arg X(k)', xlabel='k', ylabel='arg X(k) / π')
ax[0, 1].stem(fk, amp, basefmt='k-')
ax[0, 1].set(title='Амплитудный спектр |X(f)|', xlabel='f, Гц', ylabel='|X|')
ax[1, 1].stem(fk, ph / np.pi, basefmt='k-')
ax[1, 1].set(title='Фазовый спектр arg X(f)', xlabel='f, Гц', ylabel='arg X / π')
for a_ in ax[:, 0]: a_.axvline(N / 2, color='grey', ls=':')
for a_ in ax[:, 1]: a_.axvline(fs / 2, color='grey', ls=':')
plt.tight_layout()
plt.show()
```

Дискретное нормированное время $n = nT/T$ - это номер отсчёта, безразмерная величина. Дискретное время $nT$ измеряется в секундах с шагом $T = 0.25$ мс, длительность периода $NT = 16$ мс.

Частота $f$ связана с дискретной нормированной частотой $k$ формулой $f_k = k\,f_d/N = k\,\Delta f$, где $\Delta f = 62.5$ Гц. Нормированная угловая частота равна $\hat\omega_k = 2\pi k/N$. Частоте $f_1 = 500$ Гц соответствует $k_1 = 8$, частоте $f_2 = 750$ Гц соответствует $k_2 = 12$, а $k = N/2 = 32$ соответствует $f_d/2 = 2000$ Гц.

Спектр линейчатый, ненулевых отсчётов четыре: $k = 8, 12$ и симметричные им $k = 52$ и $k = 56$. Два последних отвечают отрицательным частотам $-750$ и $-500$ Гц, которые из-за периодичности спектра попадают в интервал $[f_d/2,\,f_d)$. Амплитуды равны $|X(k_i)| = NA_i/2$, то есть 32.32 и 64.64, их отношение $A_2/A_1 = 2$. Фазы в этих точках равны начальным фазам гармоник, $\arg X(8) = \pi/4$ и $\arg X(12) = \pi/8$, а на симметричных отсчётах имеют противоположный знак. В период $N$ укладывается целое число периодов каждой гармоники (8 и 12), поэтому растекания спектра нет и остальные отсчёты нулевые.

#### 4. ДПФ конечной последовательности и спектр периодической

Последовательность $x(n)$ рассматривается как конечная длины $N$. Модуль её ДПФ сравнивается с амплитудным спектром периодической последовательности с периодом $N$.

```python
# ДПФ конечной последовательности
X_fin = np.fft.fft(xn, N)
# амплитудный спектр периодической последовательности (сумма по одному периоду)
X_per = np.array([np.sum(xn * np.exp(-2j * np.pi * kk * n / N)) for kk in k])
print(f'max| |X_fin| - |X_per| | = {np.max(np.abs(np.abs(X_fin) - np.abs(X_per))):.2e}')

fig, ax = plt.subplots(1, 2, figsize=(14, 4))
ax[0].stem(k, np.abs(X_fin), basefmt='k-')
ax[0].set(title='Модуль ДПФ конечной последовательности', xlabel='k', ylabel='|X(k)|')
ax[1].stem(k, np.abs(X_per), linefmt='C1-', markerfmt='C1o', basefmt='k-')
ax[1].set(title='Амплитудный спектр периодической последовательности', xlabel='k', ylabel='|X̃(k)|')
plt.tight_layout()
plt.show()
```

Модуль ДПФ конечной последовательности совпадает с амплитудным спектром периодической последовательности, полученной её повторением с периодом $N$, потому что в обоих случаях вычисляется одна и та же сумма по одному периоду. Коэффициенты ряда Фурье периодической последовательности равны $X(k)/N$ и отличаются от отсчётов ДПФ только множителем $1/N$.

#### 5. Восстановление периодического аналогового сигнала по отсчётам ДПФ

Аналоговый сигнал вычисляется по формуле
$$x(t) = \frac{1}{N}\sum_{k=-N/2}^{N/2-1} X_a(k)\exp\!\left(j\frac{2\pi}{NT}kt\right),\qquad
X_a(k) = \begin{cases} X(N+k), & -N/2 \leqslant k \leqslant -1,\\ X(k), & 0 \leqslant k \leqslant N/2-1,\end{cases}$$
для $t \in [0,\,(N-1)T]$ с шагом $\Delta t = 0.25T$. Перестановка отсчётов в $X_a(k)$ переносит верхнюю половину ДПФ ($k = N/2 \ldots N-1$) на отрицательные частоты, и сумма берётся по основной полосе $[-f_d/2,\,f_d/2)$.

Амплитудный спектр восстановленного сигнала образуют модули слагаемых $|X_a(k)|/N$ на частотах $f_k = k/(NT) = k\,\Delta f$.

```python
k_a = np.arange(-N // 2, N // 2)
Xa = np.concatenate([X[N // 2:], X[:N // 2]])          # X(N+k) при k<0, X(k) при k>=0

def x_analog(t):
    return (np.exp(2j * np.pi * np.outer(t, k_a) / (N * T)) @ Xa) / N

dt = 0.25 * T
t = np.arange(0, (N - 1) * T + dt / 2, dt)
xt = x_analog(t)
x_true = A1 * np.cos(2 * np.pi * f1 * t + np.pi / 4) + A2 * np.cos(2 * np.pi * f2 * t + np.pi / 8)
print(f'число точек t: {len(t)},  max|Im x(t)| = {np.max(np.abs(xt.imag)):.2e}')
print(f'max|x_восст(t) - x_исх(t)| = {np.max(np.abs(xt.real - x_true)):.2e}')

# амплитудный спектр восстановленного сигнала
f_a = k_a * fs / N
amp_a = np.abs(Xa) / N

fig, ax = plt.subplots(3, 2, figsize=(14, 10))
ax[0, 0].stem(n, xn, basefmt='k-')
ax[0, 0].set(title='Периодическая последовательность x(n)', xlabel='n', ylabel='x(n)')
ax[0, 1].stem(k, np.abs(X), basefmt='k-')
ax[0, 1].set(title='Модуль ДПФ |X(k)|', xlabel='k', ylabel='|X(k)|')
ax[1, 0].plot(t * 1e3, xt.real)
ax[1, 0].set(title='Восстановленный аналоговый сигнал x(t)', xlabel='t, мс', ylabel='x(t)')
ax[1, 1].stem(f_a, amp_a, basefmt='k-')
ax[1, 1].set(title='Амплитудный спектр восстановленного x(t)', xlabel='f, Гц', ylabel='|X_a(k)| / N')
ax[2, 0].plot(t * 1e3, x_true, color='C2')
ax[2, 0].set(title='Исходный аналоговый сигнал x(t)', xlabel='t, мс', ylabel='x(t)')
ax[2, 1].plot(t * 1e3, xt.real, lw=2, label='восстановленный')
ax[2, 1].plot(t * 1e3, x_true, 'r--', lw=1, label='исходный')
ax[2, 1].plot(n * T * 1e3, xn, 'ko', ms=3, label='отсчёты x(nT)')
ax[2, 1].set(title='Сравнение (фрагмент 0…4 мс)', xlabel='t, мс', xlim=(0, 4))
ax[2, 1].legend(fontsize=8)
plt.tight_layout()
plt.show()

nz_a = amp_a > 1e-9
print('линии спектра восстановленного сигнала:', [f'{fv:g} Гц: {v:.3f}' for fv, v in zip(f_a[nz_a], amp_a[nz_a])])
```

#### 6. Интерпретация результатов п. 5

Периодический аналоговый сигнал с периодом $NT$ раскладывается в ряд Фурье с коэффициентами $c_k$ на частотах $k/(NT) = k\Delta f$. Если спектр сигнала ограничен частотой $f_d/2$ (здесь $f_2 = 750 < 2000$ Гц), то $c_k = X_a(k)/N$, то есть модуль ДПФ, делённый на $N$, равен амплитудному спектру аналогового сигнала в основной полосе. На графиках у восстановленного сигнала есть линии $\pm500$ Гц с амплитудой $A_1/2 = 0.505$ и $\pm750$ Гц с амплитудой $A_2/2 = 1.01$, что совпадает с $|X(8)|/N$ и $|X(12)|/N$.

Восстановленный и исходный сигналы совпадают, расхождение порядка $10^{-14}$. Восстановление точное, потому что выполнено условие теоремы Котельникова ($f_d = 4000$ Гц $> 2f_2 = 1500$ Гц) и на интервале $NT$ укладывается целое число периодов обеих гармоник. Восстановленная кривая проходит через все отсчёты $x(nT)$ и плавно интерполирует сигнал между ними.

#### 7. Равенство Парсеваля

```python
E_time = np.sum(np.abs(xn) ** 2)
E_freq = np.sum(np.abs(X) ** 2) / N
print(f'Σ|x(n)|²       = {E_time:.6f}')
print(f'(1/N)·Σ|X(k)|² = {E_freq:.6f}')
print(f'разность       = {abs(E_time - E_freq):.2e}')
```

Равенство Парсеваля выполняется: $\sum|x(n)|^2 = \frac{1}{N}\sum|X(k)|^2 = 163.216$.

Смысл равенства в том, что энергия последовательности на периоде, посчитанная по отсчётам во времени, равна энергии, посчитанной по спектру. ДПФ не добавляет и не теряет энергию, множитель $1/N$ связан только с нормировкой преобразования.
