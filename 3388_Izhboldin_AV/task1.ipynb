## Основные теоретические положения

### Линейные дискретные системы

Линейная дискретная система (ЛДС) обладает свойствами однородности и аддитивности: $F(c_1 x_1 + c_2 x_2) = c_1 F(x_1) + c_2 F(x_2)$. Если сдвиг воздействия во времени приводит только к такому же сдвигу реакции, система стационарна. Все характеристики ниже определяются при нулевых начальных условиях: $x(n) = 0$, $y(n) = 0$ при $n < 0$.

Импульсной характеристикой $h(n)$ называется реакция ЛДС на цифровой единичный импульс

$$u_0(n) = \begin{cases}1, & n = 0, \\ 0, & n \neq 0.\end{cases}$$

Воздействие $x(n)$ и реакция $y(n)$ связаны формулой свёртки:

$$y(n) = \sum_{m=0}^{\infty} h(n-m)\,x(m) = \sum_{m=0}^{\infty} h(m)\,x(n-m).$$

Если длина $h(n)$ равна $N_h$, а длина $x(n)$ равна $N_x$, то длина свёртки $N_y = N_x + N_h - 1$.

Рекурсивное звено 2-го порядка описывается разностным уравнением

$$y(n) = b_0 x(n) + b_1 x(n-1) + b_2 x(n-2) - a_1 y(n-1) - a_2 y(n-2).$$

Система рекурсивна, если хотя бы один коэффициент $a_k \neq 0$, то есть в ней есть обратная связь. Импульсная характеристика рекурсивной системы, вообще говоря, бесконечна (БИХ-система).

### Z-преобразование, передаточная функция, нули и полюсы

$$X(z) = \sum_{n=0}^{\infty} x(n)\,z^{-n}, \qquad H(z) = \frac{Y(z)}{X(z)} = \mathcal{Z}\{h(n)\} = \frac{b_0 + b_1 z^{-1} + b_2 z^{-2}}{1 + a_1 z^{-1} + a_2 z^{-2}}.$$

Нулями $z_{0i}$ называются корни числителя $b_0 z^2 + b_1 z + b_2 = 0$, полюсами $z_{*k}$ называются корни знаменателя $z^2 + a_1 z + a_2 = 0$. Через них передаточная функция записывается произведением простейших множителей:

$$H(z) = b_0\,\frac{(1 - z_{01} z^{-1})(1 - z_{02} z^{-1})}{(1 - z_{*1} z^{-1})(1 - z_{*2} z^{-1})}.$$

При вещественных коэффициентах комплексные нули и полюсы образуют комплексно-сопряжённые пары $r e^{\pm j\varphi}$. Алгебраическая и показательная формы связаны соотношениями

$$z = \xi + j\eta = r e^{j\varphi}, \qquad r = \sqrt{\xi^2 + \eta^2}, \qquad \varphi = \arg z, \qquad \xi = r\cos\varphi, \qquad \eta = r\sin\varphi.$$

Для пары полюсов $r_* e^{\pm j\varphi_*}$ выполняется $a_1 = -2 r_* \cos\varphi_*$, $a_2 = r_*^2$, для пары нулей $r_0 e^{\pm j\varphi_0}$ выполняется $b_1/b_0 = -2 r_0 \cos\varphi_0$, $b_2/b_0 = r_0^2$.

Импульсная характеристика звена с полюсами $r_* e^{\pm j\varphi_*}$ находится обратным Z-преобразованием:

$$h(n) = b_0 r_*^{n}\frac{\sin\big((n+1)\varphi_*\big)}{\sin\varphi_*}\,u_1(n) + b_1 r_*^{n-1}\frac{\sin(n\varphi_*)}{\sin\varphi_*}\,u_1(n-1) + b_2 r_*^{n-2}\frac{\sin\big((n-1)\varphi_*\big)}{\sin\varphi_*}\,u_1(n-2),$$

где $u_1(n)$ обозначает цифровой единичный скачок. Множители $u_1(n-1)$ и $u_1(n-2)$ учитывают нулевые начальные условия: задержанные слагаемые появляются только после прихода импульса на соответствующую линию задержки.

Если разложить передаточную функцию на простые дроби, $H(z) = A_0 + \sum_{k} \frac{A_k}{1 - \alpha_k z^{-1}}$, где $\alpha_k = z_{*k}$ (полюсы), то импульсная характеристика принимает вид $h(n) = A_0 u_0(n) + \sum_{k} A_k \alpha_k^n$.

ЛДС устойчива тогда и только тогда, когда $\sum_{n=0}^{\infty} |h(n)| < \infty$. Для передаточной функции это означает, что все полюсы лежат внутри единичного круга: $|z_{*k}| < 1$.

### Частотные характеристики

Частотная характеристика равна передаточной функции на единичной окружности:

$$H(e^{j\hat\omega}) = H(z)\Big|_{z = e^{j\hat\omega}} = A(\hat\omega)\,e^{j\varphi(\hat\omega)}, \qquad \hat\omega = \omega T = 2\pi\frac{f}{f_d},$$

Здесь $A(\hat\omega) = |H(e^{j\hat\omega})|$ называется АЧХ, $\varphi(\hat\omega) = \arg H(e^{j\hat\omega})$ называется ФЧХ, $f_d$ обозначает частоту дискретизации. Характеристики периодичны с периодом $2\pi$ по $\hat\omega$ (с периодом $f_d$ по $f$), а при вещественных коэффициентах АЧХ чётна и ФЧХ нечётна. Поэтому их рассматривают в основной полосе $\hat\omega \in [0;\ \pi]$, $f \in [0;\ f_d/2]$.

АЧХ можно выразить через расстояния от точки $e^{j\hat\omega}$ до нулей и полюсов:

$$A(\hat\omega) = |b_0|\,\frac{|e^{j\hat\omega} - z_{01}|\cdot|e^{j\hat\omega} - z_{02}|}{|e^{j\hat\omega} - z_{*1}|\cdot|e^{j\hat\omega} - z_{*2}|}.$$

Вблизи аргумента полюса (особенно при $r_*$, близком к 1) АЧХ имеет максимум, вблизи аргумента нуля имеет минимум. Нуль на единичной окружности даёт $A(\hat\omega) = 0$ и скачок ФЧХ на $\pi$.

### Дискретное преобразование Фурье

ДПФ $N$-точечной последовательности и обратное ДПФ (ОДПФ):

$$X(k) = \sum_{n=0}^{N-1} x(n)\,W_N^{nk}, \qquad x(n) = \frac{1}{N}\sum_{k=0}^{N-1} X(k)\,W_N^{-nk}, \qquad W_N^{nk} = e^{-j\frac{2\pi}{N}nk}.$$

Номер отсчёта $n$ называют дискретным нормированным временем, произведение $nT$ называют дискретным временем. Номер $k$ называют дискретной нормированной частотой. Ему соответствуют нормированная частота $\hat\omega_k = 2\pi k/N$ и абсолютная частота $f_k = k\,\Delta f$, где $\Delta f = f_d/N = 1/(NT)$ (шаг по частоте). В круговых частотах шаг равен $\Delta\omega = 2\pi/(NT)$, а отсчётам соответствуют частоты $k\,\Delta\omega$.

* Для периодической последовательности $x_p(n)$ с периодом $N$ ДПФ $X_p(k)$ определяет её дискретный (линейчатый) спектр: $|X_p(k)|$ даёт амплитудный спектр, $\arg X_p(k)$ даёт фазовый спектр. В части 2 индекс $p$ опущен везде, кроме п. 2.4, где периодическая и конечная последовательности сравниваются.
* Спектр конечной последовательности длины $N$, $X(e^{j\hat\omega}) = \sum_{n=0}^{N-1} x(n)\,e^{-j\hat\omega n}$, непрерывен, а ДПФ состоит из его отсчётов: $X(k) = X(e^{j\hat\omega})\big|_{\hat\omega = 2\pi k/N}$.
* Для вещественной $x(n)$: $|X(k)| = |X(N-k)|$, $\arg X(k) = -\arg X(N-k)$.
* Гармоника $A\cos(2\pi k_0 n/N + \varphi_0)$ с целым $k_0$ даёт два ненулевых отсчёта ДПФ: $X(k_0) = \frac{NA}{2}e^{j\varphi_0}$, $X(N-k_0) = \frac{NA}{2}e^{-j\varphi_0}$.

Если выполнено условие теоремы Котельникова $f_d \ge 2 f_{max}$, периодический аналоговый сигнал восстанавливается по ДПФ:

$$x(t) = \frac{1}{N}\sum_{k=-N/2}^{N/2-1} X_a(k)\,e^{j\frac{2\pi}{NT}kt}, \qquad X_a(k) = \begin{cases}X(N+k), & -N/2 \le k \le -1, \\ X(k), & 0 \le k \le N/2-1.\end{cases}$$

Величины $C(k) = X_a(k)/N$ являются комплексными амплитудами ряда Фурье аналогового сигнала на частотах $k/(NT)$.

Равенство Парсеваля для ДПФ:

$$\sum_{n=0}^{N-1} |x(n)|^2 = \frac{1}{N}\sum_{k=0}^{N-1} |X(k)|^2.$$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
import pandas as pd
from IPython.display import display

# Цвета серий и общий стиль графиков
C1, C2, C3 = '#2a78d6', '#eb6834', '#1baf7a'
INK, INK2, GRID = '#0b0b0b', '#52514e', '#e4e3df'
plt.rcParams.update({
    'figure.dpi': 100, 'font.size': 10,
    'axes.edgecolor': INK2, 'axes.labelcolor': INK, 'axes.titlesize': 11,
    'axes.grid': True, 'grid.color': GRID, 'grid.linewidth': 0.8,
    'axes.spines.top': False, 'axes.spines.right': False,
    'xtick.color': INK2, 'ytick.color': INK2,
    'legend.frameon': False, 'lines.linewidth': 2,
})

# Изображение дискретного сигнала "палочками"
def stem(ax, n, y, color=C1, label=None, ms=4):
    ml, sl, _ = ax.stem(n, y, basefmt=' ', label=label)
    plt.setp(sl, color=color, linewidth=1.2)
    plt.setp(ml, color=color, markersize=ms)
    ax.axhline(0, color=INK2, linewidth=0.8)

# Формат чисел в таблицах: до 4 знаков после точки, без лишних нулей
def _num(v):
    s = f'{v:.4f}'
    return '0' if float(s) == 0 else s.rstrip('0').rstrip('.')
pd.set_option('display.float_format', _num)

# Значение для таблицы из двух столбцов "Величина" и "Значение"
def fmt(v):
    if isinstance(v, (bool, np.bool_)):
        return 'да' if v else 'нет'
    if isinstance(v, (int, np.integer)):
        return str(v)
    if isinstance(v, (complex, np.complexfloating)):
        return f'{v.real:.6g} {v.imag:+.6g}j'
    if isinstance(v, (float, np.floating)):
        return f'{v:.1e}' if 0 < abs(v) < 1e-4 else f'{v:.6g}'
    return str(v)

# Таблица из списка пар (название, значение)
def table(rows, name='Величина', value='Значение'):
    df = pd.DataFrame([(r, fmt(v)) for r, v in rows], columns=[name, value])
    return df.set_index(name)
```

### Часть 1. Линейная дискретная система (рекурсивное звено 2-го порядка)

Параметры варианта вычисляются по формулам из таблицы вариантов.

```python
N_var = 7 # номер варианта

# Коэффициенты передаточной функции H(z) = (b0 + b1 z^-1 + b2 z^-2) / (1 + a1 z^-1 + a2 z^-2)
b0 = 0.5 + 0.02 * N_var
b1 = b0 * (-1)**(N_var + 1) * (0.9822 + 0.0178 * N_var)
b2 = b0 * (0.8 + 0.2 * (N_var % 5))
a0 = 1.0
a1 = (-1)**N_var * (0.7778 + 0.025 * N_var)
a2 = 0.64 + 0.006 * N_var

N1 = N_var % 10 + 20 # длина импульсной характеристики
N2 = N_var % 10 + 30 # длина воздействия
fd = 1000 * N_var # частота дискретизации, Гц

b = np.array([b0, b1, b2]) # числитель
a = np.array([a0, a1, a2]) # знаменатель
display(table([
    ('b0', b0), ('b1', b1), ('b2', b2),
    ('a0', a0), ('a1', a1), ('a2', a2),
    ('N1', N1), ('N2', N2), ('fd, Гц', fd),
], 'Параметр'))
```

#### 1.1. Импульсная характеристика

Импульсная характеристика вычисляется как реакция звена на цифровой единичный импульс $u_0(n)$ длины $N_1$ при нулевых начальных условиях. Те же отсчёты рассчитываются по аналитической формуле.

```python
n1 = np.arange(N1)
u0 = np.zeros(N1); u0[0] = 1.0 # цифровой единичный импульс
h = signal.lfilter(b, a, u0) # импульсная характеристика

# Цифровой единичный скачок
def u1(n):
    return (np.asarray(n) >= 0).astype(float)

# Полюсы z* = r* exp(+-j phi*): a1 = -2 r* cos(phi*), a2 = r*^2
r_p = np.sqrt(a2)
phi_p = np.arccos(-a1 / (2 * r_p))

# Аналитическая формула (нулевые начальные условия учтены множителями u1)
h_an = (b0 * r_p**n1 * np.sin((n1 + 1) * phi_p) * u1(n1)
        + b1 * r_p**(n1 - 1.0) * np.sin(n1 * phi_p) * u1(n1 - 1)
        + b2 * r_p**(n1 - 2.0) * np.sin((n1 - 1) * phi_p) * u1(n1 - 2)) / np.sin(phi_p)

# Разложение на простые дроби:
# H(z) = A0 + A1/(1 - alpha1 z^-1) + A2/(1 - alpha2 z^-1)  =>  h(n) = A0 u0(n) + 2|A1| r*^n cos(n phi* + arg A1)
Ak, alpha_k, A0 = signal.residuez(b, a)
i = np.argmax(alpha_k.imag) # полюс с положительной мнимой частью
C_env, theta = 2 * np.abs(Ak[i]), np.angle(Ak[i])
h_cos = A0[0] * u0 + C_env * r_p**n1 * np.cos(phi_p * n1 + theta)

display(table([
    ('r*', r_p), ('φ*, рад', phi_p), ('φ* / π', phi_p / np.pi), ('sin φ*', np.sin(phi_p)),
    ('A0 = b2/a2', A0[0]), ('2|A1|', C_env), ('arg A1, рад', theta),
    ('max |h(n) − формула|', np.abs(h - h_an).max()),
    ('max |h(n) − простые дроби|', np.abs(h - h_cos).max()),
]))
display(pd.DataFrame({'h(n), расчёт': h, 'h(n), формула': h_an, 'h(n), простые дроби': h_cos},
                     index=pd.Index(n1, name='n')))

fig, ax = plt.subplots(figsize=(10, 3.8))
n_env = np.linspace(0, N1 - 1, 300)
ax.plot(n_env, C_env * r_p**n_env, '--', color=INK2, linewidth=1, label=r'огибающая $\pm 2|A_1|\,r_*^n$')
ax.plot(n_env, -C_env * r_p**n_env, '--', color=INK2, linewidth=1)
stem(ax, n1, h, C1, label='h(n): расчёт по передаточной функции')
ax.plot(n1, h_an, 'x', color=C2, ms=8, mew=1.5, label='h(n): аналитическая формула')
ax.set_xlabel('n'); ax.set_ylabel('h(n)')
ax.set_title(f'Импульсная характеристика, N1 = {N1}')
ax.legend(loc='upper right')
plt.show()
```

Радиус и аргумент полюсов: $r_* = \sqrt{a_2} = 0.8258$, $\varphi_* = \arccos\big(-a_1/(2r_*)\big) = 0.9559$ рад $= 0.3043\pi$, $\sin\varphi_* = 0.8168$. Аналитическая формула импульсной характеристики имеет вид

$$h(n) = \frac{1}{0.8168}\Big[0.64\cdot 0.8258^{n}\sin\big(0.9559(n+1)\big)\,u_1(n) + 0.708352\cdot 0.8258^{n-1}\sin(0.9559\,n)\,u_1(n-1) + 0.768\cdot 0.8258^{n-2}\sin\big(0.9559(n-1)\big)\,u_1(n-2)\Big].$$

Эквивалентную компактную форму даёт разложение на простые дроби: $H(z) = A_0 + \frac{A_1}{1 - \alpha_1 z^{-1}} + \frac{A_2}{1 - \alpha_2 z^{-1}}$, где $\alpha_{1,2} = z_{*1,2}$, откуда $h(n) = A_0 u_0(n) + A_1 \alpha_1^n + A_2 \alpha_2^n$. Степени числителя и знаменателя равны, поэтому есть постоянное слагаемое $A_0 = b_2/a_2$. Полюсы $\alpha_{1,2}$ и коэффициенты $A_{1,2}$ комплексно-сопряжённые, поэтому сумма двух последних слагаемых равна $2|A_1|\,r_*^n\cos(\varphi_* n + \arg A_1)$:

$$h(n) = 1.1261\,u_0(n) + 2.3482\cdot 0.8258^{n}\cos(0.9559\,n - 1.7793)\,u_1(n).$$

Обе формулы совпадают с расчётом по передаточной функции с точностью до ошибок округления: расхождение порядка $10^{-16}$, отметки "×" на графике лежат на вершинах "палочек".

Звено рекурсивное ($a_1, a_2 \neq 0$), поэтому $h(n)$ представляет собой затухающую гармоническую последовательность с огибающей $\pm2.3482\cdot 0.8258^n$. Огибающая не обращается в ноль ни при каком конечном $n$, так что в действительности импульсная характеристика бесконечна (БИХ-система), а $N_1 = 27$ означает только число вычисленных отсчётов. Последний из них, $|h(26)| = 0.0076$, составляет 0.48 % от максимума $\max|h(n)| = 1.5874$ (при $n = 2$). Скорость затухания задаёт радиус полюсов $r_* = 0.8258 < 1$, а частоту колебаний задаёт их аргумент: период колебаний $2\pi/\varphi_* \approx 6.57$ отсчёта.

#### 1.2. Реакция по формуле свёртки

Воздействием служит прямоугольный импульс длины $N_2 = 37$. Реакция вычисляется по формуле свёртки $y_1(n) = \sum_{m} h(m)\,x(n-m)$ с импульсной характеристикой длины $N_1$ из п. 1.1.

```python
# округление до ближайшего целого (N2/2 = 18.5 -> 19)
L_imp = int(np.ceil(N2 / 2))
n2 = np.arange(N2)
x = np.where(n2 < L_imp, 1.0, 0.0) # воздействие

# Формула свёртки: y1(n) = sum_m h(m) x(n - m)
Ny_full = N1 + N2 - 1
y1_full = np.zeros(Ny_full)
for nn in range(Ny_full):
    for m in range(N1):
        if 0 <= nn - m < N2:
            y1_full[nn] += h[m] * x[nn - m]
conv_ok = np.allclose(y1_full, np.convolve(x, h)) # проверка библиотечной свёрткой

y1 = y1_full[:N2] # свёртка, ограниченная длиной воздействия
last_nz = np.nonzero(np.abs(y1_full) > 1e-12)[0].max()
display(table([
    ('Длина импульса [N2/2]', L_imp),
    ('Длина воздействия N2', N2),
    ('Длина свёртки N1 + N2 − 1', Ny_full),
    ('Последний ненулевой отсчёт свёртки, n', last_nz),
    ('Число ненулевых отсчётов свёртки', last_nz + 1),
    ('Совпадение с библиотечной свёрткой', conv_ok),
]))

fig, axs = plt.subplots(3, 1, figsize=(10, 8.5))
stem(axs[0], n2, x, C1)
axs[0].set_title(f'Воздействие x(n): прямоугольный импульс длины {L_imp}, N2 = {N2}')
axs[0].set_ylabel('x(n)')

stem(axs[1], np.arange(Ny_full), y1_full, C1)
axs[1].axvspan(N2 - 0.5, Ny_full - 0.5, color=GRID, alpha=0.6, lw=0)
axs[1].text(N2 + 0.5, y1_full.max() * 0.85, 'n ≥ N2: отбрасывается', color=INK2)
axs[1].axvline(last_nz, color=C2, ls='--', lw=1)
axs[1].text(last_nz + 0.5, y1_full.max() * 0.55, f'последний ненулевой\nотсчёт: n = {last_nz}', color=INK2)
axs[1].set_title(f'Полная свёртка x(n) * h(n), длина N1 + N2 − 1 = {Ny_full}')
axs[1].set_ylabel('y(n)')

stem(axs[2], n2, y1, C1)
axs[2].set_title(f'Реакция y1(n), ограниченная длиной воздействия N2 = {N2}')
axs[2].set_ylabel('y1(n)')
for ax in axs:
    ax.set_xlabel('n'); ax.set_xlim(-1, Ny_full)
fig.tight_layout()
plt.show()
```

Импульс $x$ содержит 19 единичных отсчётов ($0 \le n \le 18$). Значение $N_2/2 = 18.5$ лежит ровно между целыми, и при округлении до ближайшего целого половина округляется вверх: $[18.5] = 19$. Оставшиеся 18 отсчётов воздействия нулевые, общая длина воздействия $N_2 = 37$.

Аналитически длина свёртки равна $N_1 + N_2 - 1 = 27 + 37 - 1 = 63$ отсчёта ($0 \le n \le 62$). На графике ненулевые значения заканчиваются на $n = 44$, то есть ненулевая часть содержит 45 $= 19 + 27 - 1$ отсчётов. Последнее ненулевое произведение $h(26)\,x(18)$ попадает в момент $n = 44$, а 18 нулевых отсчётов в конце воздействия формально удлиняют свёртку, но вклада в неё не дают.

Свёртку ограничивают длиной воздействия по двум причинам. При обработке в реальном времени очередной отсчёт реакции вычисляется при поступлении очередного отсчёта воздействия, поэтому реакция наблюдается на том же интервале из $N_2$ отсчётов. Так же работает разностное уравнение в п. 1.3, и одинаковая длина позволяет сравнивать $y_1(n)$ и $y_2(n)$ поотсчётно. Кроме того, "хвост" полной свёртки появился из-за усечения $h(n)$: истинная реакция БИХ-системы после $n = 44$ продолжает затухать, а при $n \ge N_1$ свёртка с усечённой $h(n)$ уже неточна.

Пока действует импульс, реакция нарастает с перерегулированием (максимум 4.159 при $n = 3$) и колебаниями, затем устанавливается около 2.901 (при $n = 18$). Это значение близко к коэффициенту передачи на нулевой частоте $H(1) = \sum_n h(n) = 2.902$. После окончания импульса ($n \ge 19$) идут затухающие колебания (минимум -1.235 при $n = 22$) с частотой, которую задаёт аргумент полюсов.

#### 1.3. Реакция по разностному уравнению

Разностное уравнение рекурсивного звена 2-го порядка с коэффициентами варианта:

$$y(n) = 0.64\,x(n) + 0.708352\,x(n-1) + 0.768\,x(n-2) + 0.9528\,y(n-1) - 0.682\,y(n-2).$$

Реакция $y_2(n)$ вычисляется рекуррентно при нулевых начальных условиях на то же воздействие $x(n)$ из п. 1.2.

```python
# Реакция по разностному уравнению y(n) = sum b_i x(n-i) - sum a_k y(n-k), нулевые начальные условия
def diff_eq(b, a, x):
    y = np.zeros(len(x))
    for nn in range(len(x)):
        acc = 0.0
        for i in range(len(b)):
            if nn - i >= 0:
                acc += b[i] * x[nn - i]
        for k in range(1, len(a)):
            if nn - k >= 0:
                acc -= a[k] * y[nn - k]
        y[nn] = acc / a[0]
    return y

y2 = diff_eq(b, a, x)
d = np.abs(y1 - y2)
h_long = signal.lfilter(b, a, np.r_[1.0, np.zeros(199)]) # более длинный фрагмент h(n)
display(table([
    ('Длина воздействия', len(x)),
    ('Длина реакции y2', len(y2)),
    ('max |y1 − y2| при n < N1', d[:N1].max()),
    ('max |y1 − y2| при n ≥ N1', d[N1:].max()),
    ('n, где расхождение максимально', N1 + d[N1:].argmax()),
    ('max |h(n)| при n ≥ N1', np.abs(h_long[N1:]).max()),
    ('Совпадение с библиотечным расчётом', np.allclose(y2, signal.lfilter(b, a, x))),
]))

fig, axs = plt.subplots(2, 1, figsize=(10, 5.6))
stem(axs[0], n2, x, C1)
axs[0].set_title('Воздействие x(n)'); axs[0].set_ylabel('x(n)')
stem(axs[1], n2, y2, C2)
axs[1].set_title('Реакция y2(n) по разностному уравнению'); axs[1].set_ylabel('y2(n)')
for ax in axs:
    ax.set_xlabel('n')
fig.tight_layout()
plt.show()

fig, axs = plt.subplots(2, 1, figsize=(10, 5.6), gridspec_kw={'height_ratios': [2, 1]})
stem(axs[0], n2, y1, C1, label='y1(n): свёртка')
axs[0].plot(n2, y2, 'x', color=C2, ms=8, mew=1.5, label='y2(n): разностное уравнение')
axs[0].axvline(N1 - 0.5, color=INK2, ls=':', lw=1)
axs[0].text(N1 + 0.3, y2.max() * 0.45, 'n ≥ N1', color=INK2)
axs[0].set_title('Сравнение реакций y1(n) и y2(n)'); axs[0].legend(loc='upper right')
stem(axs[1], n2, y1 - y2, C3)
axs[1].set_title('Разность y1(n) − y2(n)')
for ax in axs:
    ax.set_xlabel('n')
fig.tight_layout()
plt.show()
```

Длина воздействия $N_2 = 37$, длина реакции по разностному уравнению тоже 37: каждый новый отсчёт воздействия даёт ровно один новый отсчёт реакции. Рекуррентная формула ничего не усекает, поэтому $y_2(n)$ содержит точные первые 37 отсчётов бесконечной реакции рекурсивного звена.

При $n < N_1 = 27$ реакции $y_1(n)$ и $y_2(n)$ совпадают с точностью до ошибок округления ($\max|y_1 - y_2| = 1.8\cdot 10^{-15}$). Начиная с $n = 27$ они расходятся, сильнее всего при $n = 29$ (на 0.0233). Свёртка использует импульсную характеристику, усечённую до $N_1$ отсчётов, и слагаемые $h(m)\,x(n-m)$ с $m \ge 27$ в неё не входят. Отброшенные отсчёты малы ($|h(n)| \le 0.0109$ при $n \ge 27$), поэтому и расхождение мало. Точной является реакция $y_2(n)$, а $y_1(n)$ приближается к ней с ростом $N_1$.

#### 1.4. Нули и полюсы передаточной функции

Нули находятся как корни уравнения $0.64z^2 + 0.708352z + 0.768 = 0$, полюсы как корни уравнения $z^2 - 0.9528z + 0.682 = 0$ (числитель и знаменатель $H(z)$ умножены на $z^2$).

```python
z0 = np.roots(b) # нули
zp = np.roots(a) # полюсы
z0 = z0[np.argsort(-z0.imag)]
zp = zp[np.argsort(-zp.imag)]

roots = np.r_[z0, zp]
display(pd.DataFrame({
    'Re': roots.real, 'Im': roots.imag,
    'Модуль r': np.abs(roots), 'Аргумент, рад': np.angle(roots), 'Аргумент / π': np.angle(roots) / np.pi,
}, index=['Нуль 1', 'Нуль 2', 'Полюс 1', 'Полюс 2']))

# Связь показательной формы с коэффициентами
r0_f, phi0_f = np.sqrt(b2 / b0), np.arccos(-b1 / (2 * np.sqrt(b0 * b2)))
rp_f, phip_f = np.sqrt(a2), np.arccos(-a1 / (2 * np.sqrt(a2)))
display(table([
    ('r0 = sqrt(b2/b0)', r0_f), ('φ0 / π', phi0_f / np.pi),
    ('r* = sqrt(a2)', rp_f), ('φ* / π', phip_f / np.pi),
]))

# Проверка представления H(z) произведением простейших множителей в произвольной точке
z_t = 0.7 + 0.4j
H_frac = (b0 + b1 / z_t + b2 / z_t**2) / (1 + a1 / z_t + a2 / z_t**2)
H_prod = b0 * np.prod(1 - z0 / z_t) / np.prod(1 - zp / z_t)
display(table([('H(z_t) по исходной дроби', H_frac), ('H(z_t) по произведению множителей', H_prod)]))
```

| | Алгебраическая форма | Показательная форма | Аргумент относительно $\pi$ |
|:---:|:---:|:---:|:---:|
| Нули $z_{01,2}$ | $-0.5534 \pm j\,0.9454$ | $1.0954\,e^{\pm j\,2.1004}$ | $\pm 0.6686\pi$ |
| Полюсы $z_{*1,2}$ | $0.4764 \pm j\,0.6746$ | $0.8258\,e^{\pm j\,0.9559}$ | $\pm 0.3043\pi$ |

Модуль и аргумент вычисляются по вещественной и мнимой частям: $r = \sqrt{\xi^2 + \eta^2}$, $\varphi = \mathrm{arctg}(\eta/\xi)$ при $\xi > 0$. У нулей $\xi < 0$, поэтому к арктангенсу прибавляется $\pi$ (при $\eta > 0$) или вычитается $\pi$ (при $\eta < 0$): $\mathrm{arctg}\big(0.9454/(-0.5534)\big) + \pi = 2.1004$. Обратно, $\xi = r\cos\varphi$, $\eta = r\sin\varphi$. Например, для полюса $\sqrt{0.4764^2 + 0.6746^2} = 0.8258$. Коэффициенты $H(z)$ вещественны, поэтому нули и полюсы образуют комплексно-сопряжённые пары с одинаковыми модулями и противоположными аргументами. Показательная форма выражается прямо через коэффициенты: $r_* = \sqrt{a_2} = 0.8258$, $\cos\varphi_* = -a_1/(2\sqrt{a_2})$, для нулей $r_0 = \sqrt{b_2/b_0} = \sqrt{1.2} = 1.0954$, $\cos\varphi_0 = -b_1/(2\sqrt{b_0 b_2})$.

Передаточная функция в виде произведения простейших множителей:

$$H(z) = 0.64\,\frac{\big(1 - 1.0954\,e^{j0.6686\pi}z^{-1}\big)\big(1 - 1.0954\,e^{-j0.6686\pi}z^{-1}\big)}{\big(1 - 0.8258\,e^{j0.3043\pi}z^{-1}\big)\big(1 - 0.8258\,e^{-j0.3043\pi}z^{-1}\big)}.$$

В контрольной точке значения $H(z)$ по исходной дроби и по произведению совпадают.

#### 1.6. АЧХ и ФЧХ в шкале нормированных частот

Частотная характеристика $H(e^{j\hat\omega})$ вычисляется в основной полосе $\hat\omega \in [0;\ \pi]$. Для анализа подавления частотных составляющих дополнительно сравниваются спектры воздействия $X(e^{j\hat\omega})$ и реакции $Y(e^{j\hat\omega}) = H(e^{j\hat\omega})\,X(e^{j\hat\omega})$.

```python
w = np.linspace(0, np.pi, 4096) # нормированная частота, рад
_, Hw = signal.freqz(b, a, worN=w)
A_w, Phi_w = np.abs(Hw), np.angle(Hw) # АЧХ и ФЧХ 

i_max, i_min = A_w.argmax(), A_w.argmin()
jumps = np.where(np.abs(np.diff(Phi_w)) > np.pi)[0]
points = [('ω̂ = 0', 0), ('Максимум АЧХ', i_max), ('Минимум АЧХ', i_min), ('ω̂ = π', len(w) - 1)]
display(pd.DataFrame({
    'ω̂ / π': [w[j] / np.pi for _, j in points],
    'A(ω̂)': [A_w[j] for _, j in points],
    'φ(ω̂), рад': [Phi_w[j] for _, j in points],
}, index=pd.Index([p for p, _ in points], name='Точка')))
rows = [('Аргумент полюса / π', np.angle(zp[0]) / np.pi), ('Аргумент нуля / π', np.angle(z0[0]) / np.pi)]
for j in jumps:
    rows += [('Скачок ФЧХ, рад', Phi_w[j + 1] - Phi_w[j]), ('Скачок ФЧХ при ω̂ / π', w[j] / np.pi)]
display(table(rows))

fig, axs = plt.subplots(2, 1, figsize=(10, 6.5), sharex=True)
axs[0].plot(w / np.pi, A_w, color=C1)
axs[0].axvline(np.angle(zp[0]) / np.pi, color=C2, ls='--', lw=1, label=r'аргумент полюса $\varphi_*$')
axs[0].axvline(np.angle(z0[0]) / np.pi, color=C3, ls='--', lw=1, label=r'аргумент нуля $\varphi_0$')
axs[0].plot(w[i_max] / np.pi, A_w[i_max], 'o', color=C1, ms=6)
axs[0].annotate(f'max = {A_w[i_max]:.3f}', (w[i_max] / np.pi, A_w[i_max]), xytext=(8, -4), textcoords='offset points')
axs[0].plot(w[i_min] / np.pi, A_w[i_min], 'o', color=C1, ms=6)
axs[0].annotate(f'min = {A_w[i_min]:.3f}', (w[i_min] / np.pi, A_w[i_min]), xytext=(8, 8), textcoords='offset points')
axs[0].set_ylabel(r'$A(\hat{\omega})$'); axs[0].set_title('АЧХ'); axs[0].legend(loc='upper right')

axs[1].plot(w / np.pi, Phi_w, color=C1, label='ФЧХ (главное значение аргумента)')
axs[1].plot(w / np.pi, np.unwrap(Phi_w), '--', color=INK2, lw=1, label='ФЧХ без скачков на 2π')
axs[1].set_ylabel(r'$\varphi(\hat{\omega})$, рад'); axs[1].set_title('ФЧХ')
axs[1].set_xlabel(r'$\hat{\omega}/\pi$'); axs[1].legend(loc='lower left')
fig.tight_layout()
plt.show()

# Спектры воздействия и реакции
_, Xw = signal.freqz(x, 1, worN=w)
Yw = Hw * Xw
# Доля энергии спектра в полосе w >= pi/2
def hf_share(S):
    P = np.abs(S)**2
    return P[w >= np.pi / 2].sum() / P.sum()
display(table([('Воздействие x(n)', 100 * hf_share(Xw)), ('Реакция y(n)', 100 * hf_share(Yw))],
              'Сигнал', 'Доля энергии в полосе [π/2; π], %'))

fig, ax = plt.subplots(figsize=(10, 3.8))
ax.plot(w / np.pi, 20 * np.log10(np.abs(Xw) + 1e-12), color=C1, lw=1.5, label='|X|, воздействие')
ax.plot(w / np.pi, 20 * np.log10(np.abs(Yw) + 1e-12), color=C2, lw=1.5, label='|Y| = A·|X|, реакция')
ax.set_ylim(-60, 45); ax.set_xlabel(r'$\hat{\omega}/\pi$'); ax.set_ylabel('дБ')
ax.set_title('Модули спектров воздействия и реакции'); ax.legend(loc='upper right')
plt.show()
```

Основная полоса частот: $\hat\omega \in [0;\ \pi]$. За её пределами частотная характеристика повторяется с периодом $2\pi$, а на отрицательных частотах АЧХ чётна и ФЧХ нечётна.

Полюсы с аргументом $\pm0.3043\pi$ лежат близко к единичной окружности ($r_* = 0.8258$), и АЧХ имеет резонансный максимум 5.985 при $\hat\omega \approx 0.2911\pi$. Максимум сдвинут немного левее аргумента полюса из-за сопряжённого полюса и нулей. Нули с аргументом $\pm0.6686\pi$ лежат снаружи вблизи окружности ($r_0 = 1.0954$), и им соответствует глубокий минимум 0.0603 при $\hat\omega \approx 0.6713\pi$. До нуля АЧХ не опускается, потому что нули не лежат на единичной окружности. На краях полосы $A(0) = 2.902$, $A(\pi) = 0.266$.

Скачок ФЧХ на $\pi$ соответствует нулевому значению АЧХ: при прохождении нуля, лежащего на единичной окружности, вещественная амплитудная функция меняет знак. В этом варианте АЧХ нигде не равна нулю, поэтому скачка на $\pi$ нет. Скачок на $6.282$ рад $\approx 2\pi$ при $\hat\omega \approx 0.5497\pi$ возникает из-за того, что аргумент вычисляется в интервале $(-\pi;\ \pi]$. ФЧХ без таких скачков (штриховая кривая) непрерывна. Быстрее всего фаза меняется около аргументов полюса и нуля.

Звено усиливает низкие и средние частоты: до $\hat\omega \approx 0.47\pi$ коэффициент передачи больше 1, максимум около $0.29\pi$. Высокие частоты ослабляются: на $\hat\omega = \pi$ коэффициент передачи равен 0.266, около $0.67\pi$ он падает до 0.0603. Поэтому в реакции подавлены в основном высокочастотные составляющие воздействия: доля энергии спектра в полосе $[\pi/2;\ \pi]$ уменьшилась с 1.58 % у воздействия до 0.011 % у реакции. Высокочастотные составляющие формируют резкие фронты прямоугольного импульса, и в реакции фронты сглажены, а составляющие около резонансной частоты $0.29\pi$ дают затухающие колебания, видные в п. 1.2 и 1.3.

#### 1.7. АЧХ и ФЧХ в шкале частот $f$ (Гц)

Переход от нормированной частоты к абсолютной: $f = \dfrac{\hat\omega}{2\pi}f_d$, $f_d = 7000$ Гц.

```python
f_hz = w * fd / (2 * np.pi) # частота, Гц

display(table([
    ('Верхняя граница основной полосы fd/2, Гц', fd / 2),
    ('Частота максимума АЧХ, Гц', f_hz[i_max]),
    ('Частота минимума АЧХ, Гц', f_hz[i_min]),
]))
wk = np.array([0, 0.25, 0.5, 0.75, 1.0])
display(pd.DataFrame({'f, Гц': wk * fd / 2}, index=pd.Index(wk, name='ω̂ / π')))

fig, axs = plt.subplots(2, 1, figsize=(10, 6.5), sharex=True)
axs[0].plot(f_hz, A_w, color=C1)
axs[0].plot(f_hz[i_max], A_w[i_max], 'o', color=C1, ms=6)
axs[0].annotate(f'{f_hz[i_max]:.0f} Гц', (f_hz[i_max], A_w[i_max]), xytext=(8, -4), textcoords='offset points')
axs[0].plot(f_hz[i_min], A_w[i_min], 'o', color=C1, ms=6)
axs[0].annotate(f'{f_hz[i_min]:.0f} Гц', (f_hz[i_min], A_w[i_min]), xytext=(8, 8), textcoords='offset points')
axs[0].set_ylabel('A(f)'); axs[0].set_title('АЧХ')
axs[1].plot(f_hz, Phi_w, color=C1)
axs[1].set_ylabel('φ(f), рад'); axs[1].set_title('ФЧХ'); axs[1].set_xlabel('f, Гц')
axs[1].set_xlim(0, fd / 2)
fig.tight_layout()
plt.show()
```

Основная полоса в герцах: $f \in [0;\ f_d/2] = [0;\ 3500]$ Гц, верхняя граница равна частоте Найквиста.

Частоты связаны соотношением $\hat\omega = 2\pi f/f_d$: $\hat\omega = 0$ соответствует $f = 0$, $\hat\omega = \pi/2$ соответствует 1750 Гц, $\hat\omega = \pi$ соответствует $f_d/2 = 3500$ Гц. Графики в двух шкалах отличаются только масштабом оси абсцисс. Максимум АЧХ при $0.2911\pi$ приходится на 1019 Гц, минимум при $0.6713\pi$ на 2350 Гц.

### Часть 2. Дискретное преобразование Фурье

```python
N = 64 # период (длина) последовательности
fd2 = 2000 * (N_var % 5 + 1) # частота дискретизации, Гц
T = 1 / fd2 # период дискретизации, с
A1 = 1 + 0.01 * N_var
A2 = 2 * A1
f1 = fd2 / 8
f2 = 1.5 * f1
k1, k2 = N * f1 / fd2, N * f2 / fd2 # номера отсчётов ДПФ, соответствующие f1 и f2

display(table([
    ('N', N), ('fd, Гц', fd2), ('T, мс', T * 1e3),
    ('A1', A1), ('A2', A2), ('f1, Гц', f1), ('f2, Гц', f2),
    ('f1 / fd', f1 / fd2), ('f2 / fd', f2 / fd2), ('k1', k1), ('k2', k2),
], 'Параметр'))
```

#### 2.1. Последовательность на периоде $N$

```python
n = np.arange(N) # дискретное нормированное время
nT = n * T # дискретное время, с
xn = A1 * np.cos(2 * np.pi * f1 * nT + np.pi / 4) + A2 * np.cos(2 * np.pi * f2 * nT + np.pi / 8)

fig, axs = plt.subplots(2, 1, figsize=(10, 6))
stem(axs[0], nT * 1e3, xn, C1, ms=3)
axs[0].set_xlabel('nT, мс'); axs[0].set_ylabel('x(nT)')
axs[0].set_title('Последовательность в шкале дискретного времени nT')
stem(axs[1], n, xn, C1, ms=3)
axs[1].set_xlabel('n'); axs[1].set_ylabel('x(n)')
axs[1].set_title('Последовательность в шкале дискретного нормированного времени n')
fig.tight_layout()
plt.show()
display(table([('Последний отсчёт (N−1)T, мс', nT[-1] * 1e3), ('Период NT, мс', N * T * 1e3)]))
```

Последовательность состоит из двух дискретных гармоник с нормированными частотами $f_1/f_d = 1/8$ и $f_2/f_d = 3/16$. На периоде $N = 64$ укладывается ровно $k_1 = 8$ периодов первой гармоники и $k_2 = 12$ периодов второй. Общий период суммы равен 16 отсчётам, поэтому на графике повторяются 4 одинаковых фрагмента. Графики в двух шкалах совпадают по форме и различаются только подписями оси времени: $nT$ от 0 до $63T = 10.5$ мс и $n$ от 0 до 63.

#### 2.2. ОДПФ от ДПФ последовательности

ДПФ вычисляется по определению, умножением на матрицу $W_N^{nk}$, затем к результату применяется ОДПФ.

```python
k = np.arange(N) # дискретная нормированная частота
W = np.exp(-2j * np.pi * np.outer(k, n) / N) # матрица W_N^{nk}
X = W @ xn # ДПФ по определению

x_idft = (W.conj() @ X) / N # ОДПФ: x(n) = (1/N) sum X(k) W_N^{-nk}
display(table([
    ('ДПФ по определению совпадает с библиотечным', np.allclose(X, np.fft.fft(xn))),
    ('max |Im ОДПФ|', np.abs(x_idft.imag).max()),
    ('max |ОДПФ − x(n)|', np.abs(x_idft.real - xn).max()),
]))

fig, ax = plt.subplots(figsize=(10, 3.6))
stem(ax, n, x_idft.real, C2, ms=3, label='ОДПФ{ДПФ{x(n)}}')
ax.plot(n, xn, 'x', color=C1, ms=6, mew=1.2, label='исходная x(n)')
ax.set_xlabel('n'); ax.set_ylabel('x(n)')
ax.set_title('Последовательность, восстановленная ОДПФ')
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.18), ncol=2)
plt.show()
```

ОДПФ от ДПФ возвращает исходную последовательность: отличие не превышает $2.2 \cdot 10^{-14}$, мнимая часть результата не больше $2.9 \cdot 10^{-14}$. Оба значения объясняются ошибками округления, так как прямое и обратное ДПФ взаимно однозначны.

#### 2.3. Амплитудный и фазовый спектры

```python
X_abs = np.abs(X) # амплитудный спектр
tol = 1e-9 * X_abs.max()
X_arg = np.where(X_abs > tol, np.angle(X), 0.0) # фазовый спектр (шумовые отсчёты обнулены)
f_k = k * fd2 / N # абсолютная частота, Гц
df2 = fd2 / N
nz = np.nonzero(X_abs > tol)[0] # ненулевые отсчёты спектра
display(pd.DataFrame({'f, Гц': f_k[nz], '|X(k)|': X_abs[nz], 'arg X(k) / π': X_arg[nz] / np.pi},
                     index=pd.Index(nz, name='k')))
display(table([('Шаг по частоте Δf = fd/N, Гц', df2), ('N·A1/2', N * A1 / 2), ('N·A2/2', N * A2 / 2)]))

fig, axs = plt.subplots(2, 2, figsize=(11, 6.5))
stem(axs[0, 0], k, X_abs, C1, ms=3)
axs[0, 0].set_xlabel('k'); axs[0, 0].set_ylabel('|X(k)|')
axs[0, 0].set_title('Амплитудный спектр, шкала k')
stem(axs[0, 1], f_k, X_abs, C1, ms=3)
axs[0, 1].set_xlabel('f, Гц'); axs[0, 1].set_ylabel('|X(f)|')
axs[0, 1].set_title('Амплитудный спектр, шкала f')
stem(axs[1, 0], k, X_arg / np.pi, C2, ms=3)
axs[1, 0].set_xlabel('k'); axs[1, 0].set_ylabel('arg X(k) / π')
axs[1, 0].set_title('Фазовый спектр, шкала k')
stem(axs[1, 1], f_k, X_arg / np.pi, C2, ms=3)
axs[1, 1].set_xlabel('f, Гц'); axs[1, 1].set_ylabel('arg X(f) / π')
axs[1, 1].set_title('Фазовый спектр, шкала f')
for ax in axs[:, 0]:
    ax.axvline(N / 2, color=INK2, ls=':', lw=1)
for ax in axs[:, 1]:
    ax.axvline(fd2 / 2, color=INK2, ls=':', lw=1)
fig.tight_layout()
plt.show()
```

Дискретное нормированное время связано с дискретным временем соотношением $n = nT/T$: моменты $nT$ в секундах делятся на период дискретизации $T = 1/6000$ с $\approx 0.1667$ мс и превращаются в целые номера отсчётов $n = 0, 1, \dots, 63$.

Частота $f$ связана с дискретной нормированной частотой $k$ соотношением $f_k = k\,f_d/N = k\,\Delta f$, где $\Delta f = 6000/64 = 93.75$ Гц. Нормированная частота при этом равна $\hat\omega_k = 2\pi k/N$. Гармоники $f_1 = 750$ Гц и $f_2 = 1125$ Гц попадают точно на отсчёты $k_1 = 8$ и $k_2 = 12$. Отсчётам $k = 52$ и $k = 56$ соответствуют частоты 4875 и 5250 Гц, равные $f_d - f_2$ и $f_d - f_1$. Спектр периодичен с периодом $f_d$, так что это отрицательные частоты $-f_2$ и $-f_1$. Пунктир на графиках отмечает границу основной полосы $k = N/2$, $f = f_d/2 = 3000$ Гц.

В амплитудном спектре четыре ненулевые линии: $|X(8)| = |X(56)| = NA_1/2 = 34.24$ и $|X(12)| = |X(52)| = NA_2/2 = 68.48$. Остальные отсчёты равны нулю, так как на периоде $N$ укладывается целое число периодов каждой гармоники и растекания спектра нет. Амплитудный спектр чётен: $|X(k)| = |X(N-k)|$.

Фазовый спектр нечётен: $\arg X(8) = \pi/4$ и $\arg X(12) = \pi/8$ совпадают с начальными фазами гармоник, $\arg X(56) = -\pi/4$, $\arg X(52) = -\pi/8$. В отсчётах с нулевой амплитудой фаза не определена (значения порядка $10^{-14}$ дают случайный аргумент), поэтому там она принята равной нулю.

#### 2.4. ДПФ конечной последовательности и амплитудный спектр периодической последовательности

Конечная последовательность состоит из тех же $N = 64$ отсчётов $x(n)$ и равна нулю вне интервала $0 \le n \le N-1$. Её непрерывный спектр $X(e^{j\hat\omega})$ рассчитывается на плотной сетке частот: ДПФ с дополнением нулями до $M = 4096$ точек. Периодическая последовательность получена периодическим продолжением $x(n)$, её амплитудный спектр вычисляется по одному периоду.

```python
X_fin = np.fft.fft(xn) # N-точечное ДПФ конечной последовательности
M = 4096
X_cont = np.fft.fft(xn, M) # спектр конечной последовательности на сетке w = 2*pi*m/M
k_cont = np.arange(M) * N / M # та же частота в единицах k (w = 2*pi*k/N)

x_per = np.tile(xn, 3) # три периода периодической последовательности
X_per = np.fft.fft(x_per[N:2 * N]) # ДПФ по одному периоду

display(table([
    ('|ДПФ конечной| совпадает с отсчётами непрерывного спектра', np.allclose(np.abs(X_cont[::M // N]), np.abs(X_fin))),
    ('|ДПФ конечной| совпадает с амплитудным спектром периодической', np.allclose(np.abs(X_fin), np.abs(X_per))),
], 'Проверка', 'Результат'))

fig, axs = plt.subplots(2, 1, figsize=(10, 6.5), sharex=True)
axs[0].plot(k_cont, np.abs(X_cont), color=INK2, lw=1, label=r'$|X(e^{j\hat{\omega}})|$, непрерывный спектр')
stem(axs[0], k, np.abs(X_fin), C1, ms=4, label='|X(k)|, ДПФ конечной последовательности')
axs[0].set_ylabel('|X|'); axs[0].legend(loc='upper center')
axs[0].set_title('Модуль ДПФ конечной последовательности')
stem(axs[1], k, np.abs(X_per), C2, ms=4)
axs[1].set_ylabel('|Xp(k)|'); axs[1].set_xlabel(r'k  (нормированная частота $\hat{\omega} = 2\pi k/N$)')
axs[1].set_title('Амплитудный спектр периодической последовательности')
fig.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(10, 3.4))
ax.plot(k_cont, np.abs(X_cont), color=INK2, lw=1)
stem(ax, k, np.abs(X_fin), C1, ms=4)
ax.set_xlim(0, 20); ax.set_xlabel('k'); ax.set_ylabel('|X|')
ax.set_title('Фрагмент 0 ≤ k ≤ 20: отсчёты ДПФ на непрерывном спектре')
plt.show()
```

Модуль ДПФ конечной последовательности совпадает с амплитудным спектром периодической последовательности отсчёт в отсчёт, $|X(k)| = |X_p(k)|$, хотя смысл у этих величин разный. Спектр периодической последовательности дискретный (линейчатый): он существует только на частотах $\hat\omega_k = 2\pi k/N$, и ДПФ описывает его полностью. Спектр конечной последовательности непрерывен: серая кривая на графике складывается из смещённых "лепестков" вида $\sin(N\hat\omega/2)/\sin(\hat\omega/2)$. ДПФ конечной последовательности даёт $N$ равноотстоящих отсчётов этого спектра на частотах $\hat\omega_k = 2\pi k/N$. Номера $k_1 = 8$ и $k_2 = 12$ целые, поэтому отсчёты ДПФ попадают на вершины главных лепестков и в нули боковых. В результате ненулевыми оказываются только 4 отсчёта, хотя непрерывный спектр конечной последовательности отличен от нуля почти на всех частотах.

#### 2.5. Восстановление аналогового сигнала по ДПФ

Аналоговый сигнал восстанавливается по формуле с отсчётами $X_a(k)$, $k = -N/2, \dots, N/2-1$, на интервале $t \in [0;\ (N-1)T]$ с шагом $\Delta t = 0.25T$.

```python
dt = 0.25 * T
t = np.arange(0, (N - 1) * T + dt / 2, dt) # 253 момента времени
k_a = np.arange(-N // 2, N // 2)
Xa = np.where(k_a < 0, X[(N + k_a) % N], X[k_a % N]) # X(N+k) при k < 0, X(k) при k >= 0

# x(t) = (1/N) sum_k Xa(k) exp(j 2 pi k t / (N T))
x_rec_c = (np.exp(2j * np.pi * np.outer(t, k_a) / (N * T)) @ Xa) / N
x_rec = x_rec_c.real

# Исходный аналоговый сигнал x(t) = x(nT)|_{nT = t}
x_an = A1 * np.cos(2 * np.pi * f1 * t + np.pi / 4) + A2 * np.cos(2 * np.pi * f2 * t + np.pi / 8)

C_a = Xa / N # комплексные амплитуды ряда Фурье
f_a = k_a / (N * T) # частоты гармоник, Гц
display(table([
    ('Число моментов времени', len(t)),
    ('max |Im x(t)|', np.abs(x_rec_c.imag).max()),
    ('max |x_восст(t) − x(t)|', np.abs(x_rec - x_an).max()),
]))
nzc = np.nonzero(np.abs(C_a) > 1e-9)[0] # ненулевые гармоники
display(pd.DataFrame({'f, Гц': f_a[nzc], '|C(k)|': np.abs(C_a[nzc])}, index=pd.Index(k_a[nzc], name='k')))

fig, axs = plt.subplots(3, 2, figsize=(12, 9.5))
stem(axs[0, 0], n, xn, C1, ms=3)
axs[0, 0].set_title('Периодическая последовательность x(n)'); axs[0, 0].set_xlabel('n')
stem(axs[0, 1], k, np.abs(X), C1, ms=3)
axs[0, 1].set_title('Модуль ДПФ |X(k)|'); axs[0, 1].set_xlabel('k')

axs[1, 0].plot(t * 1e3, x_rec, color=C2, lw=1.5)
axs[1, 0].set_title('Восстановленный аналоговый сигнал x(t)'); axs[1, 0].set_xlabel('t, мс')
stem(axs[1, 1], f_a, np.abs(C_a), C2, ms=4)
axs[1, 1].set_title('Амплитудный спектр восстановленного сигнала |C(k)| = |Xa(k)|/N')
axs[1, 1].set_xlabel('f, Гц')

axs[2, 0].plot(t * 1e3, x_an, color=C3, lw=2.5, label='исходный x(t)')
axs[2, 0].plot(t * 1e3, x_rec, '--', color=C2, lw=1.2, label='восстановленный')
axs[2, 0].plot(nT * 1e3, xn, 'o', color=C1, ms=3, label='отсчёты x(nT)')
axs[2, 0].set_title('Исходный аналоговый сигнал x(t) = x(nT)|nT=t'); axs[2, 0].set_xlabel('t, мс')
axs[2, 0].legend(loc='upper center', bbox_to_anchor=(0.5, -0.22), ncol=3, fontsize=9)
axs[2, 1].plot(t * 1e3, x_rec - x_an, color=C1, lw=1)
axs[2, 1].set_title('Разность восстановленного и исходного сигналов'); axs[2, 1].set_xlabel('t, мс')
fig.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(10, 3.4))
m = t <= 16 * T + dt / 2
ax.plot(t[m] * 1e3, x_an[m], color=C3, lw=2.5, label='исходный x(t)')
ax.plot(t[m] * 1e3, x_rec[m], '--', color=C2, lw=1.5, label='восстановленный x(t)')
stem(ax, nT[:17] * 1e3, xn[:17], C1, ms=4, label='отсчёты x(nT)')
ax.set_xlabel('t, мс'); ax.set_title('Фрагмент длиной 16T (один общий период суммы гармоник)')
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.2), ncol=3)
plt.show()
```

#### 2.6. Связь ДПФ со спектром аналогового сигнала и сравнение сигналов

Восстановленный сигнал периодичен с периодом $NT = 10.667$ мс. Его спектр линейчатый: гармоники на частотах $k/(NT) = k\,\Delta f$ с комплексными амплитудами $C(k) = X_a(k)/N$. Значит, модуль ДПФ, делённый на $N$, равен амплитудному спектру аналогового сигнала. Отсчёты $k = 0, \dots, N/2-1$ дают составляющие на положительных частотах от 0 до $f_d/2$, отсчёты $k = N/2, \dots, N-1$ дают отрицательные частоты от $-f_d/2$ до $-\Delta f$. У данного сигнала это линии на $\pm 750$ Гц с $|C| = A_1/2 = 0.535$ и на $\pm 1125$ Гц с $|C| = A_2/2 = 1.07$. Спектр дискретной последовательности (ДПФ) равен спектру аналогового сигнала, умноженному на $N$ и периодически продолженному с периодом $f_d$.

Восстановленный и исходный сигналы на графике неразличимы, максимальное отклонение $5.4 \cdot 10^{-14}$. Восстановление точное по двум причинам: выполнено условие теоремы Котельникова ($f_2 = 1125$ Гц $< f_d/2 = 3000$ Гц, наложения спектров нет), и на интервале $NT$ укладывается целое число периодов каждой гармоники, так что частоты гармоник совпадают с частотами $k\,\Delta f$ формулы восстановления. Мнимая часть суммы ($3.2 \cdot 10^{-14}$) появляется из-за несимметричного диапазона $k$: у слагаемого $k = -N/2$ нет пары. Но $X(N/2) \approx 0$, поэтому мнимая часть пренебрежимо мала и отброшена.

#### 2.7. Равенство Парсеваля

```python
E_time = np.sum(np.abs(xn)**2) # энергия во временной области
E_freq = np.sum(np.abs(X)**2) / N # энергия по спектру
E_theory = N * (A1**2 + A2**2) / 2 # для суммы гармоник с целым числом периодов
display(table([
    ('Σ|x(n)|²', E_time),
    ('(1/N)·Σ|X(k)|²', E_freq),
    ('N·(A1² + A2²)/2', E_theory),
    ('Относительная разница', abs(E_time - E_freq) / E_time),
    ('Средняя мощность E/N', E_time / N),
    ('A1²/2 + A2²/2', (A1**2 + A2**2) / 2),
]))
```

Равенство Парсеваля выполняется: $\sum|x(n)|^2 = \frac{1}{N}\sum|X(k)|^2 = 183.1840$, относительное различие порядка $10^{-15}$. Это совпадает с теоретическим значением $N(A_1^2 + A_2^2)/2$.

Энергия сигнала, вычисленная во временной области, равна энергии, вычисленной по спектру: с точностью до множителя $N$ ДПФ сохраняет энергию, так как это ортогональное преобразование. Каждая спектральная линия несёт энергию $|X(k)|^2/N$. После деления на $N$ получается средняя мощность $2.8623 = A_1^2/2 + A_2^2/2$, то есть мощность суммы гармоник равна сумме их мощностей.
