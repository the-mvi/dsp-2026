## Основные теоретические положения

**Передаточная функция** линейной дискретной системы определяется как отношение Z-изображений реакции и воздействия

$$
H(z)=\frac{Y(z)}{X(z)}.
$$

Для рекурсивного звена второго порядка

$$
H(z)=\frac{b_0+b_1z^{-1}+b_2z^{-2}}
{1+a_1z^{-1}+a_2z^{-2}}.
$$

**Разностное уравнение** связывает текущий отсчёт реакции с отсчётами воздействия и предыдущими отсчётами реакции

$$
y(n)=b_0x(n)+b_1x(n-1)+b_2x(n-2)-a_1y(n-1)-a_2y(n-2).
$$

**Импульсная характеристика** $h(n)$ — реакция системы на единичный цифровой импульс при нулевых начальных условиях.

**Свёртка** определяет реакцию линейной системы через воздействие и импульсную характеристику

$$
y(n)=\sum_k h(k)x(n-k).
$$

Для конечных последовательностей длиной $N_1$ и $N_2$ длина полной линейной свёртки равна

$$
N_y=N_1+N_2-1.
$$

**Нули передаточной функции** — корни её числителя. **Полюсы передаточной функции** — корни её знаменателя.

**Устойчивость линейной дискретной системы** определяется абсолютной сходимостью импульсной характеристики

$$
\sum_{n=0}^{\infty}|h(n)|<\infty.
$$

**Частотная характеристика** получается из передаточной функции на единичной окружности $z=e^{j\hat\omega}$

$$
H(e^{j\hat\omega})=H(z)\big|_{z=e^{j\hat\omega}}.
$$

**Амплитудно-частотная характеристика** определяется как

$$
A(\hat\omega)=|H(e^{j\hat\omega})|.
$$

**Фазочастотная характеристика** определяется как

$$
\varphi(\hat\omega)=\arg H(e^{j\hat\omega}).
$$

**Дискретное преобразование Фурье** последовательности длиной $N$ имеет вид

$$
X(k)=\sum_{n=0}^{N-1}x(n)e^{-j\frac{2\pi}{N}kn}.
$$

**Обратное дискретное преобразование Фурье** восстанавливает отсчёты последовательности

$$
x(n)=\frac1N\sum_{k=0}^{N-1}X(k)e^{j\frac{2\pi}{N}kn}.
$$

**Амплитудный спектр** определяется модулем спектральных коэффициентов, **фазовый спектр** — их аргументом.

**Равенство Парсеваля** имеет вид

$$
\sum_{n=0}^{N-1}|x(n)|^2=
\frac1N\sum_{k=0}^{N-1}|X(k)|^2.
$$

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
import cmath
from fractions import Fraction
```

### Часть 1. Анализ рекурсивного цифрового звена второго порядка

```python
# Номер варианта
N_var = 17

# Коэффициенты передаточной функции
b0 = 0.5 + 0.02 * N_var
b1 = b0 * (-1) ** (N_var + 1) * (0.9822 + 0.0178 * N_var)
b2 = b0 * (0.8 + 0.2 * (N_var % 5))
a0 = 1
a1 = (-1) ** N_var * (0.7778 + 0.025 * N_var)
a2 = 0.64 + 0.006 * N_var
# Длины сигналов и частота дискретизации
N1 = N_var % 10 + 20
N2 = N_var % 10 + 30
f_d1 = 1000 * N_var

print(f"b0 = {b0:.6f}, b1 = {b1:.6f}, b2 = {b2:.6f}")
print(f"a0 = {a0}, a1 = {a1:.6f}, a2 = {a2:.6f}")
print(f"N1 = {N1}, N2 = {N2}, f_d1 = {f_d1} Гц")
```

#### 1.1. Импульсная характеристика

Для вычисления импульсной характеристики на вход подаётся единичный цифровой импульс. При нулевых начальных условиях отсчёты определяются разностным уравнением

$$
h(n)=b_0u_0(n)+b_1u_0(n-1)+b_2u_0(n-2)-a_1h(n-1)-a_2h(n-2).
$$


```python
# Единичный импульс
u0 = np.zeros(N1)
u0[0] = 1

# ИХ по разностному уравнению при нулевых начальных условиях
h = np.zeros(N1)
for n in range(N1):
    h[n] = b0 * u0[n]
    if n >= 1:
        h[n] += b1 * u0[n - 1] - a1 * h[n - 1]
    if n >= 2:
        h[n] += b2 * u0[n - 2] - a2 * h[n - 2]

plt.figure(figsize=(8, 4))
plt.stem(np.arange(N1), h, basefmt="k-")
plt.title(f"Импульсная характеристика h(n), N1 = {N1}")
plt.xlabel("n")
plt.ylabel("h(n)")
plt.grid(True)
plt.show()
```

На графике показаны первые $N_1=27$ отсчётов импульсной характеристики. Последовательность имеет колебательный характер, а размах колебаний постепенно уменьшается. При этом последовательность не заканчивается на последнем показанном отсчёте.

Для рекурсивного звена второго порядка при нулевых начальных условиях аналитическая формула импульсной характеристики задаётся выражением 

$$
h(n)=\begin{cases}
 b_0r_*^n\dfrac{\sin((n+1)\varphi_*)}{\sin\varphi_*}, & n=0,\\[8pt]
 b_0r_*^n\dfrac{\sin((n+1)\varphi_*)}{\sin\varphi_*}
 +b_1r_*^{n-1}\dfrac{\sin(n\varphi_*)}{\sin\varphi_*}, & n=1,\\[10pt]
 b_0r_*^n\dfrac{\sin((n+1)\varphi_*)}{\sin\varphi_*}
 +b_1r_*^{n-1}\dfrac{\sin(n\varphi_*)}{\sin\varphi_*}
 +b_2r_*^{n-2}\dfrac{\sin((n-1)\varphi_*)}{\sin\varphi_*}, & n\geq 2.
\end{cases}
$$

С учётом коэффициентов параметры равны

$$
r_* = \sqrt{0.742} \approx 0.861394,
\qquad
\varphi_* = \arccos\left(\frac{1.2028}{2\sqrt{0.742}}\right) \approx 0.797957\text{ рад}.
$$

После подстановки полученных значений формула принимает вид

$$
h(n)\approx
\begin{cases}
0.84, & n=0,\\[5pt]
2.089584, & n=1,\\[8pt]
\begin{aligned}
\dfrac{1}{\sin(0.797957)}\Bigl[&0.84\,(0.861394)^n\sin\!\bigl((n+1)\,0.797957\bigr)\\
&+1.079232\,(0.861394)^{n-1}\sin(n\,0.797957)\\
&+1.008\,(0.861394)^{n-2}\sin\!\bigl((n-1)\,0.797957\bigr)\Bigr]
\end{aligned}, & n\geq 2.
\end{cases}
$$

Численные значения $r_*$ и $\varphi_*$ округлены, поэтому формула записана приближённо. Для рекурсивного звена импульсная характеристика теоретически бесконечна, а число $N_1=27$ означает лишь количество отсчётов, выбранных для расчёта и графика.

#### 1.2. Реакция по формуле свёртки

Воздействие длины $N_2$ задаётся прямоугольной последовательностью

$$
x(n)=
\begin{cases}
1, & 0\leq n<int(N_2/2)\\
0, & int(N_2/2)\leq n\leq N_2-1.
\end{cases}
$$


Реакция определяется свёрткой

$$
y_1(n)=\sum_{k=0}^{N_1-1}h(k)x(n-k)=\sum_{k=0}^{N_2-1}x(k)h(n-k).
$$


```python
# Прямоугольное воздействие длины N2
# Число единичных отсчётов
L = int(np.ceil(N2 / 2))
x_rect = np.zeros(N2)
x_rect[:L] = 1

# Полная свёртка и реакция на интервале воздействия
y_full = np.convolve(h, x_rect)
y1 = y_full[:N2]

print(f"Длина воздействия: {len(x_rect)}, число единиц: {L}")
print(f"Длина свёртки: {len(y_full)}")

# Воздействие с нулевым продолжением
n_show = len(y_full)
x_show = np.zeros(n_show)
x_show[:N2] = x_rect

fig, ax = plt.subplots(1, 3, figsize=(16, 4))
ax[0].stem(np.arange(n_show), x_show, basefmt="k-")
ax[0].set_title("Воздействие x(n)")
ax[1].stem(np.arange(N2), y1, basefmt="k-")
ax[1].set_title("Реакция y1(n) на интервале воздействия")
ax[2].stem(np.arange(len(y_full)), y_full, basefmt="k-")
ax[2].set_title("Полная свёртка")
for a in ax:
    a.set_xlabel("n")
    a.grid(True)
plt.tight_layout()
plt.show()
```

На графиках представлены прямоугольное воздействие, первые $37$ отсчётов реакции и полная свёртка. Воздействие равно единице на первых $19$ отсчётах, после чего становится нулевым. Реакция при этом продолжается и после окончания импульса.

**Ответы на вопросы**

**(а) Чему равна длина импульса $x$?**

Последовательность $x(n)$ содержит $N_2=37$ отсчётов. Сам прямоугольный импульс, то есть её ненулевая часть, длится $19$ отсчётов. На графике реакции видно, что выходной сигнал продолжает изменяться и после окончания импульса. Это связано с тем, что реакция системы зависит не только от текущего воздействия, но и от предыдущих состояний.

**(б) Чему равна длина свёртки аналитически и по графику?**

Длина полной свёртки составляет $N_1+N_2-1=63$ отсчёта. На графике это диапазон индексов от $0$ до $62$, при этом ненулевая часть рассчитанной свёртки занимает первые $45$ отсчётов, так как последние $18$ отсчётов воздействия равны нулю.

**(в) Почему длину свёртки ограничивают до длины воздействия?**

Длину свёртки ограничивают, чтобы рассматривать реакцию на том же временном интервале, что и воздействие. Это позволяет сравнивать результаты, полученные по формуле свёртки и разностному уравнению, отсчёт за отсчётом. При этом оставшаяся часть свёртки отбрасывается только из рассматриваемого интервала, поскольку реакция системы продолжается и после его окончания.

#### 1.3. Реакция по разностному уравнению

Для того же воздействия реакция вычисляется по разностному уравнению

$$
y_2(n)=b_0x(n)+b_1x(n-1)+b_2x(n-2)-a_1y_2(n-1)-a_2y_2(n-2).
$$


```python
# Реакция по разностному уравнению при нулевых начальных условиях
y2 = np.zeros(N2)
for n in range(N2):
    y2[n] = b0 * x_rect[n]
    if n >= 1:
        y2[n] += b1 * x_rect[n - 1] - a1 * y2[n - 1]
    if n >= 2:
        y2[n] += b2 * x_rect[n - 2] - a2 * y2[n - 2]

print(f"Длина воздействия: {len(x_rect)}, длина рассчитанной реакции y2: {len(y2)}")

# Графическое сравнение двух реакций
fig, ax = plt.subplots(1, 2, figsize=(14, 4.8))
ax[0].stem(np.arange(N2), x_rect, basefmt="k-")
ax[0].set_title("Воздействие x(n)")
ax[1].stem(np.arange(N2), y2, basefmt="k-")
ax[1].plot(np.arange(N2), y1, "r.--")
ax[1].axvline(N1 - 0.5, color="gray", linestyle=":")
ax[1].set_title("Реакции y1(n) и y2(n)")

from matplotlib.lines import Line2D
legend_handles = [
    Line2D([0], [0], color="gray", linestyle=":", linewidth=1.5, label="граница"),
    Line2D([0], [0], color="C0", marker="o", linestyle="None", markersize=7, label=r"$y_2(n)$, разностное уравнение"),
    Line2D([0], [0], color="r", marker=".", linestyle="--", markersize=7, label=r"$y_1(n)$, свёртка"),
]
fig.legend(handles=legend_handles, loc="lower center", bbox_to_anchor=(0.5, 0.015), ncol=3, fontsize=9, frameon=True)
for a in ax:
    a.set_xlabel("n")
    a.grid(True)
fig.tight_layout(rect=(0, 0.12, 1, 1))
plt.show()
```

На графике реакции совпадают до $n=26$ включительно. Затем между ними появляется разница. Для свёртки была взята импульсная характеристика из первых $27$ отсчётов (п. 1.1), а разностное уравнение описывает дальнейшее поведение звена без такого ограничения.

Длина воздействия равна $37$ отсчётам. На том же интервале, от $n=0$ до $n=36$, получена реакция $y_2(n)$ длиной $37$ отсчётов. Однако полная реакция рекурсивного звена не ограничивается этим интервалом и в общем случае имеет бесконечную длительность.

#### 1.4. Нули и полюсы передаточной функции

Нули определяются как корни числителя передаточной функции, а полюсы — как корни её знаменателя

$$
b_0z^2+b_1z+b_2=0,
\qquad
z^2+a_1z+a_2=0.
$$

Комплексное число представляется в показательной форме

$$
z=re^{j\varphi},
\qquad
r=|z|,
\qquad
\varphi=\arg z.
$$


```python
# Корни квадратного уравнения A*z^2 + B*z + C = 0
def roots2(A, B, C):
    s = cmath.sqrt(B ** 2 - 4 * A * C)
    return (-B + s) / (2 * A), (-B - s) / (2 * A)

# Нули и полюсы
z01, z02 = roots2(b0, b1, b2)
p1, p2 = roots2(1, a1, a2)

# Корни с положительной мнимой частью
z0_up = z01 if z01.imag >= 0 else z02
p_up = p1 if p1.imag >= 0 else p2

# Модули и аргументы корней
r0, phi0 = abs(z0_up), cmath.phase(z0_up)
rp, phip = abs(p_up), cmath.phase(p_up)

# Угол в долях π
def pi_fraction(angle, max_denominator=12):
    frac = Fraction(angle / np.pi).limit_denominator(max_denominator)
    if frac.denominator == 1:
        return f"{frac.numerator}π"
    if frac.numerator == 1:
        return f"π/{frac.denominator}"
    return f"{frac.numerator}π/{frac.denominator}"

print("Алгебраическая форма:")
print(f"  нули:   {z01:.6f}, {z02:.6f}")
print(f"  полюсы: {p1:.6f}, {p2:.6f}")

print("\nМодули:")
print(f"  |z0| = {r0:.6f}")
print(f"  |p|  = {rp:.6f}")

print("\nАргументы:")
print(f"  нули:   ±{phi0:.6f} рад ≈ ±{pi_fraction(phi0)}")
print(f"  полюсы: ±{phip:.6f} рад ≈ ±{pi_fraction(phip)}")

print("\nПоказательная форма:")
print(f"  нули:   {r0:.6f} * exp(±j*{pi_fraction(phi0)})")
print(f"  полюсы: {rp:.6f} * exp(±j*{pi_fraction(phip)})")
```

Вычисленные нули имеют вид

$$
z_{01,02}=-0.6424\pm j\,0.887312
$$

а полюсы —

$$
p_{1,2}=0.6014\pm j\,0.616699.
$$

В алгебраической форме указаны действительная и мнимая части корней. В показательной форме те же числа описываются модулем и аргументом. Модуль показывает расстояние до начала координат, а аргумент задаёт угол. Для полученных корней

$$
z_{01,02}\approx1.095445e^{\pm j7\pi/10},
\qquad
p_{1,2}\approx0.861394e^{\pm j\pi/4}
$$

Передаточную функцию можно записать через найденные нули и полюсы

$$
H(z)=b_0\frac{(1-z_{01}z^{-1})(1-z_{02}z^{-1})}{(1-p_1z^{-1})(1-p_2z^{-1})}.
$$

После подстановки численных значений

$$
\begin{aligned}
H(z)\approx 0.84\,\frac{
\bigl(1-1.095445\,e^{j\,0.699466\pi}z^{-1}\bigr)
\bigl(1-1.095445\,e^{-j\,0.699466\pi}z^{-1}\bigr)
}{
\bigl(1-0.861394\,e^{j\,0.253998\pi}z^{-1}\bigr)
\bigl(1-0.861394\,e^{-j\,0.253998\pi}z^{-1}\bigr)
}.
\end{aligned}
$$

#### 1.5. Карта нулей и полюсов

Карта нулей и полюсов строится на комплексной плоскости вместе с единичной окружностью.

Основной критерий устойчивости по импульсной характеристике имеет вид

$$
\sum_{n=0}^{\infty}|h(n)|<\infty.
$$

Для рекурсивного звена этот критерий эквивалентен условию расположения всех полюсов внутри единичной окружности

$$
|p_i|<1.
$$


```python
# Повторное вычисление корней полиномов для построения карты
zeros_map = np.roots([b0, b1, b2])
poles_map = np.roots([1, a1, a2])

# Сопоставление с корнями, найденными в п. 1.4
zeros_14 = np.array([z01, z02])
poles_14 = np.array([p1, p2])
zeros_equal = np.allclose(np.sort_complex(zeros_map), np.sort_complex(zeros_14), atol=1e-12)
poles_equal = np.allclose(np.sort_complex(poles_map), np.sort_complex(poles_14), atol=1e-12)

print("Нули, повторное вычисление:", np.round(zeros_map, 6))
print("Полюсы, повторное вычисление:", np.round(poles_map, 6))
print("Модули нулей:", np.round(np.abs(zeros_map), 6))
print("Модули полюсов:", np.round(np.abs(poles_map), 6))
print(f"Совпадают с п. 1.4: нули — {'да' if zeros_equal else 'нет'}, "
      f"полюсы — {'да' if poles_equal else 'нет'}")

fig, ax = plt.subplots(figsize=(6.6, 6.6))
# Точки единичной окружности
phi_circle = np.linspace(0, 2 * np.pi, 400)
ax.plot(np.cos(phi_circle), np.sin(phi_circle), "k--", linewidth=1, label="единичная окружность")
ax.plot(zeros_map.real, zeros_map.imag, "o", markersize=9, markerfacecolor="none", markeredgewidth=2, label="нули")
ax.plot(poles_map.real, poles_map.imag, "x", markersize=10, markeredgewidth=2, label="полюсы")
ax.axhline(0, color="gray", linewidth=0.8)
ax.axvline(0, color="gray", linewidth=0.8)
ax.set_xlabel("Re z")
ax.set_ylabel("Im z")
ax.set_title("Карта нулей и полюсов")
ax.set_aspect("equal")
ax.set_xlim(-1.5, 1.5)
ax.set_ylim(-1.5, 1.5)
ax.grid(True)
handles, legend_labels = ax.get_legend_handles_labels()
fig.legend(handles, legend_labels, loc="lower center", bbox_to_anchor=(0.5, 0.01), ncol=3, fontsize=9, frameon=True)
fig.tight_layout(rect=(0, 0.12, 1, 1))
plt.show()
```

На карте оба полюса находятся внутри единичной окружности, а оба нуля — за её пределами.

#### 1.6. АЧХ и ФЧХ в шкале нормированных частот

Для вычисления частотных характеристик передаточная функция рассматривается на единичной окружности

$$
H(e^{j\hat\omega})=
\frac{b_0+b_1e^{-j\hat\omega}+b_2e^{-j2\hat\omega}}
{1+a_1e^{-j\hat\omega}+a_2e^{-j2\hat\omega}}.
$$

АЧХ равна $|H(e^{j\hat\omega})|$, а ФЧХ равна $\arg H(e^{j\hat\omega})$.


```python
# Положительная половина основной полосы нормированных частот
w = np.linspace(0, np.pi, 1000)

# Части числителя и знаменателя
Re_B = b0 + b1 * np.cos(w) + b2 * np.cos(2 * w)
Im_B = -(b1 * np.sin(w) + b2 * np.sin(2 * w))
Re_A = 1 + a1 * np.cos(w) + a2 * np.cos(2 * w)
Im_A = -(a1 * np.sin(w) + a2 * np.sin(2 * w))

# Комплексная ЧХ и её модуль
H_complex = (Re_B + 1j * Im_B) / (Re_A + 1j * Im_A)
H_abs = np.abs(H_complex)

# Главное значение аргумента содержит разрыв ветви в пределах (-π, π]
phase_principal = np.angle(H_complex)

# Непрерывное представление фазы устраняет искусственные скачки на 2π
H_phase = np.unwrap(phase_principal)

# Индексы максимума и минимума АЧХ
i_max, i_min = H_abs.argmax(), H_abs.argmin()

print(f"|H(0)| = {H_abs[0]:.4f}")
print(f"|H(pi)| = {H_abs[-1]:.4f}")
print(f"Максимум АЧХ = {H_abs[i_max]:.4f}, частота близка к {pi_fraction(w[i_max])}")
print(f"Минимум АЧХ  = {H_abs[i_min]:.4f}, частота близка к {pi_fraction(w[i_min])}")

# Деления оси в долях π
ticks = np.arange(0, 1.01, 0.25) * np.pi
labels = ["0", "π/4", "π/2", "3π/4", "π"]

fig, ax = plt.subplots(1, 3, figsize=(17, 4))
ax[0].plot(w, H_abs)
ax[0].set_title("АЧХ")
ax[0].set_ylabel("|H|")
ax[1].plot(w, phase_principal)
ax[1].set_title("ФЧХ: главное значение аргумента")
ax[1].set_ylabel("arg H, рад")
ax[2].plot(w, H_phase)
ax[2].set_title("ФЧХ: непрерывное представление")
ax[2].set_ylabel("arg H, рад")
for a in ax:
    a.set_xlabel("ω, рад/отсчёт")
    a.set_xticks(ticks)
    a.set_xticklabels(labels)
    a.set_xlim(0, np.pi)
    a.grid(True)
plt.tight_layout()
plt.show()
```

На АЧХ заметен выраженный максимум около $\pi/4$ и глубокий минимум вблизи $7\pi/10$. На графике главного значения фазы есть резкий переход на $2\pi$. Он возникает из-за того, что аргумент ограничен интервалом от $-\pi$ до $\pi$, а не из-за изменения частотной характеристики. Для сравнения показана ФЧХ после применения функции `np.unwrap`: этот переход исчезает, и фаза меняется непрерывно.

**Ответы на вопросы**

**(а) Чему равны границы основной полосы частот?**

Полная основная полоса нормированных частот задаётся интервалом

$$

$$

На графиках представлена его положительная половина, от $0$ до $\pi$. Для звена с вещественными коэффициентами этого достаточно, поскольку АЧХ симметрична относительно нуля, а ФЧХ на отрицательных частотах меняет знак с точностью до полного оборота.

На карте полюсов и нулей полюсы расположены под углами около $\pm\pi/4$, а нули — около $\pm7\pi/10$. Поэтому на графике АЧХ наблюдается максимум вблизи $\pi/4$ и глубокий минимум около $7\pi/10$. Таким образом, форма АЧХ соответствует расположению нулей и полюсов.

#### 1.7. АЧХ и ФЧХ в шкале частот $f$ в герцах

Переход от нормированной частоты к частоте в герцах выполняется по формуле

$$
f=\frac{\hat\omega}{2\pi}f_{d1}.
$$

### Часть 2. Спектральный анализ периодической последовательности

```python
# Число отсчётов и частота дискретизации
N = 64
f_d2 = 2000 * (N_var % 5 + 1)
# Период дискретизации
T = 1 / f_d2
# Амплитуды и частоты гармоник
A1 = 1 + 0.01 * N_var
A2 = 2 * A1
f1 = f_d2 / 8
f2 = 1.5 * f1

print(f"f_d2 = {f_d2} Гц, T = {T * 1e6:.2f} мкс")
print(f"A1 = {A1:.2f}, A2 = {A2:.2f}")
print(f"f1 = {f1:.0f} Гц, f2 = {f2:.0f} Гц")
```

#### 2.1. Последовательность $x(n)$ на периоде

На одном периоде строится последовательность

$$
x(nT)=A_1\cos\left(2\pi f_1nT+\frac{\pi}{4}\right)
+A_2\cos\left(2\pi f_2nT+\frac{\pi}{8}\right).
$$

```python
# Последовательность на периоде N
n = np.arange(N)
t = n * T
x = A1 * np.cos(2 * np.pi * f1 * t + np.pi / 4) + A2 * np.cos(2 * np.pi * f2 * t + np.pi / 8)

fig, ax = plt.subplots(1, 2, figsize=(13, 4))
ax[0].stem(t * 1e3, x, basefmt="k-")
ax[0].set_title("x(nT) в шкале дискретного времени")
ax[0].set_xlabel("nT, мс")
ax[1].stem(n, x, basefmt="k-")
ax[1].set_title("x(n) в шкале дискретного нормированного времени")
ax[1].set_xlabel("n")
for a in ax:
    a.set_ylabel("x")
    a.grid(True)
plt.tight_layout()
plt.show()
```

На обоих графиках показаны $64$ отсчёта последовательности, содержащие четыре повторения её основного периода длиной $16$ отсчётов. На левом графике по горизонтали отложено дискретное время $nT$, на правом — номер отсчёта $n$. Поскольку каждому $n$ соответствует момент времени $t=nT$, форма сигнала и значения отсчётов остаются одинаковыми, а меняется только масштаб горизонтальной оси.

#### 2.2. ОДПФ от ДПФ последовательности

ДПФ и ОДПФ определяются формулами

$$
X(k)=\sum_{n=0}^{N-1}x(n)e^{-j\frac{2\pi}{N}kn},
\qquad
\tilde x(n)=\frac1N\sum_{k=0}^{N-1}X(k)e^{j\frac{2\pi}{N}kn}.
$$

```python
# Номера гармоник и матрица ДПФ
k = np.arange(N)
W = np.exp(-2j * np.pi * np.outer(k, n) / N)

# Прямое и обратное ДПФ
X = W @ x
x_tilde = np.conj(W).T @ X / N

print(f"Максимальное отклонение после ОДПФ: {np.max(np.abs(x_tilde - x)):.3e}")

fig, ax = plt.subplots(figsize=(9.5, 4.8))
ax.stem(n, x_tilde.real, basefmt="k-")
ax.plot(n, x, "r.", markersize=5)
ax.set_title("Сравнение исходной и восстановленной последовательностей")
ax.set_xlabel("n")
ax.set_ylabel("x(n)")
ax.grid(True)

# Условные обозначения отсчётов без вертикальной линии от stem.
from matplotlib.lines import Line2D
legend_handles = [Line2D([0], [0], color="C0", marker="o", linestyle="None", markersize=7, label="ОДПФ от ДПФ"),
       Line2D([0], [0], color="r", marker=".", linestyle="None", markersize=9, label="исходная последовательность")]
fig.legend(handles=legend_handles, loc="lower center", bbox_to_anchor=(0.5, 0.015), ncol=2, fontsize=9, frameon=True)
fig.tight_layout(rect=(0, 0.13, 1, 1))
plt.show()
```

После ОДПФ получена исходная последовательность из $64$ отсчётов. На графике её точки совпадают с исходными, поэтому практически неразличимы. Небольшая численная погрешность связана с вычислениями и не влияет на результат восстановления.

#### 2.3. Амплитудный и фазовый спектры

Для анализа спектральных составляющих вычисляются модуль и аргумент ДПФ последовательности. Каждому спектральному отсчёту $k$ соответствуют нормированная и абсолютная частоты

$$
\hat\omega_k=\frac{2\pi k}{N},
\qquad
f_k=\frac{k}{N}f_{d2}.
$$

```python
# Амплитудный и фазовый спектры (k = 0 ... N-1)
amp_X = np.abs(X)
phase_X = np.angle(X)
phase_X[amp_X < 1e-9] = 0         # в нулевых отсчётах фаза не определена

w_k = 2 * np.pi * k / N           # нормированные частоты
f_k = k * f_d2 / N                # абсолютные частоты, Гц

print("Ненулевые линии спектра:")
for kk in np.where(amp_X > 1e-9)[0]:
    phase_sign = "+" if phase_X[kk] >= 0 else "-"
    phase_text = pi_fraction(abs(phase_X[kk]))
    print(f"  k = {kk:2d}: ω = {pi_fraction(w_k[kk])}, f = {f_k[kk]:7.1f} Гц, |X| = {amp_X[kk]:.4f}, arg X = {phase_sign}{phase_text}")

w_ticks = np.arange(0, 2.01, 0.5) * np.pi
w_labels = ["0", "π/2", "π", "3π/2", "2π"]

fig, ax = plt.subplots(2, 2, figsize=(13, 8))
ax[0, 0].stem(w_k, amp_X, basefmt="k-")
ax[0, 0].set_title("Амплитудный спектр")
ax[0, 0].set_ylabel("|X(k)|")
ax[0, 1].stem(w_k, phase_X, basefmt="k-")
ax[0, 1].set_title("Фазовый спектр")
ax[0, 1].set_ylabel("arg X(k), рад")
for a in ax[0]:
    a.set_xlabel("ω, рад/отсчёт")
    a.set_xticks(w_ticks)
    a.set_xticklabels(w_labels)
ax[1, 0].stem(f_k, amp_X, basefmt="k-")
ax[1, 0].set_title("Амплитудный спектр")
ax[1, 0].set_ylabel("|X(k)|")
ax[1, 1].stem(f_k, phase_X, basefmt="k-")
ax[1, 1].set_title("Фазовый спектр")
ax[1, 1].set_ylabel("arg X(k), рад")
for a in ax[1]:
    a.set_xlabel("f, Гц")
for a in ax.ravel():
    a.grid(True)
plt.tight_layout()
plt.show()
```

В амплитудном спектре выделяются четыре ненулевые линии. Две из них соответствуют положительным частотам $\pi/4$ и $3\pi/8$, а две другие — симметричным отрицательным частотам. При отображении в диапазоне от $0$ до $2\pi$ они находятся в точках $7\pi/4$ и $13\pi/8$.

**Ответы на вопросы**

**(а) Как связано дискретное нормированное время с дискретным временем?**

Каждому номеру отсчёта $n$ соответствует момент дискретного времени $t=nT$. Поэтому нормированное время выражается как

#### 2.4. ДПФ конечной последовательности и спектр периодической последовательности

Для периодического продолжения последовательности коэффициенты спектра связаны с ДПФ формулой

$$
c(k)=\frac{X(k)}{N},
\qquad
|c(k)|=\frac{|X(k)|}{N}.
$$

```python
# Модуль ДПФ конечной последовательности и амплитудный спектр периодической
mod_X = np.abs(X)
amp_periodic = mod_X / N

print("Ненулевые отсчёты ДПФ:")
for kk in np.where(mod_X > 1e-9)[0]:
    print(f"  X({kk:2d}) = {X[kk].real:8.4f} {X[kk].imag:+8.4f}j,  |X| = {mod_X[kk]:.4f},  |X|/N = {amp_periodic[kk]:.4f}")

fig, ax = plt.subplots(1, 2, figsize=(13, 4))
ax[0].stem(w_k, mod_X, basefmt="k-")
ax[0].set_title("Модуль ДПФ конечной последовательности")
ax[0].set_ylabel("|X(k)|")
ax[1].stem(w_k, amp_periodic, basefmt="k-")
ax[1].set_title("Амплитудный спектр периодической последовательности")
ax[1].set_ylabel("|X(k)|/N")
for a in ax:
    a.set_xlabel("ω, рад/отсчёт")
    a.set_xticks(w_ticks)
    a.set_xticklabels(w_labels)
    a.grid(True)
plt.tight_layout()
plt.show()
```

На обоих графиках спектральные линии находятся на одних и тех же частотах. Различается только их высота: амплитуды спектра периодической последовательности в $N=64$ раза меньше модуля ДПФ конечной последовательности.

Это следует из соотношения между коэффициентами спектра периодической последовательности и ДПФ

$$
|c(k)|=\frac{|X(k)|}{N}.
$$

#### 2.5. Восстановление аналогового сигнала по ДПФ

Периодический аналоговый сигнал восстанавливается по формуле

$$
x(t)=\frac1N\sum_{k=-N/2}^{N/2-1}X_a(k)\exp\left(j\frac{2\pi}{NT}kt\right),
$$

где

$$
X_a(k)=
\begin{cases}
X(N+k), & -N/2\leq k\leq-1\\
X(k), & 0\leq k\leq N/2-1.
\end{cases}
$$

Время задаётся на интервале $0\leq t\leq(N-1)T$ с шагом $0.25T$.

```python
# Отсчёты X_a(k) для k = -N/2 ... N/2-1 
k_a = np.arange(-N // 2, N // 2)
X_a = X[k_a % N]                  # X(N+k) при k < 0, X(k) при k >= 0

# Время от 0 до (N-1)T с шагом 0.25T
dt = 0.25 * T
t_rec = np.arange(4 * (N - 1) + 1) * dt

# Восстановленный сигнал
x_rec = (np.exp(1j * 2 * np.pi / (N * T) * np.outer(t_rec, k_a)) @ X_a / N).real

# Исходный аналоговый сигнал
x_orig = A1 * np.cos(2 * np.pi * f1 * t_rec + np.pi / 4) + A2 * np.cos(2 * np.pi * f2 * t_rec + np.pi / 8)

# Спектр восстановленного сигнала: частоты f = k/(N*T)
# Частоты и двусторонний спектр
f_a = k_a / (N * T)
amp_a = np.abs(X_a) / N
```

**(а) Периодическая последовательность и модуль её ДПФ**

```python
# a) Последовательность и ненормированный модуль ДПФ
fig, ax = plt.subplots(1, 2, figsize=(13, 4))
ax[0].stem(n, x, basefmt="k-")
ax[0].set_title("Периодическая последовательность x(n), период N")
ax[0].set_xlabel("n")
ax[0].set_ylabel("x(n)")
ax[1].stem(2 * np.pi * k_a / N, np.abs(X_a), basefmt="k-")
ax[1].set_title("Модуль ДПФ без нормировки |Xₐ(k)|")
ax[1].set_xlabel("ω, рад/отсчёт")
ax[1].set_ylabel("|Xₐ(k)|")
ax[1].set_xticks([-np.pi, -np.pi/2, 0, np.pi/2, np.pi])
ax[1].set_xticklabels(["−π", "−π/2", "0", "π/2", "π"])
for a in ax:
    a.grid(True)
plt.tight_layout()
plt.show()
```

**(б) Восстановленный аналоговый сигнал и его амплитудный спектр**

```python
# б) Восстановленный сигнал и его амплитудный спектр
fig, ax = plt.subplots(1, 2, figsize=(13, 4.8))
ax[0].plot(t_rec * 1e3, x_rec, label="восстановленный x(t)")
ax[0].plot(t * 1e3, x, "ro", markersize=4, label="отсчёты x(nT)")
ax[0].set_title("Восстановленный аналоговый сигнал")
ax[0].set_xlabel("t, мс")
ax[0].set_ylabel("x(t)")
handles, legend_labels = ax[0].get_legend_handles_labels()
fig.legend(handles, legend_labels, loc="lower center", bbox_to_anchor=(0.5, 0.01), ncol=2, fontsize=9, frameon=True)
ax[1].stem(f_a, amp_a, basefmt="k-")
ax[1].set_title("Амплитудный спектр восстановленного сигнала")
ax[1].set_xlabel("f, Гц")
ax[1].set_ylabel("|X_a(k)|/N")
for a in ax:
    a.grid(True)
fig.tight_layout(rect=(0, 0.13, 1, 1))
plt.show()
```

**(в) Исходный аналоговый сигнал**

```python
# в) Исходный аналоговый сигнал
plt.figure(figsize=(8, 4))
plt.plot(t_rec * 1e3, x_orig)
plt.title("Исходный аналоговый сигнал x(t)")
plt.xlabel("t, мс")
plt.ylabel("x(t)")
plt.grid(True)
plt.show()
```

На графике (а) представлен модуль ДПФ без нормировки. Высоты ненулевых линий равны $37.44$ и $74.88$ (получены в п. 2.3).

В подпункте (б) спектральные линии остаются на тех же частотах, но их высоты уменьшаются в $N=64$ раза и составляют $0.585$ и $1.17$ (соотношение установлено в п. 2.4). Восстановленная кривая проходит через исходные дискретные отсчёты.

На графике (в) показан исходный аналоговый сигнал. По форме он совпадает с восстановленным сигналом на всём рассматриваемом интервале.

**Ответы на вопросы**

Каждому ненулевому отсчёту ДПФ соответствует гармоника восстановленного периодического сигнала на той же частоте. Амплитуда соответствующей комплексной спектральной составляющей равна

$$

По сравнению с модулем ДПФ она меньше в $N$ раз. Такое же соотношение получено для периодической последовательности в п. 2.4. Для каждой косинусоиды получаются две комплексно-сопряжённые линии, поэтому амплитуда косинусоиды вдвое больше высоты одной линии.

#### 2.7. Равенство Парсеваля

Для ДПФ проверяется равенство

$$
\sum_{n=0}^{N-1}|x(n)|^2
=\frac1N\sum_{k=0}^{N-1}|X(k)|^2.
$$

Для средней мощности используется эквивалентная запись

$$
\frac1N\sum_{n=0}^{N-1}|x(n)|^2
=\sum_{k=0}^{N-1}\left|\frac{X(k)}{N}\right|^2.
$$

```python
# Энергия и средняя мощность во временной и частотной областях
E_time = np.sum(np.abs(x) ** 2)
E_freq = np.sum(np.abs(X) ** 2) / N
P_time = E_time / N
P_freq = np.sum((np.abs(X) / N) ** 2)

print(f"Энергия по отсчётам x(n):      {E_time:.8f}")
print(f"Энергия по ДПФ (1/N)*sum|X|^2: {E_freq:.8f}")
print(f"Средняя мощность по x(n):      {P_time:.8f}")
print(f"Сумма |X(k)/N|^2:              {P_freq:.8f}")
```

Расчёт дал одинаковую энергию во временной и частотной областях: $219.024$. Средняя мощность также совпала и составила $3.42225$.

Это подтверждает равенство Парсеваля: энергия последовательности сохраняется при переходе от временного представления к частотному.
