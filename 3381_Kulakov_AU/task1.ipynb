## Основные теоретические положения

### Дискретные сигналы и линейные системы

Дискретный сигнал задаётся последовательностью отсчётов $x[n]$, где $n$ является целым числом. При дискретизации с периодом $T$ значение последовательности равно значению аналогового сигнала в момент отсчёта: $x[n]=x(nT)$. Частота дискретизации определяется соотношением $f_d=1/T$.

Дискретная система преобразует входную последовательность $x[n]$ в выходную последовательность $y[n]$. Система линейна, если для любых сигналов $x_1[n]$, $x_2[n]$ и постоянных $c_1$, $c_2$ выполняется принцип суперпозиции:
$$L\{c_1x_1[n]+c_2x_2[n]\}=c_1L\{x_1[n]\}+c_2L\{x_2[n]\}.$$
Система инвариантна к сдвигу, если задержка входа на $n_0$ отсчётов вызывает такую же задержку выхода. Линейная система, инвариантная к сдвигу, описывается импульсной характеристикой $h[n]$.

Единичный импульс определяется как $\delta[0]=1$ и $\delta[n]=0$ при $n\ne0$. Импульсная характеристика является откликом системы на единичный импульс. При нулевых начальных условиях выход системы можно вычислить как линейную свёртку входа с импульсной характеристикой:
$$y[n]=\sum_{m=-\infty}^{\infty}x[m]h[n-m].$$

Для конечных массивов $x[m]$ длины $L_x$ и $h[n]$ длины $L_h$ линейная свёртка вычисляется по формуле
$$y[n]=\sum_{m=0}^{L_x-1}x[m]h[n-m],\quad n=0,\ldots,L_x+L_h-2.$$
Индекс $m$ перебирает отсчёты входа, а индекс $n$ задаёт отсчёты выхода. Поэтому предел суммы по $m$ определяется длиной входа, тогда как диапазон $n$ задаётся длиной результата свёртки. Значения $h[n-m]$ за пределами диапазона $0,\ldots,L_h-1$ равны нулю. Если в конце входного массива записаны нулевые отсчёты, последние элементы результата свёртки также могут быть нулевыми.

У рекурсивного звена импульсная характеристика обычно бесконечна. Конечный расчёт содержит заданное число её отсчётов и представляет собой усечённый фрагмент. При нулевых начальных условиях вычисление отклика по свёртке и по разностному уравнению описывает одну и ту же систему.

### Рекурсивное звено второго порядка

Разностное уравнение звена второго порядка имеет вид
$$a_0y[n]+a_1y[n-1]+a_2y[n-2]=b_0x[n]+b_1x[n-1]+b_2x[n-2].$$

При нулевых начальных условиях передаточная функция равна
$$H(z)=\frac{b_0+b_1z^{-1}+b_2z^{-2}}{a_0+a_1z^{-1}+a_2z^{-2}}.$$

После умножения числителя и знаменателя на $z^2$ получаются многочлены $B(z)=b_0z^2+b_1z+b_2$ и $D(z)=a_0z^2+a_1z+a_2$. Нули передаточной функции являются корнями $B(z)$, полюса являются корнями $D(z)$. Модуль комплексного полюса определяет скорость затухания соответствующей составляющей, аргумент определяет её угловую частоту. Причинная рациональная система устойчива, если все её полюса находятся внутри единичной окружности: $|p_i|<1$.

### Частотные характеристики

Частотная характеристика получается подстановкой $z=e^{j\omega}$ в передаточную функцию:
$$H(e^{j\omega})=H(z)\big|_{z=e^{j\omega}}.$$

Амплитудно-частотная характеристика равна $|H(e^{j\omega})|$, фазо-частотная характеристика равна $\arg H(e^{j\omega})$. В установившемся режиме гармонический вход вызывает отклик той же частоты. Его амплитуда изменяется в $|H(e^{j\omega})|$ раз, а фаза изменяется на $\arg H(e^{j\omega})$.

Нормированная частота в циклах на отсчёт равна $\hat f=f/f_d$, угловая частота в радианах на отсчёт равна $\omega=2\pi f/f_d=2\pi\hat f$. Для действительного сигнала основная неповторяющаяся полоса частот находится в диапазоне от $0$ до частоты Найквиста $f_d/2$. Частоте Найквиста соответствует $\omega=\pi$.

### Дискретное преобразование Фурье

Периодическая последовательность с периодом $N$ удовлетворяет условию $x[n+N]=x[n]$. Конечная последовательность длины $N$ задаётся отсчётами при $n=0,\ldots,N-1$ и равна нулю за пределами этого интервала.

Дискретное преобразование Фурье длины $N$ связывает $N$ временных отсчётов с $N$ комплексными частотными отсчётами:
$$X[k]=\sum_{n=0}^{N-1}x[n]e^{-j2\pi kn/N},\quad k=0,\ldots,N-1.$$

Обратное дискретное преобразование Фурье восстанавливает исходную последовательность:
$$x[n]=\frac{1}{N}\sum_{k=0}^{N-1}X[k]e^{j2\pi kn/N},\quad n=0,\ldots,N-1.$$

ДПФ конечного блока соответствует спектру его периодического продолжения с периодом $N$. Коэффициенты комплексного ряда Фурье этого продолжения равны $C[k]=X[k]/N$. Модуль и аргумент частотного отсчёта образуют амплитудный и фазовый спектры: $A[k]=|X[k]|$ и $\varphi[k]=\arg X[k]$. Для действительной последовательности выполняется сопряжённая симметрия $X[N-k]=X^*[k]$.

Для косинусоидальной составляющей с частотой, точно совпадающей с частотным отсчётом $k_0$ и не равной нулю или частоте Найквиста, амплитуда гармоники равна $2|X[k_0]|/N$, а начальная фаза равна $\arg X[k_0]$. Нормированная частота отсчёта $k$ равна $\hat f_k=k/N$, угловая частота равна $\omega_k=2\pi k/N$, частота в герцах равна $f_k=kf_d/N$. Шаг частотной сетки равен $\Delta f=f_d/N$.

### Восстановление аналогового сигнала и равенство Парсеваля

Для чётного $N$ отсчёты ДПФ можно переставить в диапазон индексов от $-N/2$ до $N/2-1$:
$$X_a[k]=\begin{cases}X[N+k],&-N/2\le k<0,\\X[k],&0\le k\le N/2-1.\end{cases}$$

Периодический аналоговый сигнал восстанавливается суммой гармоник:
$$x_r(t)=\frac{1}{N}\sum_{k=-N/2}^{N/2-1}X_a[k]e^{j2\pi kt/(NT)}.$$

В моменты дискретизации $t=nT$ обратное преобразование восстанавливает исходные отсчёты. При прямом ДПФ без множителя $1/N$ равенство Парсеваля имеет вид
$$\sum_{n=0}^{N-1}|x[n]|^2=\frac{1}{N}\sum_{k=0}^{N-1}|X[k]|^2.$$
Левая часть задаёт энергию последовательности во временной области, правая часть задаёт ту же энергию через спектральные отсчёты.

#### Код для пункта 1

```python
import matplotlib.pyplot as plt
import numpy as np
from IPython.display import display

plt.rcParams.update({"font.family": "DejaVu Sans", "font.size": 11})

def show_figure(fig):
    if fig._suptitle is None:
        fig.tight_layout()
    else:
        fig.tight_layout(rect=(0, 0, 1, 0.96))
    display(fig)
    plt.close(fig)

Nv = 10
b0 = 0.5 + 0.02 * Nv
b1 = b0 * (-1) ** (Nv + 1) * (0.9822 + 0.0178 * Nv)
b2 = b0 * (0.8 + 0.2 * (Nv % 5))
a0 = 1.0
a1 = (-1) ** Nv * (0.7778 + 0.025 * Nv)
a2 = 0.64 + 0.006 * Nv
b = np.array([b0, b1, b2], dtype=float)
a = np.array([a0, a1, a2], dtype=float)
N1 = Nv % 10 + 20
N2 = Nv % 10 + 30
fd1 = 1000 * Nv
h = np.zeros(N1)
for n in range(N1):
    forcing = b[n] if n < len(b) else 0.0
    feedback = sum(a[k] * h[n - k] for k in range(1, len(a)) if n - k >= 0)
    h[n] = (forcing - feedback) / a0

np.set_printoptions(precision=10, suppress=True)
print("b =", b)
print("a =", a)
print("N1 =", N1, "N2 =", N2, "fd1 =", fd1)
print("h[n] =", h)

fig, ax = plt.subplots(figsize=(10, 5))
ax.stem(np.arange(N1), h, basefmt="k-")
ax.set(title="Рисунок 1.1. Импульсная характеристика, вариант 10", xlabel="Номер отсчёта n", ylabel="h[n]")
ax.grid(True, linestyle=":", alpha=0.7)
show_figure(fig)
```

#### Пункт 1. Импульсная характеристика

При нулевых начальных условиях передаточная функция является отношением преобразований выхода и входа:
$$H(z)=\frac{Y(z)}{X(z)}=\frac{0{,}7-0{,}81214z^{-1}+0{,}56z^{-2}}{1+1{,}0278z^{-1}+0{,}7z^{-2}}.$$

Перенос знаменателя в левую часть даёт
$$Y(z)(1+1{,}0278z^{-1}+0{,}7z^{-2})=X(z)(0{,}7-0{,}81214z^{-1}+0{,}56z^{-2}).$$

Применяя обратное $Z$-преобразование и свойство задержки $z^{-k}Y(z)\leftrightarrow y[n-k]$, $z^{-k}X(z)\leftrightarrow x[n-k]$, получаем разностное уравнение:
$$y[n]+1{,}0278y[n-1]+0{,}7y[n-2]=0{,}7x[n]-0{,}81214x[n-1]+0{,}56x[n-2].$$

Для определения импульсной характеристики на вход подаётся единичный импульс $x[n]=\delta[n]$, а выход обозначается $h[n]$. При нулевых начальных условиях $h[-1]=h[-2]=0$. После подстановки получается

$$h[n]=0{,}7\delta[n]-0{,}81214\delta[n-1]+0{,}56\delta[n-2]-1{,}0278h[n-1]-0{,}7h[n-2].$$

Начальные отсчёты:

$$h[0]=0{,}7,$$
$$h[1]=-0{,}81214-1{,}0278\cdot0{,}7=-1{,}5316,$$
$$h[2]=0{,}56-1{,}0278\cdot(-1{,}5316)-0{,}7\cdot0{,}7=1{,}64417848.$$

При $n\ge3$ слагаемые с единичным импульсом равны нулю. Последующие отсчёты рассчитываются по формуле

$$h[n]=-1{,}0278h[n-1]-0{,}7h[n-2],\quad n\ge3.$$

#### Пункт 4. Нули и полюса передаточной функции

Передаточная функция звена варианта 10:

$$H(z)=\frac{0{,}7-0{,}81214z^{-1}+0{,}56z^{-2}}{1+1{,}0278z^{-1}+0{,}7z^{-2}}.$$

Умножая числитель и знаменатель на $z^2$, получаем полиномиальную форму:

$$H(z)=\frac{0{,}7z^2-0{,}81214z+0{,}56}{z^2+1{,}0278z+0{,}7}.$$

Нулями передаточной функции являются корни числителя. Дискриминант числителя равен

$$D_z=(-0{,}81214)^2-4\cdot0{,}7\cdot0{,}56=-0{,}9084286204.$$

Корни числителя, то есть нули передаточной функции, в алгебраической форме:

$$z_{1,2}=\frac{0{,}81214\pm j\sqrt{0{,}9084286204}}{1{,}4}\approx0{,}5801\pm j\,0{,}6807965849,$$

Здесь $j$ обозначает мнимую единицу, $j^2=-1$.

Полюсами передаточной функции являются корни знаменателя. Дискриминант знаменателя равен

$$D_p=(1{,}0278)^2-4\cdot0{,}7=-1{,}74362716.$$

Корни знаменателя, то есть полюсы передаточной функции, в алгебраической форме:

$$p_{1,2}=\frac{-1{,}0278\pm j\sqrt{1{,}74362716}}{2}\approx-0{,}5139\pm j\,0{,}6602323758.$$

Для перехода от алгебраической формы $a+jb$ к показательной форме $re^{j\varphi}$ вычисляются модуль и аргумент:

$$r=\sqrt{a^2+b^2},\qquad a=r\cos\varphi,\qquad b=r\sin\varphi.$$

Для нуля $z_1=0{,}5801+j\,0{,}6807965849$ действительная и мнимая части положительны, поэтому аргумент находится в первой четверти. Угол выражен в радианах:

$$r_z=\sqrt{0{,}5801^2+0{,}6807965849^2}\approx0{,}8944271910,$$
$$\varphi_{z_1}=\arctan\left(\frac{0{,}6807965849}{0{,}5801}\right)\approx0{,}8650901282\approx0{,}2753667403\pi.$$

Второй нуль является комплексно-сопряжённым первому. Поэтому его модуль тот же, а аргумент имеет противоположный знак:

$$z_1\approx0{,}8944271910e^{j\,0{,}2753667403\pi},\qquad z_2\approx0{,}8944271910e^{-j\,0{,}2753667403\pi}.$$

Для полюса $p_1=-0{,}5139+j\,0{,}6602323758$ действительная часть отрицательна, а мнимая положительна. Полюс находится во второй четверти, поэтому аргумент вычисляется как разность между $\pi$ и острым углом. Угол выражен в радианах:

$$r_p=\sqrt{(-0{,}5139)^2+0{,}6602323758^2}\approx0{,}8366600265,$$
$$\varphi_{p_1}=\pi-\arctan\left(\frac{0{,}6602323758}{0{,}5139}\right)\approx2{,}2322036037\approx0{,}7105324750\pi.$$

Второй полюс комплексно-сопряжён первому, поэтому его аргумент имеет противоположный знак:

$$p_1\approx0{,}8366600265e^{j\,0{,}7105324750\pi},\qquad p_2\approx0{,}8366600265e^{-j\,0{,}7105324750\pi}.$$

Таким образом, алгебраическая и показательная формы задают одни и те же нули и полюсы. Поскольку коэффициенты передаточной функции действительные, комплексные корни образуют сопряжённые пары: их модули равны, а аргументы противоположны. Аргументы указаны в главном диапазоне $(-\pi,\pi]$.

Сначала запишем передаточную функцию как произведение множителей через нули $z_i$ и полюсы $p_i$:

$$H(z)=K\frac{\prod_{i=1}^{2}(z-z_i)}{\prod_{i=1}^{2}(z-p_i)},\qquad K=\frac{0{,}7}{1}=0{,}7.$$

Общий вид произведения в показательной форме имеет вид:

$$H(z)=K\frac{\prod_{i=1}^{2}(z-r_{z_i}e^{j\varphi_{z_i}})}{\prod_{i=1}^{2}(z-r_{p_i}e^{j\varphi_{p_i}})}.$$

Подставляя показательные формы нулей и полюсов, получаем:

$$H(z)\approx0{,}7\frac{\left(z-0{,}8944271910e^{j\,0{,}2753667403\pi}\right)\left(z-0{,}8944271910e^{-j\,0{,}2753667403\pi}\right)}{\left(z-0{,}8366600265e^{j\,0{,}7105324750\pi}\right)\left(z-0{,}8366600265e^{-j\,0{,}7105324750\pi}\right)}.$$

#### Пункт 6. Частотные характеристики в шкале нормированных частот

Подставляя $z=e^{j\hat{\omega}}$, то есть $z^{-1}=e^{-j\hat{\omega}}$, в передаточную функцию пункта 1, получаем частотную передаточную функцию:

$$H(e^{j\hat{\omega}})=\frac{0{,}7-0{,}81214e^{-j\hat{\omega}}+0{,}56e^{-j2\hat{\omega}}}{1+1{,}0278e^{-j\hat{\omega}}+0{,}7e^{-j2\hat{\omega}}}.$$

Амплитудно-частотная характеристика определяется как модуль частотной передаточной функции:

$$A(\hat{\omega})=|H(e^{j\hat{\omega}})|=\left|\frac{0{,}7-0{,}81214e^{-j\hat{\omega}}+0{,}56e^{-j2\hat{\omega}}}{1+1{,}0278e^{-j\hat{\omega}}+0{,}7e^{-j2\hat{\omega}}}\right|.$$

Для вычисления фазочастотной характеристики обозначим числитель через $B$, а знаменатель через $D$. При $c=\cos\hat{\omega}$, $s=\sin\hat{\omega}$, $c_2=\cos2\hat{\omega}$ и $s_2=\sin2\hat{\omega}$ их действительные и мнимые части равны:

$$B_R=0{,}7-0{,}81214c+0{,}56c_2,\qquad B_I=0{,}81214s-0{,}56s_2,$$
$$D_R=1+1{,}0278c+0{,}7c_2,\qquad D_I=-1{,}0278s-0{,}7s_2.$$

Фазочастотная характеристика равна разности аргументов числителя и знаменателя:

$$\varphi(\hat{\omega})=\operatorname{atan2}(B_I,B_R)-\operatorname{atan2}(D_I,D_R).$$

Функция $\operatorname{atan2}(v,u)$ определяет аргумент комплексного числа $u+jv$ с учётом четверти комплексной плоскости. На рисунке 1.6 представлена непрерывная ветвь фазы на интервале $0\leq\hat{\omega}\leq\pi$.

**a. Границы основной полосы частот.** Частотная характеристика дискретной системы периодична с периодом $2\pi$. Поэтому полный основной интервал задаётся как $-\pi\leq\hat{\omega}\leq\pi$. На рисунке 1.6 показана положительная половина основного интервала, $0\leq\hat{\omega}\leq\pi$. При действительных коэффициентах АЧХ симметрична относительно нуля частоты, а ФЧХ для отрицательных частот зеркальна ФЧХ для положительных. Две части графика поэтому содержат одну и ту же информацию.

**b. Связь карты нулей и полюсов с видом АЧХ.** Аргумент нуля или полюса указывает частоту, вблизи которой его влияние на АЧХ наиболее заметно. Нулю соответствует провал АЧХ, полюсу соответствует максимум. Радиус корня показывает его расстояние от начала комплексной плоскости. Нуль на единичной окружности даёт нулевое значение АЧХ на частоте своего аргумента. Полюс, расположенный близко к единичной окружности, создаёт выраженный максимум.

В данном варианте аргумент нуля равен $0{,}2753667403\pi$. Вблизи этой частоты на АЧХ наблюдается провал. Аргумент полюса равен $0{,}7105324750\pi$. Вблизи этой частоты расположен выраженный максимум. Частоты минимума и максимума не совпадают точно с аргументами корней, поскольку форма АЧХ зависит от всех нулей и полюсов. Нули и полюсы находятся внутри единичной окружности, поэтому провал и максимум конечны.

**c. Скачок ФЧХ на $\pi$.** Скачок фазы на $\pi$ возможен при нулевом значении частотной передаточной функции. В этой точке АЧХ равна нулю, а фаза не определена. В рассматриваемом варианте такого скачка нет. АЧХ не обращается в ноль, а ФЧХ непрерывна на интервале от $0$ до $\pi$.

**d. Подавление частотных составляющих воздействия.** При $A(\hat{\omega})<1$ амплитуда выходной составляющей меньше амплитуды соответствующей составляющей воздействия. При $A(\hat{\omega})>1$ она больше. В данном варианте АЧХ ниже единицы в низкочастотной части диапазона и выше единицы в высокочастотной части. Следовательно, преимущественно подавляются низкие частоты, а высокие усиливаются.

#### Пункт 7. Частотные характеристики в шкале частот

Для варианта 10 частота дискретизации равна $f_d=10000$ Гц. Нормированная угловая частота и частота в герцах связаны соотношениями

$$\hat{\omega}=\frac{2\pi f}{f_d},\qquad f=\frac{\hat{\omega}f_d}{2\pi}.$$

Подставляя $\hat{\omega}=2\pi f/f_d$ в частотную передаточную функцию, получаем

$$H_f(f)=\frac{0{,}7-0{,}81214e^{-j2\pi f/10000}+0{,}56e^{-j4\pi f/10000}}{1+1{,}0278e^{-j2\pi f/10000}+0{,}7e^{-j4\pi f/10000}}.$$

Амплитудно-частотная и фазочастотная характеристики в шкале герц равны

$$A_f(f)=|H_f(f)|,\qquad \varphi_f(f)=\arg H_f(f).$$

На рисунке 1.7 показаны обе характеристики в шкале герц.

**a. Границы основной полосы частот.** При принятом одностороннем представлении основная полоса частот задаётся интервалом $0\leq f\leq f_d/2$. При $f_d=10000$ Гц границы равны $0$ и $5000$ Гц. Полный симметричный интервал составляет от $-5000$ до $5000$ Гц.

**b. Соответствие частот $\hat{\omega}$ и $f$.** Пересчёт выполняется линейно по формуле $f=5000\hat{\omega}/\pi$. Например, частотам $0$, $\pi/4$, $\pi/2$, $3\pi/4$ и $\pi$ соответствуют $0$, $1250$, $2500$, $3750$ и $5000$ Гц.

Аргумент нуля $0{,}2753667403\pi$ соответствует частоте примерно $1376{,}83$ Гц. Аргумент полюса $0{,}7105324750\pi$ соответствует частоте примерно $3552{,}66$ Гц. Значения АЧХ и ФЧХ при пересчёте не изменяются, изменяется только шкала горизонтальной оси.

#### Код для пункта 1

```python
import matplotlib.pyplot as plt
import numpy as np
from IPython.display import display

plt.rcParams.update({"font.family": "DejaVu Sans", "font.size": 11})

def show_figure(fig):
    if fig._suptitle is None:
        fig.tight_layout()
    else:
        fig.tight_layout(rect=(0, 0, 1, 0.96))
    display(fig)
    plt.close(fig)

Nv = 10
N = 64
fd = 2000 * (Nv % 5 + 1)
T = 1.0 / fd
A1 = 1.0 + 0.01 * Nv
A2 = 2.0 * A1
f1 = fd / 8.0
f2 = 1.5 * f1
n = np.arange(N)
n_normalized = n / N
x = A1 * np.cos(2 * np.pi * f1 * n * T + np.pi / 4) + A2 * np.cos(2 * np.pi * f2 * n * T + np.pi / 8)

np.set_printoptions(precision=10, suppress=True)
print("N =", N, "fd =", fd, "T =", T)
print("A1 =", A1, "A2 =", A2, "f1 =", f1, "f2 =", f2)
print("x[n] =", x)

fig, axes = plt.subplots(2, 1, figsize=(12, 8))
fig.suptitle("Рисунок 2.1. Исходная периодическая последовательность")
axes[0].stem(n, x, basefmt="k-")
axes[0].set(title="Последовательность на заданном интервале", xlabel="Номер отсчёта n", ylabel="x[n]")
axes[0].grid(True, linestyle=":", alpha=0.7)
axes[1].stem(n_normalized, x, basefmt="k-")
axes[1].set(title="Последовательность в нормированной временной шкале", xlabel="Нормированное время n/N", ylabel="x[n]")
axes[1].grid(True, linestyle=":", alpha=0.7)
show_figure(fig)
```

#### Пункт 1. Исходная периодическая последовательность

Для варианта 10 заданы период наблюдения $(N=64)$, частота дискретизации $(f_d=2000)$ Гц и период дискретизации $(T=1/f_d=0{,}0005)$ с. Амплитуды гармоник составляют $(A_1=1{,}1)$ и $(A_2=2{,}2)$, частоты равны $(f_1=250)$ Гц и $(f_2=375)$ Гц.

Подстановка параметров в заданное выражение даёт

$$
x(nT)=1{,}1\cos\left(2\pi\cdot250\,nT+\frac{\pi}{4}\right)+2{,}2\cos\left(2\pi\cdot375\,nT+\frac{\pi}{8}\right).
$$

Так как $(T=1/2000)$ с, аргументы гармоник можно записать через номер отсчёта:

$$
2\pi f_1Tn=\frac{\pi n}{4},\qquad 2\pi f_2Tn=\frac{3\pi n}{8}.
$$

Следовательно, последовательность на заданном интервале имеет вид

$$
x[n]=1{,}1\cos\left(\frac{\pi n}{4}+\frac{\pi}{4}\right)+2{,}2\cos\left(\frac{3\pi n}{8}+\frac{\pi}{8}\right),\qquad n=0,1,\ldots,63.
$$

Период первой гармоники равен 8 отсчётам, второй гармоники равен 16 отсчётам. Поэтому период суммы равен наименьшему общему кратному 8 и 16, то есть 16 отсчётам. Заданный интервал из 64 отсчётов содержит четыре периода последовательности.

На рисунке 2.1 одна и та же последовательность представлена в шкале номера отсчёта $(n)$ и в нормированной временной шкале $\hat n=n/N$. Для второго графика последняя точка имеет координату $(63/64)$, поскольку отсчёты нумеруются от 0 до $(N-1)$.

#### Пункт 2. Обратное дискретное преобразование Фурье

Используется прямое дискретное преобразование Фурье

$$
X[k]=\sum_{n=0}^{N-1}x[n]\exp\left(-j\frac{2\pi}{N}kn\right),\qquad k=0,1,\ldots,N-1,
$$

и обратное дискретное преобразование Фурье. Полученную последовательность обозначим $\tilde{x}[n]$.

$$
\tilde{x}[n]=\frac{1}{N}\sum_{k=0}^{N-1}X[k]\exp\left(j\frac{2\pi}{N}kn\right).
$$

Для гармоники, частота которой совпадает с узлом сетки ДПФ $(k_0)$, справедливо разложение

$$
A\cos\left(\frac{2\pi k_0}{N}n+\varphi\right)=\frac{A}{2}e^{j\varphi}e^{j2\pi k_0n/N}+\frac{A}{2}e^{-j\varphi}e^{-j2\pi k_0n/N}.
$$

Вследствие ортогональности экспонент на интервале длины $(N)$ первая гармоника формирует отсчёты $(X[8])$ и $(X[56])$, а вторая гармоника формирует $(X[12])$ и $(X[52])$:

$$
X[8]=\frac{64\cdot1{,}1}{2}e^{j\pi/4}=35{,}2e^{j\pi/4},\qquad X[56]=35{,}2e^{-j\pi/4},
$$

$$
X[12]=\frac{64\cdot2{,}2}{2}e^{j\pi/8}=70{,}4e^{j\pi/8},\qquad X[52]=70{,}4e^{-j\pi/8}.
$$

Остальные отсчёты ДПФ равны нулю. При подстановке этих четырёх значений в формулу обратного преобразования комплексно-сопряжённые пары объединяются в косинусы:

$$
\begin{aligned}
\tilde{x}[n]
&=\frac{1}{64}\left(35{,}2e^{j\pi/4}e^{j2\pi8n/64}+35{,}2e^{-j\pi/4}e^{j2\pi56n/64}\right.\\
&\qquad\left.+70{,}4e^{j\pi/8}e^{j2\pi12n/64}+70{,}4e^{-j\pi/8}e^{j2\pi52n/64}\right)\\
&=\frac{2\cdot35{,}2}{64}\cos\left(\frac{2\pi\cdot8}{64}n+\frac{\pi}{4}\right)
+\frac{2\cdot70{,}4}{64}\cos\left(\frac{2\pi\cdot12}{64}n+\frac{\pi}{8}\right)\\
&=1{,}1\cos\left(\frac{\pi n}{4}+\frac{\pi}{4}\right)+2{,}2\cos\left(\frac{3\pi n}{8}+\frac{\pi}{8}\right)=x[n].
\end{aligned}
$$

На рисунке 2.2 исходные отсчёты и значения после обратного ДПФ совпадают.



#### Код для пункта 3

```python
centered_k = np.arange(-N // 2, N // 2)
Xa = np.fft.fftshift(X)
omega_centered = 2 * np.pi * centered_k / N
frequency_centered = centered_k * fd / N
coefficient_amplitude = np.abs(Xa) / N
nonzero = np.abs(Xa) > 1e-8

print("k, ω, f, |X[k]|/N, фаза")
for k, w, f, amp, phase_value in zip(
    centered_k[nonzero],
    omega_centered[nonzero],
    frequency_centered[nonzero],
    coefficient_amplitude[nonzero],
    np.angle(Xa[nonzero])
):
    print(f"{k}, {w:.10f}, {f:.2f}, {amp:.10f}, {phase_value:.10f}")

fig, axes = plt.subplots(2, 2, figsize=(14, 9))
fig.suptitle("Рисунок 2.3. Амплитудный и фазовый спектры")
for column, (freq_values, label) in enumerate([
    (omega_centered, "Нормированная угловая частота ω, радиан на отсчёт"),
    (frequency_centered, "Частота f, герц")
]):
    axes[0, column].stem(freq_values[nonzero], coefficient_amplitude[nonzero], basefmt="k-")
    axes[0, column].set(title="Двусторонний амплитудный спектр", xlabel=label, ylabel="|X[k]|/N")
    axes[0, column].grid(True, linestyle=":", alpha=0.7)
    axes[1, column].stem(freq_values[nonzero], np.angle(Xa[nonzero]), basefmt="k-")
    axes[1, column].set(title="Фазовый спектр", xlabel=label, ylabel="Фаза, радиан")
    axes[1, column].grid(True, linestyle=":", alpha=0.7)
axes[0, 0].set_xlim(-np.pi, np.pi)
axes[1, 0].set_xlim(-np.pi, np.pi)
axes[0, 1].set_xlim(-fd / 2, fd / 2)
axes[1, 1].set_xlim(-fd / 2, fd / 2)
show_figure(fig)
```

##### a. Связь дискретного нормированного времени с дискретным временем

Дискретное время задаётся целым номером отсчёта $(n)$. Нормированная временная координата определяется отношением

$$
\hat n=\frac{n}{N}.
$$

Физическое время отсчёта равно $(t=nT)$, поэтому связь между физическим и нормированным временем имеет вид

$$
\hat n=\frac{t}{NT},\qquad t=\hat nNT.
$$

При $(N=64)$ и $(T=0{,}0005)$ с длительность интервала наблюдения для блока из 64 отсчётов равна $(NT=0{,}032)$ с. Отсчёты последовательности на рисунке 2.1 показаны как по индексу $(n)$, так и по координате $\hat n$.

##### b. Связь частоты в герцах с дискретной нормированной частотой

Нормированная частота в циклах на отсчёт определяется отношением
$$
\hat f=\frac{f}{f_d}.
$$
Дискретная нормированная угловая частота измеряется в радианах на отсчёт и связана с ней равенством
$$
\hat\omega=2\pi\hat f=\frac{2\pi f}{f_d}.
$$
Для узла с номером $k$ в $N$-точечном ДПФ
$$
\hat\omega_k=\frac{2\pi k}{N},\qquad f_k=\frac{k f_d}{N},\qquad \Delta f=\frac{f_d}{N}=31{,}25.
$$
Центрированным индексам $k=-32,\ldots,31$ соответствует основная полоса $-\pi\leq\hat\omega<\pi$, или $-1000\leq f<1000$ Гц. Для заданных гармоник
$$
k_1=8,\quad \hat f_1=0{,}125,\quad \hat\omega_1=\frac{\pi}{4},\quad f_1=250;
$$
$$
k_2=12,\quad \hat f_2=0{,}1875,\quad \hat\omega_2=\frac{3\pi}{8},\quad f_2=375.
$$
Значения частот $f_k$, $f_1$, $f_2$ и шага $\Delta f$ выражены в герцах. На графиках нормированная шкала представлена через $\hat\omega$, а абсолютная шкала через частоту $f$.

#### Код для пункта 5

```python
time_step = 0.25 * T
time_points = int(round(((N - 1) * T) / time_step)) + 1
t = np.linspace(0.0, (N - 1) * T, time_points)
x_analog = A1 * np.cos(2 * np.pi * f1 * t + np.pi / 4) + A2 * np.cos(2 * np.pi * f2 * t + np.pi / 8)
reconstruction_terms = Xa[:, None] * np.exp(
    1j * 2 * np.pi * centered_k[:, None] * t[None, :] / (N * T)
)
x_reconstructed = np.real(np.sum(reconstruction_terms, axis=0) / N)
reconstruction_error = np.max(np.abs(x_analog - x_reconstructed))

print("Число точек восстановления =", time_points)
print("Шаг времени, секунды =", time_step)
print("Максимальное различие исходного и восстановленного сигналов =", reconstruction_error)

fig, axes = plt.subplots(1, 2, figsize=(14, 6))
fig.suptitle("Рисунок 2.5. Восстановление аналогового сигнала")
axes[0].plot(t * 1000, x_analog, label="Исходный аналоговый сигнал", linewidth=2)
axes[0].plot(t * 1000, x_reconstructed, "--", label="Восстановленный сигнал", linewidth=2)
axes[0].scatter(n * T * 1000, x, color="black", s=12, label="Отсчёты x[n]", zorder=3)
axes[0].set(title="Исходный и восстановленный сигналы", xlabel="Время t, миллисекунды", ylabel="Амплитуда")
axes[0].grid(True, linestyle=":", alpha=0.7)
axes[0].legend()
axes[1].stem(frequency_centered[nonzero], coefficient_amplitude[nonzero], basefmt="k-")
axes[1].set(title="Двусторонний амплитудный спектр", xlabel="Частота f, герц", ylabel="|Xa[k]|/N")
axes[1].grid(True, linestyle=":", alpha=0.7)
axes[1].set_xlim(-fd / 2, fd / 2)
show_figure(fig)
```

#### Пункт 5. Восстановление периодического аналогового сигнала

Для применения заданной формулы частотные индексы ДПФ представляются в центрированном виде $(k=-32,\ldots,31)$. Отсчёты $X_a(k)$ определяются через обычную нумерацию $X[k]$:

$$
X_a(k)=\begin{cases}
X[N+k],&-N/2\leq k\leq-1,\\
X[k],&0\leq k\leq N/2-1.
\end{cases}
$$

Ненулевые центрированные отсчёты равны

$$
X_a(-12)=70{,}4e^{-j\pi/8},\quad X_a(-8)=35{,}2e^{-j\pi/4},\quad

Интервал восстановления по 64 отсчётам равен $NT=64\cdot0{,}0005=0{,}032$ с. Он содержит четыре основных периода сигнала; основной период равен $0{,}008$ с. Восстановление по отсчётам ДПФ выполняется по формуле

x_r(t)=\frac{1}{64}\sum_{k=-32}^{31}X_a(k)\exp\left(j\frac{2\pi k}{0{,}032}t\right).

В сумме ненулевы только четыре слагаемых. Объединение пар индексов $(k=\pm8)$ и $(k=\pm12)$ даёт

$$
\begin{aligned}
x_r(t)&=\frac{2\cdot35{,}2}{64}\cos\left(\frac{2\pi\cdot8}{0{,}032}t+\frac{\pi}{4}\right)
+\frac{2\cdot70{,}4}{64}\cos\left(\frac{2\pi\cdot12}{0{,}032}t+\frac{\pi}{8}\right)\\

\end{aligned}
$$

Значения сигнала вычисляются на интервале $0\leq t\leq(N-1)T=0{,}0315$ с шагом

$$
\Delta t=0{,}25T=0{,}000125=0{,}125.
$$

#### Пункт 7. Проверка равенства Парсеваля

Для принятого определения прямого ДПФ равенство Парсеваля имеет вид

$$
\sum_{n=0}^{N-1}|x[n]|^2=\frac{1}{N}\sum_{k=0}^{N-1}|X[k]|^2.
$$

Временная сумма квадратов отсчётов определяется вкладом двух гармоник. На интервале из 64 отсчётов каждая гармоника содержит целое число периодов, сумма произведения разных гармоник равна нулю, а сумма квадратов каждой косинусоиды равна $(N/2)$. Поэтому

$$
\sum_{n=0}^{63}|x[n]|^2=\frac{64}{2}A_1^2+\frac{64}{2}A_2^2
=32(1{,}1^2+2{,}2^2)=32(1{,}21+4{,}84)=193{,}6.
$$

В спектральной сумме ненулевы четыре отсчёта: два имеют модуль $(35{,}2)$, два имеют модуль $(70{,}4)$. Следовательно,

$$
\frac{1}{64}\sum_{k=0}^{63}|X[k]|^2
=\frac{2\cdot35{,}2^2+2\cdot70{,}4^2}{64}
=\frac{12390{,}4}{64}=193{,}6.
$$

Обе части равенства Парсеваля совпадают. Средняя мощность последовательности равна

$$
\frac{1}{64}\sum_{n=0}^{63}|x[n]|^2=\frac{193{,}6}{64}=3{,}025.
$$

Это подтверждает согласованность вычисленных отсчётов последовательности и её ДПФ при выбранной нормировке преобразования.
