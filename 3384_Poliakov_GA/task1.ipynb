## Основные теоретические положения

### Линейная дискретная система

При нулевых начальных условиях звено второго порядка с $a_0=1$ описывается передаточной функцией

$$H(z)=\frac{b_0+b_1z^{-1}+b_2z^{-2}}{1+a_1z^{-1}+a_2z^{-2}}$$

и разностным уравнением

$$y(n)=b_0x(n)+b_1x(n-1)+b_2x(n-2)-a_1y(n-1)-a_2y(n-2).$$

До начала воздействия $x(n)=y(n)=0$ при $n<0$. Импульсная характеристика $h(n)$ — реакция системы на единичный импульс $u_0(n)$, равный 1 при $n=0$ и 0 иначе. Для рассматриваемого рекурсивного звена она бесконечна; в расчётах используется конечный фрагмент.

Реакция линейной стационарной системы определяется свёрткой:

$$y(n)=\sum_{k=-\infty}^{\infty}x(k)h(n-k).$$

Полная свёртка массивов длины $N_2$ и $N_1$ содержит $N_2+N_1-1$ отсчётов. Усечение бесконечной импульсной характеристики даёт приближённую реакцию.

### Нули, полюсы и устойчивость

После сокращения общих множителей нули $z_i$ являются корнями числителя, полюсы $p_i$ — корнями знаменателя:

$$H(z)=\frac{b_0}{a_0}\frac{(z-z_1)(z-z_2)}{(z-p_1)(z-p_2)}.$$

Алгебраическая и показательная формы комплексного числа связаны формулой Эйлера:

$$q=u+jv=re^{j\varphi},\qquad r=\sqrt{u^2+v^2},\qquad\varphi=\arg q,\qquad j^2=-1.$$

Для рассматриваемого рекурсивного звена условие устойчивости — расположение всех полюсов передаточной функции строго внутри единичной окружности: $|p_i|<1$.

### Время и частотные характеристики

Моменты дискретизации и частотные шкалы связаны соотношениями

$$T=\frac1{f_d},\qquad t_n=nT,\qquad \hat f=\frac{f}{f_d},\qquad \hat\omega=2\pi\hat f.$$

Нормированное время равно $n=t_n/T$. Для устойчивой системы частотная характеристика получается подстановкой $z=e^{j\hat\omega}$:

$$H(e^{j\hat\omega})=\frac{b_0+b_1e^{-j\hat\omega}+b_2e^{-j2\hat\omega}}{1+a_1e^{-j\hat\omega}+a_2e^{-j2\hat\omega}}.$$

АЧХ равна $|H(e^{j\hat\omega})|$, ФЧХ — $\arg H(e^{j\hat\omega})$. Основная полоса $-\pi\le\hat\omega<\pi$ соответствует $-f_d/2\le f<f_d/2$. При вещественных коэффициентах значения на противоположных частотах комплексно-сопряжены. В точках нулевой АЧХ фаза не определена.

### ДПФ и спектры периодической последовательности

Для $N$ отсчётов прямое и обратное ДПФ определяются формулами

$$X(k)=\sum_{n=0}^{N-1}x(n)e^{-j2\pi kn/N},\qquad
x(n)=\frac1N\sum_{k=0}^{N-1}X(k)e^{j2\pi kn/N}.$$

Коэффициенты ряда Фурье периодического продолжения равны $C(k)=X(k)/N$. Двусторонний амплитудный спектр равен $|C(k)|$, фазовый — $\arg C(k)$ при $C(k)\ne0$. Косинусоида амплитуды $A$ и начальной фазы $\varphi$, попадающая на частотную сетку строго между нулём и частотой Найквиста, даёт две линии на $\pm f$ с амплитудами $A/2$ и фазами $\pm\varphi$.

На центрированной сетке для чётного $N$

$$k=-N/2,\ldots,N/2-1,\qquad \hat f_k=\frac{k}{N},\qquad f_k=\frac{k f_d}{N},\qquad \Delta f=\frac{f_d}{N}.$$

Модуль ДПФ конечной записи и амплитудный спектр её периодического продолжения связаны соотношением $|X(k)|=N|C(k)|$.

### Восстановление непрерывного сигнала

После перестановки коэффициентов ДПФ в порядок отрицательных и положительных частот

$$X_a(k)=\begin{cases}X(N+k),&-N/2\le k<0,\\X(k),&0\le k<N/2\end{cases}$$

периодический сигнал восстанавливается по формуле

$$x_{\mathrm{rec}}(t)=\frac1N\sum_{k=-N/2}^{N/2-1}X_a(k)e^{j2\pi kt/(NT)}.$$

При $t=nT$ она совпадает с ОДПФ. Для суммы гармоник на сетке $f_k=k/(NT)$, лежащих строго ниже частоты Найквиста $f_d/2$, восстановление сохраняет исходный сигнал и между отсчётами.

### Равенство Парсеваля

Энергия записи одинакова во временном и частотном представлениях:

$$E=\sum_{n=0}^{N-1}|x(n)|^2=\frac1N\sum_{k=0}^{N-1}|X(k)|^2.$$

Средняя мощность периодического продолжения равна

$$P=\frac{E}{N}=\sum_{k=0}^{N-1}|C(k)|^2.$$

### Часть 1. Линейные дискретные системы

#### 1. Импульсная характеристика

Коэффициенты и длина вычисляемого фрагмента равны:

$$b_0=0.82,\quad b_1=-1.03894,\quad b_2=0.82,$$
$$a_0=1,\quad a_1=1.1778,\quad a_2=0.736,\quad N_1=26.$$

Следовательно,

$$H(z)=\frac{0.82-1.03894z^{-1}+0.82z^{-2}}{1+1.1778z^{-1}+0.736z^{-2}}.$$

При нулевых начальных условиях $h(-1)=h(-2)=0$:

$$\begin{aligned}
h(0)&=b_0=0.82,\\
h(1)&=b_1-a_1h(0)\approx-2.0047,\\
h(2)&=b_2-a_1h(1)-a_2h(0)\approx2.5777,\\
h(n)&=-1.1778h(n-1)-0.736h(n-2),\qquad n\ge3.
\end{aligned}$$

Для получения аналитической формулы выделим функцию

$$G(z)=\frac{1}{1+a_1z^{-1}+a_2z^{-2}},\qquad H(z)=(b_0+b_1z^{-1}+b_2z^{-2})G(z).$$

Через различные корни $p_1,p_2$ уравнения $p^2+a_1p+a_2=0$ знаменатель раскладывается на множители:

$$p_{1,2}=\frac{-a_1\pm\sqrt{a_1^2-4a_2}}{2},$$

$$G(z)=\frac{1}{(1-p_1z^{-1})(1-p_2z^{-1})}
=\frac{1}{p_1-p_2}\left(\frac{p_1}{1-p_1z^{-1}}-\frac{p_2}{1-p_2z^{-1}}\right).$$

Обратное $Z$-преобразование $1/(1-pz^{-1})$ при $|z|>|p|$ даёт последовательность $p^n$ для $n\ge0$, равную нулю при $n<0$. Поэтому

$$g(n)=\begin{cases}\dfrac{p_1^{n+1}-p_2^{n+1}}{p_1-p_2},&n\ge0,\\0,&n<0.\end{cases}$$

Умножение на $z^{-i}$ соответствует задержке на $i$ отсчётов, откуда

$$h(n)=b_0g(n)+b_1g(n-1)+b_2g(n-2)
=0.82g(n)-1.03894g(n-1)+0.82g(n-2).$$




```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import lfilter

# Коэффициенты звена и длина импульсной характеристики для варианта.
variant = 16
b0 = 0.5 + 0.02 * variant
b1 = b0 * (-1) ** (variant + 1) * (0.9822 + 0.0178 * variant)
b2 = b0 * (0.8 + 0.2 * (variant % 5))
b = np.array([b0, b1, b2])
a = np.array([1.0, (-1) ** variant * (0.7778 + 0.025 * variant),
              0.64 + 0.006 * variant])
N1 = variant % 10 + 20
n = np.arange(N1)
u0 = np.zeros(N1)
u0[0] = 1.0
h = np.zeros(N1)

# Рекурсия с нулевыми начальными условиями: отрицательные индексы исключаются.
for i in range(N1):
    for j in range(3):
        if i >= j:
            h[i] += b[j] * u0[i - j]
    for j in range(1, 3):
        if i >= j:
            h[i] -= a[j] * h[i - j]

# Аналитическая проверка через полюсы; g(k) = 0 при k < 0.
p1, p2 = np.roots(a)

def g(k):
    k = np.asarray(k)
    result = np.zeros(k.shape, dtype=complex)
    valid = k >= 0
    result[valid] = (p1 ** (k[valid] + 1) - p2 ** (k[valid] + 1)) / (p1 - p2)
    return result

h_analytic = b[0] * g(n) + b[1] * g(n - 1) + b[2] * g(n - 2)
h_filter = lfilter(b, a, u0)
np.testing.assert_allclose(h, h_analytic, rtol=0, atol=1e-12)
np.testing.assert_allclose(h, h_filter, rtol=0, atol=1e-12)
print(f"N1 = {N1}; b = {b}; a = {a}")
print(f"Расхождение с аналитической формулой: {np.max(np.abs(h - h_analytic)):.2e}")
print(" n          h(n)")
for index, value in zip(n, h):
    print(f"{index:2d}  {value: .4f}")

fig, ax = plt.subplots(figsize=(10, 4.5))
ax.stem(n, h, linefmt="C0-", markerfmt="C0o", basefmt="k-")
ax.set(title="Импульсная характеристика",
       xlabel="Номер отсчёта n", ylabel="h(n)", xlim=(-0.5, N1 - 0.5))
ax.set_xticks(np.arange(0, N1, 5))
ax.grid(True, alpha=0.3)
fig.tight_layout()
plt.show()
```

На графике наблюдаются затухающие знакопеременные колебания. Первые отсчёты равны $h(0)=0.82$, $h(1)\approx-2.0047$, $h(2)\approx2.5777$. Начальное увеличение модуля связано с воздействием коэффициентов числителя на первые три отсчёта; далее действует однородная рекурсия.

Действительная длина импульсной характеристики бесконечна: знаменатель не сокращается с числителем, а рекурсия продолжает формировать отклик после окончания входного импульса. Модули сопряжённых корней знаменателя равны $\sqrt{0.736}\approx0.8579<1$, поэтому отклик затухает, но не обрывается после 26 отсчётов.

#### 2. Реакция на прямоугольный импульс по формуле свёртки

Длина массива воздействия составляет $N_2=16\bmod10+30=36$. Прямоугольный импульс занимает первые $N_2/2=18$ отсчётов:

$$x(n)=\begin{cases}1,&0\le n<18,\\0,&18\le n\le35.\end{cases}$$

За пределами заданного массива воздействие также считается нулевым. Для расчёта используется фрагмент $h(n)$ из пункта 1 длины $N_1=26$. Обозначим его продолжение нулями через $h_{26}(n)$:

$$y_{1,\mathrm{full}}(n)=\sum_{k=0}^{35}x(k)h_{26}(n-k),\qquad 0\le n\le60.$$

Для представления реакции на том же временном интервале, что и воздействие, сохраняются первые 36 отсчётов: $y_1(n)=y_{1,\mathrm{full}}(n)$ при $0\le n\le35$. Полная свёртка также сохраняется и отображается на отдельном графике.

```python
# Воздействие: 18 единичных отсчётов, затем 18 нулевых.
N2 = variant % 10 + 30
pulse_length = N2 // 2
n_x = np.arange(N2)
x = np.zeros(N2)
x[:pulse_length] = 1.0

# Полная свёртка с 26 отсчётами h по определению.
y1_full = np.zeros(N2 + N1 - 1)
for i in range(len(y1_full)):
    for k in range(N2):
        if 0 <= i - k < N1:
            y1_full[i] += x[k] * h[i - k]

np.testing.assert_allclose(y1_full, np.convolve(x, h, mode="full"),
                           rtol=0, atol=1e-12)
# Для сравнения с воздействием берутся первые N2 отсчётов реакции.
y1 = y1_full[:N2].copy()
n_full = np.arange(len(y1_full))
assert len(y1_full) == 61 and len(y1) == 36
assert np.all(y1_full[pulse_length + N1 - 1:] == 0)
print(f"Длина массива x: {len(x)}; длительность импульса: {pulse_length} отсчётов")
print(f"Длина полной свёртки: {len(y1_full)}; длина y1: {len(y1)}")
print(f"Последний потенциально ненулевой индекс свёртки: {pulse_length + N1 - 2}")

fig, axes = plt.subplots(3, 1, figsize=(11, 9), sharex=True)
axes[0].stem(n_x, x, linefmt="C0-", markerfmt="C0o", basefmt="k-")
axes[0].set(title="Воздействие: прямоугольный импульс", ylabel="x(n)",
            ylim=(-0.15, 1.25))
axes[1].stem(n_full, y1_full, linefmt="C1-", markerfmt="C1o", basefmt="k-")
axes[1].set(title="Полная свёртка: 61 отсчёт", ylabel="y₁,full(n)")
axes[2].stem(n_x, y1, linefmt="C2-", markerfmt="C2o", basefmt="k-")
axes[2].set(title="Реакция на интервале воздействия: 36 отсчётов",
            ylabel="y₁(n)", xlabel="Номер отсчёта n")
for ax in axes:
    ax.axvline(N2 - 0.5, color="gray", linestyle="--", linewidth=1,
               label="Конец интервала воздействия")
    ax.set_xlim(-1, len(y1_full))
    ax.set_xticks(np.arange(0, len(y1_full), 5))
    ax.grid(True, alpha=0.3)
axes[1].legend(loc="upper right")
fig.tight_layout()
plt.show()
```

##### Анализ результатов

Массив $x$ содержит 36 отсчётов, из которых первые 18 равны единице. Длительность ненулевой части импульса — 18 отсчётов, или $18/16000=1.125$ мс.

Полная свёртка имеет длину $36+26-1=61$ отсчёт ($n=0,\ldots,60$). Из-за 18 нулей в конце воздействия её ненулевая часть занимает 43 отсчёта ($n=0,\ldots,42$); остальные значения равны нулю, что видно на графике.

Включение и выключение импульса вызывают колебательные переходные процессы. Для сравнения с воздействием выделяются первые 36 отсчётов реакции. Такое ограничение задаёт интервал наблюдения, а не прекращение отклика.

Свёртка с 26 отсчётами $h(n)$ даёт приближение к реакции исходной системы. Усечение начинает влиять на результат с $n=26$; истинный отклик имеет бесконечный затухающий хвост.

#### 3. Реакция по разностному уравнению

Для заданных коэффициентов разностное уравнение имеет вид:

$$y_2(n)=0.82x(n)-1.03894x(n-1)+0.82x(n-2)-1.1778y_2(n-1)-0.736y_2(n-2).$$

Используется воздействие $x(n)$ из пункта 2: 18 единичных отсчётов и 18 нулевых. Начальные условия: $y_2(-1)=y_2(-2)=0$, $x(-1)=x(-2)=0$. Реакция вычисляется последовательно для $n=0,\ldots,35$ и сравнивается с сохранённой реакцией $y_1(n)$.

```python
# Расчёт по разностному уравнению при нулевых начальных условиях.
y2 = np.zeros(N2)
for i in range(N2):
    y2[i] = b[0] * x[i]
    if i >= 1:
        y2[i] += b[1] * x[i - 1] - a[1] * y2[i - 1]
    if i >= 2:
        y2[i] += b[2] * x[i - 2] - a[2] * y2[i - 2]

np.testing.assert_allclose(y2, lfilter(b, a, x), rtol=0, atol=1e-12)
difference = y2 - y1
np.testing.assert_allclose(y2[:N1], y1[:N1], rtol=0, atol=1e-12)

# 36 отсчётов h достаточно для точной свёртки на интервале 0...35.
n_check = np.arange(N2)
h_check = np.real(b[0] * g(n_check) + b[1] * g(n_check - 1)
                  + b[2] * g(n_check - 2))
y_check = np.convolve(x, h_check, mode="full")[:N2]
np.testing.assert_allclose(y2, y_check, rtol=0, atol=1e-12)
max_index = int(np.argmax(np.abs(difference)))
print(f"Длины массивов: x — {len(x)}, y1 — {len(y1)}, y2 — {len(y2)}")
print(f"Максимальное |y2 - y1|: {abs(difference[max_index]):.4f} при n = {max_index}")
print(f"Расхождение на n = 0...25: {np.max(np.abs(difference[:N1])):.2e}")
print(f"Расхождение со свёрткой при длине h = 36: {np.max(np.abs(y2 - y_check)):.2e}")
print(f"Последний вычисленный отсчёт y2(35): {y2[-1]:.4f}")

fig, axes = plt.subplots(4, 1, figsize=(11, 12), sharex=True)
axes[0].stem(n_x, x, linefmt="C0-", markerfmt="C0o", basefmt="k-")
axes[0].set(title="Воздействие из пункта 2", ylabel="x(n)", ylim=(-0.15, 1.25))
axes[1].stem(n_x, y2, linefmt="C1-", markerfmt="C1o", basefmt="k-")
axes[1].set(title="Реакция по разностному уравнению", ylabel="y₂(n)")
axes[2].plot(n_x, y1, "o-", markersize=5, label="y₁: свёртка с 26 отсчётами h")
axes[2].plot(n_x, y2, "x--", markersize=6, label="y₂: разностное уравнение")
axes[2].set(title="Сравнение реакций (линии соединяют дискретные отсчёты)", ylabel="Амплитуда")
axes[2].legend(loc="upper right")
axes[3].stem(n_x, difference, linefmt="C3-", markerfmt="C3o", basefmt="k-")
axes[3].axvline(N1 - 0.5, color="gray", linestyle="--",
               label="Начало влияния усечения h")
axes[3].set(title="Разность реакций", ylabel="y₂(n) − y₁(n)", xlabel="Номер отсчёта n")
axes[3].legend(loc="upper left")
for ax in axes:
    ax.set_xlim(-0.5, N2 - 0.5)
    ax.set_xticks(np.arange(0, N2, 5))
    ax.grid(True, alpha=0.3)
fig.tight_layout()
plt.show()
```

##### Анализ результатов

На графиках $y_1(n)$ и $y_2(n)$ совпадают при $n=0,\ldots,25$. Далее появляется расхождение до $0.05555$: свёртка использует усечённую характеристику длины 26, а разностное уравнение сохраняет память системы. Разность равна

$$y_2(n)-y_1(n)=\sum_{m=26}^{n}h(m)x(n-m),$$

где при $n<26$ сумма пуста. Проверка со свёрткой, использующей 36 отсчётов $h(n)$, даёт совпадение на всём интервале с погрешностью менее $10^{-12}$.

Массив воздействия и рассчитанный массив $y_2$ содержат по 36 отсчётов. Это длина наблюдаемого фрагмента: после $n=35$ реакция продолжается и затухает.

#### 4. Нули и полюсы передаточной функции

После умножения числителя и знаменателя на $z^2$:

$$H(z)=\frac{0.82z^2-1.03894z+0.82}{z^2+1.1778z+0.736}.$$

Нули определяются уравнением $0.82z^2-1.03894z+0.82=0$, полюсы — уравнением $z^2+1.1778z+0.736=0$:

$$z_{1,2}=\frac{1.03894\pm\sqrt{1.03894^2-4\cdot0.82^2}}{2\cdot0.82},$$
$$p_{1,2}=\frac{-1.1778\pm\sqrt{1.1778^2-4\cdot0.736}}{2}.$$

```python
# Корни полиномов числителя и знаменателя по убыванию степеней z.
zeros = np.roots(b)
poles = np.roots(a)
gain = b[0] / a[0]
zeros = zeros[np.argsort(-zeros.imag)]
poles = poles[np.argsort(-poles.imag)]

print(f"Коэффициент перед произведением множителей: {gain:.2f}")
print(f"{'Корень':<8} {'Алгебраическая форма':<28} {'Модуль':>12} {'Аргумент, рад':>16} {'Аргумент / π':>16}")
for label, roots in [("z", zeros), ("p", poles)]:
    for index, root in enumerate(roots, start=1):
        algebraic = f"{root.real:.4f} {root.imag:+.4f}j"
        print(f"{label + str(index):<8} {algebraic:<28} {abs(root):12.4f} "
              f"{np.angle(root):16.4f} {np.angle(root) / np.pi:16.4f}")

# Проверка восстановления коэффициентов из корней.
np.testing.assert_allclose(b[0] * np.poly(zeros), b, rtol=0, atol=1e-12)
np.testing.assert_allclose(a[0] * np.poly(poles), a, rtol=0, atol=1e-12)
for roots in (zeros, poles):
    np.testing.assert_allclose(np.abs(roots) * np.exp(1j * np.angle(roots)),
                               roots, rtol=0, atol=1e-12)

# Сопоставление исходной и факторизованных передаточных функций.
z_test = np.array([0.4 + 0.2j, -0.3 + 1.1j, 1.0 + 0j, -1.0 + 0j])
H_direct = np.polyval(b, z_test) / np.polyval(a, z_test)
H_factors = gain * np.prod(z_test[:, None] - zeros, axis=1) / np.prod(
    z_test[:, None] - poles, axis=1)
H_inverse = gain * np.prod(1 - zeros / z_test[:, None], axis=1) / np.prod(
    1 - poles / z_test[:, None], axis=1)
np.testing.assert_allclose(H_direct, H_factors, rtol=0, atol=1e-12)
np.testing.assert_allclose(H_direct, H_inverse, rtol=0, atol=1e-12)
print("Восстановление коэффициентов, показательная форма и факторизация проверены.")
```

##### Результаты и связь форм представления

В таблице вычисленные значения корней и их аргументов округлены до четырёх знаков после запятой.

| Корень | Алгебраическая форма | Модуль | Аргумент относительно $\pi$ | Показательная форма |
| --- | --- | --- | --- | --- |
| $z_1$ | $0.6335+j\,0.7737$ | $1$ | $+0.2816\pi$ | $e^{j\,0.2816\pi}$ |
| $z_2$ | $0.6335-j\,0.7737$ | $1$ | $-0.2816\pi$ | $e^{-j\,0.2816\pi}$ |
| $p_1$ | $-0.5889+j\,0.6239$ | $0.8579$ | $+0.7408\pi$ | $0.8579e^{j\,0.7408\pi}$ |
| $p_2$ | $-0.5889-j\,0.6239$ | $0.8579$ | $-0.7408\pi$ | $0.8579e^{-j\,0.7408\pi}$ |

В каждой паре корни комплексно-сопряжённые. Алгебраическая форма задаёт координаты точки на комплексной плоскости, показательная — расстояние до начала координат и угол относительно положительной вещественной полуоси. Это два представления одного и того же корня; аргумент определён с точностью до $2\pi$.

Модуль нулей равен единице точно: по теореме Виета $z_1z_2=b_2/b_0=1$, а для сопряжённых корней $z_1z_2=|z_1|^2$. Аналогично, модуль каждого полюса равен $\sqrt{a_2/a_0}=\sqrt{0.736}$. Нули и полюсы различны, поэтому общие множители не сокращаются.

##### Передаточная функция в виде произведения множителей

Обозначим $\theta_z=\arg z_1\approx0.2816\pi$, $\theta_p=\arg p_1\approx0.7408\pi$ и $r_p=\sqrt{0.736}\approx0.8579$. Тогда

$$H(z)=0.82\,
\frac{(z-e^{j\theta_z})(z-e^{-j\theta_z})}
{(z-r_pe^{j\theta_p})(z-r_pe^{-j\theta_p})}.$$

Эквивалентная запись через задержки $z^{-1}$:

$$H(z)=0.82\,
\frac{(1-e^{j\theta_z}z^{-1})(1-e^{-j\theta_z}z^{-1})}
{(1-r_pe^{j\theta_p}z^{-1})(1-r_pe^{-j\theta_p}z^{-1})}.$$

Коэффициент перед произведением равен $b_0/a_0=0.82$. В формулах углы и модуль обозначают точные значения; приведённые десятичные числа являются их приближениями.

```python
from scipy.signal import tf2zpk

# Проверка корней из предыдущего пункта через tf2zpk.
zeros_map, poles_map, gain_map = tf2zpk(b, a)
zeros_map = zeros_map[np.argsort(-zeros_map.imag)]
poles_map = poles_map[np.argsort(-poles_map.imag)]
np.testing.assert_allclose(zeros_map, zeros, rtol=0, atol=1e-12)
np.testing.assert_allclose(poles_map, poles, rtol=0, atol=1e-12)
np.testing.assert_allclose(gain_map, gain, rtol=0, atol=1e-12)

# Устойчивость определяется расположением полюсов внутри единичного круга.
pole_radii = np.abs(poles_map)
is_stable = bool(np.all(pole_radii < 1.0))
print("Модули нулей:", np.round(np.abs(zeros_map), 4))
print("Модули полюсов:", np.round(pole_radii, 4))
print(f"Наибольший модуль полюса: {pole_radii.max():.4f}")
print("Исследуемое рекурсивное звено устойчиво:", "да" if is_stable else "нет")
print("Нули и полюсы совпадают с пунктом 4 с точностью 1e-12.")

fig, ax = plt.subplots(figsize=(7.5, 7.5))
angle = np.linspace(0, 2 * np.pi, 721)
ax.plot(np.cos(angle), np.sin(angle), "--", color="gray",
        linewidth=1.5, label="Единичная окружность |z| = 1")
ax.axhline(0, color="gray", linewidth=0.8)
ax.axvline(0, color="gray", linewidth=0.8)
ax.scatter(zeros_map.real, zeros_map.imag, s=100, marker="o",
           facecolors="none", edgecolors="C0", linewidths=2, label="Нули")
ax.scatter(poles_map.real, poles_map.imag, s=110, marker="x",
           color="C3", linewidths=2, label="Полюсы")
for label, roots in [("z", zeros_map), ("p", poles_map)]:
    for index, root in enumerate(roots, start=1):
        ax.annotate(f"{label}{index}", (root.real, root.imag),
                    xytext=(10, 8), textcoords="offset points", fontsize=12)
ax.set(title="Карта нулей и полюсов",
       xlabel="Re(z)", ylabel="Im(z)", xlim=(-1.3, 1.3), ylim=(-1.3, 1.3))
ax.set_aspect("equal", adjustable="box")
ax.grid(True, alpha=0.3)
ax.legend(loc="upper center", bbox_to_anchor=(0.5, -0.12), ncol=2)
fig.tight_layout()
plt.show()
```

#### 6. АЧХ и ФЧХ в шкале нормированных частот

Для заданного звена

$$H(e^{j\hat{\omega}})=\frac{0.82-1.03894e^{-j\hat{\omega}}+0.82e^{-j2\hat{\omega}}}{1+1.1778e^{-j\hat{\omega}}+0.736e^{-j2\hat{\omega}}}.$$

Графики строятся на интервале $-\pi\le\hat{\omega}\le\pi$. Горизонтальная ось показывает $\hat{\omega}/\pi$, поэтому её пределы равны $-1$ и $1$. Фаза представляется главным значением аргумента в радианах, без развёртки. В точках нулей АЧХ фазовые значения исключаются из графика, чтобы не показывать численно случайную фазу.

```python
from scipy.signal import freqz

zero_angle = float(np.angle(zeros[0]))
pole_angle = float(np.angle(poles[0]))
# В сетку включены точные углы нулей передаточной функции.
omega = np.unique(np.concatenate((np.linspace(-np.pi, np.pi, 20001),
                                  [-zero_angle, zero_angle])))
omega, H_response = freqz(b, a, worN=omega)
amplitude = np.abs(H_response)
phase = np.angle(H_response)
# При нулевой амплитуде фаза не определена.
phase[amplitude < 1e-12] = np.nan

# Проверка freqz прямой подстановкой z = exp(j * omega).
z_unit = np.exp(1j * omega)
H_direct = np.polyval(b, z_unit) / np.polyval(a, z_unit)
np.testing.assert_allclose(H_response, H_direct, rtol=1e-12, atol=1e-12)
_, H_at_zero = freqz(b, a, worN=np.array([zero_angle]))
assert abs(H_at_zero[0]) < 1e-12

positive_indices = np.flatnonzero(omega >= 0)
peak_index = positive_indices[np.argmax(amplitude[positive_indices])]
_, H_edges = freqz(b, a, worN=np.array([0.0, np.pi]))
print(f"Частота нуля АЧХ: ±{zero_angle / np.pi:.4f}π рад/отсчёт")
print(f"Угол полюса: ±{pole_angle / np.pi:.4f}π")
print(f"Максимум АЧХ на сетке: {amplitude[peak_index]:.4f} "
      f"при |ω̂| / π ≈ {omega[peak_index] / np.pi:.4f}")
print(f"A(0) = {abs(H_edges[0]):.4f}; A(π) = {abs(H_edges[1]):.4f}")

# Односторонние значения фазы для оценки скачка у нуля АЧХ.
eps = 1e-7
_, H_sides = freqz(b, a, worN=np.array([zero_angle - eps, zero_angle + eps]))
phase_sides = np.angle(H_sides)
print(f"Фаза слева и справа от нуля: {phase_sides[0]:.4f}, {phase_sides[1]:.4f} рад")
print(f"Разность фаз около нуля: {(phase_sides[1] - phase_sides[0]) / np.pi:.4f}π")

fig, axes = plt.subplots(2, 1, figsize=(11, 8), sharex=True)
axes[0].plot(omega / np.pi, amplitude, color="C0", linewidth=1.6)
axes[0].scatter([-zero_angle / np.pi, zero_angle / np.pi], [0, 0],
                color="C3", s=35, zorder=3)
axes[0].set(title="Амплитудно-частотная характеристика", ylabel="|H(exp(jω̂))|", ylim=(0, 12))
axes[1].plot(omega / np.pi, phase, color="C1", linewidth=1.6)
axes[1].set(title="Фазочастотная характеристика", ylabel="Фаза, рад",
            xlabel="Нормированная угловая частота ω̂ / π", ylim=(-np.pi, np.pi))
axes[1].set_yticks([-np.pi, -np.pi/2, 0, np.pi/2, np.pi],
                  ["−π", "−π/2", "0", "π/2", "π"])
for ax in axes:
    for index, sign in enumerate([-1, 1]):
        ax.axvline(sign * zero_angle / np.pi, color="C3", linestyle=":", alpha=0.8,
                   label="Углы нулей" if index == 0 else None)
        ax.axvline(sign * pole_angle / np.pi, color="gray", linestyle="--", alpha=0.7,
                   label="Углы полюсов" if index == 0 else None)
    ax.set_xlim(-1, 1)
    ax.set_xticks(np.linspace(-1, 1, 9))
    ax.grid(True, alpha=0.3)
axes[0].legend(loc="upper center")
fig.tight_layout()
plt.show()
```

##### Анализ результатов

Основная полоса — $-\pi\le\hat\omega<\pi$, на оси $\hat\omega/\pi$ ей соответствует интервал $[-1,1)$. На графике показаны обе граничные точки, эквивалентные из-за периодичности. Для вещественной системы достаточно положительной половины $[0,\pi]$.

Связь АЧХ с картой нулей и полюсов выражается через расстояния от точки единичной окружности:

$$A(\hat\omega)=0.82\frac{|e^{j\hat\omega}-z_1|\,|e^{j\hat\omega}-z_2|}{|e^{j\hat\omega}-p_1|\,|e^{j\hat\omega}-p_2|}.$$

Нули на единичной окружности дают провалы АЧХ до нуля при $\hat\omega\approx\pm0.2816\pi$. Вблизи аргументов полюсов $\pm0.7408\pi$ знаменатель мал и АЧХ возрастает. Максимумы $A\approx11.385$ достигаются около $\pm0.7486\pi$: их положение зависит от всех корней, а не только от одного полюса.

При переходе через нуль АЧХ фаза изменяется скачком на $\pi$; в самой точке фаза не определена. Это величина скачка, а не его частота: при $\hat\omega=\pi$ имеем $A\approx4.7992$ и нулевую фазу.

Преимущественно подавляются низкие частоты: $A(0)\approx0.2063<1$, а около $0.2816\pi$ происходит полное подавление. Высокочастотные составляющие в области максимума усиливаются, что согласуется с колебаниями реакции на фронтах импульса.

### Часть 2. Дискретное преобразование Фурье

#### 1. Последовательность в шкалах дискретного и нормированного времени

Параметры второй части равны:

Длина последовательности равна $N=64$, частота дискретизации — $f_d=2000(16\bmod5+1)=4000$ Гц, период дискретизации — $T=1/4000=0.00025$ с.
$$A_1=1+0.01\cdot16=1.16,\qquad A_2=2A_1=2.32,$$
Частоты гармоник равны $f_1=f_d/8=500$ Гц и $f_2=1.5f_1=750$ Гц.

Исследуемая последовательность:

$$x(nT)=1.16\cos\left(2\pi\cdot500\,nT+\frac{\pi}{4}\right)+2.32\cos\left(2\pi\cdot750\,nT+\frac{\pi}{8}\right).$$

После подстановки $T=1/4000$:

$$x(n)=1.16\cos\left(\frac{\pi n}{4}+\frac{\pi}{4}\right)+2.32\cos\left(\frac{3\pi n}{8}+\frac{\pi}{8}\right),\qquad n=0,\ldots,63.$$

На первом графике время $nT$ выражено в миллисекундах, на втором нормированное время равно номеру отсчёта $n=t_n/T$. Оба графика отображают те же 64 значения.

```python
# Параметры сигнала второй части; частота дискретизации отличается от первой.
N = 64
fd_part2 = 2000 * (variant % 5 + 1)
T = 1 / fd_part2
A1 = 1 + 0.01 * variant
A2 = 2 * A1
f1 = fd_part2 / 8
f2 = 1.5 * f1
n_part2 = np.arange(N)
t_part2 = n_part2 * T

x_harmonic1 = A1 * np.cos(2 * np.pi * f1 * t_part2 + np.pi / 4)
x_harmonic2 = A2 * np.cos(2 * np.pi * f2 * t_part2 + np.pi / 8)
x_part2 = x_harmonic1 + x_harmonic2

# Проверка формулы в нормированном времени и повторения через 16 отсчётов.
x_normalized = (A1 * np.cos(np.pi * n_part2 / 4 + np.pi / 4)
                + A2 * np.cos(3 * np.pi * n_part2 / 8 + np.pi / 8))
np.testing.assert_allclose(x_part2, x_normalized, rtol=0, atol=1e-12)
np.testing.assert_allclose(x_part2[16:], x_part2[:-16], rtol=0, atol=1e-12)
assert len(x_part2) == N
print(f"N = {N}; fd = {fd_part2} Гц; T = {T:.5f} с = {T * 1000:.2f} мс")
print(f"A1 = {A1:.2f}; A2 = {A2:.2f}; f1 = {f1:.0f} Гц; f2 = {f2:.0f} Гц")
print(f"Время последнего отсчёта: {t_part2[-1] * 1000:.2f} мс")
print(f"Длительность периода N отсчётов: {N * T * 1000:.2f} мс")
print(f"Число циклов гармоник на N отсчётах: {f1 * N * T:g} и {f2 * N * T:g}")
print("Основной период суммы: 16 отсчётов = 4 мс; на интервале N — 4 повторения.")

fig, axes = plt.subplots(2, 1, figsize=(11, 8), sharey=True)
axes[0].stem(t_part2 * 1000, x_part2, linefmt="C0-", markerfmt="C0o", basefmt="k-")
axes[0].set(title="Последовательность в шкале дискретного времени",
            xlabel="Время nT, мс", ylabel="x(nT)", xlim=(-0.25, N * T * 1000))
axes[0].set_xticks(np.arange(0, 16.1, 2))
axes[1].stem(n_part2, x_part2, linefmt="C1-", markerfmt="C1o", basefmt="k-")
axes[1].set(title="Последовательность в шкале дискретного нормированного времени",
            xlabel="Нормированное время n = t / T", ylabel="x(n)", xlim=(-1, N))
axes[1].set_xticks(np.arange(0, N + 1, 8))
for ax in axes:
    ax.grid(True, alpha=0.3)
fig.tight_layout()
plt.show()
```

##### Анализ результатов

Две временные шкалы показывают одинаковые отсчёты: единице $n$ соответствует $T=0.25$ мс. Последний отсчёт расположен при $63T=15.75$ мс, а длительность периода из 64 отсчётов равна $NT=16$ мс.

Частоты $f_1/f_d=1/8$ и $f_2/f_d=3/16$ задают периоды гармоник 8 и 16 отсчётов. Минимальный период суммы равен 16 отсчётам (4 мс), поэтому показаны четыре повторения. На интервале $NT$ укладываются 8 циклов первой гармоники и 12 циклов второй.

#### 2. Восстановление последовательности с помощью ОДПФ

Для последовательности $x(n)$ из пункта 1 вычисляются все $N=64$ комплексных коэффициента ДПФ, затем выполняется обратное преобразование:

$$X(k)=\operatorname{DFT}\{x(n)\},\qquad x_{\mathrm{rec}}(n)=\operatorname{IDFT}\{X(k)\}.$$


```python
# В ifft уже включена нормировка 1/N.
X_part2 = np.fft.fft(x_part2)
x_restored_complex = np.fft.ifft(X_part2)

# Проверка комплексного результата до выделения действительной части.
reconstruction_error = np.max(np.abs(x_restored_complex - x_part2))
imaginary_residual = np.max(np.abs(x_restored_complex.imag))
np.testing.assert_allclose(x_restored_complex, x_part2, rtol=0, atol=1e-12)
assert imaginary_residual < 1e-12
x_restored = x_restored_complex.real

# Независимая проверка ОДПФ прямым суммированием.
k_part2 = np.arange(N)
inverse_matrix = np.exp(2j * np.pi * np.outer(n_part2, k_part2) / N)
x_restored_sum = inverse_matrix @ X_part2 / N
np.testing.assert_allclose(x_restored_sum, x_part2, rtol=0, atol=1e-12)
print(f"Число коэффициентов ДПФ: {len(X_part2)}")
print(f"Число восстановленных отсчётов: {len(x_restored)}")
print(f"Максимальная ошибка восстановления: {reconstruction_error:.3e}")
print(f"Максимальная мнимая часть после ОДПФ: {imaginary_residual:.3e}")
print(f"Ошибка ОДПФ по прямой формуле: {np.max(np.abs(x_restored_sum - x_part2)):.3e}")

fig, axes = plt.subplots(2, 1, figsize=(11, 7.5), sharex=True)
axes[0].stem(n_part2, x_restored, linefmt="C0-", markerfmt="C0o", basefmt="k-",
             label="Восстановленная последовательность")
axes[0].plot(n_part2, x_part2, linestyle="none", marker="x", color="C1",
             markersize=6, label="Исходные отсчёты")
axes[0].set(title="Последовательность после ДПФ и ОДПФ", ylabel="x(n)")
axes[0].legend(loc="upper right")
axes[1].stem(n_part2, x_restored - x_part2, linefmt="C3-", markerfmt="C3o", basefmt="k-")
axes[1].set(title="Ошибка восстановления действительной части",
            ylabel="x_rec(n) − x(n)", xlabel="Нормированное время n = t / T")
axes[1].ticklabel_format(axis="y", style="sci", scilimits=(0, 0))
for ax in axes:
    ax.set_xlim(-1, N)
    ax.set_xticks(np.arange(0, N + 1, 8))
    ax.grid(True, alpha=0.3)
fig.tight_layout()
plt.show()
```

##### Анализ результатов

Восстановленные 64 отсчёта совпадают с исходными: максимальная комплексная ошибка составляет около $1.34\cdot10^{-15}$. Малые отклонения и остаточная мнимая часть обусловлены округлением.

Совпадение обеспечивается сохранением всех комплексных коэффициентов ДПФ, включая фазы. График построен в нормированном времени $n=0,\ldots,63$.

#### 3. Амплитудный и фазовый спектры последовательности

Используются коэффициенты ДПФ $X(k)$ из пункта 2. Строится двусторонний амплитудный спектр периодической последовательности $|C(k)|=|X(k)|/N$ и фазовый спектр $\arg C(k)$. Такая нормировка даёт для каждой косинусоиды две линии с амплитудами $A/2$.

По горизонтальным осям используются $\hat{f}_k=k/N=f_k/f_d$ (циклы на отсчёт) и $f_k$ (Гц). Связь с угловой нормировкой первой части: $\hat{\omega}_k/\pi=2\hat{f}_k$.

Фаза отображается только для значимых спектральных линий: значения фазы на уровне вычислительного шума не несут информации о сигнале.

```python
# Деление на N даёт коэффициенты ряда Фурье; fftshift центрирует спектр.
C_part2 = X_part2 / N
C_centered = np.fft.fftshift(C_part2)
fhat_part2 = np.fft.fftshift(np.fft.fftfreq(N, d=1.0))
freq_part2 = np.fft.fftshift(np.fft.fftfreq(N, d=T))
k_centered = np.arange(-N // 2, N // 2)
amplitude_spectrum = np.abs(C_centered)
phase_spectrum = np.angle(C_centered)
# Фаза отображается только для линий выше уровня численного шума.
spectral_threshold = 1e-12 * amplitude_spectrum.max()
significant = amplitude_spectrum > spectral_threshold
phase_spectrum[~significant] = np.nan

# Проверка четырёх линий: частот, амплитуд A/2 и начальных фаз.
expected_frequencies = np.array([-f2, -f1, f1, f2])
expected_coefficients = np.array([
    A2 / 2 * np.exp(-1j * np.pi / 8),
    A1 / 2 * np.exp(-1j * np.pi / 4),
    A1 / 2 * np.exp(1j * np.pi / 4),
    A2 / 2 * np.exp(1j * np.pi / 8),
])
np.testing.assert_allclose(freq_part2[significant], expected_frequencies,
                           rtol=0, atol=1e-12)
np.testing.assert_allclose(C_centered[significant], expected_coefficients,
                           rtol=0, atol=1e-12)
np.testing.assert_allclose(freq_part2, fd_part2 * fhat_part2, rtol=0, atol=1e-12)
print(f"Шаг частоты: Δf̂ = {1 / N:.6f}; Δf = {fd_part2 / N:.2f} Гц")
print("  k      f̂          f, Гц      |C(k)|       фаза / π")
for k, fhat, f, amplitude_value, phase_value in zip(
        k_centered[significant], fhat_part2[significant], freq_part2[significant],
        amplitude_spectrum[significant], phase_spectrum[significant]):
    print(f"{k:3d}  {fhat: .4f}   {f:8.2f}     {amplitude_value:.2f}     {phase_value / np.pi: .4f}")
print(f"Максимум остальных амплитуд: {amplitude_spectrum[~significant].max():.3e}")

fig, axes = plt.subplots(2, 2, figsize=(13, 8))
for column, (frequency_axis, xlabel, limits) in enumerate([
    (fhat_part2, r"Нормированная частота $\hat{f}=f/f_d$", (-0.5, 0.5)),
    (freq_part2, "Частота f, Гц", (-fd_part2 / 2, fd_part2 / 2)),
]):
    axes[0, column].stem(frequency_axis, amplitude_spectrum,
                          linefmt="C0-", markerfmt="C0o", basefmt="k-")
    axes[0, column].set(title="Двусторонний амплитудный спектр",
                        ylabel="|C(k)| = |X(k)| / N", ylim=(-0.05, 1.3))
    axes[1, column].stem(frequency_axis[significant], phase_spectrum[significant],
                          linefmt="C1-", markerfmt="C1o", basefmt="k-")
    axes[1, column].set(title="Фазы значимых спектральных линий",
                        ylabel="Фаза, рад", ylim=(-np.pi / 2, np.pi / 2))
    axes[1, column].set_yticks([-np.pi / 2, -np.pi / 4, 0, np.pi / 4, np.pi / 2],
                              ["−π/2", "−π/4", "0", "π/4", "π/2"])
    for ax in axes[:, column]:
        ax.set_xlabel(xlabel)
        ax.set_xlim(*limits)
        ax.grid(True, alpha=0.3)
    if column == 0:
        for ax in axes[:, column]:
            ax.set_xticks([-0.5, -0.25, -0.1875, -0.125, 0, 0.125, 0.1875, 0.25, 0.5])
            ax.tick_params(axis="x", labelrotation=45)
    else:
        for ax in axes[:, column]:
            ax.set_xticks([-2000, -1000, -750, -500, 0, 500, 750, 1000, 2000])
            ax.tick_params(axis="x", labelrotation=45)
fig.tight_layout()
plt.show()
```

##### Анализ результатов

Временные шкалы связаны как $t_n=nT$, где $T=0.25$ мс. Частотные шкалы связаны как $f=f_d\hat f=4000\hat f$, $\hat\omega=2\pi\hat f$. При $N=64$ шаг сетки равен $\Delta\hat f=1/64$ и $\Delta f=62.5$ Гц. Центрированные индексы $k=-32,\ldots,31$ соответствуют частотам от $-2000$ до $1937.5$ Гц.

| $k$ | $\hat f=k/N$ | $f$, Гц | $\lvert C(k)\rvert$ | Фаза |
| --- | --- | --- | --- | --- |
| $-12$ | $-0.1875$ | $-750$ | $1.16$ | $-\pi/8$ |
| $-8$ | $-0.125$ | $-500$ | $0.58$ | $-\pi/4$ |
| $8$ | $0.125$ | $500$ | $0.58$ | $\pi/4$ |
| $12$ | $0.1875$ | $750$ | $1.16$ | $\pi/8$ |

Каждая косинусоида даёт две линии с амплитудой $A/2$. Спектр вещественного сигнала обладает сопряжённой симметрией: амплитуды на противоположных частотах одинаковы, фазы противоположны.

Гармоники точно попадают на узлы ДПФ ($f_1/\Delta f=8$, $f_2/\Delta f=12$), поэтому растекания нет. Остальные коэффициенты теоретически нулевые; их численно случайные фазы не показаны.

#### 4. ДПФ конечной последовательности и спектр периодической последовательности

Рассматривается конечная последовательность $x_{\mathrm{fin}}(n)=x(n)$ при $0\le n<N$, равная нулю вне этого интервала. Её ДПФ длины $N=64$ вычисляется по формуле

$$X_{\mathrm{fin}}(k)=\sum_{n=0}^{N-1}x(n)e^{-j2\pi kn/N}.$$

Для периодического продолжения тех же 64 отсчётов коэффициенты ряда Фурье равны

$$C(k)=\frac{1}{N}\sum_{n=0}^{N-1}x(n)e^{-j2\pi kn/N}=\frac{X_{\mathrm{fin}}(k)}{N}.$$

Следовательно, при принятой двусторонней нормировке

$$|X_{\mathrm{fin}}(k)|=N|C(k)|=64|C(k)|.$$

Оба графика строятся на центрированной сетке нормированных частот $\hat{f}_k=k/N$. Дополнительный график показывает совпадение после деления модуля ДПФ на $N$.

```python
# ДПФ конечной записи без дополнения нулями.
x_finite = x_part2.copy()
X_finite = np.fft.fft(x_finite, n=N)
X_finite_centered = np.fft.fftshift(X_finite)
dft_magnitude = np.abs(X_finite_centered)

# Проверка ДПФ по определению и соотношения |X| = N|C|.
forward_matrix = np.exp(-2j * np.pi * np.outer(np.arange(N), n_part2) / N)
X_finite_sum = forward_matrix @ x_finite
np.testing.assert_allclose(X_finite, X_finite_sum, rtol=0, atol=1e-11)
np.testing.assert_allclose(dft_magnitude / N, amplitude_spectrum,
                           rtol=0, atol=1e-12)
np.testing.assert_allclose(dft_magnitude[significant] / amplitude_spectrum[significant],
                           N, rtol=0, atol=1e-12)
print("  k       f̂          |X_fin(k)|       |C(k)|      Отношение")
for k, fhat, dft_value, amplitude_value in zip(
        k_centered[significant], fhat_part2[significant],
        dft_magnitude[significant], amplitude_spectrum[significant]):
    print(f"{k:3d}   {fhat: .4f}       {dft_value:8.4f}       {amplitude_value:.4f}       {dft_value / amplitude_value:.0f}")
print(f"Ошибка прямого расчёта ДПФ: {np.max(np.abs(X_finite - X_finite_sum)):.3e}")
print(f"Расхождение |X_fin| / N и |C|: {np.max(np.abs(dft_magnitude / N - amplitude_spectrum)):.3e}")

fig, axes = plt.subplots(3, 1, figsize=(11, 10), sharex=True)
axes[0].stem(fhat_part2, dft_magnitude, linefmt="C0-", markerfmt="C0o", basefmt="k-")
axes[0].set(title="Модуль ДПФ конечной последовательности", ylabel="|X_fin(k)|")
axes[1].stem(fhat_part2, amplitude_spectrum, linefmt="C1-", markerfmt="C1o", basefmt="k-")
axes[1].set(title="Амплитудный спектр периодической последовательности", ylabel="|C(k)|")
axes[2].stem(fhat_part2, dft_magnitude / N, linefmt="C0-", markerfmt="C0o", basefmt="k-",
             label="|X_fin(k)| / N")
axes[2].plot(fhat_part2, amplitude_spectrum, linestyle="none", marker="x", color="C1",
             markersize=7, label="|C(k)|")
axes[2].set(title="Сравнение после нормировки", ylabel="Амплитуда",
            xlabel=r"Нормированная частота $\hat{f}=f/f_d$")
axes[2].legend(loc="upper right")
for ax in axes:
    ax.set_xlim(-0.5, 0.5)
    ax.set_xticks([-0.5, -0.25, -0.1875, -0.125, 0, 0.125, 0.1875, 0.25, 0.5])
    ax.set_xticklabels(["−0.5", "−0.25", "−0.1875", "−0.125", "0", "0.125", "0.1875", "0.25", "0.5"])
    ax.grid(True, alpha=0.3)
fig.tight_layout()
plt.show()
```

##### Анализ результатов

Положение линий одинаково, а их высоты отличаются в $N=64$ раза:

| Нормированная частота $\hat f$ | $\lvert X_{\mathrm{fin}}(k)\rvert$ | $\lvert C(k)\rvert$ |
| --- | --- | --- |
| $\pm0.125$ | $37.12$ | $0.58$ |
| $\pm0.1875$ | $74.24$ | $1.16$ |

Причина различия — нормировка $C(k)=X_{\mathrm{fin}}(k)/N$. После деления модуля ДПФ на 64 графики совпадают. Сравнение относится к отсчётам на сетке ДПФ; спектр конечной записи между этими частотами в общем случае ненулевой.

#### 5. Восстановление периодического аналогового сигнала

Используется формула из задания:

$$x_{\mathrm{rec}}(t)=\frac{1}{N}\sum_{k=-N/2}^{N/2-1}X_a(k)\exp\left(j\frac{2\pi}{NT}kt\right),$$

$$X_a(k)=\begin{cases}X(N+k),&-N/2\le k\le-1,\\X(k),&0\le k\le N/2-1.\end{cases}$$

Для $N=64$ индексы равны $k=-32,\ldots,31$. Время задаётся на интервале $0\le t\le(N-1)T=15.75$ мс с шагом $\Delta t=T/4=0.0625$ мс. Получается $4(N-1)+1=253$ точки, включая оба конца интервала.

Для сравнения исходный аналоговый сигнал вычисляется непосредственно:

$$x_{\mathrm{orig}}(t)=1.16\cos(2\pi\cdot500t+\pi/4)+2.32\cos(2\pi\cdot750t+\pi/8).$$

Амплитудный спектр восстановленного периодического сигнала представлен модулями коэффициентов его ряда Фурье $|X_a(k)|/N$ на частотах $k/(NT)$. Это двусторонний линейчатый спектр; он вычисляется из коэффициентов ряда, а не повторным ДПФ укороченной временной сетки из 253 точек.

```python
# Коэффициенты X_a(k) располагаются от отрицательных частот к положительным.
k_analog = np.arange(-N // 2, N // 2)
X_analog = np.fft.fftshift(X_part2)
np.testing.assert_allclose(X_analog, X_part2[k_analog % N], rtol=0, atol=1e-12)

# Сетка включает оба конца интервала [0, (N-1)T] с шагом T/4.
dt_analog = T / 4
t_analog = np.arange(4 * (N - 1) + 1) * dt_analog
frequencies_analog = k_analog / (N * T)
# Суммирование ряда Фурье на уточнённой временной сетке.
reconstruction_matrix = np.exp(2j * np.pi * np.outer(t_analog, frequencies_analog))
x_analog_complex = reconstruction_matrix @ X_analog / N
x_original_analog = (A1 * np.cos(2 * np.pi * f1 * t_analog + np.pi / 4)
                     + A2 * np.cos(2 * np.pi * f2 * t_analog + np.pi / 8))

# Проверка по исходной формуле; каждый четвёртый узел совпадает с nT.
np.testing.assert_allclose(x_analog_complex, x_original_analog, rtol=0, atol=1e-12)
np.testing.assert_allclose(x_analog_complex[::4], x_part2, rtol=0, atol=1e-12)
np.testing.assert_allclose(t_analog[[0, -1]], [0, (N - 1) * T], rtol=0, atol=1e-15)
assert len(t_analog) == 253
assert np.max(np.abs(x_analog_complex.imag)) < 1e-12
x_analog_restored = x_analog_complex.real
analog_amplitudes = np.abs(X_analog) / N
analog_significant = analog_amplitudes > 1e-12 * analog_amplitudes.max()
np.testing.assert_allclose(frequencies_analog[analog_significant], expected_frequencies,
                           rtol=0, atol=1e-12)
np.testing.assert_allclose((X_analog / N)[analog_significant], expected_coefficients,
                           rtol=0, atol=1e-12)
print(f"Временная сетка: {len(t_analog)} точек; шаг {dt_analog * 1000:.4f} мс")
print(f"Интервал: {t_analog[0] * 1000:.2f}...{t_analog[-1] * 1000:.2f} мс")
print(f"Максимальная ошибка на всей сетке: {np.max(np.abs(x_analog_complex - x_original_analog)):.3e}")
print(f"Ошибка в исходных узлах: {np.max(np.abs(x_analog_complex[::4] - x_part2)):.3e}")
print(f"Остаточная мнимая часть: {np.max(np.abs(x_analog_complex.imag)):.3e}")
print("Частоты спектральных линий, Гц:", frequencies_analog[analog_significant])
print("Амплитуды двусторонних линий:", analog_amplitudes[analog_significant])

fig, axes = plt.subplots(3, 2, figsize=(14, 11))
axes[0, 0].stem(t_part2 * 1000, x_part2, linefmt="C0-", markerfmt="C0o", basefmt="k-")
axes[0, 0].set(title="Периодическая последовательность: 64 отсчёта", ylabel="x(nT)")
axes[0, 1].stem(frequencies_analog, np.abs(X_analog),
                linefmt="C0-", markerfmt="C0o", basefmt="k-")
axes[0, 1].set(title="Модуль ДПФ последовательности", ylabel="|X_a(k)|", xlabel="Частота f, Гц")
axes[1, 0].plot(t_analog * 1000, x_analog_restored, color="C1", linewidth=1.5)
axes[1, 0].plot(t_part2 * 1000, x_part2, linestyle="none", marker=".", color="C0", markersize=5)
axes[1, 0].set(title="Восстановленный сигнал: шаг T/4", ylabel="x_rec(t)")
axes[1, 1].stem(frequencies_analog, analog_amplitudes,
                linefmt="C1-", markerfmt="C1o", basefmt="k-")
axes[1, 1].set(title="Амплитудный спектр восстановленного сигнала",
               ylabel="|X_a(k)| / N", xlabel="Частота f, Гц")
axes[2, 0].plot(t_analog * 1000, x_original_analog, color="C2", linewidth=1.5)
axes[2, 0].set(title="Исходный аналоговый сигнал", ylabel="x_orig(t)")
axes[2, 1].plot(t_analog * 1000, x_original_analog, color="C2", linewidth=1.5, label="Исходный")
axes[2, 1].plot(t_analog * 1000, x_analog_restored, color="C1", linestyle="--",
                linewidth=1.2, label="Восстановленный")
axes[2, 1].set(title="Сопоставление сигналов", ylabel="Амплитуда")
axes[2, 1].legend(loc="upper right")
for ax in [axes[0, 0], axes[1, 0], axes[2, 0], axes[2, 1]]:
    ax.set_xlabel("Время t, мс")
    ax.set_xlim(0, (N - 1) * T * 1000)
    ax.set_ylim(-3.7, 4.2)
for ax in [axes[0, 1], axes[1, 1]]:
    ax.set_xlim(-fd_part2 / 2, fd_part2 / 2)
    ax.set_xticks([-2000, -1000, -750, -500, 0, 500, 750, 1000, 2000])
    ax.tick_params(axis="x", labelrotation=45)
for ax in axes.flat:
    ax.grid(True, alpha=0.3)
fig.tight_layout()
plt.show()
```

Графики показывают исходную последовательность и модуль ДПФ, восстановленный сигнал и его спектр, исходный аналоговый сигнал и сравнение кривых. Максимальное расхождение с исходной формулой на всей сетке — около $2.97\cdot10^{-14}$.

#### 6. Связь спектров и сравнение восстановленного и исходного сигналов

В формуле восстановления комплексный коэффициент гармоники равен $C_a(k)=X_a(k)/N$, а её частота — $f_k=k/(NT)$. Поэтому амплитуды двустороннего спектра аналогового сигнала равны $|X_a(k)|/N$: на $\pm500$ Гц — $0.58$, на $\pm750$ Гц — $1.16$. Фазы сохраняются; каждая сопряжённая пара образует косинусоиду с амплитудой $2|C_a(k)|$.

На совмещённом графике восстановленный и исходный сигналы совпадают. Обе частоты ниже частоты Найквиста $2000$ Гц и точно совпадают с узлами $k/(NT)$, а при восстановлении сохранены все комплексные коэффициенты. Объединение сопряжённых гармоник даёт исходную формулу:

$$x_{\mathrm{rec}}(t)=1.16\cos(2\pi500t+\pi/4)+2.32\cos(2\pi750t+\pi/8)=x_{\mathrm{orig}}(t).$$

Следовательно, совпадение имеет место и между исходными отсчётами. Наблюдаемая ошибка порядка $10^{-14}$ обусловлена округлением; шаг $T/4$ задаёт только сетку отображения.

#### 7. Проверка равенства Парсеваля

Для $N=64$ вычисляются обе части равенства

$$\sum_{n=0}^{63}|x(n)|^2=\frac{1}{64}\sum_{k=0}^{63}|X(k)|^2.$$

Дополнительно проверяется равенство средней мощности, рассчитанной по отсчётам, по коэффициентам ряда Фурье и аналитически. Поскольку частоты гармоник различны и на интервале $NT$ укладывается целое число их периодов, взаимные слагаемые при усреднении исчезают:

$$P=\frac{A_1^2+A_2^2}{2}=\frac{1.16^2+2.32^2}{2}=3.364,$$
$$E=NP=64\cdot3.364=215.296.$$


```python
# Две части равенства Парсеваля при ненормированном прямом ДПФ.
energy_time = float(np.sum(np.abs(x_part2) ** 2))
energy_frequency = float(np.sum(np.abs(X_part2) ** 2) / N)
energy_absolute_error = abs(energy_time - energy_frequency)
energy_relative_error = energy_absolute_error / energy_time

# Средняя мощность и независимая проверка по амплитудам гармоник.
power_time = energy_time / N
power_frequency = float(np.sum(np.abs(C_part2) ** 2))
power_theoretical = (A1 ** 2 + A2 ** 2) / 2
energy_theoretical = N * power_theoretical
np.testing.assert_allclose(energy_time, energy_frequency, rtol=1e-12, atol=1e-12)
np.testing.assert_allclose(energy_time, energy_theoretical, rtol=1e-12, atol=1e-12)
np.testing.assert_allclose([power_time, power_frequency], power_theoretical,
                           rtol=1e-12, atol=1e-12)
print(f"Энергия по временным отсчётам: {energy_time:.3f}")
print(f"Энергия по ДПФ:              {energy_frequency:.3f}")
print(f"Теоретическое значение:     {energy_theoretical:.3f}")
print(f"Абсолютное расхождение: {energy_absolute_error:.3e}")
print(f"Относительное расхождение: {energy_relative_error:.3e}")
print(f"Средняя мощность по отсчётам: {power_time:.3f}")
print(f"Средняя мощность по спектру:  {power_frequency:.3f}")
print(f"Теоретическая мощность:       {power_theoretical:.3f}")
print("Равенство Парсеваля выполнено с точностью вычислений.")
```

| Средняя мощность $P$ | $3.364$ | $3.364$ | $3.364$ |

Равенство Парсеваля выполняется с точностью вычислений:

$$E=\frac{2\cdot37.12^2+2\cdot74.24^2}{64}=215.296.$$

Оно показывает, что энергия одной и той же записи одинакова во временном и частотном представлениях при учёте нормировки $1/N$. Здесь $E$ относится к массиву из 64 отсчётов, а $P=E/N$ — к средней мощности периодической последовательности.
