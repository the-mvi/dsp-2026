## Основные теоретические положения

### Линейные дискретные системы

Линейная дискретная система (ЛДС) описывается разностным уравнением. Для рекурсивного звена 2-го порядка оно имеет вид

$$
y(n) = b_0x(n) + b_1x(n-1) + b_2x(n-2) - a_1y(n-1) - a_2y(n-2),
$$

где $b_i$, $a_k$ - вещественные коэффициенты, $a_0 = 1$. Применив к обеим частям уравнения z-преобразование, получим передаточную функцию

$$
H(z) = \frac{Y(z)}{X(z)} = \frac{b_0 + b_1z^{-1} + b_2z^{-2}}{1 + a_1z^{-1} + a_2z^{-2}}.
$$

**Импульсная характеристика** $h(n)$ - реакция системы на цифровой единичный импульс

$$
u_0(n) = \begin{cases}1, & n = 0,\\ 0, & n \ne 0\end{cases}
$$

при нулевых начальных условиях. У рекурсивной системы (хотя бы один из коэффициентов $a_1, a_2$ не равен нулю) импульсная характеристика бесконечна, такие системы называют БИХ-системами.

Реакция на произвольное воздействие $x(n)$ при нулевых начальных условиях вычисляется по **формуле свёртки**

$$
y(n) = \sum_{m=0}^{n} h(m)\,x(n-m).
$$

При свёртке двух последовательностей конечной длины $L_1$ и $L_2$ результат имеет длину $L_1 + L_2 - 1$.

**Нули и полюсы.** Нули $z_i$ - корни числителя $H(z)$, полюсы $p_k$ - корни знаменателя. Через них передаточная функция записывается в виде произведения простейших множителей:

$$
H(z) = b_0\frac{(1 - z_1z^{-1})(1 - z_2z^{-1})}{(1 - p_1z^{-1})(1 - p_2z^{-1})}.
$$

ЛДС устойчива, если все её полюсы лежат внутри единичной окружности: $|p_k| < 1$. Положение нулей на устойчивость не влияет.

**Частотная характеристика** получается из передаточной функции подстановкой $z = e^{j\hat\omega}$:

$$
H(e^{j\hat\omega}) = \frac{b_0 + b_1e^{-j\hat\omega} + b_2e^{-j2\hat\omega}}{1 + a_1e^{-j\hat\omega} + a_2e^{-j2\hat\omega}} = A(\hat\omega)\,e^{j\varphi(\hat\omega)},
$$

где $A(\hat\omega) = |H(e^{j\hat\omega})|$ - АЧХ, $\varphi(\hat\omega) = \arg H(e^{j\hat\omega})$ - ФЧХ, $\hat\omega = \omega T = 2\pi f/f_д$ - нормированная частота, рад. Частотная характеристика периодична с периодом $2\pi$ (в шкале $f$ - с периодом $f_д$). При вещественных коэффициентах АЧХ - чётная, а ФЧХ - нечётная функция частоты, поэтому основную полосу $[-\pi; \pi]$ достаточно рассматривать на отрезке $[0; \pi]$, которому соответствует $[0; f_д/2]$.

Значение АЧХ пропорционально отношению произведений расстояний от точки $e^{j\hat\omega}$ на единичной окружности до нулей и до полюсов:

$$
A(\hat\omega) = |b_0|\frac{|e^{j\hat\omega} - z_1|\,|e^{j\hat\omega} - z_2|}{|e^{j\hat\omega} - p_1|\,|e^{j\hat\omega} - p_2|}.
$$

Поэтому вблизи аргумента нуля АЧХ уменьшается, а вблизи аргумента полюса - возрастает, и тем сильнее, чем ближе нуль или полюс к единичной окружности.

### Дискретное преобразование Фурье

Периодическая последовательность $\tilde x(n)$ с периодом $N$ представляется дискретным рядом Фурье:

$$
\tilde X(k) = \sum_{n=0}^{N-1}\tilde x(n)\,e^{-j\frac{2\pi}{N}kn}, \qquad
\tilde x(n) = \frac{1}{N}\sum_{k=0}^{N-1}\tilde X(k)\,e^{j\frac{2\pi}{N}kn}.
$$

Спектр периодической последовательности дискретный (линейчатый) и тоже периодический с периодом $N$.

Для последовательности $x(n)$ конечной длины $N$ дискретным преобразованием Фурье называют те же формулы при $n, k = 0, \ldots, N-1$. Спектр конечной последовательности $X(e^{j\hat\omega})$ непрерывный, а ДПФ $X(k)$ - это $N$ его отсчётов на частотах $\hat\omega_k = 2\pi k/N$. Значения ДПФ конечной последовательности совпадают с коэффициентами $\tilde X(k)$ её периодического продолжения.

Используемые шкалы:
- $nT$ - дискретное время, $n$ - дискретное нормированное время ($n = nT/T$);
- $k$ - дискретная нормированная частота, $\hat\omega_k = 2\pi k/N$ - соответствующая ей нормированная частота;
- $f_k = k\,\Delta f$ - абсолютная частота, где $\Delta f = f_д/N$ - шаг по частоте (разрешение ДПФ).

Амплитудный спектр определяется модулем, фазовый - аргументом $X(k)$. Амплитудный спектр периодической последовательности в работе нормирован на $N$: величины $|X(k)|/N$ - это модули комплексных амплитуд экспонент $e^{j2\pi kn/N}$, из которых складывается последовательность. Гармоника $A\cos(2\pi fnT + \varphi)$ дает в таком спектре две линии высотой $A/2$ с фазами $\pm\varphi$. Для вещественной последовательности $|X(k)| = |X(N-k)|$, $\arg X(k) = -\arg X(N-k)$, т.е. спектр симметричен относительно $k = N/2$ (частоты $f_д/2$), а отсчёты с $k > N/2$ соответствуют отрицательным частотам $k - N$.

Если спектр аналогового периодического сигнала ограничен частотой $f_д/2$ (выполняется теорема Котельникова), то сигнал можно точно восстановить по отсчётам ДПФ одного периода в виде конечной суммы гармоник с частотами $k/(NT)$ (формула приведена в п. 5 части 2).

**Равенство Парсеваля** для ДПФ:

$$
\sum_{n=0}^{N-1}|x(n)|^2 = \frac{1}{N}\sum_{k=0}^{N-1}|X(k)|^2.
$$

#### 1. Импульсная характеристика

Импульсную характеристику находим как реакцию звена на $u_0(n)$. Для расчёта реакции по разностному уравнению написана функция `diff_eq`: значения $x$ и $y$ с отрицательными индексами в ней считаются нулевыми, что соответствует нулевым начальным условиям. Результат сверяется со встроенной функцией `scipy.signal.lfilter`.

```python
def diff_eq(b, a, x):
    """Реакция по разностному уравнению при нулевых начальных условиях."""
    y = np.zeros(len(x))
    for n in range(len(x)):
        s = 0.0
        for i in range(len(b)):
            if n - i >= 0:
                s += b[i] * x[n - i]
        for k in range(1, len(a)):
            if n - k >= 0:
                s -= a[k] * y[n - k]
        y[n] = s / a[0]
    return y


u0 = np.zeros(N1)
u0[0] = 1
h = diff_eq(b, a, u0)

print('Расхождение с signal.lfilter:', np.max(np.abs(h - signal.lfilter(b, a, u0))))
for n in range(N1):
    print(f'h({n:2d}) = {h[n]: .6f}')
```

```python
n1 = np.arange(N1)

plt.stem(n1, h)
plt.title(f'Импульсная характеристика h(n), N1 = {N1}')
plt.xlabel('n')
plt.ylabel('h(n)')
plt.show()
```

**Аналитическая формула.** Дискриминант знаменателя $a_1^2 - 4a_2 = 0.7705 - 2.656 < 0$, поэтому полюсы комплексно-сопряжённые: $p_{1,2} = re^{\pm j\theta}$, где

$$
r = \sqrt{a_2}, \qquad \theta = \arccos\left(-\frac{a_1}{2r}\right).
$$

Степени числителя и знаменателя (по $z^{-1}$) одинаковы, поэтому при разложении $H(z)$ на простые дроби появляется целая часть $C = b_2/a_2$:

$$
H(z) = C + \frac{A}{1 - p_1z^{-1}} + \frac{A^*}{1 - p_1^*z^{-1}}.
$$

Слагаемому $\frac{1}{1 - pz^{-1}}$ соответствует последовательность $p^n u_1(n)$, где $u_1(n)$ - единичный скачок, а постоянной $C$ - импульс $C u_0(n)$. Сумма двух сопряжённых слагаемых вещественна, и импульсная характеристика принимает вид

$$
h(n) = C\,u_0(n) + 2|A|\,r^n\cos(\theta n + \psi)\,u_1(n), \qquad \psi = \arg A.
$$

Множитель $u_1(n)$ отражает нулевые начальные условия: при $n < 0$ $h(n) = 0$. Вычет $A$ найдём с помощью `signal.residuez` и проверим формулу.

```python
res, p_res, k_res = signal.residuez(b, a)
i = np.argmax(p_res.imag)          # полюс с положительной мнимой частью
A_res, p1 = res[i], p_res[i]
C = k_res[0]

r = np.abs(p1)
theta = np.angle(p1)
psi = np.angle(A_res)

print(f'C = b2/a2 = {C:.6f}')
print(f'A = {A_res:.6f}, |A| = {abs(A_res):.6f}, psi = {psi / np.pi:.6f}pi')
print(f'r = {r:.6f}, theta = {theta / np.pi:.6f}pi')

h_formula = 2 * abs(A_res) * r ** n1 * np.cos(theta * n1 + psi)
h_formula[0] += C
print('Расхождение формулы с расчётом:', np.max(np.abs(h - h_formula)))

envelope = 2 * abs(A_res) * r ** n1
plt.stem(n1, h, label='h(n) по разностному уравнению')
plt.plot(n1, h_formula, 'rx', label='h(n) по формуле')
plt.plot(n1, envelope, 'k--', lw=1, label='огибающая $\\pm 2|A|r^n$')
plt.plot(n1, -envelope, 'k--', lw=1)
plt.title('Проверка аналитической формулы')
plt.xlabel('n')
plt.legend()
plt.show()
```

С учётом найденных значений:

$$
h(n) = 1.397590\,u_0(n) + 2.304059\cdot 0.814862^{\,n}\cos(0.681054\pi n + 0.615467\pi)\,u_1(n).
$$

Значения по формуле совпадают с рассчитанными по разностному уравнению с точностью до ошибок округления.

**Длина импульсной характеристики.** Звено рекурсивное, поэтому длина его импульсной характеристики в действительности бесконечна: множитель $r^n$ убывает, но при конечном $n$ в ноль не обращается. Число $N_1 = 24$ - это только количество вычисленных отсчётов. Огибающая уменьшается в 100 раз примерно за 23 отсчёта ($0.8149^{23} \approx 0.009$), поэтому к концу рассчитанного отрезка значения $h(n)$ имеют порядок 0.01-0.03. #### 2. Реакция по формуле свёртки

Воздействие - прямоугольный импульс, $int(N_2/2) = int(17) = 17$:

$$
x(n) = \begin{cases}1, & 0 \le n < 17,\\ 0, & 17 \le n \le 33.\end{cases}
$$

Реакцию вычисляем по формуле свёртки с импульсной характеристикой длины $N_1$:

$$
y_1(n) = \sum_{m=0}^{N_1-1}h(m)\,x(n-m).
$$

Свёртка реализована по определению в функции `conv`, результат сравнивается с `np.convolve`.

```python
L = int(np.round(N2 / 2))     # длина импульса
n2 = np.arange(N2)
x = np.where(n2 < L, 1.0, 0.0)


def conv(h, x):
    """Линейная свёртка по определению."""
    y = np.zeros(len(h) + len(x) - 1)
    for n in range(len(y)):
        for m in range(len(h)):
            if 0 <= n - m < len(x):
                y[n] += h[m] * x[n - m]
    return y


y1_full = conv(h, x)
print('Расхождение с np.convolve:', np.max(np.abs(y1_full - np.convolve(h, x))))
print('Длина последовательности x(n):', len(x))
print('Длина импульса (число единичных отсчётов):', int(np.sum(x)))
print(f'Длина свёртки: {len(y1_full)} = {N1} + {N2} - 1')
last = np.nonzero(np.abs(y1_full) > 1e-12)[0][-1]
print('Последний ненулевой отсчёт свёртки: n =', last)

y1 = y1_full[:N2]             # ограничение длиной воздействия
```

```python
n_full = np.arange(len(y1_full))

fig, ax = plt.subplots(3, 1, figsize=(10, 10))

ax[0].stem(n2, x)
ax[0].set_title('Воздействие x(n)')
ax[0].set_xlabel('n')
ax[0].set_xlim(-1, len(y1_full))

ax[1].stem(n_full, y1_full)
ax[1].axvline(N2 - 0.5, color='r', ls='--', label='граница n = N2 - 1')
ax[1].axvline(last + 0.5, color='g', ls=':', label=f'последний ненулевой отсчёт n = {last}')
ax[1].set_title(f'Полная свёртка, длина {len(y1_full)}')
ax[1].set_xlabel('n')
ax[1].set_xlim(-1, len(y1_full))
ax[1].legend()

ax[2].stem(n2, y1)
ax[2].set_title(f'Реакция y1(n), ограниченная длиной воздействия N2 = {N2}')
ax[2].set_xlabel('n')
ax[2].set_xlim(-1, len(y1_full))

fig.tight_layout()
plt.show()
```

- **Длина импульса** $x(n)$ - 17 отсчётов (единицы при $n = 0, \ldots, 16$). Вся последовательность имеет длину $N_2 = 34$, оставшиеся 17 отсчётов нулевые.
- **Длина свёртки.** Аналитически длина свёртки последовательностей длины $N_1 = 24$ и $N_2 = 34$ равна $N_1 + N_2 - 1 = 57$ ($n = 0, \ldots, 56$). По графику ненулевые значения есть только при $n = 0, \ldots, 39$, т.е. на $17 + 24 - 1 = 40$ отсчётах: нулевая часть $x(n)$ в свёртку вклада не дает, поэтому отсчёты с $n = 40, \ldots, 56$ равны нулю.
- **Почему свёртку ограничивают длиной воздействия.** Реакция рекурсивного звена на самом деле бесконечна, а свёртка с усечённой $h(n)$ точна только на начальном участке: при $n \ge N_1 = 24$ в сумму должны входить отброшенные отсчёты $h(24), h(25), \ldots$. Обрыв свёртки при $n = 39$ - следствие усечения $h(n)$, а не свойство звена. Поэтому реакцию рассматривают на том же интервале $n = 0, \ldots, N_2-1$, на котором задано воздействие. На этом же интервале в п. 3 вычисляется реакция по разностному уравнению, и результаты можно сравнить.

#### 3. Реакция по разностному уравнению

Разностное уравнение звена с заданными коэффициентами:

$$
y_2(n) = 0.58x(n) - 0.610972x(n-1) + 0.928x(n-2) - 0.8778y_2(n-1) - 0.664y_2(n-2),
$$

начальные условия нулевые: $x(-1) = x(-2) = y_2(-1) = y_2(-2) = 0$.

```python
y2 = diff_eq(b, a, x)
print('Расхождение с signal.lfilter:', np.max(np.abs(y2 - signal.lfilter(b, a, x))))
print('Длина воздействия:', len(x), ', длина реакции:', len(y2))

d = y1 - y2
print(f'max|y1 - y2| при n = 0..{N1 - 1}:', np.max(np.abs(d[:N1])))
print(f'max|y1 - y2| при n = {N1}..{N2 - 1}:', np.max(np.abs(d[N1:])))

# если взять h длины N2, то свёртка совпадёт с разностным уравнением на всём интервале
u0_long = np.zeros(N2)
u0_long[0] = 1
h_long = diff_eq(b, a, u0_long)
print('max|y1 - y2| при h длины N2:', np.max(np.abs(np.convolve(h_long, x)[:N2] - y2)))

H0 = np.sum(b) / np.sum(a)    # H(z) при z = 1, т.е. на нулевой частоте
print(f'H(1) = {H0:.6f}')
```

```python
fig, ax = plt.subplots(2, 2, figsize=(13, 8))

ax[0, 0].stem(n2, x)
ax[0, 0].set_title('Воздействие x(n)')
ax[0, 0].set_xlabel('n')

ax[0, 1].stem(n2, y2)
ax[0, 1].axhline(H0, color='r', ls='--', label=f'H(1) = {H0:.3f}')
ax[0, 1].set_title('Реакция y2(n) по разностному уравнению')
ax[0, 1].set_xlabel('n')
ax[0, 1].legend()

ax[1, 0].plot(n2, y1, 'o-', ms=4, label='y1(n), свёртка')
ax[1, 0].plot(n2, y2, 'x--', ms=6, label='y2(n), разностное уравнение')
ax[1, 0].axvline(N1 - 0.5, color='gray', ls=':')
ax[1, 0].set_title('Сравнение реакций')
ax[1, 0].set_xlabel('n')
ax[1, 0].legend(loc='upper right')

ax[1, 1].stem(n2, d)
ax[1, 1].axvline(N1 - 0.5, color='gray', ls=':')
ax[1, 1].set_title('Разность y1(n) - y2(n)')
ax[1, 1].set_xlabel('n')

fig.tight_layout()
plt.show()
```

**Длины.** Длина воздействия - 34 отсчёта, длина реакции $y_2(n)$ - тоже 34, так как рекурсия вычислялась для $n = 0, \ldots, N_2-1$. Теоретически реакция бесконечна: после окончания импульса ($n \ge 17$) вход нулевой, но за счёт обратной связи звено продолжает выдавать затухающие колебания.

**Сравнение $y_1(n)$ и $y_2(n)$.** При $n = 0, \ldots, 23$ реакции совпадают с точностью до ошибок округления, так как оба способа описывают одну и ту же систему. При $n \ge 24$ появляется расхождение (до 0.0168): в свёртке использовалась импульсная характеристика, усечённая до $N_1 = 24$ отсчётов. Если взять $h(n)$ длины $N_2$, результаты совпадают на всём интервале.

**Вид реакции.** На вершине импульса реакция колеблется и приближается к значению $H(1) = \frac{b_0+b_1+b_2}{1+a_1+a_2} \approx 0.353$ - постоянная составляющая воздействия ослабляется почти в три раза. После фронтов ($n = 0$ и $n = 17$) возникают затухающие колебания с периодом около трёх отсчётов, т.е. на частоте, близкой к аргументу полюса $\theta \approx 0.68\pi$ ($2\pi/\theta \approx 2.9$).

#### 4. Нули и полюсы передаточной функции

Умножим числитель и знаменатель на $z^2$:

$$
H(z) = \frac{0.58z^2 - 0.610972z + 0.928}{z^2 + 0.8778z + 0.664}.
$$

Нули - корни числителя, полюсы - корни знаменателя. Найдём их функцией `np.roots` и запишем в алгебраической $\alpha + j\beta$ и показательной $\rho e^{j\phi}$ формах.

```python
zeros = np.roots(b)
poles = np.roots(a)
zeros = zeros[np.argsort(-zeros.imag)]
poles = poles[np.argsort(-poles.imag)]


def show_root(name, v):
    sign = '+' if v.imag >= 0 else '-'
    print(f'{name} = {v.real:.6f} {sign} j{abs(v.imag):.6f} = '
          f'{abs(v):.6f}*exp(j*{np.angle(v):.6f}) = {abs(v):.6f}*exp(j*{np.angle(v) / np.pi:.6f}pi)')


for i, v in enumerate(zeros, 1):
    show_root(f'z{i}', v)
for i, v in enumerate(poles, 1):
    show_root(f'p{i}', v)

print()
print('sqrt(b2/b0) =', np.sqrt(b2 / b0), ', sqrt(a2) =', np.sqrt(a2))

# обратная проверка: перемножаем простейшие множители
print('b по нулям:  ', b0 * np.real(np.poly(zeros)))
print('a по полюсам:', np.real(np.poly(poles)))
```

| Корни | Алгебраическая форма | Модуль $\rho$ | Аргумент $\phi$, рад | $\phi/\pi$ | Показательная форма |
| :--- | :--- | :--- | :--- | :--- | :--- |
| $z_{1,2}$ | $0.526700 \pm j1.150038$ | 1.264911 | $\pm 1.141322$ | $\pm 0.363294$ | $1.264911\,e^{\pm j0.363294\pi}$ |
| $p_{1,2}$ | $-0.438900 \pm j0.686562$ | 0.814862 | $\pm 2.139593$ | $\pm 0.681054$ | $0.814862\,e^{\pm j0.681054\pi}$ |

**Связь между формами.** Модуль и аргумент находятся по вещественной и мнимой частям: $\rho = \sqrt{\alpha^2 + \beta^2}$, $\phi = \operatorname{arctg}(\beta/\alpha)$ с учётом четверти (у полюсов $\alpha < 0$, поэтому их аргумент лежит во второй/третьей четверти и по модулю больше $\pi/2$). Обратно: $\alpha = \rho\cos\phi$, $\beta = \rho\sin\phi$.

Коэффициенты $H(z)$ вещественные, поэтому нули и полюсы образуют комплексно-сопряжённые пары: модули у них одинаковые, а аргументы отличаются знаком. По теореме Виета $z_1z_2 = \rho_z^2 = b_2/b_0$ и $p_1p_2 = r^2 = a_2$, откуда $\rho_z = \sqrt{1.6} = 1.264911$ и $r = \sqrt{0.664} = 0.814862$; суммы корней дают $2\rho_z\cos\phi_z = -b_1/b_0 = 1.0534$ и $2r\cos\theta = -a_1 = -0.8778$, т.е. удвоенные вещественные части.

Передаточная функция в виде произведения простейших множителей:

$$
H(z) = 0.58\,\frac{\left(1 - 1.264911e^{j0.363294\pi}z^{-1}\right)\left(1 - 1.264911e^{-j0.363294\pi}z^{-1}\right)}{\left(1 - 0.814862e^{j0.681054\pi}z^{-1}\right)\left(1 - 0.814862e^{-j0.681054\pi}z^{-1}\right)}.
$$

Перемножение множителей возвращает исходные коэффициенты $b$ и $a$ (см. вывод выше).

```python
z_zpk, p_zpk, k_zpk = signal.tf2zpk(b, a)
print('Нули (tf2zpk):  ', np.sort_complex(z_zpk))
print('Полюсы (tf2zpk):', np.sort_complex(p_zpk))
print('Отличие от п. 4:', np.max(np.abs(np.sort_complex(z_zpk) - np.sort_complex(zeros))),
      np.max(np.abs(np.sort_complex(p_zpk) - np.sort_complex(poles))))
print('Модули полюсов:', np.abs(p_zpk))
print('Звено устойчиво:', np.all(np.abs(p_zpk) < 1))

t_circ = np.linspace(0, 2 * np.pi, 400)
fig, ax = plt.subplots(figsize=(7, 7))
ax.plot(np.cos(t_circ), np.sin(t_circ), 'k--', lw=1, label='единичная окружность')
ax.plot(z_zpk.real, z_zpk.imag, 'o', ms=10, mfc='none', mew=2, label='нули')
ax.plot(p_zpk.real, p_zpk.imag, 'x', ms=10, mew=2, color='r', label='полюсы')
for name, v in [('z1', zeros[0]), ('z2', zeros[1]), ('p1', poles[0]), ('p2', poles[1])]:
    ax.annotate(f'{name} ({v.real:.3f}; {v.imag:.3f})', (v.real, v.imag),
                xytext=(8, 8), textcoords='offset points')
ax.axhline(0, color='gray', lw=0.8)
ax.axvline(0, color='gray', lw=0.8)
ax.set_xlim(-1.5, 1.7)
ax.set_ylim(-1.5, 1.5)
ax.set_aspect('equal')
ax.set_xlabel('Re z')
ax.set_ylabel('Im z')
ax.set_title('Карта нулей и полюсов')
ax.legend(loc='lower left')
plt.show()
```

Это согласуется с затуханием импульсной характеристики в п. 1 (множитель $r^n \to 0$). Нули находятся вне единичной окружности ($|z_{1,2}| = 1.264911 > 1$), но на устойчивость это не влияет.
- **Совпадение с п. 4.** Нули и полюсы, найденные `tf2zpk`, совпадают с вычисленными в п. 4, координаты точек на карте соответствуют таблице.

#### 6. АЧХ и ФЧХ в шкале нормированных частот

Частотную характеристику вычислим по формуле $H(e^{j\hat\omega})$ из теоретической части на отрезке $\hat\omega \in [0; \pi]$ и сверим с `signal.freqz`. Пунктиром на графиках отмечены аргументы нуля и полюса.

```python
w = np.linspace(0, np.pi, 2001)
e = np.exp(-1j * w)
H = (b0 + b1 * e + b2 * e ** 2) / (a0 + a1 * e + a2 * e ** 2)

_, H_check = signal.freqz(b, a, worN=w)
print('Расхождение с signal.freqz:', np.max(np.abs(H - H_check)))

A = np.abs(H)
phi = np.angle(H)

i_min, i_max = np.argmin(A), np.argmax(A)
print(f'A(0) = {A[0]:.4f}, A(pi) = {A[-1]:.4f}')
print(f'Минимум АЧХ:  {A[i_min]:.4f} при w = {w[i_min] / np.pi:.4f}pi')
print(f'Максимум АЧХ: {A[i_max]:.4f} при w = {w[i_max] / np.pi:.4f}pi')
i_one = np.nonzero(A > 1)[0][0]
print(f'АЧХ становится больше 1 при w = {w[i_one] / np.pi:.4f}pi')

jumps = np.nonzero(np.abs(np.diff(phi)) > np.pi)[0]
for j in jumps:
    print(f'Скачок ФЧХ при w = {w[j] / np.pi:.4f}pi: с {phi[j]:.4f} на {phi[j + 1]:.4f} рад, A = {A[j]:.4f}')

arg_z = np.angle(zeros[0]) / np.pi
arg_p = np.angle(poles[0]) / np.pi
```

```python
fig, ax = plt.subplots(2, 1, figsize=(10, 8), sharex=True)

ax[0].plot(w / np.pi, A)
ax[0].axvline(arg_z, color='g', ls='--', label=f'аргумент нуля {arg_z:.3f}$\\pi$')
ax[0].axvline(arg_p, color='r', ls='--', label=f'аргумент полюса {arg_p:.3f}$\\pi$')
ax[0].plot(w[[i_min, i_max]] / np.pi, A[[i_min, i_max]], 'ko')
ax[0].set_title('АЧХ')
ax[0].set_ylabel('$A(\\hat{\\omega})$')
ax[0].legend()

ax[1].plot(w / np.pi, phi, label='главное значение')
ax[1].plot(w / np.pi, np.unwrap(phi), '--', label='после устранения скачков на $2\\pi$')
ax[1].axvline(arg_z, color='g', ls='--')
ax[1].axvline(arg_p, color='r', ls='--')
ax[1].set_title('ФЧХ')
ax[1].set_ylabel('$\\varphi(\\hat{\\omega})$, рад')
ax[1].set_xlabel('$\\hat{\\omega}/\\pi$')
ax[1].set_xlim(0, 1)
ax[1].legend()

fig.tight_layout()
plt.show()
```

- **Границы основной полосы.** Частотная характеристика периодична по $\hat\omega$ с периодом $2\pi$, основная полоса $\hat\omega \in [-\pi; \pi]$. Так как АЧХ чётная, а ФЧХ нечётная, графики построены для $\hat\omega \in [0; \pi]$.
- **Соответствие карте нулей и полюсов.** Аргумент нулей $0.3633\pi$ - вблизи него АЧХ имеет провал: минимум $0.1911$ при $\hat\omega \approx 0.343\pi$. До нуля АЧХ не опускается, так как нули не лежат на единичной окружности ($|z| = 1.265$). Аргумент полюсов $0.6811\pi$ - вблизи него резонансный максимум $5.2642$ при $\hat\omega \approx 0.6965\pi$. Полюс расположен ближе к окружности, чем нуль ($1 - 0.815 = 0.185$ против $1.265 - 1 = 0.265$), поэтому пик выражен сильнее, чем провал. Экстремумы немного смещены относительно аргументов корней, так как на АЧХ влияют и сопряжённые корни, и соседний корень другого типа.
- **Скачок на $\pi$.** Скачок ФЧХ на $\pi$ происходит на частоте, где АЧХ равна нулю (вектор $H(e^{j\hat\omega})$ проходит через ноль и меняет направление). Такое возможно, только если нуль лежит на единичной окружности. В нашем случае $|z| \ne 1$, АЧХ нигде не обращается в ноль, поэтому скачков на $\pi$ нет. На графике есть скачок с $-\pi$ на $\pi$ (на $2\pi$) при $\hat\omega \approx 0.529\pi$, где $A \approx 1.03$, - это следствие вычисления главного значения аргумента в интервале $(-\pi; \pi]$, а не свойство звена. После устранения скачков на $2\pi$ (`np.unwrap`) ФЧХ непрерывна.
- **Какие составляющие подавлены.** $A(0) = 0.353$, и до $\hat\omega \approx 0.53\pi$ АЧХ меньше единицы, а на высоких частотах звено усиливает сигнал (до $5.26$ в резонансе, $A(\pi) = 2.695$). Значит, преимущественно подавлены низкочастотные составляющие воздействия. Это видно и по реакции в п. 3: постоянный уровень импульса передается с коэффициентом $0.353$, а фронты, содержащие высокие частоты, вызывают колебания с частотой вблизи резонанса.

#### 7. АЧХ и ФЧХ в шкале частот $f$

Связь нормированной и абсолютной частоты:

$$
f = \frac{\hat\omega}{2\pi}f_д, \qquad \hat\omega = \frac{2\pi f}{f_д}, \qquad f_д = 4000\ \text{Гц}.
$$

АЧХ и ФЧХ вычислим через `signal.freqz` с параметром `fs`, задав сетку непосредственно в герцах.

```python
f = np.linspace(0, fd / 2, 2001)
_, Hf = signal.freqz(b, a, worN=f, fs=fd)

print('Расхождение с расчётом в шкале w:', np.max(np.abs(Hf - H)))
for wn in [0, 0.25, 0.5, 0.75, 1]:
    print(f'w = {wn:4}pi  ->  f = {wn * np.pi / (2 * np.pi) * fd:6.0f} Гц')
print(f'Минимум АЧХ при f = {f[i_min]:.1f} Гц, максимум при f = {f[i_max]:.1f} Гц')

fig, ax = plt.subplots(2, 1, figsize=(10, 8), sharex=True)
ax[0].plot(f, np.abs(Hf))
ax[0].set_title(f'АЧХ, fд = {fd} Гц')
ax[0].set_ylabel('A(f)')
ax[1].plot(f, np.angle(Hf))
ax[1].set_title(f'ФЧХ, fд = {fd} Гц')
ax[1].set_ylabel('$\\varphi(f)$, рад')
ax[1].set_xlabel('f, Гц')
ax[1].set_xlim(0, fd / 2)
fig.tight_layout()
plt.show()
```

- **Границы основной полосы.** В шкале $f$ основная полоса $[-f_д/2; f_д/2] = [-2000; 2000]$ Гц, на графиках показана её половина $[0; 2000]$ Гц. Период частотной характеристики равен $f_д = 4000$ Гц.
- **Соответствие частот $\hat\omega$ и $f$.** Шкалы связаны линейно: $\hat\omega = 0 \leftrightarrow 0$ Гц, $\hat\omega = \pi/2 \leftrightarrow 1000$ Гц, $\hat\omega = \pi \leftrightarrow f_д/2 = 2000$ Гц. Провал АЧХ ($0.343\pi$) приходится на $\approx 686$ Гц, резонанс ($0.6965\pi$) - на $\approx 1393$ Гц. Сами значения АЧХ и ФЧХ при смене шкалы не меняются, меняется только подпись оси частот.

ОДПФ от ДПФ возвращает исходную последовательность: расхождение и мнимая часть результата не превышают $10^{-15}$, т.е. это ошибки вычислений с плавающей точкой. ДПФ и ОДПФ взаимно обратны, и при переходе в частотную область информация о последовательности не теряется.

#### 3. Амплитудный и фазовый спектры

Амплитудный спектр - $|X(k)|/N$, фазовый - $\arg X(k)$. Для отсчётов с нулевой (на уровне ошибок округления) амплитудой фаза не определена, поэтому она принимается равной нулю. Спектры построены в шкале дискретных нормированных частот $k$ и в шкале абсолютных частот $f_k = k\Delta f$.

```python
k = np.arange(N)
fk = k * df

amp = np.abs(X) / N
ph = np.angle(X)
ph[amp < 1e-10 * amp.max()] = 0

lines = np.nonzero(amp > 1e-10 * amp.max())[0]
print(' k     f, Гц     |X(k)|   |X(k)|/N   arg/pi')
for kk in lines:
    print(f'{kk:2d} {fk[kk]:9.2f} {np.abs(X[kk]):9.4f} {amp[kk]:9.4f} {ph[kk] / np.pi:8.4f}')

fig, ax = plt.subplots(2, 2, figsize=(13, 8))
ax[0, 0].stem(k, amp)
ax[0, 0].set_title('Амплитудный спектр')
ax[0, 0].set_xlabel('k')
ax[0, 0].set_ylabel('|X(k)|/N')
ax[1, 0].stem(k, ph / np.pi)
ax[1, 0].set_title('Фазовый спектр')
ax[1, 0].set_xlabel('k')
ax[1, 0].set_ylabel('arg X(k) / $\\pi$')
ax[0, 1].stem(fk, amp)
ax[0, 1].set_title('Амплитудный спектр')
ax[0, 1].set_xlabel('f, Гц')
ax[0, 1].set_ylabel('|X(k)|/N')
ax[1, 1].stem(fk, ph / np.pi)
ax[1, 1].set_title('Фазовый спектр')
ax[1, 1].set_xlabel('f, Гц')
ax[1, 1].set_ylabel('arg X(k) / $\\pi$')
for i in range(2):
    ax[i, 0].axvline(N / 2, color='gray', ls=':')
    ax[i, 1].axvline(fd2 / 2, color='gray', ls=':')
fig.tight_layout()
plt.show()
```

- **Связь дискретного нормированного времени с дискретным временем:** $n = nT/T$, т.е. нормированное время - это дискретное время, выраженное в периодах дискретизации ($t_n = nT$, $T = 0.1$ мс).
- **Связь частоты $f$ с дискретной нормированной частотой:** $f_k = k\,f_д/N = k\Delta f$, $\Delta f = 156.25$ Гц; через нормированную частоту $\hat\omega_k = 2\pi k/N = 2\pi f_k/f_д$. Частоте $f_1 = 1250$ Гц соответствует $k = 8$ ($\hat\omega = \pi/4$), частоте $f_2 = 1875$ Гц - $k = 12$ ($\hat\omega = 3\pi/8$). Отсчёт $k = N/2 = 32$ соответствует $f_д/2 = 5000$ Гц, отсчёты $k = 33, \ldots, 63$ - отрицательным частотам $(k - N)\Delta f$: $k = 52 \leftrightarrow -1875$ Гц, $k = 56 \leftrightarrow -1250$ Гц.
- **Вид спектров.** В спектре четыре ненулевые линии: $k = 8$ и $56$ высотой $0.52 = A_1/2$, $k = 12$ и $52$ высотой $1.04 = A_2/2$. Остальные отсчёты нулевые: частоты гармоник кратны $\Delta f$, поэтому растекания спектра нет. Амплитудный спектр симметричен относительно $k = N/2$, фазовый - антисимметричен, как и должно быть для вещественной последовательности. Фазы линий на положительных частотах равны начальным фазам гармоник: $\arg X(8) = \pi/4$, $\arg X(12) = \pi/8$, на отрицательных - противоположны по знаку. Если сложить симметричные линии, получим амплитуды гармоник: $2\cdot 0.52 = 1.04 = A_1$, $2\cdot 1.04 = 2.08 = A_2$.

#### 4. ДПФ конечной последовательности

Теперь рассматриваем те же 64 отсчёта как конечную последовательность (вне $n = 0, \ldots, 63$ она равна нулю). ДПФ вычислим непосредственно по формуле, без БПФ. Для наглядности на тот же график нанесём модуль непрерывного спектра конечной последовательности $|X(e^{j\hat\omega})|$, рассчитанный через дополнение нулями до 4096 отсчётов.

```python
X_fin = np.array([np.sum(xn * np.exp(-2j * np.pi * kk * n / N)) for kk in k])
print('Расхождение ДПФ по формуле и np.fft.fft:', np.max(np.abs(X_fin - X)))

M = 4096
X_cont = np.fft.fft(xn, M)       # спектр конечной последовательности на мелкой сетке
k_cont = np.arange(M) * N / M    # та же шкала k

fig, ax = plt.subplots(2, 1, figsize=(11, 8), sharex=True)
ax[0].plot(k_cont, np.abs(X_cont), color='lightgray', label='$|X(e^{j\\hat{\\omega}})|$ конечной последовательности')
ax[0].stem(k, np.abs(X_fin), label='|X(k)| (ДПФ)')
ax[0].set_title('Модуль ДПФ конечной последовательности')
ax[0].set_ylabel('|X(k)|')
ax[0].legend(loc='upper center')
ax[1].stem(k, amp)
ax[1].set_title('Амплитудный спектр периодической последовательности')
ax[1].set_ylabel('|X(k)|/N')
ax[1].set_xlabel('k')
fig.tight_layout()
plt.show()

print('Отношение |X(k)| к амплитудному спектру на ненулевых линиях:', np.abs(X_fin[lines]) / amp[lines])
```

ДПФ конечной последовательности длины $N$ вычисляется по той же формуле, что и коэффициенты ряда Фурье её периодического продолжения, поэтому отсчёты совпадают: $X(k) = \tilde X(k)$. Различие только в нормировке: модуль ДПФ в $N = 64$ раз больше амплитудного спектра, $|X(k)| = N\cdot\frac{|\tilde X(k)|}{N}$. Так, $33.28 = 64\cdot 0.52$ и $66.56 = 64\cdot 1.04$.

Смысл этих отсчётов разный. Спектр периодической последовательности линейчатый: кроме частот $k\Delta f$ в нём ничего нет. Спектр конечной последовательности непрерывный (серая кривая), а ДПФ - это его отсчёты в точках $\hat\omega_k = 2\pi k/N$. Между отсчётами непрерывный спектр не равен нулю (видны боковые лепестки), но в нашем случае сетка ДПФ попадает точно на максимумы главных лепестков и на нули боковых, поэтому в модуле ДПФ видны только четыре линии.

#### 5. Восстановление аналогового сигнала

Аналоговый сигнал восстанавливаем по отсчётам ДПФ:

$$
x(t) = \frac{1}{N}\sum_{k=-N/2}^{N/2-1}X_а(k)\exp\Big(j\frac{2\pi}{NT}kt\Big), \qquad
X_а(k) = \begin{cases}X(N+k), & -N/2 \le k \le -1,\\ X(k), & 0 \le k \le N/2-1,\end{cases}
$$

при $t \in [0; (N-1)T]$ с шагом $\Delta t = 0.25T$. Отсчёты второй половины ДПФ переносятся на отрицательные частоты, так что суммирование идёт по симметричному диапазону $k = -32, \ldots, 31$.

Восстановленный сигнал - сумма гармоник с частотами $f_k = k/(NT)$ и комплексными амплитудами $X_а(k)/N$, поэтому его амплитудный спектр - это линии $|X_а(k)|/N$ на этих частотах.

```python
k_a = np.arange(-N // 2, N // 2)
Xa = np.where(k_a < 0, X[(N + k_a) % N], X[k_a % N])

dt = 0.25 * T
t = np.arange(0, (N - 1) * T + dt / 2, dt)
print('Число точек по времени:', len(t), ', t от 0 до', t[-1], 'с')

x_t = np.zeros(len(t), dtype=complex)
for kk, Xk in zip(k_a, Xa):
    x_t += Xk * np.exp(1j * 2 * np.pi * kk * t / (N * T))
x_t /= N
print('max|Im x(t)| =', np.max(np.abs(x_t.imag)))
x_t = x_t.real

# исходный аналоговый сигнал на той же сетке
x_orig = A1 * np.cos(2 * np.pi * f1 * t + np.pi / 4) + A2 * np.cos(2 * np.pi * f2 * t + np.pi / 8)
print('max|x_восст(t) - x_исх(t)| =', np.max(np.abs(x_t - x_orig)))

f_a = k_a / (N * T)
amp_a = np.abs(Xa) / N
```

```python
fig, ax = plt.subplots(3, 2, figsize=(13, 12))

n_per = np.arange(2 * N)                      # два периода
ax[0, 0].stem(n_per, xn[n_per % N], markerfmt='.')
ax[0, 0].axvline(N - 0.5, color='r', ls='--')
ax[0, 0].set_title('Периодическая последовательность x(n), два периода')
ax[0, 0].set_xlabel('n')

ax[0, 1].stem(k, np.abs(X))
ax[0, 1].set_title('Модуль ДПФ |X(k)|')
ax[0, 1].set_xlabel('k')

ax[1, 0].plot(t * 1000, x_t)
ax[1, 0].plot(n * T * 1000, xn, 'r.', label='отсчёты x(nT)')
ax[1, 0].set_title('Восстановленный аналоговый сигнал')
ax[1, 0].set_xlabel('t, мс')
ax[1, 0].set_ylim(-3.5, 4.2)
ax[1, 0].legend(loc='upper right')

ax[1, 1].stem(f_a, amp_a)
ax[1, 1].set_title('Амплитудный спектр восстановленного сигнала')
ax[1, 1].set_xlabel('f, Гц')
ax[1, 1].set_ylabel('$|X_а(k)|/N$')

ax[2, 0].plot(t * 1000, x_orig, color='g')
ax[2, 0].set_title('Исходный аналоговый сигнал $x(t) = x(nT)|_{nT=t}$')
ax[2, 0].set_xlabel('t, мс')

ax[2, 1].plot(t * 1000, x_orig, color='g', lw=3, alpha=0.5, label='исходный')
ax[2, 1].plot(t * 1000, x_t, 'k--', lw=1, label='восстановленный')
ax[2, 1].set_xlim(0, 1.5)
ax[2, 1].set_title('Сравнение сигналов (начальный участок)')
ax[2, 1].set_xlabel('t, мс')
ax[2, 1].legend(loc='upper right')

fig.tight_layout()
plt.show()

print(' f, Гц    |X_a(k)|  |X_a(k)|/N')
for i in np.nonzero(amp_a > 1e-10)[0]:
    print(f'{f_a[i]:7.1f} {np.abs(Xa[i]):9.4f} {amp_a[i]:9.4f}')
```

#### 6. Анализ результатов восстановления

**Связь модуля ДПФ со спектром аналогового сигнала.** Восстановленный периодический сигнал с периодом $NT = 6.4$ мс имеет линейчатый спектр с шагом $1/(NT) = \Delta f = 156.25$ Гц, и комплексная амплитуда линии на частоте $k\Delta f$ равна $X_а(k)/N$. Таким образом, модуль ДПФ с точностью до множителя $1/N$ даёт амплитудный спектр аналогового сигнала: отсчёты $k = 0, \ldots, N/2-1$ - положительные частоты, отсчёты $k = N/2, \ldots, N-1$ - отрицательные частоты $(k-N)\Delta f$. В нашем случае это линии $0.52$ на $\pm 1250$ Гц и $1.04$ на $\pm 1875$ Гц, что соответствует $A_1/2$ и $A_2/2$. Это справедливо, потому что спектр исходного сигнала ограничен частотой $1875 < f_д/2 = 5000$ Гц (теорема Котельникова выполняется, наложения спектров нет), а частоты гармоник кратны $\Delta f$.

**Сравнение восстановленного и исходного сигналов.** Визуально сигналы совпадают, кривая восстановленного сигнала проходит через все отсчёты $x(nT)$; численное расхождение порядка $10^{-14}$. Восстановленный сигнал содержит только две исходные гармоники, поэтому восстановление точное. Если бы частоты не были кратны $\Delta f$ или превышали $f_д/2$, восстановленный сигнал отличался бы от исходного между отсчётами.

#### 7. Равенство Парсеваля

$$
\sum_{n=0}^{N-1}|x(n)|^2 = \frac{1}{N}\sum_{k=0}^{N-1}|X(k)|^2.
$$

```python
E_time = np.sum(np.abs(xn) ** 2)
E_freq = np.sum(np.abs(X) ** 2) / N

print(f'Сумма во временной области:   {E_time:.6f}')
print(f'Сумма в частотной области:    {E_freq:.6f}')
print(f'Разность: {abs(E_time - E_freq):.2e}')
print(f'N * (A1^2 + A2^2) / 2 = {N * (A1 ** 2 + A2 ** 2) / 2:.6f}')
print(f'Средняя мощность E/N = {E_time / N:.6f}, (A1^2 + A2^2)/2 = {(A1 ** 2 + A2 ** 2) / 2:.6f}')
```

Равенство Парсеваля выполняется: обе суммы равны $173.056$ (разность на уровне ошибок округления).

Смысл равенства в том, что энергия последовательности на периоде одинакова при вычислении во временной и в частотной области: ДПФ не меняет энергию сигнала, а лишь распределяет её по частотным составляющим (множитель $1/N$ обусловлен нормировкой ДПФ). Для двух гармоник энергия равна $N\frac{A_1^2 + A_2^2}{2} = 64\cdot 2.704 = 173.056$, т.е. средняя мощность $2.704$ складывается из мощностей гармоник $A^2/2$. Мощности просто суммируются, так как гармоники с частотами, кратными $\Delta f$, ортогональны на периоде $N$.
